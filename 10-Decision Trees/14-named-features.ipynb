# A question with names, not numbers

Five parts have a material instead of a length.

| Material | Label |
|---|---|
| oak | pass |
| oak | pass |
| pine | pass |
| steel | fail |
| steel | fail |

A named feature is not cut at a midpoint. Each question is a subset of the names. The smallest useful questions are "is it this name?"

Parent counts: three pass, two fail.

$$
G = 1 - \left(\frac{3}{5}\right)^2 - \left(\frac{2}{5}\right)^2
= 1 - \frac{13}{25}
= \frac{12}{25}
$$


## Three questions

"Is it steel?" puts two fails on one side and three passes on the other. Both sides are pure. The weighted Gini is $0$, so the gain is the whole parent value, $12/25$.

"Is it oak?" puts two passes on one side. The other side is pine plus two steels: one pass and two fails, $n = 3$.

$$
G = 1 - \left(\frac{1}{3}\right)^2 - \left(\frac{2}{3}\right)^2 = \frac{4}{9}
$$

$$
\frac{3}{5}\cdot\frac{4}{9} = \frac{4}{15}
$$

$$
\operatorname{Gain} = \frac{12}{25} - \frac{4}{15}
= \frac{36}{75} - \frac{20}{75}
= \frac{16}{75}
$$

"Is it pine?" leaves four parts on the other side, two pass and two fail, Gini $1/2$.

$$
\frac{4}{5}\cdot\frac{1}{2} = \frac{2}{5} = \frac{10}{25}
$$

$$
\operatorname{Gain} = \frac{12}{25} - \frac{10}{25} = \frac{2}{25}
$$

Steel wins. Oak is second. Pine is a poor question because the leftover group is a tie.


```python
# Steel separates the five rows. Oak and pine do not.
from fractions import Fraction

rows = [("oak", "pass"), ("oak", "pass"), ("pine", "pass"), ("steel", "fail"), ("steel", "fail")]

def gini(labels):
    n = len(labels)
    counts = {}
    for label in labels:
        counts[label] = counts.get(label, 0) + 1
    return 1 - sum(Fraction(c, n) ** 2 for c in counts.values())

parent = gini([label for _, label in rows])
assert parent == Fraction(12, 25)

def gain(name):
    left = [label for material, label in rows if material == name]
    right = [label for material, label in rows if material != name]
    weighted = (
        Fraction(len(left), 5) * gini(left) + Fraction(len(right), 5) * gini(right)
    )
    return parent - weighted

assert gain("steel") == Fraction(12, 25)
assert gain("oak") == Fraction(16, 75)
assert gain("pine") == Fraction(2, 25)
print("steel", gain("steel"), "oak", gain("oak"), "pine", gain("pine"))
```

## Same gain, more leaves

A three-way split, one branch per material, is also pure on every branch. Its weighted Gini is $0$, so its gain is also $12/25$. The gain ties the steel question, and the three-way split uses three leaves instead of two.

When the gains tie, keep the question with fewer branches. "Is it steel?" is that question. The oak and pine rows share a leaf, which is honest: both passed.


## A pitfall

Sorting names alphabetically and then cutting them like numbers invents an order the names do not have. Oak, pine, and steel are not positions on a line. The questions are membership questions.

## What to carry forward

On this table, "is it steel?" gains $12/25$, "is it oak?" gains $16/75$, and "is it pine?" gains $2/25$. A three-way split matches the steel gain and spends an extra leaf.
