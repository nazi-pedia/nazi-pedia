# Entropy

Entropy is another score for the same leaf. With proportions $p_k > 0$,

$$
H = - \sum_k p_k \log_2 p_k
$$

A proportion of $0$ is taken to contribute $0$. That is a limit, and it is worth a short proof so the pure leaf is allowed to contain the formula.

Let $p = 1/2^n$ for $n = 1, 2, 3, \ldots$. Then

$$
p \log_2 p = \frac{1}{2^n} \cdot (-n) = -\frac{n}{2^n}
$$

The fraction $n/2^n$ goes to $0$ as $n$ grows, because the denominator doubles each time and the numerator only adds $1$. So along this sequence, $p \log_2 p \to 0$. A pure class, the one with proportion $0$, adds nothing to $H$.

For a fair coin, $p = 1/2$:

$$
H = -\frac{1}{2}\log_2\frac{1}{2} - \frac{1}{2}\log_2\frac{1}{2}
= -\frac{1}{2}(-1) - \frac{1}{2}(-1)
= 1
$$

One bit. A pure leaf has $H = 0$.


## The same eight parts

Four and four:

$$
H = 1
$$

The best cut in the next lesson will leave six parts on one side, four pass and two fail, and two fails on the other side. The pure side has entropy $0$. The mixed side has proportions $2/3$ and $1/3$:

$$
\begin{aligned}
H
&= -\frac{2}{3}\log_2\frac{2}{3} - \frac{1}{3}\log_2\frac{1}{3} \\
&= -\frac{2}{3}(1 - \log_2 3) - \frac{1}{3}(0 - \log_2 3) \\
&= -\frac{2}{3} + \log_2 3
\end{aligned}
$$

Weight that side by $6/8 = 3/4$:

$$
\frac{3}{4}\left(-\frac{2}{3} + \log_2 3\right)
= -\frac{1}{2} + \frac{3}{4}\log_2 3
$$

The drop from the parent's $1$ bit is

$$
1 - \left(-\frac{1}{2} + \frac{3}{4}\log_2 3\right)
= \frac{3}{2} - \frac{3}{4}\log_2 3
$$

Numerically that is about $0.311$. Gini's drop on this same cut will be the fraction $1/6$. Both say the leaf got purer. The units differ, so the two drops are not subtracted from each other.


```python
# Entropy of a tie, of a pure leaf, and of the four-and-two leaf.
import math

def entropy(labels):
    n = len(labels)
    counts = {}
    for label in labels:
        counts[label] = counts.get(label, 0) + 1
    total = 0.0
    for count in counts.values():
        proportion = count / n
        total -= proportion * math.log2(proportion)
    return total

assert entropy(["pass", "fail"]) == 1
assert entropy(["pass", "pass"]) == 0
mixed = entropy(["pass"] * 4 + ["fail"] * 2)
closed_form = -2 / 3 + math.log2(3)
assert abs(mixed - closed_form) < 1e-12
gain = 1 - (3 / 4) * mixed
assert abs(gain - (1.5 - 0.75 * math.log2(3))) < 1e-12
print("mixed entropy", mixed)
print("information gain", gain)
```

## A pitfall

$\log_2(3)$ does not become a friendly fraction. Leave it in the exact answer, and use `math.log2` when you need a decimal. Replacing it by $1.58$ inside a later equality check will fail for no statistical reason.

## What to carry forward

Entropy is $-\sum p_k\log_2 p_k$. A tie is $1$ bit and a pure leaf is $0$. On the four-and-two leaf it equals $-2/3 + \log_2 3$.
