## Import SymPy

```python
from sympy import symbols, integrate, sin, cos, exp, diff, tan, sec, sqrt, ln, Rational
```

## Define the Variable

```python
x = symbols('x')
```

## Basic Polynomial Integral

$$ \int_{}^{}3x^2dx=3.\frac{x^{3}}{3}+C=x^3+C $$

```python
f = 3 * x**2
integral = integrate(f, x)
print("∫ 3x² dx =", integral, "+ C")
```

## Trigonometric Integral

$$ \int_{}^{}sin(x)dx=-cos(x)+C $$

```python
f = sin(x)
integral = integrate(f, x)
print("∫ sin(x) dx =", integral, "+ C")
```

## Multiple Terms (Linearity)

$$ \int_{}^{}(2x+3)dx=x^2+3x+C $$

```python
f = 2 * x + 3
integral = integrate(f, x)
print("∫ (2x + 3) dx =", integral, "+ C")
```

## Exponential Functions

$$ \int_{}^{}e^{x}dx=e^x+C $$

```python
f = exp(x)
integral = integrate(f, x)
print("∫ eˣ dx =", integral, "+ C")
```

## Integration by Parts

$$ \int_{}^{}xe^{x}dx $$

```python
f = x * exp(x)
integral = integrate(f, x)
print("∫ x·eˣ dx =", integral, "+ C")
```
