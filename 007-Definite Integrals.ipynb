## Python Setup

```python
import sympy as sp

# Define the symbolic variable
x = sp.symbols('x')
```

## Example 1 — A Simple Polynomial

$$ \int_{0}^{1}x^{2}dx $$

```python
f = x**2

result = sp.integrate(f, (x, 0, 1))
print("The integral of x^2 from 0 to 1 is:", result)
```

## Example 2 — Trigonometric Function

$$ \int_{0}^{\pi}sin(x)dx $$

```python
f = sp.sin(x)
result = sp.integrate(f, (x, 0, sp.pi))
print("The integral of sin(x) from 0 to pi is:", result)
```

## Example 3 — Natural Logarithm

$$ \int_{1}^{e}ln(x)dx $$

```python
f = sp.ln(x)
result = sp.integrate(f, (x, 1, sp.E))
print("The integral of ln(x) from 1 to e is:", result)
```

## Visualizing the Limit Definition

```python
import numpy as np
import matplotlib.pyplot as plt

def riemann_plot(f, a, b, n, ax):
    x_vals = np.linspace(a, b, 400)
    ax.plot(x_vals, f(x_vals), 'r-', linewidth=2, label='f(x)')
    
    dx = (b - a) / n
    x_left = np.linspace(a, b - dx, n)
    heights = f(x_left)
    
    ax.bar(x_left, heights, width=dx, alpha=0.4,
           color='steelblue', edgecolor='black', align='edge')
    
    area = np.sum(heights * dx)
    ax.set_title(f'n = {n},  Approx. area = {area:.4f}')
    ax.legend()

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
f_np = lambda x: x**2

for ax, n in zip(axes, [4, 16, 64]):
    riemann_plot(f_np, 0, 1, n, ax)

plt.tight_layout()
plt.show()
```
