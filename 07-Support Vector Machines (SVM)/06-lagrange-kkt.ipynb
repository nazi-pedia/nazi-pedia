# Lagrange multipliers and the KKT conditions

The hard-margin problem is an inequality-constrained optimization. Each training point gets a multiplier $\alpha_i \ge 0$. The Lagrangian is

$$
\mathcal{L}(\mathbf{w}, b, \boldsymbol{\alpha})
= \frac{1}{2}\|\mathbf{w}\|^2
- \sum_i \alpha_i \bigl(y_i(\mathbf{w}\cdot\mathbf{x}_i+b) - 1\bigr)
$$

The term in parentheses is the slack in the margin constraint. If a constraint is loose, its multiplier will be zero. If a multiplier is positive, its constraint will be tight. That pairing is the whole idea of a support vector.


## Stationarity

Differentiate with respect to the weight vector. The derivative of $\tfrac{1}{2}\|\mathbf{w}\|^2$ is $\mathbf{w}$. The derivative of $\mathbf{w}\cdot\mathbf{x}_i$ in the direction of the weights is $\mathbf{x}_i$. Set the gradient to zero:

$$
\mathbf{w} = \sum_i \alpha_i y_i \mathbf{x}_i
$$

The optimal weight vector is a weighted sum of the training points. Only points with $\alpha_i > 0$ appear in the sum. Those are the support vectors.

Differentiate with respect to the bias. The bias appears only inside the constraints:

$$
\frac{\partial \mathcal{L}}{\partial b}
= - \sum_i \alpha_i y_i = 0
$$

so

$$
\sum_i \alpha_i y_i = 0
$$

The positive support and the negative support pull with equal total multiplier.


## The four KKT conditions

A point $(\mathbf{w}, b, \boldsymbol{\alpha})$ solves the hard-margin problem when all four of these hold.

1. **Stationarity.** $\mathbf{w} = \sum_i \alpha_i y_i \mathbf{x}_i$ and $\sum_i \alpha_i y_i = 0$.
2. **Primal feasibility.** $y_i(\mathbf{w}\cdot\mathbf{x}_i+b) \ge 1$ for every $i$.
3. **Dual feasibility.** $\alpha_i \ge 0$ for every $i$.
4. **Complementary slackness.** $\alpha_i \bigl(y_i(\mathbf{w}\cdot\mathbf{x}_i+b) - 1\bigr) = 0$ for every $i$.

Complementary slackness is the support-vector rule. If the margin is strictly larger than $1$, then $\alpha_i = 0$. If $\alpha_i > 0$, then the margin equals $1$.


## Check the three-point solution

Claimed multipliers:

$$
\alpha_{(2,0)} = \frac{1}{2},
\qquad
\alpha_{(0,1)} = \frac{1}{4},
\qquad
\alpha_{(0,-1)} = \frac{1}{4}
$$

Stationarity for $\mathbf{w}$:

$$
\begin{aligned}
\mathbf{w}
&= \frac{1}{2}(+1)(2, 0)
+ \frac{1}{4}(-1)(0, 1)
+ \frac{1}{4}(-1)(0, -1) \\
&= (1, 0) + (0, -\tfrac{1}{4}) + (0, \tfrac{1}{4}) \\
&= (1, 0)
\end{aligned}
$$

Stationarity for $b$:

$$
\frac{1}{2}(+1) + \frac{1}{4}(-1) + \frac{1}{4}(-1) = 0
$$

Every margin equals $1$, so primal feasibility and complementary slackness hold, and every $\alpha_i > 0$. If we add $(4, 0)$ with label $+1$, its multiplier must be $0$ because its margin is $3$.


```python
# KKT check for the three support vectors, plus a zero multiplier.
from fractions import Fraction

points = [(2, 0), (0, 1), (0, -1), (4, 0)]
labels = [1, -1, -1, 1]
alpha = [Fraction(1, 2), Fraction(1, 4), Fraction(1, 4), Fraction(0)]
b = Fraction(-1)

w1 = 0
w2 = 0
balance = 0
for (x1, x2), label, a in zip(points, labels, alpha):
    w1 += a * label * x1
    w2 += a * label * x2
    balance += a * label
assert (w1, w2) == (Fraction(1), Fraction(0))
assert balance == 0

for (x1, x2), label, a in zip(points, labels, alpha):
    gap = label * (w1 * x1 + w2 * x2 + b) - 1
    assert gap >= 0
    assert a * gap == 0
print("w", (w1, w2), "balance", balance)
print("alphas", alpha)
```

## A pitfall

A zero multiplier does not mean the point was ignored during modeling. It means that, at the optimum, the point is not pulling on the line. If you move it across the street, its multiplier can become positive and the line will move.

## What to carry forward

$\mathbf{w}$ is a combination of the support vectors. Their signed multipliers sum to zero. A multiplier and a loose margin cannot be nonzero at the same time.
