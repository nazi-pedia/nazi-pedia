# ۱۴ · بهینه‌سازی

| | |
|---|---|
| **دوره** | ریاضی عمومی ۱ |
| **پیش‌نیاز** | ۱۳ |

بهینه‌سازی یعنی بیشینه یا کمینهٔ یک تابع روی دامنه‌ای که مسئله اجازه داده است.
نقاط بحرانی داخل دامنه نامزدند. اگر دامنه بسته باشد، دو انتها هم نامزدند.
مقدار تابع را در نامزدها مقایسه کنید. شیب صفر، به‌تنهایی برنده را انتخاب نمی‌کند.

تابع را تا یک متغیر کم کنید. قید هندسی، متغیر دوم را حذف می‌کند.



## مثال ۱ — مستطیل با محیط ثابت

### حل ریاضی

محیط ۲۰ است. ضلع‌ها \(w\) و \(20/2-w=10-w\) هستند، با \(0<w<10\).

\[
A(w)=w(10-w)=10w-w^2
\]
\[
A'(w)=10-2w=0\implies w=5
\]

ضلع دیگر هم ۵ است. مربع، و مساحت ۲۵.
در دو انتهای بازه مساحت به ۰ می‌رود، پس ۲۵ بیشینه است.
مشتق دوم \(A''=-2<0\) همین را می‌گوید.

### بازبینی با پایتون



```python
import sympy as sp
w = sp.symbols("w", real=True)
A = w*(10 - w)
crit = sp.solve(sp.diff(A, w), w)[0]
print("ضلع:", crit, "مساحت:", A.subs(w, crit))
assert crit == 5
assert A.subs(w, crit) == 25
assert sp.diff(A, w, 2) == -2
print("تایید شد")

```

## مثال ۲ — نزدیک‌ترین نقطه

### حل ریاضی

نزدیک‌ترین نقطهٔ خم \(y=x^2\) به نقطهٔ \((0,1)\) را می‌خواهیم.
مربع فاصله را کمینه می‌کنیم تا رادیکال مزاحم نشود.

\[
s(x)=x^2+(x^2-1)^2
\]
\[
s'(x)=2x+2(x^2-1)\cdot 2x=2x\bigl(1+2(x^2-1)\bigr)=2x(2x^2-1)
\]

\[
x=0\quad\text{یا}\quad x=\pm\frac{1}{\sqrt{2}}
\]

\[
s(0)=1
\]
\[
s\left(\frac{1}{\sqrt{2}}\right)=\frac{1}{2}+\left(\frac{1}{2}-1\right)^2=\frac{1}{2}+\frac{1}{4}=\frac{3}{4}
\]

کمینهٔ مربع فاصله \(3/4\) است و خود فاصله \(\sqrt{3}/2\).
دو نقطه، با \(x\) های قرینه، به یک اندازه نزدیک‌اند. \((0,0)\) دورتر است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", real=True)
s = x**2 + (x**2 - 1)**2
crit = sp.solve(sp.diff(s, x), x)
print(crit)
values = {point: sp.simplify(s.subs(x, point)) for point in crit}
print(values)
assert sp.simplify(s.subs(x, 0)) == 1
assert sp.simplify(s.subs(x, 1/sp.sqrt(2))) == sp.Rational(3, 4)
print("تایید شد")

```

## مثال ۳ — جعبهٔ روباز

### حل ریاضی

از ورق مربع به ضلع ۱۲، مربع‌های کوچک به ضلع \(s\) از چهار گوشه بریده و دیواره‌ها تا می‌شوند.
\(0<s<6\).

\[
V(s)=s(12-2s)^2
\]
\[
V'(s)=(12-2s)^2+s\cdot 2(12-2s)\cdot(-2)
=(12-2s)\bigl[(12-2s)-4s\bigr]
=(12-2s)(12-6s)
\]

\(s=6\) انتهای دامنه است و حجم صفر می‌دهد. \(s=2\) درون دامنه است.

\[
V(2)=2\cdot 8^2=128
\]

در انتهاها حجم ۰ است، پس ۱۲۸ بیشینه است.

### بازبینی با پایتون



```python
import sympy as sp
s = sp.symbols("s", real=True)
V = s*(12 - 2*s)**2
crit = [c for c in sp.solve(sp.diff(V, s), s) if 0 < c < 6]
print("نامزد داخل بازه:", crit)
print("حجم:", [V.subs(s, c) for c in crit])
assert crit == [2]
assert V.subs(s, 2) == 128
print("تایید شد")

```

## مثال ۴ — بخش مثبت محور و خط

### حل ریاضی

خط \(y=4-2x\) محورها را در \((2,0)\) و \((0,4)\) قطع می‌کند.
مستطیلی با اضلاع موازی محور، زیر این خط و در ربع اول، بیشینهٔ مساحت را می‌خواهیم.

\[
A(x)=x(4-2x)=4x-2x^2,\qquad 0<x<2
\]
\[
A'(x)=4-4x=0\implies x=1
\]

ارتفاع \(4-2=2\) و مساحت ۲.
انتهاها مساحت ۰ دارند.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", real=True)
A = x*(4 - 2*x)
crit = sp.solve(sp.diff(A, x), x)[0]
print(crit, A.subs(x, crit))
assert crit == 1 and A.subs(x, crit) == 2
print("تایید شد")

```

## نکته

اگر قید جمع است، مثل محیط، متغیرها را خطی حذف کنید.
اگر قید فاصله است، مربع فاصله را کمینه کنید.
واحد و دامنه را در جملهٔ جواب بیاورید: «بیشینهٔ حجم ۱۲۸، وقتی برش ۲ باشد.»

## جمع‌بندی

یک متغیر، یک دامنه، فهرست نامزدها، مقایسهٔ مقدارها.
مشتق دوم میان‌بر است. مقایسه با انتها، وقتی دامنه بسته یا جواب در مرز مشکوک است، مطمئن‌تر است.
