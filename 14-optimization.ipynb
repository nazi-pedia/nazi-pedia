Optimization asks for the maximum or the minimum of a function on the domain the problem allows. Critical points inside the domain are candidates. If the domain is closed, the two endpoints are candidates as well. Compare the function values on that list. A zero slope, by itself, does not choose the winner.

Reduce the objective to one variable. A geometric constraint removes the second variable.


## Example 1 — A rectangle of fixed perimeter

### Hand solution

The perimeter is $20$. The sides are $w$ and $10-w$, with $0<w<10$.

$$
\begin{aligned}
\dfrac{20}{2}-w &= 10-w, \\
A(w) &= w(10-w)=10w-w^2, \\
A'(w) &= 10-2w=0 \implies w=5.
\end{aligned}
$$

The other side is $5$ as well. The rectangle is a square, and the area is $25$. At both ends of the interval the area tends to $0$, so $25$ is the maximum. The second derivative $A''=-2<0$ gives the same conclusion.

### Python recheck


```python
import sympy as sp

# Area in the single variable w.
w = sp.symbols("w", real=True)
A = w*(10 - w)

# The only critical point, and the area there.
crit = sp.solve(sp.diff(A, w), w)[0]
print("side:", crit, "area:", A.subs(w, crit))
assert crit == 5
assert A.subs(w, crit) == 25
# The second derivative is negative, so the critical point is a maximum.
assert sp.diff(A, w, 2) == -2
print("Confirmed")
```

## Example 2 — The nearest point

### Hand solution

We want the point on the curve $y=x^2$ nearest to the point $(0,1)$. Minimize the square of the distance, so the square root stays out of the derivative.

$$
\begin{aligned}
s(x) &= x^2+(x^2-1)^2, \\
s'(x) &= 2x+2(x^2-1)\cdot 2x=2x\bigl(1+2(x^2-1)\bigr)=2x(2x^2-1).
\end{aligned}
$$

$$
x=0\quad\text{or}\quad x=\pm\dfrac{1}{\sqrt{2}}.
$$

$$
\begin{aligned}
s(0) &= 1, \\
s\left(\dfrac{1}{\sqrt{2}}\right) &= \dfrac{1}{2}+\left(\dfrac{1}{2}-1\right)^2=\dfrac{1}{2}+\dfrac{1}{4}=\dfrac{3}{4}.
\end{aligned}
$$

The minimum of the squared distance is $\dfrac{3}{4}$, so the distance itself is $\dfrac{\sqrt{3}}{2}$. The two points with opposite values of $x$ are equally near. The point $(0,0)$ is farther away.

### Python recheck


```python
import sympy as sp

# Squared distance from (0, 1) to (x, x^2).
x = sp.symbols("x", real=True)
s = x**2 + (x**2 - 1)**2
crit = sp.solve(sp.diff(s, x), x)
print(crit)

# Value of s at each critical point.
values = {point: sp.simplify(s.subs(x, point)) for point in crit}
print(values)
assert sp.simplify(s.subs(x, 0)) == 1
assert sp.simplify(s.subs(x, 1/sp.sqrt(2))) == sp.Rational(3, 4)
print("Confirmed")
```

## Example 3 — An open box

### Hand solution

From a square sheet of side $12$, cut squares of side $s$ from the four corners and fold the sides up. The domain is $0<s<6$.

$$
\begin{aligned}
V(s) &= s(12-2s)^2, \\
V'(s) &= (12-2s)^2+s\cdot 2(12-2s)\cdot(-2) \\
&= (12-2s)\bigl[(12-2s)-4s\bigr] \\
&= (12-2s)(12-6s).
\end{aligned}
$$

The value $s=6$ is an endpoint of the domain, and the volume there is zero. The value $s=2$ lies inside the domain.

$$
V(2)=2\cdot 8^2=128.
$$

At the endpoints the volume is $0$, so $128$ is the maximum.

### Python recheck


```python
import sympy as sp

# Volume after the corner squares are removed.
s = sp.symbols("s", real=True)
V = s*(12 - 2*s)**2

# Keep only critical points inside (0, 6).
crit = [c for c in sp.solve(sp.diff(V, s), s) if 0 < c < 6]
print("candidate inside the interval:", crit)
print("volume:", [V.subs(s, c) for c in crit])
assert crit == [2]
assert V.subs(s, 2) == 128
print("Confirmed")
```

## Example 4 — The positive axes and a line

### Hand solution

The line $y=4-2x$ meets the axes at $(2,0)$ and $(0,4)$. Among rectangles with sides parallel to the axes, lying under this line in the first quadrant, we want the greatest area.

$$
\begin{aligned}
A(x) &= x(4-2x)=4x-2x^2, \qquad 0<x<2, \\
A'(x) &= 4-4x=0 \implies x=1.
\end{aligned}
$$

The height and the area are

$$
\begin{aligned}
4-2 &= 2, \\
A(1) &= 2.
\end{aligned}
$$

The endpoints have area $0$.

### Python recheck


```python
import sympy as sp

# Area of the inscribed rectangle.
x = sp.symbols("x", real=True)
A = x*(4 - 2*x)
crit = sp.solve(sp.diff(A, x), x)[0]
print(crit, A.subs(x, crit))
assert crit == 1 and A.subs(x, crit) == 2
print("Confirmed")
```

## A pitfall

When the constraint is a sum, as a perimeter is, eliminate the extra variable linearly. When the constraint is a distance, minimize the square of the distance. Put the unit and the domain into the answer sentence: "The maximum volume is $128$, when the cut is $2$."

## Summary

One variable, one domain, a list of candidates, and a comparison of values. The second derivative is a shortcut. Comparison with the endpoints is the safer check when the domain is closed, or when a boundary answer is in doubt.
