# A complete fit, with every sum written out

The formulas are only useful if they can be carried through a fresh table without hiding a step. This lesson fits a five-point line from the raw sums, checks the centered sums, and then checks the two orthogonality conditions one product at a time.


## The table and the sums

$$
\begin{array}{c|ccccc}
x & 1 & 2 & 3 & 4 & 5 \\
\hline
y & 4 & 4 & 7 & 8 & 12
\end{array}
$$

$$
\begin{aligned}
n &= 5, \\
\sum x &= 15, \qquad \sum y = 35, \\
\sum xy &= 1\cdot 4 + 2\cdot 4 + 3\cdot 7 + 4\cdot 8 + 5\cdot 12 = 4 + 8 + 21 + 32 + 60 = 125, \\
\sum x^2 &= 1 + 4 + 9 + 16 + 25 = 55.
\end{aligned}
$$

Therefore

$$
\begin{aligned}
b_1 &= \dfrac{5\cdot 125 - 15\cdot 35}{5\cdot 55 - 15^2}
= \dfrac{625 - 525}{275 - 225}
= \dfrac{100}{50}
= 2, \\
b_0 &= \dfrac{35 - 2\cdot 15}{5} = \dfrac{5}{5} = 1.
\end{aligned}
$$

The line is $\widehat y = 1 + 2x$.


## The same answer after centering

$$
\bar x = 3, \qquad \bar y = 7.
$$

The centered $x$ values are $-2, -1, 0, 1, 2$, so

$$
S_{xx} = 4 + 1 + 0 + 1 + 4 = 10.
$$

The centered products $(x_i - \bar x)(y_i - \bar y)$ are

$$
\begin{aligned}
&(-2)(4-7) + (-1)(4-7) + 0\cdot(7-7) + (1)(8-7) + (2)(12-7) \\
&\qquad = 6 + 3 + 0 + 1 + 10 = 20.
\end{aligned}
$$

Hence $b_1 = 20/10 = 2$ and $b_0 = 7 - 2\cdot 3 = 1$. Changing the bookkeeping must not change the line.


```python
# Raw sums and centered sums on the five-point table.
from fractions import Fraction
x = [1, 2, 3, 4, 5]
y = [4, 4, 7, 8, 12]
n = len(x)
sum_x, sum_y = sum(x), sum(y)
sum_xy = sum(a * b for a, b in zip(x, y))
sum_x2 = sum(a * a for a in x)
slope = Fraction(n * sum_xy - sum_x * sum_y, n * sum_x2 - sum_x ** 2)
intercept = Fraction(sum_y - slope * sum_x, n)
xbar, ybar = Fraction(sum_x, n), Fraction(sum_y, n)
sxx = sum((a - xbar) ** 2 for a in x)
sxy = sum((a - xbar) * (b - ybar) for a, b in zip(x, y))
print("sums", n, sum_x, sum_y, sum_xy, sum_x2)
print("line", intercept, slope)
print("centered", sxx, sxy, sxy / sxx, ybar - (sxy / sxx) * xbar)
assert (n, sum_x, sum_y, sum_xy, sum_x2) == (5, 15, 35, 125, 55)
assert sxx == 10 and sxy == 20
assert intercept == 1 and slope == 2
```

## Fitted values and residuals

$$
\begin{array}{c|ccccc}
\widehat y = 1 + 2x & 3 & 5 & 7 & 9 & 11 \\
\hline
e = y - \widehat y & 1 & -1 & 0 & -1 & 1
\end{array}
$$

The normal equations, checked entry by entry:

$$
\begin{aligned}
\sum e_i &= 1 - 1 + 0 - 1 + 1 = 0, \\
\sum x_i e_i &= 1\cdot 1 + 2\cdot(-1) + 3\cdot 0 + 4\cdot(-1) + 5\cdot 1 = 1 - 2 + 0 - 4 + 5 = 0, \\
\mathrm{RSS} &= 1 + 1 + 0 + 1 + 1 = 4.
\end{aligned}
$$

There are $n - 2 = 3$ degrees of freedom left after estimating two coefficients. The unbiased estimate of the error variance, used in a later lesson, will be $4/3$. Only the residual sum is needed here.


```python
# Predictions, residuals, and the two inner products required by the normal equations.
x = [1, 2, 3, 4, 5]
y = [4, 4, 7, 8, 12]
fitted = [1 + 2 * xi for xi in x]
errors = [yi - yh for yi, yh in zip(y, fitted)]
print("fitted", fitted)
print("residuals", errors)
print("sum e", sum(errors))
print("sum x e", sum(xi * ei for xi, ei in zip(x, errors)))
print("RSS", sum(ei ** 2 for ei in errors))
assert fitted == [3, 5, 7, 9, 11]
assert errors == [1, -1, 0, -1, 1]
assert sum(errors) == 0
assert sum(xi * ei for xi, ei in zip(x, errors)) == 0
assert sum(ei ** 2 for ei in errors) == 4
```

## One change: edit a single response

Replace the last response $12$ by $10$, so the new table ends in $(5, 10)$. The sums that involve $y$ change, and the others do not.

$$
\begin{aligned}
\sum y &= 33, \qquad \sum xy = 125 - 5\cdot 2 = 115, \\
b_1 &= \dfrac{5\cdot 115 - 15\cdot 33}{5\cdot 55 - 225}
= \dfrac{575 - 495}{50} = \dfrac{80}{50} = \dfrac{8}{5}, \\
b_0 &= \dfrac{33 - \tfrac{8}{5}\cdot 15}{5}
= \dfrac{33 - 24}{5} = \dfrac{9}{5}.
\end{aligned}
$$

One observation moved the slope from $2$ to $8/5$ and the intercept from $1$ to $9/5$. A fitted line is a function of every row, not a property of the first few.


```python
# Changing the last response changes both the slope and the intercept.
from fractions import Fraction
x = [1, 2, 3, 4, 5]
y = [4, 4, 7, 8, 10]
n = 5
sum_x, sum_y = sum(x), sum(y)
sum_xy = sum(a * b for a, b in zip(x, y))
sum_x2 = sum(a * a for a in x)
slope = Fraction(n * sum_xy - sum_x * sum_y, n * sum_x2 - sum_x ** 2)
intercept = Fraction(sum_y - slope * sum_x, n)
print("edited line", intercept, slope)
assert sum_y == 33 and sum_xy == 115
assert intercept == Fraction(9, 5) and slope == Fraction(8, 5)
```

## Pitfall

Fitting $1 + 2x$ and then rounding the predictions before computing residuals does not preserve the normal equations. The exact fitted values are the integers $3, 5, 7, 9, 11$ already, so rounding is harmless here. On the three-point line the fitted values are $11/6$, $10/3$, and $29/6$. Rounding them to $2, 3, 5$ produces residuals $0, 0, 0$ and pretends the line passed through every point. It did not: the exact residual sum of squares is $1/6$, not $0$.


```python
# Rounding fitted values can invent a perfect fit.
from fractions import Fraction
y = [Fraction(2), Fraction(3), Fraction(5)]
exact = [Fraction(11, 6), Fraction(10, 3), Fraction(29, 6)]
rounded = [round(float(value)) for value in exact]
exact_rss = sum((yi - yh) ** 2 for yi, yh in zip(y, exact))
rounded_rss = sum((yi - yh) ** 2 for yi, yh in zip(y, rounded))
print("rounded fitted values", rounded)
print("exact RSS", exact_rss)
print("rounded RSS", rounded_rss)
assert exact_rss == Fraction(1, 6)
assert rounded_rss == 0
```

## Summary

- Compute $\sum x$, $\sum y$, $\sum xy$, and $\sum x^2$, then $b_1$ and $b_0$. Centering is a check, not a different estimator.
- On this table the fit is exactly $1 + 2x$, the residuals are $1, -1, 0, -1, 1$, and $\mathrm{RSS} = 4$.
- Keep fractions until the residuals have been checked. Rounding the fitted values can erase a real error.
