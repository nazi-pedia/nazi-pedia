# Centering

Subtracting $\bar x$ from the predictor does not change the fitted values of a simple regression. It moves the intercept from "the prediction at $x = 0$" to "the prediction at the average $x$," which is often the quantity the intercept was being asked to name.


## The algebra

Let $c_i = x_i - \bar x$ and fit $y = a_0 + a_1 c$. The centered predictor has mean zero, so the normal equations separate:

$$
a_1 = \dfrac{\sum c_i(y_i-\bar y)}{\sum c_i^2} = \dfrac{S_{xy}}{S_{xx}} = b_1, \qquad a_0 = \bar y.
$$

The slope is the same number as in the uncentered fit. The new intercept is the mean response. The predictions agree at every row:

$$
a_0 + a_1(x_i - \bar x) = \bar y + b_1(x_i - \bar x) = (\bar y - b_1\bar x) + b_1 x_i = b_0 + b_1 x_i.
$$

Centering is a reparameterization. It is not a different model.


## The three-point line

For $(1, 2),\ (2, 3),\ (3, 5)$, the uncentered fit is $b_0 = 1/3$, $b_1 = 3/2$. The centered predictor is $-1, 0, 1$, and $\bar y = 10/3$, so the centered fit is

$$
\widehat y = \dfrac{10}{3} + \dfrac{3}{2}\,c.
$$

Evaluate it at the three centered inputs:

$$
\begin{aligned}
c = -1 &\Longrightarrow \dfrac{10}{3} - \dfrac{3}{2} = \dfrac{11}{6}, \\
c = 0 &\Longrightarrow \dfrac{10}{3}, \\
c = 1 &\Longrightarrow \dfrac{10}{3} + \dfrac{3}{2} = \dfrac{29}{6}.
\end{aligned}
$$

These are the uncentered fitted values. What changed is the sentence one is allowed to say about the intercept. $1/3$ is the prediction at $x = 0$, a place where this table has no observation. $10/3$ is the prediction at $x = 2$, the middle of the table.


```python
# Centering replaces the intercept and preserves every fitted value.
from fractions import Fraction
x = [Fraction(1), Fraction(2), Fraction(3)]
y = [Fraction(2), Fraction(3), Fraction(5)]
xbar = sum(x) / len(x)
centered = [xi - xbar for xi in x]
old = [Fraction(1, 3) + Fraction(3, 2) * xi for xi in x]
new = [Fraction(10, 3) + Fraction(3, 2) * ci for ci in centered]
print("centered x", centered)
print("uncentered fitted", old)
print("centered fitted", new)
assert centered == [Fraction(-1), Fraction(0), Fraction(1)]
assert old == new == [Fraction(11, 6), Fraction(10, 3), Fraction(29, 6)]
```

## One change: center $y$ as well

Replace $y$ by $y - \bar y$, which is $-4/3,\ -1/3,\ 5/3$, and keep the centered $x$. The slope is still $3/2$, because centering $y$ does not change $S_{xy}$ relative to a mean that is now zero. The intercept becomes $0$. The line passes through the origin of the centered coordinates, which is the point $(\bar x, \bar y)$ in the original coordinates.


```python
# Centering both axes puts the intercept at zero and keeps the slope.
from fractions import Fraction
c = [Fraction(-1), Fraction(0), Fraction(1)]
d = [Fraction(-4, 3), Fraction(-1, 3), Fraction(5, 3)]
slope = sum(a * b for a, b in zip(c, d)) / sum(a * a for a in c)
intercept = sum(d) / len(d)
print("centered response", d)
print("intercept and slope", intercept, slope)
assert intercept == 0
assert slope == Fraction(3, 2)
```

## Pitfall

Reporting the uncentered intercept $1/3$ as "the response when the predictor is absent" adds a story the fit did not tell. The value $x = 0$ is one unit outside the table $1, 2, 3$. The number $1/3$ is the extrapolation of this line to that point. After centering, the intercept $10/3$ is inside the range of $x$ and equals $\bar y$, which is a description of the observed responses rather than of a hypothetical zero.


```python
# x = 0 is outside the observed set. The centered intercept is the mean response.
from fractions import Fraction
observed = {1, 2, 3}
print("zero was observed", 0 in observed)
print("centered intercept", Fraction(10, 3))
assert 0 not in observed
assert Fraction(10, 3) == Fraction(sum([2, 3, 5]), 3)
```

## Summary

- Replacing $x$ by $x - \bar x$ leaves the slope and the fitted values unchanged.
- The new intercept equals $\bar y$, the prediction at the average predictor.
- An intercept at $x = 0$ is an extrapolation whenever $0$ lies outside the data.
