# What a neighbor model adds, and what a factorization adds

The bias formula has no notion of "people who agreed." The neighbor formula does. For a user $u$ and a movie $i$, take other users $v$ who rated $i$, weight them by a similarity $s_{uv}$ computed on the movies both have rated, and add their centered ratings:

$$
\hat r_{ui}
= \bar r_u
+ \dfrac{\sum_{v} s_{uv}\,(r_{vi} - \bar r_v)}{\sum_{v} |s_{uv}|}.
$$

On the toy, that sum had one term. U2's similarity to U1 was 1, U2's mean was $10/3$, and U2's rating of Room was 4, so the prediction was $14/3$. The bias formula on that same cell was 4.

An item-item version swaps the roles. It looks at movies $j$ that user $u$ has rated, weights them by a similarity between movie $i$ and movie $j$, and blends those ratings. The Pearson similarity is undefined when two people share fewer than two movies or when one of them gave the same star to every shared movie. The bias formula still returns a number there.

A factorization writes each user and each movie as a short list of numbers, and predicts with a dot product plus the same biases:

$$
\hat r_{ui} = \mu + b_u + b_i + \mathbf{p}_u \cdot \mathbf{q}_i.
$$

The dot product is what lets a kind of agreement survive without storing a similarity for every pair. Fitting those lists is an optimization problem. This path does not fit them. The number it stands behind is the bias RMSE on `ua.test`, about 0.992, from means that never saw those 9430 ratings.

Neighbor models and factorizations are the usual next step in write-ups of this dataset, because they can use agreements that a pair of averages ignores. Their published errors are not the 0.992 computed here, and this path does not borrow those errors.


```python
# The toy neighbor answer and the toy bias answer are different numbers.
from fractions import Fraction
bias = 4
neighbor = Fraction(14, 3)
print(bias, neighbor, neighbor - bias)
assert neighbor - bias == Fraction(2, 3)
assert neighbor != bias
```

## Pitfall

A library call that factorizes the whole of `u.data`, then reports an error on `ua.test`, has trained on the answers. The dot-product formula does not change that. The split does.

## Summary

One neighbor moved the toy prediction from 4 to $14/3$. A factorization would add a dot product on top of the biases. The audited error in this path remains the bias RMSE, about 0.992 on `ua.test`.
