# When two people move together

The bias formula uses only averages. A neighbor formula uses the shape of the ratings. Compare U1 and U2 on the two movies they both rated, Toy and Gold. U1 rated them 5 and 3. U2 rated them 4 and 2.

Center each person by their own mean on those shared movies. U1's mean is 4, so the centered ratings are $1, -1$. U2's mean is 3, so the centered ratings are $1, -1$. The Pearson correlation is

$$
\dfrac{(1)(1) + (-1)(-1)}{\sqrt{1^2 + (-1)^2}\,\sqrt{1^2 + (-1)^2}} = \dfrac{2}{2} = 1.
$$

They move together exactly, on this pair of movies. The uncentered cosine does not say that. The raw vectors $(5, 3)$ and $(4, 2)$ have dot product 26 and lengths $\sqrt{34}$ and $\sqrt{20}$, so

$$
\cos = \dfrac{26}{\sqrt{680}}, \qquad \cos^2 = \dfrac{676}{680} = \dfrac{169}{170}.
$$

Cosine here is high and still not 1. Centering is what made the correlation exact.

With Pearson similarity $1$, the neighbor prediction for U1 on Room uses U2's departure from U2's mean over all of U2's known ratings, not only the shared ones. That mean is $10/3$, and U2 rated Room as 4.

$$
\hat r = 4 + \left(4 - \dfrac{10}{3}\right) = \dfrac{14}{3}.
$$

The bias formula, on the same cell, said 4. The two formulas are allowed to disagree. $14/3$ is one neighbor. The bias answer used every known rating of U1 and of Room.

A correlation also fails open. One shared movie gives a vector of length 1, and a centered vector of all zeros has length 0. In either case there is no Pearson similarity to weight. U3, in the toy, has no usable neighbor for Gold. The bias formula still returned 1.


```python
# Pearson 1, cosine squared 169/170, and the neighbor prediction 14/3.
from fractions import Fraction
import math
shared_u1 = [5, 3]
shared_u2 = [4, 2]
center_u1 = [rating - 4 for rating in shared_u1]
center_u2 = [rating - 3 for rating in shared_u2]
dot = sum(a * b for a, b in zip(center_u1, center_u2))
norm = math.sqrt(sum(a * a for a in center_u1)) * math.sqrt(sum(b * b for b in center_u2))
pearson = dot / norm
raw_dot = 5 * 4 + 3 * 2
cosine_square = Fraction(raw_dot ** 2, (25 + 9) * (16 + 4))
neighbor = 4 + (4 - Fraction(10, 3))
print(pearson, cosine_square, neighbor)
assert center_u1 == [1, -1] and center_u2 == [1, -1]
assert abs(pearson - 1) < 1e-12
assert raw_dot == 26
assert cosine_square == Fraction(169, 170)
assert neighbor == Fraction(14, 3)
assert Fraction(14, 3) != 4
```

## Pitfall

A cosine computed after filling missing movies with 0 is a third formula. It rewards two people for skipping the same films. The Pearson number above uses only Toy and Gold, the movies both people actually rated.

## Summary

On the shared movies, U1 and U2 have Pearson correlation 1, while the raw cosine squares to $169/170$. The one-neighbor prediction for U1 on Room is $14/3$. The bias prediction was 4.
