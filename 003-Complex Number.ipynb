## 1. Two ways to write the same point

**See.** A complex number is a point in the plane, or the arrow from the origin to that point.
The point 3 units east and 4 units north is written \(3 + 4i\).

**Name.** In Python the imaginary unit is `1j`, not `i`. `j` is an engineering convention, and it is what the language uses.
`3 + 4j` is a literal. `complex(3, 4)` is the same value built from two numbers.
`.real` and `.imag` are floats, even when you started from integers.


```python
z = 3 + 4j
print(z)
print("same value via complex():", z == complex(3, 4))
print("real part", z.real, type(z.real))
print("imaginary part", z.imag, type(z.imag))
print("i^2 =", 1j ** 2)

real_part = 5
imaginary_part = 12
w = complex(real_part, imaginary_part)
print("built from names:", w)
assert 1j ** 2 == -1
assert w == 5 + 12j
```

**Read the output.** `1j ** 2` is `-1`. That single fact is why these numbers were introduced.
The real and imaginary parts are floats: `3.0` and `4.0`, not the integers 3 and 4.


## 2. Arithmetic is vector arithmetic, with a twist on multiplication

**See.** Addition is componentwise: \((3+4i) + (1+2i) = 4 + 6i\).
It is the parallelogram law for arrows. Subtraction is the same.

Multiplication is not componentwise. Use \(i^2 = -1\):

\[
(3+4i)(1+2i) = 3 + 6i + 4i + 8i^2 = 3 + 10i - 8 = -5 + 10i
\]

Division is multiplication by the reciprocal. The practical route is "multiply by the conjugate over the conjugate", which is section 4.


```python
z1 = 3 + 4j
z2 = 1 + 2j

print("sum", z1 + z2)
print("difference", z1 - z2)
print("product", z1 * z2)
print("quotient", z1 / z2)

assert z1 + z2 == 4 + 6j
assert z1 - z2 == 2 + 2j
assert z1 * z2 == (-5 + 10j)
assert z1 / z2 == (2.2 - 0.4j)
```

### Own it

Compute \((1+i)^2\) by hand, then check it. You should get \(2i\), not \(1+i\).


```python
result = (1 + 1j) ** 2
print(result)
assert result == 2j
print("checkpoint passed")
```

## 3. Equality is defined. Order is not.

**See.** On a line, "greater than" has a direction. In the plane, it does not.
Is \(i\) greater than \(1\)? East versus north is not a ranking.

**Name.** `==` compares values and is defined. `<`, `>`, `<=`, and `>=` raise `TypeError`.
The original version of this lesson crashed on `z1 > z2`. The crash is the lesson.
Catch it, name it, and do not sort a list of complex numbers with the default ordering.


```python
z1 = 3 + 4j
z2 = 3 + 4j
z3 = 1 + 2j
print("equal values:", z1 == z2)
print("different values:", z1 == z3)

try:
    print(z1 > z3)
except TypeError as err:
    print("order is undefined:", type(err).__name__ + ":", err)
```

**Misconception.** You can sort complex numbers by writing `sorted(numbers)`.

**Correction.** The default sort uses `<`, which complex numbers do not implement.
If you need an order for a table, sort by real part and then by imaginary part,
and say that this order is yours, not the complex numbers'.


```python
numbers = [1 + 2j, -1 + 0j, 1 - 3j, 0 + 1j]
ordered = sorted(numbers, key=lambda z: (z.real, z.imag))
print(ordered)
assert ordered[0] == -1 + 0j
```

## 4. Modulus and conjugate

**See.** The modulus \(|3+4i|\) is the length of the arrow.
The 3-4-5 right triangle says that length is 5.
The conjugate \(3-4i\) is the reflection across the real axis.

**Name.** `abs(z)` is the modulus. `z.conjugate()` is the conjugate.
The identity that makes division practical:

\[
z \cdot \overline{z} = |z|^2
\]

which is a real number. For \(3+4i\) it is 25.


```python
z = 3 + 4j
print("modulus", abs(z))
print("conjugate", z.conjugate())
print("z * conjugate", z * z.conjugate())
print("modulus squared", abs(z) ** 2)

assert abs(z) == 5
assert z.conjugate() == 3 - 4j
assert z * z.conjugate() == 25
assert abs(z) ** 2 == 25
```

**Check a second way.** \(|3+4i| = \sqrt{3^2 + 4^2}\). The next cell computes that with `math.hypot`
and demands agreement. From here on, a single printed complex number is not a finished answer.


```python
import math

z = 3 + 4j
by_abs = abs(z)
by_hypot = math.hypot(z.real, z.imag)
print(by_abs, by_hypot)
assert math.isclose(by_abs, by_hypot)
print("two methods agree")
```

## 5. Polar form

**See.** A point in the plane is also a distance and an angle.
\(1+i\) sits on the line \(y = x\) in the first quadrant, at distance \(\sqrt{2}\), at angle \(\pi/4\).

**Name.** The argument is the angle from the positive real axis, in radians, returned by `cmath.phase`.
`math.degrees` converts to degrees when a human needs degrees.
`cmath.rect(r, theta)` builds the rectangular form from a modulus and an argument.
`cmath.exp(1j * theta)` is the point on the unit circle at that angle, because \(e^{i\theta} = \cos\theta + i\sin\theta\).

`math.sqrt(-1)` fails. `cmath.sqrt(-1)` returns `1j`. Use `cmath` when the radicand may be negative.


```python
import cmath
import math

print("cmath.sqrt(-1) =", cmath.sqrt(-1))
print("cmath.sqrt(-9) =", cmath.sqrt(-9))

try:
    math.sqrt(-1)
except ValueError as err:
    print("math.sqrt refuses a negative:", err)

z = 1 + 1j
radius = abs(z)
theta = cmath.phase(z)
print(f"modulus = {radius}")
print(f"argument in radians = {theta}")
print(f"argument in degrees = {math.degrees(theta)}")
assert math.isclose(radius, math.sqrt(2))
assert math.isclose(theta, math.pi / 4)

rebuilt = cmath.rect(radius, theta)
print("back to rectangular:", rebuilt)
assert math.isclose(rebuilt.real, 1) and math.isclose(rebuilt.imag, 1)
```

### Own it

\(3+4i\) has argument \(\arctan(4/3)\), not \(\pi/4\). It is a 3-4-5 triangle, not an isosceles one.
Check the modulus and the argument against `math.atan2(4, 3)`.


```python
import cmath
import math

z = 3 + 4j
theta = cmath.phase(z)
print("phase", theta)
print("atan2", math.atan2(4, 3))
assert math.isclose(abs(z), 5)
assert math.isclose(theta, math.atan2(4, 3))
print("checkpoint passed")
```

## 6. Powers

**Do.** Powers of \(2+3i\) are just repeated multiplication. The reciprocal is the power \(-1\),
and it equals the conjugate divided by the modulus squared.


```python
import math

z = 2 + 3j
print("z^2", z ** 2)
print("z^3", z ** 3)
print("z^4", z ** 4)
print("1/z", z ** -1)
print("conjugate / |z|^2", z.conjugate() / (abs(z) ** 2))

reciprocal = z ** -1
via_conjugate = z.conjugate() / (abs(z) ** 2)
assert math.isclose(reciprocal.real, via_conjugate.real)
assert math.isclose(reciprocal.imag, via_conjugate.imag)
```

## Workshop — plot the 3-4-5 arrow

A complex number you cannot see is easy to misread. The plot below draws \(3+4i\),
marks the modulus as the length of the segment, and marks the conjugate.
The title states the claim the picture is making.


```python
import matplotlib.pyplot as plt

z = 3 + 4j
conjugate = z.conjugate()

fig, ax = plt.subplots(figsize=(6, 6))
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.plot([0, z.real], [0, z.imag], color="#1f4e79", linewidth=2, label="3 + 4i")
ax.scatter([z.real], [z.imag], color="#1f4e79", zorder=3)
ax.scatter([conjugate.real], [conjugate.imag], color="#b85c38", zorder=3, label="conjugate 3 - 4i")
ax.annotate("length 5", xy=(1.5, 2.4), color="#1f4e79")
ax.set_aspect("equal")
ax.set_xlim(-1, 6)
ax.set_ylim(-6, 6)
ax.set_xlabel("real part")
ax.set_ylabel("imaginary part")
ax.set_title("Modulus of 3 + 4i is the hypotenuse 5")
ax.legend()
ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

## Workshop — a polar power

Compute \((2 e^{i\pi/4})^3\) two ways.

1. Rectangular: build \(2\cis(\pi/4)\) with `cmath.rect`, then cube it.
2. Polar rule: cube the modulus and triple the argument, then convert back.

De Moivre says those agree: \((r e^{i\theta})^n = r^n e^{in\theta}\).


```python
import cmath
import math

radius = 2
theta = math.pi / 4
rectangular = cmath.rect(radius, theta) ** 3

polar = cmath.rect(radius ** 3, 3 * theta)
print("by cubing the rectangular form:", rectangular)
print("by De Moivre:", polar)
assert math.isclose(rectangular.real, polar.real, rel_tol=1e-12, abs_tol=1e-12)
assert math.isclose(rectangular.imag, polar.imag, rel_tol=1e-12, abs_tol=1e-12)
print("two methods agree")
```

## Mistakes this lesson is meant to prevent

- Writing `i` instead of `1j`. `i` is an ordinary name and is undefined until you define it.
- Using `math.sqrt` on a negative number.
- Sorting complex numbers with `<`.
- Forgetting that `.real` and `.imag` are floats.
- Treating the argument as degrees. `cmath` uses radians. \(\pi/4\), not 45, goes into `cmath.rect`.
