# 20 Volumes: Disks, Washers, and Shells

| | |
|---|---|
| **Course** | Calculus II |
| **Prerequisite** | 19 |

Rotating a region about an axis produces a volume.

Disk: a cross-section is a disk of radius $R(x)$.

$$
V=\int_{a}^{b}\pi [R(x)]^{2}\,dx.
$$

Washer: a ring with outer radius $R$ and inner radius $r$.

$$
V=\int_{a}^{b}\pi\bigl(R^{2}-r^{2}\bigr)\,dx.
$$

Shell: coaxial cylinders of radius $x$, height $h(x)$, and thickness $dx$.

$$
V=\int_{a}^{b} 2\pi x\, h(x)\,dx.
$$

The radius is the distance to the axis of rotation, not necessarily the coordinate $y$ itself.



## Example 1 — Disk

### Hand solution

Rotate $y=\sqrt{x}$ from $0$ to $4$ about the $x$-axis. The cross-sectional radius is $\sqrt{x}$.

$$
\begin{aligned}
V &= \int_{0}^{4}\pi x\,dx \\
&= \pi\left[\dfrac{x^{2}}{2}\right]_{0}^{4} \\
&= 8\pi.
\end{aligned}
$$

### Python recheck



```python
import sympy as sp

# Disk volume for radius sqrt(x) from 0 to 4.
x = sp.symbols("x")
volume = sp.pi*sp.integrate(sp.sqrt(x)**2, (x, 0, 4))
print(volume)

# The volume is 8*pi.
assert volume == 8*sp.pi
print("Confirmed")

```

## Example 2 — Washer

### Hand solution

Rotate the region between $y=x$ and $y=x^{2}$, from $0$ to $1$, about the $x$-axis. The outer radius is $x$ and the inner radius is $x^{2}$.

$$
\begin{aligned}
V &= \pi\int_{0}^{1}\bigl(x^{2}-x^{4}\bigr)\,dx \\
&= \pi\left[\dfrac{x^{3}}{3}-\dfrac{x^{5}}{5}\right]_{0}^{1} \\
&= \pi\left(\dfrac{1}{3}-\dfrac{1}{5}\right) \\
&= \dfrac{2\pi}{15}.
\end{aligned}
$$

### Python recheck



```python
import sympy as sp

# Washer with outer radius x and inner radius x^2.
x = sp.symbols("x")
volume = sp.pi*sp.integrate(x**2 - (x**2)**2, (x, 0, 1))
print(volume)

# The volume is 2*pi/15.
assert sp.simplify(volume - 2*sp.pi/15) == 0
print("Confirmed")

```

## Example 3 — The same region with shells, about the $y$-axis

### Hand solution

The shell radius is $x$ and the height is $x-x^{2}$.

$$
\begin{aligned}
V &= 2\pi\int_{0}^{1} x(x-x^{2})\,dx \\
&= 2\pi\int_{0}^{1}(x^{2}-x^{3})\,dx \\
&= 2\pi\left[\dfrac{x^{3}}{3}-\dfrac{x^{4}}{4}\right]_{0}^{1} \\
&= 2\pi\left(\dfrac{1}{3}-\dfrac{1}{4}\right) \\
&= \dfrac{\pi}{6}.
\end{aligned}
$$

This volume is not the volume of Example 2, because the axis of rotation is different.

### Python recheck



```python
import sympy as sp

# Cylindrical shells of radius x and height x - x^2.
x = sp.symbols("x")
volume = 2*sp.pi*sp.integrate(x*(x - x**2), (x, 0, 1))
print(volume)

# The volume about the y-axis is pi/6.
assert sp.simplify(volume - sp.pi/6) == 0
print("Confirmed")

```

## Example 4 — A disk with no central hole: the sphere

### Hand solution

Rotating the semicircle $y=\sqrt{1-x^{2}}$ about the $x$-axis produces the unit sphere.

$$
\begin{aligned}
V &= \pi\int_{-1}^{1}(1-x^{2})\,dx \\
&= 2\pi\left[x-\dfrac{x^{3}}{3}\right]_{0}^{1} \\
&= 2\pi\left(1-\dfrac{1}{3}\right) \\
&= \dfrac{4\pi}{3}.
\end{aligned}
$$

This is the familiar volume of a sphere of radius $1$.

### Python recheck



```python
import sympy as sp

# Rotate the upper unit semicircle about the x-axis.
x = sp.symbols("x")
volume = sp.pi*sp.integrate(1 - x**2, (x, -1, 1))
print(volume)

# The unit sphere has volume 4*pi/3.
assert sp.simplify(volume - 4*sp.pi/3) == 0
print("Confirmed")

```

## A pitfall

Before writing $R^{2}-r^{2}$, say which curve is farther from the axis. If the axis is the line $y=1$, the radius is $\lvert y-1\rvert$, not $y$ itself.

## Summary

Disks are convenient for rotation about an axis perpendicular to the cross-section. Shells are convenient when the height is a function of the radius. One region and two axes produce two different volumes.
