# ۲۰ · حجم: قرص، واشر، پوسته

| | |
|---|---|
| **دوره** | ریاضی عمومی ۲ |
| **پیش‌نیاز** | ۱۹ |

دوران یک ناحیه حول یک محور، حجم می‌سازد.

قرص: مقطع، دایره‌ای به شعاع \(R(x)\) است.

\[
V=\int_a^b \pi R(x)^2\,dx
\]

واشر: حلقه‌ای با شعاع بیرونی \(R\) و درونی \(r\).

\[
V=\int_a^b \pi\bigl(R^2-r^2\bigr)\,dx
\]

پوسته: استوانه‌های هم‌محور به شعاع \(x\)، ارتفاع \(h(x)\) و ضخامت \(dx\).

\[
V=\int_a^b 2\pi x\, h(x)\,dx
\]

شعاع، فاصله تا محور دوران است، نه لزوماً خود \(y\).



## مثال ۱ — قرص

### حل ریاضی

\(y=\sqrt{x}\) از ۰ تا ۴، حول محور \(x\).
شعاع مقطع \(\sqrt{x}\) است.

\[
V=\int_0^4 \pi x\,dx=\pi\Bigl[\frac{x^2}{2}\Bigr]_0^4=8\pi
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
volume = sp.pi*sp.integrate(sp.sqrt(x)**2, (x, 0, 4))
print(volume)
assert volume == 8*sp.pi
print("تایید شد")

```

## مثال ۲ — واشر

### حل ریاضی

ناحیهٔ بین \(y=x\) و \(y=x^2\) از ۰ تا ۱، حول محور \(x\).
شعاع بیرونی \(x\) و شعاع درونی \(x^2\) است.

\[
V=\pi\int_0^1\bigl(x^2-x^4\bigr)\,dx
=\pi\Bigl[\frac{x^3}{3}-\frac{x^5}{5}\Bigr]_0^1
=\pi\left(\frac{1}{3}-\frac{1}{5}\right)=\frac{2\pi}{15}
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
volume = sp.pi*sp.integrate(x**2 - (x**2)**2, (x, 0, 1))
print(volume)
assert sp.simplify(volume - 2*sp.pi/15) == 0
print("تایید شد")

```

## مثال ۳ — همان ناحیه با پوسته، حول محور \(y\)

### حل ریاضی

شعاع پوسته \(x\) است و ارتفاع \(x-x^2\).

\[
V=2\pi\int_0^1 x(x-x^2)\,dx
=2\pi\int_0^1(x^2-x^3)\,dx
=2\pi\Bigl[\frac{x^3}{3}-\frac{x^4}{4}\Bigr]_0^1
=2\pi\left(\frac{1}{3}-\frac{1}{4}\right)=\frac{\pi}{6}
\]

این حجم با مثال ۲ یکی نیست، چون محور دوران فرق دارد.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
volume = 2*sp.pi*sp.integrate(x*(x - x**2), (x, 0, 1))
print(volume)
assert sp.simplify(volume - sp.pi/6) == 0
print("تایید شد")

```

## مثال ۴ — قرص با سوراخ مرکزی نیست؛ کره

### حل ریاضی

نیم‌دایرهٔ \(y=\sqrt{1-x^2}\) حول محور \(x\)، کرهٔ واحد را می‌سازد.

\[
V=\pi\int_{-1}^{1}(1-x^2)\,dx
=2\pi\Bigl[x-\frac{x^3}{3}\Bigr]_0^1
=2\pi\left(1-\frac{1}{3}\right)=\frac{4\pi}{3}
\]

این همان حجم معروف کره به شعاع ۱ است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
volume = sp.pi*sp.integrate(1 - x**2, (x, -1, 1))
print(volume)
assert sp.simplify(volume - 4*sp.pi/3) == 0
print("تایید شد")

```

## نکته

پیش از نوشتن \(R^2-r^2\)، بگویید کدام خم از محور دورتر است.
اگر محور \(y=1\) باشد، شعاع \(|y-1|\) است نه خود \(y\).

## جمع‌بندی

قرص برای دوران حول محوری که مقطع بر آن عمود است راحت است.
پوسته وقتی ارتفاع را به‌صورت تابع شعاع می‌بینید راحت است.
یک ناحیه و دو محور، دو حجم متفاوت می‌سازند.
