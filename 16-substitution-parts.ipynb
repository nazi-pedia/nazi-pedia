# ۱۶ · جانشینی و جزءبه‌جزء

| | |
|---|---|
| **دوره** | ریاضی عمومی ۲ |
| **پیش‌نیاز** | ۰۸، و معکوس دیدن مشتق |

انتگرال نامعین، خانواده‌ای از تابع‌هاست که مشتقشان انتگرال‌گیر است.

\[
\int f(x)\,dx = F(x)+C \iff F'(x)=f(x)
\]

SymPy یکی از عضوهای خانواده را برمی‌گرداند و \(C\) را نمی‌نویسد.
آزمون درست، مشتق گرفتن از جواب است، نه مقایسهٔ ظاهری دو فرمول.

**جانشینی.** اگر \(u=g(x)\) و \(du=g'(x)\,dx\) داخل انتگرال دیده شود، انتگرال به متغیر \(u\) منتقل می‌شود.

**جزءبه‌جزء** محصول قاعدهٔ ضرب است که از راست خوانده شود.

\[
\int u\,dv = uv-\int v\,du
\]

\(u\) را چیزی بگیرید که با مشتق ساده‌تر شود.



## مثال ۱ — جانشینیِ آشکار

### حل ریاضی

\[
\int 2x\cos(x^2)\,dx
\]

\(u=x^2\)، پس \(du=2x\,dx\).

\[
\int \cos u\,du=\sin u+C=\sin(x^2)+C
\]

مشتق \(\sin(x^2)\) با زنجیره برابر \(2x\cos(x^2)\) است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
found = sp.integrate(2*x*sp.cos(x**2), x)
hand = sp.sin(x**2)
print(found)
assert sp.simplify(sp.diff(found - hand, x)) == 0
print("تایید شد")

```

## مثال ۲ — جزءبه‌جزء

### حل ریاضی

\[
\int x e^{x}\,dx
\]

\(u=x\)، \(dv=e^{x}\,dx\). آن‌گاه \(du=dx\) و \(v=e^{x}\).

\[
xe^{x}-\int e^{x}\,dx=xe^{x}-e^{x}+C=e^{x}(x-1)+C
\]

مشتق \(e^{x}(x-1)\) برابر \(e^{x}(x-1)+e^{x}=xe^{x}\) است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
found = sp.integrate(x*sp.exp(x), x)
hand = sp.exp(x)*(x - 1)
print(sp.simplify(found))
assert sp.simplify(sp.diff(found - hand, x)) == 0
print("تایید شد")

```

## مثال ۳ — لگاریتم را \(u\) بگیرید

### حل ریاضی

\[
\int x\ln x\,dx,\qquad x>0
\]

\(u=\ln x\)، \(dv=x\,dx\). پس \(du=dx/x\) و \(v=x^2/2\).

\[
\frac{x^2}{2}\ln x-\int \frac{x^2}{2}\cdot\frac{1}{x}\,dx
=\frac{x^2}{2}\ln x-\int \frac{x}{2}\,dx
=\frac{x^2}{2}\ln x-\frac{x^2}{4}+C
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", positive=True)
found = sp.integrate(x*sp.log(x), x)
hand = (x**2/2)*sp.log(x) - x**2/4
print(sp.simplify(found))
assert sp.simplify(sp.diff(found - hand, x)) == 0
print("تایید شد")

```

## مثال ۴ — همان پادمشتق، این بار با کران

### حل ریاضی

\[
\int_0^1 x e^{x}\,dx
=\Bigl[e^{x}(x-1)\Bigr]_0^1
=(0)-e^{0}(0-1)=1
\]

ثابت \(C\) در تفاضل \(F(b)-F(a)\) حذف می‌شود. انتگرال معین عدد است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
exact = sp.integrate(x*sp.exp(x), (x, 0, 1))
by_ftc = sp.exp(1)*(1 - 1) - sp.exp(0)*(0 - 1)
print(exact, by_ftc)
assert exact == 1
assert sp.simplify(by_ftc - 1) == 0
print("تایید شد")

```

## مثال ۵ — جانشینی در انتگرال معین، کران هم عوض می‌شود

### حل ریاضی

\[
\int_0^{\pi/2}\sin x\cos x\,dx
\]

\(u=\sin x\). وقتی \(x=0\)، \(u=0\). وقتی \(x=\pi/2\)، \(u=1\). و \(du=\cos x\,dx\).

\[
\int_0^1 u\,du=\frac{1}{2}
\]

اگر کران‌ها را عوض نکنید و هنوز با \(x\) فکر کنید، جواب را در متغیر اشتباه ارزیابی کرده‌اید.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
exact = sp.integrate(sp.sin(x)*sp.cos(x), (x, 0, sp.pi/2))
print(exact)
assert exact == sp.Rational(1, 2)
print("تایید شد")

```

## نکته

در جزءبه‌جزء، اگر \(u\) را طوری بگیرید که درجهٔ جبری‌اش بالا برود، انتگرال بدتر می‌شود.
برای \(\int x e^{x}\)، چندجمله‌ای را \(u\) بگیرید نه نمایی را.

## جمع‌بندی

جانشینی یک زنجیره را وارونه می‌کند.
جزءبه‌جزء یک ضرب را وارونه می‌کند.
در هر دو، مشتق جواب باید انتگرال‌گیر را پس بدهد.
