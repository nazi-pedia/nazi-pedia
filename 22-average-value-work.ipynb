# ۲۲ · مقدار میانگین و کار

مقدار میانگین \(f\) روی \([a,b]\):

\[
f_{\mathrm{avg}}=\frac{1}{b-a}\int_a^b f(x)\,dx
\]

قضیهٔ مقدار میانگین برای انتگرال می‌گوید اگر \(f\) پیوسته باشد، نقطه‌ای هست که \(f(c)\) همین میانگین است.
این با قضیهٔ مقدار میانگین مشتق در درس ۱۱ هم‌نام است و هم‌معنا نیست.

کار نیروی متغیر در راستای حرکت:

\[
W=\int_a^b F(x)\,dx
\]

برای فنر، \(F(x)=kx\) اگر \(x\) را از وضعیت تعادل بگیرید.



## مثال ۱ — میانگین یک سهمی

### حل ریاضی

\[
\frac{1}{2-0}\int_0^2 x^2\,dx=\frac{1}{2}\cdot\frac{8}{3}=\frac{4}{3}
\]

نقطهٔ میانگین: \(c^2=4/3\)، پس \(c=2/\sqrt{3}\) داخل \((0,2)\).

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", positive=True)
avg = sp.integrate(x**2, (x, 0, 2))/2
c = sp.solve(sp.Eq(x**2, avg), x)[0]
print(avg, c)
assert avg == sp.Rational(4, 3)
assert 0 < c < 2
print("تایید شد")

```

## مثال ۲ — میانگین سینوس روی یک طاق

### حل ریاضی

\[
\frac{1}{\pi}\int_0^{\pi}\sin x\,dx=\frac{1}{\pi}\Bigl[-\cos x\Bigr]_0^{\pi}
=\frac{1}{\pi}(1+1)=\frac{2}{\pi}
\]

ارتفاع طاق تا ۱ می‌رود، ولی میانگین کمتر از ۱ است. \(2/\pi\approx 0.637\).

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
avg = sp.integrate(sp.sin(x), (x, 0, sp.pi))/sp.pi
print(avg, sp.N(avg, 6))
assert sp.simplify(avg - 2/sp.pi) == 0
print("تایید شد")

```

## مثال ۳ — کار فنر

### حل ریاضی

ثابت فنر \(k=300\) نیوتن بر متر است. فنر را ۲۰ سانتی‌متر، یعنی \(0.2\) متر، از تعادل می‌کشیم.

\[
W=\int_0^{0.2} 300x\,dx=150x^2\Big|_0^{0.2}=150\cdot 0.04=6
\]

ژول. اگر سانتی‌متر را به متر تبدیل نکنید، عدد ۱۵۰۰۰۰ برابر بزرگ می‌شود و واحد دروغ می‌گوید.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
work = sp.integrate(300*x, (x, 0, sp.Rational(1, 5)))
print(work)
assert work == 6
print("تایید شد")

```

## مثال ۴ — کار علیه گرانش، با نیروی ثابت به‌عنوان حالت خاص

### حل ریاضی

بالا بردن جرم ۲ کیلوگرم به ارتفاع ۳ متر، با \(g=9.8\)، نیروی \(19.6\) نیوتن است.

\[
W=\int_0^3 19.6\,dx=19.6\cdot 3=58.8
\]

ژول. انتگرال نیروی ثابت همان ضرب دبیرستانی است. فرمول انتگرال را برای وقتی نگه دارید که نیرو با مکان عوض می‌شود، مثل فنر.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
work = sp.integrate(sp.Rational(196, 10), (x, 0, 3))
print(work)
assert work == sp.Rational(294, 5)
print("تایید شد")

```

## نکته

میانگین، ضرب طول در یک ارتفاع نماینده است.
کار، جمع نیرو ضرب در جابه‌جایی‌های کوچک است.
هر دو انتگرال‌اند. واحد را پیش از عدد بنویسید.

## جمع‌بندی

میانگین را بر طول بازه تقسیم کنید.
کار فنر را از تعادل تا کشش نهایی انتگرال بگیرید، نه دو برابر آن، مگر اینکه صورت مسئله رفت و برگشت خواسته باشد.
