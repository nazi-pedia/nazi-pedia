The average value of $f$ on $[a,b]$ is

$$
f_{\text{avg}}=\dfrac{1}{b-a}\int_{a}^{b} f(x)\,dx.
$$

The Mean Value Theorem for integrals says that if $f$ is continuous, some point $c$ satisfies $f(c)=f_{\text{avg}}$. It shares a name with the Mean Value Theorem for derivatives in Lesson 11, and it does not share the same meaning.

The work of a variable force along the direction of motion is

$$
W=\int_{a}^{b} F(x)\,dx.
$$

For a spring, $F(x)=kx$ when $x$ is measured from equilibrium.



## Example 1 — Average of a parabola

### Hand solution

$$
\begin{aligned}
\dfrac{1}{2-0}\int_{0}^{2} x^{2}\,dx
&= \dfrac{1}{2}\cdot\dfrac{8}{3} \\
&= \dfrac{4}{3}.
\end{aligned}
$$

The mean-value point satisfies $c^{2}=\dfrac{4}{3}$, so $c=\dfrac{2}{\sqrt{3}}$ lies in $(0,2)$.

### Python recheck



```python
import sympy as sp

# Average value of x^2 on [0, 2], then a point where f equals it.
x = sp.symbols("x", positive=True)
avg = sp.integrate(x**2, (x, 0, 2))/2
c = sp.solve(sp.Eq(x**2, avg), x)[0]
print(avg, c)

# The average is 4/3, and c lies in (0, 2).
assert avg == sp.Rational(4, 3)
assert 0 < c < 2
print("Confirmed")

```

## Example 2 — Average of sine over one arch

### Hand solution

$$
\begin{aligned}
\dfrac{1}{\pi}\int_{0}^{\pi}\sin x\,dx
&= \dfrac{1}{\pi}\bigl[-\cos x\bigr]_{0}^{\pi} \\
&= \dfrac{1}{\pi}(1+1) \\
&= \dfrac{2}{\pi}.
\end{aligned}
$$

The arch rises to height $1$, but the average is less than $1$. Numerically, $\dfrac{2}{\pi}\approx 0.637$.

### Python recheck



```python
import sympy as sp

# Average value of sin x over one arch.
x = sp.symbols("x")
avg = sp.integrate(sp.sin(x), (x, 0, sp.pi))/sp.pi
print(avg, sp.N(avg, 6))

# The average is 2/pi.
assert sp.simplify(avg - 2/sp.pi) == 0
print("Confirmed")

```

## Example 3 — Work done by a spring

### Hand solution

The spring constant is $k=300$ newtons per meter. Stretch the spring $20$ centimeters, that is $0.2$ meters, from equilibrium.

$$
\begin{aligned}
W &= \int_{0}^{0.2} 300x\,dx \\
&= 150x^{2}\Big|_{0}^{0.2} \\
&= 150\cdot 0.04 \\
&= 6.
\end{aligned}
$$

The work is $6$ joules. If the centimeters are not converted to meters, the number becomes $150000$ times too large, and the unit is false.

### Python recheck



```python
import sympy as sp

# Work of F = 300x from equilibrium out to 1/5 meter.
x = sp.symbols("x")
work = sp.integrate(300*x, (x, 0, sp.Rational(1, 5)))
print(work)

# The work is 6 joules.
assert work == 6
print("Confirmed")

```

## Example 4 — Work against gravity, with a constant force as a special case

### Hand solution

Lifting a mass of $2$ kilograms through a height of $3$ meters, with $g=9.8$, requires a force of $19.6$ newtons.

$$
\begin{aligned}
W &= \int_{0}^{3} 19.6\,dx \\
&= 19.6\cdot 3 \\
&= 58.8.
\end{aligned}
$$

The work is $58.8$ joules. The integral of a constant force is the product learned in school. Keep the integral formula for the case in which the force changes with position, as it does for a spring.

### Python recheck



```python
import sympy as sp

# Constant force 19.6 newtons through 3 meters.
x = sp.symbols("x")
work = sp.integrate(sp.Rational(196, 10), (x, 0, 3))
print(work)

# The work is 294/5 joules.
assert work == sp.Rational(294, 5)
print("Confirmed")

```

## A pitfall

The integral behind an average is the length times one representative height. Work is the sum of force times small displacements. Both are integrals. Write the unit before the number.

## Summary

Divide by the length of the interval to obtain the average. Integrate the work of a spring from equilibrium to the final stretch, not twice that far, unless the problem asks for a round trip.
