# Pay for every extra leaf

The line of ten parts can be stored as three trees.

- $T_3$ has three leaves: six passes, three fails, and the single pass at $x = 10$. Training errors: $0$.
- $T_2$ stops after the first cut. The right leaf holds three fails and one pass, so the majority says fail and the tree makes $1$ error. Errors: $1/10$.
- $T_1$ is a single leaf. Seven passes beat three fails, so it predicts pass and makes $3$ errors. Errors: $3/10$.

Cost-complexity adds a price $\alpha \ge 0$ for each leaf:

$$
C(T) = R(T) + \alpha \,|T|
$$

$R(T)$ is the training error rate and $|T|$ is the number of leaves.

$$
\begin{aligned}
C(T_3) &= 0 + 3\alpha \\
C(T_2) &= \frac{1}{10} + 2\alpha \\
C(T_1) &= \frac{3}{10} + \alpha
\end{aligned}
$$


## Which tree wins

$T_3$ beats $T_2$ when

$$
3\alpha < \frac{1}{10} + 2\alpha
\qquad\Rightarrow\qquad
\alpha < \frac{1}{10}
$$

$T_2$ beats $T_1$ when

$$
\frac{1}{10} + 2\alpha < \frac{3}{10} + \alpha
\qquad\Rightarrow\qquad
\alpha < \frac{1}{5}
$$

So the ranking has three regions.

- If $\alpha < 1/10$, the three-leaf tree is cheapest.
- If $1/10 < \alpha < 1/5$, the two-leaf tree is cheapest.
- If $\alpha > 1/5$, the single leaf is cheapest.

At the boundaries the two neighboring trees tie.

Take $\alpha = 1/8$. It sits strictly between $1/10$ and $1/5$.

$$
\begin{aligned}
C(T_3) &= 3\cdot\frac{1}{8} = \frac{3}{8} = \frac{15}{40} \\
C(T_2) &= \frac{1}{10} + 2\cdot\frac{1}{8}
= \frac{1}{10} + \frac{1}{4}
= \frac{7}{20} = \frac{14}{40} \\
C(T_1) &= \frac{3}{10} + \frac{1}{8}
= \frac{17}{40}
\end{aligned}
$$

$14/40$ is the smallest. The price $1/8$ keeps the first cut and throws away the leaf that existed only to memorize $x = 10$.


```python
# At alpha = 1/8 the two-leaf tree is strictly cheapest.
from fractions import Fraction

alpha = Fraction(1, 8)
costs = {
    "three leaves": 0 + 3 * alpha,
    "two leaves": Fraction(1, 10) + 2 * alpha,
    "one leaf": Fraction(3, 10) + alpha,
}
assert costs["three leaves"] == Fraction(15, 40)
assert costs["two leaves"] == Fraction(14, 40)
assert costs["one leaf"] == Fraction(17, 40)
winner = min(costs, key=costs.get)
assert winner == "two leaves"
# Just below 1/10 and just above 1/5 the winner changes.
assert 0 + 3 * Fraction(1, 20) < Fraction(1, 10) + 2 * Fraction(1, 20)
assert Fraction(3, 10) + Fraction(1, 4) < Fraction(1, 10) + 2 * Fraction(1, 4)
for name, cost in costs.items():
    print(name, cost)
```

## A pitfall

$\alpha$ is not a property of the parts. It is the price you chose for a leaf. Changing it from $1/20$ to $1/4$ changes which tree is best, with the same ten rows. Report $\alpha$ next to the tree.

## What to carry forward

$C = R + \alpha|T|$. On this line, $\alpha < 1/10$ keeps three leaves, $1/10 < \alpha < 1/5$ keeps two, and $\alpha > 1/5$ keeps one. At $\alpha = 1/8$ the costs are $15/40$, $14/40$, and $17/40$.
