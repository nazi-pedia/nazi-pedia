# A leaf that predicts a number

Classification leaves name a class. Regression leaves name a number. The training score is the sum of squared errors. If the leaf predicts a constant $c$ and the observed values are $y_1, \ldots, y_n$,

$$
S(c) = \sum_{i=1}^{n} (y_i - c)^2
$$

Expand one term and differentiate. The derivative of $(y_i - c)^2$ with respect to $c$ is $-2(y_i - c)$, so

$$
S'(c) = -2\sum_{i=1}^{n} (y_i - c)
$$

Set the derivative to zero:

$$
\sum_{i=1}^{n} y_i - n c = 0
\qquad\Rightarrow\qquad
c = \frac{1}{n}\sum_{i=1}^{n} y_i
$$

The second derivative is $S''(c) = 2n > 0$, so this critical point is a minimum. The leaf predicts the mean. No other constant has a smaller sum of squares.


## Four yields

Settings $1, 2, 3, 4$ produced yields

$$
2, 2, 8, 8
$$

The mean is

$$
\frac{2+2+8+8}{4} = 5
$$

$$
S(5) = (2-5)^2 + (2-5)^2 + (8-5)^2 + (8-5)^2
= 4 \cdot 9
= 36
$$

The nearby constant $4$ is worse, and the arithmetic is short enough to see why:

$$
S(4) = 2\cdot(2-4)^2 + 2\cdot(8-4)^2 = 2\cdot 4 + 2\cdot 16 = 40
$$

$40 > 36$. The mean wins, as the derivative said it must.


```python
# The mean is the unique best constant. c = 4 costs 40, c = 5 costs 36.
from fractions import Fraction

yields = [2, 2, 8, 8]
mean = Fraction(sum(yields), len(yields))
assert mean == 5

def sse(constant):
    return sum((y - constant) ** 2 for y in yields)

assert sse(mean) == 36
assert sse(4) == 40
assert sse(6) == 40
print("mean", mean, "SSE", sse(mean))
```

## A pitfall

The mean minimizes squared error. It does not minimize the sum of absolute errors. That criterion picks a median. On $2, 2, 8, 8$ the two middle numbers are $2$ and $8$, and any constant between them ties for absolute error. Squared error still has one answer, $5$. Say which score you are minimizing before you name the leaf.

## What to carry forward

A regression leaf predicts the training mean, because $S'(c) = 0$ there and $S''(c) = 2n > 0$. These four yields have mean $5$ and squared error $36$.
