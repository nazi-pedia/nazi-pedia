# The query does not help compute the scale

The previous scale used two stored rows. Fold the query into the averages and the center moves.

Feature 1 becomes $0, 5, 0$. The mean is $5/3$, not $5/2$.

Feature 2 becomes $100, 101, 102$. The mean is $101$.

The root-mean-square spreads are now

$$
s_x = \sqrt{\frac{50}{9}} = \frac{5\sqrt{2}}{3}, \qquad
s_y = \sqrt{\frac{2}{3}}
$$

The scaled positions simplify:

$$
\begin{aligned}
A &\mapsto \left(-\frac{1}{\sqrt{2}},\ -\sqrt{\frac{3}{2}}\right) \\
B &\mapsto \left(\sqrt{2},\ 0\right) \\
Q &\mapsto \left(-\frac{1}{\sqrt{2}},\ \sqrt{\frac{3}{2}}\right)
\end{aligned}
$$

Squared distance from Q to A:

$$
\left(2\sqrt{\frac{3}{2}}\right)^2 = 4 \cdot \frac{3}{2} = 6
$$

Squared distance from Q to B:

$$
\left(-\frac{1}{\sqrt{2}} - \sqrt{2}\right)^2 + \left(\sqrt{\frac{3}{2}}\right)^2
= \left(-\frac{3}{\sqrt{2}}\right)^2 + \frac{3}{2}
= \frac{9}{2} + \frac{3}{2}
= 6
$$

The two stored visits are now exactly tied. Raw units picked A. Training-only scaling picked B. Scaling that swallowed the query picks neither.

A new query would move the mean again. The stored model would change every time someone asked a question. Compute the center and the spread once, from the stored rows, and apply those fixed numbers to every query.


```python
# Including Q makes the two squared distances both equal to 6.
from fractions import Fraction

mean_x = Fraction(0 + 5 + 0, 3)
mean_y = Fraction(100 + 101 + 102, 3)
assert mean_x == Fraction(5, 3)
assert mean_y == 101
# The algebra above already reduced both squared distances to 6.
to_a = 4 * Fraction(3, 2)
to_b = Fraction(9, 2) + Fraction(3, 2)
assert to_a == to_b == 6
print("means", mean_x, mean_y)
print("squared distances", to_a, to_b)
```

## A pitfall

A library that "scales the spreadsheet" will scale the query if the query is sitting in the same sheet. Split the stored mornings from the question before you compute a mean.

## What to carry forward

Training-only scaling made B nearer, $8$ against $16$. Putting Q into the scale ties them at squared distance $6$. The scale is part of the stored model.
