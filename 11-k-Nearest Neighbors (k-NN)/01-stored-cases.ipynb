# Look at the stored mornings

k-nearest neighbors does not fit a line and then throw the rows away. It keeps the rows. A new morning is answered by the mornings that sit nearest to it.

Six past mornings are stored. Each one records bookings, walk-ins, and whether the desk was busy or quiet.

| Case | Bookings | Walk-ins | Label |
|---|---:|---:|---|
| A | 2 | 5 | busy |
| B | 6 | 2 | busy |
| C | 5 | 6 | quiet |
| D | 2 | 8 | quiet |
| E | 10 | 2 | quiet |
| F | 7 | 14 | busy |

Today is a query, not a stored row: $2$ bookings and $2$ walk-ins. Call it $Q = (2, 2)$.

Count the memory before choosing a neighbor. Busy appears three times. Quiet appears three times.

$$
3 = 3
$$

There is no majority in the whole table. "Guess the more common label" has nothing to say. The local neighborhood is the whole method.


```python
# The memory is a tie. A global majority cannot answer today.
cases = [
    ("A", 2, 5, "busy"),
    ("B", 6, 2, "busy"),
    ("C", 5, 6, "quiet"),
    ("D", 2, 8, "quiet"),
    ("E", 10, 2, "quiet"),
    ("F", 7, 14, "busy"),
]
labels = [row[3] for row in cases]
assert labels.count("busy") == 3
assert labels.count("quiet") == 3
assert len(cases) == 6
print("busy", labels.count("busy"), "quiet", labels.count("quiet"))
```

## The four steps, before any arithmetic

1. Measure the distance from $Q$ to every stored morning.
2. Sort those distances.
3. Keep the $k$ shortest.
4. Let those $k$ mornings vote, or, if the target is a number, average them.

$k$ is a choice you write down. It is not estimated by the distance formula. This path will use $k = 1$, $k = 3$, $k = 5$, and $k = 6$ on the same six rows, so you can see the choice move the answer.

## A pitfall

The stored table is the model. If you delete case A, every future answer can change. There is no separate list of coefficients that survives the deletion.

## What to carry forward

k-NN stores the rows and consults the nearest ones. These six mornings are three busy and three quiet, so only a neighborhood can break the tie. Today is $(2, 2)$.
