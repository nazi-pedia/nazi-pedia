A vector in $\mathbb{R}^3$ is an arrow with three components.
The dot product gives length and angle:

$$
\mathbf{a}\cdot\mathbf{b}=|\mathbf{a}|\,|\mathbf{b}|\cos\theta.
$$

A zero dot product means the vectors are perpendicular.
The cross product is perpendicular to both factors, and its length is the area of the parallelogram they span.

A line is a point plus a scalar multiple of a direction vector.
A plane is given by

$$
\mathbf{n}\cdot(\mathbf{r}-\mathbf{r}_0)=0.
$$


## Example 1 — Perpendicular vectors

### Hand solution

$$
\begin{aligned}
(1,2,3)\cdot(3,0,-1)
&=3+0-3
=0.
\end{aligned}
$$

The two vectors are perpendicular. The first has length $\sqrt{14}$ and the second has length $\sqrt{10}$.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# The two vectors.
a = sp.Matrix([1, 2, 3])
b = sp.Matrix([3, 0, -1])
print(a.dot(b), a.norm(), b.norm())
# A zero dot product means the vectors are perpendicular.
assert a.dot(b) == 0
# Length of the first vector.
assert a.norm() == sp.sqrt(14)
print("Confirmed")
```

## Example 2 — Cross product and area

### Hand solution

$$
\mathbf{i}\times\mathbf{j}=\mathbf{k}.
$$

$$
\begin{aligned}
(1,2,0)\times(0,3,0)
&=
\begin{vmatrix}
\mathbf{i}&\mathbf{j}&\mathbf{k}\\
1&2&0\\
0&3&0
\end{vmatrix}
=(0,0,3).
\end{aligned}
$$

The area of the parallelogram is

$$
|(0,0,3)|=3.
$$

A base of $3$ and a height of $2$ give the same area: both vectors lie in the $xy$-plane.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Two vectors in the xy-plane.
u = sp.Matrix([1, 2, 0])
v = sp.Matrix([0, 3, 0])
# Cross product.
cross = u.cross(v)
print(cross, cross.norm())
assert cross == sp.Matrix([0, 0, 3])
# The area is the length of the cross product.
assert cross.norm() == 3
print("Confirmed")
```

## Example 3 — A point on a line

### Hand solution

The line through $(1,0,2)$ with direction $(1,1,0)$ is

$$
\mathbf{r}(t)=(1,0,2)+t(1,1,0).
$$

At $t=2$,

$$
\begin{aligned}
\mathbf{r}(2)
&=(1,0,2)+2(1,1,0)
=(3,2,2).
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Line parameter.
t = sp.symbols("t")
# Point on the line through (1, 0, 2) with direction (1, 1, 0).
point = sp.Matrix([1, 0, 2]) + t*sp.Matrix([1, 1, 0])
print(point.subs(t, 2))
# At t = 2 the point is (3, 2, 2).
assert point.subs(t, 2) == sp.Matrix([3, 2, 2])
print("Confirmed")
```

## Example 4 — Distance from the origin to a plane

### Hand solution

Write the plane $2x-y+z=4$ in the form $2x-y+z-4=0$.
The distance from the origin is

$$
\begin{aligned}
\dfrac{|-4|}{\sqrt{4+1+1}}
&=\dfrac{4}{\sqrt{6}}.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Normal vector of the plane 2x - y + z = 4.
normal = sp.Matrix([2, -1, 1])
# Distance from the origin to the plane.
distance = sp.Abs(-4)/normal.norm()
print(sp.simplify(distance))
assert sp.simplify(distance - 4/sp.sqrt(6)) == 0
print("Confirmed")
```

## A pitfall

The cross product is not commutative:

$$
\mathbf{a}\times\mathbf{b}=-\mathbf{b}\times\mathbf{a}.
$$

The area does not depend on the sign. The orientation of the plane does.

## Summary

The dot product is for angles and projections.
The cross product is for a normal vector and for area.
A line has one parameter, and a plane has one equation.
