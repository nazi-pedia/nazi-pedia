# ۲۸ · بردار، خط و صفحه

| | |
|---|---|
| **دوره** | ریاضی عمومی چندمتغیره |
| **پیش‌نیاز** | جبر مختصات |

بردار در \(\mathbb{R}^3\) یک پیکان با سه مؤلفه است.
ضرب داخلی، طول و زاویه را می‌دهد.

\[
\mathbf{a}\cdot\mathbf{b}=|\mathbf{a}||\mathbf{b}|\cos\theta
\]

صفر بودن ضرب داخلی یعنی عمود بودن.
ضرب خارجی بر هر دو عامل عمود است و طولش مساحت متوازی‌الاضلاع است.

خط: نقطه به‌علاوه مضرب یک بردار هادی.
صفحه: \(\mathbf{n}\cdot(\mathbf{r}-\mathbf{r}_0)=0\).



## مثال ۱ — عمود بودن

### حل ریاضی

\((1,2,3)\cdot(3,0,-1)=3+0-3=0\).
این دو بردار عمودند. طول اولی \(\sqrt{14}\) و دومی \(\sqrt{10}\) است.

### بازبینی با پایتون



```python
import sympy as sp
a = sp.Matrix([1, 2, 3])
b = sp.Matrix([3, 0, -1])
print(a.dot(b), a.norm(), b.norm())
assert a.dot(b) == 0
assert a.norm() == sp.sqrt(14)
print("تایید شد")

```

## مثال ۲ — ضرب خارجی و مساحت

### حل ریاضی

\[
\mathbf{i}\times\mathbf{j}=\mathbf{k}
\]

\[
(1,2,0)\times(0,3,0)=\begin{vmatrix}\mathbf{i}&\mathbf{j}&\mathbf{k}\\1&2&0\\0&3&0\end{vmatrix}=(0,0,3)
\]

مساحت متوازی‌الاضلاع \(|(0,0,3)|=3\) است. قاعده ۳ و ارتفاع ۲ هم همین را می‌دهد: دو بردار در صفحهٔ \(xy\) اند.

### بازبینی با پایتون



```python
import sympy as sp
u = sp.Matrix([1, 2, 0])
v = sp.Matrix([0, 3, 0])
cross = u.cross(v)
print(cross, cross.norm())
assert cross == sp.Matrix([0, 0, 3])
assert cross.norm() == 3
print("تایید شد")

```

## مثال ۳ — نقطه روی خط

### حل ریاضی

خط از \((1,0,2)\) با هادی \((1,1,0)\):

\[
\mathbf{r}(t)=(1,0,2)+t(1,1,0)
\]

در \(t=2\): \((3,2,2)\).

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t")
point = sp.Matrix([1, 0, 2]) + t*sp.Matrix([1, 1, 0])
print(point.subs(t, 2))
assert point.subs(t, 2) == sp.Matrix([3, 2, 2])
print("تایید شد")

```

## مثال ۴ — فاصلهٔ مبدأ تا صفحه

### حل ریاضی

صفحه \(2x-y+z=4\) را به شکل \(2x-y+z-4=0\) بنویسید.
فاصلهٔ مبدأ:

\[
\frac{|{-4}|}{\sqrt{4+1+1}}=\frac{4}{\sqrt{6}}
\]

### بازبینی با پایتون



```python
import sympy as sp
normal = sp.Matrix([2, -1, 1])
distance = sp.Abs(-4)/normal.norm()
print(sp.simplify(distance))
assert sp.simplify(distance - 4/sp.sqrt(6)) == 0
print("تایید شد")

```

## نکته

ضرب خارجی جابه‌جاپذیر نیست: \(\mathbf{a}\times\mathbf{b}=-\mathbf{b}\times\mathbf{a}\).
مساحت به علامت وابسته نیست، جهت صفحه وابسته است.

## جمع‌بندی

داخلی برای زاویه و تصویر.
خارجی برای عمود و مساحت.
خط یک پارامتر دارد و صفحه یک معادله.
