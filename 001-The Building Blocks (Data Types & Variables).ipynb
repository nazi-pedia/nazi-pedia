## 1. A name is a label

**See.** You do not put the number 25 into a box called `age`.
You attach the label `age` to an integer object whose value is 25.
A second assignment does not edit that integer. It moves the label.

**Name.** The statement `age = 25` is an assignment.
The right-hand side is evaluated first. The name on the left is then attached to the result.
`type(object)` reports the kind of object. `id` is the object's identity; you rarely need it,
but it makes the label story visible.


````python
age = 25
print("value:", age)
print("type:", type(age))
print("identity:", id(age))

previous = id(age)
age = 26
print("after rebinding, value:", age)
print("same object as before?", id(age) == previous)
````

**Read the output.** The value changed from 25 to 26, and the identity changed with it.
The integer 25 was not edited. The name `age` now labels a different object.

Several names may label one object. `x = y = 100` attaches both names to the same integer.


````python
x = y = 100
print(x, y)
print("same object?", x is y)

a, b, c = 10, 20, 30
print(a, b, c)

# Swap without a temporary name. The right-hand side is fully evaluated first.
left, right = 1, 2
left, right = right, left
print("swapped:", left, right)
assert (left, right) == (2, 1)
````

## 2. Names that Python will accept

**Name.** A name starts with a letter or underscore, then letters, digits, or underscores.
It is case-sensitive. `pi` and `Pi` are different names.
A name is not a string: quotes make text, not a label.

Constants that you promise not to rebind are written in `ALL_CAPS`. Python does not enforce that promise.
It is a signal to the next reader, including you.

The next cell tries an illegal name and catches the error, so the notebook keeps running.
That is the pattern this library uses whenever a failure is part of the lesson.


````python
user_name = "Ada"
_private = "internal"
userName = "works, but this library uses underscores"
MAX_STEPS = 30
pi = 3
Pi = 4
print(user_name, MAX_STEPS, "pi and Pi differ:", pi != Pi)

try:
    exec("user-name = 'Ada'")
except SyntaxError as err:
    print("rejected name:", err.msg)
````

**Misconception.** A hyphen is a legal character inside a name.

**Correction.** `user-name` is subtraction: `user` minus `name`. Use `user_name`.


## 3. One name, different kinds of object

**See.** Nothing in the name `value` tells you what it labels.
The object on the right decides the type. The next assignment may pick a different type.
This is dynamic typing. It is convenient, and it will hide mistakes until a later line crashes.
Print `type(...)` whenever a result surprises you.


````python
value = 42
print(value, type(value))

value = "Hello"
print(value, type(value))

value = True
print(value, type(value))

value = 3.5
print(value, type(value))
````

## 4. Integers and floats — the distinction calculus cares about

**See.** An integer is a whole number of unlimited size.
A float is a binary approximation of a real number, stored with about 15 significant decimals.

**Name.**
- `/` is true division and always returns a float
- `//` is floor division
- `%` is the remainder
- `**` is exponentiation
- parentheses, then exponent, then multiplication and division, then addition and subtraction

**The fact that will save you later.** `0.1 + 0.2` is not `0.3`.
Decimal 0.1 has no exact binary float. If a calculus answer is a clean fraction such as \(1/3\),
a float will smudge it. From lesson 004 on, SymPy keeps exact rationals until you ask for a decimal.


````python
a, b = 10, 3
print("sum", a + b)
print("difference", a - b)
print("product", a * b)
print("true division", a / b, type(a / b))
print("floor division", a // b)
print("remainder", a % b)
print("power", a ** b)
print("grouped power", (5 + 3) * 2 ** 2)

print("0.1 + 0.2 == 0.3 ?", 0.1 + 0.2 == 0.3)
print("actual sum", 0.1 + 0.2)

import math
print("close enough for a measurement?", math.isclose(0.1 + 0.2, 0.3))
````

**Read the output.** `10 / 3` is the float `3.333...`, not the integer 3.
`10 // 3` throws away the fractional part toward minus infinity, which for positive numbers is the same as truncating.
The equality `0.1 + 0.2 == 0.3` is false. `math.isclose` is the right test for measured numbers.
Exact mathematics in this library will not rely on that test; it will use SymPy.


### Own it

Predict the type and the value of `7 / 2`, `7 // 2`, and `int(7 / 2)`.


````python
print(7 / 2, type(7 / 2))
print(7 // 2, type(7 // 2))
print(int(7 / 2), type(int(7 / 2)))
assert 7 / 2 == 3.5
assert 7 // 2 == 3
assert int(7 / 2) == 3
print("checkpoint passed")
````

## 5. Strings are how a computation speaks

**See.** A string is text. A number is not text, even when its digits look like text.
`"25"` plus `"1"` is the text `"251"`. The integer 25 plus 1 is 26.

**Name.** Quotes, single or double, create a string. Triple quotes create a string that may span lines.
An f-string, written `f"..."`, evaluates expressions inside braces.
`{value:.2f}` prints a float with two digits after the decimal.
`{value:.2%}` prints a fraction as a percent.


````python
title = " Hello, Python "
print(repr(title))
print(repr(title.strip()))
print(title.strip().lower())
print(title.strip().upper())
print(title.strip().replace("Python", "calculus"))

words = title.strip().split()
print(words)
print("-".join(words))
print("length with spaces:", len(title))
print("starts with Hello after strip?", title.strip().startswith("Hello"))

measurement = 9.80665
print(f"g = {measurement:.3f} m/s^2")
ratio = 1 / 3
print(f"one third as a percent, wrongly rounded by the format: {ratio:.2%}")
````

**Misconception.** `len` is a string method, so you write `title.len()`.

**Correction.** `len` is a built-in function: `len(title)`. Methods such as `strip` and `split` use the dot.


## 6. Converting types, and refusing a bad conversion

**See.** `input()` — which this notebook does not call, because it would stop and wait —
always returns a string. A calculation needs a number. The conversion has to be explicit.

**Name.** `int(text)`, `float(text)`, and `str(number)` convert.
`int` applied to a float truncates toward zero. It does not round.
`int("3.14")` fails, because that string is not an integer literal.
A failed conversion raises `ValueError`. Catch it when the text comes from outside the program.


````python
age_text = "25"
age = int(age_text)
print(age, type(age), "next year", age + 1)

price_text = "19.99"
price = float(price_text)
print(price, type(price))

year = 2026
print(str(year), type(str(year)))

pi = 3.14159
print("truncated, not rounded:", int(pi))
print("rounded to nearest integer:", round(pi))
print("rounded to 2 decimals:", round(pi, 2))

for text in ("123", "12.5", "twelve"):
    try:
        print(text, "->", int(text))
    except ValueError:
        try:
            print(text, "-> float", float(text))
        except ValueError:
            print(text, "-> not a number")
````

### Own it

`int(3.9)` and `round(3.9)` do not agree. Which one is truncation, and which one is rounding?


````python
print("int(3.9) =", int(3.9))
print("round(3.9) =", round(3.9))
assert int(3.9) == 3
assert round(3.9) == 4
print("checkpoint passed")
````

## 7. Booleans are the answers to yes/no questions

**See.** A comparison does not change its operands. It produces a new object: `True` or `False`.
Lesson 002 will build decisions out of these objects. Here you only need to create them and read them.

**Name.** `==` tests equality. `=` assigns. `!=` tests inequality.
`<`, `>`, `<=`, `>=` compare order. Numbers of mixed `int` and `float` compare by value.


````python
a, b = 10, 20
print("a > b", a > b)
print("a < b", a < b)
print("a == b", a == b)
ten_int = 10
ten_float = 10.0
print("10 == 10.0", ten_int == ten_float)
print("10 is 10.0", ten_int is ten_float)

is_ready = True
is_blocked = False
print("ready and not blocked:", is_ready and not is_blocked)
````

**Read the output.** `10 == 10.0` is true: same numeric value.
`10 is 10.0` is false: an integer and a float are different objects.
Use `==` for values. Reserve `is` for the label story in section 1, and for `None`.


## 8. The math module, and a first scientific report

**See.** Python's arithmetic operators do not include square root, \(\pi\), or \(e\).
The standard library module `math` does. Import it by name. Do not import everything with `*`.

**Do.** The area of a circle of radius 5 is \(25\pi\). Compute it, print it to two decimals,
and check the exact relationship `area == math.pi * radius ** 2` with `math.isclose`.


````python
import math

radius = 5
PI_DISPLAY = math.pi
area = PI_DISPLAY * radius ** 2
circumference = 2 * PI_DISPLAY * radius

print(f"radius = {radius}")
print(f"area = {area:.2f}")
print(f"circumference = {circumference:.2f}")
print(f"sqrt(16) = {math.sqrt(16)}")
print(f"e = {math.e:.6f}")

assert math.isclose(area, 25 * math.pi)
assert math.isclose(math.sqrt(16), 4)
print("checkpoint passed")
````

## Workshop — a measurement card

Turn three raw numbers into a short report. This is the shape of every numerical result later:
a name, a value, a unit, and a stated precision.

- mass `0.453` kg, printed to 3 decimals
- length `2.5` m
- density = mass / length, which is not a physical density, just the quotient, printed to 4 decimals
- the quotient as an exact idea is \(0.453 / 2.5 = 0.1812\). Check it with `math.isclose`.


````python
import math

mass_kg = 0.453
length_m = 2.5
quotient = mass_kg / length_m

print(f"mass = {mass_kg:.3f} kg")
print(f"length = {length_m:.1f} m")
print(f"mass/length = {quotient:.4f} kg/m")
assert math.isclose(quotient, 0.1812)
print("workshop passed")
````

## Mistakes this lesson is meant to prevent

- Writing `=` when you meant to test equality with `==`.
- Expecting `10 / 2` to be the integer 5. It is the float `5.0`.
- Using `int` when you meant `round`.
- Comparing floats with `==` after arithmetic. Use `math.isclose`, or stay exact with SymPy.
- Treating `"25"` as a number. Concatenation and addition are different operations.
- Calling `input()` in the middle of a notebook you intend to re-run. Convert stored strings instead, until you deliberately build an interactive program.
