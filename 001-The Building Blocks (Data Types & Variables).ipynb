## 1.2 Variable Naming Rules

```python
# VALID variable names (these will work)
user_name = "Nazi"           # ✅ Underscore is fine
_user_name = "Nazi"          # ✅ Starting with underscore is fine
userName = "Nazi"            # ✅ CamelCase works but not recommended
user_name_2 = "Nazi"         # ✅ Numbers anywhere except first
USERNAME = "Nazi"            # ✅ All caps works but means constant

print(user_name)
print(_user_name)
print(userName)
print(user_name_2)
print(USERNAME)
```

```python
# INVALID - These will cause SyntaxError if uncommented
# 1user_name = "John"        # ❌ Cannot start with number
# user-name = "John"         # ❌ Hyphens are subtraction operators
# user name = "John"         # ❌ Spaces not allowed
# class = "Python"           # ❌ 'class' is a Python keyword
# @user = "John"             # ❌ Special characters not allowed
```

## 1.3 Dynamic Typing (Python's Superpower)

```python
# Now assign a string to the SAME variable
my_variable = "Hello, World!"
print(f"Value: {my_variable}, Type: {type(my_variable)}")
```

## 1.4 Multiple Assignment 

```python
# Instead of writing three lines...
# a = 10
# b = 20
# c = 30

# ...do this in one line!
a, b, c = 10, 20, 30
print(f"a={a}, b={b}, c={c}")
```

```python
# Or assign the same value to multiple variables
x = y = z = 100
print(f"x={x}, y={y}, z={z}")
```

# PART 2: THE CORE DATA TYPES

## 2.1 Strings: The Text Manipulators

### 2.1.1 Creating Strings (Quotes Matter!)

```python
# Single quotes
single_quote = 'Hello'
print(single_quote)
```

```python
# Double quotes (useful when your string contains apostrophes)
double_quote = "It's a beautiful day!"
print(double_quote)
```

```python
# Triple quotes (for multi-line strings)
multi_line = """This is a
multi-line string
that spans three lines!"""
print(multi_line)
```

### 2.1.2 String Operations

```python
# Let's start with a string
text = "  Hello, Python World!  "
```

```python
# 1. strip() - Remove whitespace from ends
print(f"After strip: '{text.strip()}'")
```

```python
# 2. lower() and upper() - Change case
print(f"Lowercase: '{text.lower()}'")
print(f"Uppercase: '{text.upper()}'")
```

```python
# 3. replace() - Replace text
print(f"Replace: '{text.replace('Python', 'Jupyter')}'")
```

```python
# 4. split() - Break into a list
words = text.strip().split()
print(f"Split into words: {words}")
```

```python
# 5. join() - Combine a list back into a string
combined = "-".join(words)
print(f"Joined with hyphens: {combined}")
```

```python
# 6. find() and count() - Search
print(f"Position of 'Python': {text.find('Python')}")
print(f"How many 'o's? : {text.count('o')}")
```

```python
# 7. len() - Get length (this is a built-in function, not a method!)
print(f"Length of text: {len(text)}")
```

```python
# 8. Check if string starts/ends with something
print(f"Starts with 'Hello'? {text.strip().startswith('Hello')}")
print(f"Ends with '!'? {text.strip().endswith('!')}")
```

## 2.2 Numbers: Integers and Floats

### 2.2.1 Integers (int) and Floats (float)

```python
# Integers (whole numbers)
age = 25
year = 2026
temperature = -10
```

```python
# Floats (decimal numbers)
price = 19.99
pi = 3.14159
weight = 75.5
```

```python
# Check the types
print(f"age is {type(age)}")
print(f"price is {type(price)}")
```

```python
# Basic arithmetic
a = 10
b = 3
```

```python
print(f"Addition: {a + b}")
print(f"Subtraction: {a - b}")
print(f"Multiplication: {a * b}")
print(f"Division: {a / b}")      # Always returns a float
print(f"Floor Division: {a // b}")  # Returns integer (rounds down)
print(f"Modulo (remainder): {a % b}")  # Returns remainder
print(f"Exponentiation: {a ** b}")     # 10^3 = 1000
```

### 2.2.2 Converting Between Types (Type Casting)

```python
# String to Integer
age_str = "25"
age_int = int(age_str)
print(f"Age as string: {age_str}, type: {type(age_str)}")
print(f"Age as integer: {age_int}, type: {type(age_int)}")
```

```python
# Float to Integer (truncates, doesn't round)
pi = 3.14159
pi_int = int(pi)
print(f"Pi as float: {pi}, as integer: {pi_int}")
```

```python
# Integer to String
year = 2026
year_str = str(year)
print(f"Year as integer: {year}, as string: '{year_str}'")
```

```python
# String to Float
price_str = "19.99"
price_float = float(price_str)
print(f"Price as string: {price_str}, as float: {price_float}")

# Dangerous! Can't convert invalid strings
# int("Hello")  # This would cause a ValueError!
```

```python
# Check if a string is a number before converting
text = "123"
if text.isdigit():
    number = int(text)
    print(f"Successfully converted {number}")
else:
    print("Not a valid number!")
```

### 2.2.3 Useful Number Operations

```python
import math
import random
```

```python
# Rounding
num = 3.14159
print(f"Round to nearest: {round(num)}")
print(f"Round to 2 decimals: {round(num, 2)}")
print(f"Ceiling (round up): {math.ceil(num)}")
print(f"Floor (round down): {math.floor(num)}")
print(f"Absolute value: {abs(-10)}")
```

```python
# Math functions
print(f"Square root of 16: {math.sqrt(16)}")
print(f"2^10: {math.pow(2, 10)}")
print(f"Pi: {math.pi}")
print(f"Euler's number: {math.e}")
```

```python
# Random numbers
print(f"Random integer between 1 and 10: {random.randint(1, 10)}")
print(f"Random float between 0 and 1: {random.random()}")
print(f"Random choice from list: {random.choice(['apple', 'banana', 'cherry'])}")
```

```python
# f-strings can format numbers
percentage = 0.8567
print(f"Percentage: {percentage:.2%}")  # 85.67%
```

## 2.3 Booleans: The Decision Makers

```python
# Direct assignment
is_student = True
is_employed = False
print(f"Student: {is_student}, Employed: {is_employed}")
```

```python
# Boolean from comparisons
a = 10
b = 20
print(f"a > b: {a > b}")    # False
print(f"a < b: {a < b}")    # True
print(f"a == b: {a == b}")  # False (equality check)
print(f"a != b: {a != b}")  # True (not equal)
print(f"a >= b: {a >= b}")  # False
print(f"a <= b: {a <= b}")  # True
```

```python
# Boolean operations (AND, OR, NOT)
is_weekend = True
has_homework = False
print(f"Can relax? {is_weekend and not has_homework}")  # True
```

```python
# Truthiness - Other values can be treated as booleans
print(f"Empty string: {bool('')}")      # False
print(f"Non-empty string: {bool('Hi')}")  # True
print(f"Zero: {bool(0)}")               # False
print(f"Non-zero: {bool(42)}")          # True
print(f"Empty list: {bool([])}")        # False
print(f"Non-empty list: {bool([1,2])}")  # True
print(f"None: {bool(None)}")            # False
```

```python
# Using bool() to check conditions
name = ""
if name:  # Equivalent to: if bool(name) == True
    print(f"Hello {name}")
else:
    print("Name is empty!")
```
