# The boundary is the set of ties

For $k = 1$ with two stored points, the prediction changes where the distances are equal. Let the points be $A$ and $B$, and let $P$ be a place in the plane. Set the squared distances equal:

$$
\|P - A\|^2 = \|P - B\|^2
$$

Expand both squares. The $\|P\|^2$ terms cancel.

$$
\|P\|^2 - 2P\cdot A + \|A\|^2 = \|P\|^2 - 2P\cdot B + \|B\|^2
$$

$$
2P\cdot(B - A) = \|B\|^2 - \|A\|^2
$$

The midpoint $M = (A + B)/2$ satisfies the same equation:

$$
M\cdot(B - A) = \frac{A+B}{2}\cdot(B-A) = \frac{\|B\|^2 - \|A\|^2}{2}
$$

So the tie set passes through the midpoint. The equation says $P\cdot(B-A)$ is constant, which means $B-A$ is perpendicular to the tie set. The boundary is the line through the midpoint, at a right angle to the segment $AB$.

Put A at $(0, 0)$, labeled bloom, and B at $(4, 0)$, labeled rest. The midpoint is $(2, 0)$. The segment is horizontal, so the boundary is the vertical line $x = 2$.

A place at $(1, 3)$ has squared distance $1 + 9 = 10$ to A and $9 + 9 = 18$ to B. It is on A's side of the line, and the distances agree.


```python
import matplotlib.pyplot as plt

# Quiet page. Busy is green, quiet is clay, and today is blue.
plt.rcParams.update({
    "figure.figsize": (7.2, 4.4),
    "figure.dpi": 120,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "axes.axisbelow": True,
    "grid.color": "#d9e2ec",
    "axes.edgecolor": "#334e68",
    "axes.labelcolor": "#243b53",
    "xtick.color": "#334e68",
    "ytick.color": "#334e68",
    "text.color": "#102a43",
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "font.size": 11,
    "savefig.facecolor": "white",
})
BLUE = "#1d4e89"
CLAY = "#c2410c"
GREEN = "#0f766e"
INK = "#102a43"
GRAY = "#9fb3c8"

# The tie line x = 2. The test place (1, 3) is on the bloom side.
fig, ax = plt.subplots(constrained_layout=True)
ax.axvline(2, color=BLUE, linestyle="--", label="equally far")
ax.scatter([0], [0], s=80, color=GREEN, zorder=3, label="A bloom")
ax.scatter([4], [0], s=80, color=CLAY, zorder=3, label="B rest")
ax.scatter([1], [3], s=80, color=INK, zorder=3, label="place (1, 3)")
ax.set_xlim(-1, 5)
ax.set_ylim(-1, 4)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("k = 1 changes its mind on the perpendicular bisector")
ax.legend(frameon=False)
assert (1 - 0) ** 2 + (3 - 0) ** 2 == 10
assert (1 - 4) ** 2 + (3 - 0) ** 2 == 18
```

## Today and its three nearest mornings

On the booking table, $k = 3$ reaches A, B, and C. The picture draws those three segments and leaves D, E, and F in the frame so the farther mornings stay visible. The vote on those three segments is busy, busy, quiet.


```python
import matplotlib.pyplot as plt

# Quiet page. Busy is green, quiet is clay, and today is blue.
plt.rcParams.update({
    "figure.figsize": (7.2, 4.4),
    "figure.dpi": 120,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "axes.axisbelow": True,
    "grid.color": "#d9e2ec",
    "axes.edgecolor": "#334e68",
    "axes.labelcolor": "#243b53",
    "xtick.color": "#334e68",
    "ytick.color": "#334e68",
    "text.color": "#102a43",
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "font.size": 11,
    "savefig.facecolor": "white",
})
BLUE = "#1d4e89"
CLAY = "#c2410c"
GREEN = "#0f766e"
INK = "#102a43"
GRAY = "#9fb3c8"

# Six stored mornings. The three shortest segments are the k = 3 vote.
points = {
    "A": (2, 5, GREEN),
    "B": (6, 2, GREEN),
    "C": (5, 6, CLAY),
    "D": (2, 8, CLAY),
    "E": (10, 2, CLAY),
    "F": (7, 14, GREEN),
}
fig, ax = plt.subplots(constrained_layout=True, figsize=(7.2, 5.4))
for name in ("A", "B", "C"):
    x, y, _ = points[name]
    ax.plot([2, x], [2, y], color=BLUE, linewidth=1.5, zorder=2)
for name, (x, y, color) in points.items():
    ax.scatter([x], [y], s=70, color=color, zorder=3)
    ax.annotate(name, (x, y), textcoords="offset points", xytext=(5, 4), color=INK)
ax.scatter([2], [2], s=90, facecolors="none", edgecolors=BLUE, linewidths=2, zorder=4)
ax.annotate("today", (2, 2), textcoords="offset points", xytext=(-42, -4), color=BLUE)
ax.set_xlabel("Bookings")
ax.set_ylabel("Walk-ins")
ax.set_title("The three nearest mornings vote busy, busy, quiet")
```

## A pitfall

With more than two stored points, $k = 1$ cuts the plane into one region per point, and the cuts are still pieces of perpendicular bisectors. They are not a single straight boundary through the whole page. The two-point picture is the local fact. The six-point picture is several of those facts competing.

## What to carry forward

The $k = 1$ tie set between two points is the perpendicular bisector. Between $(0,0)$ and $(4,0)$ it is $x = 2$. Today's three nearest mornings are A, B, and C.
