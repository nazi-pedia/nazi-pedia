# Straight-line distance

Two mornings $(x_1, y_1)$ and $(x_2, y_2)$ differ by $x_1 - x_2$ bookings and $y_1 - y_2$ walk-ins. The straight-line distance is the length of that step:

$$
d = \sqrt{(x_1 - x_2)^2 + (y_1 - y_2)^2}
$$

The square root is awkward in a ranking, and it is unnecessary there. For positive numbers $a$ and $b$,

$$
b^2 - a^2 = (b - a)(b + a)
$$

If $b > a$, then both factors on the right are positive, so $b^2 > a^2$. The converse is the same sentence read backwards. Sorting by $d$ and sorting by $d^2$ produce the same order. Compute the squares first.

Today is $(2, 2)$. Case A is $(2, 5)$.

$$
(2-2)^2 + (2-5)^2 = 0 + 9 = 9, \qquad d = 3
$$

The other five squares:

$$
\begin{aligned}
B,\ (6,2) &\colon (2-6)^2 + (2-2)^2 = 16 \\
C,\ (5,6) &\colon (2-5)^2 + (2-6)^2 = 9 + 16 = 25 \\
D,\ (2,8) &\colon (2-2)^2 + (2-8)^2 = 36 \\
E,\ (10,2) &\colon (2-10)^2 + (2-2)^2 = 64 \\
F,\ (7,14) &\colon (2-7)^2 + (2-14)^2 = 25 + 144 = 169
\end{aligned}
$$

The distances are $3, 4, 5, 6, 8, 13$. Every one is an integer, and every one is different. The order is A, B, C, D, E, F.


```python
# Squared distances rank the mornings. The squares are 9, 16, 25, 36, 64, 169.
from fractions import Fraction

query = (2, 2)
cases = [("A", 2, 5), ("B", 6, 2), ("C", 5, 6), ("D", 2, 8), ("E", 10, 2), ("F", 7, 14)]

def squared(case):
    return (query[0] - case[1]) ** 2 + (query[1] - case[2]) ** 2

squares = [squared(case) for case in cases]
assert squares == [9, 16, 25, 36, 64, 169]
distances = [int(square ** 0.5) for square in squares]
assert distances == [3, 4, 5, 6, 8, 13]
assert squares == sorted(squares)
# A smaller positive distance always has the smaller square.
assert Fraction(3) < Fraction(4) and 9 < 16
print("distances", distances)
```

## A pitfall

Adding the bookings gap to the walk-ins gap is a different measurement. From today to C the gaps are $3$ and $4$. Their sum is $7$. Their straight-line length is $5$, because $3^2 + 4^2 = 5^2$. A later lesson uses that sum on purpose. It is not this distance.

## What to carry forward

Rank by squared Euclidean distance. From $(2, 2)$ the order is A, B, C, D, E, F at distances $3, 4, 5, 6, 8, 13$.
