# Distance zero is a stored copy

Ask the procedure for case A, and leave A in the memory. The gaps are $(0, 0)$. The squared distance is $0$. $k = 1$ returns A and says busy. The label matches. Every unique stored row scores itself perfectly this way, and the score measures nothing about the other rows.

The weight $1/d$ cannot be formed. Division by zero is not a very sure neighbor. It is an undefined weight. If a distance is zero, stop and look at the row. Either the query is already stored, or two rows share a point.

Leave A out, as the previous lesson requires. The remaining squared distances from $(2, 5)$ are

$$
\begin{aligned}
B &\colon 4^2 + (-3)^2 = 25 \\
C &\colon 3^2 + 1^2 = 10 \\
D &\colon 0^2 + 3^2 = 9 \\
E &\colon 8^2 + (-3)^2 = 73 \\
F &\colon 5^2 + 9^2 = 106
\end{aligned}
$$

The smallest is $9$, case D, quiet. A's own label is busy. Leave-one-out at $k = 1$ gets A wrong. The perfect score from the zero distance was the opposite of this check.


```python
# Distance 0 reproduces A. Hiding A leaves D, at squared distance 9, labeled quiet.
from fractions import Fraction

memory = [("A", 2, 5, "busy"), ("B", 6, 2, "busy"), ("C", 5, 6, "quiet"),
          ("D", 2, 8, "quiet"), ("E", 10, 2, "quiet"), ("F", 7, 14, "busy")]
query = (2, 5)

def square(row):
    return (query[0] - row[1]) ** 2 + (query[1] - row[2]) ** 2

self_match = min(memory, key=square)
assert self_match[0] == "A"
assert square(self_match) == 0
try:
    Fraction(1, 0)
    raised = False
except ZeroDivisionError:
    raised = True
assert raised
hidden = [row for row in memory if row[0] != "A"]
nearest = min(hidden, key=square)
assert nearest[0] == "D"
assert square(nearest) == 9
assert nearest[3] == "quiet"
print("self", self_match[0], "left out", nearest[0], nearest[3])
```

## A pitfall

A dashboard that reports "k-NN training accuracy" without saying the rows were hidden is reporting this zero-distance trick. Ask whether each scored row was removed before the distances were computed.

## What to carry forward

A point stored at the query has distance $0$ and cannot be given weight $1/d$. Hiding A, the nearest remaining morning is D, quiet, so $k = 1$ misses A.
