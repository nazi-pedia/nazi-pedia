# ۴۳ · حل سری و دستگاه خطی

| | |
|---|---|
| **دوره** | معادلات دیفرانسیل |
| **پیش‌نیاز** | ۲۵ و ۳۹ |

حول یک نقطهٔ عادی، جواب را سری توانی بگیرید.

\[
y=\sum_{n=0}^{\infty}a_n x^n
\]

مشتق را جمله به جمله بگیرید، در معادله بگذارید، و توان‌های مشابه را برابر کنید.
\(a_0\) و \(a_1\) معمولاً آزاد می‌مانند؛ همان دو ثابت شرایط اولیه.

دستگاه \( \mathbf{x}'=A\mathbf{x} \) با مقدار ویژه حل می‌شود.
اگر \(\lambda\) مقدار ویژه و \(\mathbf{v}\) بردار ویژه باشد، \(e^{\lambda t}\mathbf{v}\) یک جواب است.



## مثال ۱ — سری که نمایی را پس می‌دهد

### حل ریاضی

\(y'=y\)، \(y(0)=1\).
\(y=\sum a_n x^n\)، \(a_0=1\).

\[
\sum_{n=1}^{\infty} n a_n x^{n-1}=\sum_{n=0}^{\infty}a_n x^n
\]

ضریب \(x^{n-1}\): \(n a_n=a_{n-1}\)، پس \(a_n=a_{n-1}/n\).

\[
a_n=\frac{1}{n!},\qquad y=\sum_{n=0}^{\infty}\frac{x^n}{n!}=e^{x}
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
y = sp.Function("y")
ode = sp.Eq(y(x).diff(x), y(x))
sol = sp.dsolve(ode, y(x), ics={y(0): 1})
series = sp.series(sp.exp(x), x, 0, 6).removeO()
print(sol)
print(series)
assert sp.simplify(sol.rhs - sp.exp(x)) == 0
assert series == 1 + x + x**2/2 + x**3/6 + x**4/24 + x**5/120
print("تایید شد")

```

## مثال ۲ — رابطهٔ بازگشتی برای \(y''-xy=0\)، شروع کار

### حل ریاضی

معادلهٔ ایری \(y''-xy=0\) نقطهٔ عادی در صفر دارد.
با \(y=\sum a_n x^n\)،

\[
\sum_{n=2}^{\infty}n(n-1)a_n x^{n-2}-\sum_{n=0}^{\infty}a_n x^{n+1}=0
\]

\(a_2=0\). برای توان‌های بالاتر، هر ضریب به سه پله قبل بسته است.
دو ضریب آزاد \(a_0\) و \(a_1\) دو جواب مستقل می‌سازند.
اینجا چهار جملهٔ جواب با \(a_0=1\) و \(a_1=0\) را می‌نویسیم.

\[
y=1+\frac{1}{6}x^3+\cdots
\]

چون \(y''(0)=0\) از خود معادله، و \(y'''(0)=1\) از مشتق معادله در صفر: مشتق \(y''=xy\) می‌دهد \(y'''=y+xy'\) و در صفر \(y'''(0)=1\).
پس جملهٔ \(x^3/6\) همان \(\dfrac{y'''(0)}{6}x^3\) است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
y = sp.Function("y")
ode = sp.Eq(y(x).diff(x, 2) - x*y(x), 0)
series = sp.dsolve(ode, y(x), hint="2nd_power_series_ordinary", n=6)
polynomial = series.rhs.removeO()
constants = sorted(polynomial.free_symbols - {x}, key=str)
print(series)
particular = polynomial.subs({constants[0]: 0, constants[1]: 1})
assert particular.subs(x, 0) == 1
assert sp.diff(particular, x).subs(x, 0) == 0
assert sp.diff(particular, x, 3).subs(x, 0) == 1
print(particular)
print("تایید شد")

```

## مثال ۳ — دستگاه و مقدار ویژه

### حل ریاضی

\[
x'=y,\qquad y'=-x
\]

با \(x(0)=1\) و \(y(0)=0\).
مشتق اولی: \(x''=y'=-x\). پس \(x''+x=0\).
\(x=\cos t\) و \(y=x'=-\sin t\).

ماتریس دستگاه \(\begin{pmatrix}0&1\\-1&0\end{pmatrix}\) مقدار ویژه‌های \(\pm i\) دارد،
و جواب حقیقی همان سینوس و کسینوس است.

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t")
x = sp.Function("x")
y = sp.Function("y")
system = [
    sp.Eq(x(t).diff(t), y(t)),
    sp.Eq(y(t).diff(t), -x(t)),
]
sol = sp.dsolve(system, [x(t), y(t)], ics={x(0): 1, y(0): 0})
print(sol)
matrix = sp.Matrix([[0, 1], [-1, 0]])
eigenvalues = matrix.eigenvals()
print(eigenvalues)
assert sp.simplify(sol[0].rhs - sp.cos(t)) == 0
assert sp.simplify(sol[1].rhs + sp.sin(t)) == 0
print("تایید شد")

```

## مثال ۴ — دو مقدار ویژهٔ حقیقی

### حل ریاضی

\[
\mathbf{u}'=\begin{pmatrix}1&0\\0&-2\end{pmatrix}\mathbf{u}
\]

مقدار ویژه‌ها ۱ و \(-2\) اند، با بردارهای \((1,0)\) و \((0,1)\).

\[
\mathbf{u}=c_1 e^{t}\begin{pmatrix}1\\0\end{pmatrix}+c_2 e^{-2t}\begin{pmatrix}0\\1\end{pmatrix}
\]

اگر \(\mathbf{u}(0)=(3,4)\)، آن‌گاه \(c_1=3\) و \(c_2=4\).

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t")
A = sp.Matrix([[1, 0], [0, -2]])
assert A.eigenvects()[0][0] in (1, -2)
u = sp.Matrix([3*sp.exp(t), 4*sp.exp(-2*t)])
assert sp.simplify(u.diff(t) - A*u) == sp.Matrix([0, 0])
assert u.subs(t, 0) == sp.Matrix([3, 4])
print(u)
print("تایید شد")

```

## نکته

سری را تا جمله‌ای ادامه دهید که رابطهٔ بازگشتی دیده شود.
چاپ شش جمله بدون رابطه، جواب نیست.
در دستگاه، هر مقدار ویژه یک مد جداست. شرط اولیه ترکیب مدها را تعیین می‌کند، نه خود مد را.

## جمع‌بندی

سری توانی، وقتی ضرایب معادله تحلیلی‌اند، دو جواب مستقل می‌سازد.
دستگاه خطی با ماتریس ثابت، به مقدار ویژه و بردار ویژه تقلیل پیدا می‌کند.
