# ۱۱ · خطی‌سازی، دیفرانسیل، رول و مقدار میانگین

| | |
|---|---|
| **دوره** | ریاضی عمومی ۱ |
| **پیش‌نیاز** | ۰۷ |

خط مماس، بهترین تقریب درجهٔ یک تابع در نزدیکی نقطهٔ تماس است.

\[
L(x)=f(a)+f'(a)(x-a)
\]

دیفرانسیل، همین ایده با نماد کوچک است: \(dy=f'(x)\,dx\).
تغییر واقعی \(\Delta y\) است. \(dy\) تقریب آن است.

**رول.** اگر \(f\) روی \([a,b]\) پیوسته، روی \((a,b)\) مشتق‌پذیر، و \(f(a)=f(b)\) باشد،
نقطه‌ای هست که \(f'(c)=0\).

**مقدار میانگین.** با همان پیوستگی و مشتق‌پذیری، نقطه‌ای هست که

\[
f'(c)=\frac{f(b)-f(a)}{b-a}
\]

شیب مماسی هست که با شیب قاطع دو سر برابر است.



## مثال ۱ — خطی‌سازی جذر

### حل ریاضی

\(f(x)=\sqrt{x}\) در \(a=9\). \(f(9)=3\) و \(f'(x)=\dfrac{1}{2\sqrt{x}}\)، پس \(f'(9)=\dfrac{1}{6}\).

\[
L(x)=3+\frac{1}{6}(x-9)
\]

\[
\sqrt{9.1}\approx 3+\frac{0.1}{6}=3+\frac{1}{60}=\frac{181}{60}
\]

تقریب کمی بزرگ‌تر از مقدار واقعی است، چون \(f''(x)<0\) و نمودار زیر مماس است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", positive=True)
f = sp.sqrt(x)
L = f.subs(x, 9) + sp.diff(f, x).subs(x, 9)*(x - 9)
approx = sp.simplify(L.subs(x, sp.Rational(91, 10)))
true = sp.sqrt(sp.Rational(91, 10))
print("خطی‌سازی:", approx)
print("مقدار واقعی:", true)
print("اختلاف:", sp.N(approx - true, 8))
assert approx == sp.Rational(181, 60)
assert approx > true
print("تایید شد")

```

## مثال ۲ — دیفرانسیل در برابر تغییر واقعی

### حل ریاضی

\(y=x^3\). از \(x=2\) به \(x=2.1\)، \(dx=0.1\).

\[
dy=3x^2\,dx=3\cdot 4\cdot 0.1=1.2
\]

تغییر واقعی:

\[
\Delta y=(2.1)^3-8=9.261-8=1.261
\]

اختلاف \(0.061\) از جمله‌های درجهٔ دو و سه در بسط می‌آید. دیفرانسیل آن‌ها را دور ریخته است.

### بازبینی با پایتون



```python
import sympy as sp
dx = sp.Rational(1, 10)
dy = 3*(2**2)*dx
delta = sp.Rational(21, 10)**3 - 8
print("dy =", dy)
print("Delta y =", delta)
assert dy == sp.Rational(6, 5)
assert delta == sp.Rational(1261, 1000)
print("تایید شد")

```

## مثال ۳ — رول

### حل ریاضی

\(f(x)=x^2-1\) روی \([-1,1]\).
\(f(-1)=f(1)=0\). تابع چندجمله‌ای است، پس شرط‌های رول برقرارند.

\[
f'(x)=2x=0\implies x=0
\]

و ۰ داخل بازه است. مماس آنجا افقی است، درست بین دو ریشه.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = x**2 - 1
assert f.subs(x, -1) == f.subs(x, 1) == 0
c = sp.solve(sp.diff(f, x), x)[0]
print("c =", c)
assert c == 0
print("تایید شد")

```

## مثال ۴ — مقدار میانگین

### حل ریاضی

\(f(x)=x^2\) روی \([0,2]\).

\[
\frac{f(2)-f(0)}{2-0}=\frac{4}{2}=2
\]

\[
f'(c)=2c=2\implies c=1
\]

\(c=1\) در \((0,2)\) است. شیب قاطع از \((0,0)\) به \((2,4)\) برابر ۲ است و مماس در \(x=1\) هم شیب ۲ دارد.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = x**2
secant = (f.subs(x, 2) - f.subs(x, 0))/2
c = sp.solve(sp.Eq(sp.diff(f, x), secant), x)[0]
print("شیب قاطع:", secant, " c:", c)
assert secant == 2 and c == 1
print("تایید شد")

```

## مثال ۵ — جایی که مقدار میانگین ساکت است

### حل ریاضی

\(f(x)=|x|\) روی \([-1,1]\).
\(f(-1)=f(1)=1\)، پس شیب قاطع ۰ است.
اما در \(x=0\) مشتق وجود ندارد و در هیچ نقطهٔ دیگر \(f'\) صفر نیست:
چپ \(-1\) و راست \(+1\).
قضیه وعده نمی‌دهد، چون مشتق‌پذیری روی بازهٔ باز شکسته است.
رول هم به همین دلیل ساکت است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", real=True)
f = sp.Abs(x)
print("مشتق چپ:", sp.limit((f.subs(x, 0 + sp.Symbol("h")) - 0)/sp.Symbol("h"), sp.Symbol("h"), 0, dir="-"))
left = sp.limit((sp.Abs(0 + x) - 0)/x, x, 0, dir="-")
right = sp.limit((sp.Abs(0 + x) - 0)/x, x, 0, dir="+")
print(left, right)
assert left == -1 and right == 1
print("دو مشتق یک‌طرفه برابر نیستند")
print("تایید شد")

```

## نکته

خطی‌سازی نزدیک نقطه خوب است و دور از نقطه دروغ می‌گوید.
برای \(\sqrt{16.1}\) نقطهٔ پایه را ۱۶ بگیرید، نه ۹.
