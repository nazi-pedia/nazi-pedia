# ۳۴ · انتگرال دوگانه و سه‌گانه

| | |
|---|---|
| **دوره** | ریاضی عمومی چندمتغیره |
| **پیش‌نیاز** | ۱۶ و ۱۹ |

انتگرال دوگانه، حجم زیر رویه یا جرم یک ورق است.
کران داخلی می‌تواند به متغیر بیرونی وابسته باشد. کران بیرونی باید عدد باشد.

\[
\iint_D f(x,y)\,dA
\]

انتگرال سه‌گانه همین ایده در حجم است.
حجم چهاروجهی \(x\ge 0,y\ge 0,z\ge 0,x+y+z\le 1\) برابر \(\dfrac{1}{6}\) است و این عدد، آزمون استاندارد ترتیب کران‌هاست.



## مثال ۱ — مستطیل

### حل ریاضی

\[
\int_0^1\int_0^1(x+y)\,dx\,dy
=\int_0^1\Bigl[\frac{x^2}{2}+yx\Bigr]_0^1 dy
=\int_0^1\left(\frac{1}{2}+y\right)dy
=\Bigl[\frac{y}{2}+\frac{y^2}{2}\Bigr]_0^1=1
\]

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y")
exact = sp.integrate(x + y, (x, 0, 1), (y, 0, 1))
print(exact)
assert exact == 1
print("تایید شد")

```

## مثال ۲ — مثلث، کران وابسته

### حل ریاضی

\(D\): \(0\le x\le 1\) و \(0\le y\le x\). انتگرال‌گیر ۱ است، پس جواب مساحت مثلث \(\dfrac{1}{2}\) است.

\[
\int_0^1\int_0^x 1\,dy\,dx=\int_0^1 x\,dx=\frac{1}{2}
\]

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y")
exact = sp.integrate(1, (y, 0, x), (x, 0, 1))
print(exact)
assert exact == sp.Rational(1, 2)
print("تایید شد")

```

## مثال ۳ — چهاروجهی

### حل ریاضی

\[
\int_0^1\int_0^{1-x}\int_0^{1-x-y} dz\,dy\,dx
=\int_0^1\int_0^{1-x}(1-x-y)\,dy\,dx
\]

داخلی: \(\bigl[(1-x)y-y^2/2\bigr]_0^{1-x}=\dfrac{(1-x)^2}{2}\).

\[
\int_0^1\frac{(1-x)^2}{2}\,dx=\frac{1}{2}\cdot\frac{1}{3}=\frac{1}{6}
\]

### بازبینی با پایتون



```python
import sympy as sp
x, y, z = sp.symbols("x y z")
exact = sp.integrate(1, (z, 0, 1 - x - y), (y, 0, 1 - x), (x, 0, 1))
print(exact)
assert exact == sp.Rational(1, 6)
print("تایید شد")

```

## مثال ۴ — جرم با چگالی

### حل ریاضی

ورق \(0\le x\le 1\)، \(0\le y\le 2\)، چگالی \(\rho=xy\).

\[
m=\int_0^2\int_0^1 xy\,dx\,dy
=\int_0^2 y\Bigl[\frac{x^2}{2}\Bigr]_0^1 dy
=\int_0^2 \frac{y}{2}\,dy=1
\]

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y")
mass = sp.integrate(x*y, (x, 0, 1), (y, 0, 2))
print(mass)
assert mass == 1
print("تایید شد")

```

## نکته

اگر انتگرال داخلی را نسبت به متغیری بگیرید که کرانش به همان متغیر وابسته است، ترتیب \(dx\,dy\) را با ترتیب نوشتن `integrate` یکی کنید.
در SymPy، اولین زوج \((x, a, b)\) داخلی‌ترین انتگرال است.
