# How far is a point from the line?

The score $\mathbf{w}\cdot\mathbf{x}+b$ still depends on how large you choose $\mathbf{w}$. If you double $\mathbf{w}$ and double $b$, the line $s = 0$ stays put, but every score doubles. A distance cannot behave like that.

The geometric distance from a point $\mathbf{x}_0$ to the line $\mathbf{w}\cdot\mathbf{x}+b = 0$ is

$$
\operatorname{distance}(\mathbf{x}_0)
= \frac{|\mathbf{w}\cdot\mathbf{x}_0+b|}{\|\mathbf{w}\|}
$$

This lesson proves that formula, then uses it on the line $x_1 = 1$.


## Proof

Let $\mathbf{u} = \mathbf{w}/\|\mathbf{w}\|$. This is a unit vector normal to the line. Drop a perpendicular from $\mathbf{x}_0$ to the line and call the foot $\mathbf{p}$. Then

$$
\mathbf{x}_0 = \mathbf{p} + t\mathbf{u}
$$

for some signed length $t$. The point $\mathbf{p}$ lies on the line, so $\mathbf{w}\cdot\mathbf{p}+b = 0$. Substitute:

$$
\begin{aligned}
\mathbf{w}\cdot\mathbf{x}_0 + b
&= \mathbf{w}\cdot(\mathbf{p}+t\mathbf{u}) + b \\
&= (\mathbf{w}\cdot\mathbf{p}+b) + t\,\mathbf{w}\cdot\mathbf{u} \\
&= t\|\mathbf{w}\|
\end{aligned}
$$

because $\mathbf{w}\cdot\mathbf{u} = \|\mathbf{w}\|$. Therefore

$$
t = \frac{\mathbf{w}\cdot\mathbf{x}_0+b}{\|\mathbf{w}\|},
\qquad
|t| = \frac{|\mathbf{w}\cdot\mathbf{x}_0+b|}{\|\mathbf{w}\|}
$$

The absolute value $|t|$ is the distance. The sign of $t$ says which side of the line the point is on.


## The same line, measured properly

Keep $\mathbf{w} = (1, 0)$ and $b = -1$. Then $\|\mathbf{w}\| = 1$, so the distance equals the absolute score.

$$
\begin{aligned}
(2, 0) &\mapsto |2-1|/1 = 1 \\
(0, 1) &\mapsto |0-1|/1 = 1 \\
(4, 0) &\mapsto |4-1|/1 = 3 \\
(1, 5) &\mapsto 0
\end{aligned}
$$

Now double the weights: $\mathbf{w} = (2, 0)$, $b = -2$. The zero set is still $x_1 = 1$. The scores double, but the distances do not:

$$
\frac{|2\cdot 2 - 2|}{\|(2,0)\|} = \frac{2}{2} = 1
$$

The point $(2, 0)$ is still one unit away. Scaling the weights is not the same thing as moving the line.


```python
# Check the distance formula before and after doubling (w, b).
import math

def distance(point, w, b):
    score = w[0] * point[0] + w[1] * point[1] + b
    norm = math.sqrt(w[0] ** 2 + w[1] ** 2)
    return abs(score) / norm

assert distance((2, 0), (1, 0), -1) == 1
assert distance((0, 1), (1, 0), -1) == 1
assert distance((4, 0), (1, 0), -1) == 3
assert distance((1, 5), (1, 0), -1) == 0
# Doubling the weights leaves the geometric distance unchanged.
assert distance((2, 0), (2, 0), -2) == 1
assert distance((4, 0), (2, 0), -2) == 3
print("distance of (4, 0) with unit weights", distance((4, 0), (1, 0), -1))
print("distance of (4, 0) with doubled weights", distance((4, 0), (2, 0), -2))
```

## Width of a street

Two parallel lines $\mathbf{w}\cdot\mathbf{x}+b = 1$ and $\mathbf{w}\cdot\mathbf{x}+b = -1$ form an empty street around the decision line. Take a point on the first line, so its score is $+1$. Its distance to the decision line $\mathbf{w}\cdot\mathbf{x}+b = 0$ is

$$
\frac{|1|}{\|\mathbf{w}\|} = \frac{1}{\|\mathbf{w}\|}
$$

The other edge is the same distance on the other side. The full street width is

$$
\frac{2}{\|\mathbf{w}\|}
$$

For $\mathbf{w} = (1, 0)$ the street runs from $x_1 = 0$ to $x_1 = 2$. Its width is $2$, which matches $2/\|\mathbf{w}\|$.


```python
# The street edges are the places where the score is +1 and -1.
# For w = (1, 0) and b = -1, those edges are x1 = 2 and x1 = 0.
edge_positive = 1 - (-1)  # solve x1 - 1 = +1
edge_negative = -1 - (-1)  # solve x1 - 1 = -1
width = edge_positive - edge_negative
assert edge_positive == 2
assert edge_negative == 0
assert width == 2
print("positive edge", edge_positive)
print("negative edge", edge_negative)
print("street width", width)
```

## A pitfall

$| \mathbf{w}\cdot\mathbf{x}+b |$ is not a distance until you divide by $\|\mathbf{w}\|$. Forgetting the division makes a doubled weight vector look twice as far from the data, even though the line has not moved.

## What to carry forward

Distance is $|\mathbf{w}\cdot\mathbf{x}+b|/\|\mathbf{w}\|$. The street between the scores $+1$ and $-1$ has width $2/\|\mathbf{w}\|$. A shorter weight vector means a wider street.
