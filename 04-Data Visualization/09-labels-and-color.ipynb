# Say what the marks mean

A chart without units is a sketch. The title, the axis labels, and the colors each have one job.

- The title is the sentence, or the question. It is not the chart type.
- Each axis says the quantity and the unit.
- Color repeats a meaning you already stated in a legend or a direct label.
- A legend is a key, not a second title.

Ada and Grace again: $4, 5, 5, 6, 7$ and $6, 6, 7, 7, 8$. The sentence is "Grace stayed ahead of Ada all week." The y-axis is "Jobs finished", not "Value".


```python
import matplotlib.pyplot as plt

# Quiet page: data ink first, grid behind the marks, colors that stay distinct.
plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "figure.dpi": 120,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "axes.axisbelow": True,
    "grid.color": "#d9e2ec",
    "axes.edgecolor": "#334e68",
    "axes.labelcolor": "#243b53",
    "xtick.color": "#334e68",
    "ytick.color": "#334e68",
    "text.color": "#102a43",
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "font.size": 11,
    "savefig.facecolor": "white",
})
BLUE = "#1d4e89"
CLAY = "#c2410c"
GREEN = "#0f766e"
INK = "#102a43"
GRAY = "#9fb3c8"
plt.rcParams["axes.prop_cycle"] = plt.cycler(color=[BLUE, CLAY, GREEN])

# Title is the finding. The legend only says which line is which person.
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
ada = [4, 5, 5, 6, 7]
grace = [6, 6, 7, 7, 8]
fig, ax = plt.subplots(constrained_layout=True)
ax.plot(days, ada, color=BLUE, marker="o", linewidth=2, label="Ada")
ax.plot(days, grace, color=CLAY, marker="o", linewidth=2, label="Grace")
ax.set_ylabel("Jobs finished")
ax.set_xlabel("Day")
ax.set_ylim(0, 12)
ax.legend(frameon=False, title=None)
ax.set_title("Grace stayed ahead of Ada all week")
assert ax.get_title() == "Grace stayed ahead of Ada all week"
assert ax.get_ylabel() == "Jobs finished"
```

## A label at the line can replace the legend

With two lines, the name can sit at the last point. The reader does not have to look away to a box. Leave a little room on the right so the name is not clipped. Ada ends at $7$, Grace at $8$.


```python
import matplotlib.pyplot as plt

# Quiet page: data ink first, grid behind the marks, colors that stay distinct.
plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "figure.dpi": 120,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "axes.axisbelow": True,
    "grid.color": "#d9e2ec",
    "axes.edgecolor": "#334e68",
    "axes.labelcolor": "#243b53",
    "xtick.color": "#334e68",
    "ytick.color": "#334e68",
    "text.color": "#102a43",
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "font.size": 11,
    "savefig.facecolor": "white",
})
BLUE = "#1d4e89"
CLAY = "#c2410c"
GREEN = "#0f766e"
INK = "#102a43"
GRAY = "#9fb3c8"
plt.rcParams["axes.prop_cycle"] = plt.cycler(color=[BLUE, CLAY, GREEN])

# Direct labels at Friday. No legend box, because the names touch their lines.
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
ada = [4, 5, 5, 6, 7]
grace = [6, 6, 7, 7, 8]
fig, ax = plt.subplots(constrained_layout=True)
ax.plot(days, ada, color=BLUE, marker="o", linewidth=2)
ax.plot(days, grace, color=CLAY, marker="o", linewidth=2)
ax.text(4.12, 7, "Ada", color=BLUE, va="center")
ax.text(4.12, 8, "Grace", color=CLAY, va="center")
ax.set_xlim(-0.2, 4.9)
ax.set_ylim(0, 12)
ax.set_ylabel("Jobs finished")
ax.set_xlabel("Day")
ax.set_title("Grace stayed ahead of Ada all week")
assert len(ax.get_legend_handles_labels()[0]) == 0
```

## Color that points at one bar

If the sentence is about Grace, Grace gets the working color and the other bars recede to gray. The color is the emphasis, not decoration. The lengths are unchanged: $20, 27, 29, 34$.


```python
import matplotlib.pyplot as plt

# Quiet page: data ink first, grid behind the marks, colors that stay distinct.
plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "figure.dpi": 120,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "axes.axisbelow": True,
    "grid.color": "#d9e2ec",
    "axes.edgecolor": "#334e68",
    "axes.labelcolor": "#243b53",
    "xtick.color": "#334e68",
    "ytick.color": "#334e68",
    "text.color": "#102a43",
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "font.size": 11,
    "savefig.facecolor": "white",
})
BLUE = "#1d4e89"
CLAY = "#c2410c"
GREEN = "#0f766e"
INK = "#102a43"
GRAY = "#9fb3c8"
plt.rcParams["axes.prop_cycle"] = plt.cycler(color=[BLUE, CLAY, GREEN])

# One clay bar. The other three stay gray so the eye lands on Grace.
names = ["Alan", "Ada", "Linus", "Grace"]
totals = [20, 27, 29, 34]
colors = [GRAY, GRAY, GRAY, CLAY]
fig, ax = plt.subplots(constrained_layout=True)
ax.barh(names, totals, color=colors)
ax.set_xlabel("Jobs finished")
ax.set_xlim(0, 44)
ax.set_title("Grace finished the most")
for row, total in enumerate(totals):
    ax.text(total + 0.5, row, str(total), va="center", color=INK)
assert colors[-1] == CLAY
assert colors.count(GRAY) == 3
```

## A pitfall

"Bar chart of weekly jobs" names the tool. It does not tell a reader what you found. Blue and orange are used here because they still differ if a reader has trouble separating red from green. A color that is the only carrier of a category, with no label, fails as soon as the page is printed in gray or the legend is cropped.

## What to carry forward

Title the finding, label the unit, and make color repeat a labeled meaning. Grace's line stays above Ada's, and Grace's bar is the one bar in clay.
