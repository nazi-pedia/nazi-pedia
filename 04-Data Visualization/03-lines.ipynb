# Change over time

A line chart needs an x-axis that has a real order. Days have that order. The line says what happened between one day and the next.

Ada and Grace, jobs per day:

$$
\begin{aligned}
\text{Ada} &= 4, 5, 5, 6, 7 \\
\text{Grace} &= 6, 6, 7, 7, 8
\end{aligned}
$$

Grace's lead each day:

$$
\begin{aligned}
6-4 &= 2 \\
6-5 &= 1 \\
7-5 &= 2 \\
7-6 &= 1 \\
8-7 &= 1
\end{aligned}
$$

Grace finished more than Ada every day. The lead is $1$ or $2$ jobs, and it is $1$ on Friday. Both rows rise from Monday to Friday: Ada by $7-4 = 3$, Grace by $8-6 = 2$.


```python
import matplotlib.pyplot as plt

# Quiet page: data ink first, grid behind the marks, colors that stay distinct.
plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "figure.dpi": 120,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "axes.axisbelow": True,
    "grid.color": "#d9e2ec",
    "axes.edgecolor": "#334e68",
    "axes.labelcolor": "#243b53",
    "xtick.color": "#334e68",
    "ytick.color": "#334e68",
    "text.color": "#102a43",
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "font.size": 11,
    "savefig.facecolor": "white",
})
BLUE = "#1d4e89"
CLAY = "#c2410c"
GREEN = "#0f766e"
INK = "#102a43"
GRAY = "#9fb3c8"
plt.rcParams["axes.prop_cycle"] = plt.cycler(color=[BLUE, CLAY, GREEN])

# Two ordered series. The gap is computed before the picture is trusted.
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
ada = [4, 5, 5, 6, 7]
grace = [6, 6, 7, 7, 8]
gaps = [g - a for a, g in zip(ada, grace)]
assert gaps == [2, 1, 2, 1, 1]
assert ada[-1] - ada[0] == 3
assert grace[-1] - grace[0] == 2
fig, ax = plt.subplots(constrained_layout=True)
ax.plot(days, ada, color=BLUE, marker="o", linewidth=2, label="Ada")
ax.plot(days, grace, color=CLAY, marker="o", linewidth=2, label="Grace")
ax.set_ylabel("Jobs finished")
ax.set_ylim(0, 12)
ax.legend(frameon=False)
ax.set_title("Grace stayed ahead every day")
print("daily lead", gaps)
```

## Read the bend, not only the endpoints

The lab's daily totals rise by $2$ every single day, so that line has no bend. Ada's row does. From Monday to Tuesday she gains $1$. From Tuesday to Wednesday she gains $0$. Then she gains $1$ on each of the last two days.

$$
5-4 = 1, \qquad 5-5 = 0, \qquad 6-5 = 1, \qquad 7-6 = 1
$$

A flat segment is a result. It says Wednesday matched Tuesday. It is not a missing point.


```python
import matplotlib.pyplot as plt

# Quiet page: data ink first, grid behind the marks, colors that stay distinct.
plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "figure.dpi": 120,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "axes.axisbelow": True,
    "grid.color": "#d9e2ec",
    "axes.edgecolor": "#334e68",
    "axes.labelcolor": "#243b53",
    "xtick.color": "#334e68",
    "ytick.color": "#334e68",
    "text.color": "#102a43",
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "font.size": 11,
    "savefig.facecolor": "white",
})
BLUE = "#1d4e89"
CLAY = "#c2410c"
GREEN = "#0f766e"
INK = "#102a43"
GRAY = "#9fb3c8"
plt.rcParams["axes.prop_cycle"] = plt.cycler(color=[BLUE, CLAY, GREEN])

# Mark the flat step. Wednesday minus Tuesday is 0 for Ada.
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
ada = [4, 5, 5, 6, 7]
steps = [ada[i + 1] - ada[i] for i in range(4)]
assert steps == [1, 0, 1, 1]
fig, ax = plt.subplots(constrained_layout=True)
ax.plot(days, ada, color=BLUE, marker="o", linewidth=2)
ax.annotate(
    "No gain",
    xy=(2, 5),
    xytext=(2.6, 3.2),
    arrowprops={"arrowstyle": "->", "color": INK},
    color=INK,
)
ax.set_ylabel("Jobs finished")
ax.set_ylim(0, 10)
ax.set_title("Ada's Wednesday matched her Tuesday")
```

## Markers without a line

If you only measured three days, and you do not want to imply that you know Thursday, plot the points you have and leave the line off. The order is still time. The missing day stays missing.

Monday $4$, Wednesday $5$, Friday $7$. You do not have Tuesday or Thursday, so you do not draw them.


```python
import matplotlib.pyplot as plt

# Quiet page: data ink first, grid behind the marks, colors that stay distinct.
plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "figure.dpi": 120,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "axes.axisbelow": True,
    "grid.color": "#d9e2ec",
    "axes.edgecolor": "#334e68",
    "axes.labelcolor": "#243b53",
    "xtick.color": "#334e68",
    "ytick.color": "#334e68",
    "text.color": "#102a43",
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "font.size": 11,
    "savefig.facecolor": "white",
})
BLUE = "#1d4e89"
CLAY = "#c2410c"
GREEN = "#0f766e"
INK = "#102a43"
GRAY = "#9fb3c8"
plt.rcParams["axes.prop_cycle"] = plt.cycler(color=[BLUE, CLAY, GREEN])

# Measured days only. No line, because the gaps were not measured.
fig, ax = plt.subplots(constrained_layout=True)
ax.plot(["Mon", "Wed", "Fri"], [4, 5, 7], color=BLUE, marker="o", linestyle="none", markersize=8)
ax.set_ylabel("Jobs finished")
ax.set_ylim(0, 10)
ax.set_title("Ada, on the three days that were recorded")
assert ax.lines[0].get_linestyle() == "None"
```

## A pitfall

A joined line claims that the quantity exists between the marks and that the x-axis order is the order in which it changed. That claim is fair for a week. It is not fair for a list of people, and it is not fair for a day you did not record.

## What to carry forward

Use a line for an ordered sequence you actually observed. Read the steps, including a step of $0$. Grace led Ada by $2, 1, 2, 1, 1$ jobs.
