# Mark the point that matters

An annotation is a sentence aimed at one coordinate. Use it when the reader would otherwise have to hunt.

Six temperature readings, in degrees C:

$$
16, 18, 17, 21, 20, 23
$$

The maximum is $23$, on day $6$. The step into that maximum is

$$
23 - 20 = 3
$$

A specification says the reactor should stay at or below $22\,\text{C}$. The readings above $22$ are exactly one of them: day $6$. A dashed horizontal line at $22$ makes that count visible, and the arrow names the point.


```python
import matplotlib.pyplot as plt

# Quiet page: data ink first, grid behind the marks, colors that stay distinct.
plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "figure.dpi": 120,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "axes.axisbelow": True,
    "grid.color": "#d9e2ec",
    "axes.edgecolor": "#334e68",
    "axes.labelcolor": "#243b53",
    "xtick.color": "#334e68",
    "ytick.color": "#334e68",
    "text.color": "#102a43",
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "font.size": 11,
    "savefig.facecolor": "white",
})
BLUE = "#1d4e89"
CLAY = "#c2410c"
GREEN = "#0f766e"
INK = "#102a43"
GRAY = "#9fb3c8"
plt.rcParams["axes.prop_cycle"] = plt.cycler(color=[BLUE, CLAY, GREEN])

# One arrow, one reference line. The only reading above 22 C is day 6.
days = [1, 2, 3, 4, 5, 6]
temperature = [16, 18, 17, 21, 20, 23]
above = [day for day, value in zip(days, temperature) if value > 22]
assert temperature.index(max(temperature)) == 5
assert max(temperature) - temperature[-2] == 3
assert above == [6]
fig, ax = plt.subplots(constrained_layout=True)
ax.plot(days, temperature, color=BLUE, marker="o", linewidth=2)
ax.axhline(22, color=GREEN, linestyle="--", linewidth=1.4)
ax.annotate(
    "23 C, the only reading above 22",
    xy=(6, 23),
    xytext=(2.1, 24.2),
    arrowprops={"arrowstyle": "->", "color": INK},
    color=INK,
)
ax.set_xlabel("Day")
ax.set_ylabel("Temperature (C)")
ax.set_xlim(0.7, 6.4)
ax.set_ylim(14, 26)
ax.set_title("Day 6 is 3 degrees above day 5, and past the spec")
```

## A label for the largest cell

In the North Lab table the largest cell is Grace on Friday, with $8$ jobs. The row sums are still $27, 34, 20, 29$. The annotation belongs on the cell, not on the weekly total, because the question here is "which morning was busiest", not "who won the week".

Friday's four values are $7, 8, 4, 7$. Their max is $8$, and it occurs once.


```python
import matplotlib.pyplot as plt

# Quiet page: data ink first, grid behind the marks, colors that stay distinct.
plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "figure.dpi": 120,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "axes.axisbelow": True,
    "grid.color": "#d9e2ec",
    "axes.edgecolor": "#334e68",
    "axes.labelcolor": "#243b53",
    "xtick.color": "#334e68",
    "ytick.color": "#334e68",
    "text.color": "#102a43",
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "font.size": 11,
    "savefig.facecolor": "white",
})
BLUE = "#1d4e89"
CLAY = "#c2410c"
GREEN = "#0f766e"
INK = "#102a43"
GRAY = "#9fb3c8"
plt.rcParams["axes.prop_cycle"] = plt.cycler(color=[BLUE, CLAY, GREEN])

# Point at Friday on Grace's line. That cell is 8, the unique maximum.
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
grace = [6, 6, 7, 7, 8]
friday = [7, 8, 4, 7]
assert max(friday) == 8 and friday.count(8) == 1
fig, ax = plt.subplots(constrained_layout=True)
ax.plot(days, grace, color=CLAY, marker="o", linewidth=2)
ax.annotate(
    "Busiest cell, 8 jobs",
    xy=(4, 8),
    xytext=(1.5, 9.3),
    arrowprops={"arrowstyle": "->", "color": INK},
    color=INK,
)
ax.set_ylabel("Jobs finished")
ax.set_ylim(4, 11)
ax.set_title("Grace on Friday is the busiest cell in the table")
```

## A pitfall

An arrow that covers the point, or a note on every point, turns the chart back into a table with extra clutter. Annotate the point the sentence needs. Park the text in empty space, and check that the arrow still touches the right coordinate after you resize the page.

## What to carry forward

Put a reference line where a rule lives, and an arrow where the data breaks that rule. Day $6$ is the only temperature above $22\,\text{C}$. Grace's Friday cell, $8$, is the unique busiest cell.
