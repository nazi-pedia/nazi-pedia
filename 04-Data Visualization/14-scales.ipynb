# Scales that change the story

The data can be exact and the picture can still mislead. The usual cause is a scale that does not match the claim.

Three purity readings, in percent:

$$
98, \quad 99, \quad 100
$$

The gaps are $1$ and $1$ percentage points. The largest is

$$
100/98 = 50/49
$$

times the smallest, about $2\%$ higher, not twice as high. An axis that starts at $0$ shows three similar bars. An axis that starts at $97$ turns a $1$-point gap into most of the picture.


```python
import matplotlib.pyplot as plt

# Quiet page: data ink first, grid behind the marks, colors that stay distinct.
plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "figure.dpi": 120,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "axes.axisbelow": True,
    "grid.color": "#d9e2ec",
    "axes.edgecolor": "#334e68",
    "axes.labelcolor": "#243b53",
    "xtick.color": "#334e68",
    "ytick.color": "#334e68",
    "text.color": "#102a43",
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "font.size": 11,
    "savefig.facecolor": "white",
})
BLUE = "#1d4e89"
CLAY = "#c2410c"
GREEN = "#0f766e"
INK = "#102a43"
GRAY = "#9fb3c8"
plt.rcParams["axes.prop_cycle"] = plt.cycler(color=[BLUE, CLAY, GREEN])

# Same three purities. Only the lower limit of the axis changes.
names = ["Run A", "Run B", "Run C"]
purity = [98, 99, 100]
assert purity[1] - purity[0] == 1
assert purity[2] / purity[0] == 100 / 98
fig, panels = plt.subplots(1, 2, constrained_layout=True, figsize=(8.4, 3.8))
panels[0].bar(names, purity, color=BLUE)
panels[0].set_ylim(0, 110)
panels[0].set_ylabel("Purity (%)")
panels[0].set_title("From zero, the runs nearly match")
panels[1].bar(names, purity, color=CLAY)
panels[1].set_ylim(97, 101)
panels[1].set_title("From 97, a 1-point gap fills the panel")
```

## A pie hides the comparison

Three tasks took $50$, $30$, and $20$ hours. The shares are $1/2$, $3/10$, and $1/5$. A pie asks the eye to compare angles. A bar asks it to compare lengths against a common zero, which is the comparison you can check with a ruler and with the fractions above. The hours add to $100$, so the percents and the hours are the same numbers. That convenience does not make the pie easier to read.


```python
import matplotlib.pyplot as plt

# Quiet page: data ink first, grid behind the marks, colors that stay distinct.
plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "figure.dpi": 120,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "axes.axisbelow": True,
    "grid.color": "#d9e2ec",
    "axes.edgecolor": "#334e68",
    "axes.labelcolor": "#243b53",
    "xtick.color": "#334e68",
    "ytick.color": "#334e68",
    "text.color": "#102a43",
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "font.size": 11,
    "savefig.facecolor": "white",
})
BLUE = "#1d4e89"
CLAY = "#c2410c"
GREEN = "#0f766e"
INK = "#102a43"
GRAY = "#9fb3c8"
plt.rcParams["axes.prop_cycle"] = plt.cycler(color=[BLUE, CLAY, GREEN])

# Prefer the bars. The pie is here so the harder comparison is visible.
labels = ["Setup", "Run", "Report"]
hours = [50, 30, 20]
assert sum(hours) == 100
fig, panels = plt.subplots(1, 2, constrained_layout=True, figsize=(8.4, 3.8))
panels[0].pie(hours, labels=labels, colors=[BLUE, CLAY, GREEN], startangle=90)
panels[0].set_title("Angles are hard to compare")
panels[1].barh(labels, hours, color=BLUE)
panels[1].set_xlabel("Hours")
panels[1].set_xlim(0, 70)
panels[1].set_title("Lengths from zero are not")
for row, value in enumerate(hours):
    panels[1].text(value + 1, row, str(value), va="center", color=INK)
```

## Two units on one panel

Daily jobs $18, 20, 22, 24, 26$ and the temperatures $16, 18, 17, 21, 20$ do not share a unit. A second vertical axis can be scaled until the two lines look like companions. They are not. From day $2$ to day $3$ the jobs rise by $2$ and the temperature falls by $1$:

$$
22-20 = 2, \qquad 17-18 = -1
$$

Two panels, already used for small multiples, keep that disagreement visible. A twin axis invites a sentence the differences do not support.


```python
import matplotlib.pyplot as plt

# Quiet page: data ink first, grid behind the marks, colors that stay distinct.
plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "figure.dpi": 120,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "axes.axisbelow": True,
    "grid.color": "#d9e2ec",
    "axes.edgecolor": "#334e68",
    "axes.labelcolor": "#243b53",
    "xtick.color": "#334e68",
    "ytick.color": "#334e68",
    "text.color": "#102a43",
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "font.size": 11,
    "savefig.facecolor": "white",
})
BLUE = "#1d4e89"
CLAY = "#c2410c"
GREEN = "#0f766e"
INK = "#102a43"
GRAY = "#9fb3c8"
plt.rcParams["axes.prop_cycle"] = plt.cycler(color=[BLUE, CLAY, GREEN])

# A twin axis can make unrelated units look like one motion.
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
jobs = [18, 20, 22, 24, 26]
temperature = [16, 18, 17, 21, 20]
assert jobs[2] - jobs[1] == 2
assert temperature[2] - temperature[1] == -1
fig, ax = plt.subplots(constrained_layout=True)
ax.plot(days, jobs, color=BLUE, marker="o", linewidth=2, label="Jobs")
ax.set_ylabel("Jobs finished", color=BLUE)
other = ax.twinx()
other.plot(days, temperature, color=CLAY, marker="o", linewidth=2, label="Temperature")
other.set_ylabel("Temperature (C)", color=CLAY)
other.grid(False)
ax.set_title("The lines look related. From Tue to Wed they moved apart.")
```

## A pitfall

If a difference of one percentage point is the decision, a zoomed axis can be honest, but only with the zoom written in the title and the full scale available beside it. A zoom that is silent is a different claim. Dual axes need an even stronger reason, and this lab week does not have one: use two panels.

## What to carry forward

Bars start at zero when length means amount. Pies ask for an angle comparison you rarely need. A second y-axis does not put jobs and degrees in the same unit. From Tuesday to Wednesday, jobs rose by $2$ and temperature fell by $1$.
