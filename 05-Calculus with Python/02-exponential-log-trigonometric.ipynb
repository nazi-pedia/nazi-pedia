## Definitions

For $a>0$ and $a\neq 1$,

$$
y=\log_a x \iff a^y=x,\qquad x>0.
$$

The natural logarithm is the logarithm in base $e$, written $\ln$.
In Python and in SymPy the name of this function is `log`, and its default base is $e$.

On the unit circle, $\cos\theta$ is the $x$-coordinate and $\sin\theta$ is the $y$-coordinate.
In calculus an angle is always in radians: $\pi$ radians means $180^\circ$.

$$
\sin^2\theta+\cos^2\theta=1.
$$

$$
\arcsin u=\theta \iff \sin\theta=u,\quad \theta\in\left[-\dfrac{\pi}{2},\dfrac{\pi}{2}\right].
$$


```python
import sympy as sp

# Compare a computed SymPy result with the answer found by hand.
def check(found, hand, label="Result"):
    # Simplify each side first so equivalent algebraic forms can match.
    found_s = sp.simplify(found)
    hand_s = sp.simplify(hand)
    difference = sp.simplify(found_s - hand_s)
    print(label)
    print("  Hand:", hand_s)
    print("  Python:", found_s)
    print("  Difference:", difference)
    # The methods agree only when that difference simplifies to zero.
    assert difference == 0
    print("  Confirmed")
```

## Example 1 — An exponential equation

### Hand solution

$$
2^{x+1}=8.
$$

Since $8=2^3$, the equation is $2^{x+1}=2^3$. The base is greater than $1$, so the exponential is one-to-one:

$$
\begin{aligned}
x+1 &= 3 \\
x &= 2.
\end{aligned}
$$

Check: $2^{3}=8$.

### Python recheck


```python
import sympy as sp

# Solve 2^(x + 1) = 8. Hand root: x = 2.
x = sp.symbols("x")
roots = sp.solve(sp.Eq(2**(x + 1), 8), x)
print("Root:", roots)
# The only solution is x = 2.
assert roots == [2]
# Check the hand substitution: 2^(2 + 1) = 8.
assert 2**(2 + 1) == 8
print("Confirmed")
```

## Example 2 — Logarithm laws, not a calculator

### Hand solution

$$
\begin{aligned}
\ln(e^5\cdot e^{-2})
&= \ln(e^5)+\ln(e^{-2}) \\
&= 5-2 \\
&= 3,
\end{aligned}
$$

and

$$
\begin{aligned}
\log_2 32
&= \log_2(2^5) \\
&= 5.
\end{aligned}
$$

A logarithm brings the exponent down. The result is an exact integer, not a decimal approximation.

### Python recheck


```python
import sympy as sp

# Hand answers: ln(e^5 * e^(-2)) = 3 and log base 2 of 32 equals 5.
exact_ln = sp.simplify(sp.log(sp.E**5 * sp.E**(-2)))
exact_log2 = sp.log(32, 2)
print("ln:", exact_ln)
print("log2 32:", exact_log2)
assert exact_ln == 3
assert exact_log2 == 5
print("Confirmed")
```

## Example 3 — Exact trigonometric values

### Hand solution

In an equilateral triangle of side $2$, the altitude is $\sqrt{3}$.
Half of that triangle is a $30^\circ\!-\!60^\circ\!-\!90^\circ$ triangle:

$$
\begin{aligned}
\sin\dfrac{\pi}{6}&=\dfrac{1}{2}, &
\cos\dfrac{\pi}{6}&=\dfrac{\sqrt{3}}{2}, &
\tan\dfrac{\pi}{3}&=\sqrt{3}, \\
\sin\dfrac{\pi}{3}&=\dfrac{\sqrt{3}}{2}, &
\cos\dfrac{\pi}{3}&=\dfrac{1}{2}.
\end{aligned}
$$

### Python recheck


```python
import sympy as sp

# Exact values from the 30-60-90 triangle.
# sin(pi/6) = 1/2, cos(pi/6) = sqrt(3)/2,
# sin(pi/3) = sqrt(3)/2, cos(pi/3) = 1/2, tan(pi/3) = sqrt(3).
assert sp.sin(sp.pi/6) == sp.Rational(1, 2)
assert sp.cos(sp.pi/6) == sp.sqrt(3)/2
assert sp.sin(sp.pi/3) == sp.sqrt(3)/2
assert sp.cos(sp.pi/3) == sp.Rational(1, 2)
assert sp.simplify(sp.tan(sp.pi/3) - sp.sqrt(3)) == 0
print("sin(pi/6) =", sp.sin(sp.pi/6))
print("cos(pi/6) =", sp.cos(sp.pi/6))
print("tan(pi/3) =", sp.tan(sp.pi/3))
print("Confirmed")
```

## Example 4 — Range of an inverse trigonometric function

### Hand solution

$\sin$ is one-to-one on $\left[-\dfrac{\pi}{2},\dfrac{\pi}{2}\right]$, and that same interval is the range of $\arcsin$.

$$
\arcsin\dfrac{1}{2}=\dfrac{\pi}{6},
$$

not $\dfrac{5\pi}{6}$, even though the sine of both angles is the same.
Among all possible angles, the inverse returns the representative that lies in the range.

Likewise

$$
\arccos\dfrac{1}{2}=\dfrac{\pi}{3},
$$

because the range of $\arccos$ is the interval $[0,\pi]$.

### Python recheck


```python
import sympy as sp

# Hand answers: arcsin(1/2) = pi/6 and arccos(1/2) = pi/3.
# arcsin does not return 5pi/6, and arccos does not leave [0, pi].
print("arcsin(1/2) =", sp.asin(sp.Rational(1, 2)))
print("arccos(1/2) =", sp.acos(sp.Rational(1, 2)))
assert sp.asin(sp.Rational(1, 2)) == sp.pi/6
assert sp.acos(sp.Rational(1, 2)) == sp.pi/3
print("Confirmed")
```

## Example 5 — An identity, before any limit simplification

### Hand solution

$$
\begin{aligned}
\dfrac{1-\cos^2 x}{\sin x}
&= \dfrac{\sin^2 x}{\sin x} \\
&= \sin x,
\end{aligned}
$$

only when $\sin x\neq 0$.
This is the same domain caution as in lesson 01: simplifying does not bring the zeros of the denominator back to life.

### Python recheck


```python
import sympy as sp

# Identity: (1 - cos(x)^2) / sin(x) simplifies to sin(x) when sin(x) != 0.
x = sp.symbols("x")
expression = (1 - sp.cos(x)**2)/sp.sin(x)
print("Simplified:", sp.simplify(expression))
# The difference from sin(x) simplifies to zero.
assert sp.simplify(expression - sp.sin(x)) == 0
print("Confirmed")
```

## A pitfall

`math.sin` takes radians. If you pass degrees, it computes $\sin 30$ in place of $\sin 30^\circ$,
and you get an unrelated number. In this course every angle is in radians.

## Summary

The exponential and the logarithm are inverses of each other.
Keep the values at $\pi/6$ and $\pi/3$ exact. Their decimals cause trouble in limits and derivatives.
The ranges of $\arcsin$ and $\arccos$ discard part of the answer, and that discarding is deliberate.
