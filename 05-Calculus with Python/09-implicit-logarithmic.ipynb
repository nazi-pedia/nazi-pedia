Sometimes $y$ cannot be written as an explicit formula in $x$, but an equation still ties the two variables together.
Differentiate both sides with respect to $x$, and every time you meet $y$, multiply by $y'$. That multiplication is the chain rule.
Then solve the equation for $y'$.

Logarithmic differentiation is for long products and for $y=u(x)^{v(x)}$.
Take $\ln$ of both sides, then differentiate.

$$
\begin{aligned}
 \dfrac{y'}{y} &= (\ln y)'.
\end{aligned}
$$


```python
import sympy as sp

# Compare a computed result with the hand result.
def check(found, hand, label="result"):
    # Simplify both sides, then their difference.
    found_s = sp.simplify(found)
    hand_s = sp.simplify(hand)
    difference = sp.simplify(found_s - hand_s)
    print(label)
    print("  hand solution:", hand_s)
    print("  Python:", found_s)
    print("  difference:", difference)
    # The assert checks that the simplified expressions agree.
    assert difference == 0
    print("  confirmed")
```

## Example 1 — A circle

### Hand solution

$$
\begin{aligned}
 x^2+y^2 &= 25, \\
 2x+2y\,y' &= 0, \\
 y' &= -\dfrac{x}{y}.
\end{aligned}
$$

At the point $(3,4)$, which lies on the circle because $9+16=25$,

$$
\begin{aligned}
 y' &= -\dfrac{3}{4}.
\end{aligned}
$$

The slope of the radius is $\frac{4}{3}$, and the slope of the tangent is the negative reciprocal of that slope. Geometry and the derivative say the same thing.
At $(3,-4)$ the slope is $+\frac{3}{4}$. The formula is incomplete if $y$ is missing from the denominator.

### Python recheck


```python
import sympy as sp

# Circle x**2 + y(x)**2 = 25. Hand result: y' = -x/y.
x = sp.symbols("x")
y = sp.Function("y")
slope = sp.idiff(x**2 + y(x)**2 - 25, y(x), x)
print(slope)
# The slope formula is -x/y(x); at (3, 4) it is -3/4, and at (3, -4) it is 3/4.
assert sp.simplify(slope + x/y(x)) == 0
assert slope.subs({x: 3, y(x): 4}) == -sp.Rational(3, 4)
assert slope.subs({x: 3, y(x): -4}) == sp.Rational(3, 4)
print("confirmed")
```

## Example 2 — A cubic curve

### Hand solution

$$
\begin{aligned}
 x^3+y^3 &= 6xy, \\
 3x^2+3y^2 y' &= 6y+6x y', \\
 3y^2 y'-6x y' &= 6y-3x^2, \\
 y'
 &= \dfrac{6y-3x^2}{3y^2-6x} \\
 &= \dfrac{2y-x^2}{y^2-2x}.
\end{aligned}
$$

At $(3,3)$ the numerator is $6-9=-3$ and the denominator is $9-6=3$, so the slope is $-1$.
The point lies on the curve: $27+27=54$ and $6\cdot 3\cdot 3=54$.

### Python recheck


```python
import sympy as sp

# Curve x**3 + y**3 = 6*x*y. Hand result: (2*y - x**2)/(y**2 - 2*x).
x = sp.symbols("x")
y = sp.Function("y")
slope = sp.simplify(sp.idiff(x**3 + y(x)**3 - 6*x*y(x), y(x), x))
hand = (2*y(x) - x**2)/(y(x)**2 - 2*x)
print(slope)
# The formula matches, and the slope at (3, 3) is -1.
assert sp.simplify(slope - hand) == 0
assert slope.subs({x: 3, y(x): 3}) == -1
print("confirmed")
```

## Example 3 — A variable exponent

### Hand solution

$$
\begin{aligned}
 y &= x^x, \qquad x>0, \\
 \ln y &= x\ln x, \\
 \dfrac{y'}{y}
 &= \ln x+x\cdot\dfrac{1}{x} \\
 &= \ln x+1, \\
 y' &= x^x(\ln x+1).
\end{aligned}
$$

At $x=1$, $\ln 1+1=1$ and $1^1=1$, so the slope is $1$.

### Python recheck


```python
import sympy as sp

# y = x**x for x > 0. Hand result: x**x * (log(x) + 1), and the slope at 1 is 1.
x = sp.symbols("x", positive=True)
y = x**x
found = sp.diff(y, x)
hand = x**x * (sp.log(x) + 1)
print(sp.simplify(found))
assert sp.simplify(found - hand) == 0
assert found.subs(x, 1) == 1
print("confirmed")
```

## Example 4 — Base and exponent both variable

### Hand solution

$$
\begin{aligned}
 y &= (\sin x)^x, \qquad \sin x>0, \\
 \ln y &= x\ln\sin x, \\
 \dfrac{y'}{y}
 &= \ln\sin x+x\cdot\dfrac{\cos x}{\sin x}, \\
 y' &= (\sin x)^x\left(\ln\sin x+x\cot x\right).
\end{aligned}
$$

### Python recheck


```python
import sympy as sp

# y = sin(x)**x. Hand result: y times (log(sin(x)) + x*cos(x)/sin(x)).
x = sp.symbols("x")
y = sp.sin(x)**x
found = sp.diff(y, x)
hand = y * (sp.log(sp.sin(x)) + x*sp.cos(x)/sp.sin(x))
# The assert checks that logarithmic differentiation matches SymPy.
assert sp.simplify(found - hand) == 0
print(sp.simplify(found))
print("confirmed")
```

## Example 5 — A logarithm used to break up a product

### Hand solution

$$
\begin{aligned}
 y &= \dfrac{(x^2+1)\sqrt{x+2}}{e^{x}}, \qquad x>-2, \\
 \ln y
 &= \ln(x^2+1)+\dfrac{1}{2}\ln(x+2)-x, \\
 \dfrac{y'}{y}
 &= \dfrac{2x}{x^2+1}+\dfrac{1}{2(x+2)}-1, \\
 y'
 &= y\left(\dfrac{2x}{x^2+1}+\dfrac{1}{2(x+2)}-1\right).
\end{aligned}
$$

There is no need to combine everything into a single fraction first. The product of $y$ and the parenthesis is already the derivative.

### Python recheck


```python
import sympy as sp

# y = (x**2 + 1)*sqrt(x + 2)*exp(-x), for real x.
# Hand factor from logarithmic differentiation: 2x/(x**2+1) + 1/(2(x+2)) - 1.
x = sp.symbols("x", real=True)
y = (x**2 + 1)*sp.sqrt(x + 2)*sp.exp(-x)
factor = 2*x/(x**2 + 1) + 1/(2*(x + 2)) - 1
found = sp.diff(y, x)
hand = y * factor
# The assert checks that the derivative equals y times that factor.
assert sp.simplify(found - hand) == 0
print("the derivative matches the logarithmic form")
print("confirmed")
```

## A pitfall

In implicit differentiation, $y$ is a function of $x$ even when you do not write $y(x)$.
Every term that contains $y$ picks up a factor $y'$.
A term that contains only $x$ does not.

## Summary

Implicit differentiation: differentiate both sides, collect $y'$, and solve.
Logarithmic differentiation: take $\ln$ of both sides, differentiate, and multiply by $y$.
At a point, first make sure the point lies on the curve.
