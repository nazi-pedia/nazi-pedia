# 21 Arc Length and Surfaces of Revolution

| | |
|---|---|
| **Course** | Calculus II |
| **Prerequisite** | 8 and 16 |

The length of the graph $y=f(x)$ from $a$ to $b$ is

$$
L=\int_{a}^{b}\sqrt{1+\bigl(f'(x)\bigr)^{2}}\,dx.
$$

If that arc is rotated about the $x$-axis, the surface area is

$$
S=2\pi\int_{a}^{b} y\sqrt{1+(y')^{2}}\,dx.
$$

The factor $\sqrt{1+(y')^{2}}\,dx$ is the arc-length element, and $2\pi y$ is the circumference of the circle traced by that element.



## Example 1 — A slanted line segment

### Hand solution

Take $y=x$ from $0$ to $1$. The slope is $1$.

$$
\begin{aligned}
L &= \int_{0}^{1}\sqrt{1+1}\,dx \\
&= \sqrt{2}.
\end{aligned}
$$

That is the distance from $(0,0)$ to $(1,1)$.

### Python recheck



```python
import sympy as sp

# Arc length of the segment y = x from 0 to 1.
x = sp.symbols("x")
length = sp.integrate(sp.sqrt(1 + sp.diff(x, x)**2), (x, 0, 1))
print(length)

# The length is sqrt(2).
assert length == sp.sqrt(2)
print("Confirmed")

```

## Example 2 — An arc whose radical behaves well

### Hand solution

Take $y=x^{3/2}$ from $0$ to $1$.

$$
\begin{aligned}
y' &= \dfrac{3}{2}x^{1/2}, \\
1+(y')^{2} &= 1+\dfrac{9}{4}x, \\
u &= 1+\dfrac{9}{4}x, \quad du=\dfrac{9}{4}\,dx, \quad u:1\to\dfrac{13}{4}, \\
L &= \int_{0}^{1}\sqrt{1+\dfrac{9}{4}x}\,dx \\
&= \dfrac{4}{9}\cdot\dfrac{2}{3}\left[u^{3/2}\right]_{1}^{13/4} \\
&= \dfrac{8}{27}\left(\left(\dfrac{13}{4}\right)^{3/2}-1\right) \\
&= \dfrac{13\sqrt{13}-8}{27}.
\end{aligned}
$$

### Python recheck



```python
import sympy as sp

# Arc length of y = x**(3/2) on [0, 1].
x = sp.symbols("x", nonnegative=True)
y = x**sp.Rational(3, 2)
length = sp.integrate(sp.sqrt(1 + sp.diff(y, x)**2), (x, 0, 1))

# Compare with the antiderivative evaluated by hand.
hand = (13*sp.sqrt(13) - 8)/27
print(sp.simplify(length))
assert sp.simplify(length - hand) == 0
print("Confirmed")

```

## Example 3 — Lateral surface of a cone

### Hand solution

Rotating the segment $y=x$ from $0$ to $1$ about the $x$-axis produces a cone of radius $1$ and slant height $\sqrt{2}$. The classical lateral area is $\pi r \ell=\pi\sqrt{2}$.

$$
\begin{aligned}
S &= 2\pi\int_{0}^{1} x\sqrt{2}\,dx \\
&= 2\pi\sqrt{2}\cdot\dfrac{1}{2} \\
&= \pi\sqrt{2}.
\end{aligned}
$$

The two methods agree.

### Python recheck



```python
import sympy as sp

# Surface area of the cone generated by y = x.
x = sp.symbols("x")
y = x
surface = 2*sp.pi*sp.integrate(y*sp.sqrt(1 + sp.diff(y, x)**2), (x, 0, 1))
print(sp.simplify(surface))

# The lateral area is pi*sqrt(2).
assert sp.simplify(surface - sp.pi*sp.sqrt(2)) == 0
print("Confirmed")

```

## Example 4 — A semicircle

### Hand solution

The graph $y=\sqrt{1-x^{2}}$ from $-1$ to $1$ is the unit semicircle, so its length must be $\pi$. The derivative becomes infinite at the endpoints, so the integral is improper, but the value is finite.

$$
\begin{aligned}
y' &= \dfrac{-x}{\sqrt{1-x^{2}}}, \\
1+(y')^{2} &= \dfrac{1}{1-x^{2}}, \\
L &= \int_{-1}^{1}\dfrac{1}{\sqrt{1-x^{2}}}\,dx \\
&= \arcsin x\Big|_{-1}^{1} \\
&= \pi.
\end{aligned}
$$

### Python recheck



```python
import sympy as sp

# Build the arc-length integrand of the unit semicircle.
x = sp.symbols("x")
y = sp.sqrt(1 - x**2)
raw = sp.sqrt(1 + sp.diff(y, x)**2)

# Check that it matches 1/sqrt(1-x^2), then integrate that form.
hand_integrand = 1/sp.sqrt(1 - x**2)
assert sp.simplify(raw**2 - hand_integrand**2) == 0
length = sp.integrate(hand_integrand, (x, -1, 1))

# Estimate the original improper integral numerically.
numeric = sp.N(sp.Integral(raw, (x, -1, 1)))
print(sp.simplify(raw), "->", length, "Numeric:", numeric)

# The exact length is pi, and the numeric value agrees.
assert sp.simplify(length - sp.pi) == 0
assert abs(numeric - sp.pi) < 10**(-10)
print("Confirmed")

```

## A pitfall

If $1+(y')^{2}$ is not a perfect square, the arc-length integral is often not elementary. A numerical approximation is then an answer, not a failure of the problem. First check whether the radical simplifies.

## Summary

Arc length is the integral of infinitesimal chords. A surface of revolution integrates that arc-length element against the circumference of the path of rotation.
