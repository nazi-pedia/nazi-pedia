## تعریف‌ها

برای \(a>0\) و \(a\neq 1\)،

\[
y=\log_a x \iff a^y=x,\qquad x>0
\]

لگاریتم طبیعی، لگاریتم در پایهٔ \(e\) است و \(\ln\) نوشته می‌شود.
در پایتون و SymPy نام این تابع `log` است و پایهٔ پیش‌فرضش \(e\) است.

روی دایرهٔ واحد، \(\cos\theta\) مختص \(x\) و \(\sin\theta\) مختص \(y\) است.
زاویه در حسابان همیشه بر حسب رادیان است: \(\pi\) رادیان یعنی ۱۸۰ درجه.

\[
\sin^2\theta+\cos^2\theta=1
\]

\[
\arcsin u=\theta \iff \sin\theta=u,\quad \theta\in\left[-\frac{\pi}{2},\frac{\pi}{2}\right]
\]



```python
import sympy as sp

def check(found, hand, label="نتیجه"):
    found_s = sp.simplify(found)
    hand_s = sp.simplify(hand)
    difference = sp.simplify(found_s - hand_s)
    print(label)
    print("  حل دستی:", hand_s)
    print("  پایتون:", found_s)
    print("  اختلاف:", difference)
    assert difference == 0
    print("  تایید شد")

```

## مثال ۱ — معادلهٔ نمایی

### حل ریاضی

\[
2^{x+1}=8
\]

\(8=2^3\)، پس \(2^{x+1}=2^3\) و چون پایه بزرگ‌تر از ۱ است، تابع یک‌به‌یک است:

\[
x+1=3,\qquad x=2
\]

آزمون: \(2^{3}=8\).

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
roots = sp.solve(sp.Eq(2**(x + 1), 8), x)
print("ریشه:", roots)
assert roots == [2]
assert 2**(2 + 1) == 8
print("تایید شد")

```

## مثال ۲ — قانون‌های لگاریتم، نه ماشین‌حساب

### حل ریاضی

\[
\ln(e^5\cdot e^{-2})=\ln(e^5)+\ln(e^{-2})=5-2=3
\]

و

\[
\log_2 32=\log_2(2^5)=5
\]

لگاریتم، توان را پایین می‌آورد. حاصل یک عدد صحیح دقیق است، نه یک تقریب اعشاری.

### بازبینی با پایتون



```python
import sympy as sp
exact_ln = sp.simplify(sp.log(sp.E**5 * sp.E**(-2)))
exact_log2 = sp.log(32, 2)
print("ln:", exact_ln)
print("log2 32:", exact_log2)
assert exact_ln == 3
assert exact_log2 == 5
print("تایید شد")

```

## مثال ۳ — مقدارهای دقیق مثلثاتی

### حل ریاضی

در مثلث متساوی‌الاضلاع با ضلع ۲، ارتفاع \(\sqrt{3}\) است.
نصف آن، مثلث \(30^\circ\!-\!60^\circ\!-\!90^\circ\) می‌دهد:

\[
\sin\frac{\pi}{6}=\frac{1}{2},\qquad
\cos\frac{\pi}{6}=\frac{\sqrt{3}}{2},\qquad
\tan\frac{\pi}{3}=\sqrt{3}
\]

\[
\sin\frac{\pi}{3}=\frac{\sqrt{3}}{2},\qquad
\cos\frac{\pi}{3}=\frac{1}{2}
\]

### بازبینی با پایتون



```python
import sympy as sp
assert sp.sin(sp.pi/6) == sp.Rational(1, 2)
assert sp.cos(sp.pi/6) == sp.sqrt(3)/2
assert sp.sin(sp.pi/3) == sp.sqrt(3)/2
assert sp.cos(sp.pi/3) == sp.Rational(1, 2)
assert sp.simplify(sp.tan(sp.pi/3) - sp.sqrt(3)) == 0
print("sin(pi/6) =", sp.sin(sp.pi/6))
print("cos(pi/6) =", sp.cos(sp.pi/6))
print("tan(pi/3) =", sp.tan(sp.pi/3))
print("تایید شد")

```

## مثال ۴ — بردِ وارون مثلثاتی

### حل ریاضی

\(\sin\) روی \([-\pi/2,\pi/2]\) یک‌به‌یک است و همین بازه، برد \(\arcsin\) است.

\[
\arcsin\frac{1}{2}=\frac{\pi}{6}
\]

نه \(\dfrac{5\pi}{6}\)، هرچند سینوس هر دو یکی است.
وارون، از میان همهٔ زاویه‌های ممکن، نمایندهٔ داخل برد را برمی‌گرداند.

همین‌طور \(\arccos\dfrac{1}{2}=\dfrac{\pi}{3}\)، چون برد \(\arccos\) بازهٔ \([0,\pi]\) است.

### بازبینی با پایتون



```python
import sympy as sp
print("arcsin(1/2) =", sp.asin(sp.Rational(1, 2)))
print("arccos(1/2) =", sp.acos(sp.Rational(1, 2)))
assert sp.asin(sp.Rational(1, 2)) == sp.pi/6
assert sp.acos(sp.Rational(1, 2)) == sp.pi/3
print("تایید شد")

```

## مثال ۵ — اتحاد، پیش از هر ساده‌سازی حدی

### حل ریاضی

\[
\frac{1-\cos^2 x}{\sin x}=\frac{\sin^2 x}{\sin x}=\sin x
\]

فقط وقتی \(\sin x\neq 0\).
این همان احتیاط دامنه در درس ۰۱ است: ساده‌کردن، صفرهای مخرج را زنده نمی‌کند.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
expression = (1 - sp.cos(x)**2)/sp.sin(x)
print("ساده‌شده:", sp.simplify(expression))
assert sp.simplify(expression - sp.sin(x)) == 0
print("تایید شد")

```
