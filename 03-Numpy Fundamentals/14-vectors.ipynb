A one-dimensional array is a vector. The dot product, the length, and the angle are definitions, not tricks of the library. Each example is computed by hand, then recomputed with NumPy.

$$
\mathbf{a}\cdot\mathbf{b} = \sum_i a_i b_i,
\qquad
\|\mathbf{a}\| = \sqrt{\mathbf{a}\cdot\mathbf{a}}.
$$


## Example 1 — Dot product

### Predict

$$
(1, 2, 3)\cdot(4, 5, 6)
= 1\cdot 4 + 2\cdot 5 + 3\cdot 6
= 4 + 10 + 18
= 32.
$$

For one-dimensional arrays, `@` is this same sum of products.

### Run


```python
import numpy as np

left = np.array([1, 2, 3])
right = np.array([4, 5, 6])
# The hand sum is 32. dot and @ agree on vectors.
print(np.dot(left, right))
print(left @ right)
```

## Example 2 — Length

### Predict

$$
\|(3, 4)\| = \sqrt{3^2 + 4^2} = \sqrt{9 + 16} = \sqrt{25} = 5.
$$

### Run


```python
# Example 2 — Length.
import numpy as np

vector = np.array([3.0, 4.0])
print(np.linalg.norm(vector))
print(np.sqrt(np.dot(vector, vector)))
```

## Example 3 — The angle between two vectors

### Predict

$\mathbf{a} = (1, 0)$ and $\mathbf{b} = (1, 1)$.

$$
\mathbf{a}\cdot\mathbf{b} = 1,
\qquad
\|\mathbf{a}\| = 1,
\qquad
\|\mathbf{b}\| = \sqrt{2},
\qquad
\cos\theta = \dfrac{1}{\sqrt{2}}.
$$

So $\theta = \pi/4$.

### Run


```python
# Example 3 — The angle between two vectors.
import numpy as np

a = np.array([1.0, 0.0])
b = np.array([1.0, 1.0])
cos_theta = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
theta = np.arccos(cos_theta)
print(np.allclose(cos_theta, 1 / np.sqrt(2)))
print(np.allclose(theta, np.pi / 4))
```

## Example 4 — Projection

### Predict

The projection of $\mathbf{a}$ onto $\mathbf{b}$ is

$$
\mathrm{proj}_{\mathbf{b}}\mathbf{a}
= \dfrac{\mathbf{a}\cdot\mathbf{b}}{\mathbf{b}\cdot\mathbf{b}}\,\mathbf{b}.
$$

With $\mathbf{a}=(1,0)$ and $\mathbf{b}=(1,1)$, the fraction is $1/2$, and the projection is $(1/2,\ 1/2)$.

### Run


```python
# Example 4 — Projection.
import numpy as np

a = np.array([1.0, 0.0])
b = np.array([1.0, 1.0])
scale = np.dot(a, b) / np.dot(b, b)
projection = scale * b
print(projection)
print(np.allclose(projection, [0.5, 0.5]))
```

## Example 5 — A cross product in three dimensions

### Predict

$$
(1, 2, 3) \times (0, 1, 0)
= (2\cdot 0 - 3\cdot 1,\ 3\cdot 0 - 1\cdot 0,\ 1\cdot 1 - 2\cdot 0)
= (-3,\ 0,\ 1).
$$

### Run


```python
# Example 5 — A cross product in three dimensions.
import numpy as np

a = np.array([1, 2, 3])
b = np.array([0, 1, 0])
print(np.cross(a, b))
```

## A pitfall

`*` is still elementwise. `(1, 2, 3) * (4, 5, 6)` equals `(4, 10, 18)`, not 32. The sum that makes a dot product is a separate step, or a call to `np.dot`.

### Run


```python
# A pitfall.
import numpy as np

left = np.array([1, 2, 3])
right = np.array([4, 5, 6])
print(left * right)
print(np.sum(left * right))
```

## Summary

The dot product adds the elementwise products. The norm is the square root of a vector dotted with itself. The cosine of the angle divides the dot product by the two lengths. Projection scales $\mathbf{b}$ by $(\mathbf{a}\cdot\mathbf{b})/(\mathbf{b}\cdot\mathbf{b})$. `*` does none of this addition for you.
