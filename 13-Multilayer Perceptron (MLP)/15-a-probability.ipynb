# 15 A Score Can Be Read as a Probability

| | |
|---|---|
| **Path** | Multilayer perceptron |
| **Prerequisite** | 14 |

Squared loss treats the output as a number. A pass-or-fail target can instead be a probability. Pass the score through the sigmoid of Lesson 07, and charge the logarithmic loss

$$
L = -\, y\log\sigma(s) - (1 - y)\log\bigl(1 - \sigma(s)\bigr)
$$

Its derivative with respect to the score collapses. Using $\sigma' = \sigma(1-\sigma)$,

$$
\begin{aligned}
\dfrac{\partial L}{\partial s}
&= -y\,\dfrac{\sigma'}{\sigma} + (1 - y)\,\dfrac{\sigma'}{1 - \sigma} \\
&= -y\,(1 - \sigma) + (1 - y)\,\sigma \\
&= \sigma - y
\end{aligned}
$$

At $s = 0$ and $y = 1$,

$$
\sigma(0) = \dfrac{1}{2}, \qquad
\dfrac{\partial L}{\partial s} = \dfrac{1}{2} - 1 = -\dfrac{1}{2}
$$

The loss itself is

$$
L = -\log\dfrac{1}{2} = \log 2
$$

A step of size $1$ on the score alone gives

$$
s \leftarrow 0 - 1\cdot\left(-\dfrac{1}{2}\right) = \dfrac{1}{2}
$$

Lesson 07 already showed $\sigma(1/2) > 1/2$. The probability of pass moves up from one half. The new probability is not a small integer ratio, and this page does not pretend that it is.


## Example 1 — The derivative at one half


```python
# At score 0 and target 1, the logistic gradient is sigma - y = -1/2.
from fractions import Fraction

sigma = Fraction(1, 2)
y = 1
gradient = sigma - y
print(gradient)
assert gradient == Fraction(-1, 2)
```

## Example 2 — The score moves to one half, and the probability rises


```python
# The new score is 1/2. Its sigmoid is strictly above the old probability 1/2.
import math
from fractions import Fraction

new_score = 0 - 1 * Fraction(-1, 2)
probability = 1 / (1 + math.exp(-float(new_score)))
print(new_score, probability)
assert new_score == Fraction(1, 2)
assert probability > 0.5
```

## Pitfall — $\sigma - y$ is not the squared residual

Squared loss differentiates to $s - y$. Logistic loss differentiates to $\sigma(s) - y$. At $s = 0$ and $y = 1$ those numbers are $-1$ and $-1/2$. Same direction, different length. Mixing them doubles or halves the first step.


```python
# Squared residual at s = 0, y = 1 is -1. Logistic gradient is -1/2.
from fractions import Fraction

squared = 0 - 1
logistic = Fraction(1, 2) - 1
print(squared, logistic)
assert squared == -1
assert logistic == Fraction(-1, 2)
```

## Summary

Logistic loss, after the sigmoid, has derivative $\sigma(s) - y$. At the balanced score $0$ with target pass, that derivative is $-1/2$ and the loss is $\log 2$. One unit step moves the score to $1/2$ and lifts the probability above $1/2$.
