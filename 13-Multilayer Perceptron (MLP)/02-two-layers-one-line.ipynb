# 02 Two Straight Layers Are Still One Line

| | |
|---|---|
| **Path** | Multilayer perceptron |
| **Prerequisite** | 01 |

Put a second perceptron on top of the two scores. Its weights are $v = (1, -2)$ and its bias is $c = 0$.

$$
s = z_1 - 2 z_2
$$

If $z = Wx + b$ with no bend in between, the composition is still one affine score:

$$
\begin{aligned}
s &= v\cdot(Wx + b) + c \\
&= (v^{\top} W)\, x + (v\cdot b + c)
\end{aligned}
$$

Depth without a bend does not create a new kind of function. It only renames the weights.

Here

$$
v^{\top} W = (1,\,-2)\begin{pmatrix} 1 & 1 \\ 1 & 1 \end{pmatrix} = (-1,\,-1)
$$

and the effective bias is

$$
v\cdot b + c = (1)(0) + (-2)(-1) = 2
$$

so

$$
s = -x_1 - x_2 + 2
$$


## Example 1 — The four straight scores

$$
\begin{aligned}
s(0, 0) &= 2 \\
s(1, 0) &= 1 \\
s(0, 1) &= 1 \\
s(1, 1) &= 0
\end{aligned}
$$

Gate 00 wanted target $0$ and received $2$. The straight stack misses XOR on the first gate.


```python
# Straight composition. Gate 00 scores 2, not the target 0.
gates = [(0, 0), (1, 0), (0, 1), (1, 1)]
scores = [(-x1 - x2 + 2) for x1, x2 in gates]
print(scores)
assert scores == [2, 1, 1, 0]
```

## Example 2 — The renamed weights reproduce the stack

Compute $s = z_1 - 2z_2$ from the layer, and compute $-x_1 - x_2 + 2$ from the collapsed formula. They are the same four numbers.


```python
# The matrix product and the two-layer arithmetic agree on every gate.
gates = [(0, 0), (1, 0), (0, 1), (1, 1)]
from_layers = []
from_collapse = []
for x1, x2 in gates:
    z1 = x1 + x2
    z2 = x1 + x2 - 1
    from_layers.append(1 * z1 + (-2) * z2)
    from_collapse.append(-x1 - x2 + 2)
print(from_layers)
assert from_layers == from_collapse == [2, 1, 1, 0]
```

## Pitfall — Adding a third straight layer still collapses

Three affine maps are one affine map. The same algebra absorbs the extra matrix into $v^{\top} W$. A deeper stack of straight scores cannot represent the four gates, because path 11 already showed that no single straight score can.


## Summary

Two layers with no bend compose into $s = -x_1 - x_2 + 2$. The four values are $2, 1, 1, 0$. Gate 00 is wrong. Depth starts to matter at the first function that is not affine.
