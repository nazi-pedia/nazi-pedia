Strip the network down to one input, one hidden wire, and one output. The hidden wire is the identity, so its slope is $1$. ReLU returns in the next lesson as a slope that can be $0$.

$$
\begin{aligned}
z &= wx + b \\
h &= z \\
s &= vh + c \\
L &= \dfrac{1}{2}(s - y)^{2}
\end{aligned}
$$

Take $x = 1$, $y = 0$, and

$$
w = 1,\quad b = 0,\quad v = 1,\quad c = 0
$$

Then $z = 1$, $h = 1$, $s = 1$, and

$$
L = \dfrac{1}{2}
$$

Differentiate from the loss backward.

$$
\begin{aligned}
\dfrac{\partial L}{\partial s} &= s - y = 1 \\
\dfrac{\partial L}{\partial v} &= (s - y)\, h = 1 \\
\dfrac{\partial L}{\partial c} &= s - y = 1 \\
\dfrac{\partial L}{\partial h} &= (s - y)\, v = 1 \\
\dfrac{\partial L}{\partial z} &= 1 \\
\dfrac{\partial L}{\partial w} &= x = 1 \\
\dfrac{\partial L}{\partial b} &= 1
\end{aligned}
$$

Every factor in this particular network happened to be $1$. The pattern is the thing to keep: each weight receives the loss signal times the value that weight multiplies.


## Example 1 — One step of size $1/2$

$$
w \leftarrow w - \eta\dfrac{\partial L}{\partial w}
$$

and the same rule for $b$, $v$, and $c$, with $\eta = 1/2$.

$$
\begin{aligned}
w &= 1 - \dfrac{1}{2} = \dfrac{1}{2} \\
b &= 0 - \dfrac{1}{2} = -\dfrac{1}{2} \\
v &= \dfrac{1}{2} \\
c &= -\dfrac{1}{2}
\end{aligned}
$$

The new forward pass is

$$
z = \dfrac{1}{2} - \dfrac{1}{2} = 0, \qquad s = -\dfrac{1}{2}
$$

$$
L = \dfrac{1}{2}\left(-\dfrac{1}{2}\right)^{2} = \dfrac{1}{8}
$$

The loss fell from $1/2$ to $1/8$.


```python
# Hand: every partial is 1, the step of 1/2 halves each weight, and the loss falls to 1/8.
from fractions import Fraction

x, y = 1, 0
w, b, v, c = 1, 0, 1, 0
z = w * x + b
h = z
s = v * h + c
loss = Fraction(1, 2) * (s - y) ** 2
d_s = s - y
d_v = d_s * h
d_c = d_s
d_h = d_s * v
d_w = d_h * x
d_b = d_h
print(loss, d_w, d_b, d_v, d_c)
assert loss == Fraction(1, 2)
assert (d_w, d_b, d_v, d_c) == (1, 1, 1, 1)
eta = Fraction(1, 2)
w, b = w - eta * d_w, b - eta * d_b
v, c = v - eta * d_v, c - eta * d_c
z = w * x + b
s = v * z + c
new_loss = Fraction(1, 2) * (s - y) ** 2
print(w, b, v, c, s, new_loss)
assert (w, b, v, c) == (Fraction(1, 2), Fraction(-1, 2), Fraction(1, 2), Fraction(-1, 2))
assert s == Fraction(-1, 2)
assert new_loss == Fraction(1, 8)
```

## Pitfall — A smaller loss on this point is not a promise about every step size

$\eta = 1/2$ worked here. Lesson 12 uses the four gates and shows a step size that raises the total loss. The chain rule gives the direction. The step size decides whether that direction, taken too far, walks past the bottom.


## Summary

The loss derivative arrives at a weight multiplied by the forward value that weight sees. On this one-wire network every partial equals $1$. A step of $1/2$ moves the loss from $1/2$ to $1/8$.
