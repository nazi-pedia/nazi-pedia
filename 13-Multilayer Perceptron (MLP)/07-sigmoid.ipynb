ReLU has a corner. The logistic sigmoid is a bend with a slope everywhere.

$$
\sigma(z) = \dfrac{1}{1 + e^{-z}}
$$

At zero the exponential is $1$, so

$$
\sigma(0) = \dfrac{1}{1 + 1} = \dfrac{1}{2}
$$

The derivative follows from the quotient rule. Write $e^{-z}$ as $u$.

$$
\begin{aligned}
\sigma(z) &= \dfrac{1}{1 + u} \\
1 - \sigma(z) &= \dfrac{u}{1 + u} \\
\sigma(z)\,(1 - \sigma(z)) &= \dfrac{u}{(1 + u)^{2}}
\end{aligned}
$$

And

$$
\sigma'(z) = \dfrac{u}{(1 + u)^{2}}
$$

because $du/dz = -u$ and the derivative of $(1+u)^{-1}$ contributes a factor $u$. Therefore

$$
\sigma'(z) = \sigma(z)\,\bigl(1 - \sigma(z)\bigr)
$$

At zero,

$$
\sigma'(0) = \dfrac{1}{2}\cdot\dfrac{1}{2} = \dfrac{1}{4}
$$

The steepest slope of this bend is $1/4$, not $1$. ReLU can have slope $1$. Sigmoid never does.


## Example 1 — The value and the slope at zero


```python
# sigma(0) = 1/2 and sigma'(0) = 1/4, from the identity sigma(1 - sigma).
from fractions import Fraction

sigma = Fraction(1, 2)
slope = sigma * (1 - sigma)
print(sigma, slope)
assert sigma == Fraction(1, 2)
assert slope == Fraction(1, 4)
```

## Example 2 — A positive score moves the value above one half

$$
e^{-1/2} < 1
$$

so

$$
1 + e^{-1/2} < 2
\qquad\Rightarrow\qquad
\sigma\!\left(\dfrac{1}{2}\right) > \dfrac{1}{2}
$$

The inequality is exact. The decimal is only a view of it.


```python
# A positive input puts the sigmoid strictly above 1/2.
import math

value = 1 / (1 + math.exp(-0.5))
print(value)
assert value > 0.5
```

## Pitfall — Do not replace $e^{-1/2}$ by a nearby fraction and call it exact

$\sigma(1/2)$ is not a ratio of small integers. The exact statements on this page are $\sigma(0) = 1/2$, $\sigma'(0) = 1/4$, and $\sigma(1/2) > 1/2$. The printed decimal is evidence for the inequality, not a second exact form.


## Summary

Sigmoid maps every score into $(0, 1)$, equals $1/2$ at zero, and has slope $\sigma(1-\sigma)$. That slope is $1/4$ at zero, which is its maximum. The chain rule in the next lessons will use a slope of this kind as one factor among others.
