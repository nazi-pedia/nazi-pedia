A forward pass produces a score $s$. A target is a number $y$. The loss used in the rest of this path, except the probability lesson, is

$$
L = \dfrac{1}{2}(s - y)^{2}
$$

The factor $1/2$ does not move the minimum. It cancels the $2$ from the derivative:

$$
\dfrac{\partial L}{\partial s} = s - y
$$

On the finished XOR network every training loss is zero, so every derivative with respect to $s$ is zero. A zero derivative means the output weights do not want to move. To see a nonzero lesson, move one output weight. Replace $-2$ by $-1$:

$$
s = h_1 - h_2
$$

The hidden units stay as in Lesson 04. The four scores become

$$
\begin{aligned}
s(0, 0) &= 0 \\
s(1, 0) &= 1 \\
s(0, 1) &= 1 \\
s(1, 1) &= 2 - 1 = 1
\end{aligned}
$$

Three gates are exact. Gate 11 has residual $1 - 0 = 1$, and

$$
L(1, 1) = \dfrac{1}{2}(1)^{2} = \dfrac{1}{2}
$$

The total loss, the sum over the four gates, is also $1/2$.


## Example 1 — Residuals and the total


```python
# Three gates are exact. Gate 11 contributes loss 1/2, and that is the total.
from fractions import Fraction

scores = [0, 1, 1, 1]
targets = [0, 1, 1, 0]
losses = [Fraction(1, 2) * (score - target) ** 2 for score, target in zip(scores, targets)]
print(losses, sum(losses))
assert losses == [0, 0, 0, Fraction(1, 2)]
assert sum(losses) == Fraction(1, 2)
```

## Example 2 — The derivative is the residual

At gate 11, $s - y = 1$. At each exact gate the residual is $0$. The vector of derivatives with respect to the four scores is $(0, 0, 0, 1)$.


```python
# dL/ds equals the residual because the one-half cancels the square's factor 2.
residuals = [0 - 0, 1 - 1, 1 - 1, 1 - 0]
print(residuals)
assert residuals == [0, 0, 0, 1]
```

## Pitfall — A finished network has nothing to teach its output weights

If every residual is zero, every factor $\partial L/\partial s$ is zero, and the chain rule multiplies the rest of the network by zero. The weights that already solve the table are a fixed point of gradient descent. The moved weight $-1$, with loss $1/2$, is the table this path will differentiate.


## Summary

The loss is half the squared residual, so its derivative in the score is the residual itself. With output weights $(1, -1)$ the hidden XOR features score three gates perfectly and gate 11 as $1$. Total loss $1/2$. Derivative on that gate: $1$.
