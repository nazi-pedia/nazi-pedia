Return to the four gates. The hidden units are the XOR features of Lesson 04. The output weights are the perturbed pair from Lesson 08,

$$
v_1 = 1, \quad v_2 = -1, \quad c = 0
$$

so gate 11 has hidden values $(h_1, h_2) = (2, 1)$, score $s = 1$, and target $0$.

Both pre-activations on this gate are positive ($z_1 = 2$, $z_2 = 1$), so both ReLU slopes are $1$. Nothing is dead.

$$
\begin{aligned}
\dfrac{\partial L}{\partial s} &= 1 \\
\dfrac{\partial L}{\partial v_1} &= 1 \cdot h_1 = 2 \\
\dfrac{\partial L}{\partial v_2} &= 1 \cdot h_2 = 1 \\
\dfrac{\partial L}{\partial c} &= 1 \\
\dfrac{\partial L}{\partial h_1} &= v_1 = 1 \\
\dfrac{\partial L}{\partial h_2} &= v_2 = -1
\end{aligned}
$$

The hidden gradients continue into the first layer. Gate 11 has $(x_1, x_2) = (1, 1)$.

$$
\begin{aligned}
\dfrac{\partial L}{\partial z_1} &= 1, &
\dfrac{\partial L}{\partial w_{11}} &= 1, &
\dfrac{\partial L}{\partial w_{12}} &= 1, &
\dfrac{\partial L}{\partial b_1} &= 1 \\
\dfrac{\partial L}{\partial z_2} &= -1, &
\dfrac{\partial L}{\partial w_{21}} &= -1, &
\dfrac{\partial L}{\partial w_{22}} &= -1, &
\dfrac{\partial L}{\partial b_2} &= -1
\end{aligned}
$$

The three exact gates have residual $0$, so they add nothing to a full-batch gradient. The batch gradient equals this one gate's gradient.


## Example 1 — The output weights see the hidden values


```python
# Gate 11: h = (2, 1), residual 1. Output gradients are (2, 1, 1).
h1, h2 = 2, 1
residual = 1 - 0
d_v1 = residual * h1
d_v2 = residual * h2
d_c = residual
print(d_v1, d_v2, d_c)
assert (d_v1, d_v2, d_c) == (2, 1, 1)
```

## Example 2 — The first layer sees the output weights

$$
\dfrac{\partial L}{\partial z_i} = \dfrac{\partial L}{\partial h_i}
$$

because both slopes are $1$. Then each input weight is that signal times $x_1$ or $x_2$, both of which are $1$ on this gate.


```python
# Both ReLU slopes are 1. First-layer gradients are (1, 1, 1) and (-1, -1, -1).
v1, v2 = 1, -1
residual = 1
x1, x2 = 1, 1
d_h1 = residual * v1
d_h2 = residual * v2
slope1, slope2 = 1, 1
d_z1, d_z2 = d_h1 * slope1, d_h2 * slope2
row1 = (d_z1 * x1, d_z1 * x2, d_z1)
row2 = (d_z2 * x1, d_z2 * x2, d_z2)
print(row1, row2)
assert row1 == (1, 1, 1)
assert row2 == (-1, -1, -1)
```

## Pitfall — A zero residual on the other three gates is not a zero hidden value

Gates 10 and 01 have hidden value $h_1 = 1$. Their contribution to $\partial L/\partial v_1$ is $\text{residual}\times h_1 = 0\times 1 = 0$. The hidden unit is awake. The loss signal in front of it is zero. Both facts are true, and only the product enters the weight.


## Summary

On gate 11 the output gradient is $(2, 1)$ for $(v_1, v_2)$, and $1$ for the bias. The first hidden unit receives $+1$ and passes it to both of its input weights. The second receives $-1$ and passes that. The other three gates do not change this batch gradient.
