# 01 A Layer Is Several Scores

| | |
|---|---|
| **Path** | Multilayer perceptron |
| **Prerequisite** | The perceptron path. One score $w_1 x_1 + w_2 x_2 + b$ is already familiar. |

A perceptron keeps one score. A **layer** keeps several scores at once, each with its own weights. Two scores, written as rows, are a matrix.

$$
z_1 = x_1 + x_2
\qquad
z_2 = x_1 + x_2 - 1
$$

In matrix form, with a bias vector,

$$
\begin{aligned}
W &= \begin{pmatrix} 1 & 1 \\ 1 & 1 \end{pmatrix},
&
b &= \begin{pmatrix} 0 \\ -1 \end{pmatrix},
&
z &= Wx + b
\end{aligned}
$$

The four gates in this path are

| Gate | $(x_1, x_2)$ | Target |
|---|---|---|
| 00 | $(0, 0)$ | $0$ |
| 10 | $(1, 0)$ | $1$ |
| 01 | $(0, 1)$ | $1$ |
| 11 | $(1, 1)$ | $0$ |

Opposite corners share a target. Path 11 already proved that one straight score cannot separate them. A layer does not separate them either, yet. It only prepares two numbers.

## How this path teaches

1. State the rule and work the numbers by hand.
2. A code cell repeats those numbers.
3. The printed value and the hand value have to agree.
4. One pitfall, then the sentence worth keeping.


## Example 1 — All four gates, by hand

$$
\begin{aligned}
(0, 0) &\mapsto (0,\,-1) \\
(1, 0) &\mapsto (1,\, 0) \\
(0, 1) &\mapsto (1,\, 0) \\
(1, 1) &\mapsto (2,\, 1)
\end{aligned}
$$

### Predict

The cell prints those four pairs.


```python
# Hand: (0, -1), (1, 0), (1, 0), (2, 1).
gates = [(0, 0), (1, 0), (0, 1), (1, 1)]
scores = []
for x1, x2 in gates:
    z1 = 1 * x1 + 1 * x2 + 0
    z2 = 1 * x1 + 1 * x2 - 1
    scores.append((z1, z2))
print(scores)
assert scores == [(0, -1), (1, 0), (1, 0), (2, 1)]
```

## Example 2 — Gate 10, one row at a time

$$
\begin{aligned}
z_1 &= 1\cdot 1 + 1\cdot 0 = 1 \\
z_2 &= 1\cdot 1 + 1\cdot 0 - 1 = 0
\end{aligned}
$$

The two rows can share the same weights on $x_1$ and $x_2$. The bias is what makes the second score different.


```python
# Gate 10. The rows agree on the inputs and disagree on the bias.
x1, x2 = 1, 0
z1 = x1 + x2
z2 = x1 + x2 - 1
print(z1, z2)
assert (z1, z2) == (1, 0)
```

## Pitfall — Two scores are not yet two answers

$z_2 = -1$ on gate 00 is a negative number, not a fail decision. A layer of raw scores is a change of coordinates. The next lessons decide whether those coordinates stay straight.


## Summary

A layer is a matrix of perceptron scores, plus a bias for each score. On the four gates these weights produce $(0,-1)$, $(1,0)$, $(1,0)$, and $(2,1)$. Nothing has been passed through a bend yet.
