**Place in the path.** A limit is the number the outputs approach, which may exist when the function itself does not
have a value there. Every limit in this lesson is computed twice: a numerical table, and SymPy's exact `limit`.

## Learning objectives

By the end of this lesson you can:

- estimate a two-sided limit with a table of approaching inputs
- compute the same limit exactly with SymPy
- distinguish a hole from a vertical asymptote
- compute a one-sided limit and a limit at infinity
- recognize an indeterminate form instead of plugging in and crashing


## 1. A limit you could have evaluated by substitution

**See.** \(f(x) = x^2 + 2x + 1 = (x+1)^2\). As \(x\) approaches 2, the outputs approach 9.
Here \(f(2)\) itself is 9. The limit agrees with the value. This is the calm case, and it is worth seeing once
so the exciting cases are visible as exceptions.

**Name.**

\[
\lim_{x \to 2} (x^2 + 2x + 1) = 9
\]

`sp.limit(expression, x, point)` is the exact one-liner. The table is the second method.


```python
import sympy as sp

x = sp.symbols("x")
f = x ** 2 + 2 * x + 1
exact = sp.limit(f, x, 2)
print("exact limit:", exact)
print("value f(2):", f.subs(x, 2))
assert exact == 9
assert f.subs(x, 2) == 9

print("numerical approach")
for step in (0.1, 0.01, 0.001):
    left = f.subs(x, 2 - step).evalf()
    right = f.subs(x, 2 + step).evalf()
    print(f"h = {step:<6}  f(2-h) = {left:.6f}  f(2+h) = {right:.6f}")
```

## 2. A hole: the interesting case

**See.**

\[
\frac{x^2 - 1}{x - 1}
\]

At \(x = 1\) the denominator is 0 and the numerator is 0. Substitution does not produce a number.
For every other \(x\), this expression equals \(x+1\). The graph is the line \(y = x+1\) with the point \((1, 2)\) removed.

**Name.** The form \(0/0\) is indeterminate. It is not "undefined, so the limit fails."
It is a request to rewrite. Factor: \(x^2 - 1 = (x-1)(x+1)\).


```python
import sympy as sp

x = sp.symbols("x")
f = (x ** 2 - 1) / (x - 1)
print("direct substitution:", f.subs(x, 1))
print("factored, away from x = 1:", sp.simplify(f))

print("table around the hole")
for step in (0.1, 0.01, 0.001, 0.0001):
    left = f.subs(x, 1 - step).evalf()
    right = f.subs(x, 1 + step).evalf()
    print(f"h = {step:<8}  left = {left:.6f}  right = {right:.6f}")

exact = sp.limit(f, x, 1)
print("exact limit:", exact)
assert exact == 2
```

**Read the output.** SymPy's substitution at the hole returns `nan`, not a limit.
The table closes in on 2 from both sides. The exact limit is 2.
The function still has no value at \(x = 1\). A limit does not fill the hole back into the function.
It names the height the hole sits at.

**Check a second way.** `sp.limit` and the rewritten function \(x+1\) agree at \(x = 1\).


```python
import sympy as sp

x = sp.symbols("x")
f = (x ** 2 - 1) / (x - 1)
assert sp.limit(f, x, 1) == sp.simplify(f).subs(x, 1)
print("rewrite and limit agree")
```

### Own it

\[
\lim_{x \to 0} \frac{\sin x}{x}
\]

Build a table. Direct substitution is \(0/0\). The exact limit is a famous one. Predict it, then run.


```python
import sympy as sp

x = sp.symbols("x")
f = sp.sin(x) / x
for step in (0.1, 0.01, 0.001):
    print(step, f.subs(x, step).evalf())
exact = sp.limit(f, x, 0)
print("exact:", exact)
assert exact == 1
print("checkpoint passed")
```

## 3. One side is not the other side

**See.** \(1/x\) grows without bound as \(x\) approaches 0, but the two sides go to opposite infinities.
There is no single two-sided limit in the real numbers.

**Name.** `sp.limit(f, x, 0, dir="+")` and `dir="-"` are the one-sided limits.
SymPy writes positive infinity as `oo`.

The default call `sp.limit(f, x, 0)` does **not** vote between the two sides.
Its default direction is `+`, so it repeats the right-hand limit.
A two-sided limit exists only when you have computed both sides and they are equal.
Here they are \(+\infty\) and \(-\infty\), so the two-sided limit does not exist as a single extended real number.


```python
import sympy as sp

x = sp.symbols("x")
f = 1 / x
right = sp.limit(f, x, 0, dir="+")
left = sp.limit(f, x, 0, dir="-")
default = sp.limit(f, x, 0)
print("from the right:", right)
print("from the left:", left)
print("default call, which approaches from the right:", default)
assert right == sp.oo
assert left == -sp.oo
assert default == right
assert left != right
print("the two sides disagree, so there is no two-sided limit")
```

**Read the output.** The right-hand limit is \(+\infty\) and the left-hand limit is \(-\infty\).
They disagree, so there is no two-sided limit.
SymPy's default `limit` call reports only the right-hand result.
Asking for one side at a time is the honest procedure.

On a graph, this is the vertical asymptote of lesson 005, stated as a limit.


## 4. Limits at infinity

**See.**

\[
\frac{2x + 1}{x - 3}
\]

For large \(x\), this behaves like \(2x / x = 2\). The horizontal asymptote is \(y = 2\).

**Do.** Divide numerator and denominator by \(x\) in your head, then confirm with SymPy from both \(+\infty\) and \(-\infty\).


```python
import sympy as sp

x = sp.symbols("x")
f = (2 * x + 1) / (x - 3)
print("as x -> +oo:", sp.limit(f, x, sp.oo))
print("as x -> -oo:", sp.limit(f, x, -sp.oo))
assert sp.limit(f, x, sp.oo) == 2
assert sp.limit(f, x, -sp.oo) == 2

for value in (10, 100, 1000):
    print(value, f.subs(x, value).evalf())
```

## 5. Another indeterminate form, checked numerically

**Do.**

\[
\lim_{x \to 0} \frac{1 - \cos x}{x^2} = \frac{1}{2}
\]

The table should approach `0.5`. The exact result should be the rational `1/2`, not a float.


```python
import sympy as sp

x = sp.symbols("x")
f = (1 - sp.cos(x)) / x ** 2
exact = sp.limit(f, x, 0)
print("exact:", exact)
assert exact == sp.Rational(1, 2)

for step in (0.1, 0.01, 0.001):
    value = f.subs(x, step).evalf()
    print(f"h = {step:<6} value = {value:.8f}")
    assert abs(value - sp.Rational(1, 2)) < 0.01
print("table and exact result agree")
```

## Workshop — three limits, two methods each

For each limit, print a three-row table and the exact value.

1. \(\lim_{x \to 2}(x^2 + 2x + 1)\), the calm case. Exact value 9.
2. \(\lim_{x \to 1}(x^2 - 1)/(x - 1)\), the hole. Exact value 2.
3. \(\lim_{x \to 0}(1 - \cos x)/x^2\). Exact value \(1/2\).

A helper keeps the two methods from drifting apart.


```python
import sympy as sp

x = sp.symbols("x")

def two_ways(expression, point, expected):
    exact = sp.limit(expression, x, point)
    print("expression", expression, "->", exact)
    assert sp.simplify(exact - expected) == 0
    for step in (1e-2, 1e-3, 1e-4):
        left = expression.subs(x, point - step).evalf()
        right = expression.subs(x, point + step).evalf()
        print(f"  h = {step:<8} left = {left:.8f} right = {right:.8f}")
    # Only the finest row should already be close. A coarse row is allowed to sit farther away.
    finest = expression.subs(x, point + sp.Rational(1, 10000)).evalf()
    assert abs(finest - expected) < 1e-3
    print("  finest sample agrees with the exact limit")

two_ways(x ** 2 + 2 * x + 1, 2, 9)
two_ways((x ** 2 - 1) / (x - 1), 1, 2)
two_ways((1 - sp.cos(x)) / x ** 2, 0, sp.Rational(1, 2))
print("workshop passed")
```

## Mistakes this lesson is meant to prevent

- Treating `subs` at a hole as the limit. `nan` is the function value failing, not the limit failing.
- Announcing a two-sided limit when the one-sided limits disagree.
- Reading a vertical asymptote as "the limit is a large float." It is infinity, or it does not exist.
- Cancelling a factor and then using the simplified formula at the hole as if the original function were defined there.
