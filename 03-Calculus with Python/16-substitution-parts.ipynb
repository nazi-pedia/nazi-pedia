An indefinite integral is a family of functions whose derivative is the integrand.

$$
\int f(x)\,dx = F(x)+C \iff F'(x)=f(x).
$$

SymPy returns one member of the family and does not write $C$. The right test is to differentiate the answer, not to compare the appearance of two formulas.

**Substitution.** If $u=g(x)$ and $du=g'(x)\,dx$ is visible inside the integral, the integral moves to the variable $u$.

**Integration by parts** is the product rule read from the right.

$$
\int u\,dv = uv-\int v\,du.
$$

Choose $u$ to be a factor that becomes simpler when it is differentiated.


## Example 1 — An obvious substitution

### Hand solution

$$
\int 2x\cos(x^2)\,dx.
$$

$$
u=x^2, \qquad du=2x\,dx.
$$

$$
\begin{aligned}
\int 2x\cos(x^2)\,dx &= \int \cos u\,du \\
&= \sin u+C \\
&= \sin(x^2)+C.
\end{aligned}
$$

The chain rule returns the integrand:

$$
\dfrac{d}{dx}\sin(x^2)=2x\cos(x^2).
$$

### Python recheck


```python
import sympy as sp

# Substitution u = x^2, checked by differentiating the answer.
x = sp.symbols("x")
found = sp.integrate(2*x*sp.cos(x**2), x)
hand = sp.sin(x**2)
print(found)
assert sp.simplify(sp.diff(found - hand, x)) == 0
print("Confirmed")
```

## Example 2 — Integration by parts

### Hand solution

$$
\int x e^{x}\,dx.
$$

$$
u=x, \quad dv=e^{x}\,dx, \quad du=dx, \quad v=e^{x}.
$$

$$
\begin{aligned}
\int x e^{x}\,dx &= x e^{x}-\int e^{x}\,dx \\
&= e^{x}(x-1)+C.
\end{aligned}
$$

Differentiating the answer returns the integrand:

$$
\dfrac{d}{dx}\bigl[e^{x}(x-1)\bigr]=e^{x}(x-1)+e^{x}=xe^{x}.
$$

### Python recheck


```python
import sympy as sp

# Parts with u = x, checked by differentiation.
x = sp.symbols("x")
found = sp.integrate(x*sp.exp(x), x)
hand = sp.exp(x)*(x - 1)
print(sp.simplify(found))
assert sp.simplify(sp.diff(found - hand, x)) == 0
print("Confirmed")
```

## Example 3 — Take the logarithm as $u$

### Hand solution

$$
\int x\ln x\,dx,\qquad x>0.
$$

Set $u=\ln x$ and $dv=x\,dx$. Then

$$
du=\dfrac{dx}{x}, \qquad v=\dfrac{x^2}{2}.
$$

$$
\begin{aligned}
\int x\ln x\,dx &= \dfrac{x^2}{2}\ln x-\int \dfrac{x^2}{2}\cdot\dfrac{1}{x}\,dx \\
&= \dfrac{x^2}{2}\ln x-\int \dfrac{x}{2}\,dx \\
&= \dfrac{x^2}{2}\ln x-\dfrac{x^2}{4}+C.
\end{aligned}
$$

### Python recheck


```python
import sympy as sp

# Parts with u = ln x on x > 0.
x = sp.symbols("x", positive=True)
found = sp.integrate(x*sp.log(x), x)
hand = (x**2/2)*sp.log(x) - x**2/4
print(sp.simplify(found))
assert sp.simplify(sp.diff(found - hand, x)) == 0
print("Confirmed")
```

## Example 4 — The same antiderivative, now with limits

### Hand solution

$$
\begin{aligned}
\int_0^1 x e^{x}\,dx &= \Bigl[e^{x}(x-1)\Bigr]_0^1 \\
&= (0)-e^{0}(0-1) \\
&= 1.
\end{aligned}
$$

The constant $C$ cancels in the difference $F(b)-F(a)$. A definite integral is a number.

### Python recheck


```python
import sympy as sp

# Definite integral, and the same value from the antiderivative.
x = sp.symbols("x")
exact = sp.integrate(x*sp.exp(x), (x, 0, 1))
by_ftc = sp.exp(1)*(1 - 1) - sp.exp(0)*(0 - 1)
print(exact, by_ftc)
assert exact == 1
assert sp.simplify(by_ftc - 1) == 0
print("Confirmed")
```

## Example 5 — Substitution in a definite integral; the limits change too

### Hand solution

$$
\int_0^{\pi/2}\sin x\cos x\,dx.
$$

Set $u=\sin x$. Then $du=\cos x\,dx$, and the limits change with $x$:

$$
x=0 \implies u=0, \qquad x=\dfrac{\pi}{2} \implies u=1.
$$

$$
\int_0^1 u\,du=\dfrac{1}{2}.
$$

Leave the limits in $x$ and you evaluate the antiderivative in the wrong variable.

### Python recheck


```python
import sympy as sp

# Definite integral whose substitution changes the limits to 0 and 1.
x = sp.symbols("x")
exact = sp.integrate(sp.sin(x)*sp.cos(x), (x, 0, sp.pi/2))
print(exact)
assert exact == sp.Rational(1, 2)
print("Confirmed")
```

## A pitfall

In integration by parts, a choice of $u$ that raises the algebraic degree makes the integral worse. For

$$
\int x e^{x}\,dx,
$$

take the polynomial as $u$, and leave the exponential in $dv$.

## Summary

Substitution reverses a chain rule. Integration by parts reverses a product. In both, the derivative of the answer must give back the integrand.
