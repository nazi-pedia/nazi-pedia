$$
\begin{aligned}
 \nabla f &= \left(\dfrac{\partial f}{\partial x},\dfrac{\partial f}{\partial y}\right).
\end{aligned}
$$

The directional derivative in the direction of a unit vector $\mathbf{u}$ is

$$
\begin{aligned}
 D_{\mathbf{u}}f &= \nabla f\cdot\mathbf{u}.
\end{aligned}
$$

The greatest value of this derivative is $|\nabla f|$, and it occurs in the direction of the gradient itself.
The gradient is perpendicular to a level curve $f=\text{constant}$.

The tangent plane to $z=f(x,y)$ at $(a,b)$ is

$$
\begin{aligned}
 z
 &=
 f(a,b)
 +\dfrac{\partial f}{\partial x}(a,b)\,(x-a)
 +\dfrac{\partial f}{\partial y}(a,b)\,(y-b).
\end{aligned}
$$

This is the linearization of lesson 11, now in two variables.


## Example 1 — Direction of steepest ascent

### Hand solution

For $f=x^{2}+y^{2}$,

$$
\begin{aligned}
 \nabla f
 &=
 \begin{pmatrix} 2x \\ 2y \end{pmatrix}.
\end{aligned}
$$

At $(1,1)$,

$$
\begin{aligned}
 \nabla f(1,1)
 &=
 \begin{pmatrix} 2 \\ 2 \end{pmatrix},
 \qquad
 |\nabla f|
 =
 \sqrt{8}
 =
 2\sqrt{2},
 \\
 D_{\mathbf{i}}f
 &=
 \begin{pmatrix} 2 \\ 2 \end{pmatrix}
 \cdot
 \begin{pmatrix} 1 \\ 0 \end{pmatrix}
 =
 2.
\end{aligned}
$$

The directional derivative along $\mathbf{i}$ is smaller than $2\sqrt{2}$, because $\mathbf{i}$ is not the direction of the gradient.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# The paraboloid f = x^2 + y^2.
x, y = sp.symbols("x y")
f = x**2 + y**2
# Gradient at (1, 1), then the directional derivative along i = (1, 0).
grad = sp.Matrix([sp.diff(f, x), sp.diff(f, y)]).subs({x: 1, y: 1})
directional = grad.dot(sp.Matrix([1, 0]))
print(grad, grad.norm(), directional)
assert grad == sp.Matrix([2, 2])
assert sp.simplify(grad.norm() - 2*sp.sqrt(2)) == 0
assert directional == 2
print("Confirmed")
```

## Example 2 — Tangent plane to a paraboloid

### Hand solution

The surface is $z=x^{2}+y^{2}$ at the point $(1,1,2)$.

$$
\begin{aligned}
 \dfrac{\partial z}{\partial x} &= 2x=2,\\
 \dfrac{\partial z}{\partial y} &= 2,\\
 z-2 &= 2(x-1)+2(y-1),\\
 z &= 2x+2y-2.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# The same paraboloid, expanded about (1, 1).
x, y = sp.symbols("x y")
f = x**2 + y**2
plane = f.subs({x: 1, y: 1}) + sp.diff(f, x).subs({x: 1, y: 1})*(x - 1) + sp.diff(f, y).subs({x: 1, y: 1})*(y - 1)
print(sp.expand(plane))
# The tangent plane simplifies to z = 2x + 2y - 2.
assert sp.expand(plane) == 2*x + 2*y - 2
print("Confirmed")
```

## Example 3 — Linear approximation

### Hand solution

Take $f=e^{x}\cos y$ at $(0,0)$.

$$
\begin{aligned}
 f(0,0) &= 1,\\
 \dfrac{\partial f}{\partial x} &= e^{x}\cos y,\\
 \dfrac{\partial f}{\partial x}(0,0) &= 1,\\
 \dfrac{\partial f}{\partial y} &= -e^{x}\sin y,\\
 \dfrac{\partial f}{\partial y}(0,0) &= 0,\\
 L(x,y) &= 1+x.
\end{aligned}
$$

Therefore $f(0.1,-0.1)\approx 1.1$.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# f = e^x cos y and its linear approximation L = 1 + x.
x, y = sp.symbols("x y")
f = sp.exp(x)*sp.cos(y)
L = 1 + x
# Compare L and f at (1/10, -1/10).
true = f.subs({x: sp.Rational(1, 10), y: -sp.Rational(1, 10)})
approx = L.subs({x: sp.Rational(1, 10), y: -sp.Rational(1, 10)})
print(approx, sp.N(true, 6))
assert approx == sp.Rational(11, 10)
assert abs(true - approx) < 10**(-2)
print("Confirmed")
```

## Example 4 — The gradient is normal to a level curve

### Hand solution

The level set $f=x^{2}+y^{2}=1$ is a circle. At $(1,0)$,

$$
\begin{aligned}
 \nabla f(1,0)
 &=
 \begin{pmatrix} 2 \\ 0 \end{pmatrix}.
\end{aligned}
$$

This vector is horizontal, so it is perpendicular to the vertical tangent line of the circle at that point.
The slope of the tangent to the circle at $(1,0)$ is infinite. That vector is a radius, and a radius is perpendicular to the tangent.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Gradient of f = x^2 + y^2, a radius, and a tangent vector at (1, 0).
x, y = sp.symbols("x y")
grad = sp.Matrix([sp.diff(x**2 + y**2, x), sp.diff(x**2 + y**2, y)])
radius = sp.Matrix([1, 0])
tangent = sp.Matrix([0, 1])
# The gradient is orthogonal to the tangent and parallel to the radius.
assert grad.subs({x: 1, y: 0}).dot(tangent) == 0
assert grad.subs({x: 1, y: 0}).dot(radius) == 2
print("The gradient is perpendicular to the tangent")
print("Confirmed")
```

## A pitfall

The direction vector in a directional derivative must be a unit vector.
If $\mathbf{v}$ is not a unit vector, first build

$$
\begin{aligned}
 \mathbf{u} &= \dfrac{\mathbf{v}}{|\mathbf{v}|}.
\end{aligned}
$$

Otherwise the directional derivative has been scaled by the length of $\mathbf{v}$.

## Summary

The gradient carries both the direction and the magnitude of the steepest increase.
The tangent plane is the tangent-line formula with one extra variable.
