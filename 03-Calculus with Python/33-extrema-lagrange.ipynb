A critical point is a point where $\nabla f=\mathbf{0}$, or where the gradient is not defined.

The second-derivative test uses the Hessian determinant

$$
\begin{aligned}
 H
 &=
 \dfrac{\partial^{2} f}{\partial x^{2}}\dfrac{\partial^{2} f}{\partial y^{2}}
 -
 \left(\dfrac{\partial^{2} f}{\partial y\,\partial x}\right)^{2}.
\end{aligned}
$$

If $H>0$ and $\dfrac{\partial^{2} f}{\partial x^{2}}>0$, the point is a local minimum.
If $H>0$ and $\dfrac{\partial^{2} f}{\partial x^{2}}<0$, the point is a local maximum.
If $H<0$, the point is a saddle.
If $H=0$, the test says nothing.

On a constraint $g=c$, instead of eliminating a variable, solve

$$
\begin{aligned}
 \nabla f &= \lambda\nabla g
\end{aligned}
$$

together with the constraint itself. The multiplier $\lambda$ is the shadow price: the rate at which the optimum changes when the constraint is relaxed.


## Example 1 — A definite minimum

### Hand solution

$$
\begin{aligned}
 f &= x^{2}+y^{2}-xy,\\
 \dfrac{\partial f}{\partial x} &= 2x-y=0,\\
 \dfrac{\partial f}{\partial y} &= 2y-x=0,\\
 y &= 2x,\\
 4x-x &= 0,\\
 x &= 0,\quad y=0.
\end{aligned}
$$

$$
\begin{aligned}
 \dfrac{\partial^{2} f}{\partial x^{2}} &= 2,\\
 \dfrac{\partial^{2} f}{\partial y^{2}} &= 2,\\
 \dfrac{\partial^{2} f}{\partial y\,\partial x} &= -1,\\
 H &= 4-1=3>0.
\end{aligned}
$$

Since $\dfrac{\partial^{2} f}{\partial x^{2}}>0$, the origin is a local minimum, and the value there is $0$. Because

$$
\begin{aligned}
 f &= \dfrac{1}{2}\bigl(x^{2}+y^{2}+(x-y)^{2}\bigr)
\end{aligned}
$$

is nonnegative, this local minimum is also absolute.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# f = x^2 + y^2 - xy.
x, y = sp.symbols("x y", real=True)
f = x**2 + y**2 - x*y
# The only critical point, and the Hessian determinant.
crit = sp.solve([sp.diff(f, x), sp.diff(f, y)], [x, y], dict=True)
H = sp.diff(f, x, 2)*sp.diff(f, y, 2) - sp.diff(f, x, y)**2
print(crit, H, f.subs(crit[0]))
assert crit == [{x: 0, y: 0}]
assert H == 3
print("Confirmed")
```

## Example 2 — A saddle point

### Hand solution

Take $f=x^{2}-y^{2}$. The equations $\dfrac{\partial f}{\partial x}=2x=0$ and $\dfrac{\partial f}{\partial y}=-2y=0$ hold only at the origin.
It is a mistake to write $H=(-2)(-2)$.

$$
\begin{aligned}
 \dfrac{\partial^{2} f}{\partial x^{2}} &= 2,\\
 \dfrac{\partial^{2} f}{\partial y^{2}} &= -2,\\
 \dfrac{\partial^{2} f}{\partial y\,\partial x} &= 0,\\
 H &= -4<0.
\end{aligned}
$$

The origin is a saddle. Along the $x$-axis the graph looks like a local minimum, and along the $y$-axis it looks like a local maximum. Neither behavior is stable in the plane.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# The saddle f = x^2 - y^2.
x, y = sp.symbols("x y")
f = x**2 - y**2
# Hessian determinant, the value at the origin, and the values on the two axes.
H = sp.diff(f, x, 2)*sp.diff(f, y, 2) - sp.diff(f, x, y)**2
print(H, f.subs({x: 0, y: 0}), f.subs({x: 1, y: 0}), f.subs({x: 0, y: 1}))
assert H == -4
print("Confirmed")
```

## Example 3 — Lagrange multipliers on a circle

### Hand solution

Maximize $f=x+y$ on $x^{2}+y^{2}=1$.

$$
\begin{aligned}
 \begin{pmatrix} 1 \\ 1 \end{pmatrix}
 &=
 \lambda
 \begin{pmatrix} 2x \\ 2y \end{pmatrix},
 \qquad
 x^{2}+y^{2}=1.
\end{aligned}
$$

Hence $x=y$ and $2x^{2}=1$, so $x=\pm\dfrac{1}{\sqrt{2}}$.

$$
\begin{aligned}
 f\left(\dfrac{1}{\sqrt{2}},\dfrac{1}{\sqrt{2}}\right)
 &=
 \sqrt{2},
 \\
 f\left(-\dfrac{1}{\sqrt{2}},-\dfrac{1}{\sqrt{2}}\right)
 &=
 -\sqrt{2}.
\end{aligned}
$$

The maximum is $\sqrt{2}$ and the minimum is $-\sqrt{2}$.
Geometrically, the level curves $x+y=c$ are lines, and the last contact with the circle lies along the diagonal $y=x$.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Lagrange equations for f = x + y on the unit circle.
x, y, lam = sp.symbols("x y lam", real=True)
eqs = [1 - lam*2*x, 1 - lam*2*y, x**2 + y**2 - 1]
sols = sp.solve(eqs, [x, y, lam], dict=True)
# The extreme values of x + y.
values = sorted(sp.simplify(sol[x] + sol[y]) for sol in sols)
print(values)
assert values[0] == -sp.sqrt(2)
assert values[-1] == sp.sqrt(2)
print("Confirmed")
```

## Example 4 — A linear constraint, elimination versus Lagrange

### Hand solution

Maximize $f=xy$ on $x+y=4$, with $x>0$ and $y>0$.
Elimination gives

$$
\begin{aligned}
 y &= 4-x,\\
 A &= x(4-x),\\
 A' &= 4-2x=0,\\
 x &= 2,\quad y=2,\\
 A &= 4.
\end{aligned}
$$

Lagrange multipliers give

$$
\begin{aligned}
 \begin{pmatrix} y \\ x \end{pmatrix}
 &=
 \lambda
 \begin{pmatrix} 1 \\ 1 \end{pmatrix},
\end{aligned}
$$

so $x=y$, and the constraint forces $x=y=2$. The same answer.
When the constraint is linear and simple, both routes are short. When the constraint is a circle or an ellipse, Lagrange multipliers spare you from eliminating through a radical.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Elimination: A(x) = x(4 - x).
x = sp.symbols("x", real=True)
A = x*(4 - x)
crit = sp.solve(sp.diff(A, x), x)[0]
print(crit, A.subs(x, crit))
assert crit == 2 and A.subs(x, crit) == 4
print("Confirmed")
```

## A pitfall

Do not read $H=0$ the way you read $f''=0$ in one variable. The test is not finished.
On a closed region, compare the interior critical points with the whole boundary. Lagrange multipliers see only the boundary.

## Summary

$\nabla f=\mathbf{0}$ produces the interior candidates.
The Hessian classifies a candidate, unless the Hessian determinant is zero.
On a constraint, the gradient of the objective and the gradient of the constraint are parallel.
