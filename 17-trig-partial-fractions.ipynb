توان فرد سینوس یا کسینوس را کنار بگذارید و با اتحاد فیثاغورس بقیه را بر حسب عامل باقی‌مانده بنویسید.

\(\sqrt{a^2-x^2}\) جانشینی \(x=a\sin\theta\) می‌خواهد.
\(\sqrt{a^2+x^2}\) جانشینی \(x=a\tan\theta\) می‌خواهد.
\(x^2+a^2\) در مخرج، اغلب همان آرک‌تانژانت است.

کسر گویای سره را به جمع کسرهایی با مخرج خطی یا درجهٔ دو بشکنید.
اگر صورت درجهٔ بالاتر دارد، اول تقسیم کنید.



## مثال ۱ — توان فرد

### حل ریاضی

\[
\int\sin^3 x\,dx=\int(1-\cos^2 x)\sin x\,dx
\]

\(u=\cos x\)، \(du=-\sin x\,dx\).

\[
\int -(1-u^2)\,du=-u+\frac{u^3}{3}+C=-\cos x+\frac{\cos^3 x}{3}+C
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
found = sp.integrate(sp.sin(x)**3, x)
hand = -sp.cos(x) + sp.cos(x)**3/3
print(sp.simplify(found))
assert sp.simplify(sp.diff(found - hand, x)) == 0
print("تایید شد")

```

## مثال ۲ — جانشینی مثلثاتی که به آرک‌سینوس ختم می‌شود

### حل ریاضی

\[
\int\frac{1}{\sqrt{1-x^2}}\,dx
\]

\(x=\sin\theta\)، \(dx=\cos\theta\,d\theta\)، و \(\sqrt{1-x^2}=\cos\theta\) روی \(|\theta|<\pi/2\).

\[
\int\frac{\cos\theta}{\cos\theta}\,d\theta=\theta+C=\arcsin x+C
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
found = sp.integrate(1/sp.sqrt(1 - x**2), x)
print(found)
assert sp.simplify(sp.diff(found - sp.asin(x), x)) == 0
print("تایید شد")

```

## مثال ۳ — کسر جزئی با ریشه‌های ساده

### حل ریاضی

\[
\frac{x+5}{(x-1)(x+2)}=\frac{A}{x-1}+\frac{B}{x+2}
\]

\[
A(x+2)+B(x-1)=x+5
\]

\(x=1\): \(3A=6\)، پس \(A=2\).
\(x=-2\): \(-3B=3\)، پس \(B=-1\).

\[
\int\left(\frac{2}{x-1}-\frac{1}{x+2}\right)dx
=2\ln|x-1|-\ln|x+2|+C
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
integrand = (x + 5)/((x - 1)*(x + 2))
hand_derivative = 2/(x - 1) - 1/(x + 2)
print(sp.apart(integrand, x))
assert sp.simplify(sp.apart(integrand) - hand_derivative) == 0
found = sp.integrate(integrand, x)
hand = 2*sp.log(x - 1) - sp.log(x + 2)
assert sp.simplify(sp.diff(found - hand, x)) == 0
assert sp.simplify(sp.diff(hand, x) - hand_derivative) == 0
print(found)
print("تایید شد")

```

## مثال ۴ — مخرج تحویل‌ناپذیر

### حل ریاضی

\[
\int\frac{1}{x^2+1}\,dx=\arctan x+C
\]

اگر جانشینی \(x=\tan\theta\) را تا آخر بروید، \(\theta\) همان آرک‌تانژانت است.
حالت نزدیک:

\[
\int\frac{1}{x^2+4}\,dx=\int\frac{1}{4\bigl((x/2)^2+1\bigr)}\,dx
=\frac{1}{2}\arctan\frac{x}{2}+C
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
first = sp.integrate(1/(x**2 + 1), x)
second = sp.integrate(1/(x**2 + 4), x)
print(first)
print(second)
assert sp.simplify(sp.diff(first - sp.atan(x), x)) == 0
assert sp.simplify(sp.diff(second - sp.Rational(1, 2)*sp.atan(x/2), x)) == 0
print("تایید شد")

```

## مثال ۵ — اول تقسیم، بعد کسر جزئی

### حل ریاضی

\[
\frac{x^2}{x-1}=x+1+\frac{1}{x-1}
\]

چون درجهٔ صورت کمتر نیست. تقسیم:

\[
x^2=(x+1)(x-1)+1
\]

\[
\int\left(x+1+\frac{1}{x-1}\right)dx=\frac{x^2}{2}+x+\ln|x-1|+C
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
integrand = x**2/(x - 1)
print(sp.apart(integrand))
found = sp.integrate(integrand, x)
hand = x**2/2 + x + sp.log(x - 1)
assert sp.apart(integrand) == x + 1 + 1/(x - 1)
assert sp.simplify(sp.diff(found - hand, x)) == 0
print("تایید شد")

```

## نکته

پوشش \(A/(x-r)+B/(x-s)\) فقط برای عامل‌های خطیِ تکرارنشده است.
عامل \((x-r)^2\) دو خانه می‌خواهد: \(A/(x-r)+B/(x-r)^2\).
