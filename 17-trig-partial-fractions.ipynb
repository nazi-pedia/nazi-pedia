Set an odd power of sine or cosine aside, and use the Pythagorean identity to write the rest in terms of the remaining factor.

$$
\begin{aligned}
\sqrt{a^2-x^2} &\quad\text{calls for}\quad x=a\sin\theta, \\
\sqrt{a^2+x^2} &\quad\text{calls for}\quad x=a\tan\theta.
\end{aligned}
$$

A factor $x^2+a^2$ in the denominator is often an arctangent.

Split a proper rational function into a sum of fractions with linear or quadratic denominators. If the numerator has degree at least as large as the denominator, divide first.


## Example 1 — An odd power

### Hand solution

$$
\int\sin^3 x\,dx=\int(1-\cos^2 x)\sin x\,dx.
$$

Set $u=\cos x$, so $du=-\sin x\,dx$.

$$
\begin{aligned}
\int -(1-u^2)\,du &= -u+\dfrac{u^3}{3}+C \\
&= -\cos x+\dfrac{\cos^3 x}{3}+C.
\end{aligned}
$$

### Python recheck


```python
import sympy as sp

# Odd power of sine, checked against the cosine antiderivative.
x = sp.symbols("x")
found = sp.integrate(sp.sin(x)**3, x)
hand = -sp.cos(x) + sp.cos(x)**3/3
print(sp.simplify(found))
assert sp.simplify(sp.diff(found - hand, x)) == 0
print("Confirmed")
```

## Example 2 — A trigonometric substitution that ends at arcsine

### Hand solution

$$
\int\dfrac{1}{\sqrt{1-x^2}}\,dx.
$$

Set $x=\sin\theta$. Then, for $|\theta|<\dfrac{\pi}{2}$,

$$
dx=\cos\theta\,d\theta, \qquad \sqrt{1-x^2}=\cos\theta.
$$

$$
\int\dfrac{\cos\theta}{\cos\theta}\,d\theta=\theta+C=\arcsin x+C.
$$

### Python recheck


```python
import sympy as sp

# The standard arcsine integral.
x = sp.symbols("x")
found = sp.integrate(1/sp.sqrt(1 - x**2), x)
print(found)
assert sp.simplify(sp.diff(found - sp.asin(x), x)) == 0
print("Confirmed")
```

## Example 3 — Partial fractions with simple roots

### Hand solution

$$
\dfrac{x+5}{(x-1)(x+2)}=\dfrac{A}{x-1}+\dfrac{B}{x+2}.
$$

$$
A(x+2)+B(x-1)=x+5.
$$

$$
\begin{aligned}
x=1 &\implies 3A=6 \implies A=2, \\
x=-2 &\implies -3B=3 \implies B=-1.
\end{aligned}
$$

$$
\begin{aligned}
\int\left(\dfrac{2}{x-1}-\dfrac{1}{x+2}\right)\,dx
&= 2\ln|x-1|-\ln|x+2|+C.
\end{aligned}
$$

### Python recheck


```python
import sympy as sp

# Cover-up coefficients, then the antiderivative.
x = sp.symbols("x")
integrand = (x + 5)/((x - 1)*(x + 2))
hand_derivative = 2/(x - 1) - 1/(x + 2)
print(sp.apart(integrand, x))
assert sp.simplify(sp.apart(integrand) - hand_derivative) == 0
found = sp.integrate(integrand, x)
hand = 2*sp.log(x - 1) - sp.log(x + 2)
assert sp.simplify(sp.diff(found - hand, x)) == 0
assert sp.simplify(sp.diff(hand, x) - hand_derivative) == 0
print(found)
print("Confirmed")
```

## Example 4 — An irreducible denominator

### Hand solution

$$
\int\dfrac{1}{x^2+1}\,dx=\arctan x+C.
$$

Carry the substitution $x=\tan\theta$ all the way through, and $\theta$ is the arctangent. A nearby integral is

$$
\begin{aligned}
\int\dfrac{1}{x^2+4}\,dx &= \int\dfrac{1}{4\left(\left(\dfrac{x}{2}\right)^2+1\right)}\,dx \\
&= \dfrac{1}{2}\arctan\dfrac{x}{2}+C.
\end{aligned}
$$

### Python recheck


```python
import sympy as sp

# Arctangent integrals with a^2 = 1 and a^2 = 4.
x = sp.symbols("x")
first = sp.integrate(1/(x**2 + 1), x)
second = sp.integrate(1/(x**2 + 4), x)
print(first)
print(second)
assert sp.simplify(sp.diff(first - sp.atan(x), x)) == 0
assert sp.simplify(sp.diff(second - sp.Rational(1, 2)*sp.atan(x/2), x)) == 0
print("Confirmed")
```

## Example 5 — Divide first, then partial fractions

### Hand solution

$$
\dfrac{x^2}{x-1}=x+1+\dfrac{1}{x-1},
$$

because the degree of the numerator is not smaller than the degree of the denominator. Polynomial division gives

$$
x^2=(x+1)(x-1)+1.
$$

$$
\int\left(x+1+\dfrac{1}{x-1}\right)\,dx=\dfrac{x^2}{2}+x+\ln|x-1|+C.
$$

### Python recheck


```python
import sympy as sp

# Divide first, then integrate the proper remainder.
x = sp.symbols("x")
integrand = x**2/(x - 1)
print(sp.apart(integrand))
found = sp.integrate(integrand, x)
hand = x**2/2 + x + sp.log(x - 1)
assert sp.apart(integrand) == x + 1 + 1/(x - 1)
assert sp.simplify(sp.diff(found - hand, x)) == 0
print("Confirmed")
```

## A pitfall

The splitting

$$
\dfrac{A}{x-r}+\dfrac{B}{x-s}
$$

fits unrepeated linear factors only. A repeated factor $(x-r)^2$ needs two terms:

$$
\dfrac{A}{x-r}+\dfrac{B}{(x-r)^2}.
$$

## Summary

An odd power is handled by a short substitution. A quadratic radical is handled by a trigonometric substitution, or by the known arcsine and arctangent forms. Split a rational function only after it has been made proper.
