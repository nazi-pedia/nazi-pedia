# ۳۹ · معادلهٔ خطی مرتبهٔ دوم همگن

| | |
|---|---|
| **دوره** | معادلات دیفرانسیل |
| **پیش‌نیاز** | ۳۸ |

\[
ay''+by'+cy=0
\]

معادلهٔ مشخصه \(ar^2+br+c=0\).

دو ریشهٔ حقیقی متمایز \(r_1,r_2\): \(y=C_1 e^{r_1 x}+C_2 e^{r_2 x}\).

ریشهٔ تکراری \(r\): \(y=(C_1+C_2 x)e^{rx}\).

ریشه‌های مختلط \(\alpha\pm i\beta\): \(y=e^{\alpha x}(C_1\cos\beta x+C_2\sin\beta x)\).

دو شرط اولیه لازم است، چون دو ثابت هست.



## مثال ۱ — نوسان ساده

### حل ریاضی

\(y''+y=0\)، \(y(0)=1\)، \(y'(0)=0\).
مشخصه \(r^2+1=0\)، ٰ\(r=\pm i\).

\[
y=C_1\cos x+C_2\sin x
\]

\(y(0)=C_1=1\). \(y'=-C_1\sin x+C_2\cos x\) و \(y'(0)=C_2=0\).
پس \(y=\cos x\).

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
y = sp.Function("y")
ode = sp.Eq(y(x).diff(x, 2) + y(x), 0)
sol = sp.dsolve(ode, y(x), ics={y(0): 1, y(x).diff(x).subs(x, 0): 0})
print(sol)
assert sp.simplify(sol.rhs - sp.cos(x)) == 0
print("تایید شد")

```

## مثال ۲ — دو ریشهٔ حقیقی

### حل ریاضی

\(y''-3y'+2y=0\)، \(y(0)=1\)، \(y'(0)=0\).
\((r-1)(r-2)=0\).

\[
y=C_1 e^{x}+C_2 e^{2x}
\]

\(C_1+C_2=1\). مشتق \(C_1+2C_2=0\). پس \(C_2=-1\) و \(C_1=2\).

\[
y=2e^{x}-e^{2x}
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
y = sp.Function("y")
ode = sp.Eq(y(x).diff(x, 2) - 3*y(x).diff(x) + 2*y(x), 0)
sol = sp.dsolve(ode, y(x), ics={y(0): 1, y(x).diff(x).subs(x, 0): 0})
hand = 2*sp.exp(x) - sp.exp(2*x)
print(sol)
assert sp.simplify(sol.rhs - hand) == 0
print("تایید شد")

```

## مثال ۳ — ریشهٔ تکراری

### حل ریاضی

\(y''-2y'+y=0\). مشخصه \((r-1)^2=0\).

\[
y=(C_1+C_2 x)e^{x}
\]

اگر فقط \(e^{x}\) را بنویسید، یک جواب را گم کرده‌اید. رونسکین \(e^{x}\) و \(xe^{x}\) ناصفر است، پس مستقل‌اند.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
y = sp.Function("y")
ode = sp.Eq(y(x).diff(x, 2) - 2*y(x).diff(x) + y(x), 0)
sol = sp.dsolve(ode, y(x))
print(sol)
y1 = sp.exp(x)
y2 = x*sp.exp(x)
wronskian = sp.simplify(y1*sp.diff(y2, x) - y2*sp.diff(y1, x))
print("رونسکین:", wronskian)
assert wronskian != 0
assert sp.simplify(sp.diff(y2, x, 2) - 2*sp.diff(y2, x) + y2) == 0
print("تایید شد")

```

## مثال ۴ — ریشهٔ مختلط با میرایی

### حل ریاضی

\(y''+2y'+5y=0\).
\(r^2+2r+5=0\)، \(r=-1\pm 2i\).

\[
y=e^{-x}(C_1\cos 2x+C_2\sin 2x)
\]

پوش \(e^{-x}\) دامنه را می‌خواباند و \(2\) بسامد زاویه‌ای نوسان است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
y = sp.Function("y")
ode = sp.Eq(y(x).diff(x, 2) + 2*y(x).diff(x) + 5*y(x), 0)
sol = sp.dsolve(ode, y(x))
hand = sp.exp(-x)*(sp.cos(2*x) + sp.sin(2*x))
print(sol)
assert sp.simplify((hand.diff(x, 2) + 2*hand.diff(x) + 5*hand)) == 0
print("یک جواب نمونه در معادله صدق می‌کند")
print("تایید شد")

```

## نکته

شرط \(y'(0)\) را روی مشتق جواب بگذارید، نه روی خود جواب.
دو ثابت را با یک دستگاه خطی پیدا کنید و بعد یک بار در معادله مشتق بگیرید تا دستگاه را امتحان کرده باشید.

## جمع‌بندی

مشخصه را حل کنید، شکل جواب را از نوع ریشه انتخاب کنید، بعد شرط اولیه را اعمال کنید.
