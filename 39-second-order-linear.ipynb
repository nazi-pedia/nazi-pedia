The constant-coefficient equation

$$
ay''+by'+cy=0
$$

has the characteristic equation $ar^2+br+c=0$.

Two distinct real roots $r_1$ and $r_2$ give

$$
y=C_1 e^{r_1 x}+C_2 e^{r_2 x}.
$$

A repeated real root $r$ gives

$$
y=(C_1+C_2 x)e^{rx}.
$$

A complex pair $\alpha\pm i\beta$ gives

$$
y=e^{\alpha x}(C_1\cos\beta x+C_2\sin\beta x).
$$

Two initial conditions are required, because the general solution contains two constants.


## Example 1 — Simple oscillation

### Hand solution

Solve $y''+y=0$ with $y(0)=1$ and $y'(0)=0$.
The characteristic equation is $r^2+1=0$, so $r=\pm i$.

$$
\begin{aligned}
y &= C_1\cos x+C_2\sin x, \\
y(0) &= C_1 = 1, \\
y' &= -C_1\sin x+C_2\cos x, \\
y'(0) &= C_2 = 0, \\
y &= \cos x.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Independent variable and unknown function.
x = sp.symbols("x")
y = sp.Function("y")
# Oscillator equation y'' + y = 0.
ode = sp.Eq(y(x).diff(x, 2) + y(x), 0)
# Initial conditions y(0) = 1 and y'(0) = 0.
sol = sp.dsolve(ode, y(x), ics={y(0): 1, y(x).diff(x).subs(x, 0): 0})
print(sol)
# The solution must be cos(x).
assert sp.simplify(sol.rhs - sp.cos(x)) == 0
print("Confirmed")
```

## Example 2 — Two distinct real roots

### Hand solution

Solve $y''-3y'+2y=0$ with $y(0)=1$ and $y'(0)=0$.
The characteristic polynomial factors as $(r-1)(r-2)=0$.

$$
\begin{aligned}
y &= C_1 e^{x}+C_2 e^{2x}, \\
C_1+C_2 &= 1, \\
C_1+2C_2 &= 0, \\
C_2 &= -1, \\
C_1 &= 2, \\
y &= 2e^{x}-e^{2x}.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Independent variable and unknown function.
x = sp.symbols("x")
y = sp.Function("y")
# Equation y'' - 3y' + 2y = 0.
ode = sp.Eq(y(x).diff(x, 2) - 3*y(x).diff(x) + 2*y(x), 0)
# Initial conditions y(0) = 1 and y'(0) = 0.
sol = sp.dsolve(ode, y(x), ics={y(0): 1, y(x).diff(x).subs(x, 0): 0})
# Hand solution 2*exp(x) - exp(2x).
hand = 2*sp.exp(x) - sp.exp(2*x)
print(sol)
# The solver result must match the hand solution.
assert sp.simplify(sol.rhs - hand) == 0
print("Confirmed")
```

## Example 3 — A repeated root

### Hand solution

Solve $y''-2y'+y=0$. The characteristic equation is $(r-1)^2=0$, a repeated root $r=1$.

$$
y=(C_1+C_2 x)e^{x}.
$$

The single exponential $e^{x}$ is only one solution. The second independent solution is $xe^{x}$. Their Wronskian is nonzero, so the two functions are linearly independent.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Independent variable and unknown function.
x = sp.symbols("x")
y = sp.Function("y")
# Repeated-root equation y'' - 2y' + y = 0.
ode = sp.Eq(y(x).diff(x, 2) - 2*y(x).diff(x) + y(x), 0)
# General solution, with both arbitrary constants.
sol = sp.dsolve(ode, y(x))
print(sol)
# The two basis functions exp(x) and x*exp(x).
y1 = sp.exp(x)
y2 = x*sp.exp(x)
# Wronskian of the basis.
wronskian = sp.simplify(y1*sp.diff(y2, x) - y2*sp.diff(y1, x))
print("Wronskian:", wronskian)
# A nonzero Wronskian means the two solutions are independent.
assert wronskian != 0
# The second basis function satisfies the equation.
assert sp.simplify(sp.diff(y2, x, 2) - 2*sp.diff(y2, x) + y2) == 0
print("Confirmed")
```

## Example 4 — A complex root with damping

### Hand solution

Solve $y''+2y'+5y=0$.
The characteristic equation is $r^2+2r+5=0$, with roots $r=-1\pm 2i$.

$$
y=e^{-x}(C_1\cos 2x+C_2\sin 2x).
$$

The factor $e^{-x}$ is the decaying envelope of the amplitude, and $2$ is the angular frequency of the oscillation.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Independent variable and unknown function.
x = sp.symbols("x")
y = sp.Function("y")
# Damped oscillator y'' + 2y' + 5y = 0.
ode = sp.Eq(y(x).diff(x, 2) + 2*y(x).diff(x) + 5*y(x), 0)
# General solution from the solver.
sol = sp.dsolve(ode, y(x))
# One sample solution from the real form, with C_1 = C_2 = 1.
hand = sp.exp(-x)*(sp.cos(2*x) + sp.sin(2*x))
print(sol)
# That sample solution satisfies the differential equation.
assert sp.simplify((hand.diff(x, 2) + 2*hand.diff(x) + 5*hand)) == 0
print("A sample solution satisfies the equation")
print("Confirmed")
```

## A pitfall

The condition $y'(0)$ is imposed on the derivative of the solution.
Solve for the two constants from the resulting linear system. Then differentiate the solution once more and substitute, as a check that the system was read correctly.

## Summary

Solve the characteristic equation, choose the form of the solution from the type of root, and then apply the initial conditions.
