The Heaviside function $H(t-a)$ equals $0$ for $t<a$ and equals $1$ for $t>a$.

$$
\mathcal{L}\{H(t-a)\}=\dfrac{e^{-as}}{s}.
$$

The Dirac delta $\delta(t-a)$ is the limit of a short impulse of area $1$ concentrated at $t=a$.

$$
\mathcal{L}\{\delta(t-a)\}=e^{-as}.
$$

The convolution of two functions is

$$
(f*g)(t)=\int_0^t f(\tau)g(t-\tau)\,d\tau,
$$

and the transform turns convolution into a product: $\mathcal{L}\{f*g\}=\mathcal{L}\{f\}\mathcal{L}\{g\}$.


## Example 1 — A step

### Hand solution

The unit step that switches on at $t=1$ has transform

$$
\mathcal{L}\{H(t-1)\}=\dfrac{e^{-s}}{s}.
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Time variable and transform variable.
t = sp.symbols("t", real=True)
s = sp.symbols("s", positive=True)
# Transform of the step H(t-1). Keep the first entry of the triple.
found = sp.laplace_transform(sp.Heaviside(t - 1), t, s)[0]
print(sp.simplify(found))
# The table entry is exp(-s)/s.
assert sp.simplify(found - sp.exp(-s)/s) == 0
print("Confirmed")
```

## Example 2 — An impulse

### Hand solution

$$
\mathcal{L}\{\delta(t-2)\}=e^{-2s}.
$$

An impulse at time $2$ contributes only the delay factor $e^{-2s}$ to the transform. The integral of $\delta(t-2)$ over every interval that contains $2$ equals $1$.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Time variable and transform variable.
t = sp.symbols("t", real=True)
s = sp.symbols("s", positive=True)
# Transform of the impulse delta(t-2). Keep the first entry of the triple.
found = sp.laplace_transform(sp.DiracDelta(t - 2), t, s)[0]
print(found)
# The table entry is exp(-2s).
assert sp.simplify(found - sp.exp(-2*s)) == 0
print("Confirmed")
```

## Example 3 — Convolution of 1 with 1

### Hand solution

The convolution of the constant function $1$ with itself is the integral of $1$:

$$
(1*1)(t)=\int_0^t 1\,d\tau=t.
$$

The transform route gives the same function of time:

$$
\dfrac{1}{s}\cdot\dfrac{1}{s}
=\dfrac{1}{s^2}
=\mathcal{L}\{t}.
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Time, dummy variable of the convolution, and transform variable.
t, tau = sp.symbols("t tau", nonnegative=True)
s = sp.symbols("s", positive=True)
# Direct convolution of 1 with 1 on [0, t].
convolution = sp.integrate(1, (tau, 0, t))
# Product of the two Laplace transforms of 1.
product = sp.laplace_transform(1, t, s)[0]**2
# Laplace transform of t, the candidate inverse.
of_t = sp.laplace_transform(t, t, s)[0]
print(convolution, product, of_t)
# The convolution equals t, and the product of transforms equals L{t}.
assert convolution == t
# The product of the transforms equals the transform of t.
assert sp.simplify(product - of_t) == 0
print("Confirmed")
```

## Example 4 — A force that switches on

### Hand solution

Solve $y'+y=H(t-1)$ with $y(0)=0$.
On the interval $t<1$ the right-hand side is $0$, and the initial condition keeps the solution equal to $0$.
For $t>1$ the equation is $y'+y=1$. In the transform domain,

$$
\begin{aligned}
sY+Y &= \dfrac{e^{-s}}{s}, \\
Y &= \dfrac{e^{-s}}{s(s+1)}
   = e^{-s}\left(\dfrac{1}{s}-\dfrac{1}{s+1}\right), \\
y &= H(t-1)\bigl(1-e^{-(t-1)}\bigr).
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Time variable and transform variable.
t = sp.symbols("t", real=True)
s = sp.symbols("s", positive=True)
# Transform of the delayed forcing, Y = exp(-s)/(s(s+1)).
Y = sp.exp(-s)/(s*(s + 1))
# Invert the transform.
y = sp.inverse_laplace_transform(Y, s, t)
# Hand solution: the step at t = 1 times 1 - exp(-(t-1)).
hand = sp.Heaviside(t - 1)*(1 - sp.exp(-(t - 1)))
print(sp.simplify(y))
# The inverse transform must match the hand solution.
assert sp.simplify(y - hand) == 0
print("Confirmed")
```

## A pitfall

A delay of length $a$ in time is multiplication by $e^{-as}$ in the $s$-domain.
That exponential factor belongs inside the partial-fraction calculation. Omitting it produces a solution that switches on at time $0$, while the step or the impulse occurs at time $a$.

## Summary

A Heaviside function is a switch, a Dirac delta is an impulse of unit area, and convolution is the operation whose transform is an ordinary product.
