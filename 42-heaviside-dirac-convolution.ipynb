# ۴۲ · پله، ضربه و پیچش

| | |
|---|---|
| **دوره** | معادلات دیفرانسیل |
| **پیش‌نیاز** | ۴۱ |

پلهٔ هویساید \(H(t-a)\) تا پیش از \(a\) صفر است و بعد یک.

\[
\mathcal{L}\{H(t-a)\}=\frac{e^{-as}}{s}
\]

ضربهٔ دیراک، حد یک ضربهٔ کوتاه با مساحت ۱ است.

\[
\mathcal{L}\{\delta(t-a)\}=e^{-as}
\]

پیچش

\[
(f*g)(t)=\int_0^t f(\tau)g(t-\tau)\,d\tau
\]

و \(\mathcal{L}\{f*g\}=\mathcal{L}\{f\}\mathcal{L}\{g\}\).



## مثال ۱ — پله

### حل ریاضی

\[
\mathcal{L}\{H(t-1)\}=\frac{e^{-s}}{s}
\]

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t", real=True)
s = sp.symbols("s", positive=True)
found = sp.laplace_transform(sp.Heaviside(t - 1), t, s)[0]
print(sp.simplify(found))
assert sp.simplify(found - sp.exp(-s)/s) == 0
print("تایید شد")

```

## مثال ۲ — ضربه

### حل ریاضی

\[
\mathcal{L}\{\delta(t-2)\}=e^{-2s}
\]

ضربه در لحظهٔ ۲، در تبدیل فقط یک نماییِ تأخیر است. انتگرال ضربه روی هر بازه‌ای که ۲ را در بر بگیرد برابر ۱ است.

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t", real=True)
s = sp.symbols("s", positive=True)
found = sp.laplace_transform(sp.DiracDelta(t - 2), t, s)[0]
print(found)
assert sp.simplify(found - sp.exp(-2*s)) == 0
print("تایید شد")

```

## مثال ۳ — پیچش ۱ با ۱

### حل ریاضی

\[
(1*1)(t)=\int_0^t 1\,d\tau=t
\]

از طرف تبدیل: \(\dfrac{1}{s}\cdot\dfrac{1}{s}=\dfrac{1}{s^2}=\mathcal{L}\{t\}\).
دو راه یکی هستند.

### بازبینی با پایتون



```python
import sympy as sp
t, tau = sp.symbols("t tau", nonnegative=True)
s = sp.symbols("s", positive=True)
convolution = sp.integrate(1, (tau, 0, t))
product = sp.laplace_transform(1, t, s)[0]**2
of_t = sp.laplace_transform(t, t, s)[0]
print(convolution, product, of_t)
assert convolution == t
assert sp.simplify(product - of_t) == 0
print("تایید شد")

```

## مثال ۴ — خاموش و روشن شدن نیرو

### حل ریاضی

\(y'+y=H(t-1)\)، \(y(0)=0\).
تا پیش از \(t=1\) سمت راست صفر است و شرط اولیه جواب را صفر نگه می‌دارد.
بعد از آن، معادله \(y'+y=1\) است.

\[
sY+Y=\frac{e^{-s}}{s}
\]
\[
Y=\frac{e^{-s}}{s(s+1)}=e^{-s}\left(\frac{1}{s}-\frac{1}{s+1}\right)
\]

\[
y=H(t-1)\bigl(1-e^{-(t-1)}\bigr)
\]

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t", real=True)
s = sp.symbols("s", positive=True)
Y = sp.exp(-s)/(s*(s + 1))
y = sp.inverse_laplace_transform(Y, s, t)
hand = sp.Heaviside(t - 1)*(1 - sp.exp(-(t - 1)))
print(sp.simplify(y))
assert sp.simplify(y - hand) == 0
print("تایید شد")

```

## نکته

تأخیر در زمان، ضرب در \(e^{-as}\) در فضای \(s\) است.
اگر این نمایی را در کسر جزئی فراموش کنید، جواب را از لحظهٔ صفر روشن کرده‌اید نه از لحظهٔ ضربه یا پله.

## جمع‌بندی

پله یک کلید است، ضربه یک ضربه با مساحت واحد، و پیچش قانون ضرب تبدیل‌هاست.
