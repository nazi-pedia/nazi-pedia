# ۱۲ · قاعدهٔ هوپیتال

| | |
|---|---|
| **دوره** | ریاضی عمومی ۱ |
| **پیش‌نیاز** | ۰۶ و ۰۸ |

اگر \(\lim f/g\) به شکل \(0/0\) یا \(\infty/\infty\) باشد و \(\lim f'/g'\) وجود داشته باشد،
حد اصلی با آن برابر است. قاعده را می‌توان تکرار کرد، به شرط اینکه هر بار هنوز شکل مبهم مانده باشد.

شکل‌های \(0\cdot\infty\)، \(\infty-\infty\) و \(1^\infty\) اول باید به خارج‌قسمت تبدیل شوند.
هوپیتال را روی شکلی که مبهم نیست به کار نبرید. حد \(\dfrac{x}{x+1}\) در بی‌نهایت، بعد از یک مشتق غلط می‌شود
اگر یادتان برود که این شکل \(\infty/\infty\) است و مجاز است؛ اما \(\dfrac{\sin x}{x}\) در \(x=\pi\) شکل مبهم نیست.



## مثال ۱ — یک بار کافی است

### حل ریاضی

\[
\lim_{x\to 0}\frac{\sin x}{x}
\]

شکل \(0/0\).

\[
\lim_{x\to 0}\frac{\cos x}{1}=\cos 0=1
\]

این همان حد استاندارد درس ۰۴ است. هوپیتال آن را از مشتق سینوس می‌سازد،
و مشتق سینوس خودش با همین حد ثابت می‌شود. در یک دورهٔ دقیق، یکی از این دو را باید از جای دیگری بگیرید
تا دور ایجاد نشود. اینجا حد را از قبل پذیرفته‌ایم و هوپیتال فقط آن را بازتولید می‌کند.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
assert sp.limit(sp.sin(x)/x, x, 0) == 1
assert sp.limit(sp.diff(sp.sin(x), x)/sp.diff(x, x), x, 0) == 1
print("هر دو مسیر به ۱ می‌رسند")
print("تایید شد")

```

## مثال ۲ — دو بار

### حل ریاضی

\[
\lim_{x\to 0}\frac{e^{x}-1-x}{x^2}
\]

در ۰ صورت \(1-1-0=0\). شکل \(0/0\).

\[
\lim_{x\to 0}\frac{e^{x}-1}{2x}
\]

باز \(0/0\).

\[
\lim_{x\to 0}\frac{e^{x}}{2}=\frac{1}{2}
\]

بسط \(e^{x}=1+x+x^2/2+\cdots\) هم صورت را \(x^2/2+\cdots\) می‌کند و خارج‌قسمت به \(1/2\) می‌رود.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = (sp.exp(x) - 1 - x)/x**2
exact = sp.limit(f, x, 0)
print(exact)
assert exact == sp.Rational(1, 2)
print("تایید شد")

```

## مثال ۳ — \(\infty/\infty\)

### حل ریاضی

\[
\lim_{x\to +\infty}\frac{\ln x}{x}
\]

صورت و مخرج هر دو به \(+\infty\) می‌روند.

\[
\lim_{x\to +\infty}\frac{1/x}{1}=\lim_{x\to +\infty}\frac{1}{x}=0
\]

لگاریتم از هر توان مثبتی کندتر رشد می‌کند. اینجا توان، خود \(x\) است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", positive=True)
assert sp.limit(sp.log(x)/x, x, sp.oo) == 0
assert sp.limit(sp.log(x)/sp.sqrt(x), x, sp.oo) == 0
print("ln x حتی از جذر هم کندتر است")
print("تایید شد")

```

## مثال ۴ — تبدیل \(1^\infty\)

### حل ریاضی

\[
\lim_{x\to 0^+}x^{x}
\]

پایه به ۰ و توان به ۰ می‌رود. می‌نویسیم

\[
x^{x}=e^{x\ln x}
\]

و جداگانه

\[
\lim_{x\to 0^+}x\ln x=\lim_{x\to 0^+}\frac{\ln x}{1/x}
\]

شکل \(\infty/\infty\) با علامت: صورت به \(-\infty\) و مخرج به \(+\infty\).

\[
\lim_{x\to 0^+}\frac{1/x}{-1/x^2}=\lim_{x\to 0^+}(-x)=0
\]

پس توان به ۰ می‌رود و \(e^{0}=1\).

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", positive=True)
assert sp.limit(x*sp.log(x), x, 0) == 0
assert sp.limit(x**x, x, 0) == 1
print("x ln x ->", sp.limit(x*sp.log(x), x, 0))
print("x^x ->", sp.limit(x**x, x, 0))
print("تایید شد")

```

## مثال ۵ — جایی که هوپیتال مجاز نیست

### حل ریاضی

\[
\lim_{x\to \pi}\frac{\sin x}{x}
\]

صورت ۰ است و مخرج \(\pi\neq 0\). شکل مبهم نیست.

\[
\frac{\sin\pi}{\pi}=0
\]

اگر بی‌جهت مشتق بگیرید، \(\dfrac{\cos\pi}{1}=-1\) می‌شود و جواب غلط است.
مجوز قاعده، شکل مبهم است نه کسری بودن.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
exact = sp.limit(sp.sin(x)/x, x, sp.pi)
naive = sp.limit(sp.diff(sp.sin(x), x)/1, x, sp.pi)
print("حد درست:", exact)
print("مشتق بی‌مجوز:", naive)
assert exact == 0
assert naive == -1
print("تایید شد")

```

## نکته

بعد از هر بار هوپیتال، حد تازه را از نو طبقه‌بندی کنید.
ممکن است دیگر مبهم نباشد و یک جانشینی ساده کافی باشد.

## جمع‌بندی

\(0/0\) و \(\infty/\infty\) مستقیماً مشتق می‌گیرند.
ضرب و توان را اول به خارج‌قسمت یا به نمایی تبدیل کنید.
بدون شکل مبهم، مشتق نگیرید.
