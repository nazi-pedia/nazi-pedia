# Add the person and the movie, then subtract the global mean

Write $\mu$ for the mean of the known ratings, $\bar r_u$ for the mean of user $u$, and $\bar r_i$ for the mean of movie $i$. The bias prediction is

$$
\hat r_{ui} = \mu + (\bar r_u - \mu) + (\bar r_i - \mu)
= \bar r_u + \bar r_i - \mu.
$$

The two parentheses are the user's departure from the global mean and the movie's departure from the global mean. Adding both, then leaving $\mu$ in the total, would count the global mean twice. The identity subtracts it once.

Check it on seven known ratings. Movies are Toy, Gold, and Room. The question marks are the cells we predict. They are not used in the means.

| | Toy | Gold | Room |
|---|---:|---:|---:|
| U1 | 5 | 3 | ? |
| U2 | 4 | 2 | 4 |
| U3 | 1 | ? | 2 |

$$
\begin{aligned}
\mu &= \dfrac{5+3+4+2+4+1+2}{7} = 3 \\
\bar r_{U1} &= 4, \qquad \bar r_{U2} = \dfrac{10}{3}, \qquad \bar r_{U3} = \dfrac{3}{2} \\
\bar r_{\mathrm{Toy}} &= \dfrac{10}{3}, \qquad
\bar r_{\mathrm{Gold}} = \dfrac{5}{2}, \qquad
\bar r_{\mathrm{Room}} = 3
\end{aligned}
$$

U1 on Room:

$$
4 + 3 - 3 = 4.
$$

U3 on Gold:

$$
\dfrac{3}{2} + \dfrac{5}{2} - 3 = 1.
$$

Both answers land on the scale. The formula does not promise that. A user far above $\mu$ and a movie far above $\mu$ can add to more than 5, and the opposite pair can add to less than 1.


```python
# The seven known ratings, two predictions, exact fractions.
from fractions import Fraction
known = {
    ("U1", "Toy"): 5,
    ("U1", "Gold"): 3,
    ("U2", "Toy"): 4,
    ("U2", "Gold"): 2,
    ("U2", "Room"): 4,
    ("U3", "Toy"): 1,
    ("U3", "Room"): 2,
}
mu = Fraction(sum(known.values()), len(known))
def mean_of(pairs):
    return Fraction(sum(known[pair] for pair in pairs), len(pairs))
user = {
    "U1": mean_of([("U1", "Toy"), ("U1", "Gold")]),
    "U2": mean_of([("U2", "Toy"), ("U2", "Gold"), ("U2", "Room")]),
    "U3": mean_of([("U3", "Toy"), ("U3", "Room")]),
}
movie = {
    "Toy": mean_of([("U1", "Toy"), ("U2", "Toy"), ("U3", "Toy")]),
    "Gold": mean_of([("U1", "Gold"), ("U2", "Gold")]),
    "Room": mean_of([("U2", "Room"), ("U3", "Room")]),
}
room = user["U1"] + movie["Room"] - mu
gold = user["U3"] + movie["Gold"] - mu
print(mu, user, movie, room, gold)
assert mu == 3
assert user == {"U1": 4, "U2": Fraction(10, 3), "U3": Fraction(3, 2)}
assert movie == {"Toy": Fraction(10, 3), "Gold": Fraction(5, 2), "Room": 3}
assert room == 4
assert gold == 1
```

## Pitfall

The two question marks were left out of the means. If you fill them with a guess and then recompute $\bar r_u$ from that guess, the formula is no longer a prediction. It is using the answer to build the answer.

## Summary

$\hat r_{ui} = \bar r_u + \bar r_i - \mu$. On the toy, U1's missing Room is 4 and U3's missing Gold is 1.
