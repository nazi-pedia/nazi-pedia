# ۳۸ · معادلهٔ دیفرانسیل مرتبهٔ اول

| | |

| **دوره** | معادلات دیفرانسیل |
| **پیش‌نیاز** | ۱۶ |

مرتبه، بالاترین مشتقی است که در معادله هست.

جدایی‌پذیر: \(g(y)\,y'=f(x)\). دو طرف را در متغیر خودشان انتگرال بگیرید.

خطی: \(y'+P(x)y=Q(x)\). عامل انتگرال‌ساز \(e^{\int P}\).

کامل: \(M\,dx+N\,dy=0\) وقتی \(M_y=N_x\). پتانسیل را با انتگرال \(M\) نسبت به \(x\) و اصلاح بر حسب \(y\) بسازید.

برنولی: \(y'+P y=Q y^n\). با \(v=y^{1-n}\) خطی می‌شود.



## مثال ۱ — جدایی‌پذیر

### حل ریاضی

\(y'=2xy\) و \(y(0)=1\).

\[
\frac{dy}{y}=2x\,dx,\qquad \ln|y|=x^2+C
\]

\[
y=Ae^{x^2}
\]

\(y(0)=1\) می‌دهد \(A=1\). پس \(y=e^{x^2}\).

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
y = sp.Function("y")
ode = sp.Eq(y(x).diff(x), 2*x*y(x))
sol = sp.dsolve(ode, y(x), ics={y(0): 1})
print(sol)
assert sp.simplify(sol.rhs - sp.exp(x**2)) == 0
print("تایید شد")

```

## مثال ۲ — خطی

### حل ریاضی

\(y'+2y=4\)، \(y(0)=0\).
عامل \(e^{2x}\).

\[
(ye^{2x})'=4e^{2x}
\]
\[
ye^{2x}=2e^{2x}+C,\qquad y=2+Ce^{-2x}
\]

\(y(0)=0\) می‌دهد \(C=-2\).

\[
y=2(1-e^{-2x})
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
y = sp.Function("y")
ode = sp.Eq(y(x).diff(x) + 2*y(x), 4)
sol = sp.dsolve(ode, y(x), ics={y(0): 0})
hand = 2*(1 - sp.exp(-2*x))
print(sol)
assert sp.simplify(sol.rhs - hand) == 0
print("تایید شد")

```

## مثال ۳ — کامل

### حل ریاضی

\((2x+y)\,dx+(x+2y)\,dy=0\).
\(M_y=1=N_x\). پتانسیل:

\[
\phi=\int(2x+y)\,dx=x^2+xy+h(y)
\]

\[
\phi_y=x+h'(y)=x+2y\implies h'=2y\implies h=y^2
\]

\[
x^2+xy+y^2=C
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
y = sp.Function("y")
ode = sp.Eq((2*x + y(x)) + (x + 2*y(x))*y(x).diff(x), 0)
sol = sp.dsolve(ode, y(x))
print(sol)
phi = x**2 + x*y(x) + y(x)**2
assert sp.simplify(sp.diff(phi, x)) == 0 or sp.simplify(sp.diff(phi, x).subs(y(x).diff(x), sp.solve(ode, y(x).diff(x))[0])) == 0
residual = sp.diff(x**2 + x*y(x) + y(x)**2, x)
slope = sp.solve(ode, y(x).diff(x))[0]
assert sp.simplify(residual.subs(y(x).diff(x), slope)) == 0
print("تایید شد")

```

## مثال ۴ — برنولی

### حل ریاضی

\(y'+y=y^2\)، \(y(0)=\dfrac{1}{2}\).
\(v=y^{-1}\). آن‌گاه \(y=1/v\) و \(y'=-v'/v^2\).

\[
-\frac{v'}{v^2}+\frac{1}{v}=\frac{1}{v^2}
\]

ضرب در \(-v^2\):

\[
v'-v=-1
\]

عامل \(e^{-x}\). \((ve^{-x})'=-e^{-x}\)، پس \(v=1+Ce^{x}\).

\[
y=\frac{1}{1+Ce^{x}}
\]

\(y(0)=1/2\) می‌دهد \(C=1\). \(y=\dfrac{1}{1+e^{x}}\).

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
y = sp.Function("y")
ode = sp.Eq(y(x).diff(x) + y(x), y(x)**2)
sol = sp.dsolve(ode, y(x), ics={y(0): sp.Rational(1, 2)})
hand = 1/(1 + sp.exp(x))
print(sol)
assert sp.simplify(sol.rhs - hand) == 0
print("تایید شد")

```
