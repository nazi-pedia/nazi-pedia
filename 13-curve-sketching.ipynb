# ۱۳ · رسم نمودار

| | |
|---|---|
| **دوره** | ریاضی عمومی ۱ |
| **پیش‌نیاز** | ۰۶ و ۰۷ |

رسم نمودار یعنی ترجمهٔ مشتق به جمله.

- \(f'>0\): صعودی. \(f'<0\): نزولی.
- \(f''>0\): تقعر رو به بالا. \(f''<0\): تقعر رو به پایین.
- \(f'(c)=0\) و تغییر علامت \(f'\) از مثبت به منفی: بیشینهٔ موضعی. برعکس: کمینهٔ موضعی.
- \(f''(c)=0\) به‌تنهایی هیچ چیز را تمام نمی‌کند. تغییر علامت \(f''\) نقطهٔ عطف است.
- مجانب را از درس ۰۶ بیاورید.

بیشینهٔ موضعی، بیشینهٔ مطلق روی کل خط نیست. برای حکم مطلق، بازه باید بسته باشد یا رفتار دو انتها معلوم باشد.



## مثال ۱ — خم درجهٔ سهٔ آشنا

### حل ریاضی

\[
f(x)=x^3-3x^2+4
\]
\[
f'(x)=3x^2-6x=3x(x-2),\qquad f''(x)=6x-6
\]

نقاط بحرانی \(x=0\) و \(x=2\).

\(f''(0)=-6<0\)، پس بیشینهٔ موضعی. \(f(0)=4\).
\(f''(2)=6>0\)، پس کمینهٔ موضعی. \(f(2)=8-12+4=0\).

آزمون علامت \(f'\) همین را می‌گوید: مثبت، بعد منفی، بعد مثبت.
یعنی صعود، نزول، صعود.

\(f''(x)=0\) در \(x=1\)، و \(f''\) آنجا علامت عوض می‌کند. عطف در \((1,2)\)،
چون \(f(1)=1-3+4=2\).

وقتی \(x\to\pm\infty\)، جملهٔ \(x^3\) علامت حد را تعیین می‌کند. بیشینه و کمینه مطلق روی کل خط وجود ندارد.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = x**3 - 3*x**2 + 4
fp = sp.factor(sp.diff(f, x))
fpp = sp.diff(f, x, 2)
print("f' =", fp)
print("f'' =", sp.factor(fpp))
assert sp.solve(sp.diff(f, x), x) == [0, 2]
assert fpp.subs(x, 0) < 0 and f.subs(x, 0) == 4
assert fpp.subs(x, 2) > 0 and f.subs(x, 2) == 0
assert fpp.subs(x, 1) == 0 and f.subs(x, 1) == 2
print("تایید شد")

```

## مثال ۲ — خواندن علامت، وقتی مشتق دوم صفر است

### حل ریاضی

\[
g(x)=x^3,\qquad g'(x)=3x^2,\qquad g''(x)=6x
\]

تنها نقطهٔ بحرانی \(x=0\) است و \(g''(0)=0\). آزمون مشتق دوم ساکت است.
اما \(g'(x)\ge 0\) در هر دو طرف، پس نمودار از صعود به نزول نمی‌رود.
عطف افقی است، نه اکسترمم.

مقایسه: \(h(x)=x^4\) هم \(h''(0)=0\) دارد، ولی \(h'(x)=4x^3\) از منفی به مثبت می‌رود
و \(x=0\) کمینه است. صفر بودن مشتق دوم هر دو امکان را باز می‌گذارد.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
assert sp.diff(x**3, x, 2).subs(x, 0) == 0
assert sp.diff(x**4, x, 2).subs(x, 0) == 0
gp = sp.diff(x**3, x)
hp = sp.diff(x**4, x)
assert gp.subs(x, -1) > 0 and gp.subs(x, 1) > 0
assert hp.subs(x, -1) < 0 < hp.subs(x, 1)
print("x^3 اکسترمم نیست؛ x^4 کمینه است")
print("تایید شد")

```

## مثال ۳ — مجانب در طرح نمودار

### حل ریاضی

\[
k(x)=\frac{x}{x^2+1}
\]
\[
k'(x)=\frac{(x^2+1)-x\cdot 2x}{(x^2+1)^2}=\frac{1-x^2}{(x^2+1)^2}
\]

نقاط بحرانی \(x=\pm 1\). مخرج مشتق همیشه مثبت است، پس علامت از \(1-x^2\) می‌آید.
روی \((-1,1)\) صعودی است. بیرون آن نزولی است.
پس \(x=-1\) کمینهٔ موضعی است و \(k(-1)=-1/2\).
\(x=1\) بیشینهٔ موضعی است و \(k(1)=1/2\).

\[
\lim_{x\to\pm\infty}k(x)=0
\]

مجانب افقی محور \(x\) است. مجانب قائم نداریم، چون مخرج هرگز صفر نمی‌شود.
\(k(0)=0\). نمودار از زیر محور می‌آید، در \(-1/2\) برمی‌گردد، از مبدأ می‌گذرد،
تا \(1/2\) بالا می‌رود و دوباره به محور نزدیک می‌شود.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", real=True)
k = x/(x**2 + 1)
kp = sp.together(sp.diff(k, x))
print(sp.factor(sp.numer(sp.together(kp))))
assert sp.solve(sp.diff(k, x), x) == [-1, 1]
assert k.subs(x, 1) == sp.Rational(1, 2)
assert k.subs(x, -1) == -sp.Rational(1, 2)
assert sp.limit(k, x, sp.oo) == 0
print("تایید شد")

```

## مثال ۴ — عطف بدون بحرانی بودن

### حل ریاضی

\[
p(x)=x^3-3x+1
\]
\[
p'(x)=3x^2-3=3(x^2-1),\qquad p''(x)=6x
\]

بحرانی‌ها \(x=\pm 1\) هستند. عطف در \(x=0\) است، جایی که \(p'(0)=-3\neq 0\).
مماس در عطف افقی نیست. \(p(0)=1\) و شیب مماس \(-3\) است:

\[
y-1=-3x
\]

نقطهٔ عطف و نقطهٔ بحرانی دو فهرست جدا هستند. گاهی اشتراک دارند، مثل \(x^3\)، و اغلب ندارند.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
p = x**3 - 3*x + 1
assert sp.solve(sp.diff(p, x, 2), x) == [0]
assert sp.diff(p, x).subs(x, 0) == -3
assert p.subs(x, 0) == 1
print("عطف در x=0 با شیب -3")
print("تایید شد")

```

## نکته

پیش از رسم، این پنج عدد را در یک جدول بریزید:
ریشهٔ \(f\)، ریشهٔ \(f'\)، ریشهٔ \(f''\)، مجانب قائم، مجانب افقی.
نمودار بدون این جدول تزئین است.

## جمع‌بندی

\(f'\) جهت حرکت را می‌گوید و \(f''\) خمیدگی را.
اگر \(f''(c)=0\)، به علامت دو طرف برگردید.
موضعی را مطلق ننامید مگر انتهاهای بازه را دیده باشید.
