Curve sketching is the translation of a derivative into a sentence about the graph.

- If $f'>0$, the graph is increasing. If $f'<0$, the graph is decreasing.
- If $f''>0$, the graph is concave up. If $f''<0$, the graph is concave down.
- If $f'(c)=0$ and $f'$ changes from positive to negative, then $c$ is a local maximum. The opposite sign change is a local minimum.
- The equation $f''(c)=0$ settles nothing by itself. A sign change of $f''$ is an inflection point.
- Bring the asymptotes in from lesson 06.

A local maximum on the whole line is a separate claim. An absolute statement needs a closed interval, or known behavior at both ends.


## Example 1 — A familiar cubic

### Hand solution

$$
\begin{aligned}
f(x) &= x^3-3x^2+4, \\
f'(x) &= 3x^2-6x=3x(x-2), \\
f''(x) &= 6x-6.
\end{aligned}
$$

The critical points are $x=0$ and $x=2$.

$$
\begin{aligned}
f''(0) &= -6<0, & f(0) &= 4, \\
f''(2) &= 6>0, & f(2) &= 8-12+4=0.
\end{aligned}
$$

The first is a local maximum and the second is a local minimum. The sign of $f'$ tells the same story: positive, then negative, then positive. The graph increases, decreases, then increases.

Now $f''(x)=0$ at $x=1$, and $f''$ changes sign there. The inflection point is $(1,2)$, because

$$
f(1)=1-3+4=2.
$$

As $x\to\pm\infty$, the term $x^3$ sets the sign of the limit. On the whole line there is no absolute maximum and no absolute minimum.

### Python recheck


```python
import sympy as sp

# Symbol and the cubic from the hand solution.
x = sp.symbols("x")
f = x**3 - 3*x**2 + 4

# Factored first derivative, then the second derivative.
fp = sp.factor(sp.diff(f, x))
fpp = sp.diff(f, x, 2)
print("f' =", fp)
print("f'' =", sp.factor(fpp))

# Critical points are 0 and 2.
assert sp.solve(sp.diff(f, x), x) == [0, 2]
# Local maximum at 0, with value 4.
assert fpp.subs(x, 0) < 0 and f.subs(x, 0) == 4
# Local minimum at 2, with value 0.
assert fpp.subs(x, 2) > 0 and f.subs(x, 2) == 0
# Inflection at x = 1, where the value is 2.
assert fpp.subs(x, 1) == 0 and f.subs(x, 1) == 2
print("Confirmed")
```

## Example 2 — Reading the sign when the second derivative is zero

### Hand solution

$$
g(x)=x^3,\qquad g'(x)=3x^2,\qquad g''(x)=6x.
$$

The only critical point is $x=0$, and $g''(0)=0$. The second-derivative test is silent. On both sides, $g'(x)\ge 0$, so the graph does not pass from increasing to decreasing. The origin is a horizontal inflection, not an extremum.

Compare $h(x)=x^4$. There too $h''(0)=0$, but $h'(x)=4x^3$ changes from negative to positive, so $x=0$ is a minimum. A second derivative that vanishes leaves both possibilities open.

### Python recheck


```python
import sympy as sp

# Both second derivatives vanish at the origin.
x = sp.symbols("x")
assert sp.diff(x**3, x, 2).subs(x, 0) == 0
assert sp.diff(x**4, x, 2).subs(x, 0) == 0

# First derivatives, sampled on each side of 0.
gp = sp.diff(x**3, x)
hp = sp.diff(x**4, x)
# x^3 keeps a positive derivative: no extremum.
assert gp.subs(x, -1) > 0 and gp.subs(x, 1) > 0
# x^4 changes from negative to positive: a minimum.
assert hp.subs(x, -1) < 0 < hp.subs(x, 1)
print("x^3 is not an extremum; x^4 is a minimum")
print("Confirmed")
```

## Example 3 — An asymptote in the sketch

### Hand solution

$$
\begin{aligned}
k(x) &= \dfrac{x}{x^2+1}, \\
k'(x) &= \dfrac{(x^2+1)-x\cdot 2x}{(x^2+1)^2}=\dfrac{1-x^2}{(x^2+1)^2}.
\end{aligned}
$$

The critical points are $x=\pm 1$. The denominator of $k'(x)$ is always positive, so the sign comes from $1-x^2$. On $(-1,1)$ the graph is increasing. Outside that interval it is decreasing. Thus $x=-1$ is a local minimum and $k(-1)=-\dfrac{1}{2}$. Likewise $x=1$ is a local maximum and $k(1)=\dfrac{1}{2}$.

$$
\lim_{x\to\pm\infty} k(x)=0.
$$

The horizontal asymptote is the $x$-axis. There is no vertical asymptote, because the denominator never vanishes. Also $k(0)=0$. The graph arrives from below the axis, turns at $-\dfrac{1}{2}$, passes through the origin, rises to $\dfrac{1}{2}$, and approaches the axis again.

### Python recheck


```python
import sympy as sp

# Rational function and its derivative, written as one fraction.
x = sp.symbols("x", real=True)
k = x/(x**2 + 1)
kp = sp.together(sp.diff(k, x))
print(sp.factor(sp.numer(sp.together(kp))))

# Critical points, their values, and the horizontal asymptote.
assert sp.solve(sp.diff(k, x), x) == [-1, 1]
assert k.subs(x, 1) == sp.Rational(1, 2)
assert k.subs(x, -1) == -sp.Rational(1, 2)
assert sp.limit(k, x, sp.oo) == 0
print("Confirmed")
```

## Example 4 — Inflection that is not a critical point

### Hand solution

$$
\begin{aligned}
p(x) &= x^3-3x+1, \\
p'(x) &= 3x^2-3=3(x^2-1), \\
p''(x) &= 6x.
\end{aligned}
$$

The critical points are $x=\pm 1$. The inflection is at $x=0$, where $p'(0)=-3\neq 0$. The tangent line at the inflection point is not horizontal. Since $p(0)=1$ and the slope is $-3$,

$$
y-1=-3x.
$$

Inflection points and critical points are two separate lists. They sometimes meet, as they do for $x^3$, and often they do not.

### Python recheck


```python
import sympy as sp

# Inflection from the second derivative, then the slope and the height.
x = sp.symbols("x")
p = x**3 - 3*x + 1
assert sp.solve(sp.diff(p, x, 2), x) == [0]
assert sp.diff(p, x).subs(x, 0) == -3
assert p.subs(x, 0) == 1
print("Inflection at x=0 with slope -3")
print("Confirmed")
```

## A pitfall

Before you sketch, put five pieces of data in one table: the roots of $f$, the roots of $f'$, the roots of $f''$, the vertical asymptotes, and the horizontal asymptotes. A graph without that table is decoration.

## Summary

The sign of $f'$ gives the direction of motion, and the sign of $f''$ gives the bending. If $f''(c)=0$, return to the sign of $f''$ on each side. Call an extremum absolute only after you have seen the ends of the interval.
