# ۲۶ · خم پارامتری

| | |
|---|---|
| **دوره** | ریاضی عمومی ۲ |
| **پیش‌نیاز** | ۰۸ و ۲۱ |

\[
x=x(t),\qquad y=y(t)
\]

شیب:

\[
\frac{dy}{dx}=\frac{dy/dt}{dx/dt}
\]

به شرط \(dx/dt\neq 0\). اگر صورت هم صفر باشد، مماس را با حد یا با مشتق دوم پارامتری تعیین کنید.

طول از \(t=a\) تا \(t=b\):

\[
L=\int_a^b\sqrt{\left(\frac{dx}{dt}\right)^2+\left(\frac{dy}{dt}\right)^2}\,dt
\]



## مثال ۱ — مماس بر دایره

### حل ریاضی

\(x=\cos t\)، \(y=\sin t\).

\[
\frac{dy}{dx}=\frac{\cos t}{-\sin t}=-\cot t
\]

در \(t=\pi/4\) نقطه \(\left(\dfrac{\sqrt{2}}{2},\dfrac{\sqrt{2}}{2}\right)\) است و شیب \(-1\).
شعاع این نقطه شیب \(+1\) دارد. مماس بر شعاع عمود است.

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t")
x = sp.cos(t)
y = sp.sin(t)
slope = sp.simplify(sp.diff(y, t)/sp.diff(x, t))
print(slope, slope.subs(t, sp.pi/4))
assert slope.subs(t, sp.pi/4) == -1
print("تایید شد")

```

## مثال ۲ — محیط دایره از روی پارامتر

### حل ریاضی

\[
\left(\frac{dx}{dt}\right)^2+\left(\frac{dy}{dt}\right)^2=\sin^2 t+\cos^2 t=1
\]

\[
L=\int_0^{2\pi}1\,dt=2\pi
\]

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t")
speed = sp.simplify(sp.sqrt(sp.diff(sp.cos(t), t)**2 + sp.diff(sp.sin(t), t)**2))
length = sp.integrate(speed, (t, 0, 2*sp.pi))
print(speed, length)
assert length == 2*sp.pi
print("تایید شد")

```

## مثال ۳ — یک طاق چرخ‌زاد

### حل ریاضی

\[
x=t-\sin t,\qquad y=1-\cos t
\]

از \(t=0\) تا \(t=2\pi\).

\[
dx/dt=1-\cos t,\qquad dy/dt=\sin t
\]
\[
(dx/dt)^2+(dy/dt)^2=1-2\cos t+\cos^2 t+\sin^2 t=2-2\cos t=4\sin^2(t/2)
\]

روی \([0,2\pi]\)، \(\sin(t/2)\ge 0\).

\[
L=\int_0^{2\pi}2\sin\frac{t}{2}\,dt=\Bigl[-4\cos\frac{t}{2}\Bigr]_0^{2\pi}
=-4(-1-1)=8
\]

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t", real=True)
dx = sp.diff(t - sp.sin(t), t)
dy = sp.diff(1 - sp.cos(t), t)
speed = sp.simplify(sp.sqrt(dx**2 + dy**2))
hand_speed = 2*sp.sin(t/2)
assert sp.simplify(speed**2 - hand_speed**2) == 0
length = sp.integrate(hand_speed, (t, 0, 2*sp.pi))
numeric = sp.N(sp.Integral(speed, (t, 0, 2*sp.pi)))
print(speed, "->", length, "عددی:", numeric)
assert sp.simplify(length - 8) == 0
assert abs(numeric - 8) < 10**(-10)
print("تایید شد")

```

## مثال ۴ — مشتق دوم پارامتری

### حل ریاضی

\[
\frac{d^2y}{dx^2}=\frac{d}{dt}\left(\frac{dy}{dx}\right)\Big/\frac{dx}{dt}
\]

برای دایره، \(dy/dx=-\cot t\).

\[
\frac{d}{dt}(-\cot t)=\csc^2 t
\]
\[
\frac{d^2y}{dx^2}=\frac{\csc^2 t}{-\sin t}=-\frac{1}{\sin^3 t}
\]

در \(t=\pi/2\) این مقدار \(-1\) است. دایره در بالاترین نقطه تقعر رو به پایین دارد و شعاع انحنا ۱ است.

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t")
slope = sp.diff(sp.sin(t), t)/sp.diff(sp.cos(t), t)
second = sp.simplify(sp.diff(slope, t)/sp.diff(sp.cos(t), t))
print(second, second.subs(t, sp.pi/2))
assert second.subs(t, sp.pi/2) == -1
print("تایید شد")

```

## نکته

پارامتر، زمانِ رسم است نه مختص \(x\).
یک خم می‌تواند در یک \(t\) مماس قائم داشته باشد، جایی که \(dx/dt=0\) و \(dy/dt\neq 0\).

## جمع‌بندی

طول، انتگرال تندی است.
