# ۳۳ · اکسترمم و ضرایب لاگرانژ

| | |

نقطهٔ بحرانی جایی است که \(\nabla f=\mathbf{0}\)، یا گرادیان تعریف نشود.

آزمون مشتق دوم با هسیان:

\[
H=f_{xx}f_{yy}-(f_{xy})^2
\]

اگر \(H>0\) و \(f_{xx}>0\)، کمینهٔ موضعی است.
اگر \(H>0\) و \(f_{xx}<0\)، بیشینهٔ موضعی است.
اگر \(H<0\)، زینی است.
اگر \(H=0\)، آزمون ساکت است.

روی قید \(g=c\)، به‌جای حذف متغیر،

\[
\nabla f=\lambda\nabla g
\]

را همراه خود قید حل کنید. \(\lambda\) قیمت سایه است: آهنگ تغییر بهینه نسبت به شل کردن قید.



## مثال ۱ — کمینهٔ قطعی

### حل ریاضی

\[
f=x^2+y^2-xy
\]
\[
f_x=2x-y=0,\qquad f_y=2y-x=0
\]

از اولی \(y=2x\). در دومی \(4x-x=0\)، پس \(x=0\) و \(y=0\).
\(f_{xx}=2\)، \(f_{yy}=2\)، \(f_{xy}=-1\).

\[
H=4-1=3>0
\]

و \(f_{xx}>0\). کمینهٔ موضعی، و مقدارش ۰ است. در واقع چون \(f=\frac{1}{2}(x^2+y^2+(x-y)^2)\) نامنفی است، این کمینه مطلق هم هست.

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y", real=True)
f = x**2 + y**2 - x*y
crit = sp.solve([sp.diff(f, x), sp.diff(f, y)], [x, y], dict=True)
H = sp.diff(f, x, 2)*sp.diff(f, y, 2) - sp.diff(f, x, y)**2
print(crit, H, f.subs(crit[0]))
assert crit == [{x: 0, y: 0}]
assert H == 3
print("تایید شد")

```

## مثال ۲ — نقطهٔ زینی

### حل ریاضی

\(f=x^2-y^2\).
\(f_x=2x=0\) و \(f_y=-2y=0\) فقط در مبدأ.
\(H=(-2)( -2)\) غلط است: \(f_{xx}=2\)، \(f_{yy}=-2\)، \(f_{xy}=0\).

\[
H=-4<0
\]

زینی. روی محور \(x\) کمینهٔ موضعی به نظر می‌رسد و روی محور \(y\) بیشینه. هیچ‌کدام در صفحه پایدار نیست.

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y")
f = x**2 - y**2
H = sp.diff(f, x, 2)*sp.diff(f, y, 2) - sp.diff(f, x, y)**2
print(H, f.subs({x: 0, y: 0}), f.subs({x: 1, y: 0}), f.subs({x: 0, y: 1}))
assert H == -4
print("تایید شد")

```

## مثال ۳ — لاگرانژ روی دایره

### حل ریاضی

بیشینهٔ \(f=x+y\) روی \(x^2+y^2=1\).

\[
(1,1)=\lambda(2x,2y),\qquad x^2+y^2=1
\]

پس \(x=y\) و \(2x^2=1\). \(x=\pm 1/\sqrt{2}\).

\[
f\left(\frac{1}{\sqrt{2}},\frac{1}{\sqrt{2}}\right)=\sqrt{2}
\]
\[
f\left(-\frac{1}{\sqrt{2}},-\frac{1}{\sqrt{2}}\right)=-\sqrt{2}
\]

بیشینه \(\sqrt{2}\) و کمینه \(-\sqrt{2}\) است.
هندسه: ترازهای \(x+y=c\) خط‌اند و آخرین تماس با دایره در امتداد قطر \(y=x\) است.

### بازبینی با پایتون



```python
import sympy as sp
x, y, lam = sp.symbols("x y lam", real=True)
eqs = [1 - lam*2*x, 1 - lam*2*y, x**2 + y**2 - 1]
sols = sp.solve(eqs, [x, y, lam], dict=True)
values = sorted(sp.simplify(sol[x] + sol[y]) for sol in sols)
print(values)
assert values[0] == -sp.sqrt(2)
assert values[-1] == sp.sqrt(2)
print("تایید شد")

```

## مثال ۴ — قید خطی، حذف در برابر لاگرانژ

### حل ریاضی

بیشینهٔ \(f=xy\) روی \(x+y=4\)، با \(x,y>0\).
حذف: \(y=4-x\)، \(A=x(4-x)\)، \(A'=4-2x=0\)، \(x=2\)، \(y=2\)، مقدار ۴.

لاگرانژ: \((y,x)=\lambda(1,1)\)، پس \(x=y\)، و با قید \(x=y=2\). همان جواب.
وقتی قید خطی و ساده است هر دو راه کوتاه‌اند. وقتی قید دایره یا بیضی است، لاگرانژ حذف را از رادیکال نجات می‌دهد.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", real=True)
A = x*(4 - x)
crit = sp.solve(sp.diff(A, x), x)[0]
print(crit, A.subs(x, crit))
assert crit == 2 and A.subs(x, crit) == 4
print("تایید شد")

```

## نکته

\(H=0\) را مثل \(f''=0\) در یک متغیر تعبیر نکنید. آزمون تمام نشده است.
روی ناحیهٔ بسته، بحرانی‌های داخل و همهٔ مرز را با هم مقایسه کنید. لاگرانژ فقط مرز را می‌بیند.

## جمع‌بندی

\(\nabla f=0\) نامزد داخلی است.
هسیان نامزد را طبقه‌بندی می‌کند، مگر اینکه صفر شود.
روی قید، گرادیان هدف و گرادیان قید موازی‌اند.
