# What a fitted model adds

A tree, or a logistic model with one coefficient per phrase, can use tenure, the add-on fields, and the charges in the same fit. Lesson 12 already showed the smallest logistic model: two rates, and a cut at $1/2$ that flags nobody. A larger fit can rank customers using combinations the eight-sentence menu never wrote down.

That fit is a different object from the number in lesson 11. Two habits change the object into something else again.

Scoring a rule on the same rows that chose it uses the 7043 twice. The path's $F_1$ is computed on the 1417 only.

Quoting accuracy on a table whose stay rate is $5174/7043$ lines the model up against Stay-for-everyone. On the holdout, that constant rule's accuracy is $1052/1417$, which is higher than the sentence's $1044/1417$. The sentence's $F_1$ is $488/861$, and the constant rule's recall is 0.

There is no hidden contest file of labels in this folder. The 1417 labels are in the same CSV, and the split is the MD5 rule, so anyone can recompute $488/861$. A published accuracy from a library model is that author's score under that author's split. It is not this path's number.


```python
# Locate the telco table beside this notebook, or under the course root.
import csv
from pathlib import Path

def data_path():
    here = Path.cwd()
    name = "Telco-Customer-Churn.csv"
    for folder in [here, *here.parents]:
        nested = folder / "16-Customer Churn" / "data" / name
        direct = folder / "data" / name
        if nested.exists():
            return nested
        if direct.exists():
            return direct
    raise FileNotFoundError(name)

def read_customers():
    with data_path().open(newline="", encoding="utf-8") as handle:
        return list(csv.DictReader(handle))
# The holdout number stays 488/861. Full-file accuracy is a different count.
import hashlib
from fractions import Fraction

rows = read_customers()

def held(customer_id):
    digest = hashlib.md5(customer_id.encode("utf-8")).hexdigest()
    return int(digest, 16) % 5 == 0

def leave_rule(row):
    return row["Contract"] == "Month-to-month" and (
        row["InternetService"] == "Fiber optic"
        or row["PaymentMethod"] == "Electronic check"
    )

test = [row for row in rows if held(row["customerID"])]
tp = sum(leave_rule(row) and row["Churn"] == "Yes" for row in test)
fp = sum(leave_rule(row) and row["Churn"] == "No" for row in test)
fn = sum((not leave_rule(row)) and row["Churn"] == "Yes" for row in test)
precision = Fraction(tp, tp + fp)
recall = Fraction(tp, tp + fn)
f1 = 2 * precision * recall / (precision + recall)
print(f1)
assert f1 == Fraction(488, 861)
assert Fraction(5174, 7043) > Fraction(1, 2)
assert Fraction(1052, 1417) > Fraction(1044, 1417)
```

## Pitfall

An accuracy printed near $5174/7043$ can be a small step past Stay-for-everyone. The comparison that matches the menu is held-out $F_1$.

## Summary

Larger models can combine columns the sentence leaves apart. This path reports $F_1 = 488/861$ for one sentence on 1417 held-out customers. Stay-for-everyone still has the higher held-out accuracy, $1052/1417$, and recall 0.
