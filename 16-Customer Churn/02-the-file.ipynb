# Twenty-one columns, one label

Each row is one customer. The first field is `customerID`. The last field is `Churn`. Between them are the contract, the tenure in months, the services, the payment method, and two charges.

`SeniorCitizen` is stored as `0` or `1`. The other yes/no fields are stored as the words `Yes` and `No`. Add-on fields also use a third phrase when the customer has no internet service, or no phone service. That third phrase is a description of the plan, and it is the same people in every add-on column.

`MonthlyCharges` and `TotalCharges` are written as text. Eleven totals are a single space. The next lesson reads one whole row. This lesson checks the shape.


```python
# Locate the telco table beside this notebook, or under the course root.
import csv
from pathlib import Path

def data_path():
    here = Path.cwd()
    name = "Telco-Customer-Churn.csv"
    for folder in [here, *here.parents]:
        nested = folder / "16-Customer Churn" / "data" / name
        direct = folder / "data" / name
        if nested.exists():
            return nested
        if direct.exists():
            return direct
    raise FileNotFoundError(name)

def read_customers():
    with data_path().open(newline="", encoding="utf-8") as handle:
        return list(csv.DictReader(handle))
# Twenty-one names, 7043 distinct ids, and the two label words.
rows = read_customers()
columns = list(rows[0])
print(len(columns))
print(columns)
assert len(columns) == 21
assert columns[0] == "customerID" and columns[-1] == "Churn"
assert len({row["customerID"] for row in rows}) == 7043
assert {row["Churn"] for row in rows} == {"Yes", "No"}
assert {row["SeniorCitizen"] for row in rows} == {"0", "1"}
assert {row["gender"] for row in rows} == {"Female", "Male"}
internet_none = sum(row["InternetService"] == "No" for row in rows)
for name in (
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
):
    phrase = sum(row[name] == "No internet service" for row in rows)
    assert phrase == internet_none == 1526
phone_none = sum(row["PhoneService"] == "No" for row in rows)
lines = sum(row["MultipleLines"] == "No phone service" for row in rows)
assert phone_none == lines == 682
```

## Pitfall

`OnlineSecurity = No` and `OnlineSecurity = No internet service` are different cells. The first customer has internet and declined security. The second has no internet plan, so the security field is recording that fact. Pooling the two phrases into one "no" mixes two situations.

## Summary

The header has 21 names. There are 7043 distinct customer ids. `Churn` is `Yes` or `No`. `SeniorCitizen` is `0` or `1`. The phrase `No internet service` appears on the same 1526 rows in every add-on column. `No phone service` appears on the same 682 rows as `PhoneService = No`.
