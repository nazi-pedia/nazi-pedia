# Customer 7590-VHVEG

The first row is customer `7590-VHVEG`. Read it before any rate.

She is recorded as female, not a senior, with a partner and no dependents. Tenure is 1 month. She has no phone, so `MultipleLines` says `No phone service`. Internet is DSL. Online security is `No`, online backup is `Yes`, device protection is `No`, tech support is `No`, and both streaming fields are `No`. The contract is month-to-month. Billing is paperless. Payment is an electronic check. Monthly charges are 29.85, and the total charges are 29.85. `Churn` is `No`: she stayed.

One month at 29.85 is 29.85. On this row the total matches the product. Later rows show that this match is special.


```python
# Locate the telco table beside this notebook, or under the course root.
import csv
from pathlib import Path

def data_path():
    here = Path.cwd()
    name = "Telco-Customer-Churn.csv"
    for folder in [here, *here.parents]:
        nested = folder / "16-Customer Churn" / "data" / name
        direct = folder / "data" / name
        if nested.exists():
            return nested
        if direct.exists():
            return direct
    raise FileNotFoundError(name)

def read_customers():
    with data_path().open(newline="", encoding="utf-8") as handle:
        return list(csv.DictReader(handle))
# The first record, field by field, including the one-month bill.
rows = read_customers()
row = rows[0]
print(row["customerID"], row["Churn"], row["Contract"], row["tenure"])
assert row["customerID"] == "7590-VHVEG"
assert row["gender"] == "Female"
assert row["SeniorCitizen"] == "0"
assert row["Partner"] == "Yes" and row["Dependents"] == "No"
assert row["tenure"] == "1"
assert row["PhoneService"] == "No"
assert row["MultipleLines"] == "No phone service"
assert row["InternetService"] == "DSL"
assert row["OnlineSecurity"] == "No" and row["OnlineBackup"] == "Yes"
assert row["DeviceProtection"] == "No" and row["TechSupport"] == "No"
assert row["StreamingTV"] == "No" and row["StreamingMovies"] == "No"
assert row["Contract"] == "Month-to-month"
assert row["PaperlessBilling"] == "Yes"
assert row["PaymentMethod"] == "Electronic check"
assert row["MonthlyCharges"] == "29.85" and row["TotalCharges"] == "29.85"
assert row["Churn"] == "No"
assert round(float(row["MonthlyCharges"]) * 100) * int(row["tenure"]) == 2985
```

## Pitfall

This row is a stay. A later rule that flags month-to-month electronic-check customers will flag her anyway. One row is a case, and a rate is a count over many cases. Her id is also on the fit side of the split in lesson 11, so she is part of choosing the rule. She is not one of the 1417 scores.

## Summary

`7590-VHVEG` stayed after 1 month of DSL, with no phone, a month-to-month contract, and an electronic check. The bill is 29.85 for the month and 29.85 in total.
