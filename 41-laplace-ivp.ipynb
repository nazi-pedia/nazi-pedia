# ۴۱ · لاپلاس و مسئلهٔ مقدار اولیه

| | |
|---|---|
| **دوره** | معادلات دیفرانسیل |
| **پیش‌نیاز** | ۱۸ و ۳۹ |

\[
\mathcal{L}\{f\}(s)=\int_0^{\infty}e^{-st}f(t)\,dt
\]

خاصیت اصلی، تبدیل مشتق به جبر است.

\[
\mathcal{L}\{y'\}=sY-y(0)
\]
\[
\mathcal{L}\{y''\}=s^2 Y-sy(0)-y'(0)
\]

جدول پایه، با \(a>0\) و \(s\) به‌قدر کافی بزرگ:

\[
\mathcal{L}\{1\}=\frac{1}{s},\quad
\mathcal{L}\{e^{at}\}=\frac{1}{s-a},\quad
\mathcal{L}\{\sin at\}=\frac{a}{s^2+a^2},\quad
\mathcal{L}\{\cos at\}=\frac{s}{s^2+a^2}
\]

مسئله را تبدیل کنید، \(Y\) را با کسر جزئی حل کنید، برگردانید.



## مثال ۱ — تعریف، نه فقط دکمه

### حل ریاضی

\[
\mathcal{L}\{1\}=\int_0^{\infty}e^{-st}\,dt=\frac{1}{s},\qquad s>0
\]

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t", nonnegative=True)
s = sp.symbols("s", positive=True)
by_integral = sp.integrate(sp.exp(-s*t), (t, 0, sp.oo))
by_library = sp.laplace_transform(1, t, s)[0]
print(by_integral, by_library)
assert by_integral == 1/s and by_library == 1/s
print("تایید شد")

```

## مثال ۲ — معادلهٔ مرتبهٔ اول

### حل ریاضی

\(y'+y=1\)، \(y(0)=0\).

\[
sY+Y=\frac{1}{s}
\]
\[
Y=\frac{1}{s(s+1)}=\frac{1}{s}-\frac{1}{s+1}
\]
\[
y=1-e^{-t}
\]

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t", nonnegative=True)
s = sp.symbols("s", positive=True)
Y = 1/(s*(s + 1))
y = sp.inverse_laplace_transform(Y, s, t)
hand = (1 - sp.exp(-t))*sp.Heaviside(t)
print(y)
assert sp.simplify(y - hand) == 0
print("تایید شد")

```

## مثال ۳ — نوسان از روی لاپلاس

### حل ریاضی

\(y''+y=0\)، \(y(0)=1\)، \(y'(0)=0\).

\[
s^2 Y-s+Y=0
\]
\[
Y=\frac{s}{s^2+1}
\]
\[
y=\cos t
\]

همان جواب درس ۳۹. تبدیل، شرط اولیه را از همان اول بلعیده است.

### بازبینی با پایتون



```python
import sympy as sp
t, s = sp.symbols("t s")
Y = s/(s**2 + 1)
y = sp.inverse_laplace_transform(Y, s, t)
print(y)
assert sp.simplify(y - sp.cos(t)*sp.Heaviside(t)) == 0
print("تایید شد")

```

## مثال ۴ — جدول نمایی

### حل ریاضی

\[
\mathcal{L}\{e^{2t}\}=\frac{1}{s-2}
\]

این همان عامل رشد درس ۳۸ است. برگردان \(\dfrac{1}{s-2}\) برابر \(e^{2t}\) برای \(t>0\) است.

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t", nonnegative=True)
s = sp.symbols("s", positive=True)
found = sp.laplace_transform(sp.exp(2*t), t, s)[0]
print(found)
assert sp.simplify(found - 1/(s - 2)) == 0
print("تایید شد")

```

## نکته

`laplace_transform` سه‌تایی برمی‌گرداند: تبدیل، نیم‌صفحهٔ همگرایی، و شرط.
تبدیل، مؤلفهٔ اول است. اگر سه‌تایی را به‌جای \(Y\) جبر کنید، کسر جزئی بی‌معنا می‌شود.
