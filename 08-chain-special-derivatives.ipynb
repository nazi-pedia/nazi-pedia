اگر خروجی یک تابع، ورودی تابع دیگری باشد، شیب‌ها در هم ضرب می‌شوند.

\[
(f(g(x)))'=f'(g(x))\,g'(x)
\]

جدول این درس را حفظ کنید و هر بار زنجیره را جدا بنویسید.

\[
\begin{align*}
(\sin u)'&=\cos u\cdot u'\\
(\cos u)'&=-\sin u\cdot u'\\
(e^u)'&=e^u\cdot u'\\
(\ln u)'&=\frac{1}{u}u'\\
(\arctan u)'&=\frac{1}{1+u^2}u'\\
(\arcsin u)'&=\frac{1}{\sqrt{1-u^2}}u'
\end{align*}
\]



```python
import sympy as sp

def check(found, hand, label="نتیجه"):
    found_s = sp.simplify(found)
    hand_s = sp.simplify(hand)
    difference = sp.simplify(found_s - hand_s)
    print(label)
    print("  حل دستی:", hand_s)
    print("  پایتون:", found_s)
    print("  اختلاف:", difference)
    assert difference == 0
    print("  تایید شد")

```

## مثال ۱ — سینوسِ یک مربع

### حل ریاضی

بیرون سینوس است و درون \(x^2\).

\[
\frac{d}{dx}\sin(x^2)=\cos(x^2)\cdot 2x
\]

اگر \(2x\) را جا بگذارید، مشتقِ سینوسِ \(x\) را نوشته‌اید، نه سینوسِ \(x^2\) را.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
found = sp.diff(sp.sin(x**2), x)
hand = 2*x*sp.cos(x**2)
assert sp.simplify(found - hand) == 0
print(found)
print("تایید شد")

```

## مثال ۲ — نمایی با ضریب داخل توان

### حل ریاضی

\[
\frac{d}{dx}e^{3x}=e^{3x}\cdot 3=3e^{3x}
\]

ضریب داخل توان، بعد از مشتق، به صورت ضریب بیرون می‌آید.
\(\dfrac{d}{dx}e^{x^2}=e^{x^2}\cdot 2x\) همین قاعده است با درونِ متفاوت.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
assert sp.diff(sp.exp(3*x), x) == 3*sp.exp(3*x)
assert sp.simplify(sp.diff(sp.exp(x**2), x) - 2*x*sp.exp(x**2)) == 0
print("e^(3x) ->", sp.diff(sp.exp(3*x), x))
print("e^(x^2) ->", sp.diff(sp.exp(x**2), x))
print("تایید شد")

```

## مثال ۳ — لگاریتم یک چندجمله‌ای

### حل ریاضی

\[
\frac{d}{dx}\ln(x^2+1)=\frac{1}{x^2+1}\cdot 2x=\frac{2x}{x^2+1}
\]

دامنهٔ واقعی همه‌جا است، چون \(x^2+1>0\).
اگر درون می‌توانست منفی شود، دامنهٔ لگاریتم را پیش از مشتق می‌نوشتیم.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", real=True)
found = sp.diff(sp.log(x**2 + 1), x)
hand = 2*x/(x**2 + 1)
print(sp.simplify(found))
assert sp.simplify(found - hand) == 0
print("تایید شد")

```

## مثال ۴ — آرک‌تانژانت و آرک‌سینوس

### حل ریاضی

\[
\frac{d}{dx}\arctan x=\frac{1}{1+x^2}
\]

\[
\frac{d}{dx}\arcsin(2x)=\frac{1}{\sqrt{1-(2x)^2}}\cdot 2=\frac{2}{\sqrt{1-4x^2}}
\]

دامنهٔ دومی \(|2x|\le 1\) یعنی \(|x|\le 1/2\) است، و در دو سر، مشتق نامتناهی می‌شود.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
assert sp.diff(sp.atan(x), x) == 1/(1 + x**2)
found = sp.diff(sp.asin(2*x), x)
hand = 2/sp.sqrt(1 - 4*x**2)
print(sp.simplify(found))
assert sp.simplify(found - hand) == 0
print("تایید شد")

```

## مثال ۵ — زنجیره روی یک ضرب

### حل ریاضی

\[
f(x)=e^{x}\sin x
\]

اول ضرب، و داخل آن مشتق هر عامل معمولی است.

\[
f'(x)=e^{x}\sin x+e^{x}\cos x=e^{x}(\sin x+\cos x)
\]

اگر بعداً \(e^{\sin x}\) باشد، آن وقت زنجیره لازم می‌شود: \(e^{\sin x}\cos x\).
این دو تابع را با هم عوض نکنید.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
product = sp.diff(sp.exp(x)*sp.sin(x), x)
chain = sp.diff(sp.exp(sp.sin(x)), x)
assert sp.simplify(product - sp.exp(x)*(sp.sin(x) + sp.cos(x))) == 0
assert sp.simplify(chain - sp.exp(sp.sin(x))*sp.cos(x)) == 0
print("ضرب:", sp.simplify(product))
print("زنجیره:", sp.simplify(chain))
print("تایید شد")

```

## نکته

بیرون را مشتق کنید و درون را دست‌نخورده بگذارید. بعد در مشتق درون ضرب کنید.
جا انداختن مشتق درون، رایج‌ترین خطای این درس است.
