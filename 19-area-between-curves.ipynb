# ۱۹ · مساحت بین دو خم

| | |
|---|---|
| **دوره** | ریاضی عمومی ۲ |
| **پیش‌نیاز** | ۱۶ |

مساحت بین \(y=f(x)\) و \(y=g(x)\) از \(a\) تا \(b\)، اگر در این بازه \(f\ge g\)،

\[
\int_a^b\bigl(f(x)-g(x)\bigr)\,dx
\]

است. تفاضل را طوری بنویسید که ارتفاع نوار مثبت باشد.
اگر خم‌ها جای خود را عوض کنند، بازه را در نقطهٔ برخورد بشکنید.



## مثال ۱ — سهمی و خط

### حل ریاضی

\(y=x\) و \(y=x^2\) در ۰ و ۱ به هم می‌رسند، چون \(x^2-x=x(x-1)=0\).
روی \([0,1]\) خط بالای سهمی است.

\[
\int_0^1(x-x^2)\,dx=\Bigl[\frac{x^2}{2}-\frac{x^3}{3}\Bigr]_0^1=\frac{1}{2}-\frac{1}{3}=\frac{1}{6}
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
crossings = sp.solve(sp.Eq(x, x**2), x)
area = sp.integrate(x - x**2, (x, 0, 1))
print(crossings, area)
assert crossings == [0, 1]
assert area == sp.Rational(1, 6)
print("تایید شد")

```

## مثال ۲ — سینوس و کسینوس

### حل ریاضی

روی \([0,\pi/4]\) کسینوس از سینوس بزرگ‌تر است، چون \(\tan x\le 1\).

\[
\int_0^{\pi/4}(\cos x-\sin x)\,dx
=\Bigl[\sin x+\cos x\Bigr]_0^{\pi/4}
=\left(\frac{\sqrt{2}}{2}+\frac{\sqrt{2}}{2}\right)-(0+1)
=\sqrt{2}-1
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
area = sp.integrate(sp.cos(x) - sp.sin(x), (x, 0, sp.pi/4))
print(area)
assert sp.simplify(area - (sp.sqrt(2) - 1)) == 0
print("تایید شد")

```

## مثال ۳ — زیر یک سهمی تا محور

### حل ریاضی

\(y=4-x^2\) محور را در \(x=\pm 2\) قطع می‌کند و روی این بازه نامنفی است.

\[
\int_{-2}^{2}(4-x^2)\,dx=2\int_0^2(4-x^2)\,dx
=2\Bigl[4x-\frac{x^3}{3}\Bigr]_0^2
=2\left(8-\frac{8}{3}\right)=\frac{32}{3}
\]

تقارن زوج، نیمهٔ کار را حذف کرد.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
area = sp.integrate(4 - x**2, (x, -2, 2))
print(area)
assert area == sp.Rational(32, 3)
print("تایید شد")

```

## مثال ۴ — قدر مطلق، وقتی ترتیب عوض می‌شود

### حل ریاضی

\(y=x\) و \(y=x^3\) روی \([-1,1]\).
ریشه‌های \(x-x^3=x(1-x^2)=0\) عبارت‌اند از \(-1,0,1\).
روی \([0,1]\) خط بالای خم درجهٔ سه است. روی \([-1,0]\) برعکس، چون هر دو منفی‌اند و \(x^3\) پایین‌تر نیست:
برای \(x=-1/2\)، \(x=-1/2\) و \(x^3=-1/8\). پس \(x^3>x\) روی منفی‌ها.
مساحت:

\[
\int_{-1}^{0}(x^3-x)\,dx+\int_0^1(x-x^3)\,dx
\]

هر کدام \(\dfrac{1}{4}\) است. جمع \(\dfrac{1}{2}\).
یک انتگرال بدون قدر مطلق روی کل \([-1,1]\) صفر می‌شود، چون دو بخش همدیگر را خط می‌زنند. مساحت صفر نیست.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
signed = sp.integrate(x - x**3, (x, -1, 1))
area = sp.integrate(sp.Abs(x - x**3), (x, -1, 1))
print("علامت‌دار:", signed, "مساحت:", area)
assert signed == 0
assert area == sp.Rational(1, 2)
print("تایید شد")

```

## نکته

انتگرال علامت‌دار، مساحت نیست.
اگر شکل را نمی‌کشید، دست‌کم در یک نقطهٔ آزمایشی بگویید کدام تابع بزرگ‌تر است.

## جمع‌بندی

نقطهٔ برخورد، کران است.
تفاضل باید مثبت باشد.
تقارن را بعد از نوشتن انتگرال خرج کنید، نه به‌جای آن.
