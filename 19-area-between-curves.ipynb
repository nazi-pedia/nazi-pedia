The area between $y=f(x)$ and $y=g(x)$ from $a$ to $b$, when $f\ge g$ on that interval, is

$$
\int_{a}^{b}\bigl(f(x)-g(x)\bigr)\,dx.
$$

Write the difference so that every vertical strip has positive height. If the curves trade places, split the interval at the intersection.



## Example 1 — Parabola and line

### Hand solution

The curves $y=x$ and $y=x^{2}$ meet at $0$ and $1$, because $x^{2}-x=x(x-1)=0$. On $[0,1]$ the line lies above the parabola.

$$
\begin{aligned}
\int_{0}^{1}(x-x^{2})\,dx
&= \left[\dfrac{x^{2}}{2}-\dfrac{x^{3}}{3}\right]_{0}^{1} \\
&= \dfrac{1}{2}-\dfrac{1}{3} \\
&= \dfrac{1}{6}.
\end{aligned}
$$

### Python recheck



```python
import sympy as sp

# Locate the intersections of the line and the parabola.
x = sp.symbols("x")
crossings = sp.solve(sp.Eq(x, x**2), x)

# Integrate the positive height from 0 to 1.
area = sp.integrate(x - x**2, (x, 0, 1))
print(crossings, area)

# Check the intersection points and the area.
assert crossings == [0, 1]
assert area == sp.Rational(1, 6)
print("Confirmed")

```

## Example 2 — Sine and cosine

### Hand solution

On $\left[0,\dfrac{\pi}{4}\right]$, cosine is greater than sine, because $\tan x\le 1$.

$$
\begin{aligned}
\int_{0}^{\pi/4}(\cos x-\sin x)\,dx
&= \bigl[\sin x+\cos x\bigr]_{0}^{\pi/4} \\
&= \left(\dfrac{\sqrt{2}}{2}+\dfrac{\sqrt{2}}{2}\right)-(0+1) \\
&= \sqrt{2}-1.
\end{aligned}
$$

### Python recheck



```python
import sympy as sp

# Integrate cosine minus sine up to pi/4.
x = sp.symbols("x")
area = sp.integrate(sp.cos(x) - sp.sin(x), (x, 0, sp.pi/4))
print(area)

# The exact area is sqrt(2) - 1.
assert sp.simplify(area - (sp.sqrt(2) - 1)) == 0
print("Confirmed")

```

## Example 3 — Under a parabola, down to the axis

### Hand solution

The curve $y=4-x^{2}$ meets the axis at $x=\pm 2$ and is nonnegative on that interval.

$$
\begin{aligned}
\int_{-2}^{2}(4-x^{2})\,dx
&= 2\int_{0}^{2}(4-x^{2})\,dx \\
&= 2\left[4x-\dfrac{x^{3}}{3}\right]_{0}^{2} \\
&= 2\left(8-\dfrac{8}{3}\right) \\
&= \dfrac{32}{3}.
\end{aligned}
$$

Even symmetry removes half of the work.

### Python recheck



```python
import sympy as sp

# Integrate the nonnegative parabola from -2 to 2.
x = sp.symbols("x")
area = sp.integrate(4 - x**2, (x, -2, 2))
print(area)

# The area is 32/3.
assert area == sp.Rational(32, 3)
print("Confirmed")

```

## Example 4 — Absolute value, when the order changes

### Hand solution

Take $y=x$ and $y=x^{3}$ on $[-1,1]$. The roots of $x-x^{3}=x(1-x^{2})=0$ are $-1$, $0$, and $1$. On $[0,1]$ the line lies above the cubic. On $[-1,0]$ the order reverses, because both graphs are negative and $x^{3}$ is not the lower one. At $x=-\dfrac{1}{2}$ the values are $x=-\dfrac{1}{2}$ and $x^{3}=-\dfrac{1}{8}$, so $x^{3}>x$ on the negative side.

The area is

$$
\int_{-1}^{0}(x^{3}-x)\,dx+\int_{0}^{1}(x-x^{3})\,dx.
$$

Each piece equals $\dfrac{1}{4}$. The sum is $\dfrac{1}{2}$.

A single integral with no absolute value, taken over all of $[-1,1]$, equals $0$, because the two pieces cancel. The area is not $0$.

### Python recheck



```python
import sympy as sp

# Compare the signed integral with the absolute area.
x = sp.symbols("x")
signed = sp.integrate(x - x**3, (x, -1, 1))
area = sp.integrate(sp.Abs(x - x**3), (x, -1, 1))
print("Signed:", signed, "Area:", area)

# The signed integral cancels, but the area is 1/2.
assert signed == 0
assert area == sp.Rational(1, 2)
print("Confirmed")

```

## A pitfall

A signed integral is not an area. If you do not sketch the region, at least test one sample point and say which function is larger.

## Summary

An intersection point is a limit of integration. The difference of the two functions must be positive. Use symmetry after the integral is written, not in place of it.
