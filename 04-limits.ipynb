A limit is a number that the outputs of a function approach,
even when the function itself is not defined at that point.
Substitution is a shortcut, and you may use it only when the expression is continuous at the point.
If you arrive at $\dfrac{0}{0}$, substitution has failed and rewriting begins.

## Limit laws

If $\lim f$ and $\lim g$ exist, then the limit of a sum, of a product, and of a quotient
equals the same operation on the limits, provided the limit of the denominator is not zero.
The limit of a polynomial at $a$ equals the value of that polynomial at $a$.

Two standard limits build the rest:

$$
\lim_{x\to 0}\dfrac{\sin x}{x}=1,
\qquad
\lim_{n\to\infty}\left(1+\dfrac{1}{n}\right)^n=e.
$$

Python approaches the point from the right unless you name the direction with `dir`.
You conclude a two-sided limit yourself, from the equality of the two sides. Lesson 06 opens this distinction for $\infty$.


```python
import sympy as sp

# Compare a computed SymPy result with the answer found by hand.
def check(found, hand, label="Result"):
    # Simplify each side first so equivalent algebraic forms can match.
    found_s = sp.simplify(found)
    hand_s = sp.simplify(hand)
    difference = sp.simplify(found_s - hand_s)
    print(label)
    print("  Hand:", hand_s)
    print("  Python:", found_s)
    print("  Difference:", difference)
    # The methods agree only when that difference simplifies to zero.
    assert difference == 0
    print("  Confirmed")
```

## Example 1 — Substitution, when it really is allowed

### Hand solution

$$
\begin{aligned}
\lim_{x\to 2}(x^2+3x-1)
&= 4+6-1 \\
&= 9.
\end{aligned}
$$

A polynomial is continuous everywhere. A table is not required, but a table should approach $9$,
and that approach is not itself the value.

### Python recheck


```python
import sympy as sp

# Polynomial x^2 + 3x - 1. Hand limit at x = 2, and the substituted value, are both 9.
x = sp.symbols("x")
f = x**2 + 3*x - 1
exact = sp.limit(f, x, 2)
print("Limit:", exact)
# Nearby inputs should approach 9; they are not a substitute for the exact value.
for h in (sp.Rational(1, 10), sp.Rational(1, 100)):
    print("x =", 2 + h, "->", f.subs(x, 2 + h))
assert exact == 9
assert f.subs(x, 2) == 9
print("Confirmed")
```

## Example 2 — The indeterminate form $\dfrac{0}{0}$, by factoring

### Hand solution

$$
\lim_{x\to 3}\dfrac{x^2-9}{x-3}.
$$

The numerator and the denominator are both zero at $x=3$. That is not the value of the limit. It is the signal to rewrite.

$$
\begin{aligned}
\dfrac{(x-3)(x+3)}{x-3}
&= x+3, \qquad x\neq 3, \\
\lim_{x\to 3}(x+3)
&= 6.
\end{aligned}
$$

The function stays undefined at $x=3$. The limit is the height of the hole.

### Python recheck


```python
import sympy as sp

# (x^2 - 9)/(x - 3) is 0/0 at x = 3. Hand limit after factoring: 6.
x = sp.symbols("x")
f = (x**2 - 9)/(x - 3)
print("Simplified:", sp.cancel(f))
print("Limit:", sp.limit(f, x, 3))
assert sp.limit(f, x, 3) == 6
print("Confirmed")
```

## Example 3 — Rationalizing

### Hand solution

$$
\lim_{x\to 0}\dfrac{\sqrt{x+1}-1}{x}.
$$

Again the form is $\dfrac{0}{0}$. Multiply by the conjugate:

$$
\begin{aligned}
\dfrac{(\sqrt{x+1}-1)(\sqrt{x+1}+1)}{x(\sqrt{x+1}+1)}
&= \dfrac{(x+1)-1}{x(\sqrt{x+1}+1)} \\
&= \dfrac{1}{\sqrt{x+1}+1}, \\
\lim_{x\to 0}\dfrac{1}{\sqrt{x+1}+1}
&= \dfrac{1}{2}.
\end{aligned}
$$

### Python recheck


```python
import sympy as sp

# Rationalize (sqrt(x + 1) - 1)/x. Hand limit at x = 0 is 1/2.
x = sp.symbols("x")
f = (sp.sqrt(x + 1) - 1)/x
exact = sp.limit(f, x, 0)
print("Limit:", exact)
assert exact == sp.Rational(1, 2)
print("Confirmed")
```

## Example 4 — The sine limit, with a change of scale

### Hand solution

$$
\begin{aligned}
\lim_{x\to 0}\dfrac{\sin 3x}{x}
&= \lim_{x\to 0}3\cdot\dfrac{\sin 3x}{3x} \\
&= 3\cdot 1 \\
&= 3.
\end{aligned}
$$

The factor $3$ is taken from the numerator and the denominator together, so that the argument of the sine matches the denominator.
If you forget $3x$ only in the denominator, the answer becomes $1$, and that answer is wrong.

### Python recheck


```python
import sympy as sp

# sin(3x)/x. Hand limit at 0 is 3, from the standard limit sin(u)/u -> 1.
x = sp.symbols("x")
exact = sp.limit(sp.sin(3*x)/x, x, 0)
print("Limit:", exact)
assert exact == 3
print("Confirmed")
```

## Example 5 — The definition of the number $e$

### Hand solution

$$
\lim_{n\to\infty}\left(1+\dfrac{1}{n}\right)^n=e.
$$

This limit is one definition of the value of $e$.
The sequence increases, and it is bounded above by $e$. A few terms are enough to see the approach,
but no finite term is $e$ itself:

$$
\left(1+\dfrac{1}{1}\right)^1=2,
\qquad
\left(1+\dfrac{1}{100}\right)^{100}\approx 2.70481.
$$

### Python recheck


```python
import sympy as sp

# The sequence (1 + 1/n)^n. Hand limit: e. The terms at n = 1 and n = 100 are not e.
n = sp.symbols("n", integer=True, positive=True)
exact = sp.limit((1 + 1/n)**n, n, sp.oo)
print("Limit:", exact)
print("Term n=1:", (1 + sp.Integer(1)/1)**1)
print("Term n=100:", sp.N((1 + sp.Integer(1)/100)**100, 8))
assert exact == sp.E
print("Confirmed")
```

## A pitfall

`subs` at a hole is not the limit. For

$$
\dfrac{x^2-9}{x-3},
$$

substituting $x=3$ fails.
Compute the limit by rewriting, or with `limit`.

## Summary

Try substitution first.
If you see $\dfrac{0}{0}$, factor, rationalize, or match the argument of the sine to the denominator.
Keep an exact answer as a fraction or as a symbol.
