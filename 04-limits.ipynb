# ۰۴ · حد و قانون‌های حد

| | |
|---|---|
| **دوره** | ریاضی عمومی ۱ |

حد، عددی است که خروجی‌های تابع به آن نزدیک می‌شوند،
حتی اگر خود تابع در آن نقطه تعریف نشده باشد.
جانشینی، میان‌بر است و فقط وقتی حق دارید از آن استفاده کنید که عبارت در نقطه پیوسته باشد.
اگر به \(0/0\) رسیدید، جانشینی شکست خورده و بازنویسی شروع می‌شود.

## قانون‌ها

اگر \(\lim f\) و \(\lim g\) وجود داشته باشند، حد جمع، ضرب، و خارج‌قسمت
(به شرط مخرج ناصفر) برابر است با همان عمل روی حدها.
حد چندجمله‌ای در \(a\) برابر خود چندجمله‌ای در \(a\) است.

دو حد استاندارد که بقیه را می‌سازند:

\[
\lim_{x\to 0}\frac{\sin x}{x}=1,
\qquad
\lim_{n\to\infty}\left(1+\frac{1}{n}\right)^n=e
\]

پایتون از سمت راست به نقطه نزدیک می‌شود، مگر جهت را با `dir` مشخص کنید.
حد دوطرفه را خودتان از برابری دو سمت نتیجه بگیرید. درس ۰۶ این تفاوت را برای بی‌نهایت باز می‌کند.



```python
import sympy as sp

def check(found, hand, label="نتیجه"):
    found_s = sp.simplify(found)
    hand_s = sp.simplify(hand)
    difference = sp.simplify(found_s - hand_s)
    print(label)
    print("  حل دستی:", hand_s)
    print("  پایتون:", found_s)
    print("  اختلاف:", difference)
    assert difference == 0
    print("  تایید شد")

```

## مثال ۱ — جانشینی، وقتی واقعاً مجاز است

### حل ریاضی

\[
\lim_{x\to 2}(x^2+3x-1)=4+6-1=9
\]

چندجمله‌ای همه‌جا پیوسته است. جدول لازم نیست، ولی جدول باید به ۹ نزدیک شود
و این نزدیکی، خودِ مقدار نیست.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = x**2 + 3*x - 1
exact = sp.limit(f, x, 2)
print("حد:", exact)
for h in (sp.Rational(1, 10), sp.Rational(1, 100)):
    print("x =", 2 + h, "->", f.subs(x, 2 + h))
assert exact == 9
assert f.subs(x, 2) == 9
print("تایید شد")

```

## مثال ۲ — شکل مبهم \(0/0\) با تجزیه‌کردن

### حل ریاضی

\[
\lim_{x\to 3}\frac{x^2-9}{x-3}
\]

صورت و مخرج در \(x=3\) صفرند. این عددِ حد نیست؛ علامتِ «بازنویسی کن» است.

\[
\frac{(x-3)(x+3)}{x-3}=x+3,\qquad x\neq 3
\]

\[
\lim_{x\to 3}(x+3)=6
\]

تابع در \(x=3\) تعریف نشده می‌ماند. حد، ارتفاع سوراخ است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = (x**2 - 9)/(x - 3)
print("ساده:", sp.cancel(f))
print("حد:", sp.limit(f, x, 3))
assert sp.limit(f, x, 3) == 6
print("تایید شد")

```

## مثال ۳ — گویاکردن

### حل ریاضی

\[
\lim_{x\to 0}\frac{\sqrt{x+1}-1}{x}
\]

باز هم \(0/0\). در مزدوج ضرب می‌کنیم.

\[
\frac{(\sqrt{x+1}-1)(\sqrt{x+1}+1)}{x(\sqrt{x+1}+1)}
=\frac{(x+1)-1}{x(\sqrt{x+1}+1)}
=\frac{1}{\sqrt{x+1}+1}
\]

\[
\lim_{x\to 0}\frac{1}{\sqrt{x+1}+1}=\frac{1}{2}
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = (sp.sqrt(x + 1) - 1)/x
exact = sp.limit(f, x, 0)
print("حد:", exact)
assert exact == sp.Rational(1, 2)
print("تایید شد")

```

## مثال ۴ — سینوس، با تغییر مقیاس

### حل ریاضی

\[
\lim_{x\to 0}\frac{\sin 3x}{x}
=\lim_{x\to 0}3\cdot\frac{\sin 3x}{3x}
=3\cdot 1=3
\]

ضریب ۳ از صورت و مخرج با هم می‌آید تا شناسهٔ سینوس و مخرج یکی شوند.
اگر فقط \(3x\) را در مخرج فراموش کنید، جواب ۱ می‌شود و غلط است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
exact = sp.limit(sp.sin(3*x)/x, x, 0)
print("حد:", exact)
assert exact == 3
print("تایید شد")

```

## مثال ۵ — تعریف عدد \(e\)

### حل ریاضی

\[
\lim_{n\to\infty}\left(1+\frac{1}{n}\right)^n=e
\]

این حد، تعریف یکی از مقدارهای \(e\) است.
دنباله صعودی و از بالا به \(e\) کراندار است؛ چند جمله برای دیدن نزدیک‌شدن کافی است،
ولی هیچ جملهٔ متناهی خودِ \(e\) نیست.

\[
\left(1+\frac{1}{1}\right)^1=2,\qquad
\left(1+\frac{1}{100}\right)^{100}\approx 2.70481
\]

### بازبینی با پایتون



```python
import sympy as sp
n = sp.symbols("n", integer=True, positive=True)
exact = sp.limit((1 + 1/n)**n, n, sp.oo)
print("حد:", exact)
print("جمله n=1:", (1 + sp.Integer(1)/1)**1)
print("جمله n=100:", sp.N((1 + sp.Integer(1)/100)**100, 8))
assert exact == sp.E
print("تایید شد")

```

## نکته

`subs` در نقطهٔ سوراخ، حد نیست. برای \(\dfrac{x^2-9}{x-3}\) جانشینی \(x=3\) شکست می‌خورد.
حد را یا با بازنویسی حساب کنید یا با `limit`.
