# ۴۰ · ضرایب نامعین و تغییر پارامتر

| | |
|---|---|
| **دوره** | معادلات دیفرانسیل |
| **پیش‌نیاز** | ۳۹ |

معادلهٔ ناهمگن \(y''+py'+qy=g\).
جواب عمومی، جواب همگن به‌علاوه یک جواب خصوصی است.

ضرایب نامعین وقتی \(g\) نمایی، چندجمله‌ای، سینوس یا کسینوس است.
شکل \(g\) را حدس بزنید. اگر آن شکل در جواب همگن بود، در \(x\) ضرب کنید و اگر باز بود، یک بار دیگر.

تغییر پارامتر برای \(g\) کلی‌تر است. اگر \(y_1,y_2\) پایهٔ همگن باشند و رونسکین \(W\)،

\[
u_1'=-\frac{y_2 g}{W},\qquad u_2'=\frac{y_1 g}{W}
\]

و \(y_p=u_1 y_1+u_2 y_2\).



## مثال ۱ — نمایی که ریشه نیست

### حل ریاضی

\(y''-3y'+2y=e^{3x}\).
همگن: \(C_1 e^{x}+C_2 e^{2x}\).
حدس \(Ae^{3x}\).

\[
(9A-9A+2A)e^{3x}=e^{3x}\implies A=\frac{1}{2}
\]

\[
y=C_1 e^{x}+C_2 e^{2x}+\frac{1}{2}e^{3x}
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
y = sp.Function("y")
ode = sp.Eq(y(x).diff(x, 2) - 3*y(x).diff(x) + 2*y(x), sp.exp(3*x))
sol = sp.dsolve(ode, y(x))
particular = sp.exp(3*x)/2
residual = sp.diff(particular, x, 2) - 3*sp.diff(particular, x) + 2*particular
print(sol)
assert sp.simplify(residual - sp.exp(3*x)) == 0
print("تایید شد")

```

## مثال ۲ — تشدید

### حل ریاضی

\(y''+y=\sin x\).
همگن سینوس و کسینوس است، پس \(\sin x\) خودش مجاز نیست.
حدس \(y_p=x(A\cos x+B\sin x)\).
حساب استاندارد به \(A=-\dfrac{1}{2}\) و \(B=0\) می‌رسد.

\[
y_p=-\frac{x}{2}\cos x
\]

آزمون کوتاه: مشتق دوم به‌علاوه خود تابع باید \(\sin x\) شود.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
yp = -x*sp.cos(x)/2
residual = sp.simplify(sp.diff(yp, x, 2) + yp)
print(residual)
assert residual == sp.sin(x)
print("تایید شد")

```

## مثال ۳ — تغییر پارامتر

### حل ریاضی

\(y''+y=\sec x\). پایه: \(\cos x\) و \(\sin x\). رونسکین ۱ است.

\[
u_1'=-\sin x\sec x=-\tan x,\qquad u_2'=\cos x\sec x=1
\]

\[
u_1=\ln|\cos x|,\qquad u_2=x
\]

\[
y_p=\cos x\ln|\cos x|+x\sin x
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
yp = sp.cos(x)*sp.log(sp.cos(x)) + x*sp.sin(x)
residual = sp.simplify(sp.diff(yp, x, 2) + yp)
print(residual)
assert sp.simplify(residual - 1/sp.cos(x)) == 0
print("تایید شد")

```

## مثال ۴ — چندجمله‌ای

### حل ریاضی

\(y''-y=x\).
حدس خصوصی \(y_p=Ax+B\). آن‌گاه \(-Ax-B=x\)، پس \(A=-1\) و \(B=0\).

\[
y=C_1 e^{x}+C_2 e^{-x}-x
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
yp = -x
assert sp.diff(yp, x, 2) - yp == x
print("باقیمانده برابر x است")
print("تایید شد")

```

## نکته

جواب خصوصی را با جواب عمومی قاطی نکنید و ثابت‌های همگن را از روی \(y_p\) تعیین نکنید.
اول \(y_p\) را در معادله بگذارید، بعد شرط اولیه را روی مجموع بگذارید.

## جمع‌بندی

شکل \(g\) شکل حدس را می‌گوید.
حضور آن شکل در همگن، یک ضریب \(x\) می‌خواهد.
تغییر پارامتر وقتی حدس تمام شد به کار می‌آید.
