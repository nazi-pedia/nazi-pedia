# The extra sum of squares

Nested polynomials are compared by the residual sum the higher degree still explains. Under normal errors, that extra sum, divided by the larger model's variance estimate, is an $F$ statistic. For one added coefficient it is the square of a $t$ statistic.


## The quadratic term against the line

Let $\mathrm{RSS}_1 = 136$ be the line and $\mathrm{RSS}_2 = 10$ the quadratic. The extra sum of squares for the one new coefficient is

$$
\mathrm{RSS}_1 - \mathrm{RSS}_2 = 126.
$$

The larger model has $n - p = 5 - 3 = 2$ residual degrees of freedom, so

$$
F = \dfrac{126/1}{10/2} = \dfrac{126}{5} = 25.2.
$$

The reference distribution, if the true coefficient of $x^2$ is zero and the errors are normal, is $F_{1, 2}$. Equivalently, the $t$ statistic for $b_2$ in the quadratic fit will square to this $F$. The standard-error lesson computes that $t$ as $\sqrt{126/5}$. The $0.95$ quantile of $F_{1, 2}$ is the square of the $0.975$ quantile of $t_2$, about $4.303^2 \approx 18.51$. Because $126/5 > 18.51$, a $5\%$ test rejects a quadratic coefficient of zero.

The same arithmetic does not reject every coefficient in the quadratic. That finer split is the inference lesson. Here the hypothesis is specifically "the line is enough."


```python
# F = 126/5 for the quadratic term, and it exceeds the F(1, 2) critical value.
from fractions import Fraction
import scipy.stats as stats
extra = 136 - 10
f_stat = Fraction(extra, 1) / Fraction(10, 2)
critical = stats.f.ppf(0.95, 1, 2)
t_critical = stats.t.ppf(0.975, 2)
print("F", float(f_stat), "critical", critical)
print("t critical squared", t_critical ** 2)
assert extra == 126
assert f_stat == Fraction(126, 5)
assert abs(t_critical ** 2 - critical) < 1e-9
assert float(f_stat) > critical
```

## Sequential sums on an orthogonal design

Start from the intercept alone. Its residual sum is $\mathrm{TSS} = 176$. Adding the slope reduces it to $136$, an extra sum of $40$. Adding $x^2$ reduces it from $136$ to $10$, an extra sum of $126$.

These two extra sums do not depend on the order of the two powers, because $x$ is orthogonal to $x^2$. The slope coefficient is $2$ whether or not $x^2$ is present, and the extra sum for the slope, tested against the quadratic residual variance, is

$$
F_{\text{linear}} = \dfrac{40/1}{10/2} = 8.
$$

$F_{1,2}$ still has critical value about $18.51$, so $8$ does not reject a zero slope inside the quadratic model. The curvature is detectable in this sample. The tilt, although its least-squares value is $2$, is not, once only two residual degrees of freedom remain. A test that fails to reject is a statement about this sample size, not a proof that the slope was zero in the polynomial that built the table.


```python
# Sequential extra sums: 40 for the slope, 126 for the square. Only 126 clears F(1, 2).
import scipy.stats as stats
critical = stats.f.ppf(0.95, 1, 2)
f_slope = (40 / 1) / (10 / 2)
f_square = (126 / 1) / (10 / 2)
print("F slope", f_slope, "F square", f_square, "critical", critical)
assert f_slope == 8
assert abs(f_square - 25.2) < 1e-12
assert f_slope < critical < f_square
```

## Pitfall

The extra sum $126$ is not free evidence if the degree was chosen by searching many powers and then testing only the winner as though it had been named in advance. This lesson named degree $2$ against degree $1$ before looking at a menu of higher degrees. The cubic lesson will show that these same five points lie on a cubic. That fact is a property of the table. It is not, by itself, a fifth-degree search that happened to stop at a small $p$-value.


```python
# The comparison named here is one pair of nested degrees, not a search.
named_degrees = (1, 2)
print("nested comparison", named_degrees)
assert named_degrees == (1, 2)
```

## Summary

- Adding $x^2$ to the line frees $126$ residual sum of squares. $F = 126/5$, which exceeds the $5\%$ critical value of $F_{1,2}$.
- Adding the slope to the intercept, judged from the quadratic fit, gives $F = 8$, which does not exceed that critical value.
- On an orthogonal design the extra sums do not depend on which of $x$ and $x^2$ is entered first.
