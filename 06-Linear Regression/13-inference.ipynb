# Tests and confidence intervals

With normal errors, a coefficient divided by its standard error follows a $t$ distribution. That one fact produces a test, a confidence interval, and, for a single slope, the $F$ statistic of the whole line.


## The reference distribution

Add to the Gauss–Markov assumptions the requirement that $\varepsilon$ is multivariate normal. Then $\widehat\beta$ is normal, and for any coordinate $\beta_j$,

$$
t = \dfrac{\widehat\beta_j - \beta_j}{\operatorname{se}(\widehat\beta_j)}
$$

follows Student's $t$ distribution on $n - p$ degrees of freedom when the displayed $\beta_j$ is the true one. A level-$1-\alpha$ confidence interval is

$$
\widehat\beta_j \pm t_{n-p,\, 1-\alpha/2}\,\operatorname{se}(\widehat\beta_j).
$$

For one slope in a simple regression, the square of the $t$ statistic for $H_0\!: b_1 = 0$ equals the $F$ statistic that compares the line with the intercept-only model:

$$
F = \dfrac{\mathrm{SSR}/1}{\mathrm{RSS}/(n-2)} = t^2.
$$


## The five-point slope

The fitted slope is $2$, $\operatorname{se}(b_1) = \sqrt{2/15}$, and $n - p = 3$. The statistic for $H_0\!: \beta_1 = 0$ is

$$
t = \dfrac{2 - 0}{\sqrt{2/15}} = 2\sqrt{\dfrac{15}{2}} = \sqrt{4\cdot\dfrac{15}{2}} = \sqrt{30}.
$$

The matching $F$ statistic is

$$
F = \dfrac{40/1}{4/3} = 30 = t^2.
$$

The $0.975$ quantile of $t_3$ is approximately $3.182$. Because $\sqrt{30} \approx 5.477 > 3.182$, a two-sided $5\%$ test rejects $H_0\!: \beta_1 = 0$. The confidence interval is

$$
2 \pm t_{3,\,0.975}\sqrt{\dfrac{2}{15}}.
$$

Its left endpoint is positive, which is the same rejection written as an interval.

The statistic for the true value used to build related examples, $H_0\!: \beta_1 = 2$, is

$$
t = \dfrac{2-2}{\sqrt{2/15}} = 0.
$$

The interval around $2$ contains $2$, and the test does not reject. A test that rejected the value the fit itself produced would have been evidence of an arithmetic error.


```python
# t = sqrt(30), F = 30, and the 95 percent interval excludes 0 and includes 2.
import math
import scipy.stats as stats
se = math.sqrt(2 / 15)
t_zero = 2 / se
f_stat = (40 / 1) / (4 / 3)
critical = stats.t.ppf(0.975, 3)
lower, upper = 2 - critical * se, 2 + critical * se
print("t for slope 0", t_zero)
print("F", f_stat)
print("t critical", critical)
print("interval", lower, upper)
assert abs(t_zero - math.sqrt(30)) < 1e-12
assert abs(f_stat - 30) < 1e-12
assert abs(t_zero ** 2 - f_stat) < 1e-12
assert t_zero > critical
assert lower > 0 and lower < 2 < upper
```

## The intercept

$\operatorname{se}(b_0) = \sqrt{22/15}$ and the estimate is $1$, so the statistic for $H_0\!: \beta_0 = 0$ is

$$
t = \dfrac{1}{\sqrt{22/15}} = \sqrt{\dfrac{15}{22}}.
$$

This is about $0.826$, well below $3.182$. The data do not reject a zero intercept. That is not a finding that the intercept is absent. With three residual degrees of freedom, only a large coefficient survives the critical value. The interval

$$
1 \pm t_{3,\,0.975}\sqrt{\dfrac{22}{15}}
$$

contains both negative and positive values.


```python
# The intercept interval is wide and contains zero.
import math
import scipy.stats as stats
se = math.sqrt(22 / 15)
t_stat = 1 / se
critical = stats.t.ppf(0.975, 3)
lower, upper = 1 - critical * se, 1 + critical * se
print("t for intercept 0", t_stat)
print("interval", lower, upper)
assert abs(t_stat - math.sqrt(15 / 22)) < 1e-12
assert t_stat < critical
assert lower < 0 < upper
```

## Pitfall

Failing to reject $H_0\!: \beta_0 = 0$ is not a demonstration that $\beta_0 = 0$. The $t$ statistic $\sqrt{15/22}$ is also compatible with many other values inside the interval. A confidence interval reports that whole set. A test reports only whether one nominated point is inside or outside a level-$5\%$ rejection region.

A second misreading is to treat $\sqrt{30}$ as the number of standard errors from a slope of $2$ to a slope of $0$ and then to call $2$ "significant" without naming the reference distribution. The comparison that makes the word precise is $\sqrt{30}$ against the $t_3$ quantile, not against $2$.


```python
# Not rejected is an interval, not a proof that the coefficient is zero.
import math
import scipy.stats as stats
se = math.sqrt(22 / 15)
critical = stats.t.ppf(0.975, 3)
interval = (1 - critical * se, 1 + critical * se)
print("zero is inside", interval[0] < 0 < interval[1])
print("one is inside", interval[0] < 1 < interval[1])
print("width", interval[1] - interval[0])
assert interval[0] < 0 < 1 < interval[1]
assert interval[1] - interval[0] > 2
```

## Summary

- Under normal errors, $(\widehat\beta_j - \beta_j)/\operatorname{se}(\widehat\beta_j)$ is $t_{n-p}$.
- For the five-point slope, $t = \sqrt{30}$ and $F = 30$. The $95\%$ interval excludes $0$.
- The intercept's $t$ statistic is $\sqrt{15/22}$. Its interval contains $0$, which withholds a claim rather than proving the intercept is zero.
