# The prediction line

A linear regression is a straight line used as a prediction rule. Before any data set enters the argument, the line itself has to be read fluently: the intercept is the prediction at $x = 0$, and the slope is the change in the prediction when $x$ grows by one unit.

This lesson also settles a prior question. If the rule is not allowed to use $x$ at all, the single best constant prediction, in the least-squares sense, is the arithmetic mean. Every later formula is a refinement of that fact.

The rhythm of this path is fixed. Each example is solved by hand, in display mathematics. Python then repeats the same arithmetic and the two answers are required to agree.


## Reading a line

The prediction at a value $x$ is

$$
\widehat y = b_0 + b_1 x.
$$

Here $b_0$ is the intercept and $b_1$ is the slope. Nothing has been estimated yet. These are two numbers we chose, and the formula only evaluates them.

Ada keeps a small table of study hours. Take the rule $\widehat y = 40 + 5x$, where $x$ is hours and $\widehat y$ is the predicted score.

$$
\begin{aligned}
x = 2 &\Longrightarrow \widehat y = 40 + 5\cdot 2 = 50, \\
x = 4 &\Longrightarrow \widehat y = 40 + 5\cdot 4 = 60, \\
x = 6 &\Longrightarrow \widehat y = 40 + 5\cdot 6 = 70.
\end{aligned}
$$

Raising the slope from $5$ to $6$, and leaving the intercept at $40$, moves each prediction by exactly the number of hours:

$$
\begin{aligned}
40 + 6\cdot 2 &= 52 = 50 + 2, \\
40 + 6\cdot 4 &= 64 = 60 + 4, \\
40 + 6\cdot 6 &= 76 = 70 + 6.
\end{aligned}
$$


```python
# Evaluate 40 + 5x by hand, then change only the slope.
hours = [2, 4, 6]
hand = [40 + 5 * x for x in hours]
changed = [40 + 6 * x for x in hours]
print("hand predictions", hand)
print("slope changed to 6", changed)
print("differences", [b - a for a, b in zip(hand, changed)])
assert hand == [50, 60, 70]
assert changed == [52, 64, 76]
assert [b - a for a, b in zip(hand, changed)] == hours
```

## The best constant

Suppose the prediction may not depend on $x$. It is one number $c$, and the cost of using it on observations $y_1,\ldots,y_n$ is the residual sum of squares

$$
S(c) = \sum_{i=1}^{n} (y_i - c)^2.
$$

Differentiate, and use the fact that the second derivative is positive:

$$
\begin{aligned}
S'(c) &= -2\sum_{i=1}^{n}(y_i - c), \\
S'(c) = 0 &\Longrightarrow \sum_{i=1}^{n} y_i - nc = 0 \Longrightarrow c = \bar y, \\
S''(c) &= 2n > 0.
\end{aligned}
$$

So the critical point is a minimum, and it is the sample mean. This is the least-squares fit of a model that contains only an intercept.

Take $y = (2, 4, 6)$. Then

$$
\bar y = \dfrac{2 + 4 + 6}{3} = 4.
$$

The cost at the mean, and at the nearby guess $c = 5$, is

$$
\begin{aligned}
S(4) &= (2-4)^2 + (4-4)^2 + (6-4)^2 = 8, \\
S(5) &= (2-5)^2 + (4-5)^2 + (6-5)^2 = 11.
\end{aligned}
$$

The mean wins. A constant that ignores the observations entirely, $c = 0$, costs $S(0) = 4 + 16 + 36 = 56$.


```python
# The mean minimizes the sum of squared errors among constants.
from fractions import Fraction
values = [2, 4, 6]
mean = Fraction(sum(values), len(values))

def cost(guess):
    return sum((y - guess) ** 2 for y in values)

print("hand mean", mean)
print("cost at mean", cost(mean))
print("cost at 5", cost(5))
print("cost at 0", cost(0))
assert mean == 4
assert cost(mean) == 8
assert cost(5) == 11
assert cost(0) == 56
assert cost(mean) < cost(5) < cost(0)
```

## A second table

Grace's four measurements are $10, 10, 10, 14$. The same theorem gives

$$
\bar y = \dfrac{10 + 10 + 10 + 14}{4} = 11,
$$

and

$$
S(11) = 3\cdot 1^2 + 3^2 = 12, \qquad S(10) = 0 + 0 + 0 + 16 = 16.
$$

Predicting every row by $10$, the most common value, is worse than predicting every row by the mean. Least squares does not vote for the mode.


```python
# A repeated value is not automatically the least-squares constant.
from fractions import Fraction
values = [10, 10, 10, 14]
mean = Fraction(sum(values), len(values))

def cost(guess):
    return sum((y - guess) ** 2 for y in values)

print("mean", mean)
print("cost of the mean", cost(mean))
print("cost of the mode", cost(10))
assert mean == 11
assert cost(mean) == 12
assert cost(10) == 16
```

## Pitfall

The formula $\widehat y = 40 + 5x$ will happily answer $x = 100$:

$$
40 + 5\cdot 100 = 540.
$$

The arithmetic is correct and the prediction is not therefore justified. The rule was only a way to read a line. Whether a line fitted on a short range of hours still describes hour $100$ is a later question, answered with a prediction interval rather than with more enthusiasm.


```python
# The formula evaluates outside the table; that is not evidence.
far = 40 + 5 * 100
print("prediction at 100 hours", far)
assert far == 540
```

## Summary

- $\widehat y = b_0 + b_1 x$ predicts $b_0$ at $x = 0$ and then adds $b_1$ per unit of $x$.
- Among constant predictions, the mean is the unique least-squares choice, because $S'(c) = 0$ forces $c = \bar y$ and $S''(c) = 2n > 0$.
- A formula can be evaluated anywhere. Evaluation is not yet evidence.
