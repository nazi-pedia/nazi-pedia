Name the two hidden units. After the bend,

$$
\begin{aligned}
h_1 &= \mathrm{ReLU}(x_1 + x_2) \\
h_2 &= \mathrm{ReLU}(x_1 + x_2 - 1)
\end{aligned}
$$

and the output is

$$
s = h_1 - 2 h_2
$$

Inside the first quadrant, $x_1 + x_2 \ge 0$, so $h_1$ is simply the sum. The bend that matters is the line

$$
x_1 + x_2 = 1
$$

Below that line, $h_2 = 0$ and $s = x_1 + x_2$. Above it,

$$
s = (x_1 + x_2) - 2(x_1 + x_2 - 1) = 2 - x_1 - x_2
$$

The output rises with the sum, then folds back down with slope $-1$ in each input. At the opposite corner the fold has returned the score to zero.


## Example 1 — Below the line and above it

Gate 10 sits on the line, so both formulas agree:

$$
s = 1 + 0 = 1, \qquad s = 2 - 1 - 0 = 1
$$

Gate 11 is above the line:

$$
s = 2 - 1 - 1 = 0
$$

Gate 00 is below it:

$$
s = 0 + 0 = 0
$$


```python
# Below the bend the score is the sum. Above the bend it is 2 minus the sum.
def folded(x1, x2):
    if x1 + x2 <= 1:
        return x1 + x2
    return 2 - x1 - x2

scores = [folded(0, 0), folded(1, 0), folded(0, 1), folded(1, 1)]
print(scores)
assert scores == [0, 1, 1, 0]
```

## Example 2 — A point in the middle of each region

The midpoint $(1/2,\, 0)$ is below the line.

$$
s = \dfrac{1}{2}
$$

The point $(1,\, 1/2)$ is above the line.

$$
s = 2 - 1 - \dfrac{1}{2} = \dfrac{1}{2}
$$

The network is defined on the whole plane, not only on the four corners.


```python
# Hand: s(1/2, 0) = 1/2 and s(1, 1/2) = 1/2.
from fractions import Fraction

below = Fraction(1, 2) + 0
above = 2 - 1 - Fraction(1, 2)
print(below, above)
assert below == Fraction(1, 2)
assert above == Fraction(1, 2)
```

## Pitfall — Matching four corners is not the same as matching the fold formula

Both descriptions agree on the four gates. The fold formula is the stronger statement: it says what the network does at every point, including $(1/2, 0)$ and $(1, 1/2)$. When a later lesson changes only the output weights, this hidden fold stays put.


## Summary

$h_1$ carries the sum. $h_2$ turns on along $x_1 + x_2 = 1$ and is subtracted twice. Below the line the score equals the sum. Above the line the score equals $2 - x_1 - x_2$. The four gates land on $0, 1, 1, 0$.
