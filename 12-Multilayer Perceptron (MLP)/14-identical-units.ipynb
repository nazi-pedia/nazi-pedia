# 14 Identical Units Stay Identical

| | |
|---|---|
| **Path** | Multilayer perceptron |
| **Prerequisite** | 13 |

Give both hidden units the same incoming weights, the same bias, and the same outgoing weight.

$$
z = x_1, \qquad h = \mathrm{ReLU}(z), \qquad v_1 = v_2 = 1, \qquad c = 0
$$

Both units ignore $x_2$. On gate 10,

$$
h_1 = h_2 = 1, \qquad s = 2, \qquad y = 1, \qquad L = \dfrac{1}{2}
$$

The two units see the same forward value, so they receive the same backward value.

$$
\dfrac{\partial L}{\partial v_1} = \dfrac{\partial L}{\partial v_2} = 1
\qquad
\dfrac{\partial L}{\partial w_{\text{on } x_1}} = 1
\qquad
\dfrac{\partial L}{\partial w_{\text{on } x_2}} = 0
$$

A step adds the same number to both copies. After $\eta = 1/4$,

$$
v_1 = v_2 = \dfrac{3}{4}, \quad
w = \left(\dfrac{3}{4},\, 0\right), \quad
b = -\dfrac{1}{4}, \quad
c = -\dfrac{1}{4}
$$

The new score on gate 10 is $1/2$, and the loss is $1/8$. The two units are still copies of each other, and the weight on $x_2$ is still $0$.

Gate 01 then has pre-activation $-1/4$, hidden value $0$ in both units, and score $-1/4$. Its XOR target is $1$. A pair of identical units is one unit, written twice. Lesson 06 already showed that one unit cannot hit the four targets.


## Example 1 — One step keeps the two rows equal


```python
# Both units start equal. After the step they are still equal, and x2 still has weight 0.
from fractions import Fraction

eta = Fraction(1, 4)
v = 1 - eta * 1
w1 = 1 - eta * 1
w2 = 0 - eta * 0
b = 0 - eta * 1
c = 0 - eta * 1
print(v, w1, w2, b, c)
assert v == Fraction(3, 4)
assert (w1, w2, b) == (Fraction(3, 4), 0, Fraction(-1, 4))
z = w1 * 1 + w2 * 0 + b
h = z if z > 0 else 0
s = v * h + v * h + c
loss = Fraction(1, 2) * (s - 1) ** 2
print(z, s, loss)
assert z == Fraction(1, 2) and s == Fraction(1, 2)
assert loss == Fraction(1, 8)
```

## Example 2 — Gate 01 is invisible to both copies


```python
# Gate 01 still has weight 0 on x2, so both hidden units stay off and the score is the bias c.
from fractions import Fraction

z = Fraction(3, 4) * 0 + 0 * 1 + Fraction(-1, 4)
h = z if z > 0 else 0
s = h + h + Fraction(-1, 4)
print(z, h, s)
assert z == Fraction(-1, 4) and h == 0
assert s == Fraction(-1, 4)
```

## Pitfall — A random start is not a proof, but a tied start is a trap

This page does not claim that every initial weight fails. It claims a precise trap: if the two units match, their gradients match, so gradient descent never separates them. Breaking the tie has to happen before the updates, in the initial weights.


## Summary

Equal units receive equal gradients and remain equal after the step. Here the loss on gate 10 falls from $1/2$ to $1/8$, while the weight on $x_2$ stays $0$ in both units. Gate 01 still scores $-1/4$ against a target of $1$.
