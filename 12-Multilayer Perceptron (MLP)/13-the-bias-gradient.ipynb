A bias is the weight in front of an input that is always $1$. If

$$
z = w_1 x_1 + w_2 x_2 + b \cdot 1
$$

then

$$
\dfrac{\partial L}{\partial b} = \dfrac{\partial L}{\partial z} \cdot 1 = \dfrac{\partial L}{\partial z}
$$

The bias gradient is the pre-activation gradient. Nothing else is attached.

On gate 11 in Lesson 11,

$$
\dfrac{\partial L}{\partial z_2} = -1
$$

so

$$
\dfrac{\partial L}{\partial b_2} = -1
$$

The same identity is why the dead unit in Lesson 10 could still train $c$. The output score is $s = v h + c\cdot 1$, and $\partial L/\partial c = \partial L/\partial s$, whether or not $h$ is zero.


## Example 1 — The second hidden bias on gate 11


```python
# The bias gradient equals the pre-activation gradient. Here both are -1.
d_z2 = -1
d_b2 = d_z2 * 1
print(d_b2)
assert d_b2 == -1
```

## Example 2 — One step of that bias, with every other weight held still

$$
b_2 \leftarrow -1 - \dfrac{1}{8}(-1) = -1 + \dfrac{1}{8} = -\dfrac{7}{8}
$$

The old pre-activation on gate 11 is $1 + 1 - 1 = 1$. Adding $1/8$ to the bias adds $1/8$ to every pre-activation, so the new $h_2$ on gate 11 is $9/8$. The output weight on that unit is still $-1$, and the score changes by

$$
(-1)\left(\dfrac{9}{8} - 1\right) = -\dfrac{1}{8}
$$

The old score was $1$. The new score is $7/8$. The residual shrinks, and no other weight was touched.


```python
# Moving only b2 by +1/8 lowers gate 11 from score 1 to 7/8.
from fractions import Fraction

old_h2 = 1
new_h2 = Fraction(9, 8)
old_score = 1
new_score = old_score + (-1) * (new_h2 - old_h2)
print(new_h2, new_score)
assert new_h2 == Fraction(9, 8)
assert new_score == Fraction(7, 8)
```

## Pitfall — The same bias step lands on every gate

Adding $1/8$ to $b_2$ adds $1/8$ to $z_2$ everywhere, but ReLU may swallow it. On gate 00 the old $z_2$ is $-1$. The new one is $-7/8$, still negative, so $h_2$ stays $0$. On gate 10 the old $z_2$ is $0$. The new one is $1/8$, so a unit that was off turns on. A bias step is global. The kink decides which gates feel it.


```python
# Gate 00 stays off. Gate 10, which sat on the kink, turns on.
from fractions import Fraction

def relu(value):
    return value if value > 0 else 0

gate_00 = relu(-1 + Fraction(1, 8))
gate_10 = relu(0 + Fraction(1, 8))
print(gate_00, gate_10)
assert gate_00 == 0
assert gate_10 == Fraction(1, 8)
```

## Summary

The bias gradient equals the gradient of its pre-activation, because the bias multiplies the constant $1$. On gate 11 that number is $-1$ for the second hidden unit. A step of $1/8$ turns the on-kink gate 10 on, and it leaves the deeply negative gate 00 off.
