# 06 One Hidden Unit Cannot Build XOR

| | |
|---|---|
| **Path** | Multilayer perceptron |
| **Prerequisite** | 05 |

Two hidden units were enough. One is not. Suppose

$$
h = \mathrm{ReLU}(w_1 x_1 + w_2 x_2 + b), \qquad s = v h + c
$$

and $s$ matches $0, 1, 1, 0$ on the four gates. This lesson derives a contradiction in every case.

If $v = 0$, then $s = c$ everywhere. A constant cannot be both $0$ and $1$.


## Example 1 — The hidden unit is off at the origin, and $v < 0$

Off at the origin means $b \le 0$, so $h(0, 0) = 0$ and $c = 0$. Then

$$
s = v\,\mathrm{ReLU}(\cdots) \le 0
$$

because $v$ is negative and ReLU is never negative. Gate 10 needs $s = 1$. A non-positive number cannot equal $1$.


```python
# A negative output weight cannot lift a ReLU above 0. The pass target is 1.
values = [0, 1, 2]
lifted = [-3 * value for value in values]
print(lifted)
assert all(item <= 0 for item in lifted)
assert 1 not in lifted
```

## Example 2 — Off at the origin, and $v > 0$

Again $c = 0$. The three remaining gates require

$$
\begin{aligned}
\mathrm{ReLU}(w_1 + b) &= \dfrac{1}{v} \\
\mathrm{ReLU}(w_2 + b) &= \dfrac{1}{v} \\
\mathrm{ReLU}(w_1 + w_2 + b) &= 0
\end{aligned}
$$

The first two quantities are positive, so the ReLU is inactive only if we are looking at a positive pre-activation. Thus $w_1 + b = 1/v$ and $w_2 + b = 1/v$. Add them:

$$
w_1 + w_2 + b = \dfrac{1}{v} - b \ge \dfrac{1}{v} > 0
$$

because $b \le 0$. The third ReLU is then positive, not zero.

The special case $v = 1$, $b = 0$ forces $w_1 = w_2 = 1$, and

$$
\mathrm{ReLU}(1 + 1 + 0) = 2 \ne 0
$$


```python
# The weights forced by the two pass gates make the far gate active.
relu_far = 1 + 1 + 0
print(relu_far)
assert relu_far == 2
assert relu_far != 0
```

## Example 3 — The hidden unit is on at the origin

Now $b > 0$, so $h(0, 0) = b$ and the target $0$ forces $c = -vb$.

If $v > 0$, the pass gates force $w_1 = w_2 = 1/v$. Then

$$
s(1, 1) = v\left(\mathrm{ReLU}\!\left(\dfrac{2}{v} + b\right) - b\right) = v\cdot\dfrac{2}{v} = 2
$$

The far gate needed $0$ and received $2$.

If $v < 0$, matching both pass gates and the far gate forces $w_1 + w_2 = 0$ from the far gate, while each active pass gate contributes a negative weight $1/v$. Two copies sum to $2/v \ne 0$. If either pass gate is inactive, the pre-activation at the far gate stays negative, so its ReLU is $0$, which cannot equal the positive bias $b$. Every sub-case breaks.


```python
# v = 1, b = 1, w1 = w2 = 1, c = -1. The far score is 2, not 0.
score_far = 1 * (1 + 1 + 1) + (-1)
print(score_far)
assert score_far == 2
```

## Pitfall — A wide straight layer does not replace the second unit

Ten hidden units, each a ReLU, can fail in the same way if they are forced to be one unit. Lesson 14 shows the concrete way that happens: identical weights stay identical. The proof on this page is the smaller fact that a single bend has too little variety for these four targets.


## Summary

One hidden ReLU and one output weight cannot hit $0, 1, 1, 0$. A non-positive output weight never reaches $1$. A positive weight that fits the two pass gates makes the far gate score $2$, or it leaves a ReLU active where the target needs it off. The network in Lesson 04 uses two hidden units because one is not enough.
