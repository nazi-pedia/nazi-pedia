## Setting Up Python

```python
from sympy import (symbols, laplace_transform, inverse_laplace_transform,
                   exp, sin, cos, sinh, cosh, Function, DiracDelta, Heaviside,
                   simplify, apart, factor, solve, Eq, dsolve, pprint)

# Define symbols
t, s, a, b, c, w = symbols('t s a b c w', real=True)
n = symbols('n', positive=True, integer=True)
```

## Example 1:

$$ L\left\{ 1 \right\}=\frac{1}{s} $$

```python
f_t = 1
F_s = laplace_transform(f_t, t, s)
print("L{1} =", F_s)
```

## Example 2:

$$L\left\{ t \right\}=\frac{1}{s^2}$$

```python
F_s = laplace_transform(t, t, s)
print("L{t} =", F_s)
```

## Example 3:

$$L\left\{ t^n \right\}=\frac{n!}{s^{n+1}}$$

```python
from sympy import factorial, Integer
for k in range(1, 6):
    F_s = laplace_transform(t**k, t, s)
    print(f"L{{t^{k}}} =", F_s[0])
```

## Example 4: 

$$L\left\{ e^{at} \right\}=\frac{1}{s-a}$$

```python
F_s = laplace_transform(exp(a*t), t, s)
print("L{e^(at)} =", F_s)
```

## Example 5:

$$L\left\{ sin(at) \right\}=\frac{a}{s^2+a^2}$$

```python
F_s = laplace_transform(sin(a*t), t, s)
print("L{sin(at)} =", F_s)
```

## Example 6:

$$L\left\{ cos(at) \right\}=\frac{s}{s^2+a^2}$$

```python
F_s = laplace_transform(cos(a*t), t, s)
print("L{cos(at)} =", F_s)
```
