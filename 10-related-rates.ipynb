Two quantities are tied together by one equation, and both of them change with time.
Differentiate both sides with respect to time. The unknown you want is one of those time derivatives.
The rest is supplied by the problem statement or by the geometry.

The order of the work is fixed.

1. Name the quantities, and keep the instant you care about separate from the general relation.
2. Write the geometric equation.
3. Differentiate with respect to $t$.
4. Substitute the numbers only after that. If you substitute earlier, you have killed a variable that is still changing.


## Example 1 — A ladder

### Hand solution

A ladder $5$ meters long leans against a wall. Its foot moves away from the wall at $\frac{1}{2}$ meter per second.
When the foot is $3$ meters from the wall, how fast is the top of the ladder sliding down?

$$
\begin{aligned}
 x^2+y^2 &= 25, \\
 2x\dfrac{dx}{dt}+2y\dfrac{dy}{dt} &= 0, \\
 \dfrac{dy}{dt} &= -\dfrac{x}{y}\dfrac{dx}{dt}.
\end{aligned}
$$

At that instant $x=3$ and $y=4$, so

$$
\begin{aligned}
 \dfrac{dy}{dt}
 &= -\dfrac{3}{4}\cdot\dfrac{1}{2} \\
 &= -\dfrac{3}{8}.
\end{aligned}
$$

The unit is meters per second. The negative sign means the height is decreasing.

### Python recheck


```python
import sympy as sp

# Ladder: x(t)**2 + y(t)**2 = 25. Hand result at the instant: dy/dt = -3/8.
t = sp.symbols("t", real=True)
x = sp.Function("x")
y = sp.Function("y")
# Differentiate the relation, then solve for y'(t).
relation = sp.diff(x(t)**2 + y(t)**2 - 25, t)
dydt = sp.solve(relation, y(t).diff(t))[0]
# Substitute x = 3, y = 4, and dx/dt = 1/2 only after differentiating.
value = dydt.subs({x(t): 3, y(t): 4, x(t).diff(t): sp.Rational(1, 2)})
print("dy/dt =", value)
# The assert checks the signed rate -3/8.
assert value == -sp.Rational(3, 8)
print("confirmed")
```

## Example 2 — A spherical balloon

### Hand solution

The volume of a sphere is $V=\frac{4}{3}\pi r^3$.

$$
\begin{aligned}
 \dfrac{dV}{dt} &= 4\pi r^2\dfrac{dr}{dt}.
\end{aligned}
$$

Suppose air enters at a rate of $100$ cubic centimeters per second and the radius is $5$ centimeters. Then

$$
\begin{aligned}
 100 &= 4\pi(25)\dfrac{dr}{dt}, \\
 \dfrac{dr}{dt}
 &= \dfrac{100}{100\pi} \\
 &= \dfrac{1}{\pi}.
\end{aligned}
$$

The unit is centimeters per second. The surface area $4\pi r^2$ is what converts the volume rate into the radius rate.

### Python recheck


```python
import sympy as sp

# V = (4/3)*pi*r(t)**3, with dV/dt = 100. Hand result at r = 5: dr/dt = 1/pi.
t = sp.symbols("t")
r = sp.Function("r")
V = sp.Rational(4, 3)*sp.pi*r(t)**3
equation = sp.diff(V, t) - 100
drdt = sp.solve(equation, r(t).diff(t))[0]
value = sp.simplify(drdt.subs(r(t), 5))
print(value)
# The assert checks that the radius rate is 1/pi.
assert value == 1/sp.pi
print("confirmed")
```

## Example 3 — A conical tank

### Hand solution

The tank points vertex downward. At height $10$ the free-surface radius is $4$, so the relation $r=\frac{2}{5}h$ holds at every depth.
Water enters at $3$ cubic meters per minute. When the depth is $4$ meters, how fast is the depth rising?

$$
\begin{aligned}
 V
 &= \dfrac{1}{3}\pi r^2 h \\
 &= \dfrac{1}{3}\pi\left(\dfrac{2}{5}h\right)^2 h \\
 &= \dfrac{4\pi}{75}h^3, \\
 \dfrac{dV}{dt}
 &= \dfrac{4\pi}{25}h^2\dfrac{dh}{dt}, \\
 3
 &= \dfrac{4\pi}{25}\cdot 16\cdot\dfrac{dh}{dt}, \\
 \dfrac{dh}{dt}
 &= \dfrac{75}{64\pi}.
\end{aligned}
$$

The derivative of $\frac{4\pi}{75}h^3$ is $\frac{12\pi}{75}h^2=\frac{4\pi}{25}h^2$. That factor is what sits in the rate equation.

### Python recheck


```python
import sympy as sp

# V = 4*pi/75 * h(t)**3 and dV/dt = 3. Hand result at h = 4: dh/dt = 75/(64*pi).
t = sp.symbols("t")
h = sp.Function("h")
V = 4*sp.pi/75 * h(t)**3
dVdt = sp.diff(V, t)
dhdt = sp.solve(sp.Eq(dVdt, 3), h(t).diff(t))[0]
value = sp.simplify(dhdt.subs(h(t), 4))
print(value)
# The assert checks the depth rate 75/(64*pi).
assert value == sp.Rational(75, 64)/sp.pi
print("confirmed")
```

## Example 4 — A shadow

### Hand solution

A person $2$ meters tall walks away from a lamp $6$ meters high, at $1$ meter per second.
Let $s$ be the length of the shadow, and let $x$ be the distance from the person to the base of the lamp.
The large triangle has height $6$ and base $x+s$. The shadow triangle has height $2$ and base $s$.

$$
\begin{aligned}
 \dfrac{6}{x+s} &= \dfrac{2}{s}, \\
 6s &= 2x+2s, \\
 4s &= 2x, \\
 s &= \dfrac{x}{2}, \\
 \dfrac{ds}{dt}
 &= \dfrac{1}{2}\dfrac{dx}{dt} \\
 &= \dfrac{1}{2}.
\end{aligned}
$$

The shadow is always half as long as the distance to the base, so its rate of growth is always half as well, and that rate does not depend on the instant.

### Python recheck


```python
import sympy as sp

# Similar triangles: 6/(x + s) = 2/s. Hand result: s = x/2 and ds/dt = 1/2.
x, s = sp.symbols("x s", positive=True)
relation = sp.Eq(6/(x + s), 2/s)
solved = sp.solve(relation, s)[0]
print("s =", solved)
t = sp.symbols("t")
X = sp.Function("X")
# Replace x by X(t) and differentiate, with dX/dt = 1.
S = solved.subs(x, X(t))
speed = sp.diff(S, t).subs(X(t).diff(t), 1)
print("ds/dt =", speed)
# The shadow length is x/2, and the speed is the constant 1/2.
assert solved == x/2
assert speed == sp.Rational(1, 2)
print("confirmed")
```

## A pitfall

Substitute the numbers from the given instant only after you differentiate.
Translate the sign of the answer into words: moving away, sliding down, getting deeper.
Put the unit in that same sentence.

## Summary

Take the equation from the geometry, differentiate with respect to time, and only then substitute one instant.
