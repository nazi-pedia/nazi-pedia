# ۱۰ · آهنگ‌های مرتبط

| | |
|---|---|
| **دوره** | ریاضی عمومی ۱ |

دو کمیت با یک معادله به هم بسته‌اند و هر دو با زمان عوض می‌شوند.
از دو طرف نسبت به زمان مشتق بگیرید. آن‌چه می‌خواهید، یکی از مشتق‌های زمانی است.
بقیه را صورت مسئله داده یا هندسه تأمین می‌کند.

ترتیب کار ثابت است.

1. نام‌ها را بگذارید و لحظهٔ مورد نظر را از حالت کلی جدا کنید.
2. معادلهٔ هندسی را بنویسید.
3. نسبت به \(t\) مشتق بگیرید.
4. بعد عددها را بگذارید. اگر زودتر بگذارید، متغیری که هنوز تغییر می‌کند را کشته‌اید.



## مثال ۱ — نردبان

### حل ریاضی

نردبان ۵ متری به دیوار تکیه دارد. پای آن با سرعت \(\dfrac{1}{2}\) متر بر ثانیه از دیوار دور می‌شود.
وقتی پا ۳ متر از دیوار دور است، سر نردبان با چه سرعتی پایین می‌آید؟

\[
x^2+y^2=25
\]

\[
2x\frac{dx}{dt}+2y\frac{dy}{dt}=0
\]

\[
\frac{dy}{dt}=-\frac{x}{y}\frac{dx}{dt}
\]

در آن لحظه \(x=3\) و \(y=4\). پس

\[
\frac{dy}{dt}=-\frac{3}{4}\cdot\frac{1}{2}=-\frac{3}{8}
\]

متر بر ثانیه. علامت منفی یعنی ارتفاع کم می‌شود.

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t", real=True)
x = sp.Function("x")
y = sp.Function("y")
relation = sp.diff(x(t)**2 + y(t)**2 - 25, t)
dydt = sp.solve(relation, y(t).diff(t))[0]
value = dydt.subs({x(t): 3, y(t): 4, x(t).diff(t): sp.Rational(1, 2)})
print("dy/dt =", value)
assert value == -sp.Rational(3, 8)
print("تایید شد")

```

## مثال ۲ — بادکنک کروی

### حل ریاضی

حجم کره \(V=\dfrac{4}{3}\pi r^3\) است.

\[
\frac{dV}{dt}=4\pi r^2\frac{dr}{dt}
\]

اگر هوا با آهنگ \(100\) سانتی‌متر مکعب بر ثانیه وارد شود و شعاع ۵ سانتی‌متر باشد،

\[
100=4\pi(25)\frac{dr}{dt}
\]
\[
\frac{dr}{dt}=\frac{100}{100\pi}=\frac{1}{\pi}
\]

سانتی‌متر بر ثانیه. سطح \(4\pi r^2\) تبدیل‌کنندهٔ آهنگ حجم به آهنگ شعاع است.

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t")
r = sp.Function("r")
V = sp.Rational(4, 3)*sp.pi*r(t)**3
equation = sp.diff(V, t) - 100
drdt = sp.solve(equation, r(t).diff(t))[0]
value = sp.simplify(drdt.subs(r(t), 5))
print(value)
assert value == 1/sp.pi
print("تایید شد")

```

## مثال ۳ — مخزن مخروطی

### حل ریاضی

مخزن راس‌پایین است. در ارتفاع ۱۰، شعاع سطح آزاد ۴ است، پس همیشه \(r=\dfrac{2}{5}h\).
آب با آهنگ ۳ متر مکعب بر دقیقه وارد می‌شود. وقتی عمق ۴ متر است، عمق با چه آهنگی بالا می‌آید؟

\[
V=\frac{1}{3}\pi r^2 h=\frac{1}{3}\pi\left(\frac{2}{5}h\right)^2 h=\frac{4\pi}{75}h^3
\]

\[
\frac{dV}{dt}=\frac{4\pi}{25}h^2\frac{dh}{dt}
\]

\[
3=\frac{4\pi}{25}\cdot 16\cdot\frac{dh}{dt}
\]
\[
\frac{dh}{dt}=\frac{75}{64\pi}
\]

مشتق \(\frac{4\pi}{75}h^3\) برابر \(\frac{12\pi}{75}h^2=\frac{4\pi}{25}h^2\) است. همین در معادله نشسته.

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t")
h = sp.Function("h")
V = 4*sp.pi/75 * h(t)**3
dVdt = sp.diff(V, t)
dhdt = sp.solve(sp.Eq(dVdt, 3), h(t).diff(t))[0]
value = sp.simplify(dhdt.subs(h(t), 4))
print(value)
assert value == sp.Rational(75, 64)/sp.pi
print("تایید شد")

```

## مثال ۴ — سایه

### حل ریاضی

شخص ۲ متری با سرعت ۱ متر بر ثانیه از چراغ ۶ متری دور می‌شود.
طول سایه \(s\) و فاصلهٔ شخص تا پایهٔ چراغ \(x\) است.
مثلث بزرگ ارتفاع ۶ و قاعدهٔ \(x+s\) دارد. مثلث سایه ارتفاع ۲ و قاعدهٔ \(s\).

\[
\frac{6}{x+s}=\frac{2}{s}
\]
\[
6s=2x+2s,\qquad 4s=2x,\qquad s=\frac{x}{2}
\]

\[
\frac{ds}{dt}=\frac{1}{2}\frac{dx}{dt}=\frac{1}{2}
\]

طول سایه همیشه نصف فاصله تا پایه است، پس آهنگ رشدش هم همیشه نصف است و به لحظه بستگی ندارد.

### بازبینی با پایتون



```python
import sympy as sp
x, s = sp.symbols("x s", positive=True)
relation = sp.Eq(6/(x + s), 2/s)
solved = sp.solve(relation, s)[0]
print("s =", solved)
t = sp.symbols("t")
X = sp.Function("X")
S = solved.subs(x, X(t))
speed = sp.diff(S, t).subs(X(t).diff(t), 1)
print("ds/dt =", speed)
assert solved == x/2
assert speed == sp.Rational(1, 2)
print("تایید شد")

```
