## Building Complex Numbers from Variables

```python
real = 5
imaginary = 12

z = complex(real, imaginary)

print(z)
```

```python
real = float(input("Enter real part: "))
imaginary = float(input("Enter imaginary part: "))

z = complex(real, imaginary)

print("Your complex number:", z)
```

## A Practical Complex Calculator

```python
z1 = complex(3, 4)
z2 = complex(1, 2)

print("z1 =", z1)
print("z2 =", z2)

print("\nAddition:")
print(z1 + z2)

print("\nSubtraction:")
print(z1 - z2)

print("\nMultiplication:")
print(z1 * z2)

print("\nDivision:")
print(z1 / z2)
```
