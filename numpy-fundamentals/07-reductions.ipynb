# 07 Reductions along an Axis

A reduction replaces a whole axis with one number. `axis` is the index that disappears.

For a matrix $a_{ij}$, axis 0 runs down the rows, so it removes the row index. Axis 1 runs across the columns.

$$
\sum_i a_{ij} \text{ is axis 0},
\qquad
\sum_j a_{ij} \text{ is axis 1}.
$$

Omitting `axis` reduces the whole array to one number. `keepdims=True` leaves the reduced axis in place with length 1, so the result can broadcast back onto the original.


## Example 1 — Three sums of one matrix

### Predict

$$
\begin{bmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \end{bmatrix}
$$

Axis 0 adds downward: $1+4$, $2+5$, $3+6$, so `[5, 7, 9]`. Axis 1 adds across: $1+2+3=6$ and $4+5+6=15$. The total is $21$.

### Run


```python
import numpy as np

grid = np.array([[1, 2, 3], [4, 5, 6]])
# axis 0 removes rows. axis 1 removes columns.
print(grid.sum(axis=0))
print(grid.sum(axis=1))
print(grid.sum())
```

## Example 2 — Means, checked by hand

### Predict

Each row of `[[2, 4], [6, 8]]` has mean $(2+4)/2 = 3$ and $(6+8)/2 = 7$. Each column has mean $(2+6)/2 = 4$ and $(4+8)/2 = 6$.

### Run


```python
# Example 2 — Means, checked by hand.
import numpy as np

grid = np.array([[2.0, 4.0], [6.0, 8.0]])
print(grid.mean(axis=1))
print(grid.mean(axis=0))
```

## Example 3 — keepdims so the mean can be subtracted

### Predict

Column means of `[[1, 2], [3, 6]]` are $2$ and $4$. With `keepdims=True` their shape is `(1, 2)`, which broadcasts down the rows. The centered matrix is

$$
\begin{bmatrix} 1-2 & 2-4 \\ 3-2 & 6-4 \end{bmatrix}
=
\begin{bmatrix} -1 & -2 \\ 1 & 2 \end{bmatrix}.
$$

Each centered column sums to 0.

### Run


```python
import numpy as np

grid = np.array([[1.0, 2.0], [3.0, 6.0]])
means = grid.mean(axis=0, keepdims=True)
print(means.shape)
centered = grid - means
print(centered)
# A centered column has mean zero, so its sum is zero.
print(centered.sum(axis=0))
```

## Example 4 — Population standard deviation, then the sample one

### Predict

For $1, 2, 3$ the mean is $2$. The squared deviations are $1$, $0$, and $1$, and their sum is $2$.

$$
\sigma = \sqrt{\dfrac{2}{3}},
\qquad
s = \sqrt{\dfrac{2}{2}} = 1.
$$

NumPy's default `std` divides by $n$. That is the population value $\sigma$. `ddof=1` divides by $n-1$ and returns the sample value $s = 1$.

### Run


```python
import numpy as np

values = np.array([1.0, 2.0, 3.0])
# Default ddof is 0: divide by n.
print(np.allclose(values.std(), np.sqrt(2 / 3)))
# ddof=1 divides by n - 1. Here that quotient is exactly 1.
print(values.std(ddof=1))
```

## A pitfall

`axis=0` is "down the rows," which is the first index, not "the axis I am looking along" in ordinary speech. People often say "sum each row" and then pass `axis=0`. Summing each row removes the column index, so it is `axis=1`. This cell prints both, with the row sums second.

### Run


```python
import numpy as np

grid = np.array([[1, 2, 3], [4, 5, 6]])
# Downward sums are not the row sums.
print("down", grid.sum(axis=0))
print("across", grid.sum(axis=1))
```

## Summary

`axis` names the index that disappears. Axis 0 reduces down the rows, axis 1 reduces across the columns, and a missing axis reduces everything. `keepdims=True` preserves a length-1 axis so the result broadcasts back. `std` divides by $n$ unless you pass `ddof=1`.
