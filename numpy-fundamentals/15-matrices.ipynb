A two-dimensional array is a matrix when you intend matrix multiplication. The operator is `@`, or `np.matmul`. Entry $(i, j)$ of a product is the dot product of row $i$ of the left matrix with column $j$ of the right matrix.

$$
(AB)_{ij} = \sum_k A_{ik} B_{kj}.
$$

`np.linalg.solve` solves $A\mathbf{x} = \mathbf{b}$ without forming an inverse. `det` is the determinant. A zero determinant means the system does not have one unique solution, and `solve` then raises `LinAlgError`.


## Example 1 — One matrix times one column

### Predict

$$
\begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}
\begin{bmatrix} 5 \\ 6 \end{bmatrix}
=
\begin{bmatrix} 1\cdot 5 + 2\cdot 6 \\ 3\cdot 5 + 4\cdot 6 \end{bmatrix}
=
\begin{bmatrix} 17 \\ 39 \end{bmatrix}.
$$

The elementwise product is a different matrix and is not this result.

### Run


```python
import numpy as np

matrix = np.array([[1, 2], [3, 4]])
column = np.array([5, 6])
# @ is the matrix product. The column is treated as shape (2,).
print(matrix @ column)
```

## Example 2 — Determinant and inverse, checked against the identity

### Predict

$$
\det\begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}
= 1\cdot 4 - 2\cdot 3 = -2.
$$

The inverse is

$$
\dfrac{1}{-2}
\begin{bmatrix} 4 & -2 \\ -3 & 1 \end{bmatrix}
=
\begin{bmatrix} -2 & 1 \\ 3/2 & -1/2 \end{bmatrix}.
$$

The computed determinant is a binary float, so the cell checks that it lies within a tolerance of $-2$. Multiplying the matrix by the inverse returns $I$.

### Run


```python
# Example 2 — Determinant and inverse, checked against the identity.
import numpy as np

matrix = np.array([[1.0, 2.0], [3.0, 4.0]])
# The determinant is -2. Compare with a tolerance, then print the inverse.
print(np.allclose(np.linalg.det(matrix), -2))
inverse = np.linalg.inv(matrix)
print(inverse)
print(np.allclose(matrix @ inverse, np.eye(2)))
```

## Example 3 — Solve a 2 by 2 system by hand, then with solve

### Predict

$$
\begin{aligned}
2x + y &= 5 \\
x + 3y &= 10.
\end{aligned}
$$

From the first equation, $y = 5 - 2x$. Substitute:

$$
x + 3(5 - 2x) = 10,
\qquad
x + 15 - 6x = 10,
\qquad
-5x = -5,
\qquad
x = 1,
\quad y = 3.
$$

### Run


```python
import numpy as np

coefficients = np.array([[2.0, 1.0], [1.0, 3.0]])
right_hand = np.array([5.0, 10.0])
solution = np.linalg.solve(coefficients, right_hand)
print(solution)
# Multiply back: A x should reproduce b.
print(np.allclose(coefficients @ solution, right_hand))
```

## Example 4 — A singular system has no unique solution

### Predict

The second row of

$$
\begin{bmatrix} 1 & 2 \\ 2 & 4 \end{bmatrix}
$$

is twice the first, so the determinant is $1\cdot 4 - 2\cdot 2 = 0$. `solve` raises `LinAlgError`.

### Run


```python
# Example 4 — A singular system has no unique solution.
import numpy as np

coefficients = np.array([[1.0, 2.0], [2.0, 4.0]])
print(np.linalg.det(coefficients))
try:
    print(np.linalg.solve(coefficients, np.array([1.0, 2.0])))
except np.linalg.LinAlgError as error:
    print(type(error).__name__)
```

## A pitfall

`*` multiplies entries. It does not multiply matrices. For the factors in example 1, `*` is illegal or, if you first reshape the column into a matrix of the same shape, it is a different operation. Here both factors are $2\times 2$, and `*` is entrywise while `@` is the matrix product.

$$
\begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}
\begin{bmatrix} 1 & 0 \\ 1 & 1 \end{bmatrix}
=
\begin{bmatrix} 3 & 2 \\ 7 & 4 \end{bmatrix},
\qquad
\text{entrywise }
\begin{bmatrix} 1 & 0 \\ 3 & 4 \end{bmatrix}.
$$

Check: row 1 times the columns is $(1\cdot 1 + 2\cdot 1,\ 1\cdot 0 + 2\cdot 1) = (3, 2)$.

### Run


```python
# A pitfall.
import numpy as np

left = np.array([[1, 2], [3, 4]])
right = np.array([[1, 0], [1, 1]])
print(left * right)
print(left @ right)
```

## Summary

`@` is the matrix product: each entry is a dot product of a row and a column. `det` is the determinant. `solve` returns the unique $\mathbf{x}$ in $A\mathbf{x}=\mathbf{b}$ when the determinant is not zero, and raises `LinAlgError` when it is. Prefer `solve` to computing an inverse and multiplying. `*` never means `@`.
