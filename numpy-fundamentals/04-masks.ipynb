A comparison on an array returns an array of `True` and `False`, one answer per element. Using that array as an index keeps the elements where the answer is `True`. That index is a mask.

A list of positions, such as `[2, 0, 2]`, is a fancy index. It picks those positions in the order you wrote, and it may repeat a position.

Combine masks with `&`, `|`, and `~`. Python's `and`, `or`, and `not` do not work on arrays, because an array does not have one truth value.


## Example 1 — Keep the positive numbers

### Predict

The comparison `values > 0` is `[False, True, False, True]` for $-1, 4, 0, 7$. The mask keeps `4` and `7`.

### Run


```python
import numpy as np

values = np.array([-1, 4, 0, 7])
mask = values > 0
print(mask)
# The mask selects elements, not positions you count by hand.
print(values[mask])
```

## Example 2 — Two conditions need parentheses

### Predict

Keep values that are at least 0 and strictly less than 5. From $-1, 0, 4, 5, 9$ that is `0` and `4`.

`&` binds more loosely than `<` in a way that surprises people, so each comparison gets its own parentheses.

### Run


```python
import numpy as np

values = np.array([-1, 0, 4, 5, 9])
# Each comparison is parenthesized, then & combines the two masks.
mask = (values >= 0) & (values < 5)
print(values[mask])
```

## Example 3 — Pick rows by a list of positions

### Predict

Rows `[1, 0, 1]` of

$$
\begin{bmatrix} 0 & 1 \\ 2 & 3 \\ 4 & 5 \end{bmatrix}
$$

are row 1, then row 0, then row 1 again:

$$
\begin{bmatrix} 2 & 3 \\ 0 & 1 \\ 2 & 3 \end{bmatrix}.
$$

### Run


```python
import numpy as np

grid = np.arange(6).reshape(3, 2)
# The list is the order of the rows, repeats included.
print(grid[[1, 0, 1]])
```

## Example 4 — A mask can also assign

### Predict

Every negative entry becomes 0. The positive entries stay. This is a clip at zero, written with a mask. Lesson 08 will show `np.clip` as the named tool.

### Run


```python
# Example 4 — A mask can also assign.
import numpy as np

values = np.array([-2, 5, -1, 3])
values[values < 0] = 0
print(values)
```

## A pitfall

`and` asks one object whether it is true. An array of truths has many answers, so `and` raises. Use `&` between two parenthesized comparisons.

### Run


```python
# A pitfall.
import numpy as np

values = np.array([-1, 0, 4, 5])
try:
    print((values >= 0) and (values < 5))
except ValueError as error:
    print(type(error).__name__)
print(values[(values >= 0) & (values < 5)])
```

## Summary

A comparison builds a mask, and the mask keeps the `True` positions. Combine comparisons with `&`, `|`, and `~`, each comparison in parentheses. A list of integers picks positions in the order written, repeats allowed. `and` and `or` are for single Python values, not for arrays.
