`np.nan` means "missing floating-point number." It is not `None`, and it is not a string. Any ordinary arithmetic with `nan` produces `nan`. In particular,

$$
\texttt{nan} \neq \texttt{nan}.
$$

The test is `np.isnan`. Reductions that know about missing data are spelled with a `nan` prefix: `nansum`, `nanmean`, `nanstd`. They skip missing entries. They do not treat missing as zero unless the operation is a sum of the remaining numbers, which is what `nansum` does.


## Example 1 — nan is not equal to itself

### Predict

The middle entry is missing. `values[1] == np.nan` is false. `np.isnan` is true there and false on the numbers.

### Run


```python
# Example 1 — nan is not equal to itself.
import numpy as np

values = np.array([1.0, np.nan, 3.0])
print(values[1] == np.nan)
print(np.isnan(values))
```

## Example 2 — A sum that includes nan becomes nan

### Predict

$1 + \mathrm{nan} + 3$ is `nan`. Skipping the missing entry, the sum of what remains is $1+3 = 4$, and the mean of what remains is $4/2 = 2$.

### Run


```python
# Example 2 — A sum that includes nan becomes nan.
import numpy as np

values = np.array([1.0, np.nan, 3.0])
print(np.sum(values))
print(np.nansum(values))
print(np.nanmean(values))
```

## Example 3 — Count what is present

### Predict

Three entries, one missing, so two are present. `~np.isnan(values)` is the mask of present numbers, and its sum is the count.

### Run


```python
# Example 3 — Count what is present.
import numpy as np

values = np.array([1.0, np.nan, 3.0, np.nan])
present = ~np.isnan(values)
print(present.sum())
print(values[present])
```

## Example 4 — Fill missing entries, then reduce

### Predict

Replacing `nan` with 0 changes the meaning: the sum becomes $1+0+3 = 4$, which matches `nansum`, but the mean of the filled array is $4/3$, not $2$. Filling and then calling `mean` divides by the original length. `nanmean` divides by the number of present values. They answer different questions.

### Run


```python
# Example 4 — Fill missing entries, then reduce.
import numpy as np

values = np.array([1.0, np.nan, 3.0])
filled = np.where(np.isnan(values), 0.0, values)
print(filled.mean())
print(np.nanmean(values))
```

## A pitfall

An integer array cannot store `nan`. The dtype has to be floating point. `np.array([1, np.nan, 3])` promotes the whole array to float. If you build an integer array first and then try to assign `nan`, NumPy rejects the assignment.

### Run


```python
# A pitfall.
import numpy as np

values = np.array([1, np.nan, 3])
print(values.dtype)
whole = np.array([1, 2, 3])
try:
    whole[1] = np.nan
except ValueError as error:
    print(type(error).__name__)
```

## Summary

Test for missing data with `np.isnan`, never with `==`. Ordinary sums and means become `nan` as soon as one entry is missing. `nansum` and `nanmean` skip those entries, and the mean divides by how many remain. Filling with zero and then taking the mean divides by the full length instead.
