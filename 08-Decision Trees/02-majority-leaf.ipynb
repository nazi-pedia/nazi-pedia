# The leaf should name the majority

Once a walk ends, the leaf must say a label. Suppose the leaf holds $n$ training parts, and $n_c$ of them have label $c$. If the leaf predicts $c$, it is wrong on the other parts:

$$
\text{errors}(c) = n - n_c
$$

That count is smallest when $n_c$ is largest. So the leaf predicts a majority label. If two labels tie, they produce the same number of errors. The tie is real. This course breaks it by choosing pass over fail, and it says that the choice was a rule, not a discovery in the data.


## Four pass, two fail

$$
\text{errors}(\text{pass}) = 6 - 4 = 2, \qquad \text{errors}(\text{fail}) = 6 - 2 = 4
$$

Pass wins. The error rate of that choice is $2/6 = 1/3$.

## A tie

Two pass and two fail. Either prediction makes $2$ errors. The rate is $1/2$ either way. Predicting pass is the tie-break, and the error count does not improve because of it.


```python
# Majority minimizes the error count. A tie keeps the count unchanged.
from fractions import Fraction

def errors(labels, prediction):
    return sum(label != prediction for label in labels)

mixed = ["pass", "pass", "pass", "pass", "fail", "fail"]
assert errors(mixed, "pass") == 2
assert errors(mixed, "fail") == 4
tie = ["pass", "pass", "fail", "fail"]
assert errors(tie, "pass") == errors(tie, "fail") == 2
print("mixed error rate", Fraction(errors(mixed, "pass"), len(mixed)))
print("tie error rate", Fraction(errors(tie, "pass"), len(tie)))
```

## Error rate can be blind to a useful cut

Four parts: three pass, one fail. Predicting pass makes $1$ error. The rate is $1/4$.

Cut them into $\{ \text{pass}, \text{pass} \}$ and $\{ \text{pass}, \text{fail} \}$.

The first leaf is pure, so it makes $0$ errors. The second leaf, whichever label it chooses, makes $1$ error. The total is still $1$ error out of $4$. The rate does not fall:

$$
\frac{2}{4}\cdot 0 + \frac{2}{4}\cdot \frac{1}{2} = \frac{1}{4}
$$

The cut isolated two sure passes, and the error rate stayed put. The next lesson's impurity will notice this cut. The plain error count does not.


```python
# The split changes the groups and leaves the total error count at 1.
from fractions import Fraction

parent_rate = Fraction(1, 4)
child_rate = Fraction(2, 4) * 0 + Fraction(2, 4) * Fraction(1, 2)
assert child_rate == parent_rate
print("error rate before and after", parent_rate, child_rate)
```

## A pitfall

A pure leaf has error $0$, but a leaf with one odd part out of a thousand has a tiny error rate and is still not a proof that the odd part was noise. The rate hides the count. Keep $n$ next to the rate.

## What to carry forward

A leaf predicts the label with the largest count, because that choice minimizes $n - n_c$. On three pass and one fail, a cut that looks useful can leave the error rate at $1/4$.
