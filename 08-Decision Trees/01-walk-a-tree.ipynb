# A tree is a list of questions

A decision tree does not draw a line through the data. It asks one yes-or-no question, then another, and stops at a leaf. The leaf is the prediction.

Eight parts were inspected. Two measurements: length in centimetres, weight in kilograms. The label is pass or fail.

| Length | Weight | Label |
|---:|---:|---|
| 1 | 1 | pass |
| 1 | 2 | pass |
| 2 | 1 | pass |
| 2 | 2 | pass |
| 1 | 4 | fail |
| 2 | 4 | fail |
| 5 | 1 | fail |
| 5 | 2 | fail |

This lesson does not grow the tree. It only walks a tree that is already written down.

1. If the weight is at most $3$, go left. Otherwise go right and predict fail.
2. On the left, if the length is at most $7/2$, predict pass. Otherwise predict fail.

The pass region is a rectangle: length $\le 7/2$ and weight $\le 3$. Every other part fails.


## Walk three parts by hand

Part $(2, 2)$. Weight $2 \le 3$, so turn left. Length $2 \le 7/2$, so the leaf says pass.

Part $(5, 1)$. Weight $1 \le 3$, so turn left. Length $5 > 7/2$, so the leaf says fail.

Part $(1, 4)$. Weight $4 > 3$, so turn right at the first question. The leaf says fail. The length is never consulted.

The same three walks cover the three leaves. A new part $(3, 3)$ has length $3 \le 7/2$ and weight $3 \le 3$, so the tree calls it a pass. Nothing in the training table sits at $(3, 3)$. The tree is willing to answer anyway, because the questions carve up the whole plane, not only the eight rows.


```python
# Walk the finished tree. The third number is the hand prediction.
from fractions import Fraction

def predict(length, weight):
    if weight <= 3:
        if length <= Fraction(7, 2):
            return "pass"
        return "fail"
    return "fail"

rows = [
    (2, 2, "pass"),
    (5, 1, "fail"),
    (1, 4, "fail"),
    (3, 3, "pass"),
    (4, 1, "fail"),
    (2, 4, "fail"),
]
for length, weight, expected in rows:
    assert predict(length, weight) == expected
    print((length, weight), predict(length, weight))
```

## One change: reverse the first answer

Suppose the right-hand leaf said pass instead of fail. Part $(1, 4)$ would be called a pass. Its weight is $4$, and every training part with weight $4$ failed. The leaf text is the whole prediction. A wrong word there is not a small slip in a formula. It changes the label of every part that reaches that leaf.


```python
# The same questions, with the heavy leaf flipped from fail to pass.
def predict_flipped(length, weight):
    if weight <= 3:
        if length <= 3.5:
            return "pass"
        return "fail"
    return "pass"

assert predict_flipped(1, 4) == "pass"
assert predict_flipped(2, 2) == "pass"
print("heavy part", predict_flipped(1, 4))
print("light short part", predict_flipped(2, 2))
```

## A pitfall

The number $7/2$ is not a measured length. The lengths in the table are $1$, $2$, and $5$. The cut sits between $2$ and $5$. Any cut strictly between those two training values sends the same eight parts to the same leaves. $7/2$ is a convenient representative, not a physical constant of the parts.

## What to carry forward

Prediction is a walk. Each question chooses a branch. The leaf you land on is the label. Here a part passes only when its length is at most $7/2$ and its weight is at most $3$.
