A method is a function that belongs to a value. `text.upper()` asks the string `text` to produce an uppercase copy. The call does not edit `text`.

An f-string, written with an `f` before the quotes, evaluates expressions inside braces and inserts their text.

$$
\texttt{f"\{name\} is \{age\}"}
$$


## Example 1 — Cleaning and changing case

### Predict

`strip` removes whitespace on both ends. `upper` and `lower` change case. `startswith` checks a prefix after those copies are made. The original `raw` still has its spaces.

### Run


```python
raw = "  Ada Lovelace  "
print(repr(raw.strip()))
print(raw.strip().upper())
# The checks run on the cleaned lowercase copy, not on the original spaces.
print(raw.strip().lower().startswith("ada"))
print(repr(raw))
```

## Example 2 — Split apart, then join

### Predict

`split(",")` breaks the text on each comma and returns a list of three strings. `join` is called on the separator, and the list is the argument. The separator is placed between items, not after the last one.

### Run


```python
parts = "red,green,blue".split(",")
print(parts)
# The receiver is the glue. The argument is the sequence of pieces.
print(" | ".join(parts))
```

## Example 3 — Find, count, and replace

### Predict

`find` returns the first index of `"two"`, which is 9. `count` returns 2. `replace` returns a new sentence and leaves the original in place.

### Run


```python
sentence = "one fish two fish"
print(sentence.find("two"))
print(sentence.count("fish"))
print(sentence.replace("fish", "bird"))
# replace did not rebind the name.
print(sentence)
```

## Example 4 — f-strings calculate inside the braces

### Predict

The first line inserts a name and a year. The second inserts the value of `2 + 2`. The third formats $1/3$ with two digits after the decimal point, so the printed text is `0.33`.

$$
\frac{1}{3} \approx 0.33 \text{ when kept to two decimal places.}
$$

### Run


```python
name = "Ada"
year = 1843
print(f"{name} wrote notes in {year}")
# The braces hold an expression, not only a name.
print(f"{2 + 2}")
# .2f means two digits after the decimal point.
print(f"{1/3:.2f}")
```

## Example 5 — Lines inside one string

### Predict

`\n` splits the logical lines. `splitlines` returns those lines without the newline characters.

### Run


```python
text = "first\nsecond\nthird"
# splitlines drops the newline characters and returns the pieces.
print(text.splitlines())
print(len(text.splitlines()))
```

## A pitfall

Calling a method and ignoring the result does nothing visible. `name.upper()` builds a new string and throws it away if you do not attach a name to it.

### Run


```python
name = "ada"
# The uppercase copy is discarded. name still labels the old text.
name.upper()
print(name)
name = name.upper()
print(name)
```

## Summary

String methods return new strings. `strip`, `split`, `join`, `replace`, `find`, and `startswith` cover most everyday text work. An f-string evaluates the expressions in its braces, and a format like `.2f` controls how a number is written. Save the returned string if you want the name to see the change.
