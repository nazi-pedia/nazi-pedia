A comprehension builds a collection from a loop that has one clear job. The list form is the loop written inside brackets.

$$
[\ f(x) \text{ for } x \text{ in items if } p(x)\ ]
$$

Read it as: take each `x` in `items`, keep it when `p(x)` is true, and store `f(x)`. If the loop needs several steps, a name, or a `break`, write the loop out. A comprehension is for one transformation.


## Example 1 — The same squares, two ways

### Predict

The loop and the comprehension both store $1, 4, 9, 16$. They are two spellings of one idea.

### Run


```python
squares = []
for number in range(1, 5):
    squares.append(number ** 2)
print(squares)
# The expression comes first, then the for that feeds it.
print([number ** 2 for number in range(1, 5)])
```

## Example 2 — A filter is the if at the end

### Predict

The `if` keeps even numbers only. From 0 through 7 the evens are 0, 2, 4, 6.

### Run


```python
numbers = range(8)
# The if at the end keeps only the values that pass.
evens = [number for number in numbers if number % 2 == 0]
print(evens)
```

## Example 3 — A dictionary of lengths, and a set of initial letters

### Predict

The dictionary maps each word to its length. The set keeps the first letter of each word once, so the two words that start with `p` contribute one `p`.

### Run


```python
words = ["ada", "python", "code", "plan"]
# A dict comprehension stores word to length. A set keeps first letters once.
lengths = {word: len(word) for word in words}
print(lengths)
initials = {word[0] for word in words}
print(sorted(initials))
```

## Example 4 — A comprehension can call a function

### Predict

Each temperature in Celsius becomes Fahrenheit with the formula from lesson 12. The inputs 0 and 100 become 32.0 and 212.0.

### Run


```python
def fahrenheit(celsius):
    # F = (9/5) * C + 32.
    return (9 / 5) * celsius + 32

print([fahrenheit(c) for c in (0, 100)])
```

## A pitfall

A comprehension that hides a long decision is harder to read than the loop it replaced. The loop below skips blanks, lowercases the rest, and counts them. Leave it as a loop. You can still see every step.

### Run


```python
raw_words = ["  Ada", "", "GRACE", "  "]
counts = {}
for raw in raw_words:
    cleaned = raw.strip().lower()
    # Several steps belong in a loop, not in one comprehension.
    if cleaned == "":
        continue
    counts[cleaned] = counts.get(cleaned, 0) + 1
print(counts)
```

## Summary

A comprehension is a one-job loop written inside a collection. Put the new value first, the `for` next, and the `if` last. Use a list, a dictionary, or a set. When the body would need several statements, write the loop in full.
