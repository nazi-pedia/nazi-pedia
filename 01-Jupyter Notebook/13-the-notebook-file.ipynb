This page is a file. Its name ends in `.ipynb`. The editor tab shows that name. Save writes the cells and the outputs you currently see into that file.

Give the file a real name before the first serious note. `Untitled` is a temporary name, and the next new notebook may use it too.

A code cell does not automatically know the file's name. There is no `__file__` inside the cell unless some other program put it there. The name you trust is the name on the tab and in the folder.

## Example 2 — What the file actually keeps

A saved notebook keeps three things that matter to a reader.

1. The markdown source.
2. The code source.
3. The outputs from the last time those code cells ran.

Change the code and forget to run it, and the file can hold new code beside an old output. Save would keep that pair.

### Predict

The cell prints the total it actually computed, which is `9`.


```python
# The output under a cell should be the output of the code now in the cell.
total = 2 + 3 + 4
print(total)
assert total == 9
```

## Pitfall — Save keeps the photograph you are looking at

Edit a cell from `2 + 3 + 4` to `2 + 3 + 5` and save before you run it. The file now contains the new code and the old printed `9`. The next person sees a page that disagrees with itself.

Run the cell, read the output, then save.

## Summary

A notebook is a `.ipynb` file. The tab shows its name. Save stores the sources and the outputs together. Run the cells before you save, so the photograph matches the code.
