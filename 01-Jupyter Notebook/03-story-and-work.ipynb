A notebook mixes two kinds of writing.

A markdown cell is rendered. Headings become large type, and a fenced block is only a picture of code.

A code cell is executed. A hash at the start of a line is a comment. The kernel skips that line.

## A heading is a signpost

The line above is a heading. It begins with `##` in the markdown source. The line below is a smaller signpost. It begins with `###`.

### A smaller signpost

**Bold** marks a word you must not skip. *Italic* marks a word you are introducing. A code span, such as `print`, names a piece of the language and does not run it.

To edit a markdown cell, double-click it, change the source, then press Shift+Enter to render it again.

## Pitfall — The hash does a different job in each kind of cell

In markdown, `# A title` at the start of a line is a heading.

In code, `# A title` is a comment, and the cell stays quiet.

Choose the cell type before you choose the mark. In Jupyter, Esc then M makes the selected cell markdown. Esc then Y makes it code. Lesson 15 gathers those keys.

## Summary

Markdown is rendered for a reader. Code is executed by the kernel. A fenced copy of code is still markdown. A hash in a code cell is a comment.
