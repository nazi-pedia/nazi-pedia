This last notebook is a full note. It uses the whole path on one small question.

**Question.** How many pages did the four readers finish, and what is the mean?

| Reader | Pages |
|---|---|
| Ada | 4 |
| Grace | 6 |
| Alan | 3 |
| Linus | 5 |

By hand,

$$
\begin{aligned}
4 + 6 + 3 + 5 &= 18 \\
\frac{18}{4} &= 4.5
\end{aligned}
$$

The mean is four and a half pages.


## Example 1 — The cell repeats the note

### Predict

The cell prints `18`, then `4.5`.


```python
# Hand: 4 + 6 + 3 + 5 = 18, and 18/4 = 4.5.
pages = [4, 6, 3, 5]
total = sum(pages)
mean = total / len(pages)
print(total)
print(mean)
assert total == 18
assert mean == 4.5
```

## Example 2 — Build the list inside the cell

This cell sets the four readings itself. A second press starts from those four numbers again, so the total stays 18. That is the safe shape from lesson 08.

### Predict

The print is `18`.


```python
# Safe to run again: the list is created on every run, then added.
pages = [4, 6, 3, 5]
print(sum(pages))
assert sum(pages) == 18
```

## Example 3 — One extra reading of 5

Linus's 5 is already in the list. A further 5 is a fifth reading.

$$
18 + 5 = 23
$$

### Predict

The cell prints `23`.


```python
# One extra reading of 5. Hand: 18 + 5 = 23.
pages = [4, 6, 3, 5]
pages.append(5)
print(sum(pages))
assert sum(pages) == 23
```

## Pitfall — The second press appends again

A cell that appends, and does not rebuild the list, will add another 5 on every press.

$$
\begin{aligned}
18 + 5 &= 23 \\
23 + 5 &= 28
\end{aligned}
$$

The next cell writes both presses, so a single full run shows both totals.

### Predict

The cell prints `23`, then `28`.


```python
# First press appends one 5. A second press appends another 5.
pages = [4, 6, 3, 5]
pages.append(5)
print(sum(pages))
pages.append(5)
print(sum(pages))
assert sum(pages) == 28
```

## Summary

You now have the page, before the language.

- A notebook is a stack of markdown cells and code cells.
- The output sits under the code cell that made it.
- Shift+Enter runs a cell. Restart and Run All runs the page from the top.
- The kernel remembers names. A second press sees that memory.
- Old output can outlive the kernel. Run again before you trust it.
- Write formulas in markdown, finish them by hand, then let the cell repeat them.
- Put the numbers on the page. A full run has no one to type.
- Save after the run you want the next reader to see.
