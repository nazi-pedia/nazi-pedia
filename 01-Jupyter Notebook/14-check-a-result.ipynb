This is the rhythm of every later path.

1. Write the question in markdown.
2. Finish the arithmetic by hand, in display mathematics.
3. Let a code cell repeat it.
4. The printed value and the hand value have to be the same value.

Three lengths, in the same unit: 5, 5, and 8.

$$
\begin{aligned}
5 + 5 + 8 &= 18 \\
\frac{18}{3} &= 6
\end{aligned}
$$


## Example 1 — The cell repeats the hand result

### Predict

The cell prints `18`, then `6.0`.


```python
# Hand: total 18, mean 18/3 = 6.
lengths = [5, 5, 8]
total = sum(lengths)
mean = total / len(lengths)
print(total)
print(mean)
assert total == 18
assert mean == 6
```

## Example 2 — A new length equal to the old mean

The old mean is 6. Add another length of 6.

$$
\begin{aligned}
m &= \frac{18}{3} = 6 \\
\frac{3m + m}{3 + 1} &= \frac{4m}{4} = m
\end{aligned}
$$

Adding a copy of the mean does not move the mean. In these numbers,

$$
\frac{18 + 6}{4} = \frac{24}{4} = 6
$$

### Predict

The cell prints `24`, then `6.0`.


```python
# The new length equals the old mean, so the mean stays 6.
lengths = [5, 5, 8, 6]
total = sum(lengths)
mean = total / len(lengths)
print(total)
print(mean)
assert total == 24
assert mean == 6
```

## Pitfall — The old count divides the new total

Four lengths added together make 24. Dividing by the old count, 3, produces 8.

$$
\frac{24}{3} = 8
$$

The mean uses the count of the lengths that entered the total.

### Predict

The cell prints `8.0`, then `6.0`.


```python
# 24/3 uses the old count. 24/4 is the mean of the four lengths.
wrong = (5 + 5 + 8 + 6) / 3
right = (5 + 5 + 8 + 6) / 4
print(wrong)
print(right)
assert wrong == 8
assert right == 6
```

## Summary

Compute the result by hand before you read the output. The cell's job is to repeat that result. When the list grows, divide by the new count. A value equal to the mean can join the list and leave the mean where it was.
