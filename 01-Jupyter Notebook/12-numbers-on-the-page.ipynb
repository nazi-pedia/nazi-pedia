Restart and Run All runs the notebook with nobody sitting at the keyboard. Every number the cell needs has to be written on the page.

`input` is a real Python function. Calling it would pause the kernel and wait for a typed line. A full run would sit there. This lesson names `input` and does not call it.

A reading of 12 pages is copied 3 times.

$$
12 \times 3 = 36
$$


## Example 1 — The numbers are names in the cell

### Predict

The cell prints `36`.


```python
# The reading is written here, so a full run never waits for the keyboard.
reading = 12
copies = 3
total = reading * copies
print(total)
assert total == 36
```

## Example 2 — One change: four copies

$$
12 \times 4 = 48
$$

### Predict

The cell prints `48`.


```python
# One change: copies is 4. Hand product: 12 * 4 = 48.
reading = 12
copies = 4
total = reading * copies
print(total)
assert total == 48
```

## Example 3 — Show the pause, and do not start it

### Predict

The cell prints the characters of a call it never makes. `input` exists. It is not used.


```python
# The text shows a call. This cell does not call input, so the kernel does not wait.
shown = "input('How many copies? ')"
print(shown)
assert shown.startswith("input(")
assert callable(input)
```

## Pitfall — A hidden question makes the notebook unable to finish itself

A cell that calls `input` looks interactive and friendly. In a full run it is a lock. The outputs under the later cells stay old, because those cells have not run yet.

Write the reading next to the arithmetic. The page then contains the question and the answer.

## Summary

Put every number the cell needs into the cell, or into an earlier cell of the same notebook. A full run has no one to type. `input` would wait, so these lessons do not call it.
