A story is easier to check when the numbers sit in a list or a table. This lesson writes the cups both ways, computes the total by hand, and lets one cell repeat it.

Three piles:

- First pile: 2 cups
- Second pile: 3 cups
- Third pile: 4 cups

The same facts as a table:

| Pile | Cups |
|---|---|
| First | 2 |
| Second | 3 |
| Third | 4 |

By hand,

$$
2 + 3 + 4 = 9
$$


## Example 1 — The cell repeats the table

### Predict

The cell prints `9`.


````python
# Hand total from the table: 2 + 3 + 4 = 9.
first = 2
second = 3
third = 4
total = first + second + third
print(total)
assert total == 9
````

## Example 2 — One change: the third pile gains a cup

The third pile now holds 5 cups.

$$
2 + 3 + 5 = 10
$$

### Predict

The cell prints `10`. Only the third number changed.


````python
# One change: the third pile is 5. Hand total: 2 + 3 + 5 = 10.
first = 2
second = 3
third = 5
total = first + second + third
print(total)
assert total == 10
````

1. Read the table.

3. Run the cell that repeats the addition.
