Restart and Run All runs each cell once, from the top. Pressing Shift+Enter on a cell runs **that cell again**, and the kernel still holds whatever the first press stored.

That second press is safe when the cell **sets** the numbers itself. It is a different calculation when the cell **adds to** a number that already exists.

By hand, the original total is

$$
2 + 3 + 4 = 9
$$


## Example 1 — A cell that sets its own total

### Predict

The print is `9`. Run this cell once or ten times: each run starts by storing 2, 3, and 4 again, so the print stays `9`.


```python
# Safe to run again: the total is built from these three numbers every time.
cups = 2 + 3 + 4
print(cups)
assert cups == 9
```

## Example 2 — A cell that adds one

$$
9 + 1 = 10
$$

### Predict

A full run executes this cell once, so it prints `10`. A second press on **this** cell would start from 10 and print `11`. The next example writes that second press where a full run can see it.


```python
# One press adds one cup to whatever cups currently means.
cups = cups + 1
print(cups)
assert cups == 10
```

## Example 3 — Two presses, written in order

Start again from 9. The first press reaches 10. The second press reaches 11.

$$
\begin{aligned}
9 + 1 &= 10 \\
10 + 1 &= 11
\end{aligned}
$$

### Predict

The cell prints `10`, then `11`.


```python
# Two presses of "add one", starting from the original total of 9.
seen = 9
seen = seen + 1
print(seen)
seen = seen + 1
print(seen)
assert seen == 11
```

## Pitfall — The cell that only adds is a different program on the second press

Example 1 answers the question "what is 2 + 3 + 4?" every time you run it.

Example 2 answers "what is one more than the number already stored?" The answer depends on how many times the cell has run since the total was set.

Put the starting numbers in the same cell as the calculation when a second press should mean the same thing. Use a cell that only updates when you truly want each press to move the stored value.

## Summary

A second press reuses the kernel's memory. A cell that sets its numbers can be run again and still print the same result. A cell that only adds will print a new result on every press.
