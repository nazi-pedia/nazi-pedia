A code cell can show you something in two different places.

- A call to `print` writes a line into the output.
- The **last line**, when it is a value rather than an assignment, is shown as the **result** of the cell.

An assignment stores a name and shows nothing. The value is in the kernel. The page stays quiet until a later line asks to see it.

By hand, the three piles still total

$$
2 + 3 + 4 = 9
$$


## Example 1 — An assignment is silent

### Predict

The cell stores `9` under the name `total`. Nothing is written under the cell.


```python
# The last line is an assignment, so the notebook shows no result.
total = 2 + 3 + 4
```

## Example 2 — A name on the last line is shown

### Predict

The kernel still holds `9`. A name alone on the last line shows that stored value. You will see `9`.


```python
# Showing the name asks the notebook for the value stored in example 1.
total
```

## Example 3 — A written line and a result can share a cell

By hand,

$$
10 - 4 = 6
$$

### Predict

First a written line, `written line`. Under that, the result `6`.


```python
# print writes the first line. The subtraction is the result under it.
print("written line")
10 - 4
```

## Example 4 — What `/` shows

By hand, $9 / 3 = 3$. Python's `/` produces a decimal, so the result on the page is `3.0`.

### Predict

You will see `3.0`.


```python
# The hand quotient is 3. The slash division displays 3.0.
9 / 3
```

## One change — Integer division

`//` divides and drops the remainder. $9 // 3 = 3$, and the result is the whole number `3`.


```python
# Integer division of 9 by 3 is the whole number 3.
9 // 3
```

## Pitfall — A silent cell can still be wrong

A cell that only assigns will look calm even when the arithmetic is the wrong arithmetic. Show the value, or check it.

Here the silent wrong total would be $2 + 3 + 5 = 10$ if someone had changed a pile and forgotten. The check uses the piles we actually stored.


```python
# Silence is not a check. The assert states the value the story promised.
total = 2 + 3 + 4
assert total == 9
print(total)
```

## Summary

`print` writes a line. The last value of a cell is shown as the result. An assignment does neither: it only stores. A quiet cell still needs a check when the number matters.
