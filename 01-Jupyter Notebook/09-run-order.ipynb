The page has an order: top to bottom. The kernel has a different order: the order in which you actually run the code cells.

Restart and Run All makes those two orders the same. Running the bottom cell first does not. A name is available only after the cell that creates it has run.

The little dictionary below stands in for the kernel's memory, so one full run can show the missing name and then the stored name.


## Example 1 — A fresh memory is empty

Restart clears the kernel. The next cell is that fresh start.

### Predict

The print is `0`.


```python
# A new memory, standing in for a kernel just after Restart.
memory = {}
print(len(memory))
assert len(memory) == 0
```

## Example 2 — The lower question ran too early

### Predict

The name `cups` is not in the memory yet. The cell prints `cups is not in memory yet`.


```python
# This is the lower cell, run before anything has stored cups.
try:
    print(memory["cups"])
except KeyError:
    print("cups is not in memory yet")
```

## Example 3 — The cell that creates the name

By hand,

$$
2 + 3 + 4 = 9
$$

### Predict

After this cell, the memory holds 9. The print is `9`.


```python
# Hand total 9. This is the cell the earlier question needed to have run first.
memory["cups"] = 2 + 3 + 4
print(memory["cups"])
assert memory["cups"] == 9
```

## Example 4 — Restart throws the memory away

### Predict

After the clear, `cups` is gone. The print is `False`.


```python
# Restart discards names. The print under an old cell can still show 9 until you run it again.
memory.clear()
print("cups" in memory)
assert "cups" not in memory
```

## Pitfall — The output can be a photograph of a kernel that is gone

After a real Restart, the page can still show `9` under a cell you ran yesterday. That line is the saved output. The kernel no longer has the name until you run the cell that creates it.

Read a page only after the run that produced the outputs you are reading. Restart and Run All is that run.

## Summary

Page order and run order agree after Restart and Run All. A name exists only after its cell has run in this kernel. Old output can remain on the page after the memory has been cleared.
