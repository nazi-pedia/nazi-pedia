When a code cell finishes, the kernel keeps the names that cell created. The next code cell in the **same notebook** can use them. You do not copy the number down the page unless you want to.

Each notebook has its own kernel. Opening the next file starts a new memory. Restart and Run All on **this** file runs this file from the top.

By hand, the piles are still

$$
2 + 3 + 4 = 9
$$


## Example 1 — Create the name

### Predict

The cell prints `9`.


```python
# Create the name in this cell. Later cells in this notebook can see it.
cups = 2 + 3 + 4
print(cups)
assert cups == 9
```

## Example 2 — A later cell reads the name

### Predict

This cell does not add 2, 3, and 4 again. It reads `cups` and prints `9`.


```python
# The name was created in the cell above. The kernel kept it.
print(cups)
assert cups == 9
```

## Example 3 — One change updates the stored name

One more cup is poured.

$$
9 + 1 = 10
$$

### Predict

The cell prints `10`. After it runs, `cups` means 10 for every later cell in this notebook.


```python
# One more cup. The name now means 10, not 9.
cups = cups + 1
print(cups)
assert cups == 10
```

## Example 4 — The update is what later cells see

### Predict

The print is `10`.


```python
# A later cell sees the updated value.
print(cups)
assert cups == 10
```

## Pitfall — A close spelling is a different name

`cup` was never created. The kernel does not guess that you meant `cups`. The cell below catches the failure so that Restart and Run All can continue. An uncaught failure would stop the full run. Lesson 10 stays with that fact.

### Predict

The cell prints `NameError`, then a line that names `cup`. The stored total is still 10, and the last print says so.


```python
# cup was never assigned. cups, with an s, is the name that exists.
try:
    print(cup)
except NameError as error:
    print(error.__class__.__name__)
    print(error)
print(cups)
assert cups == 10
```

## Summary

A name lives in the kernel after the cell that created it. Later cells in the same notebook see the latest value. A different spelling is a different name, and the kernel will not invent it.
