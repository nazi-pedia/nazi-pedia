You are looking at a notebook. It is a page made of blocks. Each block is a **cell**. Read the page from the top block to the bottom block.

This block is a **markdown** cell. It holds the story: headings, sentences, tables, and formulas. A markdown cell has no number in brackets on the left.

The next block is a **code** cell. It holds work. When it runs, the result appears **under that same cell**. A code cell has brackets on the left. After a full run from the top, those brackets read `[1]`, then `[2]`, then `[3]`, in the order of the code cells on the page.

## How this path teaches

Every example uses the same five steps.

1. **Predict.** Read the sentence and decide what will appear under the cell.
2. **Run.** The cell has already been run. The stored output is the answer key.
3. **One change.** The next example moves a single detail.
4. **Pitfall.** One case looks almost right.
5. **Summary.** The sentences worth keeping.

The next path teaches the Python language. The code in this path stays short on purpose. Its job is to make the page visible.

Try this with your hands before you read ahead.

1. Click inside the code cell below.
2. Press Shift+Enter.
3. Read the line that appears under the cell.
4. Look at the brackets on the left of that cell.


## Example 1 — The output belongs to the cell above it

### Predict

The cell prints one sentence. That sentence appears once, directly under the cell.


```python
# Send one sentence to the output under this cell.
print("The output sits under the cell that made it.")
```

## Example 2 — Two sentences, two lines

### Predict

`print` ends its line and then stops. Two calls make two lines, in the same order as the two calls.


```python
# The first call writes the first line. The second call writes the next line.
print("First line")
print("Second line")
```

## Example 3 — A result you can compute before the cell runs

Three piles hold 2, 3, and 4 cups.

$$
2 + 3 + 4 = 9
$$

### Predict

The cell prints `9`, the same total.


```python
# Hand total: 2 + 3 + 4 = 9. The cell repeats that addition.
total = 2 + 3 + 4
print(total)
assert total == 9
```

## Example 4 — A comment is quiet

### Predict

A line that starts with `#` is a note to a person. The kernel skips it. The brackets still appear, and nothing is written underneath.


```python
# This cell is only a note. A full run gives it brackets and no output.
```

The cell above ran. Its brackets show that it ran. There is no output line under it, because a comment does not write one.

## Pitfall — Words in the story do not run

The next sentence contains the characters `print("hello")`. They sit in a markdown cell, so they stay as words. The code cell after them is the one that runs.


```python
# This is the cell that runs. The same word in the story above stayed as text.
print("hello")
```

## Summary

A notebook is a stack of cells. Markdown tells the story and does not run. Code runs, and the output sits under the code cell that made it. A comment runs as silence.
