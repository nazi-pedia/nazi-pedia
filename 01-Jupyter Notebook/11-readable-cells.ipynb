A code cell is read by two audiences. The kernel runs it. A person has to see what it is for. A comment is for the person. A blank line is space for the person. Neither one changes the numbers.

One cell should carry one idea. The idea in this lesson is a three-pile receipt.

| Pile | Cups |
|---|---|
| First | 2 |
| Second | 3 |
| Third | 4 |

$$
\begin{aligned}
2 + 3 + 4 &= 9 \\
\frac{9}{3} &= 3
\end{aligned}
$$


## Example 1 — The receipt

### Predict

The cell prints `9`, then `3.0`. The comments and the blank line do not add extra output lines.


```python
# Monday desk: three piles of cups.
first = 2
second = 3
third = 4

# The total is the sum. The mean divides by how many piles there are.
total = first + second + third
mean = total / 3
print(total)
print(mean)
assert total == 9
assert mean == 3
```

## Example 3 — A string is a value

### Predict

The sentence appears as a result, and it wears quotes. The notebook is showing you the value, not printing a comment.


```python
# A string on the last line is a value, so the notebook shows it.
"This is a value, so it is shown."
```

## Pitfall — The quote marks are part of the display

A comment and a one-line string can look similar in the editor. The comment leaves the output empty. The string, as the last line, is shown with quotes around it.

Use a comment to explain a step. Use `print` when the reader of the output should see the words. Leave a string as the last line only when you want the value itself on display.

## Summary

Comment the step a reader needs. Leave a blank line between the facts and the calculation. Keep one idea in the cell. A string is a value; a comment is not.
