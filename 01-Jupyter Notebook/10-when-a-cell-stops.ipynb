A code cell is a small program. When that program is broken, the kernel stops **inside that cell** and reports the kind of break.

Restart and Run All also stops at the first cell that breaks. Cells below it keep whatever output they stored last time. The page can then mix a new failure with old results.

The cells in this lesson catch the break, print its name, and finish. A full run of the course can therefore continue. The story next to each cell says what an uncaught break would do.


## Example 1 — A line the kernel cannot read

The broken line is missing the closing parenthesis:

```text
print(2 + 3
```

### Predict

The cell prints `SyntaxError`, then a message that the parenthesis was never closed. The broken line is stored as text. This cell itself can be read.


````python
# Study a broken line without putting the break into this cell.
broken = "print(2 + 3"
kind = ""
message = ""
try:
    compile(broken, "<one line>", "exec")
except SyntaxError as error:
    kind = error.__class__.__name__
    message = error.msg
    print(kind)
    print(message)
assert kind == "SyntaxError"
assert "never closed" in message
````

## Example 2 — A name the kernel has never seen

### Predict

The cell prints `NameError`, then a message that includes `not_a_name`.


````python
# An unknown name is a different break from a line that cannot be read.
kind = ""
message = ""
try:
    print(not_a_name)
except NameError as error:
    kind = error.__class__.__name__
    message = str(error)
    print(kind)
    print(message)
assert kind == "NameError"
assert "not_a_name" in message
````

## Example 3 — A cell should finish

A cell that never ends, such as a loop with no exit, holds the kernel until you interrupt it. The stop button, or the Kernel menu's Interrupt, asks that cell to stop. This course does not run such a loop.

### Predict

The cell prints one line and ends.


````python
# A teaching cell finishes. An endless loop would wait for Interrupt.
print("This cell reaches its last line.")
````

## Pitfall — Cells under a break can show yesterday's answer

Suppose the total cell breaks, and a later cell still shows `9` from last week. The `9` was not recomputed today. Restart and Run All stopped before it reached that later cell.

Fix the broken cell, then run the whole notebook again. Trust the outputs of the run that finished.

## Summary

A syntax error means the line cannot be read. A name error means the name was never stored. An uncaught error stops a full run, and cells below it keep their previous output until you run them again.
