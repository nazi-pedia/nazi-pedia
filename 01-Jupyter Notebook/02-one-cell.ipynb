The **kernel** is the living session behind the page. It is the worker that reads a code cell when you ask it to. Markdown does not go to the kernel.

Two keys matter on the first day.

| Key | What you asked for |
|---|---|
| Shift+Enter | Run this cell, then move to the next cell |
| Ctrl+Enter | Run this cell, and stay here |

The number inside the brackets is a **run counter** for this kernel. It is not a line number, and it is not the cell's place on the page. After Restart and Run All, the counters follow the code cells from top to bottom, because that command runs them once each, in page order.

These keys work in Jupyter Notebook, JupyterLab, and the notebook editor in Cursor.


## Example 1 — One press, one sentence

### Predict

Shift+Enter on the next cell writes one sentence under it.


```python
# One run of this cell writes one sentence.
print("Shift+Enter runs this cell.")
```

## Example 2 — The last line can be the result

By hand,

$$
2 + 3 = 5
$$

### Predict

The cell does not call `print`. Its last line is the sum. The notebook shows that value as a result, without an extra label from us. You will see `5`.


```python
# Hand result: 2 + 3 = 5. The last line asks the notebook to show it.
2 + 3
```

## Example 3 — `print` writes a line

### Predict

The same sum, sent through `print`, appears as a written line. `print` itself does not leave a second result. You will see a single line, `5`.


```python
# print writes the sum. The call itself has nothing left to display.
print(2 + 3)
```

## Example 4 — The next code cell gets the next counter

### Predict

In a full run from the top, this cell runs after the earlier code cells. Its sentence is the next line of the story, and its bracket number is the next count.


```python
# Page order in a full run: earlier code cells already ran.
print("This cell runs after the cells above it.")
```

## Example 5 — A second press adds one more

A cell that only adds one to a stored number will print a larger number if you run **that same cell** again.

The full run executes the next cell once, so it prints `1`. The cell after it is the same addition a second time, written out so this page can show both results without asking you to press twice.

### Predict

First print: `1`. Second print: `2`.


```python
# One full run executes the addition once.
presses = 0
presses = presses + 1
print(presses)
assert presses == 1
```

```python
# A second press of the same addition, written as its own step.
presses = presses + 1
print(presses)
assert presses == 2
```

## Pitfall — The bracket number is not the cell's address

The `2` above is "this addition has happened twice." It is not "this is cell number 2 on the page." Several code cells came before it. After you press Shift+Enter on one old cell, that cell's bracket becomes the newest count, and the numbers down the page no longer match a fresh full run.

Restart and Run All puts the counters back in page order.

## Summary

The kernel runs the code cell you choose. Shift+Enter runs it and moves on. Ctrl+Enter runs it and stays. The bracket number counts runs in this kernel.
