Later paths write proofs between dollar signs. The signs belong in markdown. The kernel does not read them. A code cell then repeats the arithmetic.

Inline mathematics sits inside a sentence. The total of the three piles is $2 + 3 + 4 = 9$.

Display mathematics gets its own block. The mean is the total divided by how many piles there are.

$$
\begin{aligned}
2 + 3 + 4 &= 9 \\
\frac{9}{3} &= 3
\end{aligned}
$$

Open a markdown cell and you will see the dollars and the commands. Shift+Enter renders them. One dollar on each side makes inline mathematics. A pair of dollars makes a display.


## Example 1 — The cell agrees with the display

### Predict

The cell prints the total `9`, then the mean `3.0`. Division with `/` produces a decimal value. Here the decimal is exactly three.


```python
# Hand: total 9, mean 9/3 = 3.
total = 2 + 3 + 4
mean = total / 3
print(total)
print(mean)
assert total == 9
assert mean == 3
```

## Example 2 — One change, the same layout

Four piles: 2, 3, 4, and 7.

$$
\begin{aligned}
2 + 3 + 4 + 7 &= 16 \\
\frac{16}{4} &= 4
\end{aligned}
$$

### Predict

The cell prints `16`, then `4.0`.


```python
# Hand: 2 + 3 + 4 + 7 = 16, and 16/4 = 4.
piles = [2, 3, 4, 7]
total = sum(piles)
mean = total / len(piles)
print(total)
print(mean)
assert total == 16
assert mean == 4
```

## Pitfall — Without dollars, the marks stay as letters

The sentence "frac{9}{3} = 3" is ordinary text. A reader sees the command, not a fraction. The arithmetic is unchanged: the cell can still divide 9 by 3. Dollars change the rendering of the story. They do not change the kernel.


```python
# The formula without dollars is a rendering issue. The quotient is still 3.
quotient = 9 / 3
print(quotient)
assert quotient == 3
```

## Summary

Write a formula in markdown between dollars, render the cell, then let a code cell repeat the arithmetic. Inline math uses one dollar on each side. Display math uses a pair. The dollars are for the reader.
