Two modes share the keyboard.

**Edit mode.** The text cursor is inside the cell. Letters you type go into the cell.

**Command mode.** The cursor is not in the text. The next key talks to the notebook: add a cell, change its type, or delete it. Esc enters command mode. Enter returns to the text cursor.

| Keys | What they do |
|---|---|
| Shift+Enter | Run this cell and move to the next one |
| Ctrl+Enter | Run this cell and stay on it |
| Esc | Enter command mode |
| Enter | Return to the text cursor |
| Esc, then A | Insert a cell above |
| Esc, then B | Insert a cell below |
| Esc, then M | Make the selected cell markdown |
| Esc, then Y | Make the selected cell code |
| Esc, then D, then D | Delete the selected cell |

These run keys and command-mode keys work in Jupyter Notebook, JupyterLab, and the notebook editor in Cursor.

Practice, in order:

1. Click the code cell below.
2. Press Shift+Enter.
3. Press Esc, then B, to insert a cell under it.
4. Press Esc, then M, if that new cell should be a sentence rather than code.
5. If the extra cell was only for practice, press Esc, then D, then D.

## Example 2 — Stay on the cell

Ctrl+Enter runs the cell and leaves the selection where it is. Use it when you want to read the output before you move.

### Predict

The print is `9`, the hand total.


```python
# Ctrl+Enter runs this sum and stays on the cell. Hand total: 9.
total = 2 + 3 + 4
print(total)
assert total == 9
```

## Pitfall — In edit mode, A is the letter a

Command mode is what makes A insert a cell. In edit mode the same key types the letter into your sentence or your code. Press Esc first, look away from the text cursor, then press A or B.

Deleting with D, D has the same requirement. Two D keys in edit mode type `dd`.

## Summary

Shift+Enter runs and moves. Ctrl+Enter runs and stays. Esc lets the next key arrange cells. Enter gives you the text cursor back. Press Esc before A, B, M, Y, or D.
