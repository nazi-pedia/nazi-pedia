A partial derivative with respect to $x$ holds every other variable fixed.
Clairaut's theorem says that when the mixed partial derivatives are continuous, the order of differentiation does not matter.

$$
\begin{aligned}
 \dfrac{\partial^{2} f}{\partial y\,\partial x}
 &=
 \dfrac{\partial^{2} f}{\partial x\,\partial y}.
\end{aligned}
$$

The chain rule applies when $z=f(x,y)$ and both $x$ and $y$ are functions of $t$:

$$
\begin{aligned}
 \dfrac{dz}{dt}
 &=
 \dfrac{\partial f}{\partial x}\dfrac{dx}{dt}
 +
 \dfrac{\partial f}{\partial y}\dfrac{dy}{dt}.
\end{aligned}
$$


## Example 1 — Two partial derivatives

### Hand solution

$$
\begin{aligned}
 f &= x^{2}y+y^{3},\\
 \dfrac{\partial f}{\partial x} &= 2xy,\\
 \dfrac{\partial f}{\partial y} &= x^{2}+3y^{2}.
\end{aligned}
$$

At the point $(1,-1)$,

$$
\begin{aligned}
 \dfrac{\partial f}{\partial x}(1,-1) &= -2,\\
 \dfrac{\partial f}{\partial y}(1,-1) &= 1+3=4.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Coordinates and the function f = x^2 y + y^3.
x, y = sp.symbols("x y")
f = x**2*y + y**3
# The two first partial derivatives.
assert sp.diff(f, x) == 2*x*y
assert sp.diff(f, y) == x**2 + 3*y**2
# Value of the partial with respect to x at (1, -1).
assert sp.diff(f, x).subs({x: 1, y: -1}) == -2
print("Confirmed")
```

## Example 2 — Clairaut

### Hand solution

$$
\begin{aligned}
 f &= x^{3}y^{2},\\
 \dfrac{\partial f}{\partial x} &= 3x^{2}y^{2},\\
 \dfrac{\partial^{2} f}{\partial y\,\partial x} &= 6x^{2}y,\\
 \dfrac{\partial f}{\partial y} &= 2x^{3}y,\\
 \dfrac{\partial^{2} f}{\partial x\,\partial y} &= 6x^{2}y.
\end{aligned}
$$

The mixed partials agree.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Coordinates and the function f = x^3 y^2.
x, y = sp.symbols("x y")
f = x**3 * y**2
# Clairaut: the two mixed partials are equal.
assert sp.diff(f, x, y) == sp.diff(f, y, x)
print(sp.diff(f, x, y))
print("Confirmed")
```

## Example 3 — The chain rule on a circle

### Hand solution

Take $z=x^{2}+y^{2}$ with $x=\cos t$ and $y=\sin t$. Then $z=1$ is constant, so

$$
\begin{aligned}
 \dfrac{dz}{dt} &= 0.
\end{aligned}
$$

The chain rule gives the same conclusion:

$$
\begin{aligned}
 \dfrac{\partial z}{\partial x} &= 2x,\\
 \dfrac{\partial z}{\partial y} &= 2y,\\
 \dfrac{dz}{dt}
 &=
 2\cos t\,(-\sin t)+2\sin t\,(\cos t)
 = 0.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# The circle x = cos t, y = sin t.
t = sp.symbols("t")
x = sp.cos(t)
y = sp.sin(t)
z = x**2 + y**2
# On this path, z is the constant 1 and its derivative is 0.
assert sp.simplify(z) == 1
assert sp.simplify(sp.diff(z, t)) == 0
print(sp.diff(z, t))
print("Confirmed")
```

## Example 4 — Implicit slope from the partial derivatives

### Hand solution

The unit circle is the level set $F(x,y)=x^{2}+y^{2}-1=0$.

$$
\begin{aligned}
 \dfrac{dy}{dx}
 &=
 -\dfrac{\dfrac{\partial F}{\partial x}}{\dfrac{\partial F}{\partial y}}
 =
 -\dfrac{2x}{2y}
 =
 -\dfrac{x}{y}.
\end{aligned}
$$

This is the same slope of the circle as in lesson 09. The partial-derivative formula justifies that shortcut, provided $\dfrac{\partial F}{\partial y}\neq 0$.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# The level-set function of the unit circle.
x, y = sp.symbols("x y")
F = x**2 + y**2 - 1
# Implicit slope -F_x / F_y.
slope = sp.simplify(-sp.diff(F, x)/sp.diff(F, y))
print(slope)
assert slope == -x/y
print("Confirmed")
```

## A pitfall

In $\dfrac{\partial f}{\partial x}$, every $y$ is a number.
If, after differentiating, you also treat $y$ as a function and apply the chain rule again, you have counted one derivative twice.

## Summary

A partial derivative is a one-variable derivative taken while the other variables are held fixed.
Clairaut's theorem lets you swap the order of the mixed partials.
The chain rule weights each input path by its own partial derivative.
