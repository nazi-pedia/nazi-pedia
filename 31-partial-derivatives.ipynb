# ۳۱ · مشتق جزئی، زنجیره و کلرو

| | |
|---|---|
| **دوره** | ریاضی عمومی چندمتغیره |
| **پیش‌نیاز** | ۰۸ و ۳۰ |

مشتق جزئی نسبت به \(x\)، متغیرهای دیگر را ثابت می‌گیرد.
قضیهٔ کلرو: اگر مشتق‌های آمیخته پیوسته باشند، ترتیب مشتق مهم نیست.

\[
f_{xy}=f_{yx}
\]

زنجیره: اگر \(z=f(x,y)\) و \(x\) و \(y\) تابع \(t\) باشند،

\[
\frac{dz}{dt}=f_x\frac{dx}{dt}+f_y\frac{dy}{dt}
\]



## مثال ۱ — دو جزئی

### حل ریاضی

\[
f=x^2 y+y^3
\]
\[
f_x=2xy,\qquad f_y=x^2+3y^2
\]

در \((1,-1)\): \(f_x=-2\) و \(f_y=1+3=4\).

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y")
f = x**2*y + y**3
assert sp.diff(f, x) == 2*x*y
assert sp.diff(f, y) == x**2 + 3*y**2
assert sp.diff(f, x).subs({x: 1, y: -1}) == -2
print("تایید شد")

```

## مثال ۲ — کلرو

### حل ریاضی

\[
f=x^3 y^2
\]
\[
f_x=3x^2 y^2,\qquad f_{xy}=6x^2 y
\]
\[
f_y=2x^3 y,\qquad f_{yx}=6x^2 y
\]

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y")
f = x**3 * y**2
assert sp.diff(f, x, y) == sp.diff(f, y, x)
print(sp.diff(f, x, y))
print("تایید شد")

```

## مثال ۳ — زنجیره روی دایره

### حل ریاضی

\(z=x^2+y^2\)، \(x=\cos t\)، \(y=\sin t\). آن‌گاه \(z=1\) ثابت است و \(dz/dt=0\).

از فرمول:

\[
z_x=2x,\quad z_y=2y
\]
\[
\frac{dz}{dt}=2\cos t(-\sin t)+2\sin t(\cos t)=0
\]

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t")
x = sp.cos(t)
y = sp.sin(t)
z = x**2 + y**2
assert sp.simplify(z) == 1
assert sp.simplify(sp.diff(z, t)) == 0
print(sp.diff(z, t))
print("تایید شد")

```

## مثال ۴ — شیب ضمنی از روی جزئی‌ها

### حل ریاضی

\(F(x,y)=x^2+y^2-1=0\).

\[
\frac{dy}{dx}=-\frac{F_x}{F_y}=-\frac{2x}{2y}=-\frac{x}{y}
\]

همان شیب دایرهٔ درس ۰۹. مشتق جزئی این میان‌بر را توجیه می‌کند، به شرط \(F_y\neq 0\).

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y")
F = x**2 + y**2 - 1
slope = sp.simplify(-sp.diff(F, x)/sp.diff(F, y))
print(slope)
assert slope == -x/y
print("تایید شد")

```

## نکته

در \(f_x\)، هر \(y\) عدد است.
اگر بعد از مشتق، \(y\) را هم تابع بدانید و زنجیره را دوباره اعمال کنید، یک مشتق را دو بار شمرده‌اید.

## جمع‌بندی

جزئی، مشتق یک‌متغیره است با ثابت گرفتن بقیه.
کلرو ترتیب را عوض می‌کند.
زنجیره، هر مسیر ورودی را با جزئی خودش وزن می‌کند.
