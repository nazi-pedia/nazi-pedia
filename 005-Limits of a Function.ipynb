## Defining a Function

```python
from sympy import symbols
x = symbols('x')
f = x**2 + 2*x + 1
print(f)
```

## Finding a Basic Limit

$$ \lim_{x \to 2} (x^2+2x+1)=9 $$

```python
from sympy import symbols, limit
x = symbols('x')
f = x**2 + 2*x + 1
L = limit(f, x, 2)
print(L)
```

## Another Example

$$ \lim_{x \to 1} \left( \frac{x^2-1}{x-1} \right) $$

```python
from sympy import symbols, limit
x = symbols('x')
f = (x**2 - 1) / (x - 1)
L = limit(f, x, 1)
print(L)
```
