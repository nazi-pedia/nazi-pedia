## 1. Comparisons produce booleans

**See.** A comparison is a question. The answer is `True` or `False`, not a modified copy of the numbers.

**Name.** The six comparisons are `==`, `!=`, `<`, `>`, `<=`, `>=`.
One `=` assigns. Two `=` compare.
Python also lets you chain comparisons the way a mathematician writes them:
`70 <= score <= 90` means `score >= 70 and score <= 90`.


````python
a, b, c = 10, 20, 10
print("a == b", a == b)
print("a == c", a == c)
print("a != b", a != b)
print("a > b", a > b)
print("b >= a", b >= a)
print("a <= c", a <= c)

print("5 > 3.5", 5 > 3.5)
print("'apple' < 'banana'", "apple" < "banana")
print("'apple' > 'Apple'", "apple" > "Apple")

try:
    print(10 > "5")
except TypeError as err:
    print("numbers and strings are not ordered against each other:", err)
````

**Read the output.** Strings compare by Unicode code point, character by character.
`"apple" > "Apple"` because the code point of `a` is greater than the code point of `A`.
That is alphabetical only for characters in the same case.

A number and a string cannot be ordered. The failure is a `TypeError`, and the notebook catches it
so the rest of the lesson still runs.


### Own it

Predict whether `18 <= 25 < 65` is true, and whether `3 < 2 < 10` is true.
Chaining does not mean "compare the outer two and ignore the middle" in the way a new reader sometimes hopes.
Every consecutive pair must hold.


````python
age = 25
print("working-age test", 18 <= age < 65)
print("broken chain", 3 < 2 < 10)
assert (18 <= age < 65) is True
assert (3 < 2 < 10) is False
print("checkpoint passed")
````

## 2. Combining questions

**See.** One comparison is rarely the whole decision.
"Weekend, and no homework" is two facts joined by `and`.

**Name.**

| Operator | True when |
|---|---|
| `A and B` | both are true |
| `A or B` | at least one is true |
| `not A` | A is false |

**De Morgan's laws**, which you will use whenever you simplify a condition:

- `not (A and B)` is the same as `(not A) or (not B)`
- `not (A or B)` is the same as `(not A) and (not B)`

`and` and `or` stop early. `False and anything` is false without looking at the second expression.
`True or anything` is true without looking. Do not put a necessary computation in the skipped part.


````python
is_weekend = True
has_homework = False
print("can rest:", is_weekend and not has_homework)

A = True
B = False
print("not (A and B)", not (A and B))
print("(not A) or (not B)", (not A) or (not B))
print("not (A or B)", not (A or B))
print("(not A) and (not B)", (not A) and (not B))
assert (not (A and B)) == ((not A) or (not B))
assert (not (A or B)) == ((not A) and (not B))

def side_effect():
    print("this expression was evaluated")
    return True

print("short-circuit or:")
print(True or side_effect())
print("no short-circuit and:")
print(True and side_effect())
````

## 3. Choosing a branch

**See.** A grade, a root count, or a stopping test is a fork.
You want exactly one of the branches to run.

**Name.**

```python
if condition:
    ...
elif other_condition:
    ...
else:
    ...
```

The colon and the indent are the syntax. Four spaces, used consistently.
Conditions are tested from top to bottom. The first true one wins, and the rest are skipped.
`else` has no condition. It runs only when every test above it failed.

A nested `if` is a second question asked only inside one branch.
A conditional expression is the same idea in one line, for a value rather than a block:

```python
label = "pass" if score >= 70 else "retry"
```


````python
def grade_band(score):
    if score >= 90:
        return "excellent"
    elif score >= 70:
        return "pass"
    elif score >= 50:
        return "retry"
    else:
        return "fail"

for score in (95, 70, 50, 49):
    print(score, "->", grade_band(score))

assert grade_band(95) == "excellent"
assert grade_band(70) == "pass"
assert grade_band(49) == "fail"

score = 85
label = "pass" if score >= 70 else "retry"
print(label)
assert label == "pass"
````

**Misconception.** `elif` is tested even after an earlier branch already ran.

**Correction.** The first true test wins. In `grade_band(95)` the `score >= 70` line never runs,
even though 95 is also at least 70. Order the tests from the most specific to the most general.


### Extension — `match` / `case`

Python 3.10 added `match`, which is useful when one value is compared against several shapes.
This library's baseline is 3.10, so the tool is available. You do not need it for calculus.
The `if` form above remains the default.


````python
def sign_name(n):
    match n:
        case 0:
            return "zero"
        case _ if n > 0:
            return "positive"
        case _:
            return "negative"

assert sign_name(0) == "zero"
assert sign_name(4) == "positive"
assert sign_name(-2) == "negative"
print("match checks passed")
````

## 4. `for` — walk a sequence you already know

**See.** A Riemann sum, a table of function values, and a search through roots
are the same pattern: do this once for each item in a list.

**Name.**

```python
for name in sequence:
    ...
```

`range(stop)` produces `0, 1, ..., stop - 1`. The stop value is excluded.
`range(start, stop)` starts at `start`. `range(start, stop, step)` strides by `step`.
`range` is lazy: it does not build a giant list. `list(range(5))` forces it, which is useful while you are learning and wasteful inside a large loop.

The accumulation pattern is the one calculus will keep using. Start a total at 0. Add one term per pass.


````python
print("range(5) =", list(range(5)))
print("range(1, 5) =", list(range(1, 5)))
print("range(0, 10, 2) =", list(range(0, 10, 2)))

squares = []
for x in range(-2, 3):
    squares.append((x, x ** 2))
print("table of x^2 from -2 to 2:", squares)

total = 0
for k in range(1, 11):
    total = total + k ** 2
print("sum of squares 1^2 + ... + 10^2 =", total)
assert total == 385
````

**Read the output.** `range(-2, 3)` includes −2 and excludes 3, so the table contains the five integers from −2 to 2.
The sum \(1^2 + \cdots + 10^2 = 385\). That closed form is \(n(n+1)(2n+1)/6\) at \(n = 10\).
The loop did not need the formula. Lesson 009 will use the same accumulation to estimate an integral.


### Own it

The sum of the first 5 positive odd numbers is \(1+3+5+7+9 = 25\).
Build it with `range`. Remember that the stop value is excluded.


````python
odd_sum = 0
for n in range(1, 10, 2):
    odd_sum = odd_sum + n
print(odd_sum)
assert odd_sum == 25
print("checkpoint passed")
````

## 5. `while` — repeat until a condition fails

**See.** Sometimes you do not know how many passes you need.
You know the stopping test. "Keep improving the guess until it is close enough."

**Name.** The condition is tested before each pass. If it starts false, the body never runs.
Something inside the body must change the condition. Otherwise the loop never ends.

**Do.** The Babylonian method for \(\sqrt{2}\). Start at 1. Replace the guess \(x\) by the average of \(x\) and \(2/x\).
Stop when \(x^2\) is within \(10^{-12}\) of 2. This is a numerical algorithm with a tolerance,
which is the shape of every later numerical check in this library.


````python
import math

guess = 1.0
steps = 0
tolerance = 1e-12
while abs(guess * guess - 2) > tolerance:
    guess = 0.5 * (guess + 2 / guess)
    steps = steps + 1
    if steps > 100:
        break

print(f"sqrt(2) approximation = {guess:.15f}")
print("steps =", steps)
print(f"math.sqrt(2)        = {math.sqrt(2):.15f}")
assert abs(guess - math.sqrt(2)) < 1e-12
print("checkpoint passed")
````

**Read the output.** A handful of steps is enough. The `if steps > 100: break` is a safety rail.
Every `while` you write in this library gets one, so a mistaken condition cannot spin forever.

`break` leaves the loop immediately. `continue` skips the rest of the current pass and goes back to the test.


````python
print("odds, by skipping evens:")
for n in range(1, 11):
    if n % 2 == 0:
        continue
    print(n, end=" ")
print()

print("stop before 6:")
for n in range(1, 11):
    if n == 6:
        break
    print(n, end=" ")
print()
````

## 6. Functions — name a computation so you can reuse it

**See.** A grade rule and a square-root iteration are useless if they exist only as loose lines.
A function packages inputs, a body, and a returned value.

**Name.**

```python
def function_name(parameter):
    # Say what the function returns.
    return result
```

The definition does not run the body. A call does.
`return` sends a value back and ends the function. Code after `return` in the same block does not run.
A function without `return` returns `None`.

Parameters are names local to the function. Assigning to a parameter does not change the caller's object
when the object is an integer, float, or string.


````python
def polynomial(x):
    # Return x**2 - 5*x + 6. Lesson 004 finds where this is zero.
    return x ** 2 - 5 * x + 6

for x in (0, 2, 3, 1):
    print(f"p({x}) = {polynomial(x)}")

assert polynomial(2) == 0
assert polynomial(3) == 0
assert polynomial(0) == 6
````

**Read the output.** The polynomial is 0 at \(x = 2\) and at \(x = 3\).
Those are its roots. You just met them by evaluation. Lesson 004 will find them without guessing.


## Workshop — classify a quadratic by its discriminant

For \(ax^2 + bx + c\) with \(a \neq 0\), the discriminant is \(D = b^2 - 4ac\).

- \(D > 0\): two distinct real roots
- \(D = 0\): one repeated real root
- \(D < 0\): no real root; the two roots are complex conjugates

Lesson 003 builds those complex numbers. Lesson 004 computes the roots.
Here you only classify. Use exact integer arithmetic, so \(D = 0\) is trustworthy.
Floats would make "exactly zero" a dangerous test.


````python
def root_kind(a, b, c):
    # Classify ax^2 + bx + c by its discriminant. a must be nonzero.
    if a == 0:
        return "not quadratic"
    discriminant = b ** 2 - 4 * a * c
    if discriminant > 0:
        return "two distinct real roots"
    if discriminant == 0:
        return "one repeated real root"
    return "two complex conjugate roots"

# x^2 - 5x + 6 = (x - 2)(x - 3)
assert root_kind(1, -5, 6) == "two distinct real roots"
# (x - 1)^2 = x^2 - 2x + 1
assert root_kind(1, -2, 1) == "one repeated real root"
# x^2 + 1
assert root_kind(1, 0, 1) == "two complex conjugate roots"
# x^2 + 4x + 5, the complex example of lesson 004
assert root_kind(1, 4, 5) == "two complex conjugate roots"
assert root_kind(0, 1, 1) == "not quadratic"

samples = ((1, -5, 6), (1, -2, 1), (1, 0, 1), (1, 4, 5))
for a, b, c in samples:
    print(f"{a}x^2 + ({b})x + ({c}) -> {root_kind(a, b, c)}")
print("workshop passed")
````

## Mistakes this lesson is meant to prevent

- An off-by-one from forgetting that `range` excludes its stop.
- A `while` whose body never changes the tested names.
- Tests in the wrong order, so a specific case is swallowed by a general one.
- Using `=` inside `if`. Assignment is not a comparison.
- Forgetting `return`, then testing the function and receiving `None`.
- Testing `discriminant == 0` on floats. Classify with integers, or compare a float discriminant with a tolerance and a separate "nearly zero" label.
