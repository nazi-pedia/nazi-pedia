# ۱۸ · انتگرال ناسره

| | |
|---|---|
| **دوره** | ریاضی عمومی ۲ |
| **پیش‌نیاز** | ۱۶ و ۰۶ |

انتگرال ناسره حدِ یک انتگرال سره است.
یا یک کران بی‌نهایت است، یا تابع داخل بازه بی‌کران می‌شود.

\[
\int_1^{\infty}f=\lim_{b\to\infty}\int_1^b f
\]

اگر این حد عددی تمام شود، انتگرال همگراست. وگرنه واگراست.

آزمون \(p\): \(\int_1^{\infty} x^{-p}\,dx\) همگراست دقیقاً وقتی \(p>1\).
نزدیک صفر، \(\int_0^1 x^{-p}\,dx\) همگراست دقیقاً وقتی \(p<1\).



## مثال ۱ — \(p=2\)

### حل ریاضی

\[
\int_1^{\infty}x^{-2}\,dx
=\lim_{b\to\infty}\Bigl[-\frac{1}{x}\Bigr]_1^b
=\lim_{b\to\infty}\left(-\frac{1}{b}+1\right)=1
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", positive=True)
b = sp.symbols("b", positive=True)
partial = sp.integrate(x**(-2), (x, 1, b))
exact = sp.limit(partial, b, sp.oo)
library = sp.integrate(x**(-2), (x, 1, sp.oo))
print(partial, "->", exact)
assert exact == 1 and library == 1
print("تایید شد")

```

## مثال ۲ — تکینگی در انتهای بازه

### حل ریاضی

\[
\int_0^1 x^{-1/2}\,dx
=\lim_{a\to 0^+}\Bigl[2x^{1/2}\Bigr]_a^1
=\lim_{a\to 0^+}(2-2\sqrt{a})=2
\]

توان \(1/2<1\) است، پس انتظار همگرایی داشتیم. مساحت زیر \(1/\sqrt{x}\) نزدیک صفر بی‌نهایت بلند است، ولی نازک، و مساحتش تمام می‌شود.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", positive=True)
exact = sp.integrate(x**(-sp.Rational(1, 2)), (x, 0, 1))
print(exact)
assert exact == 2
print("تایید شد")

```

## مثال ۳ — همساز واگراست

### حل ریاضی

\[
\int_1^{\infty}\frac{1}{x}\,dx=\lim_{b\to\infty}\ln b=+\infty
\]

این مرز آزمون \(p\) است: \(p=1\) همگرا نیست.
سری همساز در درس ۲۴ از همین انتگرال تغذیه می‌کند.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", positive=True)
exact = sp.integrate(1/x, (x, 1, sp.oo))
print(exact)
assert exact == sp.oo
print("تایید شد")

```

## مثال ۴ — \(p=3/2\)

### حل ریاضی

\[
\int_1^{\infty}x^{-3/2}\,dx
=\lim_{b\to\infty}\Bigl[-2x^{-1/2}\Bigr]_1^b
=\lim_{b\to\infty}\left(-\frac{2}{\sqrt{b}}+2\right)=2
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", positive=True)
exact = sp.integrate(x**(-sp.Rational(3, 2)), (x, 1, sp.oo))
print(exact)
assert exact == 2
print("تایید شد")

```

## مثال ۵ — گاوسی، به‌عنوان حدی که پادمشتق مقدماتی ندارد

### حل ریاضی

\(\int e^{-x^2}\,dx\) پادمشتق مقدماتی ندارد. با این حال

\[
\int_{-\infty}^{\infty}e^{-x^2}\,dx=\sqrt{\pi}
\]

اثبات استاندارد از انتگرال دوگانه در مختصات قطبی می‌آید و درس ۳۵ ابزار آن را می‌سازد.
اینجا نتیجه را به‌عنوان یک حدِ تمام‌شده می‌پذیریم و با کتابخانه مقابله می‌کنیم.
نصف آن، از ۰ تا \(\infty\)، برابر \(\sqrt{\pi}/2\) است، چون تابع زوج است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
full = sp.integrate(sp.exp(-x**2), (x, -sp.oo, sp.oo))
half = sp.integrate(sp.exp(-x**2), (x, 0, sp.oo))
print(full, half)
assert sp.simplify(full - sp.sqrt(sp.pi)) == 0
assert sp.simplify(half - sp.sqrt(sp.pi)/2) == 0
print("تایید شد")

```

## نکته

«بی‌نهایت در بی‌نهایت» را جانشینی نکنید. اول پادمشتق را روی بازهٔ متناهی بنویسید، بعد حد.
واگرایی یک جواب است، نه یک خطا.
