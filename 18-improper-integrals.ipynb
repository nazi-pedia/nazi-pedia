An improper integral is the limit of a proper integral. Either one limit is infinite, or the integrand becomes unbounded on the interval.

$$
\int_1^{\infty} f=\lim_{b\to\infty}\int_1^b f.
$$

If that limit is a finite number, the integral converges. Otherwise it diverges.

The $p$-test separates the two ends. At infinity,

$$
\int_1^{\infty} x^{-p}\,dx
$$

converges exactly when $p>1$. Near zero,

$$
\int_0^1 x^{-p}\,dx
$$

converges exactly when $p<1$.


## Example 1 — $p=2$

### Hand solution

$$
\begin{aligned}
\int_1^{\infty} x^{-2}\,dx &= \lim_{b\to\infty}\left[-\dfrac{1}{x}\right]_1^b \\
&= \lim_{b\to\infty}\left(-\dfrac{1}{b}+1\right) \\
&= 1.
\end{aligned}
$$

### Python recheck


```python
import sympy as sp

# Antiderivative on [1, b], then the limit b -> infinity.
x = sp.symbols("x", positive=True)
b = sp.symbols("b", positive=True)
partial = sp.integrate(x**(-2), (x, 1, b))
exact = sp.limit(partial, b, sp.oo)
library = sp.integrate(x**(-2), (x, 1, sp.oo))
print(partial, "->", exact)
assert exact == 1 and library == 1
print("Confirmed")
```

## Example 2 — A singularity at an endpoint

### Hand solution

$$
\begin{aligned}
\int_0^1 x^{-1/2}\,dx &= \lim_{a\to 0^+}\left[2x^{1/2}\right]_a^1 \\
&= \lim_{a\to 0^+}\left(2-2\sqrt{a}\right) \\
&= 2.
\end{aligned}
$$

The power $\dfrac{1}{2}$ is less than $1$, so convergence was expected. The region under $\dfrac{1}{\sqrt{x}}$ near zero is infinitely tall and thin, and its area is finite.

### Python recheck


```python
import sympy as sp

# Singular at 0, with p = 1/2 < 1.
x = sp.symbols("x", positive=True)
exact = sp.integrate(x**(-sp.Rational(1, 2)), (x, 0, 1))
print(exact)
assert exact == 2
print("Confirmed")
```

## Example 3 — The harmonic integral diverges

### Hand solution

$$
\int_1^{\infty}\dfrac{1}{x}\,dx=\lim_{b\to\infty}\ln b=+\infty.
$$

This is the boundary of the $p$-test: the case $p=1$ does not converge. The harmonic series in lesson 24 draws on this integral.

### Python recheck


```python
import sympy as sp

# The p = 1 integral diverges to infinity.
x = sp.symbols("x", positive=True)
exact = sp.integrate(1/x, (x, 1, sp.oo))
print(exact)
assert exact == sp.oo
print("Confirmed")
```

## Example 4 — $p=\dfrac{3}{2}$

### Hand solution

$$
\begin{aligned}
\int_1^{\infty} x^{-3/2}\,dx &= \lim_{b\to\infty}\left[-2x^{-1/2}\right]_1^b \\
&= \lim_{b\to\infty}\left(-\dfrac{2}{\sqrt{b}}+2\right) \\
&= 2.
\end{aligned}
$$

### Python recheck


```python
import sympy as sp

# The p = 3/2 integral converges.
x = sp.symbols("x", positive=True)
exact = sp.integrate(x**(-sp.Rational(3, 2)), (x, 1, sp.oo))
print(exact)
assert exact == 2
print("Confirmed")
```

## Example 5 — The Gaussian, a limit with no elementary antiderivative

### Hand solution

There is no elementary antiderivative:

$$
\int e^{-x^2}\,dx.
$$

The improper integral is still a finite number,

$$
\int_{-\infty}^{\infty} e^{-x^2}\,dx=\sqrt{\pi}.
$$

The standard proof comes from a double integral in polar coordinates, and lesson 35 builds the tools for it. Here the result is accepted as a finished limit and checked against the library. The integrand is even, so the half-line integral is half of that value:

$$
\int_0^{\infty} e^{-x^2}\,dx=\dfrac{\sqrt{\pi}}{2}.
$$

### Python recheck


```python
import sympy as sp

# Full Gaussian integral, and the half-line value for an even integrand.
x = sp.symbols("x")
full = sp.integrate(sp.exp(-x**2), (x, -sp.oo, sp.oo))
half = sp.integrate(sp.exp(-x**2), (x, 0, sp.oo))
print(full, half)
assert sp.simplify(full - sp.sqrt(sp.pi)) == 0
assert sp.simplify(half - sp.sqrt(sp.pi)/2) == 0
print("Confirmed")
```

## A pitfall

Do not feed $\infty$ into an antiderivative as if it were a number. First write the antiderivative on a finite interval, then take the limit. Divergence is an answer, not an error.

## Summary

Improper means a limit. The power $p$ sets the boundary of convergence at infinity and at zero separately.
