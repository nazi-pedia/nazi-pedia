# ۳۰ · تابع چندمتغیره و حد در صفحه

| | |
|---|---|
| **دوره** | ریاضی عمومی چندمتغیره |
| **پیش‌نیاز** | ۰۴ و ۲۸ |

\(f(x,y)\) به هر نقطهٔ مجاز صفحه یک عدد نسبت می‌دهد.
حد در \((a,b)\) فقط وقتی هست که از هر راهی که به نقطه نزدیک شوید همان عدد را بگیرید.
یک مسیر موافق، حد را ثابت نمی‌کند. دو مسیر مخالف، حد را باطل می‌کند.

در مختصات قطبی، \(x=r\cos\theta\) و \(y=r\sin\theta\).
اگر عبارت بعد از جایگزینی مضربی از \(r\) باشد که نسبت به \(\theta\) کراندار است، حد صفر است.



## مثال ۱ — خواندن دامنه و یک مقدار

### حل ریاضی

\[
f(x,y)=\sqrt{4-x^2-y^2}
\]

دامنه قرص \(x^2+y^2\le 4\) است. \(f(0,0)=2\) و روی مرز صفر است.
این نیم‌کرهٔ بالایی به شعاع ۲ است.

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y", real=True)
f = sp.sqrt(4 - x**2 - y**2)
assert f.subs({x: 0, y: 0}) == 2
assert f.subs({x: 2, y: 0}) == 0
print(f.subs({x: 0, y: 0}))
print("تایید شد")

```

## مثال ۲ — حدی که واقعاً صفر است

### حل ریاضی

\[
\frac{x^2 y}{x^2+y^2}
\]

در قطبی: \(x=r\cos\theta\)، \(y=r\sin\theta\).

\[
\frac{r^2\cos^2\theta\cdot r\sin\theta}{r^2}=r\cos^2\theta\sin\theta
\]

قدر مطلق حداکثر \(r\) است، چون \(|\cos^2\sin|\le 1\). پس وقتی \(r\to 0\)، عبارت به ۰ می‌رود، مستقل از \(\theta\).

### بازبینی با پایتون



```python
import sympy as sp
r, theta = sp.symbols("r theta", real=True)
polar = sp.simplify((r*sp.cos(theta))**2 * (r*sp.sin(theta)) / ((r*sp.cos(theta))**2 + (r*sp.sin(theta))**2))
print(polar)
assert sp.limit(polar, r, 0) == 0
print("تایید شد")

```

## مثال ۳ — دو مسیر، دو جواب

### حل ریاضی

\[
\frac{x^2-y^2}{x^2+y^2}
\]

روی \(y=0\): صورت و مخرج \(x^2\)، حد ۱.
روی \(x=0\): صورت \(-y^2\) و مخرج \(y^2\)، حد \(-1\).
حد دوبعدی وجود ندارد.

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y")
f = (x**2 - y**2)/(x**2 + y**2)
along_x = sp.limit(f.subs(y, 0), x, 0)
along_y = sp.limit(f.subs(x, 0), y, 0)
print(along_x, along_y)
assert along_x == 1 and along_y == -1
print("تایید شد")

```

## مثال ۴ — مسیر سهمی کافی نیست که تنها آزمون باشد

### حل ریاضی

همان کسر مثال ۳ روی \(y=mx\) برابر \(\dfrac{1-m^2}{1+m^2}\) است.
این مقدار به \(m\) بستگی دارد. حتی خانوادهٔ همهٔ خط‌ها هم یک حد واحد نمی‌دهد.
لازم نیست مسیر عجیب بسازید. اختلاف دو خط بس است.

### بازبینی با پایتون



```python
import sympy as sp
x, m = sp.symbols("x m")
f = (x**2 - (m*x)**2)/(x**2 + (m*x)**2)
along = sp.simplify(sp.limit(f, x, 0))
print(along)
assert along.subs(m, 0) == 1
assert along.subs(m, 1) == 0
print("تایید شد")

```

## نکته

تقارن قطبی را وقتی به کار ببرید که درجهٔ صورت از درجهٔ مخرج بیشتر باشد.
اگر درجه‌ها برابر باشند، اول دو مسیر امتحان کنید.

## جمع‌بندی

یک مسیر حد را اثبات نمی‌کند.
دو مسیر متفاوت آن را رد می‌کند.
قطبی، وقتی عامل \(r\) جلوی یک عبارت کراندار نشسته باشد، اثبات است.
