تابع گویا خارج‌قسمت دو چندجمله‌ای است.
نمودارش یا خطی را در دوردست تعقیب می‌کند، یا کنار یک خط قائم می‌گریزد،
یا سوراخی دارد که از دور دیده نمی‌شود.
حد در درس ۰۶ همین تصویر را دقیق می‌کند. اینجا خود تصویر را می‌سازیم.

## تعریف‌ها

اگر درجهٔ مخرج بیشتر باشد، مجانب افقی \(y=0\) است.
اگر درجه‌ها برابر باشند، مجانب افقی نسبت ضریب‌های پیشرو است.
اگر درجهٔ صورت دقیقاً یکی بیشتر باشد، خارج‌قسمت تقسیم، مجانب مایل است.
عامل مشترک صورت و مخرج، اگر بعد از ساده شدن حذف شود، سوراخ است نه مجانب.
عامل باقی‌مانده در مخرج، مجانب قائم است.



```python
import sympy as sp

def check(found, hand, label="نتیجه"):
    found_s = sp.simplify(found)
    hand_s = sp.simplify(hand)
    difference = sp.simplify(found_s - hand_s)
    print(label)
    print("  حل دستی:", hand_s)
    print("  پایتون:", found_s)
    print("  اختلاف:", difference)
    assert difference == 0
    print("  تایید شد")

```

## مثال ۱ — مجانب افقی و قائم

### حل ریاضی

\[
f(x)=\frac{2x+1}{x-3}
\]

مخرج در \(x=3\) صفر است و صورت آنجا \(7\neq 0\) است. خط \(x=3\) مجانب قائم است.

درجهٔ صورت و مخرج هر دو ۱ است. نسبت ضریب‌های پیشرو \(2/1=2\) است،
پس \(y=2\) مجانب افقی است. تقسیم هم همین را می‌گوید:

\[
f(x)=2+\frac{7}{x-3}
\]

وقتی \(|x|\) بزرگ شود، کسر کوچک می‌شود و نمودار به خط \(y=2\) می‌چسبد.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = (2*x + 1)/(x - 3)
rewritten = sp.apart(f, x)
print("تجزیه:", rewritten)
print("حد در +بی‌نهایت:", sp.limit(f, x, sp.oo))
print("حد در -بی‌نهایت:", sp.limit(f, x, -sp.oo))
print("حد راست x=3:", sp.limit(f, x, 3, dir="+"))
assert rewritten == 2 + 7/(x - 3)
assert sp.limit(f, x, sp.oo) == 2
assert sp.limit(f, x, 3, dir="+") == sp.oo
print("تایید شد")

```

## مثال ۲ — سوراخ، نه مجانب

### حل ریاضی

\[
g(x)=\frac{x^2-1}{x-1}=\frac{(x-1)(x+1)}{x-1}
\]

برای \(x\neq 1\)، \(g(x)=x+1\).
در \(x=1\) تابع تعریف نشده، ولی صورت و مخرج هر دو صفر می‌شوند
و پس از حذف، مقدار \(2\) می‌ماند.
نمودار، خط \(y=x+1\) است با یک نقطهٔ حذف‌شده در \((1,2)\).
مجانب قائم ندارد.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
g = (x**2 - 1)/(x - 1)
print("ساده:", sp.cancel(g))
print("جایگذاری مستقیم:", g.subs(x, 1))
print("حد:", sp.limit(g, x, 1))
assert sp.cancel(g) == x + 1
assert sp.limit(g, x, 1) == 2
print("تایید شد")

```

## مثال ۳ — صورت سنگین‌تر نیست؛ مجانب افقی صفر هم نیست

### حل ریاضی

\[
h(x)=\frac{x^2}{x^2+1}
\]

مخرج همیشه دست‌کم ۱ است، پس مجانب قائم نداریم.
درجه‌ها برابرند و نسبت ضریب‌های پیشرو ۱ است:

\[
h(x)=1-\frac{1}{x^2+1}
\]

مجانب افقی \(y=1\) است و نمودار هرگز به آن نمی‌رسد، چون کسر مثبتِ کم‌شونده از ۱ جدا شده است.
بیشینه در کار نیست؛ تابع از زیر به ۱ نزدیک می‌شود. در \(x=0\) مقدار ۰ است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", real=True)
h = x**2/(x**2 + 1)
print(sp.apart(h, x))
assert sp.limit(h, x, sp.oo) == 1
assert h.subs(x, 0) == 0
assert sp.simplify(h - 1) == -1/(x**2 + 1)
print("تایید شد")

```

## مثال ۴ — مجانب مایل

### حل ریاضی

\[
k(x)=\frac{x^2+1}{x}=x+\frac{1}{x},\qquad x\neq 0
\]

درجهٔ صورت یکی بیشتر است. خارج‌قسمت تقسیم، خط \(y=x\) است.
جملهٔ \(1/x\) در بی‌نهایت محو می‌شود، پس \(y=x\) مجانب مایل است.
\(x=0\) مجانب قائم است، چون مخرج صفر و صورت ۱ است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
k = (x**2 + 1)/x
print(sp.apart(k))
assert sp.apart(k) == x + 1/x
assert sp.limit(k - x, x, sp.oo) == 0
assert sp.limit(k, x, 0, dir="+") == sp.oo
print("فاصله تا خط y=x در بی‌نهایت صفر است")
print("تایید شد")

```

## مثال ۵ — خواندن نمودار از روی تجزیه

### حل ریاضی

\[
p(x)=\frac{x-2}{(x-2)(x+1)}=\frac{1}{x+1},\qquad x\neq 2
\]

عامل \(x-2\) حذف می‌شود: سوراخ در \(x=2\)، و ارتفاع سوراخ \(\dfrac{1}{2+1}=\dfrac{1}{3}\).
عامل \(x+1\) می‌ماند: مجانب قائم \(x=-1\).
درجهٔ مخرج بیشتر است: مجانب افقی \(y=0\).

سه پدیدهٔ این درس در یک کسر جمع شده‌اند.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
p = (x - 2)/((x - 2)*(x + 1))
print("پس از حذف:", sp.cancel(p))
print("ارتفاع سوراخ:", sp.limit(p, x, 2))
print("مجانب افقی:", sp.limit(p, x, sp.oo))
print("سمت راست x=-1:", sp.limit(p, x, -1, dir="+"))
assert sp.cancel(p) == 1/(x + 1)
assert sp.limit(p, x, 2) == sp.Rational(1, 3)
assert sp.limit(p, x, sp.oo) == 0
print("تایید شد")

```

## نکته

صفر مشترک صورت و مخرج را پیش از اعلام مجانب قائم خط بزنید.
اگر بعد از خط زدن چیزی از آن عامل نماند، آنجا سوراخ است.
مجانب افقی را با «عدد بزرگ بگذار» حدس نزنید؛ نسبت درجه‌ها جواب را قطعی می‌کند.

آنچه به صفر میل می‌کند، تصحیح است. آنچه می‌ماند، خط مجانب است.
