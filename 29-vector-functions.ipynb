# ۲۹ · تابع برداری

| | |
|---|---|
| **دوره** | ریاضی عمومی چندمتغیره |
| **پیش‌نیاز** | ۲۶ و ۲۸ |

\[
\mathbf{r}(t)=(x(t),y(t),z(t))
\]

مشتق، بردار سرعت است و مشتق دوم شتاب.
تندی، طول سرعت است. طول خم، انتگرال تندی است.



## مثال ۱ — مارپیچ

### حل ریاضی

\(\mathbf{r}(t)=(\cos t,\sin t,t)\).

\[
\mathbf{r}'=(-\sin t,\cos t,1),\qquad |\mathbf{r}'|=\sqrt{2}
\]

تندی ثابت است. از ۰ تا \(2\pi\) طول \(2\pi\sqrt{2}\) است.

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t", real=True)
r = sp.Matrix([sp.cos(t), sp.sin(t), t])
speed = sp.simplify(r.diff(t).norm())
length = sp.integrate(speed, (t, 0, 2*sp.pi))
print(speed, length)
assert speed == sp.sqrt(2)
assert length == 2*sp.pi*sp.sqrt(2)
print("تایید شد")

```

## مثال ۲ — شتاب یک سهمی فضایی

### حل ریاضی

\(\mathbf{r}(t)=(t,t^2,0)\).

\[
\mathbf{v}=(1,2t,0),\qquad \mathbf{a}=(0,2,0)
\]

شتاب ثابت و رو به بالای صفحهٔ \(xy\) است. این همان \(y''=2\) برای \(y=t^2\) است.

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t")
r = sp.Matrix([t, t**2, 0])
assert r.diff(t) == sp.Matrix([1, 2*t, 0])
assert r.diff(t, 2) == sp.Matrix([0, 2, 0])
print(r.diff(t), r.diff(t, 2))
print("تایید شد")

```

## مثال ۳ — تصویر سرعت روی مماس

### حل ریاضی

در مارپیچ مثال ۱، سرعت هرگز صفر نیست، پس خم هموار است.
بردار مماس واحد \(\mathbf{r}'/|\mathbf{r}'|=(-\sin t,\cos t,1)/\sqrt{2}\) است.
در \(t=0\) این بردار \((0,1,1)/\sqrt{2}\) است.

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t")
r = sp.Matrix([sp.cos(t), sp.sin(t), t])
tangent = sp.simplify(r.diff(t)/r.diff(t).norm())
print(tangent.subs(t, 0))
assert tangent.subs(t, 0) == sp.Matrix([0, 1, 1])/sp.sqrt(2)
print("تایید شد")

```

## نکته

مشتق مؤلفه‌به‌مؤلفه است.
طول مشتق با مشتق طول یکی نیست. تندی می‌تواند ثابت باشد در حالی که جهت سرعت عوض می‌شود. دایره همین است.
