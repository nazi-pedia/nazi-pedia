$$f(x)=x^3-3x^2+4$$

```python
from sympy import symbols, diff, solve, latex

# ─────────────────────────────────────────────
# STEP 1: Define the variable and the function
# ─────────────────────────────────────────────
x = symbols('x')
f = x**3 - 3*x**2 + 4

print("Function:")
print(f"f(x) = {f}")
print()

# ─────────────────────────────────────────────
# STEP 2: Compute the first derivative
# ─────────────────────────────────────────────
f_prime = diff(f, x)
print("First derivative:")
print(f"f'(x) = {f_prime}")
print()

# ─────────────────────────────────────────────
# STEP 3: Solve f'(x) = 0 to find critical points
# ─────────────────────────────────────────────
critical_points = solve(f_prime, x)
print("Critical points (where f'(x) = 0):")
print(critical_points)
print()

# ─────────────────────────────────────────────
# STEP 4: Second derivative for classification
# ─────────────────────────────────────────────
f_double_prime = diff(f_prime, x)
print("Second derivative:")
print(f"f''(x) = {f_double_prime}")
print()

# ─────────────────────────────────────────────
# STEP 5: Classify each critical point
# ─────────────────────────────────────────────
classification = {}
for point in critical_points:
    second_val = f_double_prime.subs(x, point)
    
    if second_val > 0:
        classification[point] = 'Local Minima'
    elif second_val < 0:
        classification[point] = 'Local Maxima'
    else:
        classification[point] = 'Inconclusive (use 1st derivative test)'
    
    # Also compute the function value at the critical point
    f_val = f.subs(x, point)
    print(f"At x = {point}:")
    print(f"  f({point}) = {f_val}")
    print(f"  f''({point}) = {second_val}")
    print(f"  → {classification[point]}")
    print()

# ─────────────────────────────────────────────
# STEP 6: Summary
# ─────────────────────────────────────────────
print("=" * 50)
print("SUMMARY")
print("=" * 50)
print(f"Function: f(x) = {f}")
print(f"Critical Points: {critical_points}")
print(f"Classification: {classification}")
```
