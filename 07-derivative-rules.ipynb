# ۰۷ · مشتق: تعریف، توان، ضرب، خارج‌قسمت

| | |
|---|---|
| **دوره** | ریاضی عمومی ۱ |
| **پیش‌نیاز** | ۰۴ · حد |

مشتق در \(a\)، حد شیب خط‌های قاطع است.

\[
f'(a)=\lim_{h\to 0}\frac{f(a+h)-f(a)}{h}
\]

در \(h=0\) این کسر \(0/0\) است. اول ساده کنید، بعد حد بگیرید.
\(f'(x)\) یک تابع است. \(f'(a)\) یک عدد است: شیب خط مماس.

## قاعده‌هایی که از تعریف می‌آیند

\[
(x^n)'=nx^{n-1}
\]
\[
(uv)'=u'v+uv'
\]
\[
\left(\frac{u}{v}\right)'=\frac{u'v-uv'}{v^2}
\]

جمع، جمله به جمله مشتق می‌گیرد. عدد ثابت جلوی تابع بیرون می‌ماند.



```python
import sympy as sp

def check(found, hand, label="نتیجه"):
    found_s = sp.simplify(found)
    hand_s = sp.simplify(hand)
    difference = sp.simplify(found_s - hand_s)
    print(label)
    print("  حل دستی:", hand_s)
    print("  پایتون:", found_s)
    print("  اختلاف:", difference)
    assert difference == 0
    print("  تایید شد")

```

## مثال ۱ — تعریف، تا آخر

### حل ریاضی

\(f(x)=x^2\) و \(a=3\).

\[
\frac{(3+h)^2-9}{h}=\frac{9+6h+h^2-9}{h}=6+h
\]

\[
f'(3)=\lim_{h\to 0}(6+h)=6
\]

قاعدهٔ توان هم \(2x\) می‌دهد و در ۳ همان ۶ است.
خط مماس: \(y-9=6(x-3)\)، یعنی \(y=6x-9\).

### بازبینی با پایتون



```python
import sympy as sp
x, h = sp.symbols("x h")
f = x**2
quotient = sp.simplify(((3 + h)**2 - 9)/h)
from_definition = sp.limit(quotient, h, 0)
from_rule = sp.diff(f, x).subs(x, 3)
print("خارج‌قسمت:", quotient)
print("از تعریف:", from_definition, " از قاعده:", from_rule)
assert quotient == 6 + h
assert from_definition == from_rule == 6
print("تایید شد")

```

## مثال ۲ — توان روی یک چندجمله‌ای

### حل ریاضی

\[
f(x)=x^4-3x^2+7
\]
\[
f'(x)=4x^3-6x
\]

مشتق عدد ثابت صفر است. \(7\) در \(f'\) نیست، ولی در مقدار \(f\) هست.
\(f'(1)=4-6=-2\).

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = x**4 - 3*x**2 + 7
derivative = sp.diff(f, x)
print(derivative, "در ۱:", derivative.subs(x, 1))
assert sp.simplify(derivative - (4*x**3 - 6*x)) == 0
assert derivative.subs(x, 1) == -2
print("تایید شد")

```

## مثال ۳ — ضرب، نه ضربِ مشتق‌ها

### حل ریاضی

\[
u=x^2+1,\quad v=x-4
\]
\[
(uv)'=2x(x-4)+(x^2+1)(1)=2x^2-8x+x^2+1=3x^2-8x+1
\]

اگر فقط \(u'v'\) را حساب کنید، \(2x\cdot 1=2x\) می‌شود و غلط است.
بسط مستقیم \((x^2+1)(x-4)=x^3-4x^2+x-4\) و مشتق \(3x^2-8x+1\) همان جواب است. این بازبینی دستی دوم است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
product = (x**2 + 1)*(x - 4)
by_rule = 2*x*(x - 4) + (x**2 + 1)
by_diff = sp.diff(product, x)
print(sp.expand(by_rule))
assert sp.expand(by_rule - by_diff) == 0
assert sp.expand(by_diff) == 3*x**2 - 8*x + 1
print("تایید شد")

```

## مثال ۴ — خارج‌قسمت

### حل ریاضی

\[
f(x)=\frac{x^2+1}{x-1},\qquad x\neq 1
\]
\[
f'(x)=\frac{2x(x-1)-(x^2+1)}{(x-1)^2}
=\frac{2x^2-2x-x^2-1}{(x-1)^2}
=\frac{x^2-2x-1}{(x-1)^2}
\]

در \(x=2\): صورت \(4-4-1=-1\) و مخرج ۱، پس \(f'(2)=-1\).
مقدار تابع \(f(2)=5\) است. شیب مماس \(-1\) و نقطه \((2,5)\):

\[
y-5=-(x-2),\qquad y=-x+7
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = (x**2 + 1)/(x - 1)
derivative = sp.simplify(sp.diff(f, x))
hand = (x**2 - 2*x - 1)/(x - 1)**2
print(derivative)
assert sp.simplify(derivative - hand) == 0
assert derivative.subs(x, 2) == -1
assert f.subs(x, 2) == 5
print("تایید شد")

```

## مثال ۵ — شیب عدد است، تابع نیست

### حل ریاضی

\[
f(x)=x^3-x,\qquad f'(x)=3x^2-1
\]

سؤال «مشتق در \(x=1\)» عدد می‌خواهد: \(f'(1)=2\).
سؤال «تابع مشتق» عبارت \(3x^2-1\) را می‌خواهد.
خط مماس در \(x=1\): \(f(1)=0\)، پس \(y=2(x-1)\).

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = x**3 - x
fp = sp.diff(f, x)
print("f' =", fp)
print("f'(1) =", fp.subs(x, 1))
tangent = fp.subs(x, 1)*(x - 1) + f.subs(x, 1)
print("مماس:", sp.expand(tangent))
assert fp.subs(x, 1) == 2
assert sp.expand(tangent) == 2*x - 2
print("تایید شد")

```

## نکته

ترتیب تفریق در خارج‌قسمت \(u'v-uv'\) است، نه برعکس.
یک علامت اشتباه، همهٔ نقاط بحرانی درس ۱۳ را جابه‌جا می‌کند.

## جمع‌بندی

تعریف را یک بار تا آخر روی \(x^2\) دیده باشید تا قاعده‌ها بی‌ریشه نمانند.
ضرب و خارج‌قسمت را جمله به جمله مشتق نکنید.
در پایان بگویید عدد خواسته شده یا تابع.
