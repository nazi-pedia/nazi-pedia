The derivative at $a$ is the limit of the slopes of the secant lines.

$$
\begin{aligned}
 f'(a) &= \lim_{h\to 0}\dfrac{f(a+h)-f(a)}{h}.
\end{aligned}
$$

At $h=0$ this quotient has the form $0/0$. Simplify first, then take the limit.
$f'(x)$ is a function. $f'(a)$ is a number: the slope of the tangent line.

## Rules that come from the definition

$$
\begin{aligned}
 (x^n)' &= nx^{n-1}, \\
 (uv)' &= u'v+uv', \\
 \left(\dfrac{u}{v}\right)' &= \dfrac{u'v-uv'}{v^2}.
\end{aligned}
$$

A sum is differentiated term by term. A constant in front of a function factors out of the derivative.


```python
import sympy as sp

# Compare a computed result with the hand result.
def check(found, hand, label="result"):
    # Simplify both sides, then their difference.
    found_s = sp.simplify(found)
    hand_s = sp.simplify(hand)
    difference = sp.simplify(found_s - hand_s)
    print(label)
    print("  hand solution:", hand_s)
    print("  Python:", found_s)
    print("  difference:", difference)
    # The assert checks that the simplified expressions agree.
    assert difference == 0
    print("  confirmed")
```

## Example 1 — The definition, all the way through

### Hand solution

Take $f(x)=x^2$ and $a=3$.

$$
\begin{aligned}
 \dfrac{(3+h)^2-9}{h}
 &= \dfrac{9+6h+h^2-9}{h} \\
 &= 6+h, \\
 f'(3)
 &= \lim_{h\to 0}(6+h) \\
 &= 6.
\end{aligned}
$$

The power rule also gives $2x$, and at $3$ that value is the same $6$.
The tangent line is $y-9=6(x-3)$, that is, $y=6x-9$.

### Python recheck


```python
import sympy as sp

# f(x) = x**2 at a = 3. Hand result: the quotient is 6 + h, and f'(3) = 6.
x, h = sp.symbols("x h")
f = x**2
# Difference quotient, simplified before the limit.
quotient = sp.simplify(((3 + h)**2 - 9)/h)
# The limit from the definition, and the power rule at x = 3.
from_definition = sp.limit(quotient, h, 0)
from_rule = sp.diff(f, x).subs(x, 3)
print("quotient:", quotient)
print("from the definition:", from_definition, " from the rule:", from_rule)
# The quotient equals 6 + h, and both methods equal 6.
assert quotient == 6 + h
assert from_definition == from_rule == 6
print("confirmed")
```

## Example 2 — The power rule on a polynomial

### Hand solution

$$
\begin{aligned}
 f(x) &= x^4-3x^2+7, \\
 f'(x) &= 4x^3-6x.
\end{aligned}
$$

The derivative of a constant is zero. The number $7$ is absent from $f'$, but it is still part of the value of $f$.
$f'(1)=4-6=-2$.

### Python recheck


```python
import sympy as sp

# f(x) = x**4 - 3*x**2 + 7. Hand result: 4*x**3 - 6*x, and f'(1) = -2.
x = sp.symbols("x")
f = x**4 - 3*x**2 + 7
derivative = sp.diff(f, x)
print(derivative, "at 1:", derivative.subs(x, 1))
# The derivative matches the hand formula, and the value at 1 is -2.
assert sp.simplify(derivative - (4*x**3 - 6*x)) == 0
assert derivative.subs(x, 1) == -2
print("confirmed")
```

## Example 3 — A product, not the product of the derivatives

### Hand solution

$$
\begin{aligned}
 u &= x^2+1, \quad v = x-4, \\
 (uv)'
 &= 2x(x-4)+(x^2+1)(1) \\
 &= 2x^2-8x+x^2+1 \\
 &= 3x^2-8x+1.
\end{aligned}
$$

If you compute only $u'v'$, you get $2x\cdot 1=2x$, which is wrong.
Expanding directly gives $(x^2+1)(x-4)=x^3-4x^2+x-4$, and its derivative $3x^2-8x+1$ is the same answer. That is a second check by hand.

### Python recheck


```python
import sympy as sp

# Product (x**2 + 1)*(x - 4). Hand result of the product rule: 3*x**2 - 8*x + 1.
x = sp.symbols("x")
product = (x**2 + 1)*(x - 4)
# Product rule: u'v + uv', with u' = 2x and v' = 1.
by_rule = 2*x*(x - 4) + (x**2 + 1)
by_diff = sp.diff(product, x)
print(sp.expand(by_rule))
# The product-rule expression and SymPy's derivative expand to the same polynomial.
assert sp.expand(by_rule - by_diff) == 0
assert sp.expand(by_diff) == 3*x**2 - 8*x + 1
print("confirmed")
```

## Example 4 — A quotient

### Hand solution

$$
\begin{aligned}
 f(x) &= \dfrac{x^2+1}{x-1}, \qquad x\neq 1, \\
 f'(x)
 &= \dfrac{2x(x-1)-(x^2+1)}{(x-1)^2} \\
 &= \dfrac{2x^2-2x-x^2-1}{(x-1)^2} \\
 &= \dfrac{x^2-2x-1}{(x-1)^2}.
\end{aligned}
$$

At $x=2$ the numerator is $4-4-1=-1$ and the denominator is $1$, so $f'(2)=-1$.
The function value is $f(2)=5$. The tangent slope is $-1$ at the point $(2,5)$:

$$
\begin{aligned}
 y-5 &= -(x-2), \\
 y &= -x+7.
\end{aligned}
$$

### Python recheck


```python
import sympy as sp

# f(x) = (x**2 + 1)/(x - 1). Hand result: (x**2 - 2*x - 1)/(x - 1)**2.
x = sp.symbols("x")
f = (x**2 + 1)/(x - 1)
derivative = sp.simplify(sp.diff(f, x))
hand = (x**2 - 2*x - 1)/(x - 1)**2
print(derivative)
# The derivative matches the hand formula; at x = 2 the slope is -1 and f(2) = 5.
assert sp.simplify(derivative - hand) == 0
assert derivative.subs(x, 2) == -1
assert f.subs(x, 2) == 5
print("confirmed")
```

## Example 5 — A slope is a number, not a function

### Hand solution

$$
\begin{aligned}
 f(x) &= x^3-x, \\
 f'(x) &= 3x^2-1.
\end{aligned}
$$

The question "the derivative at $x=1$" asks for a number: $f'(1)=2$.
The question "the derivative function" asks for the expression $3x^2-1$.
The tangent line at $x=1$ uses $f(1)=0$, so $y=2(x-1)$.

### Python recheck


```python
import sympy as sp

# f(x) = x**3 - x. Hand result: f'(x) = 3*x**2 - 1, so f'(1) = 2.
x = sp.symbols("x")
f = x**3 - x
fp = sp.diff(f, x)
print("f' =", fp)
print("f'(1) =", fp.subs(x, 1))
# Tangent line at x = 1: slope f'(1) through the point (1, f(1)).
tangent = fp.subs(x, 1)*(x - 1) + f.subs(x, 1)
print("tangent:", sp.expand(tangent))
# The slope is the number 2, and the tangent line is y = 2x - 2.
assert fp.subs(x, 1) == 2
assert sp.expand(tangent) == 2*x - 2
print("confirmed")
```

## A pitfall

In the quotient rule the subtraction is $u'v-uv'$, not the reverse.
One wrong sign moves every critical point in lesson 13.

## Summary

Work the definition all the way through on $x^2$ once, so the rules stay rooted in that limit.
Do not differentiate a product or a quotient term by term.
At the end, state whether the question asked for a number or for a function.
