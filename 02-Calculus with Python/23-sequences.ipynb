A sequence is an ordered list: $a_{1},a_{2},a_{3},\ldots$. The limit, when it exists, is the number the terms settle on.

$$
\lim_{n\to\infty}a_{n}=L.
$$

A monotone bounded sequence converges. This theorem gives existence of the limit, sometimes before a formula for the limit is guessed. After the guess, the fixed-point equation tests it.



## Example 1 — A rational expression

### Hand solution

$$
\begin{aligned}
a_{n} &= \dfrac{n+1}{n} \\
&= 1+\dfrac{1}{n} \\
&\to 1.
\end{aligned}
$$

### Python recheck



```python
import sympy as sp

# Limit of (n+1)/n as n tends to infinity.
n = sp.symbols("n", integer=True, positive=True)
assert sp.limit((n + 1)/n, n, sp.oo) == 1
print(sp.limit((n + 1)/n, n, sp.oo))
print("Confirmed")

```

## Example 2 — Damped oscillation

### Hand solution

$$
a_{n}=\dfrac{(-1)^{n}}{n}.
$$

The absolute value $\dfrac{1}{n}$ tends to $0$, so the sequence itself tends to $0$. The sign changes, but the amplitude of the oscillation dies out.

### Python recheck



```python
import sympy as sp

# The absolute value is 1/n, and that bound tends to 0.
n = sp.symbols("n", integer=True, positive=True)
term = ((-1)**n)/n
assert sp.simplify(sp.Abs(term) - 1/n) == 0
assert sp.limit(1/n, n, sp.oo) == 0
print("Absolute value:", sp.simplify(sp.Abs(term)), "Bound limit:", sp.limit(1/n, n, sp.oo))
print("Confirmed")

```

## Example 3 — Geometric

### Hand solution

$$
a_{n}=\left(\dfrac{1}{2}\right)^{n}\to 0.
$$

If $\lvert r\rvert<1$, then $r^{n}\to 0$. If $\lvert r\rvert>1$, the absolute value is unbounded. The case $r=1$ stays constant, and $r=-1$ jumps between $1$ and $-1$ and has no limit.

### Python recheck



```python
import sympy as sp

# Compare the geometric cases r = 1/2, r = 1, and r = 3/2.
n = sp.symbols("n", integer=True, positive=True)
assert sp.limit((sp.Rational(1, 2))**n, n, sp.oo) == 0
assert sp.limit(1**n, n, sp.oo) == 1
assert sp.limit((sp.Rational(3, 2))**n, n, sp.oo) == sp.oo
print("1/2 goes to 0, 1 stays at 1, 3/2 goes to infinity")
print("Confirmed")

```

## Example 4 — A recursive sequence and a fixed point

### Hand solution

$$
a_{1}=\sqrt{2},\qquad a_{n+1}=\sqrt{2+a_{n}}.
$$

If the limit $L$ exists, take the limit of both sides of the recurrence.

$$
\begin{aligned}
L &= \sqrt{2+L}, \\
L^{2} &= 2+L, \\
L^{2}-L-2 &= 0, \\
(L-2)(L+1) &= 0.
\end{aligned}
$$

The limit is nonnegative, so $L=2$. The terms begin at $\sqrt{2}\approx 1.414$. While they stay below $2$,

$$
\begin{aligned}
a_{n+1}^{2}-a_{n}^{2}
&= 2+a_{n}-a_{n}^{2} \\
&= (2-a_{n})(1+a_{n}) \\
&> 0,
\end{aligned}
$$

so the sequence increases and is bounded above by $2$. The theorem on monotone bounded sequences gives existence of the limit, and the equation gives its value.

### Python recheck



```python
import sympy as sp

# Solve the fixed-point equation for a nonnegative limit.
L = sp.symbols("L", nonnegative=True)
fixed = sp.solve(sp.Eq(L, sp.sqrt(2 + L)), L)

# Start at sqrt(2) and take eight steps of the recurrence.
term = sp.sqrt(2)
print("Fixed point:", fixed)
for k in range(8):
    term = sp.sqrt(2 + term)
    print(k + 2, sp.N(term, 8))

# The fixed point is 2, and the iterates are already close to it.
assert fixed == [2]
assert abs(term - 2) < 10**(-2)
print("Confirmed")

```

## A pitfall

The limit of the general term is not the limit of the series. The terms $\dfrac{1}{n}$ tend to $0$, but the sum of $\dfrac{1}{n}$ diverges. Lesson 24 computes that distinction.

## Summary

First look at $\lvert a_{n}\rvert$. For a recurrence, monotonicity and a bound give existence, and the fixed point gives the value.
