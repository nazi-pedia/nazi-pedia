The nonhomogeneous equation is $y''+py'+qy=g$.
Its general solution is a general solution of the homogeneous equation plus one particular solution.

Undetermined coefficients applies when $g$ is an exponential, a polynomial, a sine, or a cosine.
Guess a trial form copied from $g$. If that form already solves the homogeneous equation, multiply the guess by $x$. If the new guess is still a homogeneous solution, multiply by $x$ once more.

Variation of parameters covers a wider class of $g$. If $y_1$ and $y_2$ are a basis of the homogeneous equation and $W$ is their Wronskian, the coefficient functions satisfy

$$
u_1'=-\dfrac{y_2 g}{W},\qquad u_2'=\dfrac{y_1 g}{W},
$$

and the particular solution is $y_p=u_1 y_1+u_2 y_2$.


## Example 1 — An exponential that is not a root

### Hand solution

Solve $y''-3y'+2y=e^{3x}$.
The homogeneous solution is $C_1 e^{x}+C_2 e^{2x}$.
The forcing $e^{3x}$ is not a homogeneous solution, so the trial function is $Ae^{3x}$.

$$
\begin{aligned}
(9A-9A+2A)e^{3x} &= e^{3x}, \\
A &= \dfrac{1}{2}, \\
y &= C_1 e^{x}+C_2 e^{2x}+\dfrac{1}{2}e^{3x}.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Independent variable and unknown function.
x = sp.symbols("x")
y = sp.Function("y")
# Nonhomogeneous equation y'' - 3y' + 2y = exp(3x).
ode = sp.Eq(y(x).diff(x, 2) - 3*y(x).diff(x) + 2*y(x), sp.exp(3*x))
# General solution from the solver.
sol = sp.dsolve(ode, y(x))
# Particular solution (1/2)*exp(3x).
particular = sp.exp(3*x)/2
# Substitute the particular solution back into the left-hand side.
residual = sp.diff(particular, x, 2) - 3*sp.diff(particular, x) + 2*particular
print(sol)
# The left-hand side on the particular solution must recover the forcing.
assert sp.simplify(residual - sp.exp(3*x)) == 0
print("Confirmed")
```

## Example 2 — Resonance

### Hand solution

Solve $y''+y=\sin x$.
The homogeneous solutions are spanned by $\sin x$ and $\cos x$, so the trial function $\sin x$ is already a homogeneous solution.
Multiply by $x$ and guess $y_p=x(A\cos x+B\sin x)$.
The standard expansion gives $A=-\dfrac{1}{2}$ and $B=0$, hence

$$
y_p=-\dfrac{x}{2}\cos x.
$$

A short check of this formula: the second derivative plus the function itself must equal $\sin x$.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Independent variable.
x = sp.symbols("x")
# Resonant particular solution -(x/2)*cos(x).
yp = -x*sp.cos(x)/2
# Left-hand side y'' + y on this particular solution.
residual = sp.simplify(sp.diff(yp, x, 2) + yp)
print(residual)
# The residual must be the forcing sin(x).
assert residual == sp.sin(x)
print("Confirmed")
```

## Example 3 — Variation of parameters

### Hand solution

Solve $y''+y=\sec x$. A homogeneous basis is $\cos x$ and $\sin x$, and the Wronskian of that basis is $1$.

$$
\begin{aligned}
u_1' &= -\sin x\sec x = -\tan x, \\
u_2' &= \cos x\sec x = 1, \\
u_1 &= \ln|\cos x|, \\
u_2 &= x, \\
y_p &= \cos x\ln|\cos x|+x\sin x.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Independent variable.
x = sp.symbols("x")
# Particular solution from variation of parameters.
yp = sp.cos(x)*sp.log(sp.cos(x)) + x*sp.sin(x)
# Left-hand side y'' + y on this particular solution.
residual = sp.simplify(sp.diff(yp, x, 2) + yp)
print(residual)
# The residual must be sec(x) = 1/cos(x).
assert sp.simplify(residual - 1/sp.cos(x)) == 0
print("Confirmed")
```

## Example 4 — A polynomial

### Hand solution

Solve $y''-y=x$.
The homogeneous solutions are exponentials, so a linear polynomial is a safe guess: $y_p=Ax+B$.

$$
\begin{aligned}
-Ax-B &= x, \\
A &= -1, \\
B &= 0, \\
y &= C_1 e^{x}+C_2 e^{-x}-x.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Independent variable.
x = sp.symbols("x")
# Particular solution y_p = -x.
yp = -x
# The operator y'' - y applied to -x returns the forcing x.
assert sp.diff(yp, x, 2) - yp == x
print("The residual equals x")
print("Confirmed")
```

## A pitfall

Keep the particular solution separate from the homogeneous solution. The homogeneous constants are fixed by the initial conditions on the sum $y_h+y_p$, after $y_p$ is known.
First substitute $y_p$ into the differential equation. Then apply the initial conditions to the full solution.

## Summary

The form of $g$ tells you the form of the guess.
If that form already solves the homogeneous equation, the guess needs an extra factor of $x$.
Variation of parameters is the method that remains once a finite guess is no longer available.
