A rational function is the quotient of two polynomials.
Its graph either follows a line far away, or runs off beside a vertical line,
or has a hole that cannot be seen from a distance.
The limit in lesson 06 makes this picture precise. Here we build the picture itself.

## Definitions

If the degree of the denominator is greater, the horizontal asymptote is $y=0$.
If the degrees are equal, the horizontal asymptote is the ratio of the leading coefficients.
If the degree of the numerator is exactly one greater, the quotient from division is an oblique asymptote.
A common factor of the numerator and the denominator, once it cancels, is a hole and not an asymptote.
A factor that remains in the denominator is a vertical asymptote.


```python
import sympy as sp

# Compare a computed SymPy result with the answer found by hand.
def check(found, hand, label="Result"):
    # Simplify each side first so equivalent algebraic forms can match.
    found_s = sp.simplify(found)
    hand_s = sp.simplify(hand)
    difference = sp.simplify(found_s - hand_s)
    print(label)
    print("  Hand:", hand_s)
    print("  Python:", found_s)
    print("  Difference:", difference)
    # The methods agree only when that difference simplifies to zero.
    assert difference == 0
    print("  Confirmed")
```

## Example 1 — Horizontal and vertical asymptotes

### Hand solution

$$
f(x)=\dfrac{2x+1}{x-3}.
$$

The denominator is zero at $x=3$, and the numerator there is $7\neq 0$. The line $x=3$ is a vertical asymptote.

The numerator and the denominator both have degree $1$. The ratio of the leading coefficients is $2/1=2$,
so $y=2$ is a horizontal asymptote. Division says the same thing:

$$
f(x)=2+\dfrac{7}{x-3}.
$$

When $|x|$ grows large, the fraction grows small, and the graph settles against the line $y=2$.

### Python recheck


```python
import sympy as sp

# f(x) = (2x + 1)/(x - 3). Hand rewrite: 2 + 7/(x - 3).
# Horizontal asymptote y = 2; vertical asymptote x = 3, with +infinity from the right.
x = sp.symbols("x")
f = (2*x + 1)/(x - 3)
rewritten = sp.apart(f, x)
print("Decomposition:", rewritten)
print("Limit at +infinity:", sp.limit(f, x, sp.oo))
print("Limit at -infinity:", sp.limit(f, x, -sp.oo))
print("Right-hand limit at x=3:", sp.limit(f, x, 3, dir="+"))
assert rewritten == 2 + 7/(x - 3)
assert sp.limit(f, x, sp.oo) == 2
assert sp.limit(f, x, 3, dir="+") == sp.oo
print("Confirmed")
```

## Example 2 — A hole, not an asymptote

### Hand solution

$$
\begin{aligned}
g(x)
&= \dfrac{x^2-1}{x-1} \\
&= \dfrac{(x-1)(x+1)}{x-1}.
\end{aligned}
$$

For $x\neq 1$,

$$
g(x)=x+1.
$$

At $x=1$ the function is undefined, but the numerator and the denominator are both zero,
and after cancellation the value $2$ is what remains.
The graph is the line $y=x+1$ with the single point $(1,2)$ removed.
There is no vertical asymptote.

### Python recheck


```python
import sympy as sp

# g(x) = (x^2 - 1)/(x - 1). Hand simplification: x + 1 for x != 1, and the hole has height 2.
x = sp.symbols("x")
g = (x**2 - 1)/(x - 1)
print("Simplified:", sp.cancel(g))
print("Direct substitution:", g.subs(x, 1))
print("Limit:", sp.limit(g, x, 1))
# Cancellation removes the common factor and leaves x + 1.
assert sp.cancel(g) == x + 1
# The limit, which is the height of the hole, is 2.
assert sp.limit(g, x, 1) == 2
print("Confirmed")
```

## Example 3 — The numerator is not of higher degree, and the horizontal asymptote is not zero

### Hand solution

$$
h(x)=\dfrac{x^2}{x^2+1}.
$$

The denominator is always at least $1$, so there is no vertical asymptote.
The degrees are equal, and the ratio of the leading coefficients is $1$:

$$
h(x)=1-\dfrac{1}{x^2+1}.
$$

The horizontal asymptote is $y=1$. The graph never reaches that line, because a positive shrinking fraction has been subtracted from $1$.
There is no maximum. The function approaches $1$ from below. At $x=0$ the value is $0$.

### Python recheck


```python
import sympy as sp

# h(x) = x^2 / (x^2 + 1). Hand form: 1 - 1/(x^2 + 1), so the horizontal asymptote is y = 1.
x = sp.symbols("x", real=True)
h = x**2/(x**2 + 1)
print(sp.apart(h, x))
# Both ends approach 1; the value at 0 is 0; the gap below 1 is 1/(x^2 + 1).
assert sp.limit(h, x, sp.oo) == 1
assert h.subs(x, 0) == 0
assert sp.simplify(h - 1) == -1/(x**2 + 1)
print("Confirmed")
```

## Example 4 — An oblique asymptote

### Hand solution

$$
k(x)=\dfrac{x^2+1}{x}=x+\dfrac{1}{x},\qquad x\neq 0.
$$

The degree of the numerator is one greater. The quotient from division is the line $y=x$.
The term $\dfrac{1}{x}$ fades away as $|x|\to\infty$, so $y=x$ is an oblique asymptote.
The line $x=0$ is a vertical asymptote, because the denominator is zero and the numerator is $1$.

### Python recheck


```python
import sympy as sp

# k(x) = (x^2 + 1)/x. Hand form: x + 1/x, so y = x is the oblique asymptote.
x = sp.symbols("x")
k = (x**2 + 1)/x
print(sp.apart(k))
assert sp.apart(k) == x + 1/x
# The vertical gap between k(x) and the line y = x tends to 0 at infinity.
assert sp.limit(k - x, x, sp.oo) == 0
# From the right of the vertical asymptote x = 0, the limit is +infinity.
assert sp.limit(k, x, 0, dir="+") == sp.oo
print("Distance to the line y=x at infinity is zero")
print("Confirmed")
```

## Example 5 — Reading the graph from the factorization

### Hand solution

$$
\begin{aligned}
p(x)
&= \dfrac{x-2}{(x-2)(x+1)} \\
&= \dfrac{1}{x+1},\qquad x\neq 2.
\end{aligned}
$$

The factor $x-2$ cancels. There is a hole at $x=2$, and the height of the hole is

$$
\dfrac{1}{2+1}=\dfrac{1}{3}.
$$

The factor $x+1$ remains, so $x=-1$ is a vertical asymptote.
The degree of the denominator is greater, so $y=0$ is a horizontal asymptote.

The three phenomena of this lesson sit together in one fraction.

### Python recheck


```python
import sympy as sp

# p(x) = (x - 2)/((x - 2)(x + 1)).
# Hand reading: hole at x = 2 of height 1/3, vertical asymptote x = -1, horizontal asymptote y = 0.
x = sp.symbols("x")
p = (x - 2)/((x - 2)*(x + 1))
print("After cancellation:", sp.cancel(p))
print("Height of the hole:", sp.limit(p, x, 2))
print("Horizontal asymptote:", sp.limit(p, x, sp.oo))
print("Right-hand side of x=-1:", sp.limit(p, x, -1, dir="+"))
assert sp.cancel(p) == 1/(x + 1)
assert sp.limit(p, x, 2) == sp.Rational(1, 3)
assert sp.limit(p, x, sp.oo) == 0
print("Confirmed")
```

## A pitfall

Cancel a common zero of the numerator and the denominator before you declare a vertical asymptote.
If nothing of that factor remains after the cancellation, the point is a hole.
Do not guess a horizontal asymptote by substituting a large number. The ratio of the degrees settles the answer.

## Summary

Partial fractions, or polynomial division, separate the asymptote from the fractional part.
What tends to zero is the correction. What remains is the asymptote line.
