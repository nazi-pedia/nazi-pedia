A set stores each value once, with no position you can index. Membership is the question sets answer well: `value in basket`. The operators are the ones from the algebra of sets.

$$
\begin{aligned}
A \cup B &\quad \texttt{A | B} \\
A \cap B &\quad \texttt{A \& B} \\
A \setminus B &\quad \texttt{A - B}
\end{aligned}
$$

Printed order is not part of the meaning of a set. This path sorts a set before printing it, so the output is stable.


## Example 1 — Duplicates disappear

### Predict

`"red"` was written twice and stored once. `"green"` is present. There is no index 0.

### Run


```python
# A repeated item is stored once. A set has no index.
colors = {"red", "blue", "red"}
print(sorted(colors))
print("green" in colors)
print(len(colors))
try:
    print(colors[0])
except TypeError as error:
    print(type(error).__name__)
```

## Example 2 — Union, intersection, and difference

### Predict

Union holds everything from either set: 1, 2, 3, 4. Intersection holds only 3. Difference `a - b` holds the items of `a` that are not in `b`: 1 and 2.

### Run


```python
a = {1, 2, 3}
b = {3, 4}
# Union, intersection, then the items of a that are not in b.
print(sorted(a | b))
print(sorted(a & b))
print(sorted(a - b))
```

## Example 3 — Unique, and still in first-seen order

### Predict

Turning a list into a set drops duplicates and also drops order. A loop that records the first sighting keeps `"a"`, `"b"`, `"c"` in the order they first appeared.

### Run


```python
items = ["a", "b", "a", "c", "b"]
# set drops duplicates and order. The loop keeps the first sighting.
print(sorted(set(items)))
seen = []
for item in items:
    if item not in seen:
        seen.append(item)
print(seen)
```

## Example 4 — Add and discard

### Predict

`add` of an existing item changes nothing. `discard` of a missing item is quiet. `remove` of a missing item raises.

### Run


```python
basket = {1, 2}
basket.add(2)
basket.add(3)
# discard of a missing item is quiet. remove of a missing item raises.
basket.discard(9)
print(sorted(basket))
try:
    basket.remove(9)
except KeyError as error:
    print(type(error).__name__)
```

## A pitfall

Set items must be unchangeable. A list cannot be an item of a set, and a set cannot be an item of a set. A tuple of numbers can.

### Run


```python
# A tuple of numbers can sit in a set. A list cannot.
points = {(0, 0), (1, 1)}
print(sorted(points))
try:
    bad = {[1, 2]}
    print(bad)
except TypeError as error:
    print(type(error).__name__)
```

## Summary

A set keeps each value once and does not assign positions. Use it for membership and for union, intersection, and difference. Sort it when you print it if you need a stable order. Items must be immutable: tuples can be stored, lists cannot.
