A module is a file of names. `import math` loads that file and gives you a name, `math`, through which you reach its contents. `from math import floor` copies one name into your own file. The standard library is the set of modules that arrives with Python. You do not install it.

This notebook's `__name__` is `"__main__"`, because the notebook is the program being run, not a module that some other program imported.


## Example 1 — import the module, then use its names

### Predict

`math.sqrt(16)` is `4.0`. `math.pi` is the constant. `floor(3.7)` is `3`, and the name `floor` exists because it was imported directly.

### Run


```python
import math
from math import floor

print(math.sqrt(16))
print(math.pi > 3)
# floor was imported as its own name. math.floor would have worked too.
print(floor(3.7))
```

## Example 2 — A local name and a module name can both exist

### Predict

The name `pi` in this cell is the text `"not the constant"`. `math.pi` is the number. Importing the module did not overwrite the local name. `from math import *` would have done so, which is why this path does not use it.

### Run


```python
import math

# The local name pi is unrelated to math.pi.
pi = "not the constant"
print(pi)
print(round(math.pi, 5))
```

## Example 3 — Seed the random generator so a lesson can be repeated

### Predict

Without a seed, each run can differ. `seed(1)` fixes the sequence. The roll is an integer from 1 to 6, and the color is one of the three named colors. The exact roll belongs to this seed and this Python version; the guarantees are the range and the membership.

### Run


```python
import random

# The seed fixes the sequence so this cell repeats.
random.seed(1)
roll = random.randint(1, 6)
color = random.choice(["red", "blue", "green"])
print(roll)
print(color)
print(1 <= roll <= 6)
print(color in {"red", "blue", "green"})
```

## Example 4 — A date is an object with parts

### Predict

10 December 1815 is stored as year, month, and day. `isoformat` writes them as `1815-12-10`. Asking for `.year` does not depend on the language of the operating system.

### Run


```python
from datetime import date

# Year, month, and day are attributes. weekday uses Monday as 0.
birthday = date(1815, 12, 10)
print(birthday.isoformat())
print(birthday.year, birthday.month, birthday.day)
print(birthday.weekday())
```

## Example 5 — This file is `__main__`

### Predict

The notebook is being executed as a program, so `__name__` is `"__main__"`. A module that another file imported would see its own file name instead. That test is how a file can behave one way when run and another way when imported.

### Run


```python
# The notebook is the program being run, so the name is __main__.
print(__name__)

def report():
    return "program" if __name__ == "__main__" else "imported"

print(report())
```

## A pitfall

`import math` does not copy `sqrt` into your names. After that import, `sqrt(16)` is a `NameError`. The function lives at `math.sqrt`.

### Run


```python
import math

# sqrt lives on the module. It was not copied into this file's names.
print(math.sqrt(16))
try:
    print(sqrt(16))
except NameError as error:
    print(type(error).__name__)
```

## Summary

`import module` gives you a prefix. `from module import name` gives you the name itself. Prefer the prefix when a short name could collide with one of yours. The standard library already contains `math`, `random`, and `datetime`. Seed `random` when a lesson must repeat. `__name__` is `"__main__"` in the program you started.
