A tuple is an ordered sequence that cannot be edited in place. Parentheses usually surround it. The comma is what builds the tuple. Unpacking attaches several names to the items of a sequence in one statement.

$$
\texttt{name, age = pair}
$$


## Example 1 — A fixed pair

### Predict

`point` holds 3 then 4. Indexes work. Item assignment does not.

### Run


```python
point = (3, 4)
print(point[0], point[1])
print(len(point))
try:
    # A tuple does not allow item assignment.
    point[0] = 9
except TypeError as error:
    print(type(error).__name__)
```

## Example 2 — The comma makes the tuple

### Predict

`(3)` is only the number 3 in parentheses. `(3,)` is a one-item tuple. The trailing comma is the difference.

### Run


```python
# Parentheses alone do not build a tuple. The comma does.
not_a_tuple = (3)
one_item = (3,)
print(type(not_a_tuple).__name__, not_a_tuple)
print(type(one_item).__name__, one_item)
```

## Example 3 — Unpack into names

### Predict

The left side has two names and the right side has two items, so `name` receives `"Ada"` and `year` receives `1815`. Swapping works because the right-hand tuple is built before the names move.

### Run


```python
person = ("Ada", 1815)
# One name for each item. The right-hand pair is built before the swap.
name, year = person
print(name, year)
name, year = year, name
print(name, year)
```

## Example 4 — A tuple can be a dictionary key, a list cannot

### Predict

Dictionary keys must be unchangeable. A tuple of numbers qualifies. A list does not. The lookup with the same pair returns `"origin"`. Dictionaries are the subject of the next lesson; this example only shows why immutability matters.

### Run


```python
locations = {}
# A tuple can be a key because it cannot change. A list cannot.
locations[(0, 0)] = "origin"
print(locations[(0, 0)])
try:
    locations[[0, 0]] = "nope"
except TypeError as error:
    print(type(error).__name__)
```

## Example 5 — Extended unpacking keeps the middle

### Predict

`first` takes the first item, `last` takes the last, and `middle` takes everything between them as a list.

### Run


```python
numbers = (10, 20, 30, 40)
# The starred name collects every item that is not first or last.
first, *middle, last = numbers
print(first)
print(middle)
print(last)
```

## A pitfall

Unpacking requires the counts to match, unless a starred name absorbs the extras. Two names and three items is a `ValueError`.

### Run


```python
try:
    # Two names cannot receive three items.
    left, right = (1, 2, 3)
except ValueError as error:
    print(type(error).__name__)
    print(error)
```

## Summary

Tuples are ordered and immutable. The comma builds them; parentheses only group. Unpack when a sequence has a known shape, and use a starred name when one name should collect the rest. Immutability is what lets a tuple serve as a dictionary key.
