A function packages a calculation under a name. `def` creates it. The names in parentheses are parameters. The values in the call are arguments. `return` sends a value back to the caller and ends that call. `print` only writes to the output; it does not hand a value back.

The conversion from Celsius to Fahrenheit is the running example:

$$
F = \dfrac{9}{5} C + 32.
$$


## Example 1 — Return a computed number

### Predict

Water freezes at $0^\circ\mathrm{C}$, which is $32^\circ\mathrm{F}$. It boils at $100^\circ\mathrm{C}$, which is $212^\circ\mathrm{F}$.

$$
\dfrac{9}{5}\cdot 100 + 32 = 212.
$$

### Run


```python
def fahrenheit(celsius):
    # The formula is the whole body. The name celsius is the parameter.
    return (9 / 5) * celsius + 32

print(fahrenheit(0))
print(fahrenheit(100))
```

## Example 2 — print is not return

### Predict

`announced(3)` prints `3` and returns nothing useful. `handed_back(4)` returns `4`, which is stored and printed. The last line calls `announced(5)` while preparing its own print, so `5` appears first and the returned value `None` appears after it.

### Run


```python
def announced(value):
    # print writes the value and the function returns None.
    print(value)

def handed_back(value):
    return value

announced(3)
saved = handed_back(4)
print("saved", saved)
# The call runs before the outer print, so 5 appears first.
print("announced returned", announced(5))
```

## Example 3 — Defaults and keyword arguments

### Predict

Omitting `punctuation` uses `"!"`. Passing `"."` replaces the default. Naming the arguments at the call lets them arrive in either order.

### Run


```python
def greet(name, punctuation="!"):
    return "Hello, " + name + punctuation

print(greet("Ada"))
print(greet("Ada", "."))

def rectangle(width, height):
    return width * height

# The names at the call choose the parameter, not the position.
print(rectangle(height=4, width=3))
```

## Example 4 — Local names stay inside the function

### Predict

`scaled` can read `scale`, because that name lives outside. `factor` is created inside the call and disappears when the call ends. Reading it afterward raises `NameError`.

### Run


```python
scale = 2

def scaled(number):
    factor = 10
    # scale is read from outside. factor is local.
    return number * scale + factor

print(scaled(3))
try:
    print(factor)
except NameError as error:
    print(type(error).__name__)
```

## Example 5 — Extra positional arguments

### Predict

`*numbers` collects every positional argument into a tuple. The total of 1, 2, and 3 is 6. A call with no numbers totals 0.

### Run


```python
def total(*numbers):
    # numbers is a tuple of every positional argument.
    result = 0
    for number in numbers:
        result = result + number
    return result

print(total(1, 2, 3))
print(total())
```

## A pitfall

A default value is built once, when the function is defined, not on each call. A default list is therefore shared by every call that omits the argument. The first call appends `"a"`. The second call sees that same list and appends `"b"`.

The repair is a default of `None`, and a new list created inside the call.

### Run


```python
def bad(item, box=[]):
    # box was created once. Later calls reuse it.
    box.append(item)
    return box

print(bad("a"))
print(bad("b"))

def good(item, box=None):
    if box is None:
        box = []
    box.append(item)
    return box

print(good("a"))
print(good("b"))
```

## Summary

`def` names a calculation, parameters receive arguments, and `return` hands a value back. `print` does not. Defaults fill in omitted arguments, but a mutable default is shared, so use `None` and build the list inside the function. Names created inside a call are local. `*numbers` gathers extra positional arguments.
