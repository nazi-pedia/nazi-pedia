A list is an ordered, changeable sequence. Indexes and slices work as they do for strings. The difference is that a list can be edited in place: `append`, `insert`, `pop`, and item assignment change that same list.

Two names can label one list. A slice copy, `items[:]`, builds a second list.


## Example 1 — Build, index, and edit

### Predict

The list starts as `Ada`, `Grace`. `append` adds `Alan` at the end. Assigning index 0 replaces `Ada`. `pop` removes the last item and returns it. `remove` deletes the first matching value.

### Run


```python
names = ["Ada", "Grace"]
names.append("Alan")
print(names)
names[0] = "Augusta"
print(names)
# pop returns the removed item and shortens the list.
last = names.pop()
print(last, names)
names.remove("Grace")
print(names)
```

## Example 2 — A slice copy is a different list

### Predict

`copy = names` shares one list. `independent = names[:]` builds another. Appending through `copy` changes `names`. Appending through `independent` does not.

### Run


```python
names = ["Ada", "Grace"]
# copy shares the list. A slice builds a second list.
copy = names
independent = names[:]
copy.append("Alan")
independent.append("Linus")
print("names", names)
print("independent", independent)
print(names is copy)
print(names is independent)
```

## Example 3 — sort changes the list, sorted builds a new one

### Predict

`sorted` returns a new list and leaves `numbers` in its original order. `sort` reorders `numbers` itself. The `key` uses the absolute value, so `-4` sorts as if it were 4.

### Run


```python
numbers = [3, 1, 2]
print(sorted(numbers))
print(numbers)
numbers.sort()
print(numbers)
signed = [-4, -1, 3]
# key is applied before comparison. The list is ordered by abs(item).
print(sorted(signed, key=abs))
```

## Example 4 — The largest value, first by hand

### Predict

Walk the list once. Start with the first item as the champion. Replace it whenever a later item is larger. The champion ends as 9, which is also what `max` returns.

### Run


```python
numbers = [4, 9, 2, 7]
# Start with the first item, then replace it when a later item is larger.
champion = numbers[0]
for number in numbers[1:]:
    if number > champion:
        champion = number
print(champion)
print(max(numbers))
print(min(numbers))
print(sum(numbers))
```

## Example 5 — A list of lists is not a grid until each row is its own list

### Predict

Building a grid from one shared row means every "row" is the same list. Changing `grid[0][0]` changes every row. Building each row separately keeps the edits apart.

### Run


```python
row = [0, 0]
# Both entries of shared label the same row object.
shared = [row, row]
shared[0][0] = 1
print(shared)
# Each row here is a new list, so an edit stays in one row.
separate = [[0, 0], [0, 0]]
separate[0][0] = 1
print(separate)
```

## A pitfall

`insert` and `append` change the list and return `None`. Writing `names = names.append("Ada")` replaces the list with `None`. Call the method, then use the same name.

### Run


```python
names = ["Grace"]
# append changes the list and returns None. Do not assign that result.
returned = names.append("Ada")
print(returned)
print(names)
```

## Summary

Lists are ordered and mutable. Item assignment, `append`, `pop`, and `remove` edit the same list. Assignment of the name shares that list; a slice or `list(names)` copies the top level. Methods that edit a list return `None`, so do not assign their result back onto the name.
