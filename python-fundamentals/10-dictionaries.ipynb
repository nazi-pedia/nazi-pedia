A dictionary maps keys to values. Lookup is written `table[key]`. A missing key raises `KeyError`. `get` returns a default instead of raising. Since Python 3.7, a dictionary remembers the order in which keys were inserted.

Counting is the pattern worth memorizing. For each item, the new total is the old total plus one, and a missing key starts at zero:

$$
c(x) \leftarrow c(x) + 1.
$$


## Example 1 — Store, overwrite, and read

### Predict

Two keys are inserted, then `"Ada"` is overwritten with a new number. Reading `"Grace"` returns the original number. The printed order follows insertion.

### Run


```python
phones = {"Ada": "123", "Grace": "456"}
phones["Alan"] = "789"
# A second assignment to the same key replaces the value.
phones["Ada"] = "000"
print(phones["Grace"])
print(list(phones))
```

## Example 2 — get versus a bare lookup

### Predict

`"Linus"` is absent. `get` returns the default `"missing"`. The bracket lookup raises `KeyError`.

### Run


```python
phones = {"Ada": "123", "Grace": "456"}
# get supplies a default. Brackets raise when the key is absent.
print(phones.get("Linus", "missing"))
print("Linus" in phones)
try:
    print(phones["Linus"])
except KeyError as error:
    print(type(error).__name__)
    print(error)
```

## Example 3 — Loop over items, then delete a key

### Predict

`items` yields each key together with its value. `del` removes `"Grace"`. After that the key is gone and the other two remain.

### Run


```python
phones = {"Ada": "123", "Grace": "456", "Alan": "789"}
# items yields each key together with its value.
for name, number in phones.items():
    print(name, number)
del phones["Grace"]
print("Grace" in phones)
print(sorted(phones))
```

## Example 4 — Count characters

### Predict

`"abca"` contains `a` twice and `b` and `c` once. `get(character, 0)` supplies the zero for a first sighting.

### Run


```python
counts = {}
for character in "abca":
    # Missing keys start at 0, then the new sighting adds 1.
    counts[character] = counts.get(character, 0) + 1
print(counts)
print(counts["a"])
```

## Example 5 — A small lookup table for a calculation

### Predict

The table stores the number of days contributed by a completed month in a non-leap year, before the month itself. March is month 3, so the days before March are 59.

$$
31 + 28 = 59.
$$

### Run


```python
# Days elapsed before that month number, in a non-leap year.
before = {1: 0, 2: 31, 3: 59, 4: 90}
print(before[3])
print(before[3] == 31 + 28)
```

## A pitfall

A dictionary cannot contain two entries for one key. The second assignment silently replaces the first. If you needed both values, the value should have been a list.

### Run


```python
scores = {}
# The second write to the same key replaces the first value.
scores["Ada"] = 90
scores["Ada"] = 95
print(scores["Ada"])
# A list under the key can keep every attempt.
attempts = {}
attempts["Ada"] = [90]
attempts["Ada"].append(95)
print(attempts["Ada"])
```

## Summary

A dictionary stores values under keys and keeps insertion order. Use `get` when a missing key should produce a default, and use brackets when a missing key is a real mistake. Counting is `get` plus one. A second write to the same key replaces the value.
