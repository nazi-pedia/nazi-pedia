A class is a blueprint for objects. Each object built from it is an instance. `self` is that instance. `__init__` runs when the instance is created and is the place to attach the instance's own data. A method is a function whose first parameter is `self`.

Two instances of the same class do not share those data, unless you make the mistake in the pitfall and attach a changeable value to the class instead of to the instance.


## Example 1 — Two accounts, two balances

### Predict

Ada starts at 100 and deposits 25, so her balance becomes 125. Grace starts at 40 and is never changed. `__str__` supplies the text that `print` uses.

### Run


```python
class Account:
    def __init__(self, owner, balance):
        # These names live on this instance, not on the class.
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("deposit must be positive")
        self.balance = self.balance + amount

    def __str__(self):
        return f"{self.owner}: {self.balance}"

ada = Account("Ada", 100)
grace = Account("Grace", 40)
ada.deposit(25)
print(ada)
print(grace)
```

## Example 2 — A method can refuse

### Predict

A deposit of 0 breaks the rule, so `deposit` raises `ValueError` and the balance stays 10.

### Run


```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("deposit must be positive")
        self.balance = self.balance + amount

account = Account("Ada", 10)
try:
    # Zero is refused, so the balance does not change.
    account.deposit(0)
except ValueError as error:
    print(error)
print(account.balance)
```

## Example 3 — Reading an attribute is not calling a method

### Predict

`balance` is data. `deposit` is behavior. Printing the method without calling it shows that it is a bound method, not a number.

### Run


```python
class Account:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount

account = Account(10)
# balance is data. deposit, before the call, is a bound method.
print(account.balance)
print(type(account.deposit).__name__)
account.deposit(5)
print(account.balance)
```

## A pitfall

A list attached to the class is shared by every instance that does not have its own list. `red.add("Ada")` appends to that shared list, so `blue.members` sees Ada too.

The repair creates a new list inside `__init__`, on `self`.

### Run


```python
class SharedTeam:
    members = []

    def add(self, name):
        # self.members finds the class list, because the instance has no members of its own.
        self.members.append(name)

red = SharedTeam()
blue = SharedTeam()
red.add("Ada")
print(blue.members)

class Team:
    def __init__(self):
        self.members = []

    def add(self, name):
        self.members.append(name)

red = Team()
blue = Team()
red.add("Ada")
print(blue.members)
print(red.members)
```

## Summary

A class describes instances. `__init__` attaches each instance's data to `self`. Methods take `self` first and can read or change that data. A changeable value that belongs to one instance is created in `__init__`. A changeable value created on the class is shared.
