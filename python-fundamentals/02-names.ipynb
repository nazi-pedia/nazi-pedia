A name is a label attached to a value. The symbol `=` attaches the label. It does not mean "the two sides are equal." Equality is written `==`, and it arrives in lesson 04.

$$
\texttt{name = value}
$$

Reading a name gives you the value it currently labels. Writing the name again moves the label. It does not edit the old value.


## Example 1 — Attach a label, then read it

### Predict

`city` is attached to the text `London`. `print` reads that label.

### Run


```python
# Attach the label city to the text London.
city = "London"
print(city)
# type reports the kind of value the label currently points at.
print(type(city).__name__)
```

## Example 2 — Moving a label does not edit the other label

### Predict

`b = a` copies the label, not a private copy of a changing number. Then `a = 20` points `a` somewhere new. `b` still points at `10`.

### Run


```python
a = 10
# b and a both label 10. Nothing has been copied in a way that links their future.
b = a
print("before", a, b)
# Move a. b is not consulted, so b stays on 10.
a = 20
print("after", a, b)
```

## Example 3 — Two labels, one changeable object

### Predict

A list can be changed in place. `right = left` does not build a second list. Both names label the same list, so appending through `right` is visible through `left`. `is` asks whether two names label the same object.

Lists themselves are lesson 08. The only new fact here is shared identity.

### Run


```python
left = [1, 2]
# right is another name for that same list, not a second list.
right = left
right.append(3)
print(left)
print(left is right)
```

## Example 4 — `None` means no value yet

### Predict

`None` is a real object used as a placeholder. Compare it with `is`, not with a number.

### Run


```python
score = None
# is checks identity. None is a single shared placeholder.
print(score is None)
print(type(score).__name__)
```

## Example 5 — Unpack two names in one step

### Predict

The right-hand pair is built first, then the names on the left receive those values. That is why a swap can be one line: the old pair exists before either name is overwritten.

$$
(x, y) = (y, x)
$$

### Run


```python
x, y = 1, 2
print("before", x, y)
# The pair (y, x) is evaluated before x or y is rebound.
x, y = y, x
print("after", x, y)
```

## A pitfall

A keyword is already a Python word. It cannot be used as a name. `class` is a keyword, so the sentence below is a syntax error.

Legal names use letters, digits, and underscores, and they do not start with a digit. `score_1` is legal. `1score` is not.

### Run


```python
try:
    # class is reserved, so this sentence is not an assignment at all.
    compile("class = 1", "<example>", "exec")
except SyntaxError as error:
    print(type(error).__name__)
    print(error.msg)
# A readable name beats a short one. The value is unchanged by the spelling.
score_1 = 10
print(score_1)
```

## Summary

`=` attaches a name to a value. Rebinding one name leaves other names where they are, unless those names label one shared object that is changed in place. `None` is the placeholder for "no value," and keywords cannot be names.
