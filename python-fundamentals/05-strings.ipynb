A string is a sequence of characters. Positions start at $0$. The last position of a string of length $n$ is $n-1$, and the backward index $-1$ names that same last character.

A slice `s[i:j]` keeps positions $i, i+1, \ldots, j-1$. The stop position is not included.

$$
\texttt{s[i:j]} = s_i, s_{i+1}, \ldots, s_{j-1}.
$$

Strings cannot be edited in place. Any "change" builds a new string.


## Example 1 — Forward and backward indexes

### Predict

`"Python"` has length 6. Index `0` is `P`. Index `-1` is `n`. Index `100` does not exist.

### Run


```python
word = "Python"
# Index 0 is the first character. Index -1 is the last.
print(word[0])
print(word[-1])
print(len(word))
try:
    # One index past the end is an error. A slice past the end is not.
    print(word[100])
except IndexError as error:
    print(type(error).__name__)
```

## Example 2 — Slices are half-open, and they do not raise

### Predict

`word[0:4]` is `Pyth`: positions 0, 1, 2, and 3. A missing start means "from the beginning." A missing stop means "through the end." A slice that runs past the end simply stops. It does not raise `IndexError`.

### Run


```python
word = "Python"
# Stop is excluded: positions 0, 1, 2, 3.
print(word[0:4])
print(word[:3])
print(word[2:])
# Past the end is not an error for a slice. The slice just ends.
print(word[0:100])
# A third value is the step. 2 keeps every other character.
print(word[::2])
```

## Example 3 — Building a new string, because the old one cannot change

### Predict

Assigning to `word[0]` fails. Concatenation builds a different string, `"Jython"`, and the original name still labels `"Python"` until we point it at the new one.

### Run


```python
word = "Python"
try:
    # Strings do not allow item assignment.
    word[0] = "J"
except TypeError as error:
    print(type(error).__name__)
# A new string can reuse the tail. The old string is untouched.
renamed = "J" + word[1:]
print(renamed)
print(word)
```

## Example 4 — Membership and repetition

### Predict

`in` looks for a substring, not just a single character. `*` repeats text. It does not multiply a number hidden inside the text.

### Run


```python
word = "Python"
print("th" in word)
print("py" in word)
# Case matters. "py" is not "Py".
print("Py" in word)
print("ha" * 3)
```

## Example 5 — Quotes and escape characters

### Predict

A backslash gives a character a second meaning. `\n` is a newline. `\"` is a quote inside a double-quoted string. `\\` is one backslash.

### Run


```python
print("She said \"hello\"")
# \n is one character, the newline, not the two letters n and backslash.
print("first\nsecond")
print("C:\\notes")
```

## A pitfall

Indexes and slices look similar and fail differently. `word[6]` is past the end of `"Python"` and raises. `word[6:8]` is an empty string. When you write a boundary, decide whether you are indexing one character or slicing a range.

### Run


```python
word = "Python"
# A slice past the end is empty. The same start as an index raises.
print(repr(word[6:8]))
try:
    print(word[6])
except IndexError as error:
    print(type(error).__name__)
```

## Summary

Indexes start at 0, and `-1` is the last character. A slice includes the start and excludes the stop, and a slice past the end is quiet. Strings do not change in place: build a new string and, if you want, attach the old name to it.
