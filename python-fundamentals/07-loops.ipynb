A loop repeats a block. `while` repeats as long as a test stays true. `for` repeats once for each item in a sequence. `range(n)` produces the integers $0, 1, \ldots, n-1$.

`break` leaves the loop immediately. `continue` skips the rest of the current pass and starts the next one. A `for` or `while` may have an `else` block. That block runs only when the loop finishes without `break`.


## Example 1 — A while loop needs a change that will end it

### Predict

`n` starts at 3. Each pass prints it, then subtracts 1. When `n` becomes 0 the test fails. The printed lines are 3, 2, 1.

### Run


```python
n = 3
while n > 0:
    print(n)
    # Without this update the test would stay true forever.
    n = n - 1
print("done", n)
```

## Example 2 — for walks a sequence, range builds integers

### Predict

The first loop prints each character. `range(1, 5)` is $1, 2, 3, 4$: the stop value is excluded, just as in a slice. `range(0, 10, 2)` steps by 2.

### Run


```python
for character in "Ada":
    print(character)
print("---")
# Start is included. Stop is excluded.
for number in range(1, 5):
    print(number)
print("---")
for number in range(0, 10, 2):
    print(number)
```

## Example 3 — An accumulator, checked against a formula

### Predict

Adding $1$ through $5$ by hand gives $15$. The closed form says the same thing:

$$
\sum_{k=1}^{n} k = \dfrac{n(n+1)}{2},
\qquad
\dfrac{5 \cdot 6}{2} = 15.
$$

### Run


```python
total = 0
for number in range(1, 6):
    # Fold the next integer into the running total.
    total = total + number
print(total)
n = 5
print(n * (n + 1) // 2)
```

## Example 4 — break, continue, and the else that means "not found"

### Predict

`continue` skips the print for negative numbers, so `-1` is silent. `break` stops at `0` during the search, so the `else` does not run. A second search never sees `0`, never breaks, and the `else` prints `absent`.

### Run


```python
for number in (3, -1, 4, -5):
    if number < 0:
        # Skip the rest of this pass. The loop itself continues.
        continue
    print(number)
print("---")
for number in (4, 0, 9):
    if number == 0:
        print("found")
        break
else:
    # This runs only when break did not happen.
    print("absent")
print("---")
for number in (4, 5, 9):
    if number == 0:
        print("found")
        break
else:
    print("absent")
```

## Example 5 — A nested loop is a loop inside a pass

### Predict

The outer loop takes a row. For each row, the inner loop takes every column. Two rows and two columns print four pairs.

### Run


```python
# The inner loop finishes both columns before the outer loop takes the next row.
for row in range(2):
    for column in range(2):
        print(row, column)
```

## A pitfall

A `while` whose test never becomes false does not end. The usual cause is forgetting to update the name the test depends on. This cell shows the repaired loop, with the update in plain sight. Do not delete that line to "see what happens": the notebook would not finish.

### Run


```python
n = 3
passes = 0
while n > 0:
    passes = passes + 1
    n = n - 1
# Three updates were required to make the test false.
print(passes, n)
```

## Summary

`while` repeats on a test, and something inside must eventually make that test false. `for` repeats over a sequence, and `range` builds integer sequences with an excluded stop. `continue` skips one pass, `break` leaves the loop, and `else` on a loop runs only when no `break` fired.
