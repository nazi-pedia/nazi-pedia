An exception is how a function reports that it cannot do what was asked. Uncaught, it stops the program and prints a traceback. `try` attempts a block. `except` handles one kind of failure. `else` runs when the `try` block did not raise. `finally` runs in either case, which makes it the right place for cleanup.

Raise your own exception when an argument violates a rule of your function. `ValueError` means "this value is the right kind and the wrong content."


## Example 1 — The order of else and finally

### Predict

A successful division prints `division worked`, then `cleanup runs either way`, then the value. A zero divisor prints `cannot divide by zero`, then the cleanup line, then `None`. `else` does not run after the exception.

### Run


```python
def divide(numerator, denominator):
    try:
        result = numerator / denominator
    except ZeroDivisionError:
        print("cannot divide by zero")
        return None
    else:
        # else runs only when the try block raised nothing.
        print("division worked")
        return result
    finally:
        # finally runs on the way out, including on return.
        print("cleanup runs either way")

print("value", divide(10, 2))
print("---")
print("value", divide(10, 0))
```

## Example 2 — Reject a value that breaks your rule

### Predict

Age 36 becomes 37. Age `-1` raises `ValueError` before any arithmetic. The message is the one written in the `raise`.

### Run


```python
def age_next_year(age):
    # A negative age breaks the rule of this function.
    if age < 0:
        raise ValueError("age cannot be negative")
    return age + 1

print(age_next_year(36))
try:
    print(age_next_year(-1))
except ValueError as error:
    print(error)
```

## Example 3 — Catch the error you expect, and let the others through

### Predict

`"12"` becomes the integer 12. `"nope"` is a `ValueError` from `int`, and the function returns `None`. A list is a `TypeError`. This function does not catch that, so the caller sees it. Swallowing every error would have hidden a real bug.

### Run


```python
def careful(text):
    try:
        return int(text)
    except ValueError:
        # Only a bad numeral is turned into None. Other errors stay visible.
        return None

print(careful("12"))
print(careful("nope"))
try:
    print(careful([]))
except TypeError as error:
    print(type(error).__name__)
```

## Example 4 — Look before you leap, or ask forgiveness

### Predict

Both styles avoid dividing by zero. The first checks the denominator. The second attempts the division and handles the failure. For a dictionary lookup, asking forgiveness with `try` is ordinary Python. For a rule you control, a clear `if` is often easier to read. Use the version a reader can check.

### Run


```python
def by_check(numerator, denominator):
    # Look before the division.
    if denominator == 0:
        return None
    return numerator / denominator

def by_attempt(numerator, denominator):
    # Attempt the division, and handle the one expected failure.
    try:
        return numerator / denominator
    except ZeroDivisionError:
        return None

print(by_check(10, 0), by_attempt(10, 0))
print(by_check(10, 2), by_attempt(10, 2))
```

## A pitfall

`finally` runs even when you return inside `try`. It does not replace that return value unless `finally` itself returns or raises. Here the cleanup print happens, and the caller still receives 5.

### Run


```python
def give_back():
    try:
        return 5
    finally:
        # Cleanup runs on the way out. The return value is still 5.
        print("cleanup")

print(give_back())
```

## Summary

Catch the exception you expect. Leave the others alone so real bugs stay visible. `else` means the attempt succeeded. `finally` means cleanup, and it runs on the way out. `raise ValueError` when a value breaks a rule you chose to enforce.
