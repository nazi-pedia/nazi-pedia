## 1. The cubic, done as a procedure you can reuse

**See.** \(f(x) = x^3 - 3x^2 + 4\). A critical point in this lesson means a point where \(f'(x) = 0\).
(Endpoints of a closed interval are a separate story and are not in this example.)

**Name.** The procedure:

1. Differentiate.
2. Solve \(f'(x) = 0\).
3. Compute \(f''\).
4. If \(f''(c) > 0\), the graph is concave up: a local minimum.
5. If \(f''(c) < 0\), concave down: a local maximum.
6. If \(f''(c) = 0\), the test says nothing. Use the first-derivative test.

**Do.** \(f'(x) = 3x^2 - 6x = 3x(x-2)\). The critical points are \(x = 0\) and \(x = 2\).
\(f''(x) = 6x - 6\). So \(f''(0) = -6\) and \(f''(2) = 6\).


```python
import sympy as sp

x = sp.symbols("x")
f = x ** 3 - 3 * x ** 2 + 4
f_prime = sp.factor(sp.diff(f, x))
f_second = sp.diff(f, x, 2)
critical_points = sp.solve(sp.diff(f, x), x)

print("f'(x) =", f_prime)
print("critical points:", critical_points)
print("f''(x) =", f_second)

assert critical_points == [0, 2]
assert sp.simplify(f_second - (6 * x - 6)) == 0

classification = {}
for point in critical_points:
    second_value = f_second.subs(x, point)
    height = f.subs(x, point)
    if second_value > 0:
        kind = "local minimum"
    elif second_value < 0:
        kind = "local maximum"
    else:
        kind = "inconclusive"
    classification[point] = kind
    print(f"x = {point}: f = {height}, f'' = {second_value}, {kind}")

assert classification[0] == "local maximum"
assert classification[2] == "local minimum"
assert f.subs(x, 0) == 4
assert f.subs(x, 2) == 0
```

**Read the output.** At \(x = 0\) the height is 4 and the curve is concave down: a local maximum.
At \(x = 2\) the height is 0 and the curve is concave up: a local minimum.
These are local. The cubic still goes to \(+\infty\) as \(x\) grows and to \(-\infty\) as \(x\) decreases,
so neither point is a global maximum or minimum on the whole real line.

**Check a second way.** The first-derivative test. \(f'(x) = 3x(x-2)\).

- For \(x\) slightly left of 0, \(f'\) is positive (the graph is rising).
- For \(x\) between 0 and 2, \(f'\) is negative (the graph is falling).
- For \(x\) slightly right of 2, \(f'\) is positive again.

Rising then falling is a local maximum. Falling then rising is a local minimum. Same conclusion, no second derivative required.


```python
import sympy as sp

x = sp.symbols("x")
f_prime = sp.diff(x ** 3 - 3 * x ** 2 + 4, x)

def sign_at(value):
    number = sp.sign(f_prime.subs(x, value))
    return int(number)

assert sign_at(-0.5) == 1
assert sign_at(1) == -1
assert sign_at(2.5) == 1
print("sign of f' on (-inf, 0), (0, 2), (2, inf): +  -  +")
```

## 2. Put the classification on the graph

**Do.** Plot \(f\) on \([-1.5, 3.5]\). Mark the local maximum and the local minimum with different markers.
The title states the classification. A graph without that sentence is unfinished.


```python
import matplotlib.pyplot as plt
import numpy as np

def f_numeric(t):
    return t ** 3 - 3 * t ** 2 + 4

xs = np.linspace(-1.5, 3.5, 400)
fig, ax = plt.subplots(figsize=(7, 4))
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.plot(xs, f_numeric(xs), color="#1f4e79", linewidth=2, label=r"$f(x)$")
ax.scatter([0], [4], color="#b85c38", s=40, zorder=3, label="local max (0, 4)")
ax.scatter([2], [0], color="#2e7d32", s=40, zorder=3, label="local min (2, 0)")
ax.set_title(r"$f(x)=x^3-3x^2+4$: local max at 0, local min at 2")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

Lesson 005 asked you to look at this graph and not yet name the turns.
You may name them now, because both derivative tests agree.


## 3. When the second-derivative test is silent

**See.** \(g(x) = x^3\). Then \(g'(x) = 3x^2\) and \(g''(x) = 6x\).
The only critical point is \(x = 0\), and \(g''(0) = 0\). The second-derivative test is inconclusive.

**Do.** The first-derivative test: \(g'(x) = 3x^2 \ge 0\) on both sides of 0.
The graph does not rise-then-fall, and it does not fall-then-rise. It keeps rising, with a flat instant at the origin.
That is a horizontal inflection, not a local extremum.

**Misconception.** \(f''(c) = 0\) means there is an inflection point, or means there is no extremum.

**Correction.** It means this particular test refuses to answer. \(x^4\) has \(f''(0) = 0\) and a local minimum.
Always look at the sign of \(f'\) when \(f''(c) = 0\).


```python
import sympy as sp

x = sp.symbols("x")

def classify_critical_point(f, point):
    second = sp.diff(f, x, 2).subs(x, point)
    if second > 0:
        return "local minimum"
    if second < 0:
        return "local maximum"
    prime = sp.diff(f, x)
    # Probe the sign of f' on each side. A zero derivative on an interval would need a better probe.
    left = sp.sign(prime.subs(x, point - sp.Rational(1, 10)))
    right = sp.sign(prime.subs(x, point + sp.Rational(1, 10)))
    if left > 0 and right < 0:
        return "local maximum"
    if left < 0 and right > 0:
        return "local minimum"
    return "not a local extremum"

cubic = x ** 3
assert sp.diff(cubic, x, 2).subs(x, 0) == 0
assert classify_critical_point(cubic, 0) == "not a local extremum"

flat_minimum = x ** 4
assert sp.diff(flat_minimum, x, 2).subs(x, 0) == 0
assert classify_critical_point(flat_minimum, 0) == "local minimum"
print("x**3 is not a local extremum; x**4 is a local minimum. f''(0) = 0 in both cases.")
```

## Workshop — a quadratic you can also do by completing the square

\(h(x) = x^2 - 4x + 1 = (x-2)^2 - 3\).
The vertex is at \(x = 2\), and the parabola opens upward, so that vertex is a minimum of height \(-3\).

Find the critical point with SymPy, classify it, and check the height against \(-3\).


```python
import sympy as sp

x = sp.symbols("x")
h = x ** 2 - 4 * x + 1
critical = sp.solve(sp.diff(h, x), x)
print("critical points:", critical)
assert critical == [2]
assert sp.diff(h, x, 2) == 2
assert h.subs(x, 2) == -3
print("minimum value -3 at x = 2, confirmed by the completed square and by f'' > 0")
```

## Mistakes this lesson is meant to prevent

- Solving \(f(x) = 0\) when the question asked for critical points. Roots and critical points are different equations.
- Declaring a global maximum because you found a local one. Cubics have neither, on the whole line.
- Treating \(f''(c) = 0\) as a finished answer.
- Reading the turn off the graph of lesson 005 and skipping the derivative. The picture is a hint. The test is the proof.
