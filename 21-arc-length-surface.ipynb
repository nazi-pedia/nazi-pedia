# ۲۱ · طول قوس و سطح دوار

| | |
|---|---|
| **دوره** | ریاضی عمومی ۲ |
| **پیش‌نیاز** | ۰۸ و ۱۶ |

طول نمودار \(y=f(x)\) از \(a\) تا \(b\):

\[
L=\int_a^b\sqrt{1+\bigl(f'(x)\bigr)^2}\,dx
\]

اگر همین قوس حول محور \(x\) بچرخد، مساحت سطح

\[
S=2\pi\int_a^b y\sqrt{1+(y')^2}\,dx
\]

است. عامل \(\sqrt{1+(y')^2}\,dx\) همان تکهٔ طول است و \(2\pi y\) محیط دایره‌ای است که آن تکه می‌پیماید.



## مثال ۱ — پاره‌خط مورب

### حل ریاضی

\(y=x\) از ۰ تا ۱. شیب ۱ است.

\[
L=\int_0^1\sqrt{1+1}\,dx=\sqrt{2}
\]

فاصلهٔ \((0,0)\) تا \((1,1)\) همین است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
length = sp.integrate(sp.sqrt(1 + sp.diff(x, x)**2), (x, 0, 1))
print(length)
assert length == sp.sqrt(2)
print("تایید شد")

```

## مثال ۲ — قوسی که رادیکال خوش‌رفتار دارد

### حل ریاضی

\(y=x^{3/2}\) از ۰ تا ۱.

\[
y'=\frac{3}{2}x^{1/2},\qquad 1+(y')^2=1+\frac{9}{4}x
\]

\[
L=\int_0^1\sqrt{1+\frac{9}{4}x}\,dx
\]

\(u=1+\dfrac{9}{4}x\)، \(du=\dfrac{9}{4}dx\). کران از ۱ تا \(\dfrac{13}{4}\).

\[
L=\frac{4}{9}\cdot\frac{2}{3}\Bigl[u^{3/2}\Bigr]_1^{13/4}
=\frac{8}{27}\left(\left(\frac{13}{4}\right)^{3/2}-1\right)
=\frac{13\sqrt{13}-8}{27}
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", nonnegative=True)
y = x**sp.Rational(3, 2)
length = sp.integrate(sp.sqrt(1 + sp.diff(y, x)**2), (x, 0, 1))
hand = (13*sp.sqrt(13) - 8)/27
print(sp.simplify(length))
assert sp.simplify(length - hand) == 0
print("تایید شد")

```

## مثال ۳ — سطح جانبی مخروط

### حل ریاضی

پاره‌خط \(y=x\) از ۰ تا ۱ حول محور \(x\)، مخروطی به شعاع ۱ و مولد \(\sqrt{2}\) می‌سازد.
فرمول کلاسیک سطح جانبی \(\pi r \ell=\pi\sqrt{2}\) است.

\[
S=2\pi\int_0^1 x\sqrt{2}\,dx=2\pi\sqrt{2}\cdot\frac{1}{2}=\pi\sqrt{2}
\]

دو راه یکی هستند.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
y = x
surface = 2*sp.pi*sp.integrate(y*sp.sqrt(1 + sp.diff(y, x)**2), (x, 0, 1))
print(sp.simplify(surface))
assert sp.simplify(surface - sp.pi*sp.sqrt(2)) == 0
print("تایید شد")

```

## مثال ۴ — نیمدایره

### حل ریاضی

\(y=\sqrt{1-x^2}\) از \(-1\) تا ۱ نیم‌دایرهٔ واحد است و طولش باید \(\pi\) باشد.
مشتق در دو سر نامتناهی می‌شود، پس انتگرال ناسره است، ولی مقدارش تمام است.

\[
y'=\frac{-x}{\sqrt{1-x^2}},\qquad 1+(y')^2=\frac{1}{1-x^2}
\]

\[
L=\int_{-1}^{1}\frac{1}{\sqrt{1-x^2}}\,dx=\arcsin x\Big|_{-1}^{1}=\pi
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
y = sp.sqrt(1 - x**2)
raw = sp.sqrt(1 + sp.diff(y, x)**2)
hand_integrand = 1/sp.sqrt(1 - x**2)
assert sp.simplify(raw**2 - hand_integrand**2) == 0
length = sp.integrate(hand_integrand, (x, -1, 1))
numeric = sp.N(sp.Integral(raw, (x, -1, 1)))
print(sp.simplify(raw), "->", length, "عددی:", numeric)
assert sp.simplify(length - sp.pi) == 0
assert abs(numeric - sp.pi) < 10**(-10)
print("تایید شد")

```

## نکته

اگر \(1+(y')^2\) مربع کامل نشود، انتگرال طول اغلب مقدماتی نیست.
آن وقت تقریب عددی جواب است، نه شکست مسئله. اول ببینید رادیکال ساده می‌شود یا نه.
