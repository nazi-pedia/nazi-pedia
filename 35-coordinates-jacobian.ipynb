In the plane, polar coordinates use $dA=r\,dr\,d\theta$. The factor $r$ is the Jacobian of the change of variables.

Cylindrical coordinates are polar coordinates together with $z$, and

$$
\begin{aligned}
 dV &= r\,dr\,d\theta\,dz.
\end{aligned}
$$

Spherical coordinates are

$$
\begin{aligned}
 x &= \rho\sin\varphi\cos\theta,\\
 y &= \rho\sin\varphi\sin\theta,\\
 z &= \rho\cos\varphi,\\
 dV &= \rho^{2}\sin\varphi\,d\rho\,d\varphi\,d\theta.
\end{aligned}
$$

The angle $\varphi$ runs from $0$ to $\pi$, not to $2\pi$.


## Example 1 — The polar Jacobian

### Hand solution

Take $x=r\cos\theta$ and $y=r\sin\theta$.

$$
\begin{aligned}
 \det
 \begin{pmatrix}
 \cos\theta & -r\sin\theta \\
 \sin\theta & r\cos\theta
 \end{pmatrix}
 &=
 r\cos^{2}\theta+r\sin^{2}\theta
 =
 r.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Polar change of variables.
r, theta = sp.symbols("r theta", real=True)
x = r*sp.cos(theta)
y = r*sp.sin(theta)
# Determinant of the Jacobian matrix.
J = sp.Matrix([[sp.diff(x, r), sp.diff(x, theta)], [sp.diff(y, r), sp.diff(y, theta)]]).det()
print(sp.simplify(J))
assert sp.simplify(J - r) == 0
print("Confirmed")
```

## Example 2 — A disk in polar coordinates

### Hand solution

$$
\begin{aligned}
 \iint_{x^{2}+y^{2}\le 1}(x^{2}+y^{2})\,dA
 &=
 \int_{0}^{2\pi}\int_{0}^{1} r^{2}\cdot r\,dr\,d\theta
 =
 \int_{0}^{2\pi}\dfrac{1}{4}\,d\theta
 =
 \dfrac{\pi}{2}.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Integrand r^2 times the polar factor r, over the unit disk.
r, theta = sp.symbols("r theta")
exact = sp.integrate(r**2 * r, (r, 0, 1), (theta, 0, 2*sp.pi))
print(exact)
assert exact == sp.pi/2
print("Confirmed")
```

## Example 3 — A cylinder

### Hand solution

The solid is the cylinder $r\le 2$ and $0\le z\le 3$.

$$
\begin{aligned}
 V
 &=
 \int_{0}^{2\pi}\int_{0}^{2}\int_{0}^{3} r\,dz\,dr\,d\theta
 =
 3\cdot 2\pi\cdot\dfrac{4}{2}
 =
 12\pi.
\end{aligned}
$$

The geometric volume $\pi r^{2} h=12\pi$ is the same number.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Cylindrical volume element r, with r <= 2 and 0 <= z <= 3.
r, theta, z = sp.symbols("r theta z")
volume = sp.integrate(r, (z, 0, 3), (r, 0, 2), (theta, 0, 2*sp.pi))
print(volume)
assert volume == 12*sp.pi
print("Confirmed")
```

## Example 4 — The unit sphere

### Hand solution

$$
\begin{aligned}
 V
 &=
 \int_{0}^{2\pi}\int_{0}^{\pi}\int_{0}^{1} \rho^{2}\sin\varphi\,d\rho\,d\varphi\,d\theta
 =
 \dfrac{1}{3}\cdot 2\pi\cdot\int_{0}^{\pi}\sin\varphi\,d\varphi
 =
 \dfrac{2\pi}{3}\cdot 2
 =
 \dfrac{4\pi}{3}.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Volume of the unit ball in spherical coordinates.
rho, phi, theta = sp.symbols("rho phi theta", real=True)
volume = sp.integrate(rho**2 * sp.sin(phi), (rho, 0, 1), (phi, 0, sp.pi), (theta, 0, 2*sp.pi))
print(volume)
assert sp.simplify(volume - 4*sp.pi/3) == 0
print("Confirmed")
```

## A pitfall

Forgetting the factor $r$, or the factor $\rho^{2}\sin\varphi$, replaces area and volume by a Cartesian integral that does not match the region.
Compute the Jacobian once, as a determinant, so the factor is kept because it was derived, not because it was memorized without a reason.

## Summary

Circular symmetry calls for polar or cylindrical coordinates.
Spherical symmetry calls for the angle $\varphi$ and the factor $\rho^{2}\sin\varphi$.
