# ۳۵ · استوانه‌ای، کروی و ژاکوبی

| | |
|---|---|
| **دوره** | ریاضی عمومی چندمتغیره |
| **پیش‌نیاز** | ۲۷ و ۳۴ |

قطبی در صفحه: \(dA=r\,dr\,d\theta\). عامل \(r\) ژاکوبی تبدیل است.

استوانه‌ای: همان قطبی به‌علاوه \(z\). \(dV=r\,dr\,d\theta\,dz\).

کروی:

\[
x=\rho\sin\varphi\cos\theta,\quad
y=\rho\sin\varphi\sin\theta,\quad
z=\rho\cos\varphi
\]

\[
dV=\rho^2\sin\varphi\,d\rho\,d\varphi\,d\theta
\]

\(\varphi\) از ۰ تا \(\pi\) است، نه تا \(2\pi\).



## مثال ۱ — ژاکوبی قطبی

### حل ریاضی

\(x=r\cos\theta\)، \(y=r\sin\theta\).

\[
\det\begin{pmatrix}\cos\theta&-r\sin\theta\\\sin\theta&r\cos\theta\end{pmatrix}
=r\cos^2\theta+r\sin^2\theta=r
\]

### بازبینی با پایتون



```python
import sympy as sp
r, theta = sp.symbols("r theta", real=True)
x = r*sp.cos(theta)
y = r*sp.sin(theta)
J = sp.Matrix([[sp.diff(x, r), sp.diff(x, theta)], [sp.diff(y, r), sp.diff(y, theta)]]).det()
print(sp.simplify(J))
assert sp.simplify(J - r) == 0
print("تایید شد")

```

## مثال ۲ — قرص با قطبی

### حل ریاضی

\[
\iint_{x^2+y^2\le 1}(x^2+y^2)\,dA
=\int_0^{2\pi}\int_0^1 r^2\cdot r\,dr\,d\theta
=\int_0^{2\pi}\frac{1}{4}\,d\theta=\frac{\pi}{2}
\]

### بازبینی با پایتون



```python
import sympy as sp
r, theta = sp.symbols("r theta")
exact = sp.integrate(r**2 * r, (r, 0, 1), (theta, 0, 2*sp.pi))
print(exact)
assert exact == sp.pi/2
print("تایید شد")

```

## مثال ۳ — استوانه

### حل ریاضی

استوانهٔ \(r\le 2\) و \(0\le z\le 3\).

\[
V=\int_0^{2\pi}\int_0^2\int_0^3 r\,dz\,dr\,d\theta
=3\cdot 2\pi\cdot \frac{4}{2}=12\pi
\]

حجم هندسی \(\pi r^2 h=12\pi\) همین است.

### بازبینی با پایتون



```python
import sympy as sp
r, theta, z = sp.symbols("r theta z")
volume = sp.integrate(r, (z, 0, 3), (r, 0, 2), (theta, 0, 2*sp.pi))
print(volume)
assert volume == 12*sp.pi
print("تایید شد")

```

## مثال ۴ — کرهٔ واحد

### حل ریاضی

\[
V=\int_0^{2\pi}\int_0^{\pi}\int_0^1 \rho^2\sin\varphi\,d\rho\,d\varphi\,d\theta
=\frac{1}{3}\cdot 2\pi\cdot \int_0^{\pi}\sin\varphi\,d\varphi
=\frac{2\pi}{3}\cdot 2=\frac{4\pi}{3}
\]

### بازبینی با پایتون



```python
import sympy as sp
rho, phi, theta = sp.symbols("rho phi theta", real=True)
volume = sp.integrate(rho**2 * sp.sin(phi), (rho, 0, 1), (phi, 0, sp.pi), (theta, 0, 2*sp.pi))
print(volume)
assert sp.simplify(volume - 4*sp.pi/3) == 0
print("تایید شد")

```

## نکته

فراموش کردن \(r\) یا \(\rho^2\sin\varphi\)، مساحت و حجم را با یک انتگرال دکارتیِ اشتباه عوض می‌کند.
ژاکوبی را یک بار با دترمینان حساب کنید تا عامل را حفظ کنید نه حفظیِ بی‌دلیل.
