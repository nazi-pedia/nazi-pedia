# The decomposition and $R^2$

A fitted line with an intercept splits the variation of $y$ into a part along the fitted values and a part in the residuals. The ratio of those parts is $R^2$. Adjusted $R^2$ then asks whether that ratio survives a penalty for the number of estimated coefficients.


## Why the cross term vanishes

Write each centered response as fitted deviation plus residual:

$$
y_i - \bar y = (\widehat y_i - \bar y) + (y_i - \widehat y_i).
$$

Square both sides and sum. The cross term is

$$
2\sum_{i=1}^{n}(\widehat y_i - \bar y)\, e_i.
$$

With an intercept in the model, $\sum e_i = 0$, so $\sum \bar y\, e_i = 0$. Also $\widehat y_i = b_0 + b_1 x_i$ and $\sum x_i e_i = 0$, so $\sum \widehat y_i e_i = 0$. The cross term is therefore zero, and

$$
\underbrace{\sum (y_i - \bar y)^2}_{\mathrm{TSS}}
= \underbrace{\sum (\widehat y_i - \bar y)^2}_{\mathrm{SSR}}
+ \underbrace{\sum e_i^2}_{\mathrm{RSS}}.
$$

The coefficient of determination is the share of the total sum of squares that the line carries:

$$
R^2 = \dfrac{\mathrm{SSR}}{\mathrm{TSS}} = 1 - \dfrac{\mathrm{RSS}}{\mathrm{TSS}}.
$$

Adjusted $R^2$ compares variances rather than raw sums, with $p$ coefficients estimated, here $p = 2$:

$$
\bar R^2 = 1 - \dfrac{\mathrm{RSS}/(n - p)}{\mathrm{TSS}/(n - 1)}.
$$


## The five-point line

From the previous lesson, $\bar y = 7$, the fitted values are $3, 5, 7, 9, 11$, and $\mathrm{RSS} = 4$.

$$
\begin{aligned}
\mathrm{TSS}
&= (4-7)^2 + (4-7)^2 + (7-7)^2 + (8-7)^2 + (12-7)^2 \\
&= 9 + 9 + 0 + 1 + 25 = 44, \\
\mathrm{SSR}
&= (3-7)^2 + (5-7)^2 + (7-7)^2 + (9-7)^2 + (11-7)^2 \\
&= 16 + 4 + 0 + 4 + 16 = 40.
\end{aligned}
$$

The identity holds: $44 = 40 + 4$. Therefore

$$
R^2 = \dfrac{40}{44} = \dfrac{10}{11}.
$$

With $n = 5$ and $p = 2$,

$$
\bar R^2
= 1 - \dfrac{4/3}{44/4}
= 1 - \dfrac{4/3}{11}
= 1 - \dfrac{4}{33}
= \dfrac{29}{33}.
$$

Adjusted $R^2$ is smaller than $R^2$, as it must be whenever $p > 1$. The line still explains most of the variation: $29/33$ against $10/11 = 30/33$.


```python
# TSS, SSR, RSS, R-squared, and adjusted R-squared for the five-point line.
from fractions import Fraction
y = [4, 4, 7, 8, 12]
fitted = [3, 5, 7, 9, 11]
mean = Fraction(sum(y), len(y))
tss = sum((yi - mean) ** 2 for yi in y)
ssr = sum((yh - mean) ** 2 for yh in fitted)
rss = Fraction(sum((yi - yh) ** 2 for yi, yh in zip(y, fitted)))
r2 = ssr / tss
p = 2
adjusted = 1 - (rss / (len(y) - p)) / (tss / (len(y) - 1))
print("TSS SSR RSS", tss, ssr, rss)
print("R2", r2)
print("adjusted R2", adjusted)
assert tss == ssr + rss == 44
assert r2 == Fraction(10, 11)
assert adjusted == Fraction(29, 33)
```

## The three-point line

The same identity on $(1, 2),\ (2, 3),\ (3, 5)$, where $\bar y = 10/3$, $\mathrm{RSS} = 1/6$, and the fitted values are $11/6$, $10/3$, $29/6$:

$$
\begin{aligned}
\mathrm{TSS}
&= \left(-\dfrac{4}{3}\right)^2 + \left(-\dfrac{1}{3}\right)^2 + \left(\dfrac{5}{3}\right)^2
= \dfrac{42}{9} = \dfrac{14}{3}, \\
\mathrm{SSR} &= \dfrac{14}{3} - \dfrac{1}{6} = \dfrac{9}{2}, \\
R^2 &= 1 - \dfrac{1/6}{14/3} = 1 - \dfrac{1}{28} = \dfrac{27}{28}, \\
\bar R^2
&= 1 - \dfrac{(1/6)/1}{(14/3)/2}
= 1 - \dfrac{1}{14} = \dfrac{13}{14}.
\end{aligned}
$$

Again $\bar R^2 < R^2$: $13/14 = 26/28$ while $R^2 = 27/28$.


```python
# The decomposition on the fractional three-point fit.
from fractions import Fraction
y = [Fraction(2), Fraction(3), Fraction(5)]
fitted = [Fraction(11, 6), Fraction(10, 3), Fraction(29, 6)]
mean = sum(y) / len(y)
tss = sum((yi - mean) ** 2 for yi in y)
rss = sum((yi - yh) ** 2 for yi, yh in zip(y, fitted))
ssr = sum((yh - mean) ** 2 for yh in fitted)
r2 = 1 - rss / tss
adjusted = 1 - (rss / 1) / (tss / 2)
print("TSS SSR RSS", tss, ssr, rss)
print("R2", r2, "adjusted", adjusted)
assert tss == Fraction(14, 3) and rss == Fraction(1, 6) and ssr == Fraction(9, 2)
assert tss == ssr + rss
assert r2 == Fraction(27, 28) and adjusted == Fraction(13, 14)
```

## One change: a perfect line and a useless line

If $y = (3, 5, 7)$ and $x = (1, 2, 3)$, the line $1 + 2x$ has residuals $0, 0, 0$. Then $\mathrm{RSS} = 0$, so $R^2 = 1$.

If instead the response is constant, $y = (4, 4, 4)$, then $\mathrm{TSS} = 0$. The ratio $R^2$ is undefined. There is no variation to explain, and reporting $1$ or $0$ would both be inventions.


```python
# R-squared is 1 when the residuals vanish, and undefined when TSS is 0.
from fractions import Fraction
perfect_rss = sum((yi - (1 + 2 * xi)) ** 2 for xi, yi in zip([1, 2, 3], [3, 5, 7]))
constant = [4, 4, 4]
tss_constant = sum((yi - Fraction(sum(constant), 3)) ** 2 for yi in constant)
print("perfect RSS", perfect_rss)
print("TSS of a constant response", tss_constant)
assert perfect_rss == 0
assert tss_constant == 0
```

## Pitfall

$R^2 = 10/11$ does not mean that $10/11$ of the individual points lie on the line. Four of the five residuals on that table are nonzero. It also does not mean the slope is large: $R^2$ is unchanged if both $x$ and $y$ are measured in different units, while the slope changes. $R^2$ is a statement about sums of squares, and only about sums of squares.


```python
# Rescaling x changes the slope and leaves R-squared where it was.
from fractions import Fraction
y = [4, 4, 7, 8, 12]
x = [1, 2, 3, 4, 5]
x_hours = [60 * xi for xi in x]  # the same points, measured in minutes

def r_squared(predictors, responses):
    n = len(predictors)
    sx, sy = sum(predictors), sum(responses)
    sxy = sum(a * b for a, b in zip(predictors, responses))
    sx2 = sum(a * a for a in predictors)
    slope = Fraction(n * sxy - sx * sy, n * sx2 - sx ** 2)
    intercept = Fraction(sy - slope * sx, n)
    mean = Fraction(sy, n)
    rss = sum((yi - intercept - slope * xi) ** 2 for xi, yi in zip(predictors, responses))
    tss = sum((yi - mean) ** 2 for yi in responses)
    return slope, 1 - rss / tss

slope_hours, r2_hours = r_squared(x, y)
slope_minutes, r2_minutes = r_squared(x_hours, y)
print("per hour", slope_hours, r2_hours)
print("per minute", slope_minutes, r2_minutes)
assert r2_hours == r2_minutes == Fraction(10, 11)
assert slope_minutes == slope_hours / 60
```

## Summary

- With an intercept, $\mathrm{TSS} = \mathrm{SSR} + \mathrm{RSS}$ because the residuals are orthogonal to the fitted values.
- $R^2 = 1 - \mathrm{RSS}/\mathrm{TSS}$. On the five-point line it equals $10/11$.
- Adjusted $R^2$ uses $\mathrm{RSS}/(n-p)$ and $\mathrm{TSS}/(n-1)$. It was $29/33$ on that line, a little below $R^2$.
