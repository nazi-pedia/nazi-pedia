# Estimating the error variance

The Gauss–Markov variances still contain $\sigma^2$. The data estimate it from the residual sum of squares, after giving back one degree of freedom for every estimated coefficient.


## The unbiased residual variance

If $y = X\beta + \varepsilon$, $\varepsilon$ has covariance $\sigma^2 I$, and $X$ has rank $p$, then

$$
\dfrac{\mathrm{RSS}}{\sigma^2} = \dfrac{\|y - X\widehat\beta\|^2}{\sigma^2}
$$

has expectation $n - p$. The proof uses the residual projector $I - H$: it is symmetric and idempotent, its trace is $n - p$, and the expectation of a quadratic form $\varepsilon^{\mathsf T}A\varepsilon$ is $\sigma^2\operatorname{tr}(A)$ when $E(\varepsilon)=0$. Therefore

$$
s^2 = \dfrac{\mathrm{RSS}}{n - p}
$$

is unbiased for $\sigma^2$. Dividing by $n$, or by $n-1$, leaves a bias once $p$ coefficients have been fitted. The estimated covariance matrix is $s^2(X^{\mathsf T}X)^{-1}$, and a standard error is the square root of a diagonal entry.


## The five-point line

$\mathrm{RSS} = 4$, $n = 5$, and $p = 2$, so

$$
s^2 = \dfrac{4}{3}, \qquad s = \dfrac{2}{\sqrt{3}}.
$$

The estimated variances replace $\sigma^2$ by $4/3$ in the formulas of the previous lesson:

$$
\begin{aligned}
\widehat{\operatorname{Var}}(b_1) &= \dfrac{4/3}{10} = \dfrac{2}{15}, &
\operatorname{se}(b_1) &= \sqrt{\dfrac{2}{15}}, \\[4pt]
\widehat{\operatorname{Var}}(b_0) &= \dfrac{4}{3}\cdot\dfrac{11}{10} = \dfrac{22}{15}, &
\operatorname{se}(b_0) &= \sqrt{\dfrac{22}{15}}.
\end{aligned}
$$

Numerically, $\operatorname{se}(b_1) \approx 0.365$ and $\operatorname{se}(b_0) \approx 1.211$. The slope $2$ is several of its standard errors away from zero. The formal test is the next lesson; the scale of the noise is the result here.


```python
# s^2 = RSS / (n - p), then the two standard errors.
import math
rss, n, p = 4, 5, 2
sxx = 10
s2 = rss / (n - p)
se_slope = math.sqrt(s2 / sxx)
se_intercept = math.sqrt(s2 * (1 / n + 9 / sxx))
print("s^2", s2)
print("se slope", se_slope)
print("se intercept", se_intercept)
assert abs(s2 - 4 / 3) < 1e-12
assert abs(se_slope - math.sqrt(2 / 15)) < 1e-12
assert abs(se_intercept - math.sqrt(22 / 15)) < 1e-12
```

## One change: one fewer coefficient

Suppose the slope is not estimated and the model is only an intercept. Then $p = 1$, the fitted value is $\bar y = 7$, and $\mathrm{RSS} = \mathrm{TSS} = 44$. The residual variance becomes

$$
s^2_{\text{constant}} = \dfrac{44}{5 - 1} = 11,
$$

which is much larger than $4/3$. The degrees of freedom changed by one, but the residual sum of squares changed from $4$ to $44$. The interesting movement is the residual sum, not the denominator.


```python
# The intercept-only residual variance uses n - 1 and the total sum of squares.
constant_s2 = 44 / (5 - 1)
line_s2 = 4 / (5 - 2)
print("constant model", constant_s2)
print("line model", line_s2)
assert constant_s2 == 11
assert line_s2 == 4 / 3
```

## Pitfall

$s$ is not the standard deviation of the residuals in the "divide by $n$" sense, and it is not the standard deviation of $y$. Three different numbers sit next to each other on this table:

$$
\sqrt{\dfrac{1}{n}\sum e_i^2} = \sqrt{\dfrac{4}{5}}, \qquad
s = \sqrt{\dfrac{4}{3}}, \qquad
\sqrt{\dfrac{\mathrm{TSS}}{n-1}} = \sqrt{11}.
$$

The middle one estimates $\sigma$. The last one describes the raw response. The first one is the root mean squared residual and is too small as an estimate of $\sigma$, because it does not return the two fitted degrees of freedom.


```python
# Three similar-looking square roots are three different quantities.
import math
rmse = math.sqrt(4 / 5)
s = math.sqrt(4 / 3)
sy = math.sqrt(11)
print("RMSE", rmse)
print("residual s", s)
print("sample sd of y", sy)
assert rmse < s < sy
```

## Summary

- $s^2 = \mathrm{RSS}/(n-p)$ is unbiased for $\sigma^2$ in the Gauss–Markov model.
- On the five-point line, $s^2 = 4/3$, $\operatorname{se}(b_1)=\sqrt{2/15}$, and $\operatorname{se}(b_0)=\sqrt{22/15}$.
- Do not replace $n-p$ by $n$ and still call the result an unbiased estimate of $\sigma^2$.
