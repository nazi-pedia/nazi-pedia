# Residuals and the sum of squares

Once a line is used on data, every observation leaves a residual: the response minus the prediction. Linear regression chooses the line that makes the sum of the squared residuals as small as possible. This lesson does not yet derive that line. It learns to score any proposed line, so the winning line will be recognizable when it appears.


## Definitions

For a proposed intercept $b_0$ and slope $b_1$,

$$
\widehat y_i = b_0 + b_1 x_i, \qquad e_i = y_i - \widehat y_i, \qquad \mathrm{RSS} = \sum_{i=1}^{n} e_i^2.
$$

The residual $e_i$ is positive when the point sits above the line. Squaring treats an error of $+2$ and an error of $-2$ as equally expensive, and it charges a single error of $2$ four times as much as an error of $1$.

The working table in this lesson is

$$
\begin{array}{c|ccc}
x & 1 & 2 & 3 \\
\hline
y & 2 & 3 & 5
\end{array}
$$


## Two candidate lines

Candidate A is $\widehat y = 1 + x$.

$$
\begin{aligned}
\widehat y &: 2,\ 3,\ 4, \\
e &: 0,\ 0,\ 1, \\
\mathrm{RSS}_A &= 0^2 + 0^2 + 1^2 = 1.
\end{aligned}
$$

Candidate B refuses to use $x$ and predicts the constant $3$, which is not even the mean.

$$
\begin{aligned}
\widehat y &: 3,\ 3,\ 3, \\
e &: -1,\ 0,\ 2, \\
\mathrm{RSS}_B &= 1 + 0 + 4 = 5.
\end{aligned}
$$

Candidate A is better on this criterion. That does not mean A is best. The mean, which is the best constant, scores

$$
\bar y = \dfrac{10}{3}, \qquad
S\!\left(\dfrac{10}{3}\right)
= \left(\dfrac{-4}{3}\right)^2 + \left(\dfrac{-1}{3}\right)^2 + \left(\dfrac{5}{3}\right)^2
= \dfrac{16 + 1 + 25}{9}
= \dfrac{14}{3}.
$$

So $\mathrm{RSS}_A = 1 < 14/3$, and even a rough line has beaten every constant. The best line will have to beat $1$ as well.


```python
# Score two proposed lines and the best constant on one table.
from fractions import Fraction
x = [1, 2, 3]
y = [2, 3, 5]

def rss(intercept, slope):
    errors = [yi - (intercept + slope * xi) for xi, yi in zip(x, y)]
    return errors, sum(error ** 2 for error in errors)

errors_a, rss_a = rss(1, 1)
errors_b, rss_b = rss(3, 0)
mean = Fraction(sum(y), len(y))
errors_m, rss_m = rss(mean, 0)
print("line residuals", errors_a, "RSS", rss_a)
print("constant 3 residuals", errors_b, "RSS", rss_b)
print("mean", mean, "RSS", rss_m)
assert errors_a == [0, 0, 1] and rss_a == 1
assert errors_b == [-1, 0, 2] and rss_b == 5
assert mean == Fraction(10, 3) and rss_m == Fraction(14, 3)
assert rss_a < rss_m < rss_b
```

## One change: move the slope by a tenth

Keep the intercept of candidate A and replace the slope $1$ by $6/5$.

$$
\begin{aligned}
\widehat y &= \dfrac{1}{1} + \dfrac{6}{5}x
&&\text{gives}&&
\dfrac{11}{5},\ \dfrac{17}{5},\ \dfrac{23}{5}, \\
e &= 2 - \dfrac{11}{5},\ 3 - \dfrac{17}{5},\ 5 - \dfrac{23}{5}
= -\dfrac{1}{5},\ -\dfrac{2}{5},\ \dfrac{2}{5}, \\
\mathrm{RSS} &= \dfrac{1 + 4 + 4}{25} = \dfrac{9}{25}.
\end{aligned}
$$

A small change in the slope cut the cost from $1$ to $9/25$. The surface $S(b_0, b_1)$ is smooth: nearby lines have nearby costs, and a better neighbor can exist even when a line already looks reasonable.


```python
# A one-tenth change in the slope can lower the residual sum of squares.
from fractions import Fraction
x = [1, 2, 3]
y = [Fraction(2), Fraction(3), Fraction(5)]
intercept, slope = Fraction(1), Fraction(6, 5)
errors = [yi - (intercept + slope * xi) for xi, yi in zip(x, y)]
total = sum(error ** 2 for error in errors)
print("residuals", errors)
print("RSS", total)
assert errors == [Fraction(-1, 5), Fraction(-2, 5), Fraction(2, 5)]
assert total == Fraction(9, 25)
assert total < 1
```

## Pitfall

The sums of the raw residuals can flatter a bad line. Candidate B has residuals $-1, 0, 2$. Their sum is $+1$, while their squares sum to $5$. Candidate A has residual sum $1$ and squared sum $1$. The unsquared sum does not rank the lines the same way, and it can be driven to zero by a line that is still a poor fit: positive and negative errors cancel. Least squares refuses that cancellation by squaring before adding.


```python
# The signed residual sum can hide a large squared cost.
signed_b = -1 + 0 + 2
squared_b = 1 + 0 + 4
signed_a = 0 + 0 + 1
squared_a = 1
print("signed sums", signed_a, signed_b)
print("squared sums", squared_a, squared_b)
assert signed_a == signed_b
assert squared_a < squared_b
```

## Summary

- The residual is $y_i - \widehat y_i$. The residual sum of squares is the cost of a proposed line.
- Squaring stops positive and negative errors from cancelling and charges large errors more than small ones.
- A line can beat every constant without being the best line. The next lesson finds that best line.
