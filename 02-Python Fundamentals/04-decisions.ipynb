`if` runs a block only when its test is true. `elif` tries the next test when earlier tests were false. `else` runs when every test was false. The block is the indented lines underneath the test.

Comparisons produce `True` or `False`:

$$
<,\ \le,\ >,\ \ge,\ ==,\ \ne.
$$

In Python those last two are written `==` and `!=`. `and`, `or`, and `not` combine tests. `and` stops early when the left side is false. `or` stops early when the left side is true. That early stop is short-circuit evaluation.


## Example 1 — One temperature, three branches

### Predict

`18` fails `>= 30` and passes `>= 15`, so only the middle branch runs.

### Run


```python
temperature = 18
if temperature >= 30:
    print("hot")
elif temperature >= 15:
    # 18 is not hot, and it is at least 15, so this branch is the one that runs.
    print("mild")
else:
    print("cold")
```

## Example 2 — Short-circuit evaluation

### Predict

`left` returns false and prints `left`. `and` does not need the right side, so `right` is not called. `or` does need another chance, so the second pair prints `left` and then `right`.

### Run


```python
def left():
    # A side effect in a test makes the short-circuit visible.
    print("left")
    return False

def right():
    print("right")
    return True

# False and ... never evaluates the right function.
print("and-result", left() and right())
print("---")
# False or ... must evaluate the right function.
print("or-result", left() or right())
```

## Example 3 — Which values count as false

### Predict

Zero, empty text, an empty list, and `None` are false in a test. Everything else in this list is true, including the list that contains a zero. The list is not empty, so the list itself is true.

### Run


```python
samples = (0, 1, "", "hi", [], [0], None)
for value in samples:
    # The test is the value itself. Python asks whether it counts as true.
    if value:
        print(repr(value), "true")
    else:
        print(repr(value), "false")
```

## Example 4 — `True` is also the integer 1

### Predict

`True + True` equals `2`, because `bool` is a special kind of `int` with the values `0` and `1`. That fact is a curiosity. Logic should still be written with `and` and `or`, not with addition.

### Run


```python
print(True + True)
print(isinstance(True, int))
# Prefer a real test. Adding booleans hides the decision.
print(True and False)
```

## Example 5 — A sign function, written flat

### Predict

Nested tests work, but a flat `elif` chain is easier to read. For `-4` the first test fails and the second succeeds.

### Run


```python
number = -4
# A flat chain chooses the sign: positive, negative, or zero.
if number > 0:
    sign = 1
elif number < 0:
    sign = -1
else:
    sign = 0
print(sign)
```

## A pitfall

`=` attaches a name. `==` compares values. An `if` test that uses a single `=` is a syntax error. Python even hints that you may have meant `==`.

### Run


```python
try:
    compile("if x = 1:\n    pass\n", "<example>", "exec")
except SyntaxError as error:
    print(type(error).__name__)
    print(error.msg)
# The comparison is the legal sentence.
x = 1
print(x == 1)
```

## Summary

`if`, `elif`, and `else` choose one indented block. Comparisons and boolean operators build the test. `and` and `or` stop as soon as the answer is known. Empty containers, zero, and `None` count as false. Write comparisons with `==`, and write assignment with `=`.
