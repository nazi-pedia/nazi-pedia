Python has two everyday number kinds. `int` is a whole number of any size. `float` is a number with a fractional part, stored in binary, which is why some decimal fractions are not exact.

The operators, in the order Python applies them:

$$
\begin{aligned}
x ** y &\quad \text{power, before the rest} \\
+x,\ -x &\quad \text{sign} \\
x * y,\ x / y,\ x // y,\ x \% y &\quad \text{multiply and divide} \\
x + y,\ x - y &\quad \text{add and subtract}
\end{aligned}
$$

Parentheses override that order. `/` is true division and always returns a float. `//` is floor division: it rounds toward negative infinity.


## Example 1 — The seven operations

### Predict

`7 / 2` is `3.5`, a float. `7 // 2` is `3`. `7 % 2` is the remainder `1`. `7 ** 2` is `49`.

$$
7 = 3 \cdot 2 + 1.
$$

### Run


```python
print("add", 7 + 2)
print("subtract", 7 - 2)
print("multiply", 7 * 2)
# True division always produces a float, even when the result looks whole.
print("divide", 7 / 2)
print("four over two", 4 / 2)
# Floor division discards the fractional part toward -infinity.
print("floor", 7 // 2)
print("remainder", 7 % 2)
print("power", 7 ** 2)
```

## Example 2 — Precedence, then parentheses

### Predict

Power runs first, then multiplication, then addition.

$$
2 + 3 \times 4^{2} = 2 + 3 \times 16 = 50.
$$

Parentheses pull the addition forward:

$$
(2 + 3) \times 4^{2} = 5 \times 16 = 80.
$$

### Run


```python
# ** before *, then +.
print(2 + 3 * 4 ** 2)
# The parentheses force 2 + 3 to happen first.
print((2 + 3) * 4 ** 2)
```

## Example 3 — Floor division of a negative number

### Predict

Floor means "the greatest integer that is still less than or equal to the quotient." $-3.5$ floors to $-4$, not to $-3$.

$$
-7 = (-4) \cdot 2 + 1,
$$

so the quotient is $-4$ and the remainder is $1$. The remainder takes the sign of the divisor, and here the divisor is positive.

### Run


```python
# -7 / 2 is -3.5. The floor of -3.5 is -4, not -3.
print("quotient", -7 // 2)
# Remainder satisfies dividend = quotient * divisor + remainder.
print("remainder", -7 % 2)
print("check", (-7 // 2) * 2 + (-7 % 2))
```

## Example 4 — A decimal that binary cannot hold exactly

### Predict

`0.1 + 0.2` is not stored as the decimal `0.3`. The printed sum shows the binary approximation. Rounding to ten places recovers the decimal you expected. Do not compare money with bare `==` on floats.

### Run


```python
# The extra digits are the binary approximation, not a printing glitch you should ignore blindly.
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
# round is the right tool when you want a fixed number of decimal places.
print(round(0.1 + 0.2, 10))
```

## Example 5 — Converting kinds, and a chained comparison

### Predict

`int(3.9)` chops toward zero. It does not round. `int("42")` parses text. A chained comparison `1 < x < 10` is true only when both parts are true.

### Run


```python
# int on a float truncates toward zero. 3.9 becomes 3, not 4.
print(int(3.9))
print(int(-3.9))
# int on text parses a whole-number string.
print(int("42"))
print(float("3.5"))
x = 5
# This is (1 < x) and (x < 10), written as one chain.
print(1 < x < 10)
# divmod returns the floor quotient and the remainder together.
print(divmod(17, 5))
```

## A pitfall

`round` uses bankers' rounding on a tie: a value that lands exactly halfway is rounded toward the even digit. `round(2.5)` is `2`, and `round(3.5)` is `4`. To one decimal place, `round(1.25, 1)` is `1.2`, because the digit `2` is already even.

### Run


```python
# Halves round toward the even integer, not always away from zero.
print(round(2.5))
print(round(3.5))
print(round(1.25, 1))
```

## Summary

`*` and `/` happen before `+` and `-`, and `**` happens before those. `/` always returns a float. `//` floors toward negative infinity, and `%` is the remainder that makes the division identity true. Floats are binary approximations, so round them when you need a decimal place, and do not trust `==` for a value that came from a decimal fraction.
