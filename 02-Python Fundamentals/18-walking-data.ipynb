Most everyday Python is a walk over a collection. Four tools cover that walk. `enumerate` adds a position. `zip` walks two collections together. `sorted` orders a copy. `any` and `all` answer a yes-or-no question about every item.

$$
\texttt{all}(p(x) \text{ for } x \text{ in items})
$$

is true when every item passes. `any` is true when at least one item passes.


## Example 1 — Positions that start where you choose

### Predict

`enumerate` yields pairs. `start=1` makes the first position 1 rather than 0, which is what a person means by "item 1."

### Run


```python
# start=1 numbers the people the way a list on paper would.
for position, name in enumerate(["Ada", "Grace"], start=1):
    print(position, name)
```

## Example 2 — zip stops at the shorter collection

### Predict

Two names and three scores produce two pairs. The leftover score 70 is not visited. `zip` does not invent a missing name, and it does not raise.

### Run


```python
names = ["Ada", "Grace"]
scores = [90, 95, 70]
# zip stops when the names run out. The last score is unused.
pairs = list(zip(names, scores))
print(pairs)
print(len(pairs))
```

## Example 3 — Sort by a key, not by the item itself

### Predict

Alphabetical order would put `ada` first. `key=len` orders by length, so `ada` (3), then `code` (4), then `python` (6). The original list stays in its old order.

### Run


```python
words = ["python", "ada", "code"]
# The plain sort is alphabetical. key=len sorts by length and leaves words alone.
print(sorted(words))
print(sorted(words, key=len))
print(words)
```

## Example 4 — any and all

### Predict

Every score is at least 60, so `all` is true. No score is 100, so `any` is false. An empty `all` is true: there is no item that fails. An empty `any` is false: there is no item that passes.

### Run


```python
scores = [80, 70, 90]
# all is true only when every item passes. any is true when one item passes.
print(all(score >= 60 for score in scores))
print(any(score == 100 for score in scores))
# An empty all is true. An empty any is false.
print(all(score >= 60 for score in ()))
print(any(score >= 60 for score in ()))
```

## Example 5 — The largest item under a chosen measure

### Predict

`max` with `key=len` returns the longest word, `python`, not the word that comes last in the alphabet.

### Run


```python
words = ["python", "ada", "code"]
# Without a key, max uses alphabetical order. With key=len it uses length.
print(max(words))
print(max(words, key=len))
print(min(words, key=len))
```

## A pitfall

`zip` silently stops. If a missing pair would be a bug, compare the lengths before you zip. This cell treats unequal lengths as a failed check.

### Run


```python
names = ["Ada", "Grace"]
scores = [90, 95, 70]
# Unequal lengths would drop a score, so the check refuses to zip.
print(len(names) == len(scores))
if len(names) != len(scores):
    print("refusing to zip unequal rows")
else:
    print(list(zip(names, scores)))
```

## Summary

`enumerate` numbers a walk. `zip` walks collections side by side and stops at the shorter one. `sorted` and `max` take a `key` when the order you want is not the item's own order. `all` and `any` reduce a test over every item to one boolean.
