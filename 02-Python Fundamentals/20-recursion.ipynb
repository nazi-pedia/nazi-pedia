A function may call itself. Each call must be about a smaller problem, and one branch must stop without calling again. That branch is the base case.

The factorial of a whole number is the definition used here.

$$
n! =
\begin{cases}
1 & n = 0 \\
n \cdot (n-1)! & n \ge 1
\end{cases}
$$

So $5! = 5 \cdot 4 \cdot 3 \cdot 2 \cdot 1 = 120$.


## Example 1 — Factorial, with the base case first

### Predict

`factorial(0)` returns 1 without a further call. `factorial(5)` returns 120. The loop below multiplies the same integers and agrees.

### Run


```python
def factorial(n):
    # 0! is 1. This branch does not call factorial again.
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(0))
print(factorial(5))
product = 1
for number in range(1, 6):
    product = product * number
print(product)
```

## Example 2 — The calls, written out

### Predict

Each call prints its argument before the recursive call returns. The prints appear on the way down: 3, then 2, then 1, then 0. The value is built on the way back up.

### Run


```python
def factorial(n):
    # Print on the way down. The product is built on the way back up.
    print("call", n)
    if n == 0:
        return 1
    return n * factorial(n - 1)

print("value", factorial(3))
```

## Example 3 — The sum of a list, by the same shape

### Predict

The sum of an empty list is 0. The sum of a non-empty list is its first item plus the sum of the rest.

$$
\mathrm{sum}([4, 1, 7]) = 4 + \mathrm{sum}([1, 7]) = 12.
$$

### Run


```python
def total(numbers):
    if numbers == []:
        return 0
    # The rest is a shorter list, so the problem shrinks.
    return numbers[0] + total(numbers[1:])

print(total([]))
print(total([4, 1, 7]))
```

## Example 4 — A missing base case must be stopped on purpose

### Predict

This version never reaches a base case. A depth limit stops it and reports the failure. A real call without that limit would run until Python raises `RecursionError`. The limit keeps the lesson finite while showing the same mistake.

### Run


```python
def never_ends(n, depth, limit):
    # The depth limit stands in for the missing base case.
    if depth == limit:
        return "stopped: no base case"
    return never_ends(n - 1, depth + 1, limit)

print(never_ends(5, 0, 4))
```

## A pitfall

A base case that is present but unreachable is the same bug. `factorial_bad(0)` would recurse through negative numbers. The guard below rejects a negative argument before any call, which is the check the definition requires: $n$ is a whole number, including 0.

### Run


```python
def factorial(n):
    # Reject values the definition does not cover, before any recursive call.
    if n < 0:
        raise ValueError("factorial is defined for whole numbers")
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(4))
try:
    print(factorial(-1))
except ValueError as error:
    print(error)
```

## Summary

A recursive function needs a base case and a step that gets closer to it. Factorial stops at 0. The sum of a list stops at the empty list. If every call makes another call, the program does not finish. Reject values that the definition does not cover, such as a negative factorial, before the recursion starts.
