A program is a list of instructions. Python reads that list from top to bottom and does one instruction at a time. The instruction you will use first is `print`. It sends a value to the output.

## How this path teaches

Every example has the same rhythm.

1. **Predict.** Read the code and decide what the output will be.
2. **Run.** The cell under the prediction has already been executed. Compare your guess with the stored output.
3. **One change.** The next example moves a single detail, so you can see what that detail controls.
4. **Pitfall.** One case looks almost right and is not.
5. **Summary.** Three sentences you can keep.

Write the prediction on paper before you read the output. The stored output is the answer key, not the first thing to look at.


## Example 1 — One line of output

### Predict

`print` writes its value, then moves to the next line. There is one call, so there is one line.

### Run


```python
# Send one piece of text to the output.
print("Hello, Python")
```

## Example 2 — Several values, and where the line ends

### Predict

The first call joins two words with a comma and a space, because `sep` is the separator between values. The second call sets `end` to a space, so the third call continues on the same line.

### Run


```python
# sep is placed between values. The default separator is one space.
print("Ada", "Lovelace", sep=", ")
# end replaces the usual newline. A space keeps the next print on this line.
print("same", end=" ")
print("line")
```

## Example 3 — An expression has a value

### Predict

Multiplication happens before addition:

$$
2 + 3 \times 4 = 2 + 12 = 14.
$$

The name `total` stores that value. Nothing is shown until `print` asks for it.

### Run


```python
# * runs before +, so this is 2 + 12, not 5 * 4.
total = 2 + 3 * 4
# The name holds the value. print is what makes it visible.
print(total)
```

## Example 4 — Comments are for the reader

### Predict

A line that starts with `#` is skipped. Only the `print` runs.

### Run


```python
# Python ignores this note. It exists so a person can follow the step.
print("only this runs")
```

## Example 5 — A runtime error is a message, not a mystery

### Predict

Dividing by zero is legal syntax and an illegal operation. Python raises `ZeroDivisionError` and attaches a short reason. This cell catches the error so the lesson can print that reason. Left uncaught, the program would stop at this line.

### Run


```python
try:
    # The division is attempted. It cannot produce a number.
    print(1 / 0)
except ZeroDivisionError as error:
    # type(error).__name__ is the error's family. The message is the reason.
    print(type(error).__name__)
    print(error)
```

## Example 6 — A syntax error means the sentence is not Python

### Predict

The string below is missing the closing parenthesis. Python refuses to run it at all. `compile` checks the sentence and raises `SyntaxError` before any `print` can happen.

### Run


```python
# The quote closes, but the parenthesis that opened print never closes.
source = "print('hi'"
try:
    compile(source, "<example>", "exec")
except SyntaxError as error:
    print(type(error).__name__)
    print(error.msg)
```

## A pitfall

`"2" + "3"` glues text together. It does not add numbers. `2 + "3"` is rejected, because an integer and a piece of text are different kinds of value.

### Run


```python
# Both sides are text, so + concatenates.
print("2" + "3")
try:
    # An int and a str cannot be added. The kinds do not match.
    print(2 + "3")
except TypeError as error:
    print(type(error).__name__)
    print(error)
```

## Summary

Python runs statements from top to bottom. `print` is how a value becomes visible. Comments start with `#` and are skipped. When a program stops, read the error name and the message: a syntax error means the sentence is malformed, and a runtime error means a legal sentence could not be carried out.
