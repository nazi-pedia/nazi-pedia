# ۳۷ · استوکس و دیورژانس

| | |
|---|---|
| **دوره** | ریاضی عمومی چندمتغیره |
| **پیش‌نیاز** | ۳۶ |

تاو:

\[
\nabla\times\mathbf{F}
=\left(\partial_y F_3-\partial_z F_2,\;
\partial_z F_1-\partial_x F_3,\;
\partial_x F_2-\partial_y F_1\right)
\]

دیورژانس \(\nabla\cdot\mathbf{F}=\partial_x F_1+\partial_y F_2+\partial_z F_3\).

استوکس، گرین را به رویه بلند می‌کند: شار تاو از رویه برابر گردش میدان روی مرز رویه است.
دیورژانس: شار از سطح بسته برابر انتگرال دیورژانس داخل حجم است.

آزمون‌های سریع: تاو گرادیان صفر است. دیورژانس تاو صفر است.



## مثال ۱ — تاو گرادیان

### حل ریاضی

\(f=x^2+yz\). \(\nabla f=(2x,z,y)\).
تاو این میدان باید صفر باشد، چون هر میدان گرادیانی بی‌تاو است، روی ناحیهٔ ساده.

### بازبینی با پایتون



```python
import sympy as sp
x, y, z = sp.symbols("x y z")
f = x**2 + y*z
grad = sp.Matrix([sp.diff(f, v) for v in (x, y, z)])
curl = sp.Matrix([
    sp.diff(grad[2], y) - sp.diff(grad[1], z),
    sp.diff(grad[0], z) - sp.diff(grad[2], x),
    sp.diff(grad[1], x) - sp.diff(grad[0], y),
])
print(grad, curl)
assert curl == sp.Matrix([0, 0, 0])
print("تایید شد")

```

## مثال ۲ — استوکس روی قرص

### حل ریاضی

\(\mathbf{F}=(-y,x,0)\). تاو \((0,0,2)\) است.
روی قرص واحد در صفحهٔ \(xy\)، بردار نرمال \(\mathbf{k}\) است.

\[
\iint_S (\nabla\times\mathbf{F})\cdot d\mathbf{S}
=\iint_D 2\,dA=2\pi
\]

گردش مرزی همان حساب گرین در درس ۳۶ است.

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y")
flux = sp.integrate(2, (y, -sp.sqrt(1 - x**2), sp.sqrt(1 - x**2)), (x, -1, 1))
print(flux)
assert flux == 2*sp.pi
print("تایید شد")

```

## مثال ۳ — دیورژانس روی کره

### حل ریاضی

\(\mathbf{F}=(x,y,z)\). دیورژانس ۳ است.
شار خروجی از کرهٔ واحد:

\[
\oiint\mathbf{F}\cdot d\mathbf{S}=\iiint_B 3\,dV=3\cdot\frac{4\pi}{3}=4\pi
\]

### بازبینی با پایتون



```python
import sympy as sp
rho, phi, theta = sp.symbols("rho phi theta", real=True)
volume = sp.integrate(rho**2*sp.sin(phi), (rho, 0, 1), (phi, 0, sp.pi), (theta, 0, 2*sp.pi))
flux = sp.simplify(3*volume)
print(flux)
assert flux == 4*sp.pi
print("تایید شد")

```

## مثال ۴ — دیورژانس تاو

### حل ریاضی

برای \(\mathbf{F}=(y,z,x)\)، تاو \((-1,-1,-1)\) است؟ 
\(F=(yz, xz, xy)\). آن‌وقت
\(\nabla\times F=(x-x, y-y, z-z)=\mathbf{0}\) و دیورژانس تاو بدیهی صفر است.
مثال غیرصفر: \(F=(-y, x, z)\).

\[
\nabla\times F=(0,0,2)
\]

دیورژانس این تاو صفر است، هرچند خود تاو صفر نیست.

### بازبینی با پایتون



```python
import sympy as sp
x, y, z = sp.symbols("x y z")
F = sp.Matrix([-y, x, z])
curl = sp.Matrix([
    sp.diff(F[2], y) - sp.diff(F[1], z),
    sp.diff(F[0], z) - sp.diff(F[2], x),
    sp.diff(F[1], x) - sp.diff(F[0], y),
])
div_curl = sp.diff(curl[0], x) + sp.diff(curl[1], y) + sp.diff(curl[2], z)
print(curl, div_curl)
assert curl == sp.Matrix([0, 0, 2])
assert div_curl == 0
print("تایید شد")

```

## نکته

استوکس جهت نرمال و جهت مرز را با قاعدهٔ دست راست به هم قفل می‌کند.
اگر نرمال را برعکس بگیرید، هر دو طرف معادله تغییر علامت می‌دهند و برابری می‌ماند، ولی عدد شار عوض می‌شود.

## جمع‌بندی

تاو، چرخش موضعی است و استوکس آن را به گردش مرز وصل می‌کند.
دیورژانس، چشمهٔ موضعی است و قضیهٔ دیورژانس آن را به شار سطح بسته وصل می‌کند.
