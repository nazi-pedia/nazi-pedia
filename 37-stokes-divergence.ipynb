The curl of a vector field $\mathbf{F}=(F_1,F_2,F_3)$ is

$$
\nabla\times\mathbf{F}
=\left(
\partial_y F_3-\partial_z F_2,\;
\partial_z F_1-\partial_x F_3,\;
\partial_x F_2-\partial_y F_1
\right).
$$

The divergence is the scalar $\nabla\cdot\mathbf{F}=\partial_x F_1+\partial_y F_2+\partial_z F_3$.

Stokes' theorem lifts Green's theorem from a plane region onto a surface: the flux of the curl through the surface equals the circulation of the field around the boundary of the surface.
The divergence theorem says that the flux through a closed surface equals the triple integral of the divergence over the enclosed volume.

Two quick tests follow at once. The curl of a gradient is $\mathbf{0}$. The divergence of a curl is $0$.


## Example 1 — Curl of a gradient

### Hand solution

Take the scalar field $f=x^2+yz$. Its gradient is $\nabla f=(2x,z,y)$.
The curl of this field is $\mathbf{0}$, because every gradient field is irrotational on a simply connected region.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Coordinates and the scalar field.
x, y, z = sp.symbols("x y z")
f = x**2 + y*z
# Gradient, one partial derivative per coordinate.
grad = sp.Matrix([sp.diff(f, v) for v in (x, y, z)])
# Curl of that gradient.
curl = sp.Matrix([
    sp.diff(grad[2], y) - sp.diff(grad[1], z),
    sp.diff(grad[0], z) - sp.diff(grad[2], x),
    sp.diff(grad[1], x) - sp.diff(grad[0], y),
])
print(grad, curl)
# A gradient field is irrotational.
assert curl == sp.Matrix([0, 0, 0])
print("Confirmed")
```

## Example 2 — Stokes on a disk

### Hand solution

Take $\mathbf{F}=(-y,x,0)$. The curl is $(0,0,2)$.
On the unit disk in the $xy$-plane the unit normal is $\mathbf{k}$.

$$
\iint_S (\nabla\times\mathbf{F})\cdot d\mathbf{S}
=\iint_D 2\,dA
=2\pi.
$$

The circulation around the boundary is the same Green's theorem calculation as in lesson 36.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Coordinates on the unit disk.
x, y = sp.symbols("x y")
# Flux of the constant curl component 2 through the unit disk.
flux = sp.integrate(2, (y, -sp.sqrt(1 - x**2), sp.sqrt(1 - x**2)), (x, -1, 1))
print(flux)
# Stokes predicts the area of the disk times 2, which is 2*pi.
assert flux == 2*sp.pi
print("Confirmed")
```

## Example 3 — Divergence on a sphere

### Hand solution

Take $\mathbf{F}=(x,y,z)$. The divergence is $3$.
The outward flux through the unit sphere is the triple integral of that constant:

$$
\oiint\mathbf{F}\cdot d\mathbf{S}
=\iiint_B 3\,dV
=3\cdot\dfrac{4\pi}{3}
=4\pi.
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Spherical coordinates for the unit ball.
rho, phi, theta = sp.symbols("rho phi theta", real=True)
# Volume of the unit ball, with the spherical volume element.
volume = sp.integrate(rho**2*sp.sin(phi), (rho, 0, 1), (phi, 0, sp.pi), (theta, 0, 2*sp.pi))
# Divergence is 3, so the flux is 3 times the volume.
flux = sp.simplify(3*volume)
print(flux)
# The divergence theorem gives the flux 4*pi.
assert flux == 4*sp.pi
print("Confirmed")
```

## Example 4 — Divergence of a curl

### Hand solution

For $\mathbf{F}=(y,z,x)$, is the curl $(-1,-1,-1)$?
For the field $\mathbf{F}=(yz,xz,xy)$,

$$
\nabla\times\mathbf{F}=(x-x,\, y-y,\, z-z)=\mathbf{0},
$$

and the divergence of this curl is zero for the trivial reason that the curl itself is $\mathbf{0}$.
A field whose curl is not zero is $\mathbf{F}=(-y,x,z)$:

$$
\nabla\times\mathbf{F}=(0,0,2).
$$

The divergence of this curl is still $0$.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Coordinates and the field F = (-y, x, z).
x, y, z = sp.symbols("x y z")
F = sp.Matrix([-y, x, z])
# Curl of F.
curl = sp.Matrix([
    sp.diff(F[2], y) - sp.diff(F[1], z),
    sp.diff(F[0], z) - sp.diff(F[2], x),
    sp.diff(F[1], x) - sp.diff(F[0], y),
])
# Divergence of that curl.
div_curl = sp.diff(curl[0], x) + sp.diff(curl[1], y) + sp.diff(curl[2], z)
print(curl, div_curl)
# The curl is (0, 0, 2), and the divergence of that curl is 0.
assert curl == sp.Matrix([0, 0, 2])
assert div_curl == 0
print("Confirmed")
```

## A pitfall

Stokes' theorem locks the normal direction and the boundary orientation together by the right-hand rule.
Reversing the normal changes the sign of both sides, and the equality remains. The numerical value of the flux changes with that choice of normal.

## Summary

The curl measures local rotation, and Stokes' theorem connects it to the circulation around the boundary.
The divergence measures the local source, and the divergence theorem connects it to the flux through a closed surface.
