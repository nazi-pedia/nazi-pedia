# 14 The Loss Can Rise

Give each point a loss equal to how far its margin sits below zero:

$$
L_i = \max(0,\, -m_i), \qquad m_i = y_i\, w\cdot\varphi_i
$$

The loss of the table is the sum of those pieces. It is never negative.

At the origin every score is zero, so every piece is $\max(0, 0) = 0$. The total loss is $0$. Zero is the smallest a sum of these pieces can be, so the origin is a minimum of the loss.

It is a useless minimum. Every paper is on the boundary, and the mistake count is $3$, not $0$.

For a point with $m < 0$, the piece is $-y\, w\cdot\varphi$. Its slope with respect to $w$ is $-y\varphi$. A downhill step of size $1$ is

$$
w - (-y\varphi) = w + y\varphi
$$

which is the perceptron update. On the boundary the slope is not a single vector. One legal choice is $0$, and that choice stays at the origin. The perceptron chooses $-y\varphi$ instead, the same slope it would use on the wrong side, and therefore leaves the minimum.

On the three-paper walk the loss is not a descending staircase.

| Weights | Margins of A, B, C | Loss | Mistakes |
|---|---|---|---|
| $(0, 0, 0)$ | $0, 0, 0$ | $0$ | $3$ |
| $(2, 1, 1)$ | $6, 5, -1$ | $1$ | $1$ |
| $(2, 1, 0)$ | $5, 4, 0$ | $0$ | $1$ |
| $(2, 1, -1)$ | $4, 3, 1$ | $0$ | $0$ |

The first update raises the loss from $0$ to $1$. The next weights have loss $0$ while C is still a mistake. Only the last update is a strict separator. Stopping at "the loss is zero" would stop too early, twice.


## Example 1 — The four rows

### Predict

The cell prints the losses `0, 1, 0, 0` and the mistake counts `3, 1, 1, 0`.


```python
# Hand table: the loss rises from 0 to 1, returns to 0 while C is still wrong, then the mistakes hit 0.
papers = [((2, 1), 1), ((1, 2), 1), ((0, 0), -1)]
trail = [(0, 0, 0), (2, 1, 1), (2, 1, 0), (2, 1, -1)]
losses = []
mistakes = []
for weights in trail:
    pieces = []
    wrong = 0
    for (x1, x2), label in papers:
        margin = label * (weights[0] * x1 + weights[1] * x2 + weights[2])
        pieces.append(max(0, -margin))
        wrong += margin <= 0
    losses.append(sum(pieces))
    mistakes.append(wrong)
print(losses)
print(mistakes)
assert losses == [0, 1, 0, 0]
assert mistakes == [3, 1, 1, 0]
```

## Example 2 — One badly wrong point does descend

Lesson 02 updated $x = 1$, $y = +1$ from $w = -5$, $b = 0$. That point's loss starts at $5$ and the square added each time is $2$:

$$
5 \rightarrow 3 \rightarrow 1 \rightarrow 0
$$

The last drop is only $1$, because the margin crosses zero and the loss cannot become negative. With one point and a strictly negative margin, the update reduces that point's loss. The rise in Example 1 comes from leaving a boundary, and from the other papers.


```python
# A single wrong point: losses 5, 3, 1, 0. The last drop is 1, not 2, because loss stops at 0.
margin = -5
losses = [5]
for _ in range(3):
    margin = margin + 2
    losses.append(max(0, -margin))
print(losses)
assert losses == [5, 3, 1, 0]
```

## Pitfall — Loss zero is the wrong stopping test

$(2, 1, 0)$ has loss $0$ and still updates, because C's margin is $0$. The stopping test used in this path is "every margin is positive," not "the loss is zero."


## Summary

The perceptron loss is the total amount by which margins fall below zero. The origin always has loss zero, and it can still be a table of boundary mistakes. The update can raise that loss. It stops only when every margin is strictly positive, which is a useful zero of the loss, not the only zero.
