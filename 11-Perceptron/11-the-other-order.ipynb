# 11 The Other Order

| | |
|---|---|
| **Path** | Perceptron |
| **Prerequisite** | 10 |

Return to the three papers. Lesson 07 visited A, then B, then C, and finished at

$$
s = 2x_1 + x_2 - 1
$$

Visit B first, then A, then C. B is the pass at $(1, 2)$, so the first update is

$$
(0, 0, 0) + (1, 2, 1) = (1, 2, 1)
$$

A's score is already

$$
s(A) = 2 + 2 + 1 = 5 > 0
$$

so A is skipped. C has score $1$ and label $-1$, so

$$
(1, 2, 1) - (0, 0, 1) = (1, 2, 0)
$$

C is on the boundary. One more update gives

$$
(1, 2, -1)
$$

The score is

$$
s = x_1 + 2x_2 - 1
$$

The margins are $m(A) = 3$, $m(B) = 4$, and $m(C) = 1$. This is a second legal perceptron.

The new paper $(0, 1)$ now scores

$$
s(0, 1) = 0 + 2 - 1 = 1 > 0
$$

Lesson 05 used the other line and got $0$. The training table did not contain $(0, 1)$. The order of the table decided the answer.


## Example 1 — Three updates, with B in front

### Predict

The pairs are `(1, 2, 1)`, `(1, 2, 0)`, and `(1, 2, -1)`.


```python
# Hand: B first, then C twice. The finish is (1, 2, -1).
papers = [((1, 2), 1, "B"), ((2, 1), 1, "A"), ((0, 0), -1, "C")]
weights = [0, 0, 0]
steps = []
for _ in range(5):
    changed = False
    for (x1, x2), label, name in papers:
        phi = (x1, x2, 1)
        if label * sum(weights[i] * phi[i] for i in range(3)) <= 0:
            weights = [weights[i] + label * phi[i] for i in range(3)]
            steps.append((name, tuple(weights)))
            changed = True
    if not changed:
        break
print(steps)
assert steps == [
    ("B", (1, 2, 1)),
    ("C", (1, 2, 0)),
    ("C", (1, 2, -1)),
]
```

## Example 2 — The two lines disagree on $(0, 1)$

$$
\begin{aligned}
2(0) + 1 - 1 &= 0 \\
0 + 2(1) - 1 &= 1
\end{aligned}
$$

One rule withholds a decision. The other says pass.


```python
# Order A-then-B scores (0, 1) as 0. Order B-then-A scores it as 1.
score_a_first = 2 * 0 + 1 * 1 - 1
score_b_first = 1 * 0 + 2 * 1 - 1
print(score_a_first, score_b_first)
assert score_a_first == 0
assert score_b_first == 1
```

## Pitfall — Both answers are faithful to the training rule

Neither walk has a bug. Each one stopped at a strict separator of A, B, and C. A perceptron does not have a second sentence that says "prefer the line that treats unseen papers the way a person would." If that sentence matters, it has to be added. The next lesson adds a comparison, not by changing the update, but by measuring the street each line leaves open.


## Summary

B before A finishes at $(1, 2, -1)$. A before B finished at $(2, 1, -1)$. Both separate the three papers. On the unseen paper $(0, 1)$ they do not say the same thing.
