The update in two measurements is the same sentence as in one. Augment the point with a $1$,

$$
\varphi = (x_1, x_2, 1)
$$

and add $y\varphi$ when the margin is not positive:

$$
(w_1, w_2, b) \leftarrow (w_1, w_2, b) + y\varphi
$$

Start at $(0, 0, 0)$. Visit A, then B, then C.

A is $(2, 1)$ and passed, so $y = +1$ and $\varphi_A = (2, 1, 1)$. Its margin at the origin is $0$. The first update is

$$
(0, 0, 0) + (2, 1, 1) = (2, 1, 1)
$$

The score is now $s = 2x_1 + x_2 + 1$.


## Example 1 — A's own margin rises by the square length

$$
\lVert\varphi_A\rVert^{2} = 4 + 1 + 1 = 6
$$

The old margin was $0$, so the new margin of A is $6$.

$$
s(A) = 4 + 1 + 1 = 6
$$


```python
# Hand: the first update stores (2, 1, 1), and A's margin becomes 6.
weights = (0, 0, 0)
phi = (2, 1, 1)
label = 1
weights = tuple(weights[i] + label * phi[i] for i in range(3))
margin = label * sum(weights[i] * phi[i] for i in range(3))
print(weights, margin)
assert weights == (2, 1, 1)
assert margin == 6
```

## Example 2 — B becomes correct without being touched

B is $(1, 2)$, also a pass. Under the new weights,

$$
s(B) = 2\cdot 1 + 1\cdot 2 + 1 = 5 > 0
$$

The visit to B changes nothing. A's update already put B on the pass side.

C is $(0, 0)$, a fail. Its score is the bias:

$$
s(C) = 1, \qquad m = (-1)(1) = -1
$$

C is the next update. That update belongs to the next lesson. The point of this one is the skip: a point you do not touch can still become correct.


```python
# B's margin is 5, so B is skipped. C's margin is -1, so C is still wrong.
weights = (2, 1, 1)
margin_b = 1 * (2 * 1 + 1 * 2 + 1)
margin_c = -1 * (2 * 0 + 1 * 0 + 1)
print(margin_b, margin_c)
assert margin_b == 5
assert margin_c == -1
```

## Pitfall — The order "A, then B, then C" is not three updates

B is on the list and B is not an update. An update happens only for a margin that is not positive. After this first step the walk has used one update, not three.


## Summary

The first mistake is A, and the weights become $(2, 1, 1)$. A's margin is $6$. B's margin is already $5$, so B is skipped. C's margin is $-1$, so the walk is not over.
