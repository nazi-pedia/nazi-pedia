# 04 Five Updates on a Line

| | |
|---|---|
| **Path** | Perceptron |
| **Prerequisite** | 03 |

Two points, one measurement:

| Point | $x$ | Label |
|---|---|---|
| fail | $0$ | $-1$ |
| pass | $1$ | $+1$ |

Start at $(w, b) = (0, 0)$. Visit the fail point first, then the pass point, and repeat. Update when the margin is not positive. The augmented points are $\varphi = (0, 1)$ and $\varphi = (1, 1)$.

$$
\begin{aligned}
(0, 0) &\xrightarrow{\text{fail}} (0, -1) \\
(0, -1) &\xrightarrow{\text{pass}} (1, 0) \\
(1, 0) &\xrightarrow{\text{fail}} (1, -1) \\
(1, -1) &\xrightarrow{\text{pass}} (2, 0) \\
(2, 0) &\xrightarrow{\text{fail}} (2, -1)
\end{aligned}
$$

Read the last line as a score:

$$
s = 2x - 1
$$

$$
\begin{aligned}
s(0) &= -1, & (-1)(-1) &= 1 \\
s(1) &= 1, & (1)(1) &= 1
\end{aligned}
$$

Both margins are positive, so the walk stops. This is the rule Lesson 01 knew how to read. The boundary is $x = 1/2$.


## Example 1 — Replay the five updates

### Predict

The cell prints the five pairs above, then the two final margins `(1, 1)`.


```python
# Hand walk: (0, -1), (1, 0), (1, -1), (2, 0), (2, -1). Then both margins are 1.
weight, bias = 0, 0
points = [(0, -1, "fail"), (1, 1, "pass")]
steps = []
for _ in range(6):
    changed = False
    for measurement, label, name in points:
        if label * (weight * measurement + bias) <= 0:
            weight = weight + label * measurement
            bias = bias + label
            steps.append((name, weight, bias))
            changed = True
    if not changed:
        break
margins = [
    label * (weight * measurement + bias) for measurement, label, _ in points
]
print(steps)
print(margins)
assert steps == [
    ("fail", 0, -1),
    ("pass", 1, 0),
    ("fail", 1, -1),
    ("pass", 2, 0),
    ("fail", 2, -1),
]
assert margins == [1, 1]
```

## Example 2 — The other order reaches the same line

Visit the pass point first. The pairs change, and the destination does not.

$$
\begin{aligned}
(0, 0) &\rightarrow (1, 1) \\
(1, 1) &\rightarrow (1, 0) \\
(1, 0) &\rightarrow (1, -1) \\
(1, -1) &\rightarrow (2, 0) \\
(2, 0) &\rightarrow (2, -1)
\end{aligned}
$$

Five updates again. The same score $2x - 1$. On this tiny line, order changes the path and not the finish. Lesson 11 is a plane on which order changes the finish too.


```python
# Pass first. The path differs. The last pair is still (2, -1).
weight, bias = 0, 0
points = [(1, 1), (0, -1)]
steps = []
for _ in range(6):
    changed = False
    for measurement, label in points:
        if label * (weight * measurement + bias) <= 0:
            weight = weight + label * measurement
            bias = bias + label
            steps.append((weight, bias))
            changed = True
    if not changed:
        break
print(steps)
assert len(steps) == 5
assert steps[-1] == (2, -1)
assert steps[0] == (1, 1)
```

## Pitfall — Stopping at the bound is not the same as stopping at a separator

After four updates the weights are $(2, 0)$. The fail point at $x = 0$ has score $0$. The loss of a zero margin is zero, and the point is still not correct. The fifth update is what moves the bias to $-1$. Lesson 14 stays with that distinction.


```python
# After four updates the fail point is on the boundary. The fifth update finishes it.
weight, bias = 2, 0
assert (-1) * (weight * 0 + bias) == 0
weight, bias = 2, -1
assert (-1) * (weight * 0 + bias) == 1
print(weight, bias)
```

## Summary

From zero, these two points take five updates, in either order. The learned score is $2x - 1$. Each known point has margin $1$, and the boundary sits halfway between them, at $x = 1/2$.
