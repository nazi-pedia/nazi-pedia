# 07 Three Updates Reach a Line

| | |
|---|---|
| **Path** | Perceptron |
| **Prerequisite** | 06 |

Continue from $(w_1, w_2, b) = (2, 1, 1)$, visiting A, then B, then C, and repeating that order.

C has margin $-1$. Its augmented point is $(0, 0, 1)$ and its label is $-1$, so the update subtracts $(0, 0, 1)$:

$$
(2, 1, 1) + (-1)(0, 0, 1) = (2, 1, 0)
$$

The bias is now $0$. C's new margin is

$$
(-1)\cdot 0 = 0
$$

C is still not correct. A and B are still correct:

$$
s(A) = 4 + 1 = 5, \qquad s(B) = 2 + 2 = 4
$$

The next pass skips A and skips B. C has margin $0$, so it updates again:

$$
(2, 1, 0) + (-1)(0, 0, 1) = (2, 1, -1)
$$

Now

$$
\begin{aligned}
m(A) &= 4 + 1 - 1 = 4 \\
m(B) &= 2 + 2 - 1 = 3 \\
m(C) &= (-1)(-1) = 1
\end{aligned}
$$

Every margin is positive. The walk stops. The score is the line from Lesson 05,

$$
s = 2x_1 + x_2 - 1
$$


## Example 1 — The three pairs, and then silence

### Predict

The updates are `(2, 1, 1)`, then `(2, 1, 0)`, then `(2, 1, -1)`. The next pass prints no further pair.


```python
# Hand updates: A once, then C twice. B is never the cause of an update.
papers = [((2, 1), 1, "A"), ((1, 2), 1, "B"), ((0, 0), -1, "C")]
weights = [0, 0, 0]
steps = []
for _ in range(5):
    changed = False
    for (x1, x2), label, name in papers:
        phi = (x1, x2, 1)
        margin = label * sum(weights[i] * phi[i] for i in range(3))
        if margin <= 0:
            weights = [weights[i] + label * phi[i] for i in range(3)]
            steps.append((name, tuple(weights)))
            changed = True
    if not changed:
        break
print(steps)
assert steps == [
    ("A", (2, 1, 1)),
    ("C", (2, 1, 0)),
    ("C", (2, 1, -1)),
]
```

## Example 2 — The new paper is still on the line

Training never saw $(0, 1)$. The learned score gives

$$
s(0, 1) = 1 - 1 = 0
$$

The procedure separated the three papers it was given. It did not invent a margin around a paper it was not given.


```python
# The learned rule leaves (0, 1) at score 0.
score = 2 * 0 + 1 * 1 - 1
print(score)
assert score == 0
```

## Pitfall — C needed two updates, and they were not consecutive in spirit

The first update on C moved its margin from $-1$ to $0$. That is progress, and it is not success. A later lesson will say the margin grows by $\lVert\varphi_C\rVert^{2} = 1$ on each of those updates. A square of $1$ cannot jump over a deficit of $1$ all the way to a positive margin.


```python
# C's margin before each of its two updates, then the finished margin.
assert -1 * (2 * 0 + 1 * 0 + 1) == -1
assert -1 * (2 * 0 + 1 * 0 + 0) == 0
assert -1 * (2 * 0 + 1 * 0 - 1) == 1
print(-1, 0, 1)
```

## Summary

The order A, B, C makes three updates: A, then C, then C. B is carried onto the pass side by A's update. The finished weights are $(2, 1, -1)$. C's first correction only reaches the boundary; the second correction is the one that finishes the line.
