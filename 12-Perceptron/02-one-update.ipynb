# 02 One Update

Write the bias as an extra coordinate that is always $1$. The point $x$ becomes

$$
\varphi = (x,\, 1)
$$

and the weights become a pair $(w, b)$. The score is the dot product $(w, b)\cdot\varphi$.

Suppose this point is not yet correct, so its margin is not positive. The **update** adds the label times the augmented point:

$$
(w, b) \leftarrow (w, b) + y\,\varphi
$$

Let $m$ be the margin before the update. The margin of **this same point** after the update is

$$
\begin{aligned}
m' &= y\,\big((w, b) + y\varphi\big)\cdot\varphi \\
&= y(w, b)\cdot\varphi + y^{2}\,\lVert\varphi\rVert^{2} \\
&= m + \lVert\varphi\rVert^{2}
\end{aligned}
$$

Since $y = \pm 1$, the square $y^{2}$ is $1$. The added amount $\lVert\varphi\rVert^{2} = x^{2} + 1$ is positive. One update always moves this point's margin up. It does not always push the margin across zero.


## Example 1 — A point that starts at the origin of the weights

The only point is $x = 1$, $y = +1$. Start from $(w, b) = (0, 0)$. The margin is $0$, so the point is not correct.

$$
\varphi = (1, 1), \qquad \lVert\varphi\rVert^{2} = 2
$$

$$
(w, b) \leftarrow (0, 0) + (1, 1) = (1, 1)
$$

The new margin is $0 + 2 = 2$. One update finishes the job.

### Predict

The cell prints the new weights `(1, 1)` and the new margin `2`.


```python
# Hand: one update from zero gives (1, 1), and the margin rises by 1^2 + 1 = 2.
weight, bias = 0, 0
measurement, label = 1, 1
margin = label * (weight * measurement + bias)
weight = weight + label * measurement
bias = bias + label
new_margin = label * (weight * measurement + bias)
print(weight, bias)
print(margin, new_margin)
assert (weight, bias) == (1, 1)
assert (margin, new_margin) == (0, 2)
```

## Example 2 — A point that is badly wrong

Same point, $x = 1$, $y = +1$, but the weights start at $w = -5$ and $b = 0$.

$$
m = (1)(-5) = -5
$$

Each update adds $\lVert\varphi\rVert^{2} = 2$:

$$
\begin{aligned}
-5 + 2 &= -3 \\
-3 + 2 &= -1 \\
-1 + 2 &= 1
\end{aligned}
$$

Three updates are required. After the first two, the point is still wrong. The weights walk

$$
(-5, 0) \rightarrow (-4, 1) \rightarrow (-3, 2) \rightarrow (-2, 3)
$$

### Predict

The cell prints the three margins after the updates: `-3`, `-1`, and `1`.


```python
# Hand: margins -5, -3, -1, +1. The point is correct only on the last line.
weight, bias = -5, 0
measurement, label = 1, 1
seen = []
for _ in range(3):
    weight = weight + label * measurement
    bias = bias + label
    seen.append(label * (weight * measurement + bias))
print(seen)
print(weight, bias)
assert seen == [-3, -1, 1]
assert (weight, bias) == (-2, 3)
```

## Pitfall — "One update fixes the point" is a special case

It happened in Example 1 because the margin started at $0$ and $2 > 0$. In Example 2 the margin started at $-5$. Adding $2$ three times is what crosses zero. Count the gap, then divide by $x^{2} + 1$.


```python
# How many additions of 2 are needed to climb from -5 to a positive margin?
gap = 5
step = 1 ** 2 + 1
updates = (gap // step) + 1
print(step, updates)
assert step == 2
assert updates == 3
```

## Summary

One update adds $\lVert\varphi\rVert^{2} = x^{2} + 1$ to the margin of the point you updated. The margin moves the right way every time. It reaches the correct side when the added squares finally exceed the old deficit.
