Lesson 04 stopped after five updates. This lesson proves that a separable problem cannot update forever, and it checks the proof on that same line.

Augment each point as $\varphi = (x, 1)$.

$$
\varphi_{\text{fail}} = (0, 1),\quad y = -1
\qquad
\varphi_{\text{pass}} = (1, 1),\quad y = +1
$$

The longer of the two augmented points has square length

$$
R^{2} = 1^{2} + 1^{2} = 2
$$

The finished weights $v = (2, -1)$ satisfy $\lVert v\rVert^{2} = 5$, and both points have

$$
v\cdot(y\varphi) = 1
$$

Divide by the length. The unit vector $u = v/\sqrt{5}$ then has margin

$$
\gamma = u\cdot(y\varphi) = \dfrac{1}{\sqrt{5}}
$$

on both points. The classic bound says the number of updates, starting from zero, is at most

$$
\dfrac{R^{2}}{\gamma^{2}} = \dfrac{2}{1/5} = 10
$$

Five happened. Ten is the promise. The promise is larger than the walk, and it is still a proof that the walk ends.


## The argument

Write $w_k$ for the augmented weights after $k$ mistakes, with $w_0 = 0$. A mistake on $\varphi$ with label $y$ means $y\, w\cdot\varphi \le 0$, and the step is $w \leftarrow w + y\varphi$.

**The projection grows.** Because $u$ itself separates the points by at least $\gamma$,

$$
u\cdot w_{k} \ge k\gamma
$$

On this particular line the inequality is an equality: every mistake adds exactly $1/\sqrt{5}$.

**The length grows slowly.** Expand the square and drop the cross term, which is not positive on a mistake:

$$
\begin{aligned}
\lVert w + y\varphi\rVert^{2}
&= \lVert w\rVert^{2} + 2y\, w\cdot\varphi + \lVert\varphi\rVert^{2} \\
&\le \lVert w\rVert^{2} + R^{2}
\end{aligned}
$$

After $k$ mistakes, $\lVert w_k\rVert^{2} \le k R^{2} = 2k$.

**Cauchy–Schwarz closes the gate.** A unit vector cannot have a projection longer than the vector itself:

$$
(u\cdot w_k)^{2} \le \lVert w_k\rVert^{2}
$$

Chain the two estimates:

$$
(k\gamma)^{2} \le \lVert w_k\rVert^{2} \le k R^{2}
$$

Divide by the positive number $k$:

$$
k \le \dfrac{R^{2}}{\gamma^{2}} = 10
$$

In integers, with $v = (2, -1)$ instead of the unit vector, the same chain is

$$
k^{2} = (v\cdot w_k)^{2} \le 5\,\lVert w_k\rVert^{2} \le 5\cdot 2k = 10k
$$

so again $k \le 10$. At the real stopping time $k = 5$,

$$
5^{2} = 25 = 5\cdot 5 \le 10\cdot 5 = 50
$$

The projection side is exact. The $R^{2}$ side still has room. That room is why the bound says $10$ while the walk stops at $5$.


## Example 1 — Check the five weights against the proof

The pairs from Lesson 04, and what the proof needs at each of them:

| $k$ | $w_k$ | $v\cdot w_k$ | $\lVert w_k\rVert^{2}$ | $2k$ |
|---|---|---|---|---|
| 1 | $(0, -1)$ | $1$ | $1$ | $2$ |
| 2 | $(1, 0)$ | $2$ | $1$ | $4$ |
| 3 | $(1, -1)$ | $3$ | $2$ | $6$ |
| 4 | $(2, 0)$ | $4$ | $4$ | $8$ |
| 5 | $(2, -1)$ | $5$ | $5$ | $10$ |

### Predict

Every row satisfies $v\cdot w_k = k$ and $\lVert w_k\rVert^{2} \le 2k$.


```python
# Each mistake adds exactly 1 to the dot product with (2, -1), and the square length stays within 2k.
weights = [(0, -1), (1, 0), (1, -1), (2, 0), (2, -1)]
separator = (2, -1)
for k, weights_k in enumerate(weights, start=1):
    projection = sum(separator[i] * weights_k[i] for i in range(2))
    square = sum(part * part for part in weights_k)
    print(k, weights_k, projection, square, 2 * k)
    assert projection == k
    assert square <= 2 * k
    assert projection ** 2 <= 5 * square <= 10 * k
assert 5 <= 10
```

## Pitfall — The bound counts mistakes, not passes through the list

A pass that finds every margin already positive adds nothing to $k$. The five in the table are five mistakes. A later quiet pass is not a sixth mistake, and it does not threaten the bound.


```python
# The sixth look finds two positive margins, so k stays 5, which is inside the bound 10.
weight, bias = 2, -1
margins = [(-1) * (weight * 0 + bias), (1) * (weight * 1 + bias)]
print(margins)
assert margins == [1, 1]
assert 5 <= 10
```

## Summary

If some unit weight vector separates the augmented points by a margin $\gamma > 0$, and no augmented point is longer than $R$, then the number of updates from zero is at most $R^{2}/\gamma^{2}$. On this line that ceiling is $10$, and the walk used $5$. The projection onto $(2, -1)$ grew by exactly $1$ at every mistake.
