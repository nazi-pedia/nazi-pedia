# 01 Read a Score

| **Prerequisite** | A notebook that runs, and arithmetic. |

A perceptron answers one yes-or-no question with a straight score.

It keeps a **weight** $w$ and a **bias** $b$. A new measurement is a number $x$. The score is

$$
s = wx + b
$$

The sign of the score is the answer.

- $s > 0$ means pass.
- $s < 0$ means fail.
- $s = 0$ means the point is on the boundary. The perceptron does not call that a pass.

In the arithmetic, pass is the label $+1$ and fail is the label $-1$. The **margin** of a point is the label times the score,

$$
m = y\, s = y(wx + b)
$$

The point is correct only when $m > 0$. A positive margin means the sign of the score agrees with the label.

## How this path teaches

1. State the rule and work the numbers by hand.
2. A code cell repeats those numbers.
3. The printed value and the hand value have to agree.
4. One pitfall, then the sentence worth keeping.

No perceptron library is used. The arithmetic is the lesson.


## Example 1 — A finished rule on two known points

One measurement, already learned:

$$
s = 2x - 1
$$

The point $x = 0$ failed, so $y = -1$. The point $x = 1$ passed, so $y = +1$.

$$
\begin{aligned}
s(0) &= -1, & m &= (-1)(-1) = 1 \\
s(1) &= 1, & m &= (1)(1) = 1
\end{aligned}
$$

Both margins are positive. Both points are correct. The boundary is the place where the score is zero:

$$
2x - 1 = 0 \qquad \Rightarrow \qquad x = \dfrac{1}{2}
$$

### Predict

The cell prints the two scores, the two margins, and the boundary $0.5$.


```python
# Hand: s(0) = -1, s(1) = 1, both margins 1, boundary at 1/2.
weight = 2
bias = -1
score_at_zero = weight * 0 + bias
score_at_one = weight * 1 + bias
margin_fail = (-1) * score_at_zero
margin_pass = (1) * score_at_one
boundary = -bias / weight
print(score_at_zero, score_at_one)
print(margin_fail, margin_pass)
print(boundary)
assert (score_at_zero, score_at_one) == (-1, 1)
assert (margin_fail, margin_pass) == (1, 1)
assert boundary == 0.5
```

## Example 2 — A new measurement the rule has never seen

Try $x = 2$.

$$
s(2) = 4 - 1 = 3 > 0
$$

The rule says pass. Nothing in the training table was equal to 2. The score still has an answer, because a line continues.

### Predict

The cell prints `3`.


```python
# Hand: 2*2 - 1 = 3, which is a pass.
score = 2 * 2 + (-1)
print(score)
assert score == 3
assert score > 0
```

## Pitfall — Zero is not a quiet pass

At the boundary $x = 1/2$,

$$
s = 2\cdot\dfrac{1}{2} - 1 = 0
$$

The margin is also $0$. The loss formulas later in the path treat a zero margin gently. The decision rule does not. A point on the boundary is not yet correct, and a prediction is not yet a pass.


```python
# The boundary score is 0. Correct means the margin is strictly positive.
score = 2 * (1 / 2) + (-1)
margin = 1 * score
print(score, margin)
assert score == 0
assert margin == 0
assert not margin > 0
```

## Summary

The perceptron turns one measurement into the score $wx + b$. Pass and fail are the two signs. A labeled point is correct only when $y(wx + b) > 0$. The next lesson changes $w$ and $b$ on purpose.
