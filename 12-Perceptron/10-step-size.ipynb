A step size $\eta > 0$ scales the update:

$$
w \leftarrow w + \eta\, y\,\varphi
$$

Lesson 04 used $\eta = 1$. Take the same two points, the same order, and $\eta = 1/2$.

The claim is: if both walks start at zero, then after the same sequence of mistakes the half-step weights are exactly half of the unit-step weights. The scores are half as well. A positive score stays positive, a negative score stays negative, and a zero stays zero. So the walks choose the same points, and the claim survives the next mistake.

Half of the old finish $(2, -1)$ is

$$
\left(1,\, -\dfrac{1}{2}\right)
$$

The score is $s = x - 1/2$. The boundary is still $x = 1/2$, because

$$
x - \dfrac{1}{2} = 0 \qquad\text{and}\qquad 2x - 1 = 0
$$

are the same equation.


## Example 1 — Five half-steps

$$
\begin{aligned}
(0, 0) &\rightarrow \left(0, -\dfrac{1}{2}\right) \\
&\rightarrow \left(\dfrac{1}{2}, 0\right) \\
&\rightarrow \left(\dfrac{1}{2}, -\dfrac{1}{2}\right) \\
&\rightarrow (1, 0) \\
&\rightarrow \left(1, -\dfrac{1}{2}\right)
\end{aligned}
$$

### Predict

The cell prints those five pairs and then the two margins `1/2` and `1/2`.


```python
# Hand: eta = 1/2 copies the five-update walk at half scale. Both final margins are 1/2.
from fractions import Fraction

weight, bias = Fraction(0), Fraction(0)
step = Fraction(1, 2)
points = [(0, -1), (1, 1)]
steps = []
for _ in range(6):
    changed = False
    for measurement, label in points:
        if label * (weight * measurement + bias) <= 0:
            weight = weight + step * label * measurement
            bias = bias + step * label
            steps.append((weight, bias))
            changed = True
    if not changed:
        break
margins = [label * (weight * measurement + bias) for measurement, label in points]
print(steps)
print(margins)
assert steps[-1] == (Fraction(1), Fraction(-1, 2))
assert margins == [Fraction(1, 2), Fraction(1, 2)]
assert len(steps) == 5
```

## Example 2 — The scores really are half

Compare $2x - 1$ with $x - 1/2$ at the two training points and at the boundary.


```python
# At every listed x, the half-step score is exactly half of the unit-step score.
from fractions import Fraction

for measurement in (Fraction(0), Fraction(1, 2), Fraction(1), Fraction(2)):
    full = 2 * measurement - 1
    half = measurement - Fraction(1, 2)
    print(measurement, full, half)
    assert half == full / 2
```

## Pitfall — A smaller step does not find a different line here

It is tempting to read $\eta$ as a way to aim the boundary. On this walk, from zero, $\eta$ only scales the weights. The zero set of the score is unchanged. A different line appears when the order changes, or when the points change, not when this positive step size changes.


## Summary

With $\eta = 1/2$ the five updates still happen, on the same points, and the weights are half of $(2, -1)$. The boundary stays at $x = 1/2$. From zero, a positive step size scales the score and does not change its sign.
