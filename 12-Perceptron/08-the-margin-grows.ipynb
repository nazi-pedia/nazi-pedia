This is the algebra behind every update in the walk, written once.

The weights and the point both live in the augmented space, $w = (w_1, w_2, b)$ and $\varphi = (x_1, x_2, 1)$. If the margin before the update is

$$
m = y\, w\cdot\varphi
$$

the update $w \leftarrow w + y\varphi$ produces

$$
\begin{aligned}
m' &= y(w + y\varphi)\cdot\varphi \\
&= y\, w\cdot\varphi + (y\varphi)\cdot\varphi \\
&= m + y^{2}\,\lVert\varphi\rVert^{2} \\
&= m + \lVert\varphi\rVert^{2}
\end{aligned}
$$

The other points are not in this identity. Their margins can rise or fall. Only the point you just touched is guaranteed to gain $\lVert\varphi\rVert^{2}$.


## Example 1 — A's square is 6

$$
\varphi_A = (2, 1, 1), \qquad \lVert\varphi_A\rVert^{2} = 6
$$

A started at margin $0$ and moved to margin $6$. The identity says $0 + 6 = 6$.


```python
# A's augmented square is 4 + 1 + 1 = 6, which is exactly the new margin.
square = 2 ** 2 + 1 ** 2 + 1 ** 2
old_margin = 0
new_margin = 1 * (2 * 2 + 1 * 1 + 1)
print(square, new_margin)
assert square == 6
assert new_margin == old_margin + square
```

## Example 2 — C's square is 1, so C takes two updates

$$
\varphi_C = (0, 0, 1), \qquad \lVert\varphi_C\rVert^{2} = 1
$$

C is first updated from margin $-1$:

$$
-1 + 1 = 0
$$

It is updated again from margin $0$:

$$
0 + 1 = 1
$$

The identity holds both times. The first result is still not a success.


```python
# Each update on C adds exactly 1 to C's margin: -1, then 0, then 1.
square = 0 ** 2 + 0 ** 2 + 1 ** 2
margins = [-1]
margins.append(margins[-1] + square)
margins.append(margins[-1] + square)
print(square, margins)
assert square == 1
assert margins == [-1, 0, 1]
```

## Example 3 — B gains, and B was not the updated point

When A is updated, B's margin goes from $0$ to $5$. The identity does not say "$5$". It only describes A. B happened to land on the correct side because A's step $(2, 1, 1)$ has a positive dot product with B's point $(1, 2, 1)$:

$$
(2, 1, 1)\cdot(1, 2, 1) = 5
$$


```python
# B's gain of 5 is the dot product of the two augmented points, not B's own square.
step = (2, 1, 1)
phi_b = (1, 2, 1)
gained = sum(step[i] * phi_b[i] for i in range(3))
own_square = 1 ** 2 + 2 ** 2 + 1 ** 2
print(gained, own_square)
assert gained == 5
assert own_square == 6
```

## Pitfall — A rising margin on the touched point can push another point the wrong way

C's update subtracts $(0, 0, 1)$. Every point's score drops by $1$, because every augmented point ends in $1$. A's margin falls from $6$ to $5$, and B's falls from $5$ to $4$. Both stay positive, so the damage is affordable. Lesson 15 meets a point whose update is not affordable.


```python
# Subtracting the bias coordinate lowers A from 6 to 5 and B from 5 to 4.
assert 6 - 1 == 5
assert 5 - 1 == 4
print(5, 4)
```

## Summary

The updated point gains exactly the square of its augmented length. A, with square $6$, is finished in one update from zero. C, with square $1$, needs two updates to climb from $-1$ through $0$ to $1$. Other points receive whatever dot product that step happens to give them.
