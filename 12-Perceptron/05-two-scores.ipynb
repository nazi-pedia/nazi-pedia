# 05 Two Scores, One Line

| | |
|---|---|
| **Path** | Perceptron |
| **Prerequisite** | 04 |

A point can have two measurements. Call them $(x_1, x_2)$. The perceptron keeps two weights and a bias:

$$
s = w_1 x_1 + w_2 x_2 + b
$$

The boundary is the line where the score is zero. The vector $(w_1, w_2)$ points toward the pass side, because moving that way increases the score.

Three practice papers:

| Paper | Scores | Label |
|---|---|---|
| A | $(2, 1)$ | pass, $+1$ |
| B | $(1, 2)$ | pass, $+1$ |
| C | $(0, 0)$ | fail, $-1$ |

Here is one legal rule. Lessons 06 and 07 will grow it from zero. First, read it.

$$
s = 2x_1 + x_2 - 1
$$


## Example 1 — Score the three papers

$$
\begin{aligned}
s(A) &= 4 + 1 - 1 = 4 \\
s(B) &= 2 + 2 - 1 = 3 \\
s(C) &= -1
\end{aligned}
$$

A and B are positive, so they pass. C is negative, so it fails. The margins are $4$, $3$, and $(-1)(-1) = 1$. Every margin is positive.

### Predict

The cell prints `4`, `3`, and `1`.


```python
# Hand margins of 2x + y - 1: A is 4, B is 3, C is 1.
def margin(x1, x2, label):
    return label * (2 * x1 + x2 - 1)

margins = [margin(2, 1, 1), margin(1, 2, 1), margin(0, 0, -1)]
print(margins)
assert margins == [4, 3, 1]
```

## Example 2 — A new paper on the pass side

A new pair of scores is $(1, 1)$.

$$
s(1, 1) = 2 + 1 - 1 = 2 > 0
$$

The rule says pass.


```python
# Hand: 2 + 1 - 1 = 2, a pass.
score = 2 * 1 + 1 - 1
print(score)
assert score == 2
```

## Example 3 — A new paper on the line

The pair $(0, 1)$ gives

$$
s(0, 1) = 0 + 1 - 1 = 0
$$

The paper sits on the boundary $2x_1 + x_2 = 1$. The rule refuses to call it a pass. Lesson 11 finds a second legal line that does call it a pass. Both lines are perceptrons. They were reached by different orders.


```python
# (0, 1) is on 2x + y = 1. The score is zero, so this is not a pass.
score = 2 * 0 + 1 - 1
print(score)
assert score == 0
```

## Pitfall — A legal line is not the only legal line

$x_1 + 2x_2 - 1 = 0$ is a different line, and it also puts A and B on the pass side and C on the fail side:

$$
s(A) = 3, \qquad s(B) = 4, \qquad s(C) = -1
$$

Reading one line correctly is not a reason to believe the training procedure had only one possible answer.


```python
# The swapped weights (1, 2, -1) are also a strict separator.
def margin(x1, x2, label):
    return label * (1 * x1 + 2 * x2 - 1)

margins = [margin(2, 1, 1), margin(1, 2, 1), margin(0, 0, -1)]
print(margins)
assert margins == [3, 4, 1]
assert all(item > 0 for item in margins)
```

## Summary

Two measurements make a line, $w_1 x_1 + w_2 x_2 + b = 0$. The rule $2x_1 + x_2 - 1$ separates the three papers. The new paper $(1, 1)$ passes. The new paper $(0, 1)$ lands on the line. Another weight pair separates the same three papers and will treat $(0, 1)$ differently.
