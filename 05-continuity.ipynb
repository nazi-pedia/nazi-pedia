# ۰۵ · پیوستگی، فشردگی، مقدار میانی

| | |
|---|---|
| **دوره** | ریاضی عمومی ۱ |
| **پیش‌نیاز** | ۰۴ · حد |

تابع در \(a\) پیوسته است اگر سه چیز با هم برابر باشند:
مقدار \(f(a)\)، حد چپ، و حد راست.
حد می‌تواند باشد و مقدار نباشد. آن وقت ناپیوستگی رفع‌شدنی است:
سوراخ را با تعریف دوبارهٔ یک نقطه پر می‌کنیم و تابع جدید پیوسته می‌شود.
تابع قدیم همان‌جا تعریف‌نشده می‌ماند.

## دو قضیه

**فشردگی.** اگر \(g(x)\le f(x)\le h(x)\) نزدیک \(a\) و حد \(g\) و \(h\) هر دو \(L\) باشد، حد \(f\) هم \(L\) است.

**مقدار میانی.** اگر \(f\) روی \([a,b]\) پیوسته باشد و \(k\) بین \(f(a)\) و \(f(b)\) باشد،
دست‌کم یک \(c\) در \((a,b)\) هست که \(f(c)=k\).
قضیه ریشه را نشان می‌دهد، جای ریشه را نه.



```python
import sympy as sp

def check(found, hand, label="نتیجه"):
    found_s = sp.simplify(found)
    hand_s = sp.simplify(hand)
    difference = sp.simplify(found_s - hand_s)
    print(label)
    print("  حل دستی:", hand_s)
    print("  پایتون:", found_s)
    print("  اختلاف:", difference)
    assert difference == 0
    print("  تایید شد")

```

## مثال ۱ — سه نوع شکست

### حل ریاضی

\[
f(x)=\frac{x^2-1}{x-1}
\]

در \(x=1\) تعریف نشده، ولی حد ۲ است. ناپیوستگی رفع‌شدنی.

\[
g(x)=\frac{1}{x}
\]

حد چپ و راست در ۰ بی‌نهایت‌های مخالف‌اند. ناپیوستگی نامتناهی.

\[
h(x)=\begin{cases}0&x<0\\1&x\ge 0\end{cases}
\]

حد چپ ۰ و حد راست ۱ است. ناپیوستگی جهشی. مقدار \(h(0)=1\) فقط سمت راست را نجات می‌دهد.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = (x**2 - 1)/(x - 1)
assert sp.limit(f, x, 1) == 2
left = sp.limit(1/x, x, 0, dir="-")
right = sp.limit(1/x, x, 0, dir="+")
print("1/x از چپ و راست:", left, right)
assert left == -sp.oo and right == sp.oo
def h(value):
    return 0 if value < 0 else 1

left_samples = [h(-10**(-k)) for k in (1, 2, 3)]
right_samples = [h(10**(-k)) for k in (1, 2, 3)]
print("نمونه‌های چپ:", left_samples)
print("نمونه‌های راست:", right_samples)
print("h(0) =", h(0))
assert left_samples == [0, 0, 0]
assert right_samples == [1, 1, 1]
assert h(0) == 1
print("جهش تایید شد")

```

## مثال ۲ — پر کردن سوراخ

### حل ریاضی

تعریف تازه:

\[
F(x)=\begin{cases}
\dfrac{x^2-1}{x-1}&x\neq 1\\
2&x=1
\end{cases}
\]

برای \(x\neq 1\)، \(F(x)=x+1\)، و \(F(1)=2\) همان حد است.
پس \(F\) همه‌جا با \(x+1\) برابر است و پیوسته است.
\(f\) و \(F\) دو تابع‌اند.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
def F(value):
    if value == 1:
        return 2
    return (value**2 - 1)/(value - 1)

print("F(1) =", F(1))
print("حد =", sp.limit((x**2 - 1)/(x - 1), x, 1))
print("F(4) =", F(4))
assert F(1) == 2
assert sp.limit((x**2 - 1)/(x - 1), x, 1) == 2
assert F(4) == 5
print("تایید شد")

```

## مثال ۳ — قضیه مقدار میانی، فقط وجود

### حل ریاضی

\[
f(x)=x^3+x-1
\]

روی \([0,1]\) پیوسته است. \(f(0)=-1<0\) و \(f(1)=1>0\).
صفر بین این دو مقدار است، پس ریشه‌ای در \((0,1)\) وجود دارد.
قضیه مختص آن ریشه را نمی‌دهد.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = x**3 + x - 1
print("f(0) =", f.subs(x, 0), " f(1) =", f.subs(x, 1))
root = sp.nsolve(f, x, sp.Rational(1, 2))
print("یک ریشه عددی:", root)
assert f.subs(x, 0) < 0 < f.subs(x, 1)
assert 0 < root < 1
print("تایید شد")

```

## مثال ۴ — فشردگی

### حل ریاضی

\[
-x^2\le x^2\sin\frac{1}{x}\le x^2,\qquad x\neq 0
\]

چون \(-1\le\sin(1/x)\le 1\).
حد دو دیواره در ۰ برابر ۰ است، پس

\[
\lim_{x\to 0}x^2\sin\frac{1}{x}=0
\]

خود \(\sin(1/x)\) حد ندارد. ضریب \(x^2\) آن را خفه می‌کند.
در \(x=0\) تابع اصلی تعریف نشده؛ حد دارد.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = x**2 * sp.sin(1/x)
exact = sp.limit(f, x, 0)
print("حد:", exact)
assert exact == 0
print("تایید شد")

```

## مثال ۵ — پیوستگی چندضابطه‌ای در نقطهٔ دوخت

### حل ریاضی

\[
p(x)=\begin{cases}
\dfrac{\sin x}{x}&x\neq 0\\
1&x=0
\end{cases}
\]

حد استاندارد \(\sin x/x\) برابر ۱ است و مقدار تعریف‌شده هم ۱ است.
پس \(p\) در ۰ پیوسته است. اگر به‌جای ۱ عدد دیگری می‌گذاشتیم، حد همان ۱ می‌ماند و پیوستگی می‌شکست.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
assert sp.limit(sp.sin(x)/x, x, 0) == 1
print("حد sin(x)/x =", sp.limit(sp.sin(x)/x, x, 0))
print("مقدار تعریف‌شده در صفر = 1")
print("تایید شد")

```

## نکته

پیوستگی خاصیت نقطه است، نه خاصیت فرمول.
یک فرمول می‌تواند همه‌جا جز یک نقطه پیوسته باشد.
قضیه مقدار میانی روی بازهٔ باز یا روی تابع ناپیوسته وعده نمی‌دهد.

## جمع‌بندی

سه عدد را با هم بسنجید: حد چپ، حد راست، مقدار.
رفع‌شدنی یعنی حد هست و مقدار نیست یا مقدار غلط است.
فشردگی وقتی به کار می‌آید که عبارت میان دو چیز با حد معلوم گیر کرده باشد.
