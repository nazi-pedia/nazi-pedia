## Import Required Modules

```python
from sympy import symbols, diff, limit, sin, cos, exp, log, sqrt
```

## Define a Symbol and Function

```python
x = symbols('x')          # Define x as a symbolic variable
f = x**2                  # Define the function f(x) = x^2
```

## Compute the Derivative Using SymPy's diff

```python
derivative = diff(f, x)
print("Derivative of x^2:", derivative)
```

## Example 1: Basic Polynomial

$$ f(x)=x^3+2x^2+x+1 $$

```python
f = x**3 + 2*x**2 + x + 1
derivative = diff(f, x)
print("Derivative:", derivative)
```

## Example 2: Higher-Order Derivative

$$ f(x)=x^4+3x^3-5x^2+7 $$

```python
f = x**4 + 3*x**3 - 5*x**2 + 7
second_derivative = diff(f, x, 2)
print("Second Derivative:", second_derivative)
```

## Example 3: Partial Derivative

$$ f(x,y)=x^2y+y^3\Rightarrow \frac{\partial f}{\partial x}=? $$

```python
y = symbols('y')
f = x**2 * y + y**3
partial_derivative = diff(f, x)
print("Partial Derivative with respect to x:", partial_derivative)
```

## Example 4: Derivative at a Point

$$ f(x)=x^2+3x+5\Rightarrow \acute{f}(2)=? $$

```python
f = x**2 + 3*x + 5
derivative = diff(f, x)
value_at_point = derivative.subs(x, 2)
print("Derivative at x=2:", value_at_point)
```

## Example 5: Trigonometric and Exponential Functions

$$ f(x)=sin(x)\times e^{x} $$

```python
f = sin(x) * exp(x)
derivative = diff(f, x)
print("Derivative of sin(x)*exp(x):", derivative)
```
