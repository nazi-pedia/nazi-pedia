## 1. Slope of a secant, then the limit

**See.** The secant through \((a, f(a))\) and \((a+h, f(a+h))\) has slope

\[
\frac{f(a+h) - f(a)}{h}
\]

The derivative \(f'(a)\) is the limit of that quotient as \(h \to 0\). At \(h = 0\) the quotient is \(0/0\).
Lesson 006 is the reason we do not substitute.

**Do.** Take \(f(x) = x^2\) at \(a = 3\). By algebra the quotient is \(6 + h\), whose limit is 6.
A numerical table and `sp.diff` must both say 6.


```python
import sympy as sp

x = sp.symbols("x")
h = sp.symbols("h")
f = x ** 2
a = 3
quotient = (f.subs(x, a + h) - f.subs(x, a)) / h
print("difference quotient, simplified:", sp.simplify(quotient))
print("limit as h -> 0:", sp.limit(quotient, h, 0))
print("sympy diff at x, then at 3:", sp.diff(f, x), sp.diff(f, x).subs(x, a))

assert sp.limit(quotient, h, 0) == 6
assert sp.diff(f, x).subs(x, a) == 6

print("numerical slopes")
for step in (0.1, 0.01, 0.001):
    slope = ((a + step) ** 2 - a ** 2) / step
    print(f"h = {step:<6} slope = {slope:.6f}")
```

**Read the output.** The simplified quotient is \(6 + h\). No limit was required after that algebra,
but the limit is what the definition asks for. The numerical slopes approach 6 and never need to hit it at a finite \(h\).


## 2. Polynomials, including a second derivative

**Name.** `sp.diff(f, x)` is the first derivative. `sp.diff(f, x, 2)` differentiates twice.
You can also write `sp.diff(f, x, x)`.

**Do.**

\[
f(x) = x^3 + 2x^2 + x + 1, \qquad f'(x) = 3x^2 + 4x + 1
\]

\[
g(x) = x^4 + 3x^3 - 5x^2 + 7, \qquad g''(x) = 12x^2 + 18x - 10
\]

Check each result by differentiating a second way: expand `sp.diff` and compare to the hand formula with `sp.simplify(... ) == 0`.


```python
import sympy as sp

x = sp.symbols("x")

f = x ** 3 + 2 * x ** 2 + x + 1
f_prime = sp.diff(f, x)
print("f' =", f_prime)
assert sp.simplify(f_prime - (3 * x ** 2 + 4 * x + 1)) == 0

g = x ** 4 + 3 * x ** 3 - 5 * x ** 2 + 7
g_second = sp.diff(g, x, 2)
print("g'' =", g_second)
assert sp.simplify(g_second - (12 * x ** 2 + 18 * x - 10)) == 0
print("hand formulas agree with SymPy")
```

## 3. The value at a point, and the tangent line

**See.** \(f'(2)\) is a number: the slope of the tangent line at \(x = 2\).
It is not a function. The call `subs` is what turns the function \(f'\) into that number.

**Do.** \(f(x) = x^2 + 3x + 5\). Then \(f'(x) = 2x + 3\), so \(f'(2) = 7\).
\(f(2) = 15\). The tangent line is

\[
y - 15 = 7(x - 2), \qquad y = 7x + 1
\]

The picture shows the curve and the line. The second check is algebraic: at \(x = 2\) both equal 15, and the line's slope is 7.


```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

x = sp.symbols("x")
f = x ** 2 + 3 * x + 5
slope_function = sp.diff(f, x)
slope = slope_function.subs(x, 2)
height = f.subs(x, 2)
print("f'(2) =", slope, " f(2) =", height)
assert slope == 7
assert height == 15

tangent = slope * (x - 2) + height
print("tangent:", sp.expand(tangent))
assert sp.expand(tangent) == 7 * x + 1

xs = np.linspace(-1, 4, 300)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(xs, xs ** 2 + 3 * xs + 5, color="#1f4e79", linewidth=2, label=r"$f(x)$")
ax.plot(xs, 7 * xs + 1, color="#b85c38", linewidth=2, label="tangent at x = 2")
ax.scatter([2], [15], color="black", zorder=3)
ax.set_title(r"Tangent to $x^2 + 3x + 5$ at $x = 2$ has slope 7")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

## 4. A product, checked by the product rule

**Name.** \((uv)' = u'v + uv'\).

**Do.** \(f(x) = \sin(x)\, e^{x}\).
The product rule gives \(e^{x}\sin x + e^{x}\cos x = e^{x}(\sin x + \cos x)\).
Ask SymPy, then simplify the difference. Zero means the two expressions are the same.


```python
import sympy as sp

x = sp.symbols("x")
f = sp.sin(x) * sp.exp(x)
derivative = sp.diff(f, x)
by_hand = sp.exp(x) * (sp.sin(x) + sp.cos(x))
print(derivative)
print(sp.simplify(derivative))
assert sp.simplify(derivative - by_hand) == 0
print("product rule agrees")
```

### Own it

Differentiate \(x e^{x}\) by the product rule. Lesson 008 will integrate that same function,
and the check there is "differentiate the antiderivative and get this back."


```python
import sympy as sp

x = sp.symbols("x")
derivative = sp.diff(x * sp.exp(x), x)
by_hand = sp.exp(x) * (x + 1)
print(sp.simplify(derivative))
assert sp.simplify(derivative - by_hand) == 0
print("checkpoint passed")
```

## 5. Extension — a partial derivative

Everything above is single-variable calculus. You can skip this section and go to lesson 010 or 008.

If \(f\) depends on \(x\) and \(y\), the partial derivative with respect to \(x\) treats \(y\) as a constant.

\[
f(x, y) = x^2 y + y^3, \qquad \frac{\partial f}{\partial x} = 2xy
\]


```python
import sympy as sp

x, y = sp.symbols("x y")
f = x ** 2 * y + y ** 3
partial = sp.diff(f, x)
print(partial)
assert partial == 2 * x * y
print("partial derivative agrees")
```

## Workshop — definition against `diff` at three points

For \(f(x) = x^3 + 2x^2 + x + 1\), compute \(f'(a)\) from the limit definition and from `sp.diff`
at \(a = -1, 0, 2\). The two numbers must match at each point.


```python
import sympy as sp

x = sp.symbols("x")
h = sp.symbols("h")
f = x ** 3 + 2 * x ** 2 + x + 1
fast = sp.diff(f, x)

for a in (-1, 0, 2):
    quotient = (f.subs(x, a + h) - f.subs(x, a)) / h
    from_definition = sp.limit(sp.simplify(quotient), h, 0)
    from_diff = fast.subs(x, a)
    print(f"a = {a}: definition = {from_definition}, diff = {from_diff}")
    assert sp.simplify(from_definition - from_diff) == 0

print("workshop passed")
```

## Mistakes this lesson is meant to prevent

- Substituting \(h = 0\) into the difference quotient before simplifying.
- Reporting \(f'(x)\) when the question asked for the number \(f'(2)\).
- Differentiating a product term by term without the product rule. The derivative of \(\sin x \cdot e^{x}\) is not \(\cos x \cdot e^{x}\).
- Trusting a plotted tangent line whose slope was typed by hand and never checked against `diff`.
