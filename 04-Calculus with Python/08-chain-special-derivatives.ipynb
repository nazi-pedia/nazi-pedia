When the output of one function is the input of another, the slopes multiply.

$$
\begin{aligned}
 \bigl(f(g(x))\bigr)' &= f'(g(x))\,g'(x).
\end{aligned}
$$

Memorize the table in this lesson, and write the chain out separately each time.

$$
\begin{aligned}
 (\sin u)' &= \cos u\cdot u', \\
 (\cos u)' &= -\sin u\cdot u', \\
 (e^u)' &= e^u\cdot u', \\
 (\ln u)' &= \dfrac{1}{u}\,u', \\
 (\arctan u)' &= \dfrac{1}{1+u^2}\,u', \\
 (\arcsin u)' &= \dfrac{1}{\sqrt{1-u^2}}\,u'.
\end{aligned}
$$


```python
import sympy as sp

# Compare a computed result with the hand result.
def check(found, hand, label="result"):
    # Simplify both sides, then their difference.
    found_s = sp.simplify(found)
    hand_s = sp.simplify(hand)
    difference = sp.simplify(found_s - hand_s)
    print(label)
    print("  hand solution:", hand_s)
    print("  Python:", found_s)
    print("  difference:", difference)
    # The assert checks that the simplified expressions agree.
    assert difference == 0
    print("  confirmed")
```

## Example 1 — The sine of a square

### Hand solution

The outside function is sine, and the inside function is $x^2$.

$$
\begin{aligned}
 \dfrac{d}{dx}\sin(x^2) &= \cos(x^2)\cdot 2x.
\end{aligned}
$$

If you leave out the factor $2x$, you have written the derivative of $\sin x$, not the derivative of $\sin(x^2)$.

### Python recheck


```python
import sympy as sp

# Derivative of sin(x**2). Hand result: 2*x*cos(x**2).
x = sp.symbols("x")
found = sp.diff(sp.sin(x**2), x)
hand = 2*x*sp.cos(x**2)
# The assert checks that the two expressions differ by zero.
assert sp.simplify(found - hand) == 0
print(found)
print("confirmed")
```

## Example 2 — An exponential with a coefficient in the exponent

### Hand solution

$$
\begin{aligned}
 \dfrac{d}{dx}e^{3x}
 &= e^{3x}\cdot 3 \\
 &= 3e^{3x}.
\end{aligned}
$$

A coefficient inside the exponent comes back out as a factor after differentiation.
The same rule, with a different inside function, gives

$$
\begin{aligned}
 \dfrac{d}{dx}e^{x^2} &= e^{x^2}\cdot 2x.
\end{aligned}
$$

### Python recheck


```python
import sympy as sp

# Hand results: d/dx exp(3x) = 3*exp(3x), and d/dx exp(x**2) = 2*x*exp(x**2).
x = sp.symbols("x")
assert sp.diff(sp.exp(3*x), x) == 3*sp.exp(3*x)
assert sp.simplify(sp.diff(sp.exp(x**2), x) - 2*x*sp.exp(x**2)) == 0
print("e^(3x) ->", sp.diff(sp.exp(3*x), x))
print("e^(x^2) ->", sp.diff(sp.exp(x**2), x))
print("confirmed")
```

## Example 3 — The logarithm of a polynomial

### Hand solution

$$
\begin{aligned}
 \dfrac{d}{dx}\ln(x^2+1)
 &= \dfrac{1}{x^2+1}\cdot 2x \\
 &= \dfrac{2x}{x^2+1}.
\end{aligned}
$$

The real domain is the whole line, because $x^2+1>0$.
If the inside could be negative, we would record the domain of the logarithm before differentiating.

### Python recheck


```python
import sympy as sp

# x is real. Derivative of log(x**2 + 1). Hand result: 2*x/(x**2 + 1).
x = sp.symbols("x", real=True)
found = sp.diff(sp.log(x**2 + 1), x)
hand = 2*x/(x**2 + 1)
print(sp.simplify(found))
# The assert checks agreement with the hand formula.
assert sp.simplify(found - hand) == 0
print("confirmed")
```

## Example 4 — Arctangent and arcsine

### Hand solution

$$
\begin{aligned}
 \dfrac{d}{dx}\arctan x &= \dfrac{1}{1+x^2}.
\end{aligned}
$$

$$
\begin{aligned}
 \dfrac{d}{dx}\arcsin(2x)
 &= \dfrac{1}{\sqrt{1-(2x)^2}}\cdot 2 \\
 &= \dfrac{2}{\sqrt{1-4x^2}}.
\end{aligned}
$$

The domain of the second function is $|2x|\le 1$, that is $|x|\le \frac{1}{2}$, and at the two endpoints the derivative becomes infinite.

### Python recheck


```python
import sympy as sp

# Derivative of atan(x) is 1/(1 + x**2).
# Derivative of asin(2x). Hand result: 2/sqrt(1 - 4*x**2).
x = sp.symbols("x")
assert sp.diff(sp.atan(x), x) == 1/(1 + x**2)
found = sp.diff(sp.asin(2*x), x)
hand = 2/sp.sqrt(1 - 4*x**2)
print(sp.simplify(found))
# The assert checks the arcsine derivative against the hand formula.
assert sp.simplify(found - hand) == 0
print("confirmed")
```

## Example 5 — The chain rule beside a product

### Hand solution

$$
\begin{aligned}
 f(x) &= e^{x}\sin x.
\end{aligned}
$$

Use the product rule first. Inside it, each factor takes its ordinary derivative.

$$
\begin{aligned}
 f'(x)
 &= e^{x}\sin x+e^{x}\cos x \\
 &= e^{x}(\sin x+\cos x).
\end{aligned}
$$

If the function were $e^{\sin x}$ instead, the chain rule would be required, and the derivative would be $e^{\sin x}\cos x$.
Do not interchange these two functions.

### Python recheck


```python
import sympy as sp

# Product exp(x)*sin(x), and the composite exp(sin(x)).
x = sp.symbols("x")
product = sp.diff(sp.exp(x)*sp.sin(x), x)
chain = sp.diff(sp.exp(sp.sin(x)), x)
# Hand result for the product: exp(x)*(sin(x) + cos(x)).
# Hand result for the chain: exp(sin(x))*cos(x).
assert sp.simplify(product - sp.exp(x)*(sp.sin(x) + sp.cos(x))) == 0
assert sp.simplify(chain - sp.exp(sp.sin(x))*sp.cos(x)) == 0
print("product:", sp.simplify(product))
print("chain:", sp.simplify(chain))
print("confirmed")
```

## A pitfall

Differentiate the outside function and leave the inside function untouched. Then multiply by the derivative of the inside.
Leaving out the derivative of the inside is the most common mistake in this lesson.

## Summary

Name both parts of every composite function: the outside and the inside.
The derivative table supplies the derivative of the outside. The chain rule multiplies by the derivative of the inside.
