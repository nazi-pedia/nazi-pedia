$$
\mathbf{r}(t)=\bigl(x(t),y(t),z(t)\bigr).
$$

The derivative is the velocity vector, and the second derivative is the acceleration.
Speed is the length of the velocity. The length of the curve is the integral of the speed.


## Example 1 — A helix

### Hand solution

$$
\mathbf{r}(t)=(\cos t,\sin t,t).
$$

$$
\begin{aligned}
\mathbf{r}'(t)
&=(-\sin t,\cos t,1), \\
|\mathbf{r}'(t)|
&=\sqrt{2}.
\end{aligned}
$$

The speed is constant. From $0$ to $2\pi$ the length is

$$
\begin{aligned}
L
&=\int_0^{2\pi}\sqrt{2}\,dt
=2\pi\sqrt{2}.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Real parameter.
t = sp.symbols("t", real=True)
# Helix.
r = sp.Matrix([sp.cos(t), sp.sin(t), t])
# Speed |r'(t)|.
speed = sp.simplify(r.diff(t).norm())
# Length from 0 to 2*pi.
length = sp.integrate(speed, (t, 0, 2*sp.pi))
print(speed, length)
assert speed == sp.sqrt(2)
assert length == 2*sp.pi*sp.sqrt(2)
print("Confirmed")
```

## Example 2 — Acceleration of a space parabola

### Hand solution

$$
\mathbf{r}(t)=(t,t^2,0).
$$

$$
\begin{aligned}
\mathbf{v}(t)
&=(1,2t,0), \\
\mathbf{a}(t)
&=(0,2,0).
\end{aligned}
$$

The acceleration is constant and points in the positive $y$-direction of the $xy$-plane. This is the same statement as $y''=2$ for $y=t^2$.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Parameter.
t = sp.symbols("t")
# Space parabola in the xy-plane.
r = sp.Matrix([t, t**2, 0])
# Velocity is (1, 2t, 0).
assert r.diff(t) == sp.Matrix([1, 2*t, 0])
# Acceleration is the constant vector (0, 2, 0).
assert r.diff(t, 2) == sp.Matrix([0, 2, 0])
print(r.diff(t), r.diff(t, 2))
print("Confirmed")
```

## Example 3 — Projection of the velocity onto the tangent

### Hand solution

On the helix of Example 1 the velocity is never zero, so the curve is smooth.
The unit tangent vector is

$$
\dfrac{\mathbf{r}'(t)}{|\mathbf{r}'(t)|}
=\dfrac{(-\sin t,\cos t,1)}{\sqrt{2}}.
$$

At $t=0$,

$$
\begin{aligned}
\left.\dfrac{\mathbf{r}'(t)}{|\mathbf{r}'(t)|}\right|_{t=0}
&=\dfrac{(0,1,1)}{\sqrt{2}}.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Parameter.
t = sp.symbols("t")
# Helix from Example 1.
r = sp.Matrix([sp.cos(t), sp.sin(t), t])
# Unit tangent r'(t) / |r'(t)|.
tangent = sp.simplify(r.diff(t)/r.diff(t).norm())
print(tangent.subs(t, 0))
# At t = 0 the unit tangent is (0, 1, 1) / sqrt(2).
assert tangent.subs(t, 0) == sp.Matrix([0, 1, 1])/sp.sqrt(2)
print("Confirmed")
```

## A pitfall

Differentiation of a vector function is component by component.
The length of the derivative is not the derivative of the length. The speed can stay constant while the direction of the velocity changes. A circle is the standard case.

## Summary

$\mathbf{r}'$ is the velocity, $|\mathbf{r}'|$ is the speed, and the integral of the speed is the length of the path.
