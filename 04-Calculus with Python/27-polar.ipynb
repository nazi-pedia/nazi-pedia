$$
x=r\cos\theta,\qquad y=r\sin\theta.
$$

The area enclosed by $r=r(\theta)$ from $\alpha$ to $\beta$ is

$$
A=\dfrac{1}{2}\int_{\alpha}^{\beta}r(\theta)^2\,d\theta.
$$

The arc length is

$$
L=\int_{\alpha}^{\beta}\sqrt{r^2+\left(\dfrac{dr}{d\theta}\right)^2}\,d\theta.
$$

The slope in Cartesian coordinates is still $\dfrac{dy}{dx}$. Differentiate both $x$ and $y$ with respect to $\theta$ before you divide.


## Example 1 — A circle of radius 2

### Hand solution

Take $r=2$.

$$
\begin{aligned}
A
&=\dfrac{1}{2}\int_0^{2\pi}4\,d\theta
=4\pi.
\end{aligned}
$$

Convert the point with $\theta=\dfrac{\pi}{3}$ and $r=2$:

$$
\begin{aligned}
x
&=2\cdot\dfrac{1}{2}
=1, \\
y
&=2\cdot\dfrac{\sqrt{3}}{2}
=\sqrt{3}.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Polar angle.
theta = sp.symbols("theta")
# Area of the circle r = 2.
area = sp.Rational(1, 2)*sp.integrate(2**2, (theta, 0, 2*sp.pi))
# Cartesian coordinates at theta = pi/3.
x = 2*sp.cos(sp.pi/3)
y = 2*sp.sin(sp.pi/3)
print(area, x, y)
assert area == 4*sp.pi
assert x == 1 and sp.simplify(y - sp.sqrt(3)) == 0
print("Confirmed")
```

## Example 2 — A cardioid

### Hand solution

Take $r=1+\cos\theta$.

$$
\begin{aligned}
r^2
&=1+2\cos\theta+\cos^2\theta
=1+2\cos\theta+\dfrac{1+\cos 2\theta}{2} \\
&=\dfrac{3}{2}+2\cos\theta+\dfrac{1}{2}\cos 2\theta, \\
A
&=\dfrac{1}{2}\int_0^{2\pi}r^2\,d\theta
=\dfrac{1}{2}\cdot\dfrac{3}{2}\cdot 2\pi
=\dfrac{3\pi}{2}.
\end{aligned}
$$

The integrals of the cosines over a full period are zero.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Polar angle.
theta = sp.symbols("theta")
# Cardioid r = 1 + cos(theta).
r = 1 + sp.cos(theta)
# Area over one full turn.
area = sp.simplify(sp.Rational(1, 2)*sp.integrate(r**2, (theta, 0, 2*sp.pi)))
print(area)
assert area == 3*sp.pi/2
print("Confirmed")
```

## Example 3 — Where is the tangent horizontal?

### Hand solution

Take $r=1+\cos\theta$. The slope is

$$
\dfrac{dy}{dx}
=\dfrac{r'\sin\theta+r\cos\theta}{r'\cos\theta-r\sin\theta},
\qquad r'=-\sin\theta.
$$

At $\theta=\dfrac{\pi}{2}$ one has $r=1$ and $r'=-1$, which is not zero:

$$
\begin{aligned}
r'\sin\theta+r\cos\theta
&=(-1)(1)+(1)(0)
=-1, \\
r'\cos\theta-r\sin\theta
&=(-1)(0)-(1)(1)
=-1, \\
\dfrac{dy}{dx}
&=1.
\end{aligned}
$$

The point is $(0,1)$, but it is not the highest point of the curve.

A horizontal tangent means the numerator is zero and the denominator is not. The numerator is $\cos\theta+\cos 2\theta$:

$$
\begin{aligned}
\cos\theta+\cos 2\theta
&=2\cos^2\theta+\cos\theta-1 \\
&=(2\cos\theta-1)(\cos\theta+1)
=0.
\end{aligned}
$$

One solution is $\theta=\dfrac{\pi}{3}$. There $r=\dfrac{3}{2}$ and $r'=-\dfrac{\sqrt{3}}{2}$. The numerator is zero and the denominator is $-\sqrt{3}$, so the slope is $0$:

$$
\begin{aligned}
r'\sin\theta+r\cos\theta
&=0, \\
r'\cos\theta-r\sin\theta
&=-\sqrt{3}, \\
\dfrac{dy}{dx}
&=0.
\end{aligned}
$$

The angle $\theta=\pi$ also makes the numerator zero, but there the numerator and the denominator both vanish. That point is the cusp of the cardioid, not a horizontal tangent.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Polar angle.
theta = sp.symbols("theta")
# Cardioid.
r = 1 + sp.cos(theta)
# Cartesian coordinates of the cardioid.
x = r*sp.cos(theta)
y = r*sp.sin(theta)
# Derivatives with respect to theta.
dy = sp.diff(y, theta)
dx = sp.diff(x, theta)
# At theta = pi/2 both derivatives equal -1, so the slope is 1.
at_half = {theta: sp.pi/2}
print("At pi/2:", dy.subs(at_half), dx.subs(at_half))
assert dy.subs(at_half) == -1 and dx.subs(at_half) == -1
# At theta = pi/3 the numerator vanishes and the denominator is -sqrt(3).
at_top = {theta: sp.pi/3}
print("At pi/3:", dy.subs(at_top), dx.subs(at_top))
assert sp.simplify(dy.subs(at_top)) == 0
assert sp.simplify(dx.subs(at_top) + sp.sqrt(3)) == 0
print("Confirmed")
```

## Example 4 — Length of the cardioid

### Hand solution

$$
\begin{aligned}
r^2+(r')^2
&=(1+\cos\theta)^2+\sin^2\theta \\
&=1+2\cos\theta+\cos^2\theta+\sin^2\theta \\
&=2+2\cos\theta
=4\cos^2\dfrac{\theta}{2}.
\end{aligned}
$$

On $[0,2\pi]$ the square root must keep the absolute value. The standard result is that the length is $8$: from $0$ to $\pi$, $\cos\dfrac{\theta}{2}\ge 0$, and by symmetry the full length is twice the half-length.

$$
\begin{aligned}
L
&=2\int_0^{\pi}2\cos\dfrac{\theta}{2}\,d\theta
=4\cdot 2\sin\dfrac{\theta}{2}\Big|_0^{\pi}
=8.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Real polar angle.
theta = sp.symbols("theta", real=True)
# Cardioid.
r = 1 + sp.cos(theta)
# Speed in polar coordinates.
speed = sp.simplify(sp.sqrt(r**2 + sp.diff(r, theta)**2))
# Hand speed 2 cos(theta/2).
hand_speed = 2*sp.cos(theta/2)
# The two speeds agree up to sign.
assert sp.simplify(speed**2 - hand_speed**2) == 0
# Length by doubling the integral over the upper half.
length = 2*sp.integrate(hand_speed, (theta, 0, sp.pi))
# Numerical integral over a full turn.
numeric = sp.N(sp.Integral(speed, (theta, 0, 2*sp.pi)))
print(speed, "->", length, "numeric:", numeric)
assert sp.simplify(length - 8) == 0
assert abs(numeric - 8) < 10**(-10)
print("Confirmed")
```

## A pitfall

A negative $r$ in polar coordinates sends the point in the opposite direction.
Before you compute an area, choose an interval that covers the curve exactly once. Tracing the curve twice falsely doubles the area.

## Summary

Polar area is half the integral of $r^2$.
Arc length is the speed in the coordinates $(r,\theta)$.
Bring the slope back to Cartesian form, then substitute the number.
