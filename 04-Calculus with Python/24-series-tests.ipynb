# 24 Series and Convergence Tests

| | |
|---|---|
| **Course** | Calculus II |
| **Prerequisite** | 18 and 23 |

A series is the limit of its partial sums.

$$
\sum_{n=1}^{\infty}a_{n}=\lim_{N\to\infty}\sum_{n=1}^{N}a_{n}.
$$

If $a_{n}$ does not tend to $0$, the series diverges. The converse is false: the harmonic series has terms that tend to $0$ and still diverges.

The geometric series $\sum r^{n}$, starting at $n=0$, equals $\dfrac{1}{1-r}$, exactly when $\lvert r\rvert<1$.

The $p$-series $\sum n^{-p}$ converges exactly when $p>1$.

The ratio test uses

$$
\lim_{n\to\infty}\left\lvert\dfrac{a_{n+1}}{a_{n}}\right\rvert=L.
$$

If $L<1$, the series converges absolutely. If $L>1$, it diverges. If $L=1$, the test is inconclusive.

An alternating series whose absolute values decrease to $0$ converges, and the truncation error is at most the first omitted term.



## Example 1 — Geometric

### Hand solution

$$
\begin{aligned}
\sum_{n=0}^{\infty}\left(\dfrac{1}{2}\right)^{n}
&= \dfrac{1}{1-\dfrac{1}{2}} \\
&= 2.
\end{aligned}
$$

The partial sum $s_{N}=2-2^{-N}$ approaches $2$.

### Python recheck



```python
import sympy as sp

# Exact geometric sum, and the partial sum through n = 10.
n = sp.symbols("n", integer=True, nonnegative=True)
exact = sp.summation((sp.Rational(1, 2))**n, (n, 0, sp.oo))
partial = sp.summation((sp.Rational(1, 2))**n, (n, 0, 10))
print(exact, partial)

# The sum is 2, and s_10 = 2 - 2^(-10).
assert exact == 2
assert partial == 2 - sp.Rational(1, 2)**10
print("Confirmed")

```

## Example 2 — A $p$-series and the number $\pi$

### Hand solution

Since $p=2>1$, the series $\sum n^{-2}$ converges. The exact value, which comes from Fourier series and is accepted here as a closed form, is

$$
\sum_{n=1}^{\infty}\dfrac{1}{n^{2}}=\dfrac{\pi^{2}}{6}.
$$

The comparison integral

$$
\int_{1}^{\infty}x^{-2}\,dx=1
$$

supplies a bound, not the value itself.

### Python recheck



```python
import sympy as sp

# Closed form of the Basel series.
n = sp.symbols("n", integer=True, positive=True)
exact = sp.summation(1/n**2, (n, 1, sp.oo))
print(exact)

# The sum equals pi^2/6.
assert sp.simplify(exact - sp.pi**2/6) == 0
print("Confirmed")

```

## Example 3 — Harmonic

### Hand solution

$$
\begin{aligned}
\sum_{n=1}^{N}\dfrac{1}{n}
&\ge \int_{1}^{N+1}\dfrac{1}{x}\,dx \\
&= \ln(N+1).
\end{aligned}
$$

The right-hand side tends to infinity, so the harmonic series diverges. The general term tends to $0$, and that is not enough to rescue the series.

### Python recheck



```python
import sympy as sp

# The partial harmonic sum is the harmonic number H_N.
n = sp.symbols("n", integer=True, positive=True)
N = sp.symbols("N", integer=True, positive=True)
partial = sp.summation(1/n, (n, 1, N))
print(partial)

# The partial sums tend to infinity.
assert sp.limit(partial, N, sp.oo) == sp.oo
print("Confirmed")

```

## Example 4 — Alternating harmonic

### Hand solution

$$
\begin{aligned}
\sum_{n=1}^{\infty}\dfrac{(-1)^{n+1}}{n}
&= 1-\dfrac{1}{2}+\dfrac{1}{3}-\dfrac{1}{4}+\cdots \\
&= \ln 2.
\end{aligned}
$$

The absolute values of the terms decrease to $0$, so the series converges. It does not converge absolutely, because the series of absolute values is the harmonic series. The convergence is conditional.

### Python recheck



```python
import sympy as sp

# Sum of the alternating harmonic series.
n = sp.symbols("n", integer=True, positive=True)
exact = sp.summation(((-1)**(n + 1))/n, (n, 1, sp.oo))
print(exact)

# The sum is ln 2.
assert sp.simplify(exact - sp.log(2)) == 0
print("Confirmed")

```

## Example 5 — Ratio test on the exponential series

### Hand solution

$$
\begin{aligned}
a_{n} &= \dfrac{1}{n!}, \\
\left\lvert\dfrac{a_{n+1}}{a_{n}}\right\rvert
&= \dfrac{1}{n+1} \\
&\to 0<1.
\end{aligned}
$$

Thus $\sum\dfrac{1}{n!}$ converges absolutely. From $n=0$, its sum is

$$
\sum_{n=0}^{\infty}\dfrac{1}{n!}=e.
$$

### Python recheck



```python
import sympy as sp

# Ratio of successive terms, and the exponential series from n = 0.
n = sp.symbols("n", integer=True, nonnegative=True)
ratio = sp.limit(sp.factorial(n)/sp.factorial(n + 1), n, sp.oo)
exact = sp.summation(1/sp.factorial(n), (n, 0, sp.oo))
print("Ratio:", ratio, "Sum:", exact)

# The ratio limit is 0, and the sum is e.
assert ratio == 0
assert exact == sp.E
print("Confirmed")

```

## A pitfall

The ratio test is inconclusive for a $p$-series, because the ratio tends to $1$. For powers, use the $p$-test or a comparison. For factorials, use the ratio test.

## Summary

First see whether the general term tends to $0$. Then match the term with a geometric series, a $p$-series, the ratio test, or an alternating series. Separate conditional convergence from absolute convergence.
