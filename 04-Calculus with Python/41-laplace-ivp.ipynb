The Laplace transform of a function $f$ is

$$
\mathcal{L}\{f\}(s)=\int_0^{\infty} e^{-st}f(t)\,dt.
$$

Its main property turns differentiation into algebra:

$$
\begin{aligned}
\mathcal{L}\{y'\} &= sY-y(0), \\
\mathcal{L}\{y''\} &= s^2 Y-sy(0)-y'(0).
\end{aligned}
$$

The basic table, for $a>0$ and for $s$ large enough, is

$$
\begin{aligned}
\mathcal{L}\{1\} &= \dfrac{1}{s}, &
\mathcal{L}\{e^{at}\} &= \dfrac{1}{s-a}, \\
\mathcal{L}\{\sin at\} &= \dfrac{a}{s^2+a^2}, &
\mathcal{L}\{\cos at\} &= \dfrac{s}{s^2+a^2}.
\end{aligned}
$$

Transform the initial-value problem, solve the resulting algebra problem for $Y$ by partial fractions, and invert the transform.


## Example 1 — The integral definition

### Hand solution

Compute the transform of the constant function from the definition:

$$
\mathcal{L}\{1\}
=\int_0^{\infty} e^{-st}\,dt
=\dfrac{1}{s},
\qquad s>0.
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Time variable and transform variable, with the usual sign restrictions.
t = sp.symbols("t", nonnegative=True)
s = sp.symbols("s", positive=True)
# Improper integral in the definition of L{1}.
by_integral = sp.integrate(sp.exp(-s*t), (t, 0, sp.oo))
# The same transform from the library. Keep the first entry of the triple.
by_library = sp.laplace_transform(1, t, s)[0]
print(by_integral, by_library)
# Both routes equal 1/s.
assert by_integral == 1/s and by_library == 1/s
print("Confirmed")
```

## Example 2 — A first-order equation

### Hand solution

Solve $y'+y=1$ with $y(0)=0$. Transform both sides and invert:

$$
\begin{aligned}
sY+Y &= \dfrac{1}{s}, \\
Y &= \dfrac{1}{s(s+1)}
   = \dfrac{1}{s}-\dfrac{1}{s+1}, \\
y &= 1-e^{-t}.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Time variable and transform variable.
t = sp.symbols("t", nonnegative=True)
s = sp.symbols("s", positive=True)
# Transform Y = 1/(s(s+1)).
Y = 1/(s*(s + 1))
# Invert the transform.
y = sp.inverse_laplace_transform(Y, s, t)
# Hand solution (1 - exp(-t)) for t > 0, written with the unit step.
hand = (1 - sp.exp(-t))*sp.Heaviside(t)
print(y)
# The inverse transform must match the hand solution.
assert sp.simplify(y - hand) == 0
print("Confirmed")
```

## Example 3 — Oscillation recovered from the Laplace transform

### Hand solution

Solve $y''+y=0$ with $y(0)=1$ and $y'(0)=0$.

$$
\begin{aligned}
s^2 Y-s+Y &= 0, \\
Y &= \dfrac{s}{s^2+1}, \\
y &= \cos t.
\end{aligned}
$$

This is the same solution as in lesson 39. The transform has already absorbed the initial data.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Time and transform variables.
t, s = sp.symbols("t s")
# Transform Y = s/(s^2+1).
Y = s/(s**2 + 1)
# Invert the transform.
y = sp.inverse_laplace_transform(Y, s, t)
print(y)
# The inverse is cos(t) for t > 0.
assert sp.simplify(y - sp.cos(t)*sp.Heaviside(t)) == 0
print("Confirmed")
```

## Example 4 — The exponential entry in the table

### Hand solution

$$
\mathcal{L}\{e^{2t}\}=\dfrac{1}{s-2}.
$$

This is the same exponential growth factor as in lesson 38. The inverse transform of $\dfrac{1}{s-2}$ is $e^{2t}$ for $t>0$.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Time variable and transform variable.
t = sp.symbols("t", nonnegative=True)
s = sp.symbols("s", positive=True)
# Library transform of exp(2t). Keep the first entry of the triple.
found = sp.laplace_transform(sp.exp(2*t), t, s)[0]
print(found)
# The table entry is 1/(s-2).
assert sp.simplify(found - 1/(s - 2)) == 0
print("Confirmed")
```

## A pitfall

The call `laplace_transform` returns a triple: the transform, the half-plane of convergence, and a condition.
The transform is the first component. The algebra for $Y$ uses that component. Feeding the whole triple into a partial-fraction step produces an expression that is no longer a function of $s$ alone.

## Summary

Under the Laplace transform, a derivative becomes multiplication by $s$, with the initial value subtracted.
Partial fractions are the route back to the time domain.
