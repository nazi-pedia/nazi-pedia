A function $f(x,y)$ assigns a number to each allowed point of the plane.
The limit at $(a,b)$ exists only when every path approaching the point produces the same number.
One agreeing path does not prove the limit. Two disagreeing paths refute it.

In polar coordinates,

$$
x=r\cos\theta,\qquad y=r\sin\theta.
$$

If, after that substitution, the expression is a multiple of $r$ times a quantity bounded in $\theta$, then the limit is zero.


## Example 1 — Reading the domain and one value

### Hand solution

$$
f(x,y)=\sqrt{4-x^2-y^2}.
$$

The domain is the disk $x^2+y^2\le 4$. Also $f(0,0)=2$, and $f$ is zero on the boundary.
This is the upper hemisphere of radius $2$.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Real variables.
x, y = sp.symbols("x y", real=True)
# Upper hemisphere of radius 2.
f = sp.sqrt(4 - x**2 - y**2)
# Value at the origin.
assert f.subs({x: 0, y: 0}) == 2
# Value at a boundary point.
assert f.subs({x: 2, y: 0}) == 0
print(f.subs({x: 0, y: 0}))
print("Confirmed")
```

## Example 2 — A limit that really is zero

### Hand solution

$$
\dfrac{x^2 y}{x^2+y^2}.
$$

Substitute $x=r\cos\theta$ and $y=r\sin\theta$:

$$
\begin{aligned}
\dfrac{r^2\cos^2\theta\cdot r\sin\theta}{r^2}
&=r\cos^2\theta\sin\theta, \\
\bigl|\cos^2\theta\sin\theta\bigr|
&\le 1, \\
\bigl|r\cos^2\theta\sin\theta\bigr|
&\le r.
\end{aligned}
$$

As $r\to 0$ the expression tends to $0$, independently of $\theta$:

$$
\lim_{r\to 0} r\cos^2\theta\sin\theta=0.
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Polar coordinates.
r, theta = sp.symbols("r theta", real=True)
# The quotient written in polar form.
polar = sp.simplify((r*sp.cos(theta))**2 * (r*sp.sin(theta)) / ((r*sp.cos(theta))**2 + (r*sp.sin(theta))**2))
print(polar)
# The limit as r -> 0 is 0.
assert sp.limit(polar, r, 0) == 0
print("Confirmed")
```

## Example 3 — Two paths, two answers

### Hand solution

$$
\dfrac{x^2-y^2}{x^2+y^2}.
$$

Along $y=0$ the numerator and the denominator are both $x^2$. Along $x=0$ the numerator is $-y^2$ and the denominator is $y^2$:

$$
\begin{aligned}
\lim_{x\to 0}\dfrac{x^2}{x^2}
&=1, \\
\lim_{y\to 0}\dfrac{-y^2}{y^2}
&=-1.
\end{aligned}
$$

The two-variable limit does not exist.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Variables.
x, y = sp.symbols("x y")
# The two-variable quotient.
f = (x**2 - y**2)/(x**2 + y**2)
# Limit along the x-axis, y = 0.
along_x = sp.limit(f.subs(y, 0), x, 0)
# Limit along the y-axis, x = 0.
along_y = sp.limit(f.subs(x, 0), y, 0)
print(along_x, along_y)
assert along_x == 1 and along_y == -1
print("Confirmed")
```

## Example 4 — A parabolic path is not enough as the only test

### Hand solution

The same quotient as in Example 3, along the line $y=mx$, equals

$$
\begin{aligned}
\dfrac{x^2-(mx)^2}{x^2+(mx)^2}
&=\dfrac{1-m^2}{1+m^2}.
\end{aligned}
$$

The value depends on $m$. Even the family of all straight lines fails to give a single limit.
There is no need to build an exotic path. Two lines that disagree are enough.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Variable and slope of the line.
x, m = sp.symbols("x m")
# The quotient along y = mx.
f = (x**2 - (m*x)**2)/(x**2 + (m*x)**2)
# Limit along that line.
along = sp.simplify(sp.limit(f, x, 0))
print(along)
# The horizontal line gives 1; the line y = x gives 0.
assert along.subs(m, 0) == 1
assert along.subs(m, 1) == 0
print("Confirmed")
```

## A pitfall

Use polar coordinates when the degree of the numerator is higher than the degree of the denominator.
If the degrees are equal, try two paths first.

## Summary

One path does not prove a limit.
Two different paths refute it.
Polar coordinates are a proof when a factor of $r$ stands in front of a bounded expression.
