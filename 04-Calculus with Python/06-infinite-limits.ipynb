Keep the two phrases apart.

An infinite limit means that $x$ approaches a finite number while $|f(x)|$ grows without bound. That is a vertical asymptote.
A limit at $\infty$ means that $x$ itself goes far away and $f(x)$ settles on a finite number. That is a horizontal asymptote.

Infinity is not a real number. Writing $\lim=+\infty$ is a name for a kind of behavior.

If you do not give a direction, SymPy approaches from the right.
Declare a two-sided limit only after you have computed the two sides separately and found that they agree.


```python
import sympy as sp

# Compare a computed SymPy result with the answer found by hand.
def check(found, hand, label="Result"):
    # Simplify each side first so equivalent algebraic forms can match.
    found_s = sp.simplify(found)
    hand_s = sp.simplify(hand)
    difference = sp.simplify(found_s - hand_s)
    print(label)
    print("  Hand:", hand_s)
    print("  Python:", found_s)
    print("  Difference:", difference)
    # The methods agree only when that difference simplifies to zero.
    assert difference == 0
    print("  Confirmed")
```

## Example 1 — Two infinities of opposite sign

### Hand solution

$$
\lim_{x\to 0^+}\dfrac{1}{x}=+\infty,
\qquad
\lim_{x\to 0^-}\dfrac{1}{x}=-\infty.
$$

The two-sided limit does not exist, neither as a number nor as one single $\infty$.
The line $x=0$ is a vertical asymptote.

### Python recheck


```python
import sympy as sp

# 1/x at 0. Hand answers: +infinity from the right, -infinity from the left.
# SymPy's default direction is from the right, so it matches the right-hand limit.
x = sp.symbols("x")
right = sp.limit(1/x, x, 0, dir="+")
left = sp.limit(1/x, x, 0, dir="-")
default = sp.limit(1/x, x, 0)
print("Right:", right)
print("Left:", left)
print("Python default:", default)
assert right == sp.oo and left == -sp.oo and default == right
print("Confirmed")
```

## Example 2 — Equal degrees

### Hand solution

$$
\begin{aligned}
\lim_{x\to\pm\infty}\dfrac{3x^2+1}{x^2-4}
&= \lim_{x\to\pm\infty}\dfrac{3+\dfrac{1}{x^2}}{1-\dfrac{4}{x^2}} \\
&= 3.
\end{aligned}
$$

Divide the numerator and the denominator by $x^2$. Every term that still has a denominator tends to zero.
The horizontal asymptote $y=3$ holds from both sides.

### Python recheck


```python
import sympy as sp

# (3x^2 + 1)/(x^2 - 4). Equal degrees, so both infinite ends have limit 3.
x = sp.symbols("x")
f = (3*x**2 + 1)/(x**2 - 4)
assert sp.limit(f, x, sp.oo) == 3
assert sp.limit(f, x, -sp.oo) == 3
print("Both sides:", sp.limit(f, x, sp.oo))
print("Confirmed")
```

## Example 3 — Linear numerator, linear denominator

### Hand solution

$$
\dfrac{2x+1}{x-3}=2+\dfrac{7}{x-3}.
$$

$$
\lim_{x\to\pm\infty}\dfrac{2x+1}{x-3}=2,
$$

but

$$
\lim_{x\to 3^+}\dfrac{2x+1}{x-3}=+\infty,
\qquad
\lim_{x\to 3^-}\dfrac{2x+1}{x-3}=-\infty.
$$

Near $3$ the numerator is positive, while the denominator changes from positive to negative.
One function can have both a horizontal asymptote and a vertical asymptote. The two are compatible.

### Python recheck


```python
import sympy as sp

# (2x + 1)/(x - 3). Hand limits: 2 at both infinities,
# +infinity from the right of 3, and -infinity from the left of 3.
x = sp.symbols("x")
f = (2*x + 1)/(x - 3)
assert sp.limit(f, x, sp.oo) == 2
assert sp.limit(f, x, 3, dir="+") == sp.oo
assert sp.limit(f, x, 3, dir="-") == -sp.oo
print("Horizontal 2; vertical +oo from the right and -oo from the left")
print("Confirmed")
```

## Example 4 — A heavier numerator, and an infinite limit as $x\to\infty$

### Hand solution

$$
\lim_{x\to +\infty}\dfrac{x^3+x}{x^2+1}.
$$

The degree of the numerator is one greater. Divide:

$$
\begin{aligned}
\dfrac{x^3+x}{x^2+1}
&= x\cdot\dfrac{1+\dfrac{1}{x^2}}{1+\dfrac{1}{x^2}} \\
&\to +\infty.
\end{aligned}
$$

More precisely, the polynomial quotient is linear and tends to $+\infty$.
From the side $-\infty$, the same expression tends to $-\infty$, because the odd-degree factor changes the sign.

### Python recheck


```python
import sympy as sp

# (x^3 + x)/(x^2 + 1). Hand limits: +infinity as x -> +infinity, -infinity as x -> -infinity.
x = sp.symbols("x")
f = (x**3 + x)/(x**2 + 1)
print(sp.limit(f, x, sp.oo), sp.limit(f, x, -sp.oo))
assert sp.limit(f, x, sp.oo) == sp.oo
assert sp.limit(f, x, -sp.oo) == -sp.oo
print("Confirmed")
```

## Example 5 — The tangent beside an asymptote

### Hand solution

At $x=\dfrac{\pi}{2}$, the denominator of $\tan x=\dfrac{\sin x}{\cos x}$ is zero and the numerator is $1$.

$$
\lim_{x\to (\pi/2)^-}\tan x=+\infty.
$$

Cosine reaches zero from the positive side, and sine is positive.
From the right, cosine is a small negative number, and the limit is $-\infty$.
A graphing window has to stop before $\dfrac{\pi}{2}$. Otherwise a nearly vertical segment falsely joins the two branches.

### Python recheck


```python
import sympy as sp

# tan(x) at pi/2. Hand limits: +infinity from the left, -infinity from the right.
x = sp.symbols("x")
assert sp.limit(sp.tan(x), x, sp.pi/2, dir="-") == sp.oo
assert sp.limit(sp.tan(x), x, sp.pi/2, dir="+") == -sp.oo
print("From the left +oo and from the right -oo")
print("Confirmed")
```

## A pitfall

In the calculus of the real numbers, "the limit equals $\infty$" is not the same statement as "the limit exists."
In this course, a two-sided limit exists when the two sides are equal and finite,
unless we explicitly speak of a limit in the extended sense.

## Summary

Write the direction.
As $x\to\infty$, factor out the highest degree.
A vertical asymptote and a horizontal asymptote are two separate questions. Ask both.
