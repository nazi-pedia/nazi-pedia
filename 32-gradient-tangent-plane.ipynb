# ۳۲ · گرادیان، مشتق سویی و صفحهٔ مماس

| | |
|---|---|
| **دوره** | ریاضی عمومی چندمتغیره |
| **پیش‌نیاز** | ۳۱ |

\[
\nabla f=(f_x,f_y)
\]

مشتق سویی در جهت بردار واحد \(\mathbf{u}\):

\[
D_{\mathbf{u}}f=\nabla f\cdot\mathbf{u}
\]

بیشینهٔ این مقدار \(|\nabla f|\) است و در جهت خود گرادیان رخ می‌دهد.
گرادیان بر تراز \(f=\text{ثابت}\) عمود است.

صفحهٔ مماس بر \(z=f(x,y)\) در \((a,b)\):

\[
z=f(a,b)+f_x(a,b)(x-a)+f_y(a,b)(y-b)
\]

این خطی‌سازی درس ۱۱ در دو متغیر است.



## مثال ۱ — جهت تندترین صعود

### حل ریاضی

\(f=x^2+y^2\). \(\nabla f=(2x,2y)\).
در \((1,1)\): \((2,2)\)، و \(|\nabla f|=\sqrt{8}=2\sqrt{2}\).

مشتق سویی در جهت \(\mathbf{i}=(1,0)\):

\[
(2,2)\cdot(1,0)=2
\]

کمتر از \(2\sqrt{2}\) است، چون \(\mathbf{i}\) جهت گرادیان نیست.

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y")
f = x**2 + y**2
grad = sp.Matrix([sp.diff(f, x), sp.diff(f, y)]).subs({x: 1, y: 1})
directional = grad.dot(sp.Matrix([1, 0]))
print(grad, grad.norm(), directional)
assert grad == sp.Matrix([2, 2])
assert sp.simplify(grad.norm() - 2*sp.sqrt(2)) == 0
assert directional == 2
print("تایید شد")

```

## مثال ۲ — صفحهٔ مماس سهمی‌گون

### حل ریاضی

\(z=x^2+y^2\) در \((1,1,2)\).
\(z_x=2x=2\) و \(z_y=2\).

\[
z-2=2(x-1)+2(y-1)
\]
\[
z=2x+2y-2
\]

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y")
f = x**2 + y**2
plane = f.subs({x: 1, y: 1}) + sp.diff(f, x).subs({x: 1, y: 1})*(x - 1) + sp.diff(f, y).subs({x: 1, y: 1})*(y - 1)
print(sp.expand(plane))
assert sp.expand(plane) == 2*x + 2*y - 2
print("تایید شد")

```

## مثال ۳ — تقریب خطی

### حل ریاضی

\(f=e^{x}\cos y\) در \((0,0)\).
\(f(0,0)=1\).
\(f_x=e^{x}\cos y\) و در مبدأ ۱ است.
\(f_y=-e^{x}\sin y\) و در مبدأ ۰ است.

\[
L(x,y)=1+x
\]

پس \(f(0.1,-0.1)\approx 1.1\).

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y")
f = sp.exp(x)*sp.cos(y)
L = 1 + x
true = f.subs({x: sp.Rational(1, 10), y: -sp.Rational(1, 10)})
approx = L.subs({x: sp.Rational(1, 10), y: -sp.Rational(1, 10)})
print(approx, sp.N(true, 6))
assert approx == sp.Rational(11, 10)
assert abs(true - approx) < 10**(-2)
print("تایید شد")

```

## مثال ۴ — گرادیان عمود بر تراز

### حل ریاضی

تراز \(f=x^2+y^2=1\) دایره است.
در \((1,0)\)، \(\nabla f=(2,0)\) که افقی است، یعنی عمود بر مماس قائم دایره در آن نقطه.
شیب مماس دایره در \((1,0)\) بی‌نهایت است. بردار \((2,0)\) واقعاً شعاع است و شعاع بر مماس عمود است.

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y")
grad = sp.Matrix([sp.diff(x**2 + y**2, x), sp.diff(x**2 + y**2, y)])
radius = sp.Matrix([1, 0])
tangent = sp.Matrix([0, 1])
assert grad.subs({x: 1, y: 0}).dot(tangent) == 0
assert grad.subs({x: 1, y: 0}).dot(radius) == 2
print("گرادیان بر مماس عمود است")
print("تایید شد")

```

## نکته

بردار جهت در مشتق سویی باید واحد باشد.
اگر \(\mathbf{v}\) واحد نباشد، اول \(\mathbf{u}=\mathbf{v}/|\mathbf{v}|\) را بسازید. وگرنه عدد سویی را به اندازهٔ طول \(\mathbf{v}\) بزرگ کرده‌اید.
