# Support vectors are the points that hold the street

In the three-point fit, every point lies on a street edge. Add a fourth point that is safely on the correct side:

$$
(4, 0), \qquad y = +1
$$

Its score under the old solution is $4 - 1 = 3$, so its functional margin is $3 > 1$. It does not touch the street. The old line is still feasible, and it is still optimal: the closest positive point is still $(2, 0)$, and the closest negative location is still $(0, 0)$.

A training point with functional margin strictly greater than $1$ is **not** a support vector. You can delete it and the optimal line does not move.


```python
# The extra positive point is outside the street. The old line still fits.
from fractions import Fraction

w1, b = Fraction(1), Fraction(-1)
extra_margin = 1 * (w1 * 4 + b)
assert extra_margin == 3
original = [((2, 0), 1), ((0, 1), -1), ((0, -1), -1)]
for (x1, x2), label in original:
    assert label * (w1 * x1 + b) == 1
print("extra functional margin", extra_margin)
```

## Move that point onto the street and the fit changes

Now put the extra point at $(1, 0)$, still with label $+1$. Under the old line its score is $0$, so it sits on the decision line, inside the street. The old solution is no longer feasible.

The closest positive point is now $(1, 0)$. The closest negative location is still $(0, 0)$. The gap has length $1$, so each side of the street has width $1/2$. The center is $x_1 = 1/2$.

We need functional margin $1$ at $x_1 = 1$ and at $x_1 = 0$:

$$
\begin{aligned}
w\cdot 1 + b &= 1 \\
-(b) &= 1
\end{aligned}
$$

So $b = -1$ and $w - 1 = 1$, hence $w = 2$. The new solution is

$$
\mathbf{w} = (2, 0), \qquad b = -1
$$

The old support vector $(2, 0)$ now has score $4 - 1 = 3$ and margin $3 > 1$. It has stopped holding the street. The new holders are $(1, 0)$, $(0, 1)$, and $(0, -1)$.


```python
# After the positive point moves to x1 = 1, the weight doubles.
from fractions import Fraction

w1, b = Fraction(2), Fraction(-1)
rows = [
    ((1, 0), 1, Fraction(1)),
    ((2, 0), 1, Fraction(3)),
    ((0, 1), -1, Fraction(1)),
    ((0, -1), -1, Fraction(1)),
]
for (x1, x2), label, expected in rows:
    margin = label * (w1 * x1 + b)
    assert margin == expected
    print((x1, x2), "margin", margin)
assert Fraction(1, 2) * w1 ** 2 == 2
```

## The definition to keep

For a hard-margin solution, point $i$ is a support vector when

$$
y_i(\mathbf{w}\cdot\mathbf{x}_i+b) = 1
$$

Points with a larger margin can be removed without changing $\mathbf{w}$ or $b$. Points on the edge cannot.

Later, the dual variables will say the same thing in numbers: a support vector has a positive multiplier, and a non-support vector has multiplier zero.


## A pitfall

"Support vector" does not mean "a point whose label we care about." It means "a point whose removal would let the street get wider or shift." A point far on the correct side is important as data, but it is not holding the boundary.

## What to carry forward

On the original three points, all three are support vectors. The point $(4, 0)$ is not. Moving a positive point from $x_1 = 4$ to $x_1 = 1$ changes the solution from $\mathbf{w} = (1, 0)$ to $\mathbf{w} = (2, 0)$.
