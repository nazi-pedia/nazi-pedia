# Three points, solved by hand

Use the same three points and finish the algebra, not only the picture.

$$
\begin{aligned}
&(2, 0), && y = +1 \\
&(0, 1), && y = -1 \\
&(0, -1), && y = -1
\end{aligned}
$$

The negative pair is symmetric across the horizontal axis. The positive point lies on that axis. The optimal street should respect the same symmetry, so the weight on $x_2$ should be zero. We still check that guess against the constraints instead of trusting the picture alone.


## Reduce the problem

Set $\mathbf{w} = (w, 0)$. The three inequalities become

$$
\begin{aligned}
2w + b &\ge 1 \\
-(b) &\ge 1 \\
-(b) &\ge 1
\end{aligned}
$$

The last two are the same statement: $b \le -1$. The objective is $\tfrac{1}{2}w^2$. Making $|w|$ large only hurts the objective, so the positive-point inequality and the bias inequality should both be tight. Otherwise we could shrink $|w|$ or move $b$ and still be feasible.

$$
b = -1, \qquad 2w + b = 1
$$

Hence $2w - 1 = 1$, so $w = 1$. The candidate is

$$
\mathbf{w} = (1, 0), \qquad b = -1
$$

Both negative points give the same margin, so the symmetry assumption did not throw a point away.


## It really is the nearest-point argument

The convex hull of the two negative points is the vertical segment from $(0, -1)$ to $(0, 1)$. The closest point on that segment to $(2, 0)$ is $(0, 0)$. The gap has length $2$. Half of that gap is $1$, so the margin on each side is $1$ and the decision line passes through $(1, 0)$.

The direction of the gap is $(2, 0) - (0, 0) = (2, 0)$. A unit step in that direction, scaled so the functional margin is $1$, is exactly $\mathbf{w} = (1, 0)$. This is why a point that is not itself the closest point can still sit on the margin: both $(0, 1)$ and $(0, -1)$ are at distance $1$ from the line $x_1 = 1$.


```python
# Hand solution: w = (1, 0), b = -1. Every constraint is tight.
from fractions import Fraction

w1 = Fraction(1)
b = Fraction(-1)
points = [((2, 0), 1), ((0, 1), -1), ((0, -1), -1)]
for (x1, x2), label in points:
    margin = label * (w1 * x1 + b)
    assert margin == 1
objective = Fraction(1, 2) * w1 ** 2
assert objective == Fraction(1, 2)
print("w", (w1, 0), "b", b)
print("objective", objective)
```

## What if the symmetry is dropped?

Suppose someone proposes a slanted line with $w_2 \ne 0$. The two negative points are mirror images. Any slant that helps one of them hurts the other. For example, $\mathbf{w} = (1, \tfrac{1}{2})$ and $b = -1$ gives

$$
\begin{aligned}
(2, 0) &: 2 - 1 = 1 \\
(0, 1) &: (-1)\left(\tfrac{1}{2}-1\right) = \tfrac{1}{2} < 1 \\
(0, -1) &: (-1)\left(-\tfrac{1}{2}-1\right) = \tfrac{3}{2}
\end{aligned}
$$

The upper negative point has fallen inside the street. To push it back out you need a larger $w_1$, which raises the objective above $1/2$. The level line is better.


```python
# A slanted candidate breaks the upper negative point.
from fractions import Fraction

w = (Fraction(1), Fraction(1, 2))
b = Fraction(-1)
rows = [((2, 0), 1), ((0, 1), -1), ((0, -1), -1)]
margins = []
for (x1, x2), label in rows:
    margins.append(label * (w[0] * x1 + w[1] * x2 + b))
assert margins == [Fraction(1), Fraction(1, 2), Fraction(3, 2)]
print("slanted margins", margins)
```

## A pitfall

"The line goes halfway between the nearest points" needs the nearest points of the two convex hulls, not an arbitrary pair. Here the nearest negative location is $(0, 0)$, which is not one of the training points. Averaging $(2, 0)$ with $(0, 1)$ would aim the street at $(1, 0.5)$ and tilt it for no reason.

## What to carry forward

For these three points the hard-margin solution is $\mathbf{w} = (1, 0)$, $b = -1$. All three functional margins equal $1$, and the objective equals $1/2$.
