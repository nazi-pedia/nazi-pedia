# Precision, recall, and a tie on accuracy

Return to the 10-customer sketch. Three leave. The useful rule has $tp = 2$, $fp = 2$, $fn = 1$, $tn = 5$.

$$
\mathrm{accuracy} = \dfrac{2+5}{10} = \dfrac{7}{10}.
$$

Stay-for-everyone also scores $7/10$, with $tp = 0$. Precision of that constant rule would divide by $tp + fp = 0$, so it is left undefined. Recall is $0/3 = 0$.

On the fit rows, the winning sentence is ahead of Stay-for-everyone on accuracy as well as on $F_1$:

$$
\dfrac{1123 + 3070}{5626} = \dfrac{4193}{5626},
\qquad
\dfrac{4122}{5626}
$$

for the constant Stay rule, because the fit set has 4122 stayers. The comparison that decides the path is the held-out $F_1$ in the next lesson. Fit-set accuracy is part of choosing the sentence, and it is the same rows the sentence was ranked on.


```python
# The sketch ties on accuracy. Recall is what moves.
from fractions import Fraction

useful = Fraction(2 + 5, 10)
constant = Fraction(7, 10)
precision = Fraction(2, 2 + 2)
recall = Fraction(2, 2 + 1)
f1 = 2 * precision * recall / (precision + recall)
print(useful, constant, precision, recall, f1)
assert useful == constant == Fraction(7, 10)
assert f1 == Fraction(4, 7)
assert Fraction(0, 3) == 0
```

```python
# Locate the telco table beside this notebook, or under the course root.
import csv
from pathlib import Path

def data_path():
    here = Path.cwd()
    name = "Telco-Customer-Churn.csv"
    for folder in [here, *here.parents]:
        nested = folder / "17-Customer Churn" / "data" / name
        direct = folder / "data" / name
        if nested.exists():
            return nested
        if direct.exists():
            return direct
    raise FileNotFoundError(name)

def read_customers():
    with data_path().open(newline="", encoding="utf-8") as handle:
        return list(csv.DictReader(handle))
# Fit-set accuracy of the winner, beside predicting Stay.
import hashlib
from fractions import Fraction

rows = read_customers()

def held(customer_id):
    digest = hashlib.md5(customer_id.encode("utf-8")).hexdigest()
    return int(digest, 16) % 5 == 0

def leave_rule(row):
    return row["Contract"] == "Month-to-month" and (
        row["InternetService"] == "Fiber optic"
        or row["PaymentMethod"] == "Electronic check"
    )

fit = [row for row in rows if not held(row["customerID"])]
correct = sum((leave_rule(row) == (row["Churn"] == "Yes")) for row in fit)
stayers = sum(row["Churn"] == "No" for row in fit)
print(Fraction(correct, len(fit)), Fraction(stayers, len(fit)))
assert len(fit) == 5626
assert correct == 4193 and stayers == 4122
assert Fraction(4193, 5626) > Fraction(4122, 5626)
```

## Pitfall

A tie at $7/10$ on the sketch is the small version of a later fact: on the held-out customers, Stay-for-everyone wins the accuracy comparison, and the readable rule wins $F_1$. Fit-set accuracy pointing the other way is why the held-out count is required.

## Summary

$F_1 = 4/7$ on the sketch, with accuracy tied at $7/10$. On the 5626 fit rows the rule is correct on 4193 customers, and Stay-for-everyone is correct on 4122.
