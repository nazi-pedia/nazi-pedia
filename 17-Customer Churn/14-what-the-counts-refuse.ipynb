# What the counts refuse

Gender moves the leave rate by less than one percentage point. Female: 939 of 3488, which is $939/3488$. Male: 930 of 3555, which is $62/237$. And $939 + 930 = 1869$.

$$
\left|\dfrac{939}{3488} - \dfrac{62}{237}\right| < \dfrac{1}{100}.
$$

Phone service is similarly quiet. With a phone: 1699 of 6361. Without a phone: 170 of 682, which is $85/341$.

Senior citizens leave more often: 476 of 1142, which is $238/571$, against 1393 of 5901, which is $199/843$. Partner and dependents move as well. No partner: 1200 of 3641. Partner: 669 of 3402, which is $223/1134$. No dependents: 1543 of 4933. Dependents: 326 of 2110, which is $163/1055$.

Those columns were not on the eight-sentence menu. They stay out of the reported score. The mean monthly charge, in cents, is $4637695/623$ among leavers and $31698575/5174$ among stayers. Leavers' bills are higher on average, and lesson 08 already ties the higher bill to fiber optic service.


```python
# Locate the telco table beside this notebook, or under the course root.
import csv
from pathlib import Path

def data_path():
    here = Path.cwd()
    name = "Telco-Customer-Churn.csv"
    for folder in [here, *here.parents]:
        nested = folder / "17-Customer Churn" / "data" / name
        direct = folder / "data" / name
        if nested.exists():
            return nested
        if direct.exists():
            return direct
    raise FileNotFoundError(name)

def read_customers():
    with data_path().open(newline="", encoding="utf-8") as handle:
        return list(csv.DictReader(handle))
# Gender gap, phone, senior, partner, dependents, and mean monthly cents.
from fractions import Fraction

rows = read_customers()

def rate(predicate):
    chosen = [row for row in rows if predicate(row)]
    left = sum(row["Churn"] == "Yes" for row in chosen)
    return len(chosen), left, Fraction(left, len(chosen))

female = rate(lambda row: row["gender"] == "Female")
male = rate(lambda row: row["gender"] == "Male")
print(female, male, abs(female[2] - male[2]))
assert female == (3488, 939, Fraction(939, 3488))
assert male == (3555, 930, Fraction(62, 237))
assert abs(female[2] - male[2]) < Fraction(1, 100)
phone = rate(lambda row: row["PhoneService"] == "Yes")
no_phone = rate(lambda row: row["PhoneService"] == "No")
assert phone == (6361, 1699, Fraction(1699, 6361))
assert no_phone == (682, 170, Fraction(85, 341))
senior = rate(lambda row: row["SeniorCitizen"] == "1")
other = rate(lambda row: row["SeniorCitizen"] == "0")
assert senior == (1142, 476, Fraction(238, 571))
assert other == (5901, 1393, Fraction(199, 843))
no_partner = rate(lambda row: row["Partner"] == "No")
partner = rate(lambda row: row["Partner"] == "Yes")
assert no_partner == (3641, 1200, Fraction(1200, 3641))
assert partner == (3402, 669, Fraction(223, 1134))
no_dep = rate(lambda row: row["Dependents"] == "No")
dep = rate(lambda row: row["Dependents"] == "Yes")
assert no_dep == (4933, 1543, Fraction(1543, 4933))
assert dep == (2110, 326, Fraction(163, 1055))

def cents(text):
    return int(round(float(text) * 100))

for label, expected in (("Yes", Fraction(4637695, 623)), ("No", Fraction(31698575, 5174))):
    values = [cents(row["MonthlyCharges"]) for row in rows if row["Churn"] == label]
    print(label, Fraction(sum(values), len(values)))
    assert Fraction(sum(values), len(values)) == expected
```

## Pitfall

A senior-citizen sentence has a visible gap, $238/571$ against $199/843$, and it was not one of the eight sentences ranked on the fit rows. Putting it into the holdout after seeing $488/861$ would change the contest. The same restraint covers partner and dependents.
