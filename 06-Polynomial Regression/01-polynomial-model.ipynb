# A polynomial is linear in its coefficients

Polynomial regression fits a curve with a polynomial and estimates that curve by least squares. The curve bends. The unknown coefficients still enter the prediction as a sum of known columns, so the normal equations of linear regression apply without a new theory.

The running polynomial in this path is

$$
\mu(x) = 1 + 2x + 3x^2.
$$

Every later numerical example either evaluates this polynomial or fits a table built from it.


## Reading the formula

At $x = -2, -1, 0, 1, 2$,

$$
\begin{aligned}
\mu(-2) &= 1 + 2(-2) + 3(4) = 9, \\
\mu(-1) &= 1 + 2(-1) + 3(1) = 2, \\
\mu(0) &= 1, \\
\mu(1) &= 1 + 2 + 3 = 6, \\
\mu(2) &= 1 + 4 + 12 = 17.
\end{aligned}
$$

The graph is a parabola. The prediction is nevertheless $b_0$ times the column $1$, plus $b_1$ times the column $x$, plus $b_2$ times the column $x^2$. Those three columns can be written down before any coefficient is known. That is what "linear in the parameters" means.

Completing the square locates the vertex without calculus:

$$
\begin{aligned}
1 + 2x + 3x^2
&= 3\left(x^2 + \dfrac{2}{3}x\right) + 1 \\
&= 3\left[\left(x + \dfrac{1}{3}\right)^2 - \dfrac{1}{9}\right] + 1 \\
&= 3\left(x + \dfrac{1}{3}\right)^2 + \dfrac{2}{3}.
\end{aligned}
$$

The parabola bottoms at $x = -1/3$, and the value there is $2/3$.


```python
# Evaluate 1 + 2x + 3x^2, then check the completed square.
from fractions import Fraction
xs = [-2, -1, 0, 1, 2]
values = [1 + 2 * x + 3 * x * x for x in xs]
vertex = Fraction(-1, 3)
bottom = 3 * (vertex + Fraction(1, 3)) ** 2 + Fraction(2, 3)
identity = [
    3 * (Fraction(x) + Fraction(1, 3)) ** 2 + Fraction(2, 3) for x in xs
]
print("values", values)
print("bottom", bottom)
print("identity", identity)
assert values == [9, 2, 1, 6, 17]
assert bottom == Fraction(2, 3)
assert identity == [Fraction(v) for v in values]
```

## One change: raise the quadratic coefficient

Replace $3$ by $4$. Every prediction increases by exactly $x^2$:

$$
\begin{aligned}
x = -2 &\Longrightarrow +4, \\
x = -1 &\Longrightarrow +1, \\
x = 0 &\Longrightarrow +0, \\
x = 1 &\Longrightarrow +1, \\
x = 2 &\Longrightarrow +4.
\end{aligned}
$$

The new values are $13, 3, 1, 7, 21$. A change in $b_2$ does not add the same number to every row. It adds the column that $b_2$ multiplies.


```python
# Increasing the quadratic coefficient adds the column x squared.
old = [1 + 2 * x + 3 * x * x for x in (-2, -1, 0, 1, 2)]
new = [1 + 2 * x + 4 * x * x for x in (-2, -1, 0, 1, 2)]
added = [x * x for x in (-2, -1, 0, 1, 2)]
print("old", old)
print("new", new)
print("difference", [b - a for a, b in zip(old, new)])
assert [b - a for a, b in zip(old, new)] == added
assert new == [13, 3, 1, 7, 21]
```

## What is not a polynomial regression

The formula $y = b_0 + x^{\,b_1}$ has a coefficient in the exponent. At $x = 2$ and $b_1 = 3$ the value is $8$, but the column that multiplies a coefficient cannot be built until $b_1$ is already known. Least squares for that model is nonlinear regression. The loss is no longer a quadratic function of a coefficient vector, and $X^{\mathsf T}X\,\beta = X^{\mathsf T}y$ is not the estimating equation.

Powers that are known in advance are different. The column $x^2$ for $x = -2, -1, 0, 1, 2$ is $4, 1, 0, 1, 4$, whether $b_2$ is $3$ or $40$.


```python
# x squared is a column. An unknown power is not.
column = [x * x for x in (-2, -1, 0, 1, 2)]
print("known column x^2", column)
print("2 to an unknown power, here 3", 2 ** 3)
assert column == [4, 1, 0, 1, 4]
assert 2 ** 3 == 8
```

## Pitfall

A bent graph does not make the regression nonlinear. $1 + 2x + 3x^2$ is linear in $(1, 2, 3)$ and quadratic in $x$. The test is whether every unknown number multiplies a column that could have been typed before the fit. If it does, polynomial regression is ordinary least squares with those columns.


```python
# The three columns are available before the coefficients are chosen.
xs = [-2, -1, 0, 1, 2]
columns = [[1, 1, 1, 1, 1], xs, [x * x for x in xs]]
coefficients = [1, 2, 3]
prediction = [
    sum(coefficients[j] * columns[j][i] for j in range(3)) for i in range(5)
]
print("prediction from columns", prediction)
assert prediction == [9, 2, 1, 6, 17]
```

## Summary

- $\mu(x) = 1 + 2x + 3x^2$ equals $9, 2, 1, 6, 17$ at $x = -2, -1, 0, 1, 2$.
- Completing the square writes it as $3(x + 1/3)^2 + 2/3$.
- The fit is linear regression on the columns $1$, $x$, and $x^2$. A power that depends on an unknown coefficient is a different problem.
