# ۳۶ · انتگرال خطی و قضیهٔ گرین

| | |
|---|---|
| **دوره** | ریاضی عمومی چندمتغیره |
| **پیش‌نیاز** | ۲۹ و ۳۴ |

میدان \(\mathbf{F}=(P,Q)\). انتگرال خطی

\[
\int_C P\,dx+Q\,dy
\]

کار میدان در طول خم است.
اگر \(\mathbf{F}=\nabla\phi\)، میدان پایستار است و کار فقط به دو سر مسیر بستگی دارد:

\[
\int_C\mathbf{F}\cdot d\mathbf{r}=\phi(B)-\phi(A)
\]

روی خم بستهٔ ساده، با جهت مثبت، قضیهٔ گرین می‌گوید

\[
\int_C P\,dx+Q\,dy=\iint_D(Q_x-P_y)\,dA
\]



## مثال ۱ — میدان پایستار

### حل ریاضی

\(\mathbf{F}=(2x,2y)=\nabla(x^2+y^2)\).
از \((0,0)\) تا \((1,1)\):

\[
\phi(1,1)-\phi(0,0)=2
\]

مسیر مهم نیست.

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y")
phi = x**2 + y**2
F = sp.Matrix([sp.diff(phi, x), sp.diff(phi, y)])
assert F == sp.Matrix([2*x, 2*y])
assert phi.subs({x: 1, y: 1}) - phi.subs({x: 0, y: 0}) == 2
print(F)
print("تایید شد")

```

## مثال ۲ — گرین و مساحت

### حل ریاضی

\(\mathbf{F}=(-y,x)\). آن‌گاه \(Q_x-P_y=1-(-1)=2\).
روی مرز دایرهٔ واحد،

\[
\int_C -y\,dx+x\,dy=\iint_D 2\,dA=2\pi
\]

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t")
x = sp.cos(t)
y = sp.sin(t)
P = -y
Q = x
integrand = sp.simplify(P*sp.diff(x, t) + Q*sp.diff(y, t))
line = sp.integrate(integrand, (t, 0, 2*sp.pi))
print(integrand, line)
assert line == 2*sp.pi
print("تایید شد")

```

## مثال ۳ — میدان دقیق روی خم بسته

### حل ریاضی

\(P=y\)، \(Q=x\). آن‌گاه \(Q_x-P_y=1-1=0\).
پتانسیل \(xy\) است. روی هر خم بستهٔ هموار، انتگرال صفر است.
روی دایرهٔ واحد هم باید صفر شود.

### بازبینی با پایتون



```python
import sympy as sp
t = sp.symbols("t")
x = sp.cos(t)
y = sp.sin(t)
integrand = y*sp.diff(x, t) + x*sp.diff(y, t)
line = sp.integrate(sp.simplify(integrand), (t, 0, 2*sp.pi))
print(sp.simplify(integrand), line)
assert sp.simplify(line) == 0
print("تایید شد")

```

## مثال ۴ — کار روی پاره‌خط، وقتی پایستار نیست؟ 

### حل ریاضی

\(\mathbf{F}=(y,0)\) پایستار نیست، چون \(Q_x-P_y=-1\neq 0\).
روی پاره‌خط از \((0,0)\) تا \((1,0)\)، که \(y=0\) و \(dy=0\)، انتگرال صفر است.
روی پاره‌خط از \((0,0)\) تا \((0,1)\)، \(dx=0\) و \(P\) هرچه باشد در \(dx\) ضرب می‌شود و باز صفر است.
صفر بودن روی دو مسیر خاص، پایستاری را ثابت نمی‌کند. گرین روی مربع واحد، \(\iint -1\,dA=-1\) را می‌دهد.

### بازبینی با پایتون



```python
import sympy as sp
x, y = sp.symbols("x y")
curl = sp.diff(0, x) - sp.diff(y, y)
square = sp.integrate(curl, (x, 0, 1), (y, 0, 1))
print(curl, square)
assert curl == -1 and square == -1
print("تایید شد")

```

## نکته

پایستاری را با \(Q_x=P_y\) روی ناحیهٔ ساده بسنجید، نه با صفر شدن یک مسیر.
جهت مثبت گرین، ناحیه را در چپ مرز می‌گذارد: برای دایره، خلاف ساعت.

## جمع‌بندی

پتانسیل، مسیر را حذف می‌کند.
گرین، خم بسته را به انتگرال دوگانهٔ تاو تبدیل می‌کند.
