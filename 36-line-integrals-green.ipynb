For a field $\mathbf{F}=(P,Q)$, the line integral

$$
\begin{aligned}
 \int_{C} P\,dx+Q\,dy
\end{aligned}
$$

is the work done by the field along the curve.
If $\mathbf{F}=\nabla\phi$, the field is conservative, and the work depends only on the endpoints:

$$
\begin{aligned}
 \int_{C}\mathbf{F}\cdot d\mathbf{r}
 &=
 \phi(B)-\phi(A).
\end{aligned}
$$

On a simple closed curve with positive orientation, Green's theorem says

$$
\begin{aligned}
 \int_{C} P\,dx+Q\,dy
 &=
 \iint_{D}\left(\dfrac{\partial Q}{\partial x}-\dfrac{\partial P}{\partial y}\right)\,dA.
\end{aligned}
$$


## Example 1 — A conservative field

### Hand solution

$$
\begin{aligned}
 \mathbf{F}
 &=
 \begin{pmatrix} 2x \\ 2y \end{pmatrix}
 =
 \nabla(x^{2}+y^{2}).
\end{aligned}
$$

From $(0,0)$ to $(1,1)$,

$$
\begin{aligned}
 \phi(1,1)-\phi(0,0) &= 2.
\end{aligned}
$$

The path does not matter.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Potential phi = x^2 + y^2 and its gradient.
x, y = sp.symbols("x y")
phi = x**2 + y**2
F = sp.Matrix([sp.diff(phi, x), sp.diff(phi, y)])
assert F == sp.Matrix([2*x, 2*y])
# The change in potential from (0, 0) to (1, 1).
assert phi.subs({x: 1, y: 1}) - phi.subs({x: 0, y: 0}) == 2
print(F)
print("Confirmed")
```

## Example 2 — Green and area

### Hand solution

Take $\mathbf{F}=(-y,x)$. Then

$$
\begin{aligned}
 \dfrac{\partial Q}{\partial x}-\dfrac{\partial P}{\partial y}
 &=
 1-(-1)
 =
 2.
\end{aligned}
$$

On the boundary of the unit disk,

$$
\begin{aligned}
 \int_{C} -y\,dx+x\,dy
 &=
 \iint_{D} 2\,dA
 =
 2\pi.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# The unit circle, traversed once.
t = sp.symbols("t")
x = sp.cos(t)
y = sp.sin(t)
P = -y
Q = x
# Line integrand P dx/dt + Q dy/dt.
integrand = sp.simplify(P*sp.diff(x, t) + Q*sp.diff(y, t))
line = sp.integrate(integrand, (t, 0, 2*sp.pi))
print(integrand, line)
assert line == 2*sp.pi
print("Confirmed")
```

## Example 3 — An exact field on a closed curve

### Hand solution

Take $P=y$ and $Q=x$. Then

$$
\begin{aligned}
 \dfrac{\partial Q}{\partial x}-\dfrac{\partial P}{\partial y}
 &=
 1-1
 =
 0.
\end{aligned}
$$

A potential is $xy$. On every smooth closed curve the integral is $0$.
On the unit circle it must be $0$ as well.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# The unit circle, with P = y and Q = x.
t = sp.symbols("t")
x = sp.cos(t)
y = sp.sin(t)
integrand = y*sp.diff(x, t) + x*sp.diff(y, t)
line = sp.integrate(sp.simplify(integrand), (t, 0, 2*sp.pi))
print(sp.simplify(integrand), line)
# An exact field integrates to 0 around this closed curve.
assert sp.simplify(line) == 0
print("Confirmed")
```

## Example 4 — Work on a segment when the field is not conservative

### Hand solution

The field $\mathbf{F}=(y,0)$ is not conservative, because

$$
\begin{aligned}
 \dfrac{\partial Q}{\partial x}-\dfrac{\partial P}{\partial y}
 &=
 -1
 \neq
 0.
\end{aligned}
$$

On the segment from $(0,0)$ to $(1,0)$, one has $y=0$ and $dy=0$, so the integral is $0$.
On the segment from $(0,0)$ to $(0,1)$, one has $dx=0$, so $P$ is multiplied by $dx$ and the integral is $0$ again.
Vanishing on two particular paths does not prove that the field is conservative. Green's theorem on the unit square gives

$$
\begin{aligned}
 \iint_{D} (-1)\,dA &= -1.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Planar curl of F = (y, 0): differentiate Q = 0 and P = y.
x, y = sp.symbols("x y")
curl = sp.diff(0, x) - sp.diff(y, y)
# Green's theorem over the unit square.
square = sp.integrate(curl, (x, 0, 1), (y, 0, 1))
print(curl, square)
assert curl == -1 and square == -1
print("Confirmed")
```

## A pitfall

Test whether a field is conservative by checking $\dfrac{\partial Q}{\partial x}=\dfrac{\partial P}{\partial y}$ on a simply connected region, not by the vanishing of one path.
The positive orientation in Green's theorem keeps the region on the left of the boundary. For a circle, that direction is counterclockwise.

## Summary

A potential removes the path.
Green's theorem turns a closed curve into a double integral of the curl.
