# ۰۶ · حد نامتناهی و حد در بی‌نهایت

| | |
|---|---|
| **دوره** | ریاضی عمومی ۱ |
| **پیش‌نیاز** | ۰۳ و ۰۴ |

دو جمله را از هم جدا نگه دارید.

حد نامتناهی یعنی \(x\) به عدد متناهی نزدیک می‌شود و \(|f(x)|\) بی‌کران رشد می‌کند. این مجانب قائم است.
حد در بی‌نهایت یعنی \(x\) خودش دور می‌شود و \(f(x)\) به عددی متناهی می‌نشیند. این مجانب افقی است.

بی‌نهایت عدد حقیقی نیست. نوشتن \(\lim=+\infty\) نامی برای یک رفتار است.

SymPy اگر جهت را نگویید از راست نزدیک می‌شود.
حد دوطرفه فقط وقتی اعلام می‌شود که دو طرف را جدا حساب کرده باشید و برابر باشند.



```python
import sympy as sp

def check(found, hand, label="نتیجه"):
    found_s = sp.simplify(found)
    hand_s = sp.simplify(hand)
    difference = sp.simplify(found_s - hand_s)
    print(label)
    print("  حل دستی:", hand_s)
    print("  پایتون:", found_s)
    print("  اختلاف:", difference)
    assert difference == 0
    print("  تایید شد")

```

## مثال ۱ — دو بی‌نهایتِ مخالف

### حل ریاضی

\[
\lim_{x\to 0^+}\frac{1}{x}=+\infty,
\qquad
\lim_{x\to 0^-}\frac{1}{x}=-\infty
\]

حد دوطرفه وجود ندارد، نه به‌صورت عدد و نه به‌صورت یک بی‌نهایت واحد.
خط \(x=0\) مجانب قائم است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
right = sp.limit(1/x, x, 0, dir="+")
left = sp.limit(1/x, x, 0, dir="-")
default = sp.limit(1/x, x, 0)
print("راست:", right)
print("چپ:", left)
print("پیش‌فرض پایتون:", default)
assert right == sp.oo and left == -sp.oo and default == right
print("تایید شد")

```

## مثال ۲ — درجه‌های برابر

### حل ریاضی

\[
\lim_{x\to\pm\infty}\frac{3x^2+1}{x^2-4}
=\lim_{x\to\pm\infty}\frac{3+1/x^2}{1-4/x^2}=3
\]

صورت و مخرج را بر \(x^2\) تقسیم کردیم. جمله‌های با مخرج، صفر می‌شوند.
مجانب افقی \(y=3\) از هر دو سو برقرار است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = (3*x**2 + 1)/(x**2 - 4)
assert sp.limit(f, x, sp.oo) == 3
assert sp.limit(f, x, -sp.oo) == 3
print("هر دو سو:", sp.limit(f, x, sp.oo))
print("تایید شد")

```

## مثال ۳ — صورت خطی، مخرج خطی

### حل ریاضی

\[
\frac{2x+1}{x-3}=2+\frac{7}{x-3}
\]

\[
\lim_{x\to\pm\infty}\frac{2x+1}{x-3}=2
\]

اما

\[
\lim_{x\to 3^+}\frac{2x+1}{x-3}=+\infty,
\qquad
\lim_{x\to 3^-}\frac{2x+1}{x-3}=-\infty
\]

چون صورت در نزدیکی ۳ مثبت است و مخرج از مثبت به منفی عوض می‌شود.
یک تابع هم مجانب افقی دارد و هم مجانب قائم. این دو با هم ناسازگار نیستند.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = (2*x + 1)/(x - 3)
assert sp.limit(f, x, sp.oo) == 2
assert sp.limit(f, x, 3, dir="+") == sp.oo
assert sp.limit(f, x, 3, dir="-") == -sp.oo
print("افقی 2، قائم از راست +oo و از چپ -oo")
print("تایید شد")

```

## مثال ۴ — صورت سنگین‌تر، حد نامتناهی در بی‌نهایت

### حل ریاضی

\[
\lim_{x\to +\infty}\frac{x^3+x}{x^2+1}
\]

درجهٔ صورت یکی بیشتر است. تقسیم:

\[
\frac{x^3+x}{x^2+1}=x\cdot\frac{1+1/x^2}{1+1/x^2}\to +\infty
\]

دقیق‌تر، خارج‌قسمت چندجمله‌ای خطی است و به \(+\infty\) می‌رود.
از سمت \(-\infty\) همین عبارت به \(-\infty\) می‌رود، چون عامل درجهٔ فرد علامت را عوض می‌کند.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = (x**3 + x)/(x**2 + 1)
print(sp.limit(f, x, sp.oo), sp.limit(f, x, -sp.oo))
assert sp.limit(f, x, sp.oo) == sp.oo
assert sp.limit(f, x, -sp.oo) == -sp.oo
print("تایید شد")

```

## مثال ۵ — تانژانت کنار مجانب

### حل ریاضی

\(\tan x=\sin x/\cos x\) در \(x=\pi/2\) مخرج صفر و صورت ۱ است.

\[
\lim_{x\to (\pi/2)^-}\tan x=+\infty
\]

چون کسینوس از سمت مثبت به صفر می‌رسد و سینوس مثبت است.
از راست، کسینوس منفیِ کوچک است و حد \(-\infty\) است.
پنجرهٔ نمودار باید پیش از \(\pi/2\) بایستد؛ وگرنه یک پاره‌خط تقریباً قائم، دو شاخه را به دروغ به هم وصل می‌کند.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
assert sp.limit(sp.tan(x), x, sp.pi/2, dir="-") == sp.oo
assert sp.limit(sp.tan(x), x, sp.pi/2, dir="+") == -sp.oo
print("از چپ +oo و از راست -oo")
print("تایید شد")

```

## نکته
