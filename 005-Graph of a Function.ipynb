## 1. A plotting function you will reuse

**See.** Matplotlib draws arrays of numbers. You sample \(x\), you compute \(y\), you draw the polyline.
A fine sample looks like a curve. It is still a sample.

**Name.** `np.linspace(a, b, n)` returns `n` evenly spaced numbers from `a` to `b`, and it includes both ends.
That is different from `range`, which excludes its stop.

The helper below is the house style for this library: axes through the origin when they help,
a title that states the formula, and no top or right spine.


```python
import matplotlib.pyplot as plt
import numpy as np

def plot_function(f, a, b, title, n=500):
    xs = np.linspace(a, b, n)
    ys = f(xs)
    fig, ax = plt.subplots(figsize=(7, 4))
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.plot(xs, ys, color="#1f4e79", linewidth=2)
    ax.set_title(title)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.grid(True, alpha=0.3)
    fig.tight_layout()
    plt.show()
    return xs, ys

xs, ys = plot_function(lambda x: 2 * x + 1, -5, 5, r"$f(x) = 2x + 1$")
# The line crosses y = 0 at x = -1/2. Check the sample nearest that root.
root_index = np.argmin(np.abs(ys))
print(f"sample closest to the intercept: x = {xs[root_index]:.3f}, y = {ys[root_index]:.3f}")
assert abs(xs[root_index] + 0.5) < 0.05
```

**Read the output.** The intercept of \(2x+1\) is exactly \(x = -1/2\).
The sample is only near it. A graph suggests; an equation decides. Lesson 004 already taught the deciding step.


## 2. Four shapes worth recognizing on sight

**Do.** Plot each function on an interval that shows its character.

- \(x^2 - 4\): a parabola, roots at \(\pm 2\), vertex at \((0, -4)\)
- \(x^3\): odd, flat at the origin, increasing
- \(|x|\): a corner at the origin. Corners will matter when derivatives begin.
- \(\sqrt{x}\): defined for \(x \ge 0\) only. The interval must respect the domain.


```python
import matplotlib.pyplot as plt
import numpy as np

def plot_function(f, a, b, title, n=500):
    xs = np.linspace(a, b, n)
    ys = f(xs)
    fig, ax = plt.subplots(figsize=(7, 4))
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.plot(xs, ys, color="#1f4e79", linewidth=2)
    ax.set_title(title)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.grid(True, alpha=0.3)
    fig.tight_layout()
    plt.show()

plot_function(lambda x: x ** 2 - 4, -5, 5, r"$f(x) = x^2 - 4$")
plot_function(lambda x: x ** 3, -2, 2, r"$f(x) = x^3$")
plot_function(np.abs, -5, 5, r"$f(x) = |x|$")
plot_function(np.sqrt, 0, 10, r"$f(x) = \sqrt{x}$ on its domain $[0, \infty)$")
```

**Misconception.** `np.sqrt` of a negative number is a complex result you can plot.

**Correction.** For a real array it produces `nan` and a warning. The real square root is a partial function.
If you want \(\sqrt{-1} = i\), you are back in lesson 003, and you are no longer plotting a real graph.


## 3. Sine, cosine, and a window that tells the truth about tangent

**See.** Sine and cosine on \([-2\pi, 2\pi]\) show two full periods.
Tangent has vertical asymptotes at \(\pm \pi/2\). A naive sample that includes those points spikes to huge numbers
and flattens the rest of the curve into a horizontal smudge.

**Name.** Choose the interval to stop short of the asymptote, and set `ylim` so one branch is readable.
The picture must not connect the two sides of an asymptote with a nearly vertical segment.
A gap in the sample, or a broken line, is more honest. Here the open interval \((-\pi/2, \pi/2)\) keeps one branch.


```python
import matplotlib.pyplot as plt
import numpy as np

def plot_function(f, a, b, title, n=500, ylim=None):
    xs = np.linspace(a, b, n)
    ys = f(xs)
    fig, ax = plt.subplots(figsize=(7, 4))
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.plot(xs, ys, color="#1f4e79", linewidth=2)
    if ylim is not None:
        ax.set_ylim(*ylim)
    ax.set_title(title)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.grid(True, alpha=0.3)
    fig.tight_layout()
    plt.show()

plot_function(np.sin, -2 * np.pi, 2 * np.pi, r"$f(x) = \sin x$")
plot_function(np.cos, -2 * np.pi, 2 * np.pi, r"$f(x) = \cos x$")

margin = 0.08
plot_function(
    np.tan,
    -np.pi / 2 + margin,
    np.pi / 2 - margin,
    r"$f(x) = \tan x$ on $(-\pi/2,\ \pi/2)$, with a honest vertical window",
    n=1000,
    ylim=(-8, 8),
)
```

## 4. Two functions on one pair of axes

**See.** \(x^2\) and \(2x+3\) meet where \(x^2 - 2x - 3 = 0\), that is at \(x = -1\) and \(x = 3\).
The picture suggests the crossings. Solving confirms them.

**Check a second way.** Evaluate both formulas at the solved \(x\)-values. They must agree.


```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

xs = np.linspace(-5, 5, 400)
fig, ax = plt.subplots(figsize=(7, 4))
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.plot(xs, xs ** 2, color="#1f4e79", linewidth=2, label=r"$x^2$")
ax.plot(xs, 2 * xs + 3, color="#b85c38", linewidth=2, label=r"$2x + 3$")
ax.set_title(r"Intersections of $x^2$ and $2x+3$")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()

x = sp.symbols("x")
crossings = sp.solve(sp.Eq(x ** 2, 2 * x + 3), x)
print("solved crossings:", crossings)
for root in crossings:
    left = root ** 2
    right = 2 * root + 3
    print(f"at x = {root}: both functions equal {left}")
    assert sp.simplify(left - right) == 0
```

## 5. The same picture in one SymPy call

**Name.** `sp.plot(expression, (x, a, b), show=False)` is the right tool for a quick exact formula.
It is the wrong tool when you need a broken asymptote, a custom window, or two carefully labeled series.
Use it to confirm a Matplotlib picture, not to replace a picture whose window was a decision.


```python
import sympy as sp

x = sp.symbols("x")
picture = sp.plot(x ** 2 - 4, (x, -5, 5), show=False, title=r"$x^2 - 4$ via SymPy")
picture.show()
```

## Workshop — a graph with one sentence attached

Plot \(f(x) = x^3 - 3x^2 + 4\) on \([-2, 4]\). This is the function of lesson 010.
On paper, say whether the real graph seems to cross the axis, and where it seems to turn.
Then check the axis crossings by solving. Leave the turning points for lesson 010; the derivative is not available yet.
Do not claim a turn from the picture alone beyond "the sample rises, then falls, then rises" if that is what you see.


```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

def f_numeric(x):
    return x ** 3 - 3 * x ** 2 + 4

xs = np.linspace(-2, 4, 500)
fig, ax = plt.subplots(figsize=(7, 4))
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.plot(xs, f_numeric(xs), color="#1f4e79", linewidth=2)
ax.set_title(r"$f(x) = x^3 - 3x^2 + 4$")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()

x = sp.symbols("x")
real_roots = [root for root in sp.real_roots(x ** 3 - 3 * x ** 2 + 4)]
print("real roots:", real_roots)
for root in real_roots:
    assert sp.simplify(f_numeric(root)) == 0
print("every listed real root satisfies the equation")
```

**Read the output.** There is a real root at \(x = -1\), because \((-1)^3 - 3(1) + 4 = -1 - 3 + 4 = 0\).
The other two roots of the cubic may be real or not; `real_roots` is the honest query.
Looking at a turn and calling it a maximum, without a derivative, is a guess. Lesson 007 earns the right to that word.

## Mistakes this lesson is meant to prevent

- Plotting \(\sqrt{x}\) on an interval that includes negatives.
- Letting an asymptote set the vertical scale for the whole figure.
- Reading a crossing off a picture to more precision than the sample supports.
- Connecting both sides of a vertical asymptote with one line segment.
- Treating SymPy's default window as a neutral choice. Every window is a claim about what matters.
