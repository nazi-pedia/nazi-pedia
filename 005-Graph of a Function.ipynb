## Symbolic Mathematics

```python
import sympy as sp
```

## Creating a Mathematical Variable

```python
x = sp.symbols('x')
```

## Linear function

```python
f = 2*x + 1
sp.plot(f, (x, -5, 5))
```

## Quadratic function

```python
f = x**2 - 4
sp.plot(f, (x, -5, 5))
```

## Cubic function

```python
f = x**3
sp.plot(f, (x, -5, 5))
```

## Absolute-value function

```python
f = sp.Abs(x)
sp.plot(f, (x, -5, 5))
```

## Square-root function

```python
f = sp.sqrt(x)
sp.plot(f, (x, 0, 10))
```

## Sine function

```python
f = sp.sin(x)

sp.plot(f, (x, -2*sp.pi, 2*sp.pi))
```

## Cosine function

```python
f = sp.cos(x)

sp.plot(f, (x, -2*sp.pi, 2*sp.pi))
```

## Tangent function

```python
import sympy as sp

x = sp.symbols('x')

f = sp.tan(x)

sp.plot(
    f,
    (x, -sp.pi/2, sp.pi/2),
    title="Graph of f(x) = tan(x)",
    xlabel="x",
    ylabel="tan(x)"
)
```

## Graphing Multiple Functions

```python
f = x**2
g = 2*x + 3

sp.plot(
    f,
    g,
    (x, -5, 5)
)
```
