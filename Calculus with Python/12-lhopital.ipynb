# 12 L'Hôpital's Rule

| | |
|---|---|
| **Course** | Calculus I |
| **Prerequisite** | Lessons 06 and 08 |

If $\lim f/g$ has the form $0/0$ or $\infty/\infty$, and if $\lim f'/g'$ exists, then the original limit equals that new limit. The rule may be repeated, provided the form is still indeterminate each time.

The forms $0\cdot\infty$, $\infty-\infty$, and $1^\infty$ must first be rewritten as a quotient.
Do not apply L'Hôpital's rule to a form that is not indeterminate. The limit of $\frac{x}{x+1}$ at infinity comes out wrong after one differentiation if you forget that this form is $\infty/\infty$ and that the rule is allowed. The quotient $\frac{\sin x}{x}$ at $x=\pi$, by contrast, is not an indeterminate form.


## Example 1 — Once is enough

### Hand solution

$$
\begin{aligned}
 \lim_{x\to 0}\dfrac{\sin x}{x}.
\end{aligned}
$$

The form is $0/0$.

$$
\begin{aligned}
 \lim_{x\to 0}\dfrac{\cos x}{1}
 &= \cos 0 \\
 &= 1.
\end{aligned}
$$

This is the standard limit from lesson 04. L'Hôpital's rule builds it from the derivative of sine, and that derivative is itself proved from this same limit. In a careful course one of the two facts has to be taken from somewhere else, so the argument does not go in a circle. Here the limit has already been accepted, and L'Hôpital's rule only reproduces it.

### Python recheck


```python
import sympy as sp

# Standard limit sin(x)/x as x -> 0. Hand result: 1, also after differentiating numerator and denominator.
x = sp.symbols("x")
assert sp.limit(sp.sin(x)/x, x, 0) == 1
# The second assert checks the same limit along the L'Hopital path.
assert sp.limit(sp.diff(sp.sin(x), x)/sp.diff(x, x), x, 0) == 1
print("both paths give 1")
print("confirmed")
```

## Example 2 — Twice

### Hand solution

$$
\begin{aligned}
 \lim_{x\to 0}\dfrac{e^{x}-1-x}{x^2}.
\end{aligned}
$$

At $0$ the numerator is $1-1-0=0$. The form is $0/0$.

$$
\begin{aligned}
 \lim_{x\to 0}\dfrac{e^{x}-1}{2x}.
\end{aligned}
$$

The form is $0/0$ again.

$$
\begin{aligned}
 \lim_{x\to 0}\dfrac{e^{x}}{2} &= \dfrac{1}{2}.
\end{aligned}
$$

The expansion $e^{x}=1+x+\frac{x^2}{2}+\cdots$ also turns the numerator into $\frac{x^2}{2}+\cdots$, and the quotient tends to $\frac{1}{2}$.

### Python recheck


```python
import sympy as sp

# Limit of (exp(x) - 1 - x)/x**2 as x -> 0. Hand result: 1/2.
x = sp.symbols("x")
f = (sp.exp(x) - 1 - x)/x**2
exact = sp.limit(f, x, 0)
print(exact)
# The assert checks that the limit is the rational number 1/2.
assert exact == sp.Rational(1, 2)
print("confirmed")
```

## Example 3 — The form $\infty/\infty$

### Hand solution

$$
\begin{aligned}
 \lim_{x\to +\infty}\dfrac{\ln x}{x}.
\end{aligned}
$$

The numerator and the denominator both tend to $+\infty$.

$$
\begin{aligned}
 \lim_{x\to +\infty}\dfrac{1/x}{1}
 &= \lim_{x\to +\infty}\dfrac{1}{x} \\
 &= 0.
\end{aligned}
$$

A logarithm grows more slowly than every positive power. Here that power is $x$ itself.

### Python recheck


```python
import sympy as sp

# log(x)/x and log(x)/sqrt(x) as x -> infinity. Hand result: both limits are 0.
x = sp.symbols("x", positive=True)
assert sp.limit(sp.log(x)/x, x, sp.oo) == 0
assert sp.limit(sp.log(x)/sp.sqrt(x), x, sp.oo) == 0
print("ln x grows even more slowly than the square root")
print("confirmed")
```

## Example 4 — Rewriting the form $1^\infty$

### Hand solution

$$
\begin{aligned}
 \lim_{x\to 0^+}x^{x}.
\end{aligned}
$$

The base tends to $0$ and the exponent tends to $0$. Write

$$
\begin{aligned}
 x^{x} &= e^{x\ln x},
\end{aligned}
$$

and treat the exponent separately:

$$
\begin{aligned}
 \lim_{x\to 0^+}x\ln x
 &= \lim_{x\to 0^+}\dfrac{\ln x}{1/x}.
\end{aligned}
$$

The form is $\infty/\infty$, with signs: the numerator tends to $-\infty$ and the denominator tends to $+\infty$.

$$
\begin{aligned}
 \lim_{x\to 0^+}\dfrac{1/x}{-1/x^2}
 &= \lim_{x\to 0^+}(-x) \\
 &= 0.
\end{aligned}
$$

So the exponent tends to $0$, and $e^{0}=1$.

### Python recheck


```python
import sympy as sp

# As x -> 0+, x*log(x) tends to 0 and x**x tends to 1.
x = sp.symbols("x", positive=True)
assert sp.limit(x*sp.log(x), x, 0) == 0
assert sp.limit(x**x, x, 0) == 1
print("x ln x ->", sp.limit(x*sp.log(x), x, 0))
print("x^x ->", sp.limit(x**x, x, 0))
print("confirmed")
```

## Example 5 — Where L'Hôpital's rule is not allowed

### Hand solution

$$
\begin{aligned}
 \lim_{x\to\pi}\dfrac{\sin x}{x}.
\end{aligned}
$$

The numerator is $0$ and the denominator is $\pi\neq 0$. The form is not indeterminate.

$$
\begin{aligned}
 \dfrac{\sin\pi}{\pi} &= 0.
\end{aligned}
$$

Differentiating without a license produces $\frac{\cos\pi}{1}=-1$, which is the wrong answer.
What licenses the rule is an indeterminate form, not the mere fact of being a quotient.

### Python recheck


```python
import sympy as sp

# Limit of sin(x)/x at x = pi. Hand result: the true limit is 0; a blind derivative gives -1.
x = sp.symbols("x")
exact = sp.limit(sp.sin(x)/x, x, sp.pi)
naive = sp.limit(sp.diff(sp.sin(x), x)/1, x, sp.pi)
print("true limit:", exact)
print("unauthorized derivative:", naive)
# The true limit is 0, and the unauthorized derivative is -1.
assert exact == 0
assert naive == -1
print("confirmed")
```

## A pitfall

After every use of L'Hôpital's rule, classify the new limit from the start.
It may no longer be indeterminate, and a plain substitution may then be enough.

## Summary

The forms $0/0$ and $\infty/\infty$ may be differentiated directly.
First rewrite a product or a power as a quotient, or as an exponential.
If the form is not indeterminate, do not differentiate.
