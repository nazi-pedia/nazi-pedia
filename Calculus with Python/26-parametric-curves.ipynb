# 26 Parametric Curves

| | |
|---|---|
| **Course** | Calculus II |
| **Prerequisite** | 08 and 21 |

$$
x=x(t),\qquad y=y(t).
$$

The slope is

$$
\dfrac{dy}{dx}=\dfrac{\dfrac{dy}{dt}}{\dfrac{dx}{dt}},
\qquad \dfrac{dx}{dt}\neq 0.
$$

If the numerator vanishes as well, determine the tangent by a limit or by the parametric second derivative.

The length from $t=a$ to $t=b$ is

$$
L=\int_a^b\sqrt{\left(\dfrac{dx}{dt}\right)^2+\left(\dfrac{dy}{dt}\right)^2}\,dt.
$$


## Example 1 — Tangent to the circle

### Hand solution

$$
x=\cos t,\qquad y=\sin t.
$$

$$
\begin{aligned}
\dfrac{dy}{dx}
&=\dfrac{\cos t}{-\sin t}
=-\cot t.
\end{aligned}
$$

At $t=\dfrac{\pi}{4}$ the point and the slope are

$$
\begin{aligned}
(x,y)
&=\left(\dfrac{\sqrt{2}}{2},\dfrac{\sqrt{2}}{2}\right), \\
\dfrac{dy}{dx}
&=-1.
\end{aligned}
$$

The radius to this point has slope $+1$. The tangent is perpendicular to the radius.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Parameter.
t = sp.symbols("t")
# Unit circle.
x = sp.cos(t)
y = sp.sin(t)
# Parametric slope: derivative of y divided by derivative of x.
slope = sp.simplify(sp.diff(y, t)/sp.diff(x, t))
print(slope, slope.subs(t, sp.pi/4))
# At t = pi/4 the slope is -1.
assert slope.subs(t, sp.pi/4) == -1
print("Confirmed")
```

## Example 2 — Circumference from the parameter

### Hand solution

$$
\begin{aligned}
\left(\dfrac{dx}{dt}\right)^2+\left(\dfrac{dy}{dt}\right)^2
&=\sin^2 t+\cos^2 t
=1, \\
L
&=\int_0^{2\pi}1\,dt
=2\pi.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Parameter.
t = sp.symbols("t")
# Speed of the unit circle.
speed = sp.simplify(sp.sqrt(sp.diff(sp.cos(t), t)**2 + sp.diff(sp.sin(t), t)**2))
# Arc length over one full turn.
length = sp.integrate(speed, (t, 0, 2*sp.pi))
print(speed, length)
assert length == 2*sp.pi
print("Confirmed")
```

## Example 3 — One arch of a cycloid

### Hand solution

$$
x=t-\sin t,\qquad y=1-\cos t,
$$

from $t=0$ to $t=2\pi$.

$$
\begin{aligned}
\dfrac{dx}{dt}
&=1-\cos t,
\qquad
\dfrac{dy}{dt}
=\sin t, \\
\left(\dfrac{dx}{dt}\right)^2+\left(\dfrac{dy}{dt}\right)^2
&=1-2\cos t+\cos^2 t+\sin^2 t \\
&=2-2\cos t
=4\sin^2\dfrac{t}{2}.
\end{aligned}
$$

On $[0,2\pi]$, $\sin\dfrac{t}{2}\ge 0$, so

$$
\begin{aligned}
L
&=\int_0^{2\pi}2\sin\dfrac{t}{2}\,dt
=\left[-4\cos\dfrac{t}{2}\right]_0^{2\pi} \\
&=-4(-1-1)
=8.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Real parameter.
t = sp.symbols("t", real=True)
# Cycloid velocity components.
dx = sp.diff(t - sp.sin(t), t)
dy = sp.diff(1 - sp.cos(t), t)
# Speed from those components.
speed = sp.simplify(sp.sqrt(dx**2 + dy**2))
# Nonnegative hand speed used on one arch.
hand_speed = 2*sp.sin(t/2)
# The two speeds agree up to sign.
assert sp.simplify(speed**2 - hand_speed**2) == 0
# Length of one arch from the hand speed.
length = sp.integrate(hand_speed, (t, 0, 2*sp.pi))
# Numerical integral of the simplified speed.
numeric = sp.N(sp.Integral(speed, (t, 0, 2*sp.pi)))
print(speed, "->", length, "numeric:", numeric)
assert sp.simplify(length - 8) == 0
assert abs(numeric - 8) < 10**(-10)
print("Confirmed")
```

## Example 4 — The parametric second derivative

### Hand solution

$$
\dfrac{d^2y}{dx^2}
=\dfrac{\dfrac{d}{dt}\!\left(\dfrac{dy}{dx}\right)}{\dfrac{dx}{dt}}.
$$

For the circle, $\dfrac{dy}{dx}=-\cot t$, and

$$
\begin{aligned}
\dfrac{d}{dt}(-\cot t)
&=\csc^2 t, \\
\dfrac{d^2y}{dx^2}
&=\dfrac{\csc^2 t}{-\sin t}
=-\dfrac{1}{\sin^3 t}.
\end{aligned}
$$

At $t=\dfrac{\pi}{2}$ this value is $-1$. At its highest point the circle is concave downward, and the radius of curvature is $1$.

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Parameter.
t = sp.symbols("t")
# First derivative of y with respect to x on the unit circle.
slope = sp.diff(sp.sin(t), t)/sp.diff(sp.cos(t), t)
# Parametric second derivative.
second = sp.simplify(sp.diff(slope, t)/sp.diff(sp.cos(t), t))
print(second, second.subs(t, sp.pi/2))
# At the top the second derivative is -1.
assert second.subs(t, sp.pi/2) == -1
print("Confirmed")
```

## A pitfall

The parameter is the time of tracing, not the $x$-coordinate.
A curve can have a vertical tangent at a value of $t$ where

$$
\dfrac{dx}{dt}=0
\qquad\text{and}\qquad
\dfrac{dy}{dt}\neq 0.
$$

## Summary

The slope is the quotient of the two velocities.
The length is the integral of the speed.
Tame the cycloid with the identity

$$
1-\cos t=2\sin^2\dfrac{t}{2}.
$$
