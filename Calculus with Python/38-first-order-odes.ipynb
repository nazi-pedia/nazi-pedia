The order of a differential equation is the order of the highest derivative that appears in it.

A separable equation has the form $g(y)\,y'=f(x)$. Integrate each side with respect to its own variable.

A linear equation has the form $y'+P(x)y=Q(x)$. Multiply through by the integrating factor $e^{\int P}$.

An exact equation $M\,dx+N\,dy=0$ is exact when $M_y=N_x$. Build a potential by integrating $M$ with respect to $x$, then correct the result by a function of $y$ alone.

A Bernoulli equation has the form $y'+Py=Qy^n$. The substitution $v=y^{1-n}$ turns it into a linear equation.


## Example 1 — Separable

### Hand solution

Solve $y'=2xy$ with $y(0)=1$. Separate the variables, integrate, and apply the initial condition:

$$
\begin{aligned}
\dfrac{dy}{y} &= 2x\,dx, \\
\ln|y| &= x^2+C, \\
y &= Ae^{x^2}, \\
y(0)=1 &\implies A=1, \\
y &= e^{x^2}.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Independent variable and unknown function.
x = sp.symbols("x")
y = sp.Function("y")
# Separable equation y' = 2xy.
ode = sp.Eq(y(x).diff(x), 2*x*y(x))
# Solve the initial-value problem y(0) = 1.
sol = sp.dsolve(ode, y(x), ics={y(0): 1})
print(sol)
# The solution must be exp(x**2).
assert sp.simplify(sol.rhs - sp.exp(x**2)) == 0
print("Confirmed")
```

## Example 2 — Linear

### Hand solution

Solve $y'+2y=4$ with $y(0)=0$. The integrating factor is $e^{2x}$.

$$
\begin{aligned}
(ye^{2x})' &= 4e^{2x}, \\
ye^{2x} &= 2e^{2x}+C, \\
y &= 2+Ce^{-2x}, \\
y(0)=0 &\implies C=-2, \\
y &= 2\bigl(1-e^{-2x}\bigr).
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Independent variable and unknown function.
x = sp.symbols("x")
y = sp.Function("y")
# Linear equation y' + 2y = 4.
ode = sp.Eq(y(x).diff(x) + 2*y(x), 4)
# Solve the initial-value problem y(0) = 0.
sol = sp.dsolve(ode, y(x), ics={y(0): 0})
# Hand solution 2(1 - exp(-2x)).
hand = 2*(1 - sp.exp(-2*x))
print(sol)
# The solver result must match the hand solution.
assert sp.simplify(sol.rhs - hand) == 0
print("Confirmed")
```

## Example 3 — Exact

### Hand solution

Consider $(2x+y)\,dx+(x+2y)\,dy=0$.
Here $M_y=1=N_x$, so the equation is exact. Integrate $M$ in $x$ and match $\phi_y$ with $N$:

$$
\begin{aligned}
\phi &= \int (2x+y)\,dx = x^2+xy+h(y), \\
\phi_y &= x+h'(y) = x+2y, \\
h'(y) &= 2y, \\
h(y) &= y^2, \\
x^2+xy+y^2 &= C.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Independent variable and unknown function.
x = sp.symbols("x")
y = sp.Function("y")
# Exact equation (2x + y) + (x + 2y) y' = 0.
ode = sp.Eq((2*x + y(x)) + (x + 2*y(x))*y(x).diff(x), 0)
# General solution from the solver.
sol = sp.dsolve(ode, y(x))
print(sol)
# Potential phi = x^2 + xy + y^2.
phi = x**2 + x*y(x) + y(x)**2
# The total x-derivative of the potential vanishes on solutions.
assert sp.simplify(sp.diff(phi, x)) == 0 or sp.simplify(sp.diff(phi, x).subs(y(x).diff(x), sp.solve(ode, y(x).diff(x))[0])) == 0
# The potential is constant along solutions of the equation.
residual = sp.diff(x**2 + x*y(x) + y(x)**2, x)
slope = sp.solve(ode, y(x).diff(x))[0]
# Substitute that slope into the derivative of the potential.
assert sp.simplify(residual.subs(y(x).diff(x), slope)) == 0
print("Confirmed")
```

## Example 4 — Bernoulli

### Hand solution

Solve $y'+y=y^2$ with $y(0)=\dfrac{1}{2}$.
Set $v=y^{-1}$. Then $y=\dfrac{1}{v}$ and $y'=-\dfrac{v'}{v^2}$. Substitute, multiply through by $-v^2$, and use the integrating factor $e^{-x}$:

$$
\begin{aligned}
-\dfrac{v'}{v^2}+\dfrac{1}{v} &= \dfrac{1}{v^2}, \\
v'-v &= -1, \\
(ve^{-x})' &= -e^{-x}, \\
v &= 1+Ce^{x}, \\
y &= \dfrac{1}{1+Ce^{x}}, \\
y(0)=\dfrac{1}{2} &\implies C=1, \\
y &= \dfrac{1}{1+e^{x}}.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Independent variable and unknown function.
x = sp.symbols("x")
y = sp.Function("y")
# Bernoulli equation y' + y = y^2.
ode = sp.Eq(y(x).diff(x) + y(x), y(x)**2)
# Solve with y(0) = 1/2.
sol = sp.dsolve(ode, y(x), ics={y(0): sp.Rational(1, 2)})
# Hand solution 1/(1 + exp(x)).
hand = 1/(1 + sp.exp(x))
print(sol)
# The solver result must match the hand solution.
assert sp.simplify(sol.rhs - hand) == 0
print("Confirmed")
```

## A pitfall

Apply the initial condition to the family of solutions. Apply it before the integration only when separation has already isolated the constants from the variables.
The solution of the initial-value problem is a solution of the differential equation that also passes through the initial point.

## Summary

Classify the equation first: separable, linear, exact, or Bernoulli.
Then apply one method. At the end, differentiate the answer and substitute it back into the equation.
