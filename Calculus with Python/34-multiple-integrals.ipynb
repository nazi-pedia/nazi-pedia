A double integral is the volume under a surface, or the mass of a lamina.
The inner limit may depend on the outer variable. The outer limit must be a number.

$$
\begin{aligned}
 \iint_{D} f(x,y)\,dA.
\end{aligned}
$$

A triple integral is the same idea over a solid region.
The volume of the tetrahedron $x\ge 0$, $y\ge 0$, $z\ge 0$, $x+y+z\le 1$ equals $\dfrac{1}{6}$, and that number is the standard check on the order of the limits.


## Example 1 — A rectangle

### Hand solution

$$
\begin{aligned}
 \int_{0}^{1}\int_{0}^{1}(x+y)\,dx\,dy
 &=
 \int_{0}^{1}\left[\dfrac{x^{2}}{2}+yx\right]_{0}^{1}\,dy
 =
 \int_{0}^{1}\left(\dfrac{1}{2}+y\right)\,dy
 =
 \left[\dfrac{y}{2}+\dfrac{y^{2}}{2}\right]_{0}^{1}
 =
 1.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Integrate x + y over the unit square, x inner.
x, y = sp.symbols("x y")
exact = sp.integrate(x + y, (x, 0, 1), (y, 0, 1))
print(exact)
assert exact == 1
print("Confirmed")
```

## Example 2 — A triangle, with a dependent limit

### Hand solution

The region $D$ is $0\le x\le 1$ and $0\le y\le x$. The integrand is $1$, so the value is the area of the triangle, $\dfrac{1}{2}$.

$$
\begin{aligned}
 \int_{0}^{1}\int_{0}^{x} 1\,dy\,dx
 &=
 \int_{0}^{1} x\,dx
 =
 \dfrac{1}{2}.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Area of the triangle 0 <= y <= x <= 1.
x, y = sp.symbols("x y")
exact = sp.integrate(1, (y, 0, x), (x, 0, 1))
print(exact)
assert exact == sp.Rational(1, 2)
print("Confirmed")
```

## Example 3 — A tetrahedron

### Hand solution

Integrate $1$ over $x\ge 0$, $y\ge 0$, $z\ge 0$, $x+y+z\le 1$.

$$
\begin{aligned}
 \int_{0}^{1}\int_{0}^{1-x}\int_{0}^{1-x-y} 1\,dz\,dy\,dx
 &=
 \int_{0}^{1}\int_{0}^{1-x}(1-x-y)\,dy\,dx,\\
 \left[(1-x)y-\dfrac{y^{2}}{2}\right]_{0}^{1-x}
 &=
 \dfrac{(1-x)^{2}}{2},\\
 \int_{0}^{1}\dfrac{(1-x)^{2}}{2}\,dx
 &=
 \dfrac{1}{2}\cdot\dfrac{1}{3}
 =
 \dfrac{1}{6}.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Volume of the tetrahedron x, y, z >= 0 and x + y + z <= 1.
x, y, z = sp.symbols("x y z")
exact = sp.integrate(1, (z, 0, 1 - x - y), (y, 0, 1 - x), (x, 0, 1))
print(exact)
assert exact == sp.Rational(1, 6)
print("Confirmed")
```

## Example 4 — Mass from a density

### Hand solution

The lamina is $0\le x\le 1$, $0\le y\le 2$, with density $\rho=xy$.

$$
\begin{aligned}
 m
 &=
 \int_{0}^{2}\int_{0}^{1} xy\,dx\,dy
 =
 \int_{0}^{2} y\left[\dfrac{x^{2}}{2}\right]_{0}^{1}\,dy
 =
 \int_{0}^{2}\dfrac{y}{2}\,dy
 =
 1.
\end{aligned}
$$

### Python recheck


```python
# Load SymPy for the symbolic check.
import sympy as sp
# Mass of the rectangle with density rho = xy.
x, y = sp.symbols("x y")
mass = sp.integrate(x*y, (x, 0, 1), (y, 0, 2))
print(mass)
assert mass == 1
print("Confirmed")
```

## A pitfall

If the inner integral is taken with respect to a variable on which its own limit depends, the order $dx\,dy$ has been confused with the order of the pairs passed to `integrate`.
In SymPy, the first pair $(x,a,b)$ is the innermost integral.

## Summary

Write the limits from the geometry, not from habit.
The integrand $1$ returns area or volume, and it checks the limits.
