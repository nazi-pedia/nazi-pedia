The tangent line is the best linear approximation of a function near the point of tangency.

$$
\begin{aligned}
 L(x) &= f(a)+f'(a)(x-a).
\end{aligned}
$$

A differential is the same idea written with a small increment: $dy=f'(x)\,dx$.
The true change is $\Delta y$. The quantity $dy$ approximates that change.

**Rolle's theorem.** If $f$ is continuous on $[a,b]$, differentiable on $(a,b)$, and $f(a)=f(b)$, then there is a point at which $f'(c)=0$.

**The Mean Value Theorem.** Under the same continuity and differentiability, there is a point at which

$$
\begin{aligned}
 f'(c) &= \dfrac{f(b)-f(a)}{b-a}.
\end{aligned}
$$

Some tangent line has the same slope as the secant through the two endpoints.



## Example 1 — Linearizing a square root

### Hand solution

Take $f(x)=\sqrt{x}$ at $a=9$. Then $f(9)=3$ and $f'(x)=\frac{1}{2\sqrt{x}}$, so $f'(9)=\frac{1}{6}$.

$$
\begin{aligned}
 L(x) &= 3+\dfrac{1}{6}(x-9), \\
 \sqrt{9.1}
 &\approx 3+\dfrac{0.1}{6} \\
 &= 3+\dfrac{1}{60} \\
 &= \dfrac{181}{60}.
\end{aligned}
$$

The approximation is a little larger than the true value, because $f''(x)<0$ and the graph lies under the tangent line.

### Python recheck


```python
import sympy as sp

# f(x) = sqrt(x) at a = 9. Hand approximation of sqrt(9.1): 181/60, and it lies above the true value.
x = sp.symbols("x", positive=True)
f = sp.sqrt(x)
L = f.subs(x, 9) + sp.diff(f, x).subs(x, 9)*(x - 9)
approx = sp.simplify(L.subs(x, sp.Rational(91, 10)))
true = sp.sqrt(sp.Rational(91, 10))
print("linearization:", approx)
print("true value:", true)
print("difference:", sp.N(approx - true, 8))
# The linearization equals 181/60 and is larger than the true square root.
assert approx == sp.Rational(181, 60)
assert approx > true
print("confirmed")
```

## Example 2 — A differential against the true change

### Hand solution

Take $y=x^3$. From $x=2$ to $x=2.1$, the increment is $dx=0.1$.

$$
\begin{aligned}
 dy
 &= 3x^2\,dx \\
 &= 3\cdot 4\cdot 0.1 \\
 &= 1.2.
\end{aligned}
$$

The true change is

$$
\begin{aligned}
 \Delta y
 &= (2.1)^3-8 \\
 &= 9.261-8 \\
 &= 1.261.
\end{aligned}
$$

The difference $0.061$ comes from the degree-two and degree-three terms in the expansion. The differential has discarded those terms.

### Python recheck


```python
import sympy as sp

# y = x**3, dx = 1/10 at x = 2. Hand results: dy = 6/5 and Delta y = 1261/1000.
dx = sp.Rational(1, 10)
dy = 3*(2**2)*dx
delta = sp.Rational(21, 10)**3 - 8
print("dy =", dy)
print("Delta y =", delta)
# The assert checks both the differential and the true change.
assert dy == sp.Rational(6, 5)
assert delta == sp.Rational(1261, 1000)
print("confirmed")
```

## Example 3 — Rolle's theorem

### Hand solution

Take $f(x)=x^2-1$ on $[-1,1]$.
Then $f(-1)=f(1)=0$. The function is a polynomial, so the hypotheses of Rolle's theorem hold.

$$
\begin{aligned}
 f'(x) &= 2x, \\
 2x &= 0, \\
 x &= 0,
\end{aligned}
$$

and $0$ lies inside the interval. The tangent line there is horizontal, right between the two roots.

### Python recheck


```python
import sympy as sp

# f(x) = x**2 - 1 on [-1, 1]. Hand result: the Rolle point is c = 0.
x = sp.symbols("x")
f = x**2 - 1
# The endpoint values agree, as Rolle's theorem requires.
assert f.subs(x, -1) == f.subs(x, 1) == 0
c = sp.solve(sp.diff(f, x), x)[0]
print("c =", c)
# The assert checks that the critical point is 0.
assert c == 0
print("confirmed")
```

## Example 4 — The Mean Value Theorem

### Hand solution

Take $f(x)=x^2$ on $[0,2]$.

$$
\begin{aligned}
 \dfrac{f(2)-f(0)}{2-0}
 &= \dfrac{4}{2} \\
 &= 2, \\
 f'(c)
 &= 2c \\
 &= 2, \\
 c &= 1.
\end{aligned}
$$

The point $c=1$ lies in $(0,2)$. The secant from $(0,0)$ to $(2,4)$ has slope $2$, and the tangent line at $x=1$ has slope $2$ as well.

### Python recheck


```python
import sympy as sp

# f(x) = x**2 on [0, 2]. Hand result: secant slope 2, and c = 1.
x = sp.symbols("x")
f = x**2
secant = (f.subs(x, 2) - f.subs(x, 0))/2
c = sp.solve(sp.Eq(sp.diff(f, x), secant), x)[0]
print("secant slope:", secant, " c:", c)
# The secant slope is 2 and the Mean Value Theorem point is 1.
assert secant == 2 and c == 1
print("confirmed")
```

## Example 5 — Where the Mean Value Theorem stays silent

### Hand solution

Take $f(x)=|x|$ on $[-1,1]$.
Then $f(-1)=f(1)=1$, so the secant slope is $0$.
The derivative does not exist at $x=0$, and $f'$ is zero at no other point: the left-hand value is $-1$ and the right-hand value is $+1$.
The theorem makes no promise, because differentiability on the open interval has failed.
Rolle's theorem is silent for the same reason.

### Python recheck


```python
import sympy as sp

# f(x) = |x|. Hand result: the one-sided derivatives at 0 are -1 and +1.
x = sp.symbols("x", real=True)
f = sp.Abs(x)
print("left-hand derivative:", sp.limit((f.subs(x, 0 + sp.Symbol("h")) - 0)/sp.Symbol("h"), sp.Symbol("h"), 0, dir="-"))
left = sp.limit((sp.Abs(0 + x) - 0)/x, x, 0, dir="-")
right = sp.limit((sp.Abs(0 + x) - 0)/x, x, 0, dir="+")
print(left, right)
# The assert checks that the one-sided derivatives are not equal.
assert left == -1 and right == 1
print("the two one-sided derivatives are not equal")
print("confirmed")
```

## A pitfall

A linearization is trustworthy near the base point and misleading far from that point.
For $\sqrt{16.1}$, take the base point to be $16$, not $9$.

## Summary

The tangent line is a machine for approximation.
Rolle's theorem is the special case of the Mean Value Theorem in which the secant slope is zero.
Both theorems assume differentiability on the interior of the interval.
