# ۲۳ · دنباله

| | |
|---|---|
| **دوره** | ریاضی عمومی ۲ |
| **پیش‌نیاز** | ۰۶ |

دنباله فهرستی مرتب است: \(a_1,a_2,a_3,\ldots\).
حد دنباله، اگر باشد، عددی است که جمله‌ها به آن می‌چسبند.

\[
\lim_{n\to\infty}a_n=L
\]

دنبالهٔ یکنوا و کراندار همگراست. این قضیه وجود حد را می‌دهد، گاهی پیش از آنکه فرمول حد را حدس بزنید.
بعد از حدس، معادلهٔ نقطهٔ ثابت حدس را امتحان می‌کند.



## مثال ۱ — یک کسر گویا

### حل ریاضی

\[
a_n=\frac{n+1}{n}=1+\frac{1}{n}\to 1
\]

### بازبینی با پایتون



```python
import sympy as sp
n = sp.symbols("n", integer=True, positive=True)
assert sp.limit((n + 1)/n, n, sp.oo) == 1
print(sp.limit((n + 1)/n, n, sp.oo))
print("تایید شد")

```

## مثال ۲ — نوسانِ میرا

### حل ریاضی

\[
a_n=\frac{(-1)^n}{n}
\]

قدر مطلق \(1/n\) به صفر می‌رود، پس خود دنباله هم به صفر می‌رود.
علامت عوض می‌شود، ولی دامنهٔ نوسان می‌میرد.

### بازبینی با پایتون



```python
import sympy as sp
n = sp.symbols("n", integer=True, positive=True)
term = ((-1)**n)/n
assert sp.simplify(sp.Abs(term) - 1/n) == 0
assert sp.limit(1/n, n, sp.oo) == 0
print("قدر مطلق:", sp.simplify(sp.Abs(term)), "حد کران:", sp.limit(1/n, n, sp.oo))
print("تایید شد")

```

## مثال ۳ — هندسی

### حل ریاضی

\[
a_n=\left(\frac{1}{2}\right)^n\to 0
\]

اگر \(|r|<1\)، آن‌گاه \(r^n\to 0\).
اگر \(|r|>1\)، قدر مطلق بی‌کران می‌شود.
\(r=1\) ثابت می‌ماند و \(r=-1\) بین ۱ و \(-1\) می‌پرد و حد ندارد.

### بازبینی با پایتون



```python
import sympy as sp
n = sp.symbols("n", integer=True, positive=True)
assert sp.limit((sp.Rational(1, 2))**n, n, sp.oo) == 0
assert sp.limit(1**n, n, sp.oo) == 1
assert sp.limit((sp.Rational(3, 2))**n, n, sp.oo) == sp.oo
print("1/2 به صفر، 1 به یک، 3/2 به بی‌نهایت")
print("تایید شد")

```

## مثال ۴ — دنبالهٔ بازگشتی و نقطهٔ ثابت

### حل ریاضی

\[
a_1=\sqrt{2},\qquad a_{n+1}=\sqrt{2+a_n}
\]

اگر حد \(L\) وجود داشته باشد، از دو طرف معادلهٔ بازگشتی حد می‌گیریم.

\[
L=\sqrt{2+L},\qquad L^2=2+L,\qquad L^2-L-2=0
\]
\[
(L-2)(L+1)=0
\]

حد نامنفی است، پس \(L=2\).
جمله‌ها از \(\sqrt{2}\approx 1.414\) شروع می‌شوند و تا وقتی کمتر از ۲ باشند،
\(a_{n+1}^2-a_n^2=2+a_n-a_n^2=(2-a_n)(1+a_n)>0\)، پس صعودی و از بالا به ۲ کراندارند.
قضیه وجود حد را تمام می‌کند و معادله مقدارش را.

### بازبینی با پایتون



```python
import sympy as sp
L = sp.symbols("L", nonnegative=True)
fixed = sp.solve(sp.Eq(L, sp.sqrt(2 + L)), L)
term = sp.sqrt(2)
print("نقطه ثابت:", fixed)
for k in range(8):
    term = sp.sqrt(2 + term)
    print(k + 2, sp.N(term, 8))
assert fixed == [2]
assert abs(term - 2) < 10**(-2)
print("تایید شد")

```

## نکته

حد جملهٔ عمومی با حد سری فرق دارد.
\(1/n\) به صفر می‌رود، ولی جمع \(1/n\) واگراست. درس ۲۴ این تفاوت را حساب می‌کند.

## جمع‌بندی

اول \(|a_n|\) را ببینید.
برای بازگشتی، یکنوایی و کران وجود را می‌دهند و نقطهٔ ثابت مقدار را.
