# ۱۵ · روش نیوتن

| | |
|---|---|
| **دوره** | ریاضی عمومی ۱ |
| **پیش‌نیاز** | ۱۱ |

اگر \(f(r)=0\) و مماس در حدس \(x_n\) را تا برخورد با محور \(x\) ادامه دهید، حدس بعدی به دست می‌آید.

\[
x_{n+1}=x_n-\frac{f(x_n)}{f'(x_n)}
\]

شرط کار کردن: \(f'(x_n)\neq 0\)، و حدس اول آن‌قدر به ریشه نزدیک باشد که مماس شما را پرت نکند.
هر گام، خطی‌سازی درس ۱۱ است که صفرِ خط را به‌جای صفرِ خم می‌گیرد.

روش، ریشه را تقریب می‌زند. اگر ریشه فرمول بسته داشته باشد، تقریب را با آن بسنجید.
اگر نداشته باشد، \(|f(x_n)|\) را کوچک شدن بخواهید.



## مثال ۱ — جذر ۲، با فرمول بسته در کنار تقریب

### حل ریاضی

\(f(x)=x^2-2\) و \(f'(x)=2x\).

\[
x_{n+1}=x_n-\frac{x_n^2-2}{2x_n}=\frac{x_n+2/x_n}{2}
\]

از \(x_0=1\):

\[
x_1=\frac{1+2}{2}=\frac{3}{2}
\]
\[
x_2=\frac{3/2+2/(3/2)}{2}=\frac{3/2+4/3}{2}=\frac{17/6}{2}=\frac{17}{12}
\]

\(\sqrt{2}\approx 1.41421356\) و \(17/12=1.4166\ldots\). دو گام، سه رقم درست داده است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = x**2 - 2
fp = sp.diff(f, x)

def newton(guess, steps):
    values = [guess]
    for _ in range(steps):
        values.append(sp.simplify(values[-1] - f.subs(x, values[-1])/fp.subs(x, values[-1])))
    return values

seq = newton(1, 4)
for n, value in enumerate(seq):
    print(n, value, sp.N(value, 12))
assert seq[1] == sp.Rational(3, 2)
assert seq[2] == sp.Rational(17, 12)
assert abs(seq[4] - sp.sqrt(2)) < 10**(-8)
print("تایید شد")

```

## مثال ۲ — ریشه‌ای بدون فرمول رادیکالی ساده

### حل ریاضی

\(f(x)=\cos x-x\). مشتق \(f'(x)=-\sin x-1\).

\[
x_{n+1}=x_n-\frac{\cos x_n-x_n}{-\sin x_n-1}
=x_n+\frac{\cos x_n-x_n}{\sin x_n+1}
\]

از \(x_0=0.5\) یک گام را با دست، با چهار رقم، حساب می‌کنیم.

\[
\cos 0.5\approx 0.8776,\qquad \sin 0.5\approx 0.4794
\]
\[
x_1\approx 0.5+\frac{0.8776-0.5}{0.4794+1}=0.5+\frac{0.3776}{1.4794}\approx 0.755
\]

ریشهٔ واقعی نزدیک \(0.739085\) است. گام‌های بعدی در پایتون دقیق می‌شوند.
وجود ریشه از درس ۰۵ می‌آید: \(f(0)=1>0\) و \(f(1)=\cos 1-1<0\).

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = sp.cos(x) - x
fp = sp.diff(f, x)
guess = sp.Rational(1, 2)
for n in range(5):
    guess = sp.N(guess - f.subs(x, guess)/fp.subs(x, guess), 20)
    print(n + 1, guess, "f =", sp.N(f.subs(x, guess), 8))
assert f.subs(x, 0) > 0 > f.subs(x, 1)
assert abs(f.subs(x, guess)) < 10**(-12)
print("تایید شد")

```

## مثال ۳ — شکست وقتی مماس افقی است

### حل ریاضی

\(f(x)=x^3-3x+3\)؟ ساده‌تر: \(f(x)=(x-1)^3+1=x^3-3x^2+3x\).
بهتر است مثالی بگیریم که حدس اول دقیقاً مماس افقی بدهد.

\(f(x)=x^3-x-1\). \(f'(x)=3x^2-1=0\) در \(x=\pm 1/\sqrt{3}\).
اگر حدس را همان‌جا بگذارید، مخرج صفر است و روش تعریف نشده است.
ریشهٔ واقعی بین ۱ و ۲ است، چون \(f(1)=-1\) و \(f(2)=5\).
از \(x_0=1.5\) روش سالم است، چون \(f'(1.5)=3\cdot 2.25-1=5.75\neq 0\).

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = x**3 - x - 1
fp = sp.diff(f, x)
bad = 1/sp.sqrt(3)
print("مشتق در حدس بد:", sp.simplify(fp.subs(x, bad)))
guess = sp.Rational(3, 2)
for n in range(6):
    guess = sp.simplify(guess - f.subs(x, guess)/fp.subs(x, guess))
print("از 1.5:", sp.N(guess, 12), "f =", sp.N(f.subs(x, guess), 5))
assert fp.subs(x, bad) == 0
assert abs(f.subs(x, guess)) < 10**(-10)
print("تایید شد")

```

## مثال ۴ — مقایسه با حل دقیق درجهٔ دو

### حل ریاضی

\(x^2-5x+6=0\) ریشه‌های ۲ و ۳ دارد.
روش نیوتن از \(x_0=0\):

\[
f'=2x-5
\]
\[
x_1=0-\frac{6}{-5}=\frac{6}{5}
\]
\[
x_2=\frac{6}{5}-\frac{(36/25-30/5+6)}{12/5-5}
=\frac{6}{5}-\frac{(36/25-150/25+150/25)}{(12/5-25/5)}
\]

صورت: \(\left(\frac{6}{5}\right)^2-5\cdot\frac{6}{5}+6=\frac{36}{25}-6+6=\frac{36}{25}\).
مخرج مشتق: \(\frac{12}{5}-5=\frac{12-25}{5}=-\frac{13}{5}\).

\[
x_2=\frac{6}{5}-\frac{36/25}{-13/5}=\frac{6}{5}+\frac{36}{25}\cdot\frac{5}{13}
=\frac{6}{5}+\frac{36}{65}=\frac{78+36}{65}=\frac{114}{65}
\]

\(114/65\approx 1.753\). به سمت ریشهٔ ۲ می‌رود، نه ۳، چون حدس اول سمت چپ سهمی است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
f = x**2 - 5*x + 6
fp = sp.diff(f, x)
guess = 0
values = [guess]
for _ in range(2):
    guess = sp.simplify(guess - f.subs(x, guess)/fp.subs(x, guess))
    values.append(guess)
print(values)
assert values[1] == sp.Rational(6, 5)
assert values[2] == sp.Rational(114, 65)
print("تایید شد")

```

## نکته

یک گام را با دست حساب کنید تا فرمول در انگشتان بماند.
بعد پایتون گام‌های بعدی را دقیق ادامه دهد.
اگر \(|f|\) کوچک نشود، حدس اول را عوض کنید. الگوریتم را متهم نکنید پیش از آنکه مخرج را دیده باشید.

## جمع‌بندی

نیوتن صفرِ خط مماس را دنبال می‌کند.
کنار هر تقریب، یا مقدار دقیق را بگذارید یا \(|f(x_n)|\) را.
مماس افقی روش را متوقف می‌کند.
