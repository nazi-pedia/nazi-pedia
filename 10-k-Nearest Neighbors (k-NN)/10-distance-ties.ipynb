# Two mornings at the same distance

The six main distances were all different, so sorting had one answer. Add a second morning at distance $3$ from $(2, 2)$.

- A is $(2, 5)$, gaps $(0, 3)$, distance $3$, busy.
- H is $(5, 2)$, gaps $(3, 0)$, distance $3$, quiet.

$k = 1$ asks for the nearest morning, and there are two. Neither is nearer.

A silent sort will still return one of them. Python's sort is stable, so the row that appears earlier in the list survives a tie if the key is only the distance. Store A first and the function says busy. Store H first and the same function says quiet. The mornings did not change. The file order did.

The written rule here: if the nearest distance is shared, report a tie. Do not let storage order become a hidden tie-break. A later, explicit rule can break the tie, but it has to be visible in the same sentence as the prediction.


```python
# Equal distances. The earlier row wins a sort that does not know it is tied.
query = (2, 2)
first = [("A", 2, 5, "busy"), ("H", 5, 2, "quiet")]
second = [("H", 5, 2, "quiet"), ("A", 2, 5, "busy")]

def squared(row):
    return (query[0] - row[1]) ** 2 + (query[1] - row[2]) ** 2

assert squared(first[0]) == squared(first[1]) == 9
order_a = sorted(first, key=squared)[0][0]
order_h = sorted(second, key=squared)[0][0]
assert order_a == "A"
assert order_h == "H"
print("A stored first ->", order_a)
print("H stored first ->", order_h)
```

## A pitfall

Looking at more decimal places will not separate $3$ from $3$. The tie is exact. Adding a tiny jitter to one coordinate to "fix" it invents a neighbor the measurements do not support.

## What to carry forward

A and H are equally far from today. $k = 1$ has two legal answers until you write a tie rule. A distance-only sort follows the storage order.
