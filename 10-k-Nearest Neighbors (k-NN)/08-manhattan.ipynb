# Another way to add the gaps

Euclidean distance uses the straight line. Manhattan distance adds the absolute gaps, as if you could only move horizontally and then vertically:

$$
d_1 = |x_1 - x_2| + |y_1 - y_2|
$$

Two candidates for a query at $(2, 2)$:

| Case | Point | Label | Straight line | Sum of gaps |
|---|---|---|---:|---:|
| P | $(5, 6)$ | quiet | $5$ | $7$ |
| R | $(8, 2)$ | busy | $6$ | $6$ |

The straight line to P is the $3$-$4$-$5$ triangle:

$$
3^2 + 4^2 = 9 + 16 = 25 = 5^2
$$

The sum of those same gaps is $3 + 4 = 7$.

R is a horizontal step of $6$. Both measurements agree that its length is $6$.

Straight line: $5 < 6$, so P is nearer, and the label is quiet.

Sum of gaps: $7 > 6$, so R is nearer, and the label is busy.

The metric is part of the model, in the same way $k$ is part of the model.


```python
# Euclidean picks P. Manhattan picks R. Both arithmetic checks are exact.
query = (2, 2)
candidates = {"P": (5, 6, "quiet"), "R": (8, 2, "busy")}

def euclidean_sq(point):
    return (query[0] - point[0]) ** 2 + (query[1] - point[1]) ** 2

def manhattan(point):
    return abs(query[0] - point[0]) + abs(query[1] - point[1])

assert euclidean_sq(candidates["P"]) == 25
assert euclidean_sq(candidates["R"]) == 36
assert manhattan(candidates["P"]) == 7
assert manhattan(candidates["R"]) == 6
assert euclidean_sq(candidates["P"]) < euclidean_sq(candidates["R"])
assert manhattan(candidates["R"]) < manhattan(candidates["P"])
print("Euclidean nearer", "P", "Manhattan nearer", "R")
```

## A pitfall

Manhattan is not a rough approximation to the straight line. On this pair it reverses the neighbor. If you change the metric, recompute the order. Do not reuse a Euclidean ranking.

## What to carry forward

$d_1$ adds absolute gaps. Here it ranks R ahead of P, while the straight line ranks P ahead of R. Quiet or busy depends on which length you promised to use.
