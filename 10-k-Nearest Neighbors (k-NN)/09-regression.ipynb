# Neighbors can average a number

Replace the labels with a busy-index, a number rather than a word. The distances stay exactly the ones already ranked.

| Case | Distance | Busy-index |
|---|---:|---:|
| A | 3 | 9 |
| B | 4 | 8 |
| C | 5 | 3 |
| D | 6 | 2 |
| E | 8 | 1 |
| F | 13 | 6 |

For an unweighted neighborhood, the prediction is the mean. That is the same fact as a regression leaf. If the prediction is a constant $c$ and the weights are all $1$,

$$
S(c) = \sum_{i=1}^{k} (y_i - c)^2, \qquad
S'(c) = -2\sum_{i=1}^{k} (y_i - c)
$$

$S'(c) = 0$ forces $c$ to be the mean, and $S''(c) = 2k > 0$, so the mean is the minimum.

The three nearest indexes are $9, 8, 3$:

$$
\frac{9+8+3}{3} = \frac{20}{3}
$$


## A weighted mean

With positive weights $w_i = 1/d_i$, the same derivative has a weight in every term:

$$
S'(c) = -2\sum_i w_i(y_i - c) = 0
\qquad\Rightarrow\qquad
c = \frac{\sum_i w_i y_i}{\sum_i w_i}
$$

The second derivative $2\sum w_i$ is positive. The weighted mean is the unique best constant.

For A, B, and C,

$$
\sum w = \frac{1}{3} + \frac{1}{4} + \frac{1}{5} = \frac{20+15+12}{60} = \frac{47}{60}
$$

$$
\sum w y = 9\cdot\frac{1}{3} + 8\cdot\frac{1}{4} + 3\cdot\frac{1}{5}
= 3 + 2 + \frac{3}{5} = \frac{28}{5}
$$

$$
c = \frac{28/5}{47/60} = \frac{28}{5}\cdot\frac{60}{47} = \frac{336}{47}
$$

$336/47 = 7 + 7/47$, which is larger than $20/3 = 6 + 2/3$. The nearer indexes are $9$ and $8$, so giving them more weight pulls the prediction up. The unweighted mean lets C's $3$ pull equally.


```python
# Unweighted mean 20/3. Weighted mean 336/47. The nearer indexes pull it up.
from fractions import Fraction

neighbors = [(3, 9), (4, 8), (5, 3)]
unweighted = Fraction(sum(value for _, value in neighbors), 3)
assert unweighted == Fraction(20, 3)
weight_sum = sum(Fraction(1, distance) for distance, _ in neighbors)
weighted_values = sum(Fraction(1, distance) * value for distance, value in neighbors)
assert weight_sum == Fraction(47, 60)
assert weighted_values == Fraction(28, 5)
predicted = weighted_values / weight_sum
assert predicted == Fraction(336, 47)
assert predicted > unweighted
print("unweighted", unweighted)
print("weighted", predicted)
```

## A pitfall

A weighted mean uses $1/d$, so a distance of zero is still illegal. It also uses the indexes of the neighbors, not their labels. Mixing the busy/quiet vote with this average answers two different questions at once.

## What to carry forward

Average the neighbor values. Equal weights give $20/3$. Weights $1/d$ give $336/47$.
