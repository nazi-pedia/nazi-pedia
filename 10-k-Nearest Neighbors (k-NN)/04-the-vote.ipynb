# Let several neighbors vote

Keep the $k$ nearest labels and count them. The prediction is the label with the larger count.

For two labels, an odd $k$ cannot tie. If the counts were equal, then $k = 2n$ for some whole number $n$, so $k$ would be even. The contrapositive is the useful sentence: an odd $k$ has a strict majority.

The order from the distance lesson is fixed:

$$
\text{busy}, \text{busy}, \text{quiet}, \text{quiet}, \text{quiet}, \text{busy}
$$


## Three concrete values of $k$

$k = 3$ keeps A, B, and C: busy, busy, quiet. Busy has $2$ votes and quiet has $1$. Today is busy.

$k = 5$ also keeps D and E, both quiet. Busy still has $2$ votes. Quiet has $3$. Today is quiet.

Same query, same table, different $k$, different answer. Nothing was miscalculated. $k$ changed the neighborhood.

$k = 6$ keeps F as well. F is busy, so the vote is $3$ against $3$. Six is even, and the odd-$k$ guarantee does not apply. The tie is real.

The written rule for this path: when the counts tie, predict the label of the nearest morning inside that tied neighborhood. The nearest of the six is A, and A is busy. The tie-break says busy. It is a rule, not a third vote. Say that when you report it.


```python
# k = 3 and k = 5 disagree. k = 6 ties, then the nearest label breaks the tie.
order = ["busy", "busy", "quiet", "quiet", "quiet", "busy"]

def vote(k):
    chosen = order[:k]
    return chosen.count("busy"), chosen.count("quiet")

assert vote(3) == (2, 1)
assert vote(5) == (2, 3)
assert vote(6) == (3, 3)
# Odd k cannot split evenly.
assert vote(3)[0] != vote(3)[1]
assert vote(5)[0] != vote(5)[1]
tie_break = order[0]
assert tie_break == "busy"
print("k=3", vote(3), "k=5", vote(5), "k=6", vote(6), "tie-break", tie_break)
```

## A pitfall

Do not describe $k = 5$ as "more accurate" than $k = 3$ because it consulted more rows. It consulted more rows and it changed the answer. Which neighborhood you trust is a decision. A later lesson scores that decision by leave-one-out, not by the size of $k$.

## What to carry forward

$k = 3$ says busy, $2$ to $1$. $k = 5$ says quiet, $3$ to $2$. $k = 6$ is a $3$–$3$ tie, and the nearest-label rule calls it busy. An odd $k$ cannot tie a two-label vote.
