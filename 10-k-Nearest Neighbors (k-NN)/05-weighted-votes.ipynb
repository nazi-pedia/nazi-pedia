# A nearer morning gets a heavier vote

An unweighted vote treats distance $3$ and distance $8$ as the same kind of evidence. A weighted vote gives each neighbor a weight $1/d$. Closer means heavier. The prediction is the label whose weights add up to more.

These weights are not probabilities. They do not need to add to $1$. Compare the sums.

For the five nearest mornings,

$$
\begin{aligned}
\text{busy} &= \frac{1}{3} + \frac{1}{4} = \frac{4}{12} + \frac{3}{12} = \frac{7}{12} \\
\text{quiet} &= \frac{1}{5} + \frac{1}{6} + \frac{1}{8}
\end{aligned}
$$

The common denominator of $5$, $6$, and $8$ is $120$:

$$
\frac{1}{5} = \frac{24}{120}, \quad
\frac{1}{6} = \frac{20}{120}, \quad
\frac{1}{8} = \frac{15}{120}
$$

$$
\text{quiet} = \frac{59}{120}
$$

And $7/12 = 70/120$. Busy outweighs quiet, $70$ to $59$.

The unweighted $k = 5$ vote said quiet, because quiet had three neighbors and busy had two. The weighted vote says busy, because the two busy mornings are the close ones and the three quiet mornings are farther. Both calculations are right. They answer different rules.


```python
# Weighted k = 5: busy 70/120, quiet 59/120. Unweighted k = 5 was quiet.
from fractions import Fraction

neighbors = [("busy", 3), ("busy", 4), ("quiet", 5), ("quiet", 6), ("quiet", 8)]
totals = {"busy": Fraction(0), "quiet": Fraction(0)}
for label, distance in neighbors:
    totals[label] += Fraction(1, distance)
assert totals["busy"] == Fraction(7, 12) == Fraction(70, 120)
assert totals["quiet"] == Fraction(59, 120)
assert totals["busy"] > totals["quiet"]
assert totals["busy"] + totals["quiet"] != 1
print("busy", totals["busy"], "quiet", totals["quiet"])
```

## The same weights at $k = 3$

Only A, B, and C:

$$
\text{busy} = \frac{7}{12}, \qquad \text{quiet} = \frac{1}{5} = \frac{12}{60}, \qquad \frac{7}{12} = \frac{35}{60}
$$

Busy still leads. At $k = 3$ the weighted and unweighted rules agree. The disagreement appears when the farther quiet mornings are allowed in and then down-weighted.


```python
# At k = 3 the weight comparison agrees with the 2-to-1 count.
from fractions import Fraction

busy = Fraction(1, 3) + Fraction(1, 4)
quiet = Fraction(1, 5)
assert busy == Fraction(35, 60)
assert quiet == Fraction(12, 60)
assert busy > quiet
print("k=3 weights", busy, quiet)
```

## A pitfall

$1/d$ is undefined when $d = 0$. A stored morning sitting on top of the query has no finite weight. Do not drop a zero distance into this formula. If a distance is zero, that row is the query itself, or a duplicate, and it needs the rule in the zero-distance lesson.

## What to carry forward

Weight each neighbor by $1/d$ and add. For these five neighbors, busy has weight $70/120$ and quiet has $59/120$. The unweighted count said the opposite.
