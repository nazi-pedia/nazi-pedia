# Score $k$ by hiding one stored row

You cannot score a method by how well it copies a row it is allowed to see. For $k = 1$, the nearest row to a stored morning is that morning itself, at distance $0$, and the label matches by construction. The score would be perfect and empty.

Leave-one-out hides the row being scored. The other rows answer, and the hidden label is revealed afterward.

Four mornings on a line, so every distance is an absolute difference:

| Position | 0 | 1 | 3 | 6 |
|---|---:|---:|---:|---:|
| Label | busy | busy | quiet | quiet |


## $k = 1$, with the row itself removed

Position $0$: the nearest remaining point is $1$, busy. Correct.

Position $1$: distance to $0$ is $1$, distance to $3$ is $2$. The neighbor is busy. Correct.

Position $3$: distance to $1$ is $2$, distance to $6$ is $3$. The neighbor is busy. The hidden label is quiet. Wrong.

Position $6$: the nearest remaining point is $3$, quiet. Correct.

One error out of four. The rate is $1/4$.


## $k = 3$ uses everybody else

Hiding one row of four leaves three. $k = 3$ is a vote of the entire remainder.

- Hide $0$: busy, quiet, quiet. Prediction quiet. Wrong.
- Hide $1$: busy, quiet, quiet. Prediction quiet. Wrong.
- Hide $3$: busy, busy, quiet. Prediction busy. Wrong.
- Hide $6$: quiet, busy, busy. Prediction busy. Wrong.

Four errors out of four. On this line, $k = 1$ beats $k = 3$ by $1/4$ against $1$. The larger neighborhood is not the safer one. It is large enough to see the other class every time.


```python
# Leave-one-out. k = 1 misses once. k = 3 misses every row.
rows = [(0, "busy"), (1, "busy"), (3, "quiet"), (6, "quiet")]

def predict(hidden, k):
    others = [(abs(x - hidden[0]), label) for x, label in rows if x != hidden[0]]
    others.sort()
    chosen = [label for _, label in others[:k]]
    return "busy" if chosen.count("busy") > chosen.count("quiet") else "quiet"

k1_errors = sum(predict(row, 1) != row[1] for row in rows)
k3_errors = sum(predict(row, 3) != row[1] for row in rows)
assert k1_errors == 1
assert k3_errors == 4
print("k=1 errors", k1_errors, "k=3 errors", k3_errors)
```

## A pitfall

Leave-one-out on four points cannot support a subtle choice of $k$. It can reject a $k$ that swallows the whole line. With more rows, repeat the same hide-and-ask procedure. Do not replace it by the training accuracy of $k = 1$, which is automatically perfect whenever every stored position is unique.

## What to carry forward

Hide the row you are scoring. On this line, $k = 1$ has error $1/4$ and $k = 3$ has error $1$.
