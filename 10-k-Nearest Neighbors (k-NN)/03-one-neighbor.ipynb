# One neighbor

$k = 1$ keeps only the nearest stored morning. That morning is A, at distance $3$, and A was busy. Today is called busy.

The other five mornings are not consulted. C is quiet and closer than E, and neither fact enters the prediction. With one neighbor, the prediction is a copy of a single label.


```python
# k = 1 copies the label of the unique nearest case.
cases = [("A", 3, "busy"), ("B", 4, "busy"), ("C", 5, "quiet"),
         ("D", 6, "quiet"), ("E", 8, "quiet"), ("F", 13, "busy")]
nearest = min(cases, key=lambda row: row[1])
assert nearest[0] == "A"
assert nearest[2] == "busy"
print("k = 1", nearest[0], nearest[2])
```

## Shifting the origin does nothing

Add $10$ to every booking count, including today's. Case A moves from $(2, 5)$ to $(12, 15)$. Today moves from $(2, 2)$ to $(12, 12)$. The gaps are

$$
(12 - 12,\ 15 - 12) = (0,\ 3)
$$

the same gaps as before. In symbols, for any shift $t$,

$$
(x + t) - (u + t) = x - u
$$

Every coordinate difference is unchanged, so every distance is unchanged, so the nearest neighbor is unchanged. Moving the zero of the ruler is not a change of neighborhood.

Multiplying one coordinate by $10$ is different. That lesson comes after the vote, because it can change which morning is nearest.


```python
# A shift of 10 bookings leaves every squared distance alone.
query = (2, 2)
raw = (2, 5)
shifted_query = (query[0] + 10, query[1])
shifted_case = (raw[0] + 10, raw[1])
before = (query[0] - raw[0]) ** 2 + (query[1] - raw[1]) ** 2
after = (shifted_query[0] - shifted_case[0]) ** 2 + (shifted_query[1] - shifted_case[1]) ** 2
assert before == after == 9
print("squared distance", before)
```

## A pitfall

$k = 1$ will repeat a strange stored label if that one row happens to be nearest. It has no second vote to outnumber the strange row. The next lesson is the reason people use $k = 3$ or $k = 5$.

## What to carry forward

With $k = 1$, today copies A and is called busy. Adding the same constant to a measurement does not change the neighbor.
