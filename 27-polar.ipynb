# ۲۷ · مختصات قطبی

| | |
|---|---|
| **دوره** | ریاضی عمومی ۲ |
| **پیش‌نیاز** | ۲۶ |

\[
x=r\cos\theta,\qquad y=r\sin\theta
\]

مساحت محصور به \(r=r(\theta)\) از \(\alpha\) تا \(\beta\):

\[
A=\frac{1}{2}\int_{\alpha}^{\beta}r(\theta)^2\,d\theta
\]

طول:

\[
L=\int_{\alpha}^{\beta}\sqrt{r^2+\left(\frac{dr}{d\theta}\right)^2}\,d\theta
\]

شیب در مختصات دکارتی هنوز \(dy/dx\) است و هر دو را نسبت به \(\theta\) می‌گیرید.



## مثال ۱ — دایره به شعاع ۲

### حل ریاضی

\(r=2\).

\[
A=\frac{1}{2}\int_0^{2\pi}4\,d\theta=4\pi
\]

تبدیل یک نقطه: \(\theta=\pi/3\)، \(r=2\).

\[
x=2\cdot\frac{1}{2}=1,\qquad y=2\cdot\frac{\sqrt{3}}{2}=\sqrt{3}
\]

### بازبینی با پایتون



```python
import sympy as sp
theta = sp.symbols("theta")
area = sp.Rational(1, 2)*sp.integrate(2**2, (theta, 0, 2*sp.pi))
x = 2*sp.cos(sp.pi/3)
y = 2*sp.sin(sp.pi/3)
print(area, x, y)
assert area == 4*sp.pi
assert x == 1 and sp.simplify(y - sp.sqrt(3)) == 0
print("تایید شد")

```

## مثال ۲ — قلب‌نما

### حل ریاضی

\(r=1+\cos\theta\).

\[
r^2=1+2\cos\theta+\cos^2\theta=1+2\cos\theta+\frac{1+\cos 2\theta}{2}
=\frac{3}{2}+2\cos\theta+\frac{1}{2}\cos 2\theta
\]

\[
A=\frac{1}{2}\int_0^{2\pi}r^2\,d\theta
=\frac{1}{2}\cdot\frac{3}{2}\cdot 2\pi=\frac{3\pi}{2}
\]

انتگرال کسینوس‌ها روی دورهٔ کامل صفر است.

### بازبینی با پایتون



```python
import sympy as sp
theta = sp.symbols("theta")
r = 1 + sp.cos(theta)
area = sp.simplify(sp.Rational(1, 2)*sp.integrate(r**2, (theta, 0, 2*sp.pi)))
print(area)
assert area == 3*sp.pi/2
print("تایید شد")

```

## مثال ۳ — کجا مماس افقی است؟

### حل ریاضی

\(r=1+\cos\theta\). فرمول شیب:

\[
\frac{dy}{dx}=\frac{r'\sin\theta+r\cos\theta}{r'\cos\theta-r\sin\theta},\qquad r'=-\sin\theta
\]

در \(\theta=\pi/2\)، \(r=1\) و \(r'=-1\)، نه صفر.
صورت \((-1)(1)+(1)(0)=-1\) و مخرج \((-1)(0)-(1)(1)=-1\). شیب \(1\) است.
این نقطه \((0,1)\) است، ولی بالاترین نقطهٔ خم نیست.

مماس افقی یعنی صورت صفر باشد و مخرج نباشد.

\[
\cos\theta+\cos 2\theta=2\cos^2\theta+\cos\theta-1=(2\cos\theta-1)(\cos\theta+1)=0
\]

\(\theta=\pi/3\) یک جواب است. آنجا \(r=3/2\) و \(r'=-\sqrt{3}/2\).
صورت صفر می‌شود و مخرج \(-\sqrt{3}\) است، پس شیب \(0\).
\(\theta=\pi\) هم صورت را صفر می‌کند، ولی آنجا صورت و مخرج هر دو صفرند؛ آن نقطه نوک قلب‌نماست، نه مماس افقی.

### بازبینی با پایتون



```python
import sympy as sp
theta = sp.symbols("theta")
r = 1 + sp.cos(theta)
x = r*sp.cos(theta)
y = r*sp.sin(theta)
dy = sp.diff(y, theta)
dx = sp.diff(x, theta)
at_half = {theta: sp.pi/2}
print("در پی دوم:", dy.subs(at_half), dx.subs(at_half))
assert dy.subs(at_half) == -1 and dx.subs(at_half) == -1
at_top = {theta: sp.pi/3}
print("در پی سوم:", dy.subs(at_top), dx.subs(at_top))
assert sp.simplify(dy.subs(at_top)) == 0
assert sp.simplify(dx.subs(at_top) + sp.sqrt(3)) == 0
print("تایید شد")

```

## مثال ۴ — طول قلب‌نما

### حل ریاضی

\[
r^2+(r')^2=(1+\cos\theta)^2+\sin^2\theta
=1+2\cos\theta+\cos^2\theta+\sin^2\theta
=2+2\cos\theta=4\cos^2(\theta/2)
\]

روی \([0,2\pi]\) باید قدر مطلق را مراعات کرد. نتیجهٔ استاندارد طول \(8\) است:
از ۰ تا \(\pi\)، \(\cos(\theta/2)\ge 0\) و از تقارن، کل طول دو برابر نیمه است.

\[
L=2\int_0^{\pi}2\cos\frac{\theta}{2}\,d\theta
=4\cdot 2\sin\frac{\theta}{2}\Big|_0^{\pi}=8
\]

### بازبینی با پایتون



```python
import sympy as sp
theta = sp.symbols("theta", real=True)
r = 1 + sp.cos(theta)
speed = sp.simplify(sp.sqrt(r**2 + sp.diff(r, theta)**2))
hand_speed = 2*sp.cos(theta/2)
assert sp.simplify(speed**2 - hand_speed**2) == 0
length = 2*sp.integrate(hand_speed, (theta, 0, sp.pi))
numeric = sp.N(sp.Integral(speed, (theta, 0, 2*sp.pi)))
print(speed, "->", length, "عددی:", numeric)
assert sp.simplify(length - 8) == 0
assert abs(numeric - 8) < 10**(-10)
print("تایید شد")

```

## نکته

\(r\) منفی در قطبی، نقطه را به سمت مقابل می‌فرستد.
پیش از مساحت، بازه‌ای را انتخاب کنید که خم را درست یک بار بپوشاند. دو بار پیمودن، مساحت را دو برابر دروغین می‌کند.
