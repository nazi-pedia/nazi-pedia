# Order, and a tie

Sorting the four totals from largest to smallest puts Grace (30), then Linus (23), then Ada (20), then Alan (16).

The day totals are Monday 18, Tuesday 18, Wednesday 19, Thursday 17, Friday 17. Ranked so that the busiest day is rank 1, Wednesday stands alone. The two 18s share the next two places, 2 and 3, and the average of those places is $2.5$. The two 17s share places 4 and 5, and the average is $4.5$.

With `method="min"`, both 18s receive 2, and both 17s receive 4. The next rank after a tie jumps.

The default rank runs in the other direction: the smallest value receives rank 1. Wednesday, the busiest day, then has rank 5. Thursday and Friday, the two 17s, share rank $1.5$.


```python
# The reading desk for the week of 5 October 2026. Monday is the morning note.
import pandas as pd

def week_frame():
    return pd.DataFrame(
        {
            "Mon": [4, 6, 3, 5],
            "Tue": [5, 6, 0, 7],
            "Wed": [3, 6, 8, 2],
            "Thu": [6, 6, 1, 4],
            "Fri": [2, 6, 4, 5],
        },
        index=["Ada", "Grace", "Alan", "Linus"],
    )
# Largest person first. Day ranks with the busy day as 1, then the default.
week = week_frame()
order = week.sum(axis=1).sort_values(ascending=False)
days = week.sum(axis=0)
busy = days.rank(ascending=False, method="average")
busy_min = days.rank(ascending=False, method="min")
quiet_first = days.rank(method="average")
print(list(order.index))
print(list(busy.astype(float)), list(quiet_first.astype(float)))
assert list(order.index) == ["Grace", "Linus", "Ada", "Alan"]
assert list(order.astype(int)) == [30, 23, 20, 16]
assert float(busy["Wed"]) == 1
assert float(busy["Mon"]) == float(busy["Tue"]) == 2.5
assert float(busy["Thu"]) == float(busy["Fri"]) == 4.5
assert float(busy_min["Mon"]) == 2 and float(busy_min["Thu"]) == 4
assert float(quiet_first["Wed"]) == 5
assert float(quiet_first["Thu"]) == float(quiet_first["Fri"]) == 1.5
```

## Pitfall

`rank()` with no arguments gives rank 1 to Thursday and Friday. Those are the quiet days. A sentence that says "rank 1 read the most" needs `ascending=False`. The default answers "who has the smallest number?"

## Summary

Grace, Linus, Ada, Alan is the order of the week totals. Wednesday is the only day with busy-rank 1. Monday and Tuesday share 2.5. The default rank gives Wednesday a 5.
