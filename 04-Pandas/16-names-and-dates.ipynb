# Names and the five dates

The string methods live on `.str`. `"ada"` becomes `"Ada"` with `title`. Names that start with A are Ada and Alan, two of the four. The lengths are 3, 5, 4, and 5:

$$
3 + 5 + 4 + 5 = 17.
$$

The weekdays are real dates. 9 October 2026 is a Friday, so the Monday that opens the week is 5 October 2026.

| Day | Date | Name |
|---|---|---|
| Mon | 2026-10-05 | Monday |
| Tue | 2026-10-06 | Tuesday |
| Wed | 2026-10-07 | Wednesday |
| Thu | 2026-10-08 | Thursday |
| Fri | 2026-10-09 | Friday |

`.dt.day_name()` reads the name from the timestamp. The Friday date's day-of-month is 9.


```python
# Titles, two names that start with A, and the five dates of the week.
import pandas as pd

names = pd.Series(["ada", "grace", "alan", "linus"])
titled = names.str.title()
print(list(titled), int(titled.str.startswith("A").sum()), int(names.str.len().sum()))
assert list(titled) == ["Ada", "Grace", "Alan", "Linus"]
assert int(titled.str.startswith("A").sum()) == 2
assert int(names.str.len().sum()) == 17
when = pd.Series(
    pd.to_datetime(
        ["2026-10-05", "2026-10-06", "2026-10-07", "2026-10-08", "2026-10-09"]
    )
)
assert list(when.dt.day_name()) == [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
]
assert int(when.dt.day.iloc[4]) == 9
assert int(when.dt.month.iloc[0]) == 10
```

## Pitfall

`startswith` on a plain string works. On a Series it is `names.str.startswith`. Without `.str`, pandas looks for a Series method of that name and does not find the character test. The date accessor is the same idea: `.dt`, not `.str`, once the values are timestamps.

## Summary

`title` restores Ada, Grace, Alan, and Linus. Two names start with A. The lengths sum to 17. The desk week is Monday 5 October 2026 through Friday 9 October 2026.
