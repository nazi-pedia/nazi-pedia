# A mask is a column of yes and no

A comparison returns a Series of True and False, one entry per label. That Series can pick rows.

Monday pages above 5: only Grace, with 6. Ada's 4, Alan's 3, and Linus's 5 stay out. The cut is strict, so 5 is not above 5.

Cells of at least 6, counted across the whole week:

Ada has Thursday's 6. Grace has five sixes. Alan has Wednesday's 8. Linus has Tuesday's 7.

$$
1 + 5 + 1 + 1 = 8
$$

cells, and their pages sum to

$$
6 + 5 \times 6 + 8 + 7 = 51.
$$


```python
# The reading desk for the week of 5 October 2026. Monday is the morning note.
import pandas as pd

def week_frame():
    return pd.DataFrame(
        {
            "Mon": [4, 6, 3, 5],
            "Tue": [5, 6, 0, 7],
            "Wed": [3, 6, 8, 2],
            "Thu": [6, 6, 1, 4],
            "Fri": [2, 6, 4, 5],
        },
        index=["Ada", "Grace", "Alan", "Linus"],
    )
# Monday above 5, then every cell of at least 6 pages.
week = week_frame()
monday = week.loc[week["Mon"] > 5]
mask = week >= 6
print(list(monday.index), int(mask.sum().sum()), int(week[mask].sum().sum()))
assert list(monday.index) == ["Grace"]
assert int(monday.loc["Grace", "Mon"]) == 6
assert int(mask.sum().sum()) == 8
assert int(week[mask].sum().sum()) == 51
assert list(week.query("Mon > 5").index) == ["Grace"]
```

## Pitfall

`week["Mon"] > 5` is a Series aligned to the readers. Using it on the rows keeps every column of Grace's week, not only Monday. `week >= 6` is a whole table of yes and no. Summing that table of yes and no counts the True cells. It does not add the page numbers. The page numbers are `week[mask]`.

## Summary

Grace is the only reader above 5 pages on Monday. Eight cells in the week are at least 6, and those cells hold 51 pages.
