# Two weeks, one stack

A second week, with the same five columns:

| | Mon | Tue | Wed | Thu | Fri | Total |
|---|---:|---:|---:|---:|---:|---:|
| Ada | 1 | 1 | 1 | 1 | 1 | 5 |
| Grace | 2 | 0 | 2 | 0 | 2 | 6 |
| Alan | 4 | 4 | 0 | 0 | 0 | 8 |
| Linus | 1 | 2 | 3 | 4 | 5 | 15 |

$$
5 + 6 + 8 + 15 = 34,
\qquad
89 + 34 = 123.
$$

`concat` of the two tables along the default axis stacks rows. The result has $4 + 4 = 8$ rows and the same 5 columns.

Stacking a Saturday column beside the week aligns on the reader. Ada, Grace, and Edsger have a Saturday of 1. Alan and Linus have no Saturday cell, so those cells are blank. Edsger has no weekday cell, so his Monday is blank. Ada's Saturday is 1, and her Monday stays 4.


```python
# The reading desk for the week of 5 October 2026. Monday is the morning note.
import pandas as pd

def week_frame():
    return pd.DataFrame(
        {
            "Mon": [4, 6, 3, 5],
            "Tue": [5, 6, 0, 7],
            "Wed": [3, 6, 8, 2],
            "Thu": [6, 6, 1, 4],
            "Fri": [2, 6, 4, 5],
        },
        index=["Ada", "Grace", "Alan", "Linus"],
    )
# Stack a second week under the first. Then align a Saturday column.
week = week_frame()
week2 = pd.DataFrame(
    {
        "Mon": [1, 2, 4, 1],
        "Tue": [1, 0, 4, 2],
        "Wed": [1, 2, 0, 3],
        "Thu": [1, 0, 0, 4],
        "Fri": [1, 2, 0, 5],
    },
    index=["Ada", "Grace", "Alan", "Linus"],
)
both = pd.concat([week, week2])
print(both.shape, int(week2.sum().sum()), int(both.sum().sum()))
assert int(week2.sum().sum()) == 34
assert both.shape == (8, 5)
assert int(both.sum().sum()) == 123
saturday = pd.DataFrame({"Sat": [1, 1, 1]}, index=["Ada", "Grace", "Edsger"])
wide = pd.concat([week, saturday], axis=1)
assert int(wide.loc["Ada", "Mon"]) == 4
assert int(wide.loc["Ada", "Sat"]) == 1
assert pd.isna(wide.loc["Alan", "Sat"])
assert pd.isna(wide.loc["Edsger", "Mon"])
```

## Pitfall

`concat` along columns aligns labels. It does not place Saturday beside Alan just because Alan is the third row of the week and the Saturday table has three rows. Edsger is a label the week does not have, so his weekdays are blank. Alan is a label Saturday does not have, so his Saturday is blank.

## Summary

The second week sums to 34. Under the first week, the stack has 8 rows and 123 pages. A Saturday column joined by name gives Ada a 1 and leaves Alan's Saturday blank.
