# Three ways to build the same table

The week can be built from columns, from row records, or from a rectangular list. The values do not change. What changes is which part of the Python object you name first.

A column dict gives pandas the weekday lists, and the index is passed beside them. A list of records gives one dictionary per reader, so `name` starts as a column and becomes the index with `set_index`. A list of lists gives the rectangle only, so the column names and the index are arguments.

All three objects compare equal with `equals`. Equal means the same index, the same columns, the same values, and the same dtypes.


```python
# Three constructors, one week.
import pandas as pd

from_columns = pd.DataFrame(
    {
        "Mon": [4, 6, 3, 5],
        "Tue": [5, 6, 0, 7],
        "Wed": [3, 6, 8, 2],
        "Thu": [6, 6, 1, 4],
        "Fri": [2, 6, 4, 5],
    },
    index=["Ada", "Grace", "Alan", "Linus"],
)
from_records = pd.DataFrame(
    [
        {"name": "Ada", "Mon": 4, "Tue": 5, "Wed": 3, "Thu": 6, "Fri": 2},
        {"name": "Grace", "Mon": 6, "Tue": 6, "Wed": 6, "Thu": 6, "Fri": 6},
        {"name": "Alan", "Mon": 3, "Tue": 0, "Wed": 8, "Thu": 1, "Fri": 4},
        {"name": "Linus", "Mon": 5, "Tue": 7, "Wed": 2, "Thu": 4, "Fri": 5},
    ]
).set_index("name")
from_lists = pd.DataFrame(
    [
        [4, 5, 3, 6, 2],
        [6, 6, 6, 6, 6],
        [3, 0, 8, 1, 4],
        [5, 7, 2, 4, 5],
    ],
    index=["Ada", "Grace", "Alan", "Linus"],
    columns=["Mon", "Tue", "Wed", "Thu", "Fri"],
)
print(from_columns.equals(from_records), from_columns.equals(from_lists))
assert from_columns.equals(from_records)
assert from_columns.equals(from_lists)
assert list(from_columns.columns) == ["Mon", "Tue", "Wed", "Thu", "Fri"]
assert int(from_columns.sum().sum()) == 89
```

## Pitfall

`set_index("name")` returns a new table. The name column leaves the columns and becomes the index. A later `from_records["name"]` fails, because `name` is no longer a column. `from_records.index` is where those four strings live.

## Summary

Columns, records, and a bare rectangle can build the same week. `equals` is the check. After `set_index`, the reader names are the index.
