# A new column is a sum you already know

The person totals are a new column, `Total`, made from the five days:

$$
20,\ 30,\ 16,\ 23.
$$

Their mean is $89/4$.

Dropping Friday removes the column whose day total is 17, so the remaining pages sum to $89 - 17 = 72$. `drop` returns a new table. The original week still has Friday.

A copy is a second table. Setting Ada's Monday to 10 on the copy changes that cell by $10 - 4 = 6$, so the copy's grand total is $89 + 6 = 95$. The original Monday for Ada stays 4, and the original grand total stays 89.


```python
# The reading desk for the week of 5 October 2026. Monday is the morning note.
import pandas as pd

def week_frame():
    return pd.DataFrame(
        {
            "Mon": [4, 6, 3, 5],
            "Tue": [5, 6, 0, 7],
            "Wed": [3, 6, 8, 2],
            "Thu": [6, 6, 1, 4],
            "Fri": [2, 6, 4, 5],
        },
        index=["Ada", "Grace", "Alan", "Linus"],
    )
# Add the row total, drop Friday on a new table, and edit a copy.
week = week_frame()
week = week.copy()
week["Total"] = week[["Mon", "Tue", "Wed", "Thu", "Fri"]].sum(axis=1)
print(list(week["Total"].astype(int)))
assert list(week["Total"].astype(int)) == [20, 30, 16, 23]
assert int(week["Total"].sum()) == 89
without_friday = week.drop(columns="Fri")
assert "Fri" not in without_friday.columns
assert "Fri" in week.columns
assert int(without_friday[["Mon", "Tue", "Wed", "Thu"]].sum().sum()) == 72
edited = week.copy()
edited.loc["Ada", "Mon"] = 10
assert int(week.loc["Ada", "Mon"]) == 4
assert int(edited.loc["Ada", "Mon"]) == 10
assert int(edited[["Mon", "Tue", "Wed", "Thu", "Fri"]].sum().sum()) == 95
```

## Pitfall

`week.drop("Fri")` looks up the index label `Fri`. The readers are Ada, Grace, Alan, and Linus, so that call raises `KeyError`. Friday is a column. The call that removes it is `drop(columns="Fri")`.

## Summary

The new column is 20, 30, 16, 23. Dropping Friday from a new table leaves 72 pages and leaves the original week intact. A copy can change Ada's Monday to 10 while the desk's Monday stays 4.
