# The label slice includes the end

`loc` selects by label. `iloc` selects by position, counting from zero.

Ada's Monday through Wednesday is

$$
4 + 5 + 3 = 12.
$$

Monday through Tuesday is $4 + 5 = 9$.

`loc["Ada", "Mon":"Wed"]` includes Wednesday, because both ends of a label slice are kept. `iloc[0, 0:3]` also reaches Wednesday, and it does so because position 3 is the first column *not* taken: positions 0, 1, and 2 are Monday, Tuesday, and Wednesday. `iloc[0, 0:2]` stops before Wednesday and sums to 9.

The last column is Friday, position 4. `iloc[3, 4]` is Linus's Friday, 5.


```python
# The reading desk for the week of 5 October 2026. Monday is the morning note.
import pandas as pd

def week_frame():
    return pd.DataFrame(
        {
            "Mon": [4, 6, 3, 5],
            "Tue": [5, 6, 0, 7],
            "Wed": [3, 6, 8, 2],
            "Thu": [6, 6, 1, 4],
            "Fri": [2, 6, 4, 5],
        },
        index=["Ada", "Grace", "Alan", "Linus"],
    )
# Label slices keep the end. Position slices stop before the end.
week = week_frame()
ada_through_wednesday = week.loc["Ada", "Mon":"Wed"]
ada_by_position = week.iloc[0, 0:3]
ada_two_days = week.iloc[0, 0:2]
print(int(ada_through_wednesday.sum()), int(ada_by_position.sum()), int(ada_two_days.sum()))
assert int(ada_through_wednesday.sum()) == 12
assert int(ada_by_position.sum()) == 12
assert int(ada_two_days.sum()) == 9
assert int(week.iloc[3, 4]) == 5
assert list(ada_through_wednesday.index) == ["Mon", "Tue", "Wed"]
```

## Pitfall

Writing `iloc[0, 0:2]` when you mean "through Wednesday" drops Wednesday and returns 9. The label slice that matches the spoken phrase "Monday through Wednesday" is `loc`, and its end is included. Position slices follow the Python rule: the end index is the first position left out.

## Summary

Ada, Monday through Wednesday, is 12. `loc["Ada", "Mon":"Wed"]` includes Wednesday. `iloc[0, 0:3]` is the same three days. `iloc[0, 0:2]` is Monday and Tuesday, summing to 9.
