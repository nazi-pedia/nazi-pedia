# A blank is not a zero

Alan's Tuesday is 0. He was at the desk and read nothing. Linus's Friday, in this lesson's copy, is blank: the page count was not recorded. The other four of Linus's days are $5 + 7 + 2 + 4 = 18$.

The week without that one recorded Friday is

$$
89 - 5 = 84
$$

known pages, in 19 known cells.

`sum` skips blanks. `count` skips blanks. Filling the blank with 0 makes Linus's five-day sum still 18, because $18 + 0 = 18$, and the known-page total stays 84. The count changes from 4 days to 5, so the mean changes:

$$
\dfrac{18}{4} = \dfrac{9}{2},
\qquad
\dfrac{18}{5}.
$$

`dropna()` drops every row that contains a blank. Linus's whole week leaves, and the remaining pages are $20 + 30 + 16 = 66$.


```python
# The reading desk for the week of 5 October 2026. Monday is the morning note.
import pandas as pd

def week_frame():
    return pd.DataFrame(
        {
            "Mon": [4, 6, 3, 5],
            "Tue": [5, 6, 0, 7],
            "Wed": [3, 6, 8, 2],
            "Thu": [6, 6, 1, 4],
            "Fri": [2, 6, 4, 5],
        },
        index=["Ada", "Grace", "Alan", "Linus"],
    )
# One blank Friday. Zero stays zero. Filling and dropping answer different questions.
week = week_frame()
blank = week.copy()
blank.loc["Linus", "Fri"] = float("nan")
print(int(blank.loc["Linus"].sum()), int(blank.loc["Linus"].count()), int(blank.sum().sum()))
assert int(blank.loc["Alan", "Tue"]) == 0
assert int(blank.loc["Linus"].sum()) == 18
assert int(blank.loc["Linus"].count()) == 4
assert int(blank.sum().sum()) == 84
filled = blank.fillna(0)
assert int(filled.loc["Linus"].sum()) == 18
assert int(filled.loc["Linus"].count()) == 5
assert int(filled.sum().sum()) == 84
kept = blank.dropna()
assert list(kept.index) == ["Ada", "Grace", "Alan"]
assert int(kept.sum().sum()) == 66
```

## Pitfall

After `fillna(0)`, Linus's sum is still 18. The sum cannot tell you that Friday was invented. `count` can: 4 recorded days became 5. Alan's Tuesday remains 0 in every version, because 0 was a recorded number.

## Summary

A blank Friday removes 5 pages from the known total, leaving 84. Skipping the blank and filling it with 0 agree on Linus's sum, 18, and disagree on his count, 4 against 5. Dropping incomplete rows leaves 66 pages.
