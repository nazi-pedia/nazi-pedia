# One column, with names

A pandas Series is one column. Each value keeps the label it belongs to. The label is the index. Position is only a convenience.

Monday at the reading desk is the same morning note as the Jupyter path: Ada 4 pages, Grace 6, Alan 3, Linus 5.

$$
4 + 6 + 3 + 5 = 18,
\qquad
\dfrac{18}{4} = \dfrac{9}{2}.
$$

A second column arrives with the labels in another order: Linus 1, Ada 1. Adding by label, and filling a missing label with 0, gives

$$
\text{Ada } 4+1 = 5,\quad
\text{Grace } 6+0 = 6,\quad
\text{Alan } 3+0 = 3,\quad
\text{Linus } 5+1 = 6.
$$

The aligned sum is $5+6+3+6 = 20$. A list would have added by position and paired Linus's 1 with Ada's 4.


```python
# Monday, then a second column aligned by name rather than by position.
import pandas as pd

pages = pd.Series({"Ada": 4, "Grace": 6, "Alan": 3, "Linus": 5}, name="Monday")
print(int(pages.sum()), float(pages.mean()))
assert list(pages.index) == ["Ada", "Grace", "Alan", "Linus"]
assert int(pages["Grace"]) == 6
assert int(pages.sum()) == 18
assert float(pages.mean()) == 4.5
other = pd.Series({"Linus": 1, "Ada": 1})
added = pages.add(other, fill_value=0)
print(int(added["Ada"]), int(added["Grace"]), int(added["Alan"]), int(added["Linus"]))
assert [int(added[name]) for name in ["Ada", "Grace", "Alan", "Linus"]] == [5, 6, 3, 6]
assert int(added.sum()) == 20
```

## Pitfall

`pages.add(other)` without `fill_value` leaves Grace and Alan as missing, because `other` has no row for them. The sum of that result skips those missing labels. `fill_value=0` is the instruction that a missing partner means zero. It is a choice, and the next lessons will keep a real zero separate from a blank.

## Summary

Monday is 4, 6, 3, and 5. The sum is 18 and the mean is $9/2$. Adding another Series matches names. Linus's 1 joins Linus's 5, even when Linus is stored first in the second column.
