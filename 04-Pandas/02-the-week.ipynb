# Five days, four people

A DataFrame is several Series that share an index. Here the index is the reader, and the columns are the five weekdays of 5–9 October 2026.

| | Mon | Tue | Wed | Thu | Fri | Total |
|---|---:|---:|---:|---:|---:|---:|
| Ada | 4 | 5 | 3 | 6 | 2 | 20 |
| Grace | 6 | 6 | 6 | 6 | 6 | 30 |
| Alan | 3 | 0 | 8 | 1 | 4 | 16 |
| Linus | 5 | 7 | 2 | 4 | 5 | 23 |

Monday is $4+6+3+5 = 18$, Tuesday is $5+6+0+7 = 18$, Wednesday is $3+6+8+2 = 19$, Thursday is $6+6+1+4 = 17$, and Friday is $2+6+4+5 = 17$.

$$
20 + 30 + 16 + 23 = 89,
\qquad
18 + 18 + 19 + 17 + 17 = 89.
$$

The shape is 4 rows by 5 columns, so the table holds 20 numbers. Alan's Tuesday is the number 0.


```python
# The reading desk for the week of 5 October 2026. Monday is the morning note.
import pandas as pd

def week_frame():
    return pd.DataFrame(
        {
            "Mon": [4, 6, 3, 5],
            "Tue": [5, 6, 0, 7],
            "Wed": [3, 6, 8, 2],
            "Thu": [6, 6, 1, 4],
            "Fri": [2, 6, 4, 5],
        },
        index=["Ada", "Grace", "Alan", "Linus"],
    )
# Row totals, day totals, and the shape of the week.
week = week_frame()
by_person = week.sum(axis=1)
by_day = week.sum(axis=0)
print(list(by_person.astype(int)), list(by_day.astype(int)), week.shape)
assert list(by_person.astype(int)) == [20, 30, 16, 23]
assert list(by_day.astype(int)) == [18, 18, 19, 17, 17]
assert int(by_person.sum()) == int(by_day.sum()) == 89
assert week.shape == (4, 5)
assert int(week.loc["Alan", "Tue"]) == 0
assert int(week["Mon"].sum()) == 18
```

## Pitfall

`week.sum()` with no axis sums each column, because the default axis is the index. `week.sum(axis=1)` sums each row. Both answers are 89 only after you add the five day totals, or the four person totals. The number 89 is not sitting in a single cell of the table.

## Summary

Four readers, five days, twenty cells, grand total 89. Ada 20, Grace 30, Alan 16, Linus 23. Monday's 18 is the morning note, now sitting in a column of a week.
