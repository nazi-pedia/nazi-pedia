# Edsger bought nothing

The roster has five readers: Ada, Grace, Alan, Linus, and Edsger. The bill from the previous lesson has four names. A left join keeps every roster row. Edsger's spend is missing.

`sum` skips that missing spend, so the roster total is still 38. `count` of the spend column is 4.

Filling Edsger's missing spend with 0 leaves the total at 38, because $38 + 0 = 38$, and the count becomes 5. The outer join of purchases to the full price list adds the grid row, with a missing quantity. Its spend is missing too, and the sum of the spends that exist is still 38.


```python
# Left-join the roster. Edsger has no spend. Grid has no buyer.
import pandas as pd

purchases = pd.DataFrame(
    {
        "name": ["Ada", "Grace", "Alan", "Linus", "Ada"],
        "item": ["lined", "plain", "lined", "plain", "plain"],
        "qty": [2, 1, 4, 2, 1],
    }
)
prices = pd.DataFrame({"item": ["lined", "plain", "grid"], "price": [3, 5, 4]})
roster = pd.DataFrame({"name": ["Ada", "Grace", "Alan", "Linus", "Edsger"]})
bill = purchases.merge(prices, on="item", how="inner")
bill["spend"] = bill["qty"] * bill["price"]
by_name = bill.groupby("name", as_index=False)["spend"].sum()
joined = roster.merge(by_name, on="name", how="left")
edsger = joined.loc[joined["name"] == "Edsger", "spend"]
print(int(joined["spend"].sum()), int(joined["spend"].count()), edsger.isna().all())
assert edsger.isna().all()
assert int(joined["spend"].sum()) == 38
assert int(joined["spend"].count()) == 4
filled = joined.fillna({"spend": 0})
assert int(filled.loc[filled["name"] == "Edsger", "spend"].iloc[0]) == 0
assert int(filled["spend"].sum()) == 38
assert int(filled["spend"].count()) == 5
outer = purchases.merge(prices, on="item", how="outer")
assert outer.loc[outer["item"] == "grid", "qty"].isna().all()
```

## Pitfall

An inner join of the roster to the bill would drop Edsger. The desk would then look like four people, and the question "who bought nothing?" would have no row left to answer it. The left join keeps the person and leaves the spend blank.

## Summary

Edsger is on the roster and absent from the bill. The sum of known spends is 38 across 4 people. Filling his spend with 0 keeps the sum at 38 and changes the count to 5. Grid is the price with no purchase.
