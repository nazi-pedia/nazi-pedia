# Who read the most

The four week totals, once more, before the picture:

$$
\text{Grace } 30,\quad
\text{Linus } 23,\quad
\text{Ada } 20,\quad
\text{Alan } 16.
$$

The axis starts at 0. Grace's bar is the longest. Alan's is the shortest. The gap from Alan to Grace is $30 - 16 = 14$ pages, drawn on a scale that also has room above 30 so the longest bar is not cut off.


```python
# The reading desk for the week of 5 October 2026. Monday is the morning note.
import pandas as pd

def week_frame():
    return pd.DataFrame(
        {
            "Mon": [4, 6, 3, 5],
            "Tue": [5, 6, 0, 7],
            "Wed": [3, 6, 8, 2],
            "Thu": [6, 6, 1, 4],
            "Fri": [2, 6, 4, 5],
        },
        index=["Ada", "Grace", "Alan", "Linus"],
    )
# The order the chart will show.
week = week_frame()
totals = week.sum(axis=1).sort_values(ascending=False)
print(list(totals.index), list(totals.astype(int)))
assert list(totals.astype(int)) == [30, 23, 20, 16]
assert int(totals.iloc[0] - totals.iloc[-1]) == 14
```

```python
import matplotlib.pyplot as plt
plt.rcParams.update({
    "axes.titleweight": "bold",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.edgecolor": "#102a43",
    "axes.labelcolor": "#102a43",
    "xtick.color": "#102a43",
    "ytick.color": "#102a43",
    "text.color": "#102a43",
    "axes.titlecolor": "#102a43",
})
# Four week totals, drawn from zero.
labels = ["Grace   30", "Linus   23", "Ada   20", "Alan   16"]
counts = [30, 23, 20, 16]
colors = ["#0f766e", "#1d4e89", "#1d4e89", "#c2410c"]
positions = [3, 2, 1, 0]
figure, axis = plt.subplots(figsize=(8.2, 3.6))
axis.barh(positions, counts, color=colors, height=0.62)
axis.set_yticks(positions, labels)
axis.set_xlim(0, 40)
axis.set_xlabel("Pages in the week of 5 October 2026")
axis.set_title("Who read the most")
figure.tight_layout()
```

## Pitfall

Starting the axis at 16 would make Alan's bar disappear and would make Grace's lead look larger than 14 pages. The counts are already printed on the labels. The length of the bar is the count, measured from zero.

## Summary

Grace 30, Linus 23, Ada 20, Alan 16. The difference between the longest week and the shortest week is 14 pages.
