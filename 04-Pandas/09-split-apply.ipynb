# Split, then add

`melt` turns the week into 20 rows: one reader, one day, one page count. Four readers times five days is 20.

Grouping those rows by reader adds the five days back:

$$
20,\ 30,\ 16,\ 23.
$$

Grouping by day adds the four readers:

$$
18,\ 18,\ 19,\ 17,\ 17.
$$

The default groupby sorts the labels. Reader order becomes Ada, Alan, Grace, Linus. Day order becomes Fri, Mon, Thu, Tue, Wed. The sums for a given label do not change. `sort=False` keeps the order in which the labels first appear, which for this melted table is Ada, Grace, Alan, Linus, and then Mon through Fri.


```python
# The reading desk for the week of 5 October 2026. Monday is the morning note.
import pandas as pd

def week_frame():
    return pd.DataFrame(
        {
            "Mon": [4, 6, 3, 5],
            "Tue": [5, 6, 0, 7],
            "Wed": [3, 6, 8, 2],
            "Thu": [6, 6, 1, 4],
            "Fri": [2, 6, 4, 5],
        },
        index=["Ada", "Grace", "Alan", "Linus"],
    )
# Long rows, then the same totals grouped two ways.
week = week_frame()
long = week.reset_index(names="name").melt(
    id_vars="name", var_name="day", value_name="pages"
)
by_name = long.groupby("name")["pages"].sum()
by_name_kept = long.groupby("name", sort=False)["pages"].sum()
by_day = long.groupby("day")["pages"].sum()
print(len(long), list(by_name.index), list(by_name_kept.index))
assert len(long) == 20
assert [int(by_name[name]) for name in ["Ada", "Grace", "Alan", "Linus"]] == [20, 30, 16, 23]
assert list(by_name.index) == ["Ada", "Alan", "Grace", "Linus"]
assert list(by_name_kept.index) == ["Ada", "Grace", "Alan", "Linus"]
assert [int(by_day[day]) for day in ["Mon", "Tue", "Wed", "Thu", "Fri"]] == [18, 18, 19, 17, 17]
assert list(by_day.index) == ["Fri", "Mon", "Thu", "Tue", "Wed"]
```

## Pitfall

The printed group order is alphabetical unless you pass `sort=False`. Alan appearing before Grace does not mean Alan read more. It means the letter A in Alan comes before the letter G. Read the sums by label.

## Summary

The long table has 20 rows. Grouped by reader it returns 20, 30, 16, and 23. Grouped by day it returns 18, 18, 19, 17, and 17. Default group labels are sorted.
