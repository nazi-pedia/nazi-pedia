# The bill is quantity times price

Three notebooks are on the price list: lined at 3, plain at 5, and grid at 4. Five purchases are in the ledger.

| Reader | Item | Qty | Price | Spend |
|---|---|---:|---:|---:|
| Ada | lined | 2 | 3 | 6 |
| Grace | plain | 1 | 5 | 5 |
| Alan | lined | 4 | 3 | 12 |
| Linus | plain | 2 | 5 | 10 |
| Ada | plain | 1 | 5 | 5 |

$$
6 + 5 + 12 + 10 + 5 = 38.
$$

Ada's two lines are $6 + 5 = 11$. An inner join keeps a purchase only when its item is on the price list. All five purchases match. Grid is on the price list and in no purchase, so the inner join has no grid row. Grid's price of 4 is not part of 38.


```python
# Join purchases to prices and multiply.
import pandas as pd

purchases = pd.DataFrame(
    {
        "name": ["Ada", "Grace", "Alan", "Linus", "Ada"],
        "item": ["lined", "plain", "lined", "plain", "plain"],
        "qty": [2, 1, 4, 2, 1],
    }
)
prices = pd.DataFrame(
    {"item": ["lined", "plain", "grid"], "price": [3, 5, 4]}
)
bill = purchases.merge(prices, on="item", how="inner")
bill["spend"] = bill["qty"] * bill["price"]
by_name = bill.groupby("name")["spend"].sum()
print(int(bill["spend"].sum()), int(by_name["Ada"]), sorted(bill["item"].unique()))
assert len(bill) == 5
assert list(bill["spend"].astype(int)) == [6, 5, 12, 10, 5]
assert int(bill["spend"].sum()) == 38
assert int(by_name["Ada"]) == 11
assert int(by_name["Grace"]) == 5
assert int(by_name["Alan"]) == 12
assert int(by_name["Linus"]) == 10
assert "grid" not in set(bill["item"])
```

## Pitfall

Multiplying the whole price column, $3 + 5 + 4 = 12$, is not the bill. The bill multiplies each purchase's own quantity by that item's price. Grid contributes nothing here because nobody bought one, not because its price is zero.

## Summary

Five purchases meet two prices. The spends are 6, 5, 12, 10, and 5, totaling 38. Ada spent 11. Grid stays off the inner join.
