# Count, sum, and the largest day

One grouping can report several numbers. For each reader the week has 5 days. The sums are again 20, 30, 16, and 23. The largest day is Ada 6, Grace 6, Alan 8, and Linus 7.

The means are the sums divided by 5:

$$
\dfrac{20}{5} = 4,
\qquad
\dfrac{30}{5} = 6,
\qquad
\dfrac{16}{5},
\qquad
\dfrac{23}{5}.
$$

The mean of all 20 cells is $89/20$. That is not the mean of the four person-means weighted the wrong way: each person has the same 5 days, so the mean of the four means is also $89/20$. Check:

$$
\dfrac{4 + 6 + 16/5 + 23/5}{4} = \dfrac{20/5 + 30/5 + 16/5 + 23/5}{4} = \dfrac{89/5}{4} = \dfrac{89}{20}.
$$


```python
# The reading desk for the week of 5 October 2026. Monday is the morning note.
import pandas as pd

def week_frame():
    return pd.DataFrame(
        {
            "Mon": [4, 6, 3, 5],
            "Tue": [5, 6, 0, 7],
            "Wed": [3, 6, 8, 2],
            "Thu": [6, 6, 1, 4],
            "Fri": [2, 6, 4, 5],
        },
        index=["Ada", "Grace", "Alan", "Linus"],
    )
# count, sum, and max for each reader. The mean follows from the sum.
from fractions import Fraction

week = week_frame()
long = week.reset_index(names="name").melt(
    id_vars="name", var_name="day", value_name="pages"
)
summary = long.groupby("name", sort=False)["pages"].agg(["count", "sum", "max"])
print(summary.astype(int))
assert [int(summary.loc[name, "count"]) for name in ["Ada", "Grace", "Alan", "Linus"]] == [5, 5, 5, 5]
assert [int(summary.loc[name, "sum"]) for name in ["Ada", "Grace", "Alan", "Linus"]] == [20, 30, 16, 23]
assert [int(summary.loc[name, "max"]) for name in ["Ada", "Grace", "Alan", "Linus"]] == [6, 6, 8, 7]
assert Fraction(int(summary.loc["Alan", "sum"]), 5) == Fraction(16, 5)
assert Fraction(int(summary["sum"].sum()), 20) == Fraction(89, 20)
```

## Pitfall

`max` is the largest single day, not the week total. Grace's max is 6 and her total is 30. Alan's max is 8 and his total is 16. The reader with the largest day is not the reader with the largest week.

## Summary

Each reader has 5 days. Totals 20, 30, 16, 23. Largest days 6, 6, 8, 7. The mean of the twenty cells is $89/20$, and it equals the mean of the four person-means because every person has the same number of days.
