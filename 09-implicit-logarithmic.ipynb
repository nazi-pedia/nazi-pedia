# ۰۹ · مشتق ضمنی و مشتق لگاریتمی

| | |
|---|---|
| **دوره** | ریاضی عمومی ۱ |
| **پیش‌نیاز** | ۰۸ |

گاهی \(y\) را نمی‌توان به‌صورت یک فرمول صریح از \(x\) نوشت،
ولی معادله‌ای هست که هر دو را به هم بسته است.
از دو طرف نسبت به \(x\) مشتق بگیرید و هر بار که به \(y\) رسیدید، در \(y'\) ضرب کنید. این زنجیره است.
بعد معادله را برای \(y'\) حل کنید.

مشتق لگاریتمی برای حاصل‌ضرب‌های بلند و برای \(y=u(x)^{v(x)}\) است.
از دو طرف \(\ln\) بگیرید، بعد مشتق.

\[
\frac{y'}{y}=(\ln y)'
\]



```python
import sympy as sp

def check(found, hand, label="نتیجه"):
    found_s = sp.simplify(found)
    hand_s = sp.simplify(hand)
    difference = sp.simplify(found_s - hand_s)
    print(label)
    print("  حل دستی:", hand_s)
    print("  پایتون:", found_s)
    print("  اختلاف:", difference)
    assert difference == 0
    print("  تایید شد")

```

## مثال ۱ — دایره

### حل ریاضی

\[
x^2+y^2=25
\]

\[
2x+2y\,y'=0,\qquad y'=-\frac{x}{y}
\]

در نقطهٔ \((3,4)\) که روی دایره است، چون \(9+16=25\)،

\[
y'=-\frac{3}{4}
\]

شیب شعاع \(4/3\) است و شیب مماس منفیِ وارون آن است. هندسه و مشتق یکی می‌گویند.
در \((3,-4)\) شیب \(+3/4\) است. فرمول بدون \(y\) در مخرج کامل نیست.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
y = sp.Function("y")
slope = sp.idiff(x**2 + y(x)**2 - 25, y(x), x)
print(slope)
assert sp.simplify(slope + x/y(x)) == 0
assert slope.subs({x: 3, y(x): 4}) == -sp.Rational(3, 4)
assert slope.subs({x: 3, y(x): -4}) == sp.Rational(3, 4)
print("تایید شد")

```

## مثال ۲ — خم درجه سه

### حل ریاضی

\[
x^3+y^3=6xy
\]

\[
3x^2+3y^2 y'=6y+6x y'
\]

\[
3y^2 y'-6x y'=6y-3x^2
\]

\[
y'=\frac{6y-3x^2}{3y^2-6x}=\frac{2y-x^2}{y^2-2x}
\]

در \((3,3)\): صورت \(6-9=-3\) و مخرج \(9-6=3\)، پس شیب \(-1\).
نقطه روی خم است: \(27+27=54\) و \(6\cdot 3\cdot 3=54\).

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
y = sp.Function("y")
slope = sp.simplify(sp.idiff(x**3 + y(x)**3 - 6*x*y(x), y(x), x))
hand = (2*y(x) - x**2)/(y(x)**2 - 2*x)
print(slope)
assert sp.simplify(slope - hand) == 0
assert slope.subs({x: 3, y(x): 3}) == -1
print("تایید شد")

```

## مثال ۳ — توان متغیر

### حل ریاضی

\[
y=x^x,\qquad x>0
\]

\[
\ln y=x\ln x
\]

\[
\frac{y'}{y}=\ln x+x\cdot\frac{1}{x}=\ln x+1
\]

\[
y'=x^x(\ln x+1)
\]

در \(x=1\): \(\ln 1+1=1\) و \(1^1=1\)، پس شیب ۱ است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", positive=True)
y = x**x
found = sp.diff(y, x)
hand = x**x * (sp.log(x) + 1)
print(sp.simplify(found))
assert sp.simplify(found - hand) == 0
assert found.subs(x, 1) == 1
print("تایید شد")

```

## مثال ۴ — پایه و توان هر دو متغیر

### حل ریاضی

\[
y=(\sin x)^x,\qquad \sin x>0
\]

\[
\ln y=x\ln\sin x
\]

\[
\frac{y'}{y}=\ln\sin x+x\cdot\frac{\cos x}{\sin x}
\]

\[
y'=(\sin x)^x\left(\ln\sin x+x\cot x\right)
\]

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x")
y = sp.sin(x)**x
found = sp.diff(y, x)
hand = y * (sp.log(sp.sin(x)) + x*sp.cos(x)/sp.sin(x))
assert sp.simplify(found - hand) == 0
print(sp.simplify(found))
print("تایید شد")

```

## مثال ۵ — لگاریتم برای شکستن ضرب

### حل ریاضی

\[
y=\frac{(x^2+1)\sqrt{x+2}}{e^{x}},\qquad x>-2
\]

\[
\ln y=\ln(x^2+1)+\frac{1}{2}\ln(x+2)-x
\]

\[
\frac{y'}{y}=\frac{2x}{x^2+1}+\frac{1}{2(x+2)}-1
\]

\[
y'=y\left(\frac{2x}{x^2+1}+\frac{1}{2(x+2)}-1\right)
\]

لازم نیست اول همه چیز را در یک کسر تنها جمع کنید. شکل ضرب در پرانتز، همان مشتق است.

### بازبینی با پایتون



```python
import sympy as sp
x = sp.symbols("x", real=True)
y = (x**2 + 1)*sp.sqrt(x + 2)*sp.exp(-x)
factor = 2*x/(x**2 + 1) + 1/(2*(x + 2)) - 1
found = sp.diff(y, x)
hand = y * factor
assert sp.simplify(found - hand) == 0
print("مشتق با شکل لگاریتمی یکی است")
print("تایید شد")

```

## نکته

در مشتق ضمنی، \(y\) تابع \(x\) است حتی اگر ننوشته باشید \(y(x)\).
هر جملهٔ شامل \(y\) یک عامل \(y'\) می‌گیرد.
جملهٔ شامل فقط \(x\) نمی‌گیرد.

## جمع‌بندی

ضمنی: مشتق دو طرف، جمع کردن \(y'\)، حل کردن.
