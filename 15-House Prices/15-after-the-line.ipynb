# What the published solutions add

The line in the previous lesson uses two clues: `OverallQual` and `GrLivArea`. The file has 79. The usual write-ups, read against the counts in this path, add the following kinds of work.

They split the token `NA` the way lesson 4 did: no alley, no pool, no garage, and no fireplace are real categories, while lot frontage is an unknown number. Houses 949 and 333 are the reason a single fill-in rule for every basement `NA` is too coarse.

They keep `MSSubClass` as a category. The means of codes 20, 30, and 60 already refuse a straight line through the code.

They treat `Ex`, `Gd`, `TA`, `Fa` as an order, and they often add neighborhood, basement size, and garage size. Neighborhood has 25 names. Replacing a name by a mean price computed from the prices you then score is a copy of the answer. The grade table in this path was refit on the 1168 for that reason.

They model $\ln$ price, which is the contest's scale, and they look at the two Edwards partials, ids 524 and 1299, before trusting a slope per square foot.

After that, a boosted tree or a stack of several models can use interactions the two-feature line does not have: a large house in Northridge Heights is not the same object as a large partial sale in Edwards. Public write-ups of this contest commonly report a public score near 0.12 after that machinery. This path does not fit those models and does not claim that score.

The number this path stands behind is the holdout RMSLE of about 0.183, on 292 training houses, for $\ln$ price on quality and living area. The sample file remains a separate benchmark. Its first entry, for house 1461, is 169277.0524984 dollars.


```python
# Locate the contest tables beside this notebook, or under the course root.
from pathlib import Path
import csv

def manifest_path(name):
    here = Path.cwd()
    for folder in [here, *here.parents]:
        nested = folder / "15-House Prices" / "data" / name
        direct = folder / "data" / name
        if nested.exists():
            return nested
        if direct.exists() and (folder / "data" / "data_description.txt").exists():
            return direct
    raise FileNotFoundError(name)

def read_manifest(name):
    with manifest_path(name).open(newline="", encoding="utf-8") as handle:
        return list(csv.DictReader(handle))

train = read_manifest("train.csv")
test = read_manifest("test.csv")
submission = read_manifest("sample_submission.csv")
print(len(train), len(test), len(submission))
print(train[0]["Id"], train[0]["SalePrice"])
```

```python
# The sample benchmark is a different number from the training mean.
from fractions import Fraction
prices = [int(row["SalePrice"]) for row in train]
mean = Fraction(sum(prices), len(prices))
sample = float(submission[0]["SalePrice"])
print(float(mean), sample, submission[0]["Id"])
assert submission[0]["Id"] == "1461"
assert abs(sample - 169277.0524984) < 1e-6
assert abs(float(mean) - sample) > 1000
assert [row["Id"] for row in test] == [row["Id"] for row in submission]
```

## Pitfall

A public score near 0.12 is someone else's result on the hidden test prices. The 0.183 in the previous lesson is reproducible from `train.csv` because both the fit and the score stay inside that file. Those two numbers answer different questions.

## Summary

The audited model is a log line on overall quality and living area, with holdout RMSLE about 0.183. Filling every `NA` alike, numbering `MSSubClass`, and copying neighborhood means are the shortcuts the counts do not support. The sample file's first price is a benchmark, not the training mean.
