# The 1460 prices

Add the training prices before fitting anything.

$$
\begin{aligned}
\sum \mathrm{SalePrice} &= 264144946 \\
\dfrac{264144946}{1460} &= \dfrac{132072473}{730}
\end{aligned}
$$

The fraction $132072473/730$ is the mean, about 180921 dollars. The smallest price is 34900 and the largest is 755000.

There are 1460 prices, an even count, so the median is the average of the 730th and 731st prices in sorted order. Those two prices are both 163000, so the median is 163000 exactly. The mean sits above the median: a few expensive sales pull the average up.

The constant prediction "every house costs the mean" is the best single dollar amount if the score is the squared dollar error. The contest score is a squared log error. On the three toy prices, that score preferred the geometric mean. The same comparison on these 1460 prices is the content of the holdout lesson; here the two summaries are simply both true. The mean is $132072473/730$. The median is 163000.


```python
# Locate the contest tables beside this notebook, or under the course root.
from pathlib import Path
import csv

def manifest_path(name):
    here = Path.cwd()
    for folder in [here, *here.parents]:
        nested = folder / "15-House Prices" / "data" / name
        direct = folder / "data" / name
        if nested.exists():
            return nested
        if direct.exists() and (folder / "data" / "data_description.txt").exists():
            return direct
    raise FileNotFoundError(name)

def read_manifest(name):
    with manifest_path(name).open(newline="", encoding="utf-8") as handle:
        return list(csv.DictReader(handle))

train = read_manifest("train.csv")
test = read_manifest("test.csv")
submission = read_manifest("sample_submission.csv")
print(len(train), len(test), len(submission))
print(train[0]["Id"], train[0]["SalePrice"])
```

```python
# Sum, mean, and the two central sorted prices.
from fractions import Fraction
prices = sorted(int(row["SalePrice"]) for row in train)
total = sum(prices)
mean = Fraction(total, len(prices))
print(total, prices[0], prices[-1])
print(prices[729], prices[730], mean)
assert len(prices) == 1460
assert total == 264144946
assert prices[0] == 34900
assert prices[-1] == 755000
assert prices[729] == prices[730] == 163000
assert mean == Fraction(132072473, 730)
assert mean > 163000
```
